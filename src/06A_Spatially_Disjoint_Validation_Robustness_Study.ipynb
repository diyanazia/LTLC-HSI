```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# GOOGLE COLAB RECOVERY — CORRECTION CELL 1A
# LOCATE + AUTHENTICATE MIGRATION ZIP
#
# CORRECTS:
#   Cell 1 assumed:
#       /content/drive/MyDrive/CSE498/LTLC/migration/
#
#   Actual upload location is currently unknown.
#
# THIS CELL:
#   - searches Google Drive for the exact migration filename
#   - authenticates size + SHA-256
#   - performs NO extraction
#   - performs NO training
#   - performs NO inference
#   - performs NO project writes
#   - official test remains SEALED
# ==================================================================================================

from pathlib import Path
import hashlib


print("=" * 164)
print("LTLC NOTEBOOK 06A / EXPERIMENT B")
print("GOOGLE COLAB RECOVERY — CORRECTION CELL 1A")
print("LOCATE + AUTHENTICATE MIGRATION ZIP")
print("=" * 164)


# ==================================================================================================
# 1. EXPECTED ARTIFACT
# ==================================================================================================

DRIVE_ROOT = Path(
    "/content/drive/MyDrive"
)

EXPECTED_FILENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "28of36_PreLA_Kaggle_to_Colab_"
    "SELF_CONTAINED_MIGRATION_v1.zip"
)

EXPECTED_SIZE_BYTES = 358222272

EXPECTED_SHA256 = (
    "8d2798b570eefcc991308455ef5d28cea"
    "3e03c280250ccda0a8a9b356726b363"
)


# ==================================================================================================
# 2. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as f:

        for chunk in iter(
            lambda: f.read(
                chunk_size
            ),
            b"",
        ):

            h.update(
                chunk
            )

    return h.hexdigest()


# ==================================================================================================
# 3. DRIVE ROOT GATE
# ==================================================================================================

print()
print("-" * 164)
print("[1] GOOGLE DRIVE ROOT")
print("-" * 164)


print(
    "Drive root:"
)

print(
    DRIVE_ROOT
)


if not DRIVE_ROOT.is_dir():

    raise FileNotFoundError(
        "STOP: /content/drive/MyDrive is not available."
    )


print()
print(
    "Google Drive root available: PASS"
)


# ==================================================================================================
# 4. EXACT FILENAME SEARCH
# ==================================================================================================

print()
print("-" * 164)
print("[2] EXACT MIGRATION ZIP SEARCH")
print("-" * 164)

print(
    "Searching Google Drive for:"
)

print(
    EXPECTED_FILENAME
)

print()


exact_matches = sorted(
    p
    for p in DRIVE_ROOT.rglob(
        EXPECTED_FILENAME
    )
    if p.is_file()
)


print(
    "Exact matches found:",
    len(
        exact_matches
    )
)


for path in exact_matches:

    print(
        " -",
        path
    )


# ==================================================================================================
# 5. FALLBACK CANDIDATE DISCOVERY IF EXACT NAME WAS CHANGED
# ==================================================================================================

if len(
    exact_matches
) == 0:

    print()
    print("-" * 164)
    print("[3] FALLBACK MIGRATION CANDIDATE SEARCH")
    print("-" * 164)


    fallback_matches = sorted(
        p
        for p in DRIVE_ROOT.rglob(
            "*.zip"
        )
        if (
            p.is_file()
            and
            "28of36" in p.name
            and
            "Colab" in p.name
        )
    )


    print(
        "Possible migration ZIP candidates:",
        len(
            fallback_matches
        )
    )


    for path in fallback_matches:

        print()
        print(
            "Candidate:"
        )

        print(
            path
        )

        print(
            "Size:",
            int(
                path.stat().st_size
            )
        )


    raise FileNotFoundError(
        "STOP: exact migration filename was not found.\n"
        "Candidate paths, if any, are printed above.\n"
        "Do NOT extract anything yet."
    )


if len(
    exact_matches
) > 1:

    raise RuntimeError(
        "STOP: more than one exact migration ZIP exists.\n"
        "Do not choose one silently."
    )


# Exactly one match.
MIGRATION_ZIP = exact_matches[
    0
]

MIGRATION_DIR = MIGRATION_ZIP.parent


# ==================================================================================================
# 6. AUTHENTICATE SIZE
# ==================================================================================================

print()
print("-" * 164)
print("[3] LOCATED ARTIFACT")
print("-" * 164)

print(
    "Actual migration directory:"
)

print(
    MIGRATION_DIR
)

print()
print(
    "Actual migration ZIP:"
)

print(
    MIGRATION_ZIP
)


actual_size = int(
    MIGRATION_ZIP.stat().st_size
)


print()
print(
    "Expected size:",
    EXPECTED_SIZE_BYTES
)

print(
    "Actual size  :",
    actual_size
)


if actual_size != EXPECTED_SIZE_BYTES:

    raise RuntimeError(
        "STOP: located ZIP has the correct filename "
        "but the wrong file size."
    )


print()
print(
    "Migration ZIP size: PASS"
)


# ==================================================================================================
# 7. AUTHENTICATE SHA-256
# ==================================================================================================

print()
print("-" * 164)
print("[4] SHA-256 AUTHENTICATION")
print("-" * 164)

print(
    "Calculating SHA-256..."
)


actual_sha256 = sha256_file(
    MIGRATION_ZIP
)


print()
print(
    "Expected SHA-256:"
)

print(
    EXPECTED_SHA256
)

print()
print(
    "Actual SHA-256:"
)

print(
    actual_sha256
)


if actual_sha256 != EXPECTED_SHA256:

    raise RuntimeError(
        "STOP: located migration ZIP SHA-256 mismatch.\n"
        "Do NOT extract."
    )


print()
print(
    "Migration ZIP SHA-256: PASS"
)


# ==================================================================================================
# 8. RETAIN AUTHENTICATED LOCATION
# ==================================================================================================

COLAB_MIGRATION_ZIP = (
    MIGRATION_ZIP
)

COLAB_MIGRATION_DIR = (
    MIGRATION_DIR
)


# ==================================================================================================
# 9. SAFETY BOUNDARY
# ==================================================================================================

print()
print("-" * 164)
print("[5] RECOVERY SAFETY BOUNDARY")
print("-" * 164)

print(
    "Preserved scientific state : 28/36"
)

print(
    "Balanced Softmax            : 6/6 COMPLETE"
)

print(
    "LA-loss                     : 0/8"
)

print(
    "Extraction performed        : False"
)

print(
    "Training performed          : False"
)

print(
    "Validation inference        : False"
)

print(
    "Official test              : SEALED"
)


# ==================================================================================================
# 10. FINAL GATE
# ==================================================================================================

print()
print("=" * 164)
print("COLAB RECOVERY — CORRECTION CELL 1A FINAL GATE")
print("=" * 164)

print(
    "Google Drive root available             : PASS"
)

print(
    "Exact migration ZIP uniquely located    : PASS"
)

print(
    "Migration ZIP size                      : PASS"
)

print(
    "Migration ZIP SHA-256                   : PASS"
)

print(
    "Authenticated Kaggle artifact           : True"
)

print(
    "Scientific state expected               : 28/36"
)

print(
    "Extraction authorized next              : YES"
)

print(
    "Training authorized                     : NO"
)

print(
    "Official test remains SEALED            : True"
)

print()
print(
    "FINAL COLAB RECOVERY CELL 1A PASS: True"
)

print()
print(
    "STATUS: "
    "MIGRATION_ZIP_LOCATED_AND_AUTHENTICATED__"
    "READY_FOR_VERIFIED_EXTRACTION"
)

print("=" * 164)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# GOOGLE COLAB RECOVERY — CELL 2
# VERIFIED EXTRACTION + 449-FILE RECONSTRUCTION AUDIT
#
# INPUT:
#   Authenticated Kaggle -> Colab migration ZIP from Cell 1A.
#
# THIS CELL:
#   - reauthenticates ZIP identity
#   - reads embedded migration manifest BEFORE extraction
#   - rejects duplicate / unsafe / symlink archive members
#   - extracts into a temporary staging directory
#   - verifies all 449 LTLC files by size + SHA-256
#   - verifies exact preserved runtime assets
#   - promotes ONLY verified content to stable /content paths
#
# NO:
#   - PyTorch installation/change
#   - model construction
#   - DataLoader construction
#   - GPU work
#   - training
#   - validation inference
#   - official-test access
# ==================================================================================================

from pathlib import Path, PurePosixPath
import hashlib
import json
import os
import shutil
import stat
import zipfile


print("=" * 166)
print("LTLC NOTEBOOK 06A / EXPERIMENT B")
print("GOOGLE COLAB RECOVERY — CELL 2")
print("VERIFIED EXTRACTION + 449-FILE RECONSTRUCTION AUDIT")
print("=" * 166)


# ==================================================================================================
# 1. EXACT AUTHENTICATED SOURCE
# ==================================================================================================

MIGRATION_ZIP = Path(
    "/content/drive/MyDrive/CSE498/LTLC/migrations/"
    "LTLC_Notebook06A_ExperimentB_"
    "28of36_PreLA_Kaggle_to_Colab_"
    "SELF_CONTAINED_MIGRATION_v1.zip"
)

EXPECTED_ZIP_SIZE = 358222272

EXPECTED_ZIP_SHA256 = (
    "8d2798b570eefcc991308455ef5d28cea"
    "3e03c280250ccda0a8a9b356726b363"
)


EXPECTED_INTERNAL_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a4947973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_BOOTSTRAP_SHA256 = (
    "f2355458ad09c494bf7a6b04bdcff76eacc26ecab63c5d67d15a3d8b0fd590e9"
)

EXPECTED_LA_RUNTIME_PREFIX_SHA256 = (
    "2b79d77c087038c54dfde043df2f8165d27efd33069a8d57cfec1a6d23cfa624"
)

EXPECTED_RECOVERY_CONTRACT_SHA256 = (
    "6d3a61664b09e3ae1114d213d39a2e510aaf971a080ec75d1e689b21d3b0024a"
)


EXPECTED_PROJECT_FILE_COUNT = 449

EXPECTED_EXPERIMENT_PROGRESS = 28

EXPECTED_TOTAL_JOBS = 36


# ==================================================================================================
# 2. EXTRACTION / FINAL PATHS
# ==================================================================================================

STAGING_ROOT = Path(
    "/content/LTLC_RECOVERY_STAGING_v1"
)

FINAL_ROOT = Path(
    "/content/LTLC"
)

FINAL_RUNTIME_DIR = Path(
    "/content/LTLC_migration_runtime"
)

FINAL_MIGRATION_MANIFEST = Path(
    "/content/LTLC_Kaggle_to_Colab_Migration_Manifest.json"
)


INTERNAL_MANIFEST_NAME = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)

INTERNAL_BOOTSTRAP_NAME = (
    "migration_runtime/R6_SUBPROCESS_BOOTSTRAP.py"
)

INTERNAL_LA_PREFIX_NAME = (
    "migration_runtime/R6_LA_RUNTIME_PREFIX.py"
)

INTERNAL_RECOVERY_CONTRACT_NAME = (
    "migration_runtime/COLAB_RECOVERY_CONTRACT.json"
)


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as f:

        for chunk in iter(
            lambda: f.read(
                chunk_size
            ),
            b"",
        ):
            h.update(
                chunk
            )

    return h.hexdigest()


def sha256_bytes(
    payload,
):

    return hashlib.sha256(
        payload
    ).hexdigest()


def safe_relative_path(
    value,
):

    if not isinstance(
        value,
        str,
    ):

        raise RuntimeError(
            "STOP: archive path is not a string."
        )


    if "\\" in value:

        raise RuntimeError(
            f"STOP: backslash archive path forbidden: {value}"
        )


    p = PurePosixPath(
        value
    )


    if p.is_absolute():

        raise RuntimeError(
            f"STOP: absolute archive path forbidden: {value}"
        )


    if ".." in p.parts:

        raise RuntimeError(
            f"STOP: parent traversal forbidden: {value}"
        )


    if value.strip() == "":

        raise RuntimeError(
            "STOP: empty archive member name."
        )


    return str(
        p
    )


def count_completion_manifests(
    root,
    method_name,
):

    path = (
        root
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )


    if not path.is_dir():
        return 0


    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 4. NO-OVERWRITE RECOVERY GATE
# ==================================================================================================

print()
print("-" * 166)
print("[1] RECOVERY DESTINATION SAFETY")
print("-" * 166)


for path in [
    STAGING_ROOT,
    FINAL_ROOT,
    FINAL_RUNTIME_DIR,
    FINAL_MIGRATION_MANIFEST,
]:

    print(
        f"{str(path):58s}:",
        "ABSENT"
        if not path.exists()
        else "EXISTS",
    )


    if path.exists():

        raise RuntimeError(
            "STOP: a Colab recovery destination already exists:\n"
            f"{path}\n"
            "Nothing will be deleted or overwritten silently."
        )


print()
print(
    "Clean recovery destination: PASS"
)


# ==================================================================================================
# 5. REAUTHENTICATE MIGRATION ZIP
# ==================================================================================================

print()
print("-" * 166)
print("[2] MIGRATION ZIP REAUTHENTICATION")
print("-" * 166)


if not MIGRATION_ZIP.is_file():

    raise FileNotFoundError(
        "STOP: authenticated migration ZIP is no longer available."
    )


actual_zip_size = int(
    MIGRATION_ZIP.stat().st_size
)

actual_zip_sha = sha256_file(
    MIGRATION_ZIP
)


print(
    "Expected size:",
    EXPECTED_ZIP_SIZE
)

print(
    "Actual size  :",
    actual_zip_size
)

print()
print(
    "Expected SHA-256:"
)

print(
    EXPECTED_ZIP_SHA256
)

print()
print(
    "Actual SHA-256:"
)

print(
    actual_zip_sha
)


if actual_zip_size != EXPECTED_ZIP_SIZE:

    raise RuntimeError(
        "STOP: migration ZIP size changed."
    )


if actual_zip_sha != EXPECTED_ZIP_SHA256:

    raise RuntimeError(
        "STOP: migration ZIP SHA-256 changed."
    )


print()
print(
    "Migration ZIP identity: PASS"
)


# ==================================================================================================
# 6. READ + AUTHENTICATE EMBEDDED MANIFEST BEFORE EXTRACTION
# ==================================================================================================

print()
print("-" * 166)
print("[3] EMBEDDED MIGRATION MANIFEST")
print("-" * 166)


with zipfile.ZipFile(
    MIGRATION_ZIP,
    mode="r",
) as zf:

    archive_names = zf.namelist()


    if INTERNAL_MANIFEST_NAME not in archive_names:

        raise RuntimeError(
            "STOP: embedded migration manifest is missing."
        )


    manifest_bytes = zf.read(
        INTERNAL_MANIFEST_NAME
    )


manifest_sha = sha256_bytes(
    manifest_bytes
)


print(
    "Expected manifest SHA-256:"
)

print(
    EXPECTED_INTERNAL_MANIFEST_SHA256
)

print()
print(
    "Actual manifest SHA-256:"
)

print(
    manifest_sha
)


if manifest_sha != EXPECTED_INTERNAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: embedded migration manifest SHA mismatch."
    )


migration_manifest = json.loads(
    manifest_bytes.decode(
        "utf-8"
    )
)


project_snapshot = migration_manifest.get(
    "project_snapshot",
    {}
)

project_records = project_snapshot.get(
    "records",
    []
)


if int(
    project_snapshot.get(
        "file_count",
        -1,
    )
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: manifest project-file count changed."
    )


if len(
    project_records
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: embedded project-record count changed."
    )


scientific_boundary = migration_manifest.get(
    "scientific_boundary",
    {}
)


if int(
    scientific_boundary.get(
        "completed_jobs",
        -1,
    )
) != EXPECTED_EXPERIMENT_PROGRESS:

    raise RuntimeError(
        "STOP: migration scientific progress changed."
    )


if int(
    scientific_boundary.get(
        "frozen_total",
        -1,
    )
) != EXPECTED_TOTAL_JOBS:

    raise RuntimeError(
        "STOP: migration frozen total changed."
    )


if scientific_boundary.get(
    "official_test_status"
) != "SEALED":

    raise RuntimeError(
        "STOP: migration manifest does not preserve SEALED test status."
    )


print()
print(
    "Embedded migration manifest: PASS"
)

print(
    "Manifest project records   : 449 PASS"
)

print(
    "Manifest scientific state  : 28/36 PASS"
)

print(
    "Manifest official test     : SEALED PASS"
)


# ==================================================================================================
# 7. VALIDATE PROJECT RECORDS
# ==================================================================================================

print()
print("-" * 166)
print("[4] PROJECT RECORD SCHEMA")
print("-" * 166)


record_by_archive_name = {}

seen_relative_paths = set()


for record in project_records:

    rel = safe_relative_path(
        record[
            "relative_path"
        ]
    )


    if rel in seen_relative_paths:

        raise RuntimeError(
            f"STOP: duplicate project manifest path: {rel}"
        )


    seen_relative_paths.add(
        rel
    )


    archive_name = (
        "LTLC/"
        +
        rel
    )


    record_by_archive_name[
        archive_name
    ] = {
        "relative_path":
            rel,

        "size_bytes":
            int(
                record[
                    "size_bytes"
                ]
            ),

        "sha256":
            str(
                record[
                    "sha256"
                ]
            ),
    }


if len(
    record_by_archive_name
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: project archive-name record count changed."
    )


print(
    "Unique project paths : 449 PASS"
)


# ==================================================================================================
# 8. ARCHIVE MEMBER-SET + SAFETY AUDIT
# ==================================================================================================

print()
print("-" * 166)
print("[5] ZIP MEMBER-SET + SAFETY AUDIT")
print("-" * 166)


expected_member_names = set(
    record_by_archive_name.keys()
)


expected_member_names.update(
    {
        INTERNAL_MANIFEST_NAME,
        INTERNAL_BOOTSTRAP_NAME,
        INTERNAL_LA_PREFIX_NAME,
        INTERNAL_RECOVERY_CONTRACT_NAME,
    }
)


with zipfile.ZipFile(
    MIGRATION_ZIP,
    mode="r",
) as zf:

    infos = zf.infolist()

    names = [
        info.filename
        for info in infos
    ]


    if len(
        names
    ) != len(
        set(
            names
        )
    ):

        raise RuntimeError(
            "STOP: duplicate member names exist in migration ZIP."
        )


    observed_file_names = set()


    for info in infos:

        safe_name = safe_relative_path(
            info.filename
        )


        mode = (
            info.external_attr
            >>
            16
        )


        if stat.S_ISLNK(
            mode
        ):

            raise RuntimeError(
                f"STOP: symbolic link forbidden in migration ZIP: {safe_name}"
            )


        if info.is_dir():

            continue


        observed_file_names.add(
            safe_name
        )


    missing = sorted(
        expected_member_names
        -
        observed_file_names
    )

    extra = sorted(
        observed_file_names
        -
        expected_member_names
    )


    if missing or extra:

        raise RuntimeError(
            "STOP: migration ZIP member set differs from manifest.\n"
            f"Missing: {missing[:20]}\n"
            f"Extra: {extra[:20]}"
        )


print(
    "Expected file members :",
    len(
        expected_member_names
    ),
)

print(
    "Observed file members :",
    len(
        observed_file_names
    ),
)

print(
    "Duplicate members     : 0 PASS"
)

print(
    "Unsafe paths          : 0 PASS"
)

print(
    "Symbolic links        : 0 PASS"
)

print(
    "Exact member set      : PASS"
)


# ==================================================================================================
# 9. AUTHENTICATE RUNTIME ASSETS DIRECTLY FROM ZIP
# ==================================================================================================

print()
print("-" * 166)
print("[6] PRESERVED RUNTIME-ASSET AUTHENTICATION")
print("-" * 166)


with zipfile.ZipFile(
    MIGRATION_ZIP,
    mode="r",
) as zf:

    bootstrap_bytes = zf.read(
        INTERNAL_BOOTSTRAP_NAME
    )

    la_prefix_bytes = zf.read(
        INTERNAL_LA_PREFIX_NAME
    )

    recovery_contract_bytes = zf.read(
        INTERNAL_RECOVERY_CONTRACT_NAME
    )


bootstrap_sha = sha256_bytes(
    bootstrap_bytes
)

la_prefix_sha = sha256_bytes(
    la_prefix_bytes
)

recovery_contract_sha = sha256_bytes(
    recovery_contract_bytes
)


checks = [
    (
        "R6 subprocess bootstrap",
        bootstrap_sha,
        EXPECTED_BOOTSTRAP_SHA256,
    ),
    (
        "R6 LA runtime prefix",
        la_prefix_sha,
        EXPECTED_LA_RUNTIME_PREFIX_SHA256,
    ),
    (
        "Colab recovery contract",
        recovery_contract_sha,
        EXPECTED_RECOVERY_CONTRACT_SHA256,
    ),
]


for label, observed, expected in checks:

    print()
    print(
        label
    )

    print(
        "  expected:",
        expected
    )

    print(
        "  actual  :",
        observed
    )


    if observed != expected:

        raise RuntimeError(
            f"STOP: {label} SHA mismatch."
        )


print()
print(
    "Preserved runtime assets: 3/3 PASS"
)


# ==================================================================================================
# 10. CREATE TEMPORARY STAGING AREA
# ==================================================================================================

print()
print("-" * 166)
print("[7] SAFE STAGING EXTRACTION")
print("-" * 166)


STAGING_ROOT.mkdir(
    parents=True,
    exist_ok=False,
)


staging_resolved = STAGING_ROOT.resolve()


extracted_files = 0


with zipfile.ZipFile(
    MIGRATION_ZIP,
    mode="r",
) as zf:

    for info in zf.infolist():

        safe_name = safe_relative_path(
            info.filename
        )


        target = (
            STAGING_ROOT
            /
            PurePosixPath(
                safe_name
            )
        )


        target_resolved = target.resolve()


        if (
            target_resolved != staging_resolved
            and
            staging_resolved not in target_resolved.parents
        ):

            raise RuntimeError(
                "STOP: extraction target escaped staging directory."
            )


        if info.is_dir():

            target.mkdir(
                parents=True,
                exist_ok=True,
            )

            continue


        target.parent.mkdir(
            parents=True,
            exist_ok=True,
        )


        with zf.open(
            info,
            mode="r",
        ) as source, target.open(
            "wb"
        ) as destination:

            shutil.copyfileobj(
                source,
                destination,
                length=1024 * 1024,
            )


        extracted_files += 1


        if (
            extracted_files % 50 == 0
            or
            extracted_files
            ==
            len(
                expected_member_names
            )
        ):

            print(
                "Extracted file members:",
                f"{extracted_files}/"
                f"{len(expected_member_names)}"
            )


if extracted_files != len(
    expected_member_names
):

    raise RuntimeError(
        "STOP: extracted file count changed."
    )


print()
print(
    "Safe staging extraction: PASS"
)


# ==================================================================================================
# 11. VERIFY ALL 449 PROJECT FILES IN STAGING
# ==================================================================================================

print()
print("-" * 166)
print("[8] 449-FILE PROJECT CRYPTOGRAPHIC AUDIT")
print("-" * 166)


STAGED_PROJECT_ROOT = (
    STAGING_ROOT
    /
    "LTLC"
)


if not STAGED_PROJECT_ROOT.is_dir():

    raise RuntimeError(
        "STOP: staged LTLC project root missing."
    )


verified_project_files = 0


for archive_name in sorted(
    record_by_archive_name
):

    record = record_by_archive_name[
        archive_name
    ]


    path = (
        STAGED_PROJECT_ROOT
        /
        record[
            "relative_path"
        ]
    )


    if not path.is_file():

        raise FileNotFoundError(
            "STOP: staged project file missing:\n"
            f"{path}"
        )


    actual_size = int(
        path.stat().st_size
    )


    if actual_size != record[
        "size_bytes"
    ]:

        raise RuntimeError(
            "STOP: staged project file-size mismatch:\n"
            f"{record['relative_path']}"
        )


    actual_sha = sha256_file(
        path
    )


    if actual_sha != record[
        "sha256"
    ]:

        raise RuntimeError(
            "STOP: staged project SHA mismatch:\n"
            f"{record['relative_path']}"
        )


    verified_project_files += 1


    if (
        verified_project_files % 50 == 0
        or
        verified_project_files
        ==
        EXPECTED_PROJECT_FILE_COUNT
    ):

        print(
            "Project SHA verified:",
            f"{verified_project_files}/"
            f"{EXPECTED_PROJECT_FILE_COUNT}"
        )


if verified_project_files != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: project verification count changed."
    )


staged_project_files = [
    p
    for p in STAGED_PROJECT_ROOT.rglob("*")
    if p.is_file()
]


if len(
    staged_project_files
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: staged project contains unexpected file count."
    )


print()
print(
    "Project files verified : 449/449 PASS"
)

print(
    "Staged project count   : 449 PASS"
)


# ==================================================================================================
# 12. VERIFY STAGED RUNTIME FILES
# ==================================================================================================

print()
print("-" * 166)
print("[9] STAGED RECOVERY-METADATA AUDIT")
print("-" * 166)


STAGED_RUNTIME_DIR = (
    STAGING_ROOT
    /
    "migration_runtime"
)

STAGED_MANIFEST = (
    STAGING_ROOT
    /
    INTERNAL_MANIFEST_NAME
)


staged_runtime_checks = [
    (
        STAGED_RUNTIME_DIR
        /
        "R6_SUBPROCESS_BOOTSTRAP.py",
        EXPECTED_BOOTSTRAP_SHA256,
    ),
    (
        STAGED_RUNTIME_DIR
        /
        "R6_LA_RUNTIME_PREFIX.py",
        EXPECTED_LA_RUNTIME_PREFIX_SHA256,
    ),
    (
        STAGED_RUNTIME_DIR
        /
        "COLAB_RECOVERY_CONTRACT.json",
        EXPECTED_RECOVERY_CONTRACT_SHA256,
    ),
    (
        STAGED_MANIFEST,
        EXPECTED_INTERNAL_MANIFEST_SHA256,
    ),
]


for path, expected_sha in staged_runtime_checks:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: staged recovery artifact missing:\n{path}"
        )


    actual_sha = sha256_file(
        path
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: staged recovery artifact SHA mismatch:\n{path}"
        )


    print(
        f"{path.name:52s}: PASS"
    )


# ==================================================================================================
# 13. SCIENTIFIC LEDGER FROM STAGED PROJECT
# ==================================================================================================

print()
print("-" * 166)
print("[10] STAGED SCIENTIFIC LEDGER")
print("-" * 166)


ledger = {
    "CE":
        count_completion_manifests(
            STAGED_PROJECT_ROOT,
            "ce",
        ),

    "Focal":
        count_completion_manifests(
            STAGED_PROJECT_ROOT,
            "focal",
        ),

    "LDAM-DRW":
        count_completion_manifests(
            STAGED_PROJECT_ROOT,
            "ldam_drw",
        ),

    "Balanced Softmax":
        count_completion_manifests(
            STAGED_PROJECT_ROOT,
            "balanced_softmax",
        ),

    "LA-loss":
        count_completion_manifests(
            STAGED_PROJECT_ROOT,
            "la_loss",
        ),
}


expected_ledger = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 0,
}


for method_name, expected in expected_ledger.items():

    observed = ledger[
        method_name
    ]


    print(
        f"{method_name:18s}:",
        observed,
    )


    if observed != expected:

        raise RuntimeError(
            f"STOP: staged {method_name} completion count changed."
        )


if sum(
    ledger.values()
) != 28:

    raise RuntimeError(
        "STOP: staged Experiment-B progress is not 28/36."
    )


print()
print(
    "Staged Experiment-B progress : 28/36 PASS"
)

print(
    "Staged LA-loss progress       : 0/8 PASS"
)


# ==================================================================================================
# 14. PROMOTE VERIFIED ARTIFACTS TO STABLE COLAB PATHS
# ==================================================================================================

print()
print("-" * 166)
print("[11] VERIFIED PROMOTION TO /content")
print("-" * 166)


os.replace(
    STAGED_PROJECT_ROOT,
    FINAL_ROOT,
)


os.replace(
    STAGED_RUNTIME_DIR,
    FINAL_RUNTIME_DIR,
)


os.replace(
    STAGED_MANIFEST,
    FINAL_MIGRATION_MANIFEST,
)


# Staging directory should now be empty.
remaining_staging = list(
    STAGING_ROOT.iterdir()
)


if remaining_staging:

    raise RuntimeError(
        "STOP: unexpected files remain in recovery staging directory:\n"
        +
        "\n".join(
            str(p)
            for p in remaining_staging
        )
    )


STAGING_ROOT.rmdir()


print(
    "Verified LTLC project promoted       : PASS"
)

print(
    "Verified migration runtime promoted  : PASS"
)

print(
    "Verified migration manifest promoted : PASS"
)


# ==================================================================================================
# 15. POST-PROMOTION 449-FILE RECHECK
# ==================================================================================================

print()
print("-" * 166)
print("[12] FINAL /content/LTLC RECHECK")
print("-" * 166)


final_project_files = [
    p
    for p in FINAL_ROOT.rglob("*")
    if p.is_file()
]


if len(
    final_project_files
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: final /content/LTLC file count is not 449."
    )


post_verified = 0


for record in project_records:

    rel = record[
        "relative_path"
    ]


    path = (
        FINAL_ROOT
        /
        rel
    )


    if not path.is_file():

        raise FileNotFoundError(
            "STOP: final project file missing:\n"
            f"{rel}"
        )


    if int(
        path.stat().st_size
    ) != int(
        record[
            "size_bytes"
        ]
    ):

        raise RuntimeError(
            "STOP: final project file-size mismatch:\n"
            f"{rel}"
        )


    if sha256_file(
        path
    ) != record[
        "sha256"
    ]:

        raise RuntimeError(
            "STOP: final project SHA mismatch:\n"
            f"{rel}"
        )


    post_verified += 1


    if (
        post_verified % 100 == 0
        or
        post_verified
        ==
        EXPECTED_PROJECT_FILE_COUNT
    ):

        print(
            "Final SHA verified:",
            f"{post_verified}/"
            f"{EXPECTED_PROJECT_FILE_COUNT}"
        )


print()
print(
    "Final /content/LTLC files: 449 PASS"
)

print(
    "Final SHA verification   : 449/449 PASS"
)


# ==================================================================================================
# 16. RETAIN COLAB RECOVERY STATE
# ==================================================================================================

COLAB_ROOT = (
    FINAL_ROOT
)

COLAB_RUNTIME_DIR = (
    FINAL_RUNTIME_DIR
)

COLAB_MIGRATION_MANIFEST = (
    FINAL_MIGRATION_MANIFEST
)

COLAB_MIGRATION_ZIP = (
    MIGRATION_ZIP
)


# ==================================================================================================
# 17. FINAL SAFETY GATE
# ==================================================================================================

print()
print("=" * 166)
print("COLAB RECOVERY — CELL 2 FINAL GATE")
print("=" * 166)

print(
    "Authenticated migration ZIP                : PASS"
)

print(
    "Embedded migration manifest                : PASS"
)

print(
    "Archive path-safety audit                   : PASS"
)

print(
    "Duplicate archive members                  : 0 PASS"
)

print(
    "Archive symbolic links                     : 0 PASS"
)

print(
    "Preserved runtime assets                   : 3/3 PASS"
)

print(
    "Project files extracted                    : 449/449"
)

print(
    "Project files SHA-verified before promotion: 449/449 PASS"
)

print(
    "Project files SHA-verified after promotion : 449/449 PASS"
)

print(
    "Final project root                         : /content/LTLC"
)

print(
    "Experiment-B scientific state              : 28/36"
)

print(
    "CE                                          : 6/6"
)

print(
    "Focal                                       : 8/8"
)

print(
    "LDAM-DRW                                    : 8/8"
)

print(
    "Balanced Softmax                            : 6/6"
)

print(
    "LA-loss                                     : 0/8"
)

print(
    "PyTorch environment changed                 : False"
)

print(
    "Model constructed                           : False"
)

print(
    "DataLoader constructed                      : False"
)

print(
    "Training performed                          : False"
)

print(
    "Validation inference performed              : False"
)

print(
    "Official test remains SEALED                : True"
)

print(
    "Training authorized next                    : NO"
)

print()
print(
    "FINAL COLAB RECOVERY CELL 2 PASS: True"
)

print()
print(
    "STATUS: "
    "KAGGLE_449_FILE_STATE_EXTRACTED_AND_VERIFIED_IN_COLAB__"
    "28_OF_36_SCIENTIFIC_BOUNDARY_RESTORED__"
    "READY_FOR_COLAB_RUNTIME_ENVIRONMENT_PREFLIGHT"
)

print("=" * 166)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# GOOGLE COLAB RECOVERY — CELL 4
# FREEZE COLAB PATH-ADAPTED RUNTIME PACKAGE
#
# INPUT:
#   /content/LTLC_migration_runtime/R6_SUBPROCESS_BOOTSTRAP.py
#   /content/LTLC_migration_runtime/R6_LA_RUNTIME_PREFIX.py
#
# PATH-ONLY ADAPTATION:
#
#   /kaggle/working/LTLC
#           ->
#   /content/LTLC
#
# EXPECTED SUBSTITUTIONS:
#   bootstrap : 1
#   LA prefix : 4
#   total     : 5
#
# IMPORTANT:
#   - Original preserved Kaggle runtime files are NOT modified.
#   - /content/LTLC is NOT modified.
#   - Adapted runtime package is written to Google Drive so it survives
#     the future CPU -> T4 runtime switch.
#   - NO PyTorch installation/change
#   - NO GPU work
#   - NO model
#   - NO DataLoader
#   - NO training
#   - NO inference
#   - official test remains SEALED
# ==================================================================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import os


print("=" * 168)
print("LTLC NOTEBOOK 06A / EXPERIMENT B")
print("GOOGLE COLAB RECOVERY — CELL 4")
print("FREEZE COLAB PATH-ADAPTED RUNTIME PACKAGE")
print("=" * 168)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

SOURCE_RUNTIME_DIR = Path(
    "/content/LTLC_migration_runtime"
)

SOURCE_BOOTSTRAP = (
    SOURCE_RUNTIME_DIR
    / "R6_SUBPROCESS_BOOTSTRAP.py"
)

SOURCE_LA_PREFIX = (
    SOURCE_RUNTIME_DIR
    / "R6_LA_RUNTIME_PREFIX.py"
)


DRIVE_MIGRATION_DIR = Path(
    "/content/drive/MyDrive/CSE498/LTLC/migrations"
)

ADAPTED_RUNTIME_DIR = (
    DRIVE_MIGRATION_DIR
    / "colab_runtime_adapter_v1"
)

ADAPTED_BOOTSTRAP = (
    ADAPTED_RUNTIME_DIR
    / "R6_SUBPROCESS_BOOTSTRAP_COLAB_v1.py"
)

ADAPTED_LA_PREFIX = (
    ADAPTED_RUNTIME_DIR
    / "R6_LA_RUNTIME_PREFIX_COLAB_v1.py"
)

ADAPTATION_MANIFEST = (
    ADAPTED_RUNTIME_DIR
    / "COLAB_RUNTIME_ADAPTATION_MANIFEST_v1.json"
)

ADAPTATION_SHA_SIDECAR = (
    ADAPTED_RUNTIME_DIR
    / "COLAB_RUNTIME_ADAPTATION_SHA256_v1.txt"
)


# ==================================================================================================
# 2. EXPECTED SOURCE IDENTITIES
# ==================================================================================================

EXPECTED_SOURCE_BOOTSTRAP_SHA256 = (
    "f2355458ad09c494bf7a6b04bdcff76eacc26ecab63c5d67d15a3d8b0fd590e9"
)

EXPECTED_SOURCE_LA_PREFIX_SHA256 = (
    "2b79d77c087038c54dfde043df2f8165d27efd33069a8d57cfec1a6d23cfa624"
)

EXPECTED_PROJECT_FILE_COUNT = 449


OLD_ROOT = (
    "/kaggle/working/LTLC"
)

NEW_ROOT = (
    "/content/LTLC"
)


EXPECTED_BOOTSTRAP_REPLACEMENTS = 1
EXPECTED_LA_PREFIX_REPLACEMENTS = 4
EXPECTED_TOTAL_REPLACEMENTS = 5


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as f:

        for chunk in iter(
            lambda: f.read(
                chunk_size
            ),
            b"",
        ):

            h.update(
                chunk
            )

    return h.hexdigest()


def sha256_bytes(
    payload,
):

    return hashlib.sha256(
        payload
    ).hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(
        path
    )

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(
            path
        )
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open(
        "wb"
    ) as f:

        f.write(
            payload
        )

    os.replace(
        tmp,
        path,
    )


# ==================================================================================================
# 4. REASSERT 449-FILE PROJECT BOUNDARY
# ==================================================================================================

print()
print("-" * 168)
print("[1] SCIENTIFIC PROJECT BOUNDARY")
print("-" * 168)


if not ROOT.is_dir():

    raise FileNotFoundError(
        "STOP: /content/LTLC is missing."
    )


project_files_before = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if len(
    project_files_before
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: project-file count is not 449 before path adaptation."
    )


project_snapshot_before = {
    str(
        path.relative_to(
            ROOT
        )
    ):
        sha256_file(
            path
        )
    for path in project_files_before
}


print(
    "Project files before adaptation : 449 PASS"
)

print(
    "Scientific project writes planned: False"
)


# ==================================================================================================
# 5. AUTHENTICATE PRESERVED KAGGLE RUNTIME SOURCES
# ==================================================================================================

print()
print("-" * 168)
print("[2] PRESERVED KAGGLE RUNTIME IDENTITIES")
print("-" * 168)


source_contract = [
    (
        "R6 subprocess bootstrap",
        SOURCE_BOOTSTRAP,
        EXPECTED_SOURCE_BOOTSTRAP_SHA256,
    ),
    (
        "R6 LA runtime prefix",
        SOURCE_LA_PREFIX,
        EXPECTED_SOURCE_LA_PREFIX_SHA256,
    ),
]


for label, path, expected_sha in source_contract:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: preserved runtime source missing:\n{path}"
        )


    actual_sha = sha256_file(
        path
    )


    print()
    print(
        label
    )

    print(
        "  expected:",
        expected_sha
    )

    print(
        "  actual  :",
        actual_sha
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: preserved runtime source SHA mismatch: {label}"
        )


print()
print(
    "Preserved source runtime identity: 2/2 PASS"
)


# ==================================================================================================
# 6. READ SOURCES + AUDIT EXACT OLD-ROOT OCCURRENCES
# ==================================================================================================

print()
print("-" * 168)
print("[3] PATH-SUBSTITUTION SOURCE AUDIT")
print("-" * 168)


bootstrap_text = SOURCE_BOOTSTRAP.read_text(
    encoding="utf-8"
)

la_prefix_text = SOURCE_LA_PREFIX.read_text(
    encoding="utf-8"
)


bootstrap_old_count = bootstrap_text.count(
    OLD_ROOT
)

la_prefix_old_count = la_prefix_text.count(
    OLD_ROOT
)


print(
    "Bootstrap old-root occurrences:",
    bootstrap_old_count
)

print(
    "LA-prefix old-root occurrences:",
    la_prefix_old_count
)

print(
    "Total old-root occurrences:",
    (
        bootstrap_old_count
        +
        la_prefix_old_count
    )
)


if (
    bootstrap_old_count
    !=
    EXPECTED_BOOTSTRAP_REPLACEMENTS
):

    raise RuntimeError(
        "STOP: preserved bootstrap Kaggle-root count changed."
    )


if (
    la_prefix_old_count
    !=
    EXPECTED_LA_PREFIX_REPLACEMENTS
):

    raise RuntimeError(
        "STOP: preserved LA-prefix Kaggle-root count changed."
    )


if (
    bootstrap_old_count
    +
    la_prefix_old_count
    !=
    EXPECTED_TOTAL_REPLACEMENTS
):

    raise RuntimeError(
        "STOP: total path-replacement count is not exactly five."
    )


# ==================================================================================================
# 7. PRINT EXACT SOURCE LINES THAT WILL CHANGE
# ==================================================================================================

print()
print("-" * 168)
print("[4] EXACT PATH LINES TO BE ADAPTED")
print("-" * 168)


def matching_lines(
    text,
    token,
):

    return [
        (
            index,
            line,
        )
        for index, line in enumerate(
            text.splitlines(),
            start=1,
        )
        if token in line
    ]


bootstrap_lines = matching_lines(
    bootstrap_text,
    OLD_ROOT,
)

la_prefix_lines = matching_lines(
    la_prefix_text,
    OLD_ROOT,
)


print()
print(
    "BOOTSTRAP:"
)


for line_number, line in bootstrap_lines:

    print(
        f"  line {line_number}: {line}"
    )


print()
print(
    "LA RUNTIME PREFIX:"
)


for line_number, line in la_prefix_lines:

    print(
        f"  line {line_number}: {line}"
    )


# ==================================================================================================
# 8. CREATE PATH-ONLY ADAPTED TEXT
# ==================================================================================================

adapted_bootstrap_text = bootstrap_text.replace(
    OLD_ROOT,
    NEW_ROOT,
)

adapted_la_prefix_text = la_prefix_text.replace(
    OLD_ROOT,
    NEW_ROOT,
)


# ==================================================================================================
# 9. EXACT DIFFERENCE CONTRACT
# ==================================================================================================

print()
print("-" * 168)
print("[5] PATH-ONLY DIFFERENCE CONTRACT")
print("-" * 168)


if OLD_ROOT in adapted_bootstrap_text:

    raise RuntimeError(
        "STOP: Kaggle project root remains in adapted bootstrap."
    )


if OLD_ROOT in adapted_la_prefix_text:

    raise RuntimeError(
        "STOP: Kaggle project root remains in adapted LA prefix."
    )


remaining_kaggle_tokens = []


for label, text in [
    (
        "bootstrap",
        adapted_bootstrap_text,
    ),
    (
        "LA-prefix",
        adapted_la_prefix_text,
    ),
]:

    for token in [
        "/kaggle/working",
        "/kaggle/input",
        "/kaggle/",
    ]:

        if token in text:

            remaining_kaggle_tokens.append(
                (
                    label,
                    token,
                )
            )


if remaining_kaggle_tokens:

    raise RuntimeError(
        "STOP: Kaggle-specific path token remains after adaptation:\n"
        +
        "\n".join(
            f"{label}: {token}"
            for label, token
            in remaining_kaggle_tokens
        )
    )


# Reverse the intended path substitution.
# If it does not produce the exact original bytes, something besides
# the root path changed.

if adapted_bootstrap_text.replace(
    NEW_ROOT,
    OLD_ROOT,
) != bootstrap_text:

    raise RuntimeError(
        "STOP: bootstrap adaptation changed more than the root path."
    )


if adapted_la_prefix_text.replace(
    NEW_ROOT,
    OLD_ROOT,
) != la_prefix_text:

    raise RuntimeError(
        "STOP: LA-prefix adaptation changed more than the root path."
    )


adapted_bootstrap_new_count = adapted_bootstrap_text.count(
    NEW_ROOT
)

adapted_la_prefix_new_count = adapted_la_prefix_text.count(
    NEW_ROOT
)


if (
    adapted_bootstrap_new_count
    !=
    EXPECTED_BOOTSTRAP_REPLACEMENTS
):

    raise RuntimeError(
        "STOP: adapted bootstrap /content root count changed."
    )


if (
    adapted_la_prefix_new_count
    !=
    EXPECTED_LA_PREFIX_REPLACEMENTS
):

    raise RuntimeError(
        "STOP: adapted LA-prefix /content root count changed."
    )


print(
    "Only intended root substitution occurred : PASS"
)

print(
    "Remaining Kaggle path tokens             : 0 PASS"
)

print(
    "Adapted bootstrap /content refs          : 1 PASS"
)

print(
    "Adapted LA-prefix /content refs          : 4 PASS"
)


# ==================================================================================================
# 10. SYNTAX-COMPILE ADAPTED RUNTIME WITHOUT EXECUTING IT
# ==================================================================================================

print()
print("-" * 168)
print("[6] ADAPTED SOURCE SYNTAX GATE")
print("-" * 168)


compile(
    adapted_bootstrap_text,
    str(
        ADAPTED_BOOTSTRAP
    ),
    "exec",
)


compile(
    adapted_la_prefix_text,
    str(
        ADAPTED_LA_PREFIX
    ),
    "exec",
)


print(
    "Adapted bootstrap compile : PASS"
)

print(
    "Adapted LA prefix compile : PASS"
)

print(
    "Adapted runtime executed  : False"
)


# ==================================================================================================
# 11. COMPUTE ADAPTED SOURCE IDENTITIES
# ==================================================================================================

adapted_bootstrap_bytes = adapted_bootstrap_text.encode(
    "utf-8"
)

adapted_la_prefix_bytes = adapted_la_prefix_text.encode(
    "utf-8"
)


adapted_bootstrap_sha = sha256_bytes(
    adapted_bootstrap_bytes
)

adapted_la_prefix_sha = sha256_bytes(
    adapted_la_prefix_bytes
)


print()
print("-" * 168)
print("[7] ADAPTED RUNTIME IDENTITIES")
print("-" * 168)

print(
    "Adapted bootstrap SHA-256:"
)

print(
    adapted_bootstrap_sha
)

print()
print(
    "Adapted LA-prefix SHA-256:"
)

print(
    adapted_la_prefix_sha
)


# ==================================================================================================
# 12. BUILD DURABLE ADAPTATION MANIFEST
# ==================================================================================================

adaptation_manifest = {
    "schema_version":
        1,

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "artifact_type":
        "COLAB_PATH_ADAPTED_RUNTIME_PACKAGE",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "scientific_boundary": {
        "completed_jobs":
            28,

        "frozen_total":
            36,

        "la_loss_completed":
            0,

        "official_test_status":
            "SEALED",
    },

    "adaptation": {
        "scope":
            "path_only",

        "old_root":
            OLD_ROOT,

        "new_root":
            NEW_ROOT,

        "bootstrap_replacements":
            EXPECTED_BOOTSTRAP_REPLACEMENTS,

        "la_runtime_prefix_replacements":
            EXPECTED_LA_PREFIX_REPLACEMENTS,

        "total_replacements":
            EXPECTED_TOTAL_REPLACEMENTS,

        "scientific_logic_changed":
            False,

        "training_protocol_changed":
            False,

        "data_changed":
            False,

        "model_changed":
            False,

        "loss_changed":
            False,

        "checkpoint_policy_changed":
            False,
    },

    "source_identity": {
        "kaggle_bootstrap_sha256":
            EXPECTED_SOURCE_BOOTSTRAP_SHA256,

        "kaggle_la_runtime_prefix_sha256":
            EXPECTED_SOURCE_LA_PREFIX_SHA256,
    },

    "adapted_identity": {
        "colab_bootstrap_sha256":
            adapted_bootstrap_sha,

        "colab_la_runtime_prefix_sha256":
            adapted_la_prefix_sha,
    },

    "runtime_policy": {
        "required_python":
            "3.12.13",

        "required_torch":
            "2.11.0+cu128",

        "required_cuda":
            "12.8",

        "required_cudnn":
            91900,

        "required_device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,

        "gpu_environment_authenticated":
            False,
    },

    "next_step": {
        "requires_t4_gpu":
            True,

        "action":
            (
                "Switch Colab runtime to T4 GPU, "
                "restore authenticated migration state, "
                "then authenticate CUDA/PyTorch environment "
                "before any training."
            ),

        "training_authorized":
            False,
    },
}


manifest_bytes = json.dumps(
    adaptation_manifest,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 13. WRITE-ONCE DURABLE DRIVE PACKAGE
# ==================================================================================================

print()
print("-" * 168)
print("[8] WRITE-ONCE GOOGLE DRIVE RUNTIME PACKAGE")
print("-" * 168)


ADAPTED_RUNTIME_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


write_contract = [
    (
        ADAPTED_BOOTSTRAP,
        adapted_bootstrap_bytes,
        "Adapted bootstrap",
    ),
    (
        ADAPTED_LA_PREFIX,
        adapted_la_prefix_bytes,
        "Adapted LA prefix",
    ),
    (
        ADAPTATION_MANIFEST,
        manifest_bytes,
        "Adaptation manifest",
    ),
]


for path, payload, label in write_contract:

    if path.exists():

        existing = path.read_bytes()


        if existing != payload:

            raise RuntimeError(
                f"STOP: existing durable {label} differs:\n{path}"
            )


        print(
            f"{label:24s}: EXACT MATCH"
        )

    else:

        atomic_write_bytes(
            path,
            payload,
        )

        print(
            f"{label:24s}: CREATED"
        )


# ==================================================================================================
# 14. PHYSICAL SHA VERIFICATION
# ==================================================================================================

physical_bootstrap_sha = sha256_file(
    ADAPTED_BOOTSTRAP
)

physical_la_prefix_sha = sha256_file(
    ADAPTED_LA_PREFIX
)

physical_manifest_sha = sha256_file(
    ADAPTATION_MANIFEST
)


if physical_bootstrap_sha != adapted_bootstrap_sha:

    raise RuntimeError(
        "STOP: durable adapted bootstrap SHA mismatch."
    )


if physical_la_prefix_sha != adapted_la_prefix_sha:

    raise RuntimeError(
        "STOP: durable adapted LA-prefix SHA mismatch."
    )


# ==================================================================================================
# 15. WRITE SHA SIDECAR
# ==================================================================================================

sha_sidecar_text = (
    f"{physical_bootstrap_sha}  {ADAPTED_BOOTSTRAP.name}\n"
    f"{physical_la_prefix_sha}  {ADAPTED_LA_PREFIX.name}\n"
    f"{physical_manifest_sha}  {ADAPTATION_MANIFEST.name}\n"
)


sha_sidecar_bytes = sha_sidecar_text.encode(
    "utf-8"
)


if ADAPTATION_SHA_SIDECAR.exists():

    if (
        ADAPTATION_SHA_SIDECAR.read_bytes()
        !=
        sha_sidecar_bytes
    ):

        raise RuntimeError(
            "STOP: existing adaptation SHA sidecar differs."
        )

else:

    atomic_write_bytes(
        ADAPTATION_SHA_SIDECAR,
        sha_sidecar_bytes,
    )


print()
print(
    "Adapted bootstrap physical SHA:"
)

print(
    physical_bootstrap_sha
)

print()
print(
    "Adapted LA-prefix physical SHA:"
)

print(
    physical_la_prefix_sha
)

print()
print(
    "Adaptation manifest SHA:"
)

print(
    physical_manifest_sha
)


# ==================================================================================================
# 16. RECHECK SCIENTIFIC PROJECT IS BYTE-IDENTICAL
# ==================================================================================================

print()
print("-" * 168)
print("[9] POST-ADAPTATION SCIENTIFIC PROJECT RECHECK")
print("-" * 168)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if len(
    project_files_after
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: path-adaptation cell changed LTLC project-file count."
    )


project_snapshot_after = {
    str(
        path.relative_to(
            ROOT
        )
    ):
        sha256_file(
            path
        )
    for path in project_files_after
}


if (
    project_snapshot_after
    !=
    project_snapshot_before
):

    raise RuntimeError(
        "STOP: path-adaptation cell changed one or more scientific project files."
    )


print(
    "LTLC project files             : 449 PASS"
)

print(
    "LTLC project byte-identical    : PASS"
)

print(
    "Scientific project files written: False"
)


# ==================================================================================================
# 17. FINAL GATE
# ==================================================================================================

print()
print("=" * 168)
print("COLAB RECOVERY — CELL 4 FINAL GATE")
print("=" * 168)

print(
    "Preserved Kaggle bootstrap identity      : PASS"
)

print(
    "Preserved Kaggle LA-prefix identity      : PASS"
)

print(
    "Expected path substitutions              : 5/5 PASS"
)

print(
    "Scientific logic changed                 : False"
)

print(
    "Training protocol changed                : False"
)

print(
    "Data changed                             : False"
)

print(
    "LA implementation changed                : False"
)

print(
    "Checkpoint policy changed                : False"
)

print(
    "Adapted bootstrap syntax                 : PASS"
)

print(
    "Adapted LA-prefix syntax                 : PASS"
)

print(
    "Durable Drive runtime package            : PASS"
)

print(
    "Restored LTLC project                    : 449 files PASS"
)

print(
    "Restored LTLC project byte-identical     : PASS"
)

print(
    "Experiment-B progress                    : 28/36"
)

print(
    "LA-loss progress                          : 0/8"
)

print(
    "GPU operation performed                  : False"
)

print(
    "Training performed                       : False"
)

print(
    "Official test remains SEALED             : True"
)

print()
print(
    "DRIVE ADAPTED RUNTIME DIRECTORY:"
)

print(
    ADAPTED_RUNTIME_DIR
)

print()
print(
    "COLAB-ADAPTED BOOTSTRAP SHA-256:"
)

print(
    physical_bootstrap_sha
)

print()
print(
    "COLAB-ADAPTED LA PREFIX SHA-256:"
)

print(
    physical_la_prefix_sha
)

print()
print(
    "ADAPTATION MANIFEST SHA-256:"
)

print(
    physical_manifest_sha
)

print()
print(
    "T4 GPU required for THIS cell : False"
)

print(
    "T4 GPU required NEXT          : TRUE"
)

print(
    "Training authorized next      : NO"
)

print()
print(
    "FINAL COLAB RECOVERY CELL 4 PASS: True"
)

print()
print(
    "STATUS: "
    "COLAB_PATH_ADAPTED_RUNTIME_FROZEN_IN_DRIVE__"
    "SCIENTIFIC_STATE_UNCHANGED__"
    "READY_TO_SWITCH_TO_T4_FOR_GPU_RUNTIME_AUTHENTICATION"
)

print("=" * 168)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# GOOGLE COLAB RECOVERY — CELL 5
# T4 RESTORE + EXACT SCIENTIFIC RUNTIME AUTHENTICATION
#
# OPTIMIZATION POLICY:
#   - single Drive read of migration ZIP while copying + SHA hashing
#   - single-pass extraction + per-file SHA verification
#   - no redundant second 449-file hash traversal
#   - no model construction
#   - no DataLoader construction
#   - no training
#   - no inference
#   - no GPU tensors
#
# REQUIRED SCIENTIFIC RUNTIME:
#   Python  : 3.12.13
#   PyTorch : 2.11.0+cu128
#   CUDA    : 12.8
#   cuDNN   : 91900
#   GPU     : Tesla T4
#   device  : cuda:0 only
#
# OFFICIAL TEST REMAINS SEALED.
# ==================================================================================================

from google.colab import drive

from pathlib import Path, PurePosixPath
import hashlib
import inspect
import json
import os
import platform
import shutil
import sys
import zipfile


print("=" * 170)
print("LTLC NOTEBOOK 06A / EXPERIMENT B")
print("GOOGLE COLAB RECOVERY — CELL 5")
print("T4 RESTORE + EXACT SCIENTIFIC RUNTIME AUTHENTICATION")
print("=" * 170)


# ==================================================================================================
# 1. MOUNT DRIVE
# ==================================================================================================

drive.mount(
    "/content/drive"
)


# ==================================================================================================
# 2. DURABLE DRIVE INPUTS
# ==================================================================================================

DRIVE_MIGRATION_DIR = Path(
    "/content/drive/MyDrive/CSE498/LTLC/migrations"
)

MIGRATION_ZIP = (
    DRIVE_MIGRATION_DIR
    /
    "LTLC_Notebook06A_ExperimentB_"
    "28of36_PreLA_Kaggle_to_Colab_"
    "SELF_CONTAINED_MIGRATION_v1.zip"
)


ADAPTER_DIR = (
    DRIVE_MIGRATION_DIR
    /
    "colab_runtime_adapter_v1"
)

ADAPTED_BOOTSTRAP = (
    ADAPTER_DIR
    /
    "R6_SUBPROCESS_BOOTSTRAP_COLAB_v1.py"
)

ADAPTED_LA_PREFIX = (
    ADAPTER_DIR
    /
    "R6_LA_RUNTIME_PREFIX_COLAB_v1.py"
)

ADAPTATION_MANIFEST = (
    ADAPTER_DIR
    /
    "COLAB_RUNTIME_ADAPTATION_MANIFEST_v1.json"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    /
    "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 3. LOCAL RESTORE PATHS
# ==================================================================================================

LOCAL_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

ROOT = Path(
    "/content/LTLC"
)


# ==================================================================================================
# 4. EXPECTED IDENTITIES
# ==================================================================================================

EXPECTED_ZIP_SIZE = 358222272

EXPECTED_ZIP_SHA256 = (
    "8d2798b570eefcc991308455ef5d28cea"
    "3e03c280250ccda0a8a9b356726b363"
)

EXPECTED_INTERNAL_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a4947973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_ADAPTED_BOOTSTRAP_SHA256 = (
    "608d1a883d6366876138ed65f75dd531f62cc9bd33c514333047842828ff7bc7"
)

EXPECTED_ADAPTED_LA_PREFIX_SHA256 = (
    "39d5f2307666a1b220c2d8c4bc4747697fba78f69c806b3552231781533defae"
)

EXPECTED_ADAPTATION_MANIFEST_SHA256 = (
    "ff7306f19c1f02234659b128146d6d46c2dcaadd322b2450f3224cde549ad1cf"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed9633ef5d8a2f1b926744493e3adf0e841"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_DATALOADER_PREFLIGHT_SHA256 = (
    "90e5949c076059afa538697e0bf8a92711afecc97f453342182f7d3134a5d629"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f204a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_PROJECT_FILE_COUNT = 449


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for chunk in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(chunk)

    return h.hexdigest()


def sha256_bytes(
    payload,
):

    return hashlib.sha256(
        payload
    ).hexdigest()


def safe_relative_path(
    value,
):

    if not isinstance(value, str):

        raise RuntimeError(
            "STOP: non-string project path."
        )


    if "\\" in value:

        raise RuntimeError(
            f"STOP: backslash path forbidden: {value}"
        )


    p = PurePosixPath(value)


    if (
        p.is_absolute()
        or
        ".." in p.parts
    ):

        raise RuntimeError(
            f"STOP: unsafe project path: {value}"
        )


    return str(p)


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    tmp = Path(
        str(path)
        +
        ".tmp"
    )


    if tmp.exists():
        tmp.unlink()


    with tmp.open("wb") as f:
        f.write(payload)


    os.replace(
        tmp,
        path,
    )


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        /
        "runs"
        /
        "experiment_b_spatial_validation_v1"
        /
        method_name
    )


    if not path.is_dir():
        return 0


    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. CLEAN NEW-RUNTIME DESTINATION GATE
# ==================================================================================================

print()
print("-" * 170)
print("[1] NEW T4 RUNTIME DESTINATION")
print("-" * 170)


print(
    "/content/LTLC exists:",
    ROOT.exists()
)

print(
    "Local migration ZIP exists:",
    LOCAL_ZIP.exists()
)


if ROOT.exists():

    raise RuntimeError(
        "STOP: /content/LTLC unexpectedly already exists.\n"
        "Do not overwrite a possibly partial recovery."
    )


if LOCAL_ZIP.exists():

    raise RuntimeError(
        "STOP: local migration ZIP unexpectedly already exists."
    )


print()
print(
    "Fresh T4 recovery destination: PASS"
)


# ==================================================================================================
# 7. AUTHENTICATE DURABLE PATH-ADAPTED RUNTIME PACKAGE
# ==================================================================================================

print()
print("-" * 170)
print("[2] DRIVE PATH-ADAPTED RUNTIME PACKAGE")
print("-" * 170)


adapter_contract = [
    (
        "Adapted bootstrap",
        ADAPTED_BOOTSTRAP,
        EXPECTED_ADAPTED_BOOTSTRAP_SHA256,
    ),
    (
        "Adapted LA prefix",
        ADAPTED_LA_PREFIX,
        EXPECTED_ADAPTED_LA_PREFIX_SHA256,
    ),
    (
        "Adaptation manifest",
        ADAPTATION_MANIFEST,
        EXPECTED_ADAPTATION_MANIFEST_SHA256,
    ),
]


for label, path, expected_sha in adapter_contract:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: missing Drive runtime artifact:\n{path}"
        )


    actual_sha = sha256_file(path)


    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL",
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: Drive runtime identity changed: {label}"
        )


print()
print(
    "Drive-adapted runtime package: 3/3 PASS"
)


# ==================================================================================================
# 8. SINGLE-PASS DRIVE -> LOCAL COPY + ZIP SHA AUTHENTICATION
#
# This avoids:
#   copy first
#   then reread the 358 MB Drive file to hash it.
#
# We hash during the single Drive read.
# ==================================================================================================

print()
print("-" * 170)
print("[3] SINGLE-PASS MIGRATION COPY + AUTHENTICATION")
print("-" * 170)


if not MIGRATION_ZIP.is_file():

    raise FileNotFoundError(
        "STOP: migration ZIP missing from Google Drive."
    )


drive_size = int(
    MIGRATION_ZIP.stat().st_size
)


if drive_size != EXPECTED_ZIP_SIZE:

    raise RuntimeError(
        "STOP: Drive migration ZIP size changed."
    )


copy_hasher = hashlib.sha256()

copied_bytes = 0


with MIGRATION_ZIP.open(
    "rb"
) as source, LOCAL_ZIP.open(
    "xb"
) as destination:

    while True:

        block = source.read(
            4 * 1024 * 1024
        )


        if not block:
            break


        destination.write(
            block
        )

        copy_hasher.update(
            block
        )

        copied_bytes += len(
            block
        )


copied_sha = copy_hasher.hexdigest()


print(
    "Copied bytes:",
    copied_bytes
)

print()
print(
    "Expected SHA-256:"
)

print(
    EXPECTED_ZIP_SHA256
)

print()
print(
    "Copy-stream SHA-256:"
)

print(
    copied_sha
)


if copied_bytes != EXPECTED_ZIP_SIZE:

    raise RuntimeError(
        "STOP: copied migration size mismatch."
    )


if copied_sha != EXPECTED_ZIP_SHA256:

    raise RuntimeError(
        "STOP: Drive -> local migration copy SHA mismatch."
    )


print()
print(
    "Authenticated local migration copy: PASS"
)


# ==================================================================================================
# 9. READ AUTHENTICATED EMBEDDED MANIFEST
# ==================================================================================================

print()
print("-" * 170)
print("[4] EMBEDDED 449-FILE MANIFEST")
print("-" * 170)


INTERNAL_MANIFEST_NAME = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


with zipfile.ZipFile(
    LOCAL_ZIP,
    mode="r",
) as zf:

    manifest_bytes = zf.read(
        INTERNAL_MANIFEST_NAME
    )


manifest_sha = sha256_bytes(
    manifest_bytes
)


if manifest_sha != EXPECTED_INTERNAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: embedded migration-manifest SHA changed."
    )


manifest = json.loads(
    manifest_bytes.decode(
        "utf-8"
    )
)


project_snapshot = manifest[
    "project_snapshot"
]

project_records = project_snapshot[
    "records"
]


if int(
    project_snapshot[
        "file_count"
    ]
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: migration project count changed."
    )


if len(
    project_records
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: migration record count changed."
    )


print(
    "Migration manifest SHA : PASS"
)

print(
    "Project records        : 449 PASS"
)

print(
    "Scientific state       : 28/36"
)

print(
    "Official test          : SEALED"
)


# ==================================================================================================
# 10. SINGLE-PASS EXTRACTION + SHA VERIFICATION
#
# Each project file is:
#   decompressed -> written -> hashed
# in ONE pass.
#
# We do not reread the extracted 449 files afterward.
# ==================================================================================================

print()
print("-" * 170)
print("[5] SINGLE-PASS 449-FILE RESTORE + SHA AUDIT")
print("-" * 170)


ROOT.mkdir(
    parents=True,
    exist_ok=False,
)


seen_relative_paths = set()

verified_count = 0


with zipfile.ZipFile(
    LOCAL_ZIP,
    mode="r",
) as zf:

    available_members = set(
        zf.namelist()
    )


    for record in project_records:

        rel = safe_relative_path(
            record[
                "relative_path"
            ]
        )


        if rel in seen_relative_paths:

            raise RuntimeError(
                f"STOP: duplicate project path: {rel}"
            )


        seen_relative_paths.add(rel)


        archive_name = (
            "LTLC/"
            +
            rel
        )


        if archive_name not in available_members:

            raise FileNotFoundError(
                "STOP: expected migration member missing:\n"
                f"{archive_name}"
            )


        target = (
            ROOT
            /
            rel
        )


        target.parent.mkdir(
            parents=True,
            exist_ok=True,
        )


        expected_size = int(
            record[
                "size_bytes"
            ]
        )

        expected_sha = str(
            record[
                "sha256"
            ]
        )


        h = hashlib.sha256()

        written = 0


        with zf.open(
            archive_name,
            mode="r",
        ) as source, target.open(
            "xb"
        ) as destination:

            while True:

                block = source.read(
                    1024 * 1024
                )


                if not block:
                    break


                destination.write(block)

                h.update(block)

                written += len(block)


        if written != expected_size:

            raise RuntimeError(
                f"STOP: restored size mismatch:\n{rel}"
            )


        actual_sha = h.hexdigest()


        if actual_sha != expected_sha:

            raise RuntimeError(
                f"STOP: restored SHA mismatch:\n{rel}"
            )


        verified_count += 1


        if (
            verified_count % 50 == 0
            or
            verified_count
            ==
            EXPECTED_PROJECT_FILE_COUNT
        ):

            print(
                "Restored + SHA verified:",
                f"{verified_count}/"
                f"{EXPECTED_PROJECT_FILE_COUNT}"
            )


if verified_count != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: restored verification count changed."
    )


actual_restored_files = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    actual_restored_files
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: restored project-file count is not 449."
    )


print()
print(
    "Restored project files: 449/449 PASS"
)

print(
    "Per-file SHA audit    : 449/449 PASS"
)


# ==================================================================================================
# 11. CRITICAL SCIENTIFIC ARTIFACT GATE
# ==================================================================================================

print()
print("-" * 170)
print("[6] CRITICAL SCIENTIFIC ARTIFACT GATE")
print("-" * 170)


AUDIT_ROOT = (
    ROOT
    /
    "audit"
    /
    "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        /
        "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        /
        "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "DataLoader preflight",
        AUDIT_ROOT
        /
        "experiment_b_dataset_dataloader_preflight_v1.json",
        EXPECTED_DATALOADER_PREFLIGHT_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        /
        "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        /
        "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        /
        "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        /
        "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        /
        "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    actual_sha = sha256_file(path)


    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL",
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: critical scientific SHA changed: {label}"
        )


# ==================================================================================================
# 12. SCIENTIFIC LEDGER
# ==================================================================================================

print()
print("-" * 170)
print("[7] RESTORED SCIENTIFIC LEDGER")
print("-" * 170)


ledger = {
    "CE":
        count_completion_manifests(
            "ce"
        ),

    "Focal":
        count_completion_manifests(
            "focal"
        ),

    "LDAM-DRW":
        count_completion_manifests(
            "ldam_drw"
        ),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_ledger = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 0,
}


for method, expected in expected_ledger.items():

    observed = ledger[method]

    print(
        f"{method:18s}: {observed}"
    )


    if observed != expected:

        raise RuntimeError(
            f"STOP: restored {method} count changed."
        )


if sum(
    ledger.values()
) != 28:

    raise RuntimeError(
        "STOP: restored Experiment-B boundary is not 28/36."
    )


# ==================================================================================================
# 13. EXACT T4 SCIENTIFIC ENVIRONMENT AUTHENTICATION
# ==================================================================================================

print()
print("-" * 170)
print("[8] T4 SCIENTIFIC RUNTIME AUTHENTICATION")
print("-" * 170)


import torch


python_version = platform.python_version()

torch_version = torch.__version__

torch_cuda = torch.version.cuda

cuda_available = bool(
    torch.cuda.is_available()
)

cudnn_version = (
    torch.backends.cudnn.version()
    if cuda_available
    else None
)

gpu_count = (
    torch.cuda.device_count()
    if cuda_available
    else 0
)

gpu0 = (
    torch.cuda.get_device_name(0)
    if cuda_available
    and
    gpu_count >= 1
    else None
)


print(
    "Python               :",
    python_version
)

print(
    "PyTorch              :",
    torch_version
)

print(
    "CUDA build           :",
    torch_cuda
)

print(
    "CUDA available       :",
    cuda_available
)

print(
    "cuDNN                :",
    cudnn_version
)

print(
    "Visible GPU count    :",
    gpu_count
)

print(
    "GPU0                 :",
    gpu0
)


if python_version != "3.12.13":

    raise RuntimeError(
        "STOP: Python runtime differs from frozen scientific runtime."
    )


if torch_version != "2.11.0+cu128":

    raise RuntimeError(
        "STOP: PyTorch is not exact 2.11.0+cu128.\n"
        "Do NOT train."
    )


if torch_cuda != "12.8":

    raise RuntimeError(
        "STOP: torch CUDA build is not 12.8."
    )


if not cuda_available:

    raise RuntimeError(
        "STOP: CUDA is unavailable."
    )


if int(
    cudnn_version
) != 91900:

    raise RuntimeError(
        "STOP: cuDNN is not exact 91900."
    )


if gpu_count < 1:

    raise RuntimeError(
        "STOP: no CUDA GPU is visible."
    )


if "T4" not in str(gpu0):

    raise RuntimeError(
        "STOP: GPU0 is not a Tesla T4."
    )


print()
print(
    "Exact scientific T4 environment: PASS"
)


# ==================================================================================================
# 14. LOAD PATH-ADAPTED AUTHENTICATED RUNTIME
#
# No import-from-file mechanism is used, so no __pycache__ is created.
# ==================================================================================================

print()
print("-" * 170)
print("[9] COLAB-ADAPTED SCIENTIFIC RUNTIME REHYDRATION")
print("-" * 170)


sys.dont_write_bytecode = True

os.environ[
    "PYTHONDONTWRITEBYTECODE"
] = "1"


bootstrap_source = ADAPTED_BOOTSTRAP.read_text(
    encoding="utf-8"
)

la_prefix_source = ADAPTED_LA_PREFIX.read_text(
    encoding="utf-8"
)


exec(
    compile(
        bootstrap_source,
        str(
            ADAPTED_BOOTSTRAP
        ),
        "exec",
    ),
    globals(),
)


exec(
    compile(
        la_prefix_source,
        str(
            ADAPTED_LA_PREFIX
        ),
        "exec",
    ),
    globals(),
)


# ==================================================================================================
# 15. RUNTIME INTERFACE GATE
# ==================================================================================================

print()
print("-" * 170)
print("[10] LA RUNTIME INTERFACE GATE")
print("-" * 170)


interface_contract = {
    "ExperimentBLALoss":
        "(class_counts, tau)",

    "validate_la_loss":
        "(model, loader, criterion, dataset_name)",

    "better_la_tau_candidate":
        "(candidate, incumbent)",

    "select_experiment_b_la_tau":
        "(dataset_name)",

    "run_experiment_b_la_loss_job":
        "(dataset_name, tau, seed)",
}


for name, expected_signature in interface_contract.items():

    if name not in globals():

        raise RuntimeError(
            f"STOP: LA runtime symbol missing: {name}"
        )


    observed_signature = str(
        inspect.signature(
            globals()[name]
        )
    )


    print(
        f"{name:38s}:",
        observed_signature
    )


    if observed_signature != expected_signature:

        raise RuntimeError(
            f"STOP: LA runtime signature changed: {name}"
        )


if str(
    DEVICE
) != "cuda:0":

    raise RuntimeError(
        "STOP: scientific runtime DEVICE is not cuda:0."
    )


print()
print(
    "Scientific DEVICE:",
    DEVICE
)

print(
    "Runtime interface gate: PASS"
)


# ==================================================================================================
# 16. NO PRETRAINING GPU COMPUTE
# ==================================================================================================

LA_RUN_ROOT = (
    ROOT
    /
    "runs"
    /
    "experiment_b_spatial_validation_v1"
    /
    "la_loss"
)


existing_la_files = []


if LA_RUN_ROOT.is_dir():

    existing_la_files = [
        p
        for p in LA_RUN_ROOT.rglob("*")
        if p.is_file()
    ]


if existing_la_files:

    raise RuntimeError(
        "STOP: LA training artifacts unexpectedly exist "
        "before first Colab LA job."
    )


# ==================================================================================================
# 17. DURABLE T4 RUNTIME GATE
# ==================================================================================================

t4_gate = {
    "schema_version":
        1,

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "status":
        "COLAB_T4_RUNTIME_AUTHENTICATED",

    "scientific_boundary": {
        "completed_jobs":
            28,

        "frozen_total":
            36,

        "la_loss_completed":
            0,

        "official_test_status":
            "SEALED",
    },

    "environment": {
        "python":
            python_version,

        "torch":
            torch_version,

        "cuda":
            torch_cuda,

        "cudnn":
            int(
                cudnn_version
            ),

        "gpu0":
            str(
                gpu0
            ),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "identity": {
        "migration_zip_sha256":
            EXPECTED_ZIP_SHA256,

        "colab_bootstrap_sha256":
            EXPECTED_ADAPTED_BOOTSTRAP_SHA256,

        "colab_la_prefix_sha256":
            EXPECTED_ADAPTED_LA_PREFIX_SHA256,

        "la_engine_gate_sha256":
            EXPECTED_LA_ENGINE_GATE_SHA256,
    },

    "pretraining": {
        "model_constructed":
            False,

        "dataloader_constructed":
            False,

        "gpu_tensor_operation_performed":
            False,

        "training_performed":
            False,

        "validation_inference_performed":
            False,

        "next_authorized_job":
            {
                "dataset":
                    "Pingan",

                "method":
                    "LA_Loss",

                "tau":
                    0.5,

                "seed":
                    42,
            },
    },
}


t4_gate_bytes = json.dumps(
    t4_gate,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


if T4_GATE_PATH.exists():

    if T4_GATE_PATH.read_bytes() != t4_gate_bytes:

        raise RuntimeError(
            "STOP: existing durable T4 runtime gate differs."
        )

else:

    atomic_write_bytes(
        T4_GATE_PATH,
        t4_gate_bytes,
    )


T4_GATE_SHA256 = sha256_file(
    T4_GATE_PATH
)


# ==================================================================================================
# 18. FINAL PROJECT FOOTPRINT
# ==================================================================================================

final_project_files = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    final_project_files
) != EXPECTED_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: runtime rehydration changed the 449-file LTLC tree."
    )


# ==================================================================================================
# 19. RETAIN RUNTIME STATE
# ==================================================================================================

COLAB_ROOT = ROOT

COLAB_ADAPTED_BOOTSTRAP = (
    ADAPTED_BOOTSTRAP
)

COLAB_ADAPTED_LA_PREFIX = (
    ADAPTED_LA_PREFIX
)

COLAB_T4_GATE_PATH = (
    T4_GATE_PATH
)

COLAB_T4_GATE_SHA256 = (
    T4_GATE_SHA256
)


# ==================================================================================================
# 20. FINAL GATE
# ==================================================================================================

print()
print("=" * 170)
print("COLAB RECOVERY — CELL 5 FINAL GATE")
print("=" * 170)

print(
    "Migration ZIP Drive read                : ONE PASS"
)

print(
    "Drive -> local ZIP SHA verification     : PASS"
)

print(
    "449-file restore                        : PASS"
)

print(
    "449-file SHA verification               : PASS"
)

print(
    "Redundant second 449-file hash pass     : SKIPPED"
)

print(
    "Critical scientific artifact gate       : PASS"
)

print(
    "Experiment-B progress                   : 28/36"
)

print(
    "LA-loss progress                         : 0/8"
)

print(
    "Python 3.12.13                           : PASS"
)

print(
    "PyTorch 2.11.0+cu128                    : PASS"
)

print(
    "CUDA 12.8                               : PASS"
)

print(
    "cuDNN 91900                             : PASS"
)

print(
    "Tesla T4                                : PASS"
)

print(
    "Scientific device                       : cuda:0 ONLY"
)

print(
    "Colab-adapted bootstrap                 : PASS"
)

print(
    "Colab-adapted LA runtime                : PASS"
)

print(
    "LA runtime interfaces                   : PASS"
)

print(
    "Project-file count after rehydration    : 449 PASS"
)

print(
    "Model constructed                       : False"
)

print(
    "DataLoader constructed                  : False"
)

print(
    "GPU tensor operation performed          : False"
)

print(
    "Training performed                      : False"
)

print(
    "Validation inference performed          : False"
)

print(
    "Official test remains SEALED            : True"
)

print()
print(
    "T4 RUNTIME GATE SHA-256:"
)

print(
    T4_GATE_SHA256
)

print()
print(
    "NEXT AUTHORIZED SCIENTIFIC JOB:"
)

print(
    "Pingan / LA-loss / tau=0.5 / seed42"
)

print()
print(
    "FINAL COLAB RECOVERY CELL 5 PASS: True"
)

print()
print(
    "STATUS: "
    "COLAB_T4_EXACT_SCIENTIFIC_RUNTIME_AUTHENTICATED__"
    "449_FILE_STATE_RESTORED__"
    "READY_FOR_FIRST_LA_GPU_JOB"
)

print("=" * 170)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# COLAB CELL 6
# PINGAN LA-LOSS tau=0.5 / SEED42
# + IMMEDIATE VERIFIED GOOGLE-DRIVE PERSISTENCE
#
# SCIENTIFIC JOB:
#   Dataset : Pingan
#   Method  : LA_Loss
#   tau     : 0.5
#   Seed    : 42
#   Epochs  : EXACTLY 100
#
# SCIENTIFIC PROTOCOL IS UNCHANGED:
#   - Adam
#   - lr = 0.001 / (1 + 1e-6 * global_step)
#   - batch = 256
#   - dropout = 0.4
#   - no augmentation
#   - no mixed precision
#   - deterministic
#   - authoritative checkpoint = AA -> Macro-F1 -> OA -> earlier epoch
#   - prediction = argmax(raw logits)
#   - official test SEALED
#
# GPU-EFFICIENCY POLICY:
#   - no dummy GPU benchmark
#   - no redundant model construction
#   - no extra validation beyond the frozen runner
#   - no full 449-file post-training rehash
#   - persistence contains ONLY the cumulative post-migration delta
#   - ZIP_STORED used to avoid unnecessary compression CPU overhead
#
# EXPECTED AFTER SUCCESS:
#   Experiment-B : 29/36
#   LA-loss      : 1/8
#   baseline     : 449 files
#   new delta    : exactly 7 files
#   project      : 456 files
#
# AFTER PERSISTENCE:
#   STOP.
#   DO NOT RUN tau=1.0 UNTIL SAVE GATE IS ACKNOWLEDGED.
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 172)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 6")
print("PINGAN LA-LOSS tau=0.5 / SEED42 + IMMEDIATE DRIVE PERSISTENCE")
print("=" * 172)


# ==================================================================================================
# 1. FIXED PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_MIGRATION_DIR = Path(
    "/content/drive/MyDrive/CSE498/LTLC/migrations"
)

ADAPTER_DIR = (
    DRIVE_MIGRATION_DIR
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


PERSISTENCE_DIR = Path(
    "/content/drive/MyDrive/CSE498/LTLC/"
    "experiment_b_colab_persistence"
)


PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "29of36_Post_Pingan_LA_tau0p5_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    /
    (
        PERSISTENCE_BASENAME
        +
        ".zip"
    )
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    /
    (
        PERSISTENCE_BASENAME
        +
        "__MANIFEST.json"
    )
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    /
    (
        PERSISTENCE_BASENAME
        +
        "__SHA256.txt"
    )
)


# ==================================================================================================
# 2. EXACT IDENTITIES
# ==================================================================================================

EXPECTED_MIGRATION_ZIP_SHA256 = (
    "8d2798b570eefcc991308455ef5d28cea"
    "3e03c280250ccda0a8a9b356726b363"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a4947973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71bd69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f204a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed9633ef5d8a2f1b926744493e3adf0e841"
)


BASE_PROJECT_FILE_COUNT = 449

EXPECTED_NEW_DELTA_COUNT = 7

EXPECTED_RESULTING_PROJECT_FILE_COUNT = 456


INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def local_sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as f:

        for chunk in iter(
            lambda: f.read(
                chunk_size
            ),
            b"",
        ):

            h.update(
                chunk
            )

    return h.hexdigest()


def stream_sha256(
    stream,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    while True:

        block = stream.read(
            chunk_size
        )

        if not block:
            break

        h.update(
            block
        )

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(
        path
    )

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open(
        "wb"
    ) as f:

        f.write(
            payload
        )

    os.replace(
        tmp,
        path,
    )


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 4. PRETRAINING T4 / RUNTIME GATE
# ==================================================================================================

print()
print("-" * 172)
print("[1] PRETRAINING T4 RUNTIME GATE")
print("-" * 172)


if torch.__version__ != "2.11.0+cu128":

    raise RuntimeError(
        "STOP: PyTorch runtime changed."
    )


if torch.version.cuda != "12.8":

    raise RuntimeError(
        "STOP: CUDA build changed."
    )


if not torch.cuda.is_available():

    raise RuntimeError(
        "STOP: CUDA unavailable."
    )


if int(
    torch.backends.cudnn.version()
) != 91900:

    raise RuntimeError(
        "STOP: cuDNN changed."
    )


if torch.cuda.device_count() < 1:

    raise RuntimeError(
        "STOP: no GPU visible."
    )


gpu0_name = torch.cuda.get_device_name(
    0
)


if "T4" not in gpu0_name:

    raise RuntimeError(
        "STOP: GPU0 is not Tesla T4."
    )


if str(
    DEVICE
) != "cuda:0":

    raise RuntimeError(
        "STOP: scientific DEVICE changed from cuda:0."
    )


if not T4_GATE_PATH.is_file():

    raise FileNotFoundError(
        "STOP: durable T4 runtime gate missing."
    )


if local_sha256_file(
    T4_GATE_PATH
) != EXPECTED_T4_GATE_SHA256:

    raise RuntimeError(
        "STOP: durable T4 runtime-gate SHA changed."
    )


print(
    "PyTorch            : 2.11.0+cu128 PASS"
)

print(
    "CUDA               : 12.8 PASS"
)

print(
    "cuDNN              : 91900 PASS"
)

print(
    "GPU0               :",
    gpu0_name,
    "PASS"
)

print(
    "Scientific DEVICE  : cuda:0 ONLY PASS"
)

print(
    "T4 runtime gate    : PASS"
)


# ==================================================================================================
# 5. PRETRAINING SCIENTIFIC BOUNDARY
# ==================================================================================================

print()
print("-" * 172)
print("[2] PRETRAINING SCIENTIFIC BOUNDARY")
print("-" * 172)


pre_ledger = {
    "CE":
        count_completion_manifests(
            "ce"
        ),

    "Focal":
        count_completion_manifests(
            "focal"
        ),

    "LDAM-DRW":
        count_completion_manifests(
            "ldam_drw"
        ),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_pre_ledger = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 0,
}


for method, expected in expected_pre_ledger.items():

    observed = pre_ledger[
        method
    ]

    print(
        f"{method:18s}:",
        observed
    )

    if observed != expected:

        raise RuntimeError(
            f"STOP: pretraining {method} count changed."
        )


if sum(
    pre_ledger.values()
) != 28:

    raise RuntimeError(
        "STOP: pretraining Experiment-B state is not 28/36."
    )


current_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    current_before
) != BASE_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: pretraining project-file count is not 449."
    )


# Exact first job must not already exist.
first_completion = la_completion_path(
    "Pingan",
    0.5,
    42,
)


if first_completion.exists():

    raise RuntimeError(
        "STOP: Pingan tau=0.5 seed42 is already complete."
    )


if la_selection_path(
    "Pingan"
).exists():

    raise RuntimeError(
        "STOP: Pingan selected-tau manifest exists before both candidates."
    )


print()
print(
    "Experiment-B      : 28/36 PASS"
)

print(
    "LA-loss           : 0/8 PASS"
)

print(
    "Project files     : 449 PASS"
)

print(
    "Target run absent : PASS"
)

print(
    "Tau selection     : NOT YET CREATED PASS"
)


# ==================================================================================================
# 6. AUTHENTICATE CRITICAL SCIENTIFIC SOURCES
# ==================================================================================================

print()
print("-" * 172)
print("[3] CRITICAL SOURCE IDENTITY")
print("-" * 172)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    if local_sha256_file(
        path
    ) != expected_sha:

        raise RuntimeError(
            f"STOP: critical SHA changed: {label}"
        )

    print(
        f"{label:24s}: PASS"
    )


# ==================================================================================================
# 7. AUTHENTICATE MIGRATION BASE RECORD SET
#
# We use the already-local authenticated ZIP.
# No additional 358 MB Google Drive read.
# ==================================================================================================

print()
print("-" * 172)
print("[4] AUTHENTICATED 449-FILE BASELINE")
print("-" * 172)


if not LOCAL_MIGRATION_ZIP.is_file():

    raise FileNotFoundError(
        "STOP: authenticated local migration ZIP from Cell 5 is missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


migration_manifest_sha = hashlib.sha256(
    migration_manifest_bytes
).hexdigest()


if (
    migration_manifest_sha
    !=
    EXPECTED_MIGRATION_MANIFEST_SHA256
):

    raise RuntimeError(
        "STOP: local migration manifest identity changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    migration_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(
    base_records
) != BASE_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: authenticated migration baseline is not 449 files."
    )


base_paths = {
    str(
        PurePosixPath(
            record[
                "relative_path"
            ]
        )
    )
    for record in base_records
}


if len(
    base_paths
) != BASE_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: baseline contains duplicate project paths."
    )


print(
    "Migration-manifest SHA : PASS"
)

print(
    "Authenticated base     : 449 files PASS"
)


# ==================================================================================================
# 8. EXACT GPU TRAINING JOB
# ==================================================================================================

print()
print("=" * 172)
print("STARTING AUTHORIZED GPU JOB")
print("Pingan / LA-loss / tau=0.5 / seed42")
print("=" * 172)


training_start = time.time()


completion = run_experiment_b_la_loss_job(
    "Pingan",
    0.5,
    42,
)


training_elapsed = (
    time.time()
    -
    training_start
)


print()
print(
    "Authorized runner returned successfully."
)

print(
    "Runner elapsed seconds:",
    round(
        training_elapsed,
        2,
    )
)


# ==================================================================================================
# 9. COMPLETION-MANIFEST AUDIT
# ==================================================================================================

print()
print("-" * 172)
print("[5] COMPLETION-MANIFEST AUDIT")
print("-" * 172)


completion_path = la_completion_path(
    "Pingan",
    0.5,
    42,
)


if not completion_path.is_file():

    raise FileNotFoundError(
        "STOP: expected LA completion manifest missing."
    )


with completion_path.open(
    "r",
    encoding="utf-8",
) as f:

    completion_disk = json.load(
        f
    )


completion_checks = {
    "status":
        completion_disk.get(
            "status"
        )
        ==
        "COMPLETE",

    "dataset":
        completion_disk.get(
            "dataset"
        )
        ==
        "Pingan",

    "method":
        completion_disk.get(
            "method"
        )
        ==
        "LA_Loss",

    "tau":
        float(
            completion_disk.get(
                "tau"
            )
        )
        ==
        0.5,

    "seed":
        int(
            completion_disk.get(
                "seed"
            )
        )
        ==
        42,

    "epochs":
        int(
            completion_disk.get(
                "epochs_completed"
            )
        )
        ==
        100,
}


for label, passed in completion_checks.items():

    print(
        f"{label:16s}:",
        "PASS"
        if passed
        else "FAIL"
    )


if not all(
    completion_checks.values()
):

    raise RuntimeError(
        "STOP: LA completion identity gate failed."
    )


safety = completion_disk[
    "safety"
]


safety_checks = {
    "official_test_status":
        safety.get(
            "official_test_status"
        )
        ==
        "SEALED",

    "official_test_values_accessed":
        safety.get(
            "official_test_values_accessed"
        )
        is False,

    "test_dataset_created":
        safety.get(
            "test_dataset_created"
        )
        is False,

    "test_dataloader_created":
        safety.get(
            "test_dataloader_created"
        )
        is False,

    "test_inference_performed":
        safety.get(
            "test_inference_performed"
        )
        is False,
}


for label, passed in safety_checks.items():

    print(
        f"{label:30s}:",
        "PASS"
        if passed
        else "FAIL"
    )


if not all(
    safety_checks.values()
):

    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


checkpoint_policy = completion_disk[
    "checkpoint_policy"
]


print()
print(
    "Best-AA epoch :",
    checkpoint_policy[
        "best_AA_epoch"
    ]
)

print(
    "Best-AA metrics:"
)

print(
    checkpoint_policy[
        "best_AA_metrics"
    ]
)

print()
print(
    "Best-OA epoch :",
    checkpoint_policy[
        "best_OA_epoch"
    ]
)

print(
    "Best-OA metrics:"
)

print(
    checkpoint_policy[
        "best_OA_metrics"
    ]
)


# ==================================================================================================
# 10. COMPLETION-RECORDED FILE HASH GATE
# ==================================================================================================

print()
print("-" * 172)
print("[6] COMPLETION-RECORDED ARTIFACT HASHES")
print("-" * 172)


completion_files = completion_disk[
    "files"
]


if len(
    completion_files
) != 6:

    raise RuntimeError(
        "STOP: completion manifest must record exactly six run artifacts."
    )


for role, record in completion_files.items():

    relative_path = record[
        "path"
    ]

    expected_sha = record[
        "sha256"
    ]


    path = (
        ROOT
        /
        relative_path
    )


    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: completion-recorded artifact missing:\n{path}"
        )


    actual_sha = local_sha256_file(
        path
    )


    print(
        f"{role:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL"
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: completion-recorded SHA mismatch: {role}"
        )


completion_sha = local_sha256_file(
    completion_path
)


print()
print(
    "Completion manifest SHA-256:"
)

print(
    completion_sha
)


# ==================================================================================================
# 11. INDEPENDENT AUDIT RECORD
# ==================================================================================================

independent_record = completion_files[
    "independent_audit"
]


independent_audit_path = (
    ROOT
    /
    independent_record[
        "path"
    ]
)


with independent_audit_path.open(
    "r",
    encoding="utf-8",
) as f:

    independent_audit = json.load(
        f
    )


if independent_audit.get(
    "pass"
) is not True:

    raise RuntimeError(
        "STOP: independent LA run audit is not PASS."
    )


if independent_audit.get(
    "official_test_status"
) != "SEALED":

    raise RuntimeError(
        "STOP: independent audit test seal changed."
    )


print()
print(
    "Independent checkpoint audit : PASS"
)

print(
    "Independent audit SHA-256    :",
    independent_record[
        "sha256"
    ]
)


# ==================================================================================================
# 12. POST-RUN SCIENTIFIC LEDGER
# ==================================================================================================

print()
print("-" * 172)
print("[7] POST-RUN SCIENTIFIC LEDGER")
print("-" * 172)


post_ledger = {
    "CE":
        count_completion_manifests(
            "ce"
        ),

    "Focal":
        count_completion_manifests(
            "focal"
        ),

    "LDAM-DRW":
        count_completion_manifests(
            "ldam_drw"
        ),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_post_ledger = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 1,
}


for method, expected in expected_post_ledger.items():

    observed = post_ledger[
        method
    ]

    print(
        f"{method:18s}:",
        observed
    )


    if observed != expected:

        raise RuntimeError(
            f"STOP: post-run {method} completion count changed."
        )


if sum(
    post_ledger.values()
) != 29:

    raise RuntimeError(
        "STOP: post-run Experiment-B progress is not 29/36."
    )


if la_selection_path(
    "Pingan"
).exists():

    raise RuntimeError(
        "STOP: tau selection was created before the tau=1.0 candidate."
    )


print()
print(
    "Experiment-B progress : 29/36 PASS"
)

print(
    "LA-loss progress       : 1/8 PASS"
)

print(
    "Tau selected yet       : False PASS"
)


# ==================================================================================================
# 13. EXACT CUMULATIVE DELTA RELATIVE TO 449-FILE MIGRATION BASE
#
# No SHA re-read of the 449 baseline files.
# They were authenticated in Cell 5.
# Only NEW files are hashed.
# ==================================================================================================

print()
print("-" * 172)
print("[8] CUMULATIVE POST-MIGRATION DELTA")
print("-" * 172)


current_project_files = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if len(
    current_project_files
) != EXPECTED_RESULTING_PROJECT_FILE_COUNT:

    raise RuntimeError(
        f"STOP: expected {EXPECTED_RESULTING_PROJECT_FILE_COUNT} "
        f"project files after run, found {len(current_project_files)}."
    )


current_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in current_project_files
}


delta_paths = sorted(
    current_paths
    -
    base_paths
)


if len(
    delta_paths
) != EXPECTED_NEW_DELTA_COUNT:

    raise RuntimeError(
        "STOP: first LA job must create exactly seven "
        "post-migration project files.\n"
        f"Observed delta count: {len(delta_paths)}\n"
        +
        "\n".join(
            delta_paths
        )
    )


expected_run_dir_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/pingan/tau0p5/seed42/"
)


if not all(
    relative_path.startswith(
        expected_run_dir_prefix
    )
    for relative_path in delta_paths
):

    raise RuntimeError(
        "STOP: post-migration delta contains a file outside "
        "the authorized Pingan tau0.5 seed42 run directory."
    )


print(
    "Baseline project files : 449"
)

print(
    "Current project files  : 456"
)

print(
    "Cumulative delta files : 7 PASS"
)


for relative_path in delta_paths:

    print(
        " +",
        relative_path
    )


# ==================================================================================================
# 14. BUILD DELTA RECORDS
# ==================================================================================================

delta_records = []


for relative_path in delta_paths:

    path = (
        ROOT
        /
        relative_path
    )


    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(
                    path.stat().st_size
                ),

            "sha256":
                local_sha256_file(
                    path
                ),
        }
    )


# ==================================================================================================
# 15. BUILD CHAINED PERSISTENCE MANIFEST
# ==================================================================================================

persistence_manifest = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent": {
        "migration_zip_sha256":
            EXPECTED_MIGRATION_ZIP_SHA256,

        "migration_manifest_sha256":
            EXPECTED_MIGRATION_MANIFEST_SHA256,

        "base_project_file_count":
            BASE_PROJECT_FILE_COUNT,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            28,

        "after":
            29,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            1,
    },

    "completed_job": {
        "dataset":
            "Pingan",

        "method":
            "LA_Loss",

        "tau":
            0.5,

        "seed":
            42,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                completion_path.relative_to(
                    ROOT
                )
            ),

        "completion_manifest_sha256":
            completion_sha,

        "best_AA_epoch":
            int(
                checkpoint_policy[
                    "best_AA_epoch"
                ]
            ),

        "best_AA_metrics":
            checkpoint_policy[
                "best_AA_metrics"
            ],

        "best_OA_epoch":
            int(
                checkpoint_policy[
                    "best_OA_epoch"
                ]
            ),

        "best_OA_metrics":
            checkpoint_policy[
                "best_OA_metrics"
            ],
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(
                torch.backends.cudnn.version()
            ),

        "gpu0":
            torch.cuda.get_device_name(
                0
            ),

        "device":
            "cuda:0",
    },

    "full_delta": {
        "base_project_file_count":
            BASE_PROJECT_FILE_COUNT,

        "delta_file_count":
            len(
                delta_records
            ),

        "resulting_project_file_count":
            len(
                current_project_files
            ),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "tau_selected":
            False,
    },

    "next_authorized_job_after_external_save_acknowledgement": {
        "dataset":
            "Pingan",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            42,
    },
}


manifest_bytes = json.dumps(
    persistence_manifest,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 16. NO-OVERWRITE DRIVE PERSISTENCE GATE
# ==================================================================================================

print()
print("-" * 172)
print("[9] IMMEDIATE GOOGLE-DRIVE PERSISTENCE")
print("-" * 172)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for output_path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if output_path.exists():

        raise RuntimeError(
            "STOP: persistence output already exists; "
            "will not overwrite:\n"
            f"{output_path}"
        )


# Write manifest first.
atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = local_sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 17. CREATE SMALL VERIFIED CUMULATIVE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(
        PERSISTENCE_ZIP
    )
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",

    # Intentional:
    # torch checkpoints/resume payloads are already containerized.
    # ZIP_STORED avoids wasting T4-session wall time on compression.
    compression=zipfile.ZIP_STORED,

    allowZip64=True,
) as zf:


    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record[
            "relative_path"
        ]


        source = (
            ROOT
            /
            relative_path
        )


        # Recheck only the seven delta artifacts immediately before archive.
        if local_sha256_file(
            source
        ) != record[
            "sha256"
        ]:

            raise RuntimeError(
                "STOP: delta artifact changed before archive:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=
                (
                    "LTLC_DELTA/"
                    +
                    relative_path
                ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 18. VERIFY DRIVE ZIP DIRECTLY
# ==================================================================================================

verified_delta = 0


record_by_archive_name = {
    (
        "LTLC_DELTA/"
        +
        record[
            "relative_path"
        ]
    ):
        record

    for record in delta_records
}


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:


    expected_members = (
        set(
            record_by_archive_name.keys()
        )
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )


    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:

        raise RuntimeError(
            "STOP: Drive persistence ZIP member-set mismatch."
        )


    archived_manifest = zf.read(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
    )


    if archived_manifest != manifest_bytes:

        raise RuntimeError(
            "STOP: archived persistence manifest differs."
        )


    for archive_name, record in record_by_archive_name.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(
                stream
            )


        if archived_sha != record[
            "sha256"
        ]:

            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified_delta += 1


if verified_delta != EXPECTED_NEW_DELTA_COUNT:

    raise RuntimeError(
        "STOP: Drive persistence verification count changed."
    )


persistence_zip_sha = local_sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 19. SHA SIDECAR
# ==================================================================================================

sidecar = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar.encode(
        "utf-8"
    ),
)


# ==================================================================================================
# 20. FINAL GATE
# ==================================================================================================

print()
print("=" * 172)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 6 FINAL GATE")
print("=" * 172)

print(
    "Authorized scientific job                  : Pingan / LA / tau=0.5 / seed42"
)

print(
    "Epochs completed                           : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction        : PASS"
)

print(
    "Official test                              : SEALED"
)

print(
    "Experiment-B progress                      : 29/36"
)

print(
    "LA-loss progress                           : 1/8"
)

print(
    "Tau selected                               : False"
)

print(
    "Authenticated migration base               : 449 files"
)

print(
    "Current project                            : 456 files"
)

print(
    "Cumulative post-migration delta            : 7 files PASS"
)

print(
    "Drive delta files archived                 : 7/7"
)

print(
    "Drive delta SHA verification               : 7/7 PASS"
)

print(
    "Full 449-file post-training rehash         : SKIPPED"
)

print(
    "Scientific training protocol changed       : False"
)

print(
    "Mixed precision enabled                    : False"
)

print(
    "Multi-GPU used                             : False"
)

print(
    "Scientific device                          : cuda:0 ONLY"
)

print()
print(
    "Best-AA epoch:"
)

print(
    checkpoint_policy[
        "best_AA_epoch"
    ]
)

print(
    "Best-AA metrics:"
)

print(
    checkpoint_policy[
        "best_AA_metrics"
    ]
)

print()
print(
    "Best-OA epoch:"
)

print(
    checkpoint_policy[
        "best_OA_epoch"
    ]
)

print(
    "Best-OA metrics:"
)

print(
    checkpoint_policy[
        "best_OA_metrics"
    ]
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(
        training_elapsed,
        2,
    )
)

print()
print(
    "PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL COLAB CELL 6 PASS: True"
)

print()
print(
    "STATUS: "
    "PINGAN_LA_TAU0P5_SEED42_COMPLETE__"
    "29_OF_36__"
    "7_FILE_FULL_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_TAU1P0"
)

print("=" * 172)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# COLAB CELL 7
# PINGAN LA-LOSS tau=1.0 / SEED42
# + FROZEN TAU SELECTION
# + IMMEDIATE VERIFIED 30/36 GOOGLE-DRIVE PERSISTENCE
#
# AUTHORIZED SCIENTIFIC JOB:
#   Dataset : Pingan
#   Method  : LA_Loss
#   tau     : 1.0
#   Seed    : 42
#   Epochs  : 100
#
# AFTER THIS JOB:
#   Frozen selection rule:
#       validation AA
#       -> Macro-F1
#       -> OA
#       -> smaller tau
#   metric tolerance = 1e-12
#
# SCIENTIFIC PROTOCOL IS NOT MODIFIED.
# official test remains SEALED.
#
# GPU OPTIMIZATION:
#   - no dummy benchmark
#   - no extra training
#   - no AMP
#   - no torch.compile
#   - no multi-GPU
#   - no redundant 449-file SHA traversal
#   - immediate cumulative-delta persistence to Drive
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 7")
print("PINGAN LA-LOSS tau=1.0 / SEED42 + TAU SELECTION + DRIVE PERSISTENCE")
print("=" * 174)


# ==================================================================================================
# 1. FIXED PATHS / IDENTITIES
# ==================================================================================================

ROOT = Path("/content/LTLC")

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

ADAPTER_DIR = (
    DRIVE_ROOT
    / "migrations"
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# Previous externally acknowledged 29/36 persistence artifact.
PREVIOUS_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "29of36_Post_Pingan_LA_tau0p5_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PREVIOUS_ZIP = (
    PERSISTENCE_DIR
    / f"{PREVIOUS_BASENAME}.zip"
)

PREVIOUS_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PREVIOUS_BASENAME}__MANIFEST.json"
)


EXPECTED_PREVIOUS_ZIP_SHA256 = (
    "3eab9b10c6959ae4b01dce4bef6854ad"
    "81cdfb2a813b5ef7e9fb17efac703f22"
)

EXPECTED_PREVIOUS_MANIFEST_SHA256 = (
    "e8acd0f835ec9a65a4525f1bdad2ead"
    "609ac006c091c16d879c44133dc32396e"
)

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "43f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)


BASE_PROJECT_FILE_COUNT = 449
PRE_JOB_PROJECT_FILE_COUNT = 456
EXPECTED_POST_JOB_PROJECT_FILE_COUNT = 464

PRE_JOB_DELTA_COUNT = 7
EXPECTED_POST_JOB_CUMULATIVE_DELTA_COUNT = 15


# ==================================================================================================
# 2. NEW 30/36 PERSISTENCE PATHS
# ==================================================================================================

PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "30of36_Post_Pingan_LA_Tuning_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}.zip"
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__MANIFEST.json"
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__SHA256.txt"
)


INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for chunk in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(chunk)

    return h.hexdigest()


def stream_sha256(stream, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    while True:

        block = stream.read(chunk_size)

        if not block:
            break

        h.update(block)

    return h.hexdigest()


def atomic_write_bytes(path, payload):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(str(path) + ".tmp")

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(tmp, path)


def count_completion_manifests(method_name):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob("*__completion_manifest.json")
        )
    )


# ==================================================================================================
# 4. T4 / RUNTIME GATE
# ==================================================================================================

print()
print("-" * 174)
print("[1] T4 SCIENTIFIC RUNTIME GATE")
print("-" * 174)


if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError("STOP: PyTorch runtime changed.")

if torch.version.cuda != "12.8":
    raise RuntimeError("STOP: CUDA build changed.")

if not torch.cuda.is_available():
    raise RuntimeError("STOP: CUDA unavailable.")

if int(torch.backends.cudnn.version()) != 91900:
    raise RuntimeError("STOP: cuDNN changed.")

if torch.cuda.device_count() < 1:
    raise RuntimeError("STOP: no CUDA GPU visible.")

gpu0 = torch.cuda.get_device_name(0)

if "T4" not in gpu0:
    raise RuntimeError("STOP: GPU0 is not Tesla T4.")

if str(DEVICE) != "cuda:0":
    raise RuntimeError("STOP: scientific DEVICE is not cuda:0.")

if not T4_GATE_PATH.is_file():
    raise FileNotFoundError("STOP: T4 runtime gate missing.")

if sha256_file(T4_GATE_PATH) != EXPECTED_T4_GATE_SHA256:
    raise RuntimeError("STOP: T4 runtime gate SHA changed.")


print("PyTorch           : 2.11.0+cu128 PASS")
print("CUDA              : 12.8 PASS")
print("cuDNN             : 91900 PASS")
print("GPU0              :", gpu0, "PASS")
print("Scientific DEVICE : cuda:0 ONLY PASS")
print("T4 gate           : PASS")


# ==================================================================================================
# 5. EXTERNAL 29/36 SAVE ACKNOWLEDGEMENT GATE
# ==================================================================================================

print()
print("-" * 174)
print("[2] EXTERNALLY ACKNOWLEDGED 29/36 PARENT")
print("-" * 174)


if not PREVIOUS_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: acknowledged 29/36 persistence ZIP is missing."
    )

if not PREVIOUS_MANIFEST.is_file():
    raise FileNotFoundError(
        "STOP: acknowledged 29/36 persistence manifest is missing."
    )


previous_zip_sha = sha256_file(PREVIOUS_ZIP)
previous_manifest_sha = sha256_file(PREVIOUS_MANIFEST)


if previous_zip_sha != EXPECTED_PREVIOUS_ZIP_SHA256:
    raise RuntimeError(
        "STOP: acknowledged 29/36 ZIP SHA changed."
    )

if previous_manifest_sha != EXPECTED_PREVIOUS_MANIFEST_SHA256:
    raise RuntimeError(
        "STOP: acknowledged 29/36 manifest SHA changed."
    )


print("29/36 persistence ZIP      : PASS")
print("29/36 persistence manifest : PASS")
print("External save acknowledged : PASS")


# ==================================================================================================
# 6. PRE-JOB SCIENTIFIC STATE
# ==================================================================================================

print()
print("-" * 174)
print("[3] PRE-JOB SCIENTIFIC BOUNDARY")
print("-" * 174)


pre_ledger = {
    "CE": count_completion_manifests("ce"),
    "Focal": count_completion_manifests("focal"),
    "LDAM-DRW": count_completion_manifests("ldam_drw"),
    "Balanced Softmax": count_completion_manifests("balanced_softmax"),
    "LA-loss": count_completion_manifests("la_loss"),
}


expected_pre = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 1,
}


for method, expected in expected_pre.items():

    observed = pre_ledger[method]

    print(f"{method:18s}: {observed}")

    if observed != expected:
        raise RuntimeError(
            f"STOP: pre-job {method} count changed."
        )


if sum(pre_ledger.values()) != 29:
    raise RuntimeError(
        "STOP: experiment is not at 29/36."
    )


project_files_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(project_files_before) != PRE_JOB_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: pre-job project-file count is not 456."
    )


tau05_completion_path = la_completion_path(
    "Pingan",
    0.5,
    42,
)

tau10_completion_path = la_completion_path(
    "Pingan",
    1.0,
    42,
)

pingan_selection_path = la_selection_path(
    "Pingan"
)


if not tau05_completion_path.is_file():
    raise RuntimeError(
        "STOP: Pingan tau=0.5 seed42 completion missing."
    )

if tau10_completion_path.exists():
    raise RuntimeError(
        "STOP: Pingan tau=1.0 seed42 already exists."
    )

if pingan_selection_path.exists():
    raise RuntimeError(
        "STOP: Pingan tau-selection already exists prematurely."
    )


print()
print("Experiment-B       : 29/36 PASS")
print("LA-loss            : 1/8 PASS")
print("Project files      : 456 PASS")
print("tau=0.5 seed42     : COMPLETE PASS")
print("tau=1.0 seed42     : ABSENT PASS")
print("Pingan tau selected: False PASS")


# ==================================================================================================
# 7. CRITICAL FROZEN SOURCE GATE
# ==================================================================================================

print()
print("-" * 174)
print("[4] FROZEN SCIENTIFIC SOURCE IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    if sha256_file(path) != expected_sha:
        raise RuntimeError(
            f"STOP: critical SHA changed: {label}"
        )

    print(f"{label:24s}: PASS")


# ==================================================================================================
# 8. AUTHENTICATED 449-FILE BASELINE RECORD SET
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATED 28/36 MIGRATION BASE")
print("-" * 174)


if not LOCAL_MIGRATION_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: local authenticated migration ZIP missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


if (
    hashlib.sha256(
        migration_manifest_bytes
    ).hexdigest()
    != EXPECTED_MIGRATION_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: migration-manifest identity changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode("utf-8")
)


base_records = (
    migration_manifest["project_snapshot"]["records"]
)


if len(base_records) != BASE_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: migration baseline count changed."
    )


base_paths = {
    str(PurePosixPath(record["relative_path"]))
    for record in base_records
}


print("Migration manifest : PASS")
print("Base project files : 449 PASS")


# ==================================================================================================
# 9. VERIFY CURRENT 29/36 DELTA SHAPE
# ==================================================================================================

current_before_paths = {
    str(p.relative_to(ROOT))
    for p in project_files_before
}

pre_delta_paths = sorted(
    current_before_paths - base_paths
)


if len(pre_delta_paths) != PRE_JOB_DELTA_COUNT:
    raise RuntimeError(
        "STOP: current 29/36 delta is not exactly seven files."
    )


tau05_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/pingan/tau0p5/seed42/"
)


if not all(
    path.startswith(tau05_prefix)
    for path in pre_delta_paths
):
    raise RuntimeError(
        "STOP: current 29/36 delta contains an unauthorized path."
    )


print("Existing cumulative delta : 7 files PASS")


# ==================================================================================================
# 10. RUN SECOND AUTHORIZED PINGAN TUNING JOB
# ==================================================================================================

print()
print("=" * 174)
print("STARTING AUTHORIZED GPU JOB")
print("Pingan / LA-loss / tau=1.0 / seed42")
print("=" * 174)


training_start = time.time()


completion_tau10 = run_experiment_b_la_loss_job(
    "Pingan",
    1.0,
    42,
)


training_elapsed = time.time() - training_start


print()
print("Authorized runner returned successfully.")
print(
    "Runner elapsed seconds:",
    round(training_elapsed, 2)
)


# ==================================================================================================
# 11. SECOND-CANDIDATE COMPLETION AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[6] tau=1.0 COMPLETION AUDIT")
print("-" * 174)


if not tau10_completion_path.is_file():
    raise FileNotFoundError(
        "STOP: tau=1.0 completion manifest missing."
    )


with tau10_completion_path.open(
    "r",
    encoding="utf-8",
) as f:

    tau10_disk = json.load(f)


identity_checks = {
    "status":
        tau10_disk.get("status") == "COMPLETE",

    "dataset":
        tau10_disk.get("dataset") == "Pingan",

    "method":
        tau10_disk.get("method") == "LA_Loss",

    "tau":
        float(tau10_disk.get("tau")) == 1.0,

    "seed":
        int(tau10_disk.get("seed")) == 42,

    "epochs":
        int(tau10_disk.get("epochs_completed")) == 100,
}


for label, passed in identity_checks.items():

    print(
        f"{label:14s}:",
        "PASS" if passed else "FAIL"
    )


if not all(identity_checks.values()):
    raise RuntimeError(
        "STOP: tau=1.0 completion identity failed."
    )


safety = tau10_disk["safety"]


if not (
    safety.get("official_test_status") == "SEALED"
    and
    safety.get("official_test_values_accessed") is False
    and
    safety.get("test_dataset_created") is False
    and
    safety.get("test_dataloader_created") is False
    and
    safety.get("test_inference_performed") is False
):
    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


tau10_files = tau10_disk["files"]


if len(tau10_files) != 6:
    raise RuntimeError(
        "STOP: tau=1.0 completion must record six run artifacts."
    )


for role, record in tau10_files.items():

    path = ROOT / record["path"]

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: completion artifact missing: {role}"
        )

    if sha256_file(path) != record["sha256"]:
        raise RuntimeError(
            f"STOP: completion SHA mismatch: {role}"
        )

    print(f"{role:24s}: PASS")


if tau10_disk["files"]["independent_audit"]["sha256"] is None:
    raise RuntimeError(
        "STOP: independent audit SHA missing."
    )


print()
print(
    "tau=1.0 Best-AA epoch:",
    tau10_disk["checkpoint_policy"]["best_AA_epoch"]
)

print(
    "tau=1.0 Best-AA metrics:"
)

print(
    tau10_disk["checkpoint_policy"]["best_AA_metrics"]
)

print()
print(
    "tau=1.0 Best-OA epoch:",
    tau10_disk["checkpoint_policy"]["best_OA_epoch"]
)

print(
    "tau=1.0 Best-OA metrics:"
)

print(
    tau10_disk["checkpoint_policy"]["best_OA_metrics"]
)


# ==================================================================================================
# 12. LOAD tau=0.5 COMPLETION FOR SIDE-BY-SIDE TUNING RECORD
# ==================================================================================================

with tau05_completion_path.open(
    "r",
    encoding="utf-8",
) as f:

    tau05_disk = json.load(f)


tau05_best = (
    tau05_disk["checkpoint_policy"]["best_AA_metrics"]
)

tau10_best = (
    tau10_disk["checkpoint_policy"]["best_AA_metrics"]
)


print()
print("-" * 174)
print("[7] PINGAN SEED42 TUNING CANDIDATES")
print("-" * 174)


print("tau=0.5")
print(
    "  epoch    :",
    tau05_disk["checkpoint_policy"]["best_AA_epoch"]
)
print("  AA       :", tau05_best["AA"])
print("  Macro-F1 :", tau05_best["Macro_F1"])
print("  OA       :", tau05_best["OA"])

print()

print("tau=1.0")
print(
    "  epoch    :",
    tau10_disk["checkpoint_policy"]["best_AA_epoch"]
)
print("  AA       :", tau10_best["AA"])
print("  Macro-F1 :", tau10_best["Macro_F1"])
print("  OA       :", tau10_best["OA"])


# ==================================================================================================
# 13. INDEPENDENT FROZEN TAU COMPARATOR
# ==================================================================================================

TOL = 1e-12


def choose_tau_from_best_aa(a_tau, a, b_tau, b):

    if a["AA"] > b["AA"] + TOL:
        return a_tau

    if b["AA"] > a["AA"] + TOL:
        return b_tau

    if a["Macro_F1"] > b["Macro_F1"] + TOL:
        return a_tau

    if b["Macro_F1"] > a["Macro_F1"] + TOL:
        return b_tau

    if a["OA"] > b["OA"] + TOL:
        return a_tau

    if b["OA"] > a["OA"] + TOL:
        return b_tau

    return min(a_tau, b_tau)


independent_selected_tau = choose_tau_from_best_aa(
    0.5,
    tau05_best,
    1.0,
    tau10_best,
)


print()
print(
    "Independent frozen-rule selected tau:",
    independent_selected_tau
)


# ==================================================================================================
# 14. RUN AUTHENTICATED FROZEN SELECTION FUNCTION
# ==================================================================================================

print()
print("-" * 174)
print("[8] AUTHENTICATED PINGAN TAU SELECTION")
print("-" * 174)


selection_result = select_experiment_b_la_tau(
    "Pingan"
)


if not pingan_selection_path.is_file():
    raise FileNotFoundError(
        "STOP: Pingan selection manifest was not created."
    )


selection_sha = sha256_file(
    pingan_selection_path
)


with pingan_selection_path.open(
    "r",
    encoding="utf-8",
) as f:

    selection_disk = json.load(f)


print("Selection function result:")
print(selection_result)

print()
print("Selection manifest:")
print(selection_disk)

print()
print("Selection manifest SHA-256:")
print(selection_sha)

print()
print(
    "Independent expected selected tau:",
    independent_selected_tau
)


# ==================================================================================================
# 15. POST-SELECTION SCIENTIFIC LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[9] POST-SELECTION SCIENTIFIC LEDGER")
print("-" * 174)


post_ledger = {
    "CE": count_completion_manifests("ce"),
    "Focal": count_completion_manifests("focal"),
    "LDAM-DRW": count_completion_manifests("ldam_drw"),
    "Balanced Softmax": count_completion_manifests("balanced_softmax"),
    "LA-loss": count_completion_manifests("la_loss"),
}


expected_post = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 2,
}


for method, expected in expected_post.items():

    observed = post_ledger[method]

    print(f"{method:18s}: {observed}")

    if observed != expected:
        raise RuntimeError(
            f"STOP: post-selection {method} count changed."
        )


if sum(post_ledger.values()) != 30:
    raise RuntimeError(
        "STOP: Experiment-B state is not 30/36."
    )


print()
print("Experiment-B progress : 30/36 PASS")
print("LA-loss progress       : 2/8 PASS")
print("Pingan tuning          : COMPLETE")


# ==================================================================================================
# 16. POST-JOB FILE-FOOTPRINT AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[10] 30/36 PROJECT-FOOTPRINT AUDIT")
print("-" * 174)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if (
    len(project_files_after)
    != EXPECTED_POST_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 464 project files after Pingan tuning.\n"
        f"Observed: {len(project_files_after)}"
    )


after_paths = {
    str(p.relative_to(ROOT))
    for p in project_files_after
}


new_since_29 = sorted(
    after_paths - current_before_paths
)


if len(new_since_29) != 8:
    raise RuntimeError(
        "STOP: second candidate + selection must create exactly "
        "8 new files relative to 29/36."
    )


tau10_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/pingan/tau1p0/seed42/"
)

selection_rel = str(
    pingan_selection_path.relative_to(ROOT)
)


allowed_new = []


for path in new_since_29:

    if (
        path.startswith(tau10_prefix)
        or
        path == selection_rel
    ):
        allowed_new.append(path)

    else:
        raise RuntimeError(
            "STOP: unauthorized new project path:\n"
            f"{path}"
        )


print(
    "Files added since 29/36:",
    len(new_since_29),
    "PASS"
)


for path in new_since_29:
    print(" +", path)


# ==================================================================================================
# 17. CUMULATIVE DELTA FROM ORIGINAL 449-FILE BASE
# ==================================================================================================

cumulative_delta_paths = sorted(
    after_paths - base_paths
)


if (
    len(cumulative_delta_paths)
    != EXPECTED_POST_JOB_CUMULATIVE_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: expected cumulative 30/36 delta of 15 files."
    )


print()
print("Migration base files      : 449")
print("Current project files     : 464")
print("Cumulative delta files    : 15 PASS")


# ==================================================================================================
# 18. HASH ONLY THE 15 DELTA FILES
# ==================================================================================================

delta_records = []


for relative_path in cumulative_delta_paths:

    path = ROOT / relative_path

    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(path.stat().st_size),

            "sha256":
                sha256_file(path),
        }
    )


# ==================================================================================================
# 19. BUILD 30/36 CHAINED PERSISTENCE MANIFEST
# ==================================================================================================

tau10_completion_sha = sha256_file(
    tau10_completion_path
)


persistence_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent_chain": {
        "base_project_file_count":
            449,

        "previous_progress":
            "29/36",

        "previous_persistence_zip_sha256":
            previous_zip_sha,

        "previous_persistence_manifest_sha256":
            previous_manifest_sha,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            29,

        "after":
            30,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            2,
    },

    "completed_job": {
        "dataset":
            "Pingan",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            42,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                tau10_completion_path.relative_to(ROOT)
            ),

        "completion_manifest_sha256":
            tau10_completion_sha,

        "best_AA_epoch":
            int(
                tau10_disk[
                    "checkpoint_policy"
                ][
                    "best_AA_epoch"
                ]
            ),

        "best_AA_metrics":
            tau10_best,

        "best_OA_epoch":
            int(
                tau10_disk[
                    "checkpoint_policy"
                ][
                    "best_OA_epoch"
                ]
            ),

        "best_OA_metrics":
            tau10_disk[
                "checkpoint_policy"
            ][
                "best_OA_metrics"
            ],
    },

    "pingan_tuning": {
        "candidate_tau_0p5_best_AA":
            tau05_best,

        "candidate_tau_1p0_best_AA":
            tau10_best,

        "independent_selected_tau":
            independent_selected_tau,

        "authenticated_selection_manifest_relative_path":
            selection_rel,

        "authenticated_selection_manifest_sha256":
            selection_sha,

        "selection_manifest":
            selection_disk,
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(torch.backends.cudnn.version()),

        "gpu0":
            torch.cuda.get_device_name(0),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "full_delta": {
        "base_project_file_count":
            449,

        "delta_file_count":
            len(delta_records),

        "resulting_project_file_count":
            len(project_files_after),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "pingan_tau_selection_complete":
            True,
    },
}


manifest_bytes = json.dumps(
    persistence_payload,
    indent=2,
    sort_keys=True,
).encode("utf-8")


# ==================================================================================================
# 20. NO-OVERWRITE DRIVE PERSISTENCE
# ==================================================================================================

print()
print("-" * 174)
print("[11] IMMEDIATE 30/36 GOOGLE-DRIVE PERSISTENCE")
print("-" * 174)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if path.exists():
        raise RuntimeError(
            "STOP: 30/36 persistence output already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 21. CREATE CUMULATIVE 15-FILE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(PERSISTENCE_ZIP)
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_STORED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record["relative_path"]

        source = ROOT / relative_path


        if sha256_file(source) != record["sha256"]:
            raise RuntimeError(
                "STOP: delta artifact changed before ZIP creation:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=(
                "LTLC_DELTA/"
                +
                relative_path
            ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 22. DIRECTLY VERIFY DRIVE ZIP
# ==================================================================================================

record_map = {
    (
        "LTLC_DELTA/"
        +
        record["relative_path"]
    ):
        record
    for record in delta_records
}


verified = 0


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:

    expected_members = (
        set(record_map.keys())
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )

    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:
        raise RuntimeError(
            "STOP: 30/36 persistence ZIP member set changed."
        )


    if (
        zf.read(
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        )
        != manifest_bytes
    ):
        raise RuntimeError(
            "STOP: archived 30/36 manifest differs."
        )


    for archive_name, record in record_map.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(stream)


        if archived_sha != record["sha256"]:
            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified += 1


if verified != EXPECTED_POST_JOB_CUMULATIVE_DELTA_COUNT:
    raise RuntimeError(
        "STOP: expected 15/15 archived delta verification."
    )


persistence_zip_sha = sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 23. SHA SIDECAR
# ==================================================================================================

sidecar_text = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar_text.encode("utf-8"),
)


os.sync()


# ==================================================================================================
# 24. FINAL GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 7 FINAL GATE")
print("=" * 174)

print(
    "Authorized job                           : Pingan / LA / tau=1.0 / seed42"
)

print(
    "Epochs                                   : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction      : PASS"
)

print(
    "Pingan tau=0.5 candidate                 : COMPLETE"
)

print(
    "Pingan tau=1.0 candidate                 : COMPLETE"
)

print(
    "Pingan frozen tau selection              : COMPLETE"
)

print(
    "Independent selected tau                 :",
    independent_selected_tau
)

print(
    "Experiment-B progress                    : 30/36"
)

print(
    "LA-loss progress                         : 2/8"
)

print(
    "Original migration base                  : 449 files"
)

print(
    "Current project                          : 464 files"
)

print(
    "Cumulative post-migration delta          : 15 files PASS"
)

print(
    "Drive delta SHA verification             : 15/15 PASS"
)

print(
    "Scientific protocol changed              : False"
)

print(
    "Mixed precision                          : False"
)

print(
    "Multi-GPU                                : False"
)

print(
    "Scientific device                        : cuda:0 ONLY"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "tau=1.0 BEST-AA EPOCH:"
)

print(
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

print(
    "tau=1.0 BEST-AA METRICS:"
)

print(
    tau10_best
)

print()
print(
    "TAU SELECTION MANIFEST:"
)

print(
    selection_disk
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(training_elapsed, 2)
)

print()
print(
    "30/36 PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL COLAB CELL 7 PASS: True"
)

print()
print(
    "STATUS: "
    "PINGAN_LA_TAU1P0_SEED42_COMPLETE__"
    "PINGAN_TAU_SELECTION_FROZEN__"
    "30_OF_36__"
    "15_FILE_CUMULATIVE_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_NEXT_GPU_JOB"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CORRECTED COLAB CELL 7B
#
# PINGAN LA-LOSS tau=1.0 / SEED42
# + AUTHENTICATED FROZEN TAU SELECTION
# + IMMEDIATE VERIFIED 30/36 GOOGLE-DRIVE PERSISTENCE
#
# CORRECTION FROM FAILED CELL 7:
#   ONLY the expected LA executable SHA constant is corrected:
#
#   e80747b1a26266b3adc88d7be2ec3f204a3f50bc833546ae7a83484b67cecae9
#
# SCIENTIFIC PROTOCOL:
#   Dataset      : Pingan
#   Method       : LA_Loss
#   tau          : 1.0
#   seed         : 42
#   epochs       : 100
#
# TAU SELECTION:
#   validation AA
#   -> Macro-F1
#   -> OA
#   -> smaller tau
#   tolerance = 1e-12
#
# NO:
#   - AMP
#   - torch.compile
#   - multi-GPU
#   - changed batch size
#   - changed epochs
#   - changed optimizer
#   - changed validation
#   - official-test access
#
# EXPECTED AFTER SUCCESS:
#   Experiment-B : 30/36
#   LA-loss      : 2/8
#   Project      : 464 files
#   Cumulative delta from migration base : 15 files
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTED COLAB CELL 7B")
print("PINGAN LA-LOSS tau=1.0 / SEED42 + TAU SELECTION + DRIVE PERSISTENCE")
print("=" * 174)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

ADAPTER_DIR = (
    DRIVE_ROOT
    / "migrations"
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 2. AUTHENTICATED 29/36 PARENT
# ==================================================================================================

PREVIOUS_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "29of36_Post_Pingan_LA_tau0p5_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PREVIOUS_ZIP = (
    PERSISTENCE_DIR
    / f"{PREVIOUS_BASENAME}.zip"
)

PREVIOUS_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PREVIOUS_BASENAME}__MANIFEST.json"
)


EXPECTED_PREVIOUS_ZIP_SHA256 = (
    "3eab9b10c6959ae4b01dce4bef6854ad"
    "81cdfb2a813b5ef7e9fb17efac703f22"
)

EXPECTED_PREVIOUS_MANIFEST_SHA256 = (
    "e8acd0f835ec9a65a4525f1bdad2ead"
    "609ac006c091c16d879c44133dc32396e"
)


# ==================================================================================================
# 3. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

# CORRECT 64-character SHA.
EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "4a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)


BASE_PROJECT_FILE_COUNT = 449
PRE_JOB_PROJECT_FILE_COUNT = 456

EXPECTED_POST_PROJECT_FILE_COUNT = 464

EXPECTED_PRE_DELTA_COUNT = 7
EXPECTED_POST_CUMULATIVE_DELTA_COUNT = 15


INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 4. 30/36 PERSISTENCE PATHS
# ==================================================================================================

PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "30of36_Post_Pingan_LA_Tuning_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}.zip"
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__MANIFEST.json"
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__SHA256.txt"
)


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as f:

        for chunk in iter(
            lambda: f.read(
                chunk_size
            ),
            b"",
        ):
            h.update(chunk)

    return h.hexdigest()


def stream_sha256(
    stream,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    while True:

        block = stream.read(
            chunk_size
        )

        if not block:
            break

        h.update(block)

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open(
        "wb"
    ) as f:

        f.write(payload)

    os.replace(
        tmp,
        path,
    )


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. T4 GATE
# ==================================================================================================

print()
print("-" * 174)
print("[1] T4 SCIENTIFIC RUNTIME GATE")
print("-" * 174)


if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError(
        "STOP: PyTorch runtime changed."
    )

if torch.version.cuda != "12.8":
    raise RuntimeError(
        "STOP: CUDA build changed."
    )

if not torch.cuda.is_available():
    raise RuntimeError(
        "STOP: CUDA unavailable."
    )

if int(
    torch.backends.cudnn.version()
) != 91900:
    raise RuntimeError(
        "STOP: cuDNN changed."
    )

if torch.cuda.device_count() < 1:
    raise RuntimeError(
        "STOP: no GPU visible."
    )


gpu0 = torch.cuda.get_device_name(
    0
)


if "T4" not in gpu0:
    raise RuntimeError(
        "STOP: GPU0 is not Tesla T4."
    )

if str(DEVICE) != "cuda:0":
    raise RuntimeError(
        "STOP: scientific DEVICE changed."
    )

if not T4_GATE_PATH.is_file():
    raise FileNotFoundError(
        "STOP: T4 runtime gate missing."
    )

if (
    sha256_file(T4_GATE_PATH)
    !=
    EXPECTED_T4_GATE_SHA256
):
    raise RuntimeError(
        "STOP: T4 runtime-gate SHA changed."
    )


print("PyTorch           : 2.11.0+cu128 PASS")
print("CUDA              : 12.8 PASS")
print("cuDNN             : 91900 PASS")
print("GPU0              :", gpu0, "PASS")
print("Scientific DEVICE : cuda:0 ONLY PASS")
print("T4 gate           : PASS")


# ==================================================================================================
# 7. AUTHENTICATED 29/36 PARENT
# ==================================================================================================

print()
print("-" * 174)
print("[2] EXTERNALLY SAVED 29/36 PARENT")
print("-" * 174)


if not PREVIOUS_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: saved 29/36 ZIP missing."
    )

if not PREVIOUS_MANIFEST.is_file():
    raise FileNotFoundError(
        "STOP: saved 29/36 manifest missing."
    )


previous_zip_sha = sha256_file(
    PREVIOUS_ZIP
)

previous_manifest_sha = sha256_file(
    PREVIOUS_MANIFEST
)


if (
    previous_zip_sha
    !=
    EXPECTED_PREVIOUS_ZIP_SHA256
):
    raise RuntimeError(
        "STOP: saved 29/36 ZIP SHA changed."
    )

if (
    previous_manifest_sha
    !=
    EXPECTED_PREVIOUS_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: saved 29/36 manifest SHA changed."
    )


print("29/36 ZIP      : PASS")
print("29/36 manifest : PASS")
print("Save boundary  : CONFIRMED")


# ==================================================================================================
# 8. PRE-JOB STATE
# ==================================================================================================

print()
print("-" * 174)
print("[3] PRE-JOB SCIENTIFIC STATE")
print("-" * 174)


pre_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_pre = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 1,
}


for method, expected in expected_pre.items():

    observed = pre_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: pre-job {method} count changed."
        )


if sum(
    pre_ledger.values()
) != 29:
    raise RuntimeError(
        "STOP: experiment is not 29/36."
    )


project_files_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(project_files_before)
    !=
    PRE_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 456 files before job."
    )


tau05_completion_path = (
    la_completion_path(
        "Pingan",
        0.5,
        42,
    )
)

tau10_completion_path = (
    la_completion_path(
        "Pingan",
        1.0,
        42,
    )
)

pingan_selection_path = (
    la_selection_path(
        "Pingan"
    )
)


if not tau05_completion_path.is_file():
    raise RuntimeError(
        "STOP: tau=0.5 completion missing."
    )

if tau10_completion_path.exists():
    raise RuntimeError(
        "STOP: tau=1.0 completion already exists."
    )

if pingan_selection_path.exists():
    raise RuntimeError(
        "STOP: Pingan selection exists prematurely."
    )


print()
print("Experiment-B         : 29/36 PASS")
print("LA-loss              : 1/8 PASS")
print("Project files        : 456 PASS")
print("tau=0.5 / seed42     : COMPLETE")
print("tau=1.0 / seed42     : NOT STARTED")
print("Pingan tau selection : ABSENT")


# ==================================================================================================
# 9. FROZEN SOURCE HASH GATE
# ==================================================================================================

print()
print("-" * 174)
print("[4] FROZEN SCIENTIFIC SOURCE IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    actual_sha = sha256_file(
        path
    )

    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL",
    )

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP: genuine SHA mismatch: {label}"
        )


# ==================================================================================================
# 10. LOAD ORIGINAL 449-FILE BASE RECORDS
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATED 28/36 MIGRATION BASE")
print("-" * 174)


if not LOCAL_MIGRATION_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: local authenticated migration ZIP missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


migration_manifest_sha = hashlib.sha256(
    migration_manifest_bytes
).hexdigest()


if (
    migration_manifest_sha
    !=
    EXPECTED_MIGRATION_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: migration manifest SHA changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    migration_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(
    base_records
) != BASE_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: migration baseline is not 449 files."
    )


base_paths = {
    str(
        PurePosixPath(
            record[
                "relative_path"
            ]
        )
    )
    for record in base_records
}


print(
    "Migration manifest : PASS"
)

print(
    "Base project files : 449 PASS"
)


# ==================================================================================================
# 11. VERIFY CURRENT 7-FILE DELTA
# ==================================================================================================

before_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_before
}


pre_delta_paths = sorted(
    before_paths
    -
    base_paths
)


if (
    len(pre_delta_paths)
    !=
    EXPECTED_PRE_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: current 29/36 delta is not seven files."
    )


expected_tau05_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/pingan/tau0p5/seed42/"
)


if not all(
    path.startswith(
        expected_tau05_prefix
    )
    for path in pre_delta_paths
):
    raise RuntimeError(
        "STOP: unauthorized path exists in 29/36 delta."
    )


print(
    "Current cumulative delta : 7 files PASS"
)


# ==================================================================================================
# 12. START REAL GPU JOB
# ==================================================================================================

print()
print("=" * 174)
print("STARTING AUTHORIZED GPU JOB")
print("Pingan / LA-loss / tau=1.0 / seed42")
print("=" * 174)


training_start = time.time()


completion_tau10 = (
    run_experiment_b_la_loss_job(
        "Pingan",
        1.0,
        42,
    )
)


training_elapsed = (
    time.time()
    -
    training_start
)


print()
print(
    "Authorized runner returned successfully."
)

print(
    "Runner elapsed seconds:",
    round(
        training_elapsed,
        2,
    )
)


# ==================================================================================================
# 13. tau=1.0 COMPLETION AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[6] tau=1.0 COMPLETION AUDIT")
print("-" * 174)


if not tau10_completion_path.is_file():
    raise FileNotFoundError(
        "STOP: tau=1.0 completion manifest missing."
    )


with tau10_completion_path.open(
    "r",
    encoding="utf-8",
) as f:

    tau10_disk = json.load(f)


completion_identity = {
    "status":
        tau10_disk.get(
            "status"
        ) == "COMPLETE",

    "dataset":
        tau10_disk.get(
            "dataset"
        ) == "Pingan",

    "method":
        tau10_disk.get(
            "method"
        ) == "LA_Loss",

    "tau":
        float(
            tau10_disk.get(
                "tau"
            )
        ) == 1.0,

    "seed":
        int(
            tau10_disk.get(
                "seed"
            )
        ) == 42,

    "epochs":
        int(
            tau10_disk.get(
                "epochs_completed"
            )
        ) == 100,
}


for label, passed in completion_identity.items():

    print(
        f"{label:14s}:",
        "PASS"
        if passed
        else "FAIL",
    )


if not all(
    completion_identity.values()
):
    raise RuntimeError(
        "STOP: tau=1.0 completion identity failed."
    )


safety = tau10_disk[
    "safety"
]


if not (
    safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    safety.get(
        "official_test_values_accessed"
    ) is False
    and
    safety.get(
        "test_dataset_created"
    ) is False
    and
    safety.get(
        "test_dataloader_created"
    ) is False
    and
    safety.get(
        "test_inference_performed"
    ) is False
):
    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


tau10_files = tau10_disk[
    "files"
]


if len(
    tau10_files
) != 6:
    raise RuntimeError(
        "STOP: tau=1.0 run must contain six recorded artifacts."
    )


for role, record in tau10_files.items():

    path = (
        ROOT
        /
        record[
            "path"
        ]
    )

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: run artifact missing: {role}"
        )

    if (
        sha256_file(
            path
        )
        !=
        record[
            "sha256"
        ]
    ):
        raise RuntimeError(
            f"STOP: run artifact SHA mismatch: {role}"
        )

    print(
        f"{role:24s}: PASS"
    )


tau10_best_aa_epoch = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

tau10_best_aa = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)

tau10_best_oa_epoch = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_OA_epoch"
    ]
)

tau10_best_oa = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_OA_metrics"
    ]
)


print()
print(
    "tau=1.0 Best-AA epoch:",
    tau10_best_aa_epoch
)

print(
    "tau=1.0 Best-AA metrics:"
)

print(
    tau10_best_aa
)

print()
print(
    "tau=1.0 Best-OA epoch:",
    tau10_best_oa_epoch
)

print(
    "tau=1.0 Best-OA metrics:"
)

print(
    tau10_best_oa
)


# ==================================================================================================
# 14. LOAD tau=0.5 CANDIDATE
# ==================================================================================================

with tau05_completion_path.open(
    "r",
    encoding="utf-8",
) as f:

    tau05_disk = json.load(f)


tau05_best_aa_epoch = (
    tau05_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

tau05_best_aa = (
    tau05_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)


print()
print("-" * 174)
print("[7] PINGAN SEED42 CANDIDATE COMPARISON")
print("-" * 174)


print("tau=0.5")
print(
    "  epoch    :",
    tau05_best_aa_epoch
)
print(
    "  AA       :",
    tau05_best_aa["AA"]
)
print(
    "  Macro-F1 :",
    tau05_best_aa["Macro_F1"]
)
print(
    "  OA       :",
    tau05_best_aa["OA"]
)


print()
print("tau=1.0")
print(
    "  epoch    :",
    tau10_best_aa_epoch
)
print(
    "  AA       :",
    tau10_best_aa["AA"]
)
print(
    "  Macro-F1 :",
    tau10_best_aa["Macro_F1"]
)
print(
    "  OA       :",
    tau10_best_aa["OA"]
)


# ==================================================================================================
# 15. INDEPENDENT FROZEN TAU RULE
# ==================================================================================================

METRIC_TOL = 1e-12


def independent_choose_tau(
    tau_a,
    metric_a,
    tau_b,
    metric_b,
):

    if (
        metric_a["AA"]
        >
        metric_b["AA"]
        +
        METRIC_TOL
    ):
        return tau_a

    if (
        metric_b["AA"]
        >
        metric_a["AA"]
        +
        METRIC_TOL
    ):
        return tau_b

    if (
        metric_a["Macro_F1"]
        >
        metric_b["Macro_F1"]
        +
        METRIC_TOL
    ):
        return tau_a

    if (
        metric_b["Macro_F1"]
        >
        metric_a["Macro_F1"]
        +
        METRIC_TOL
    ):
        return tau_b

    if (
        metric_a["OA"]
        >
        metric_b["OA"]
        +
        METRIC_TOL
    ):
        return tau_a

    if (
        metric_b["OA"]
        >
        metric_a["OA"]
        +
        METRIC_TOL
    ):
        return tau_b

    return min(
        tau_a,
        tau_b,
    )


independent_selected_tau = (
    independent_choose_tau(
        0.5,
        tau05_best_aa,
        1.0,
        tau10_best_aa,
    )
)


print()
print(
    "Independent frozen-rule selected tau:",
    independent_selected_tau
)


# ==================================================================================================
# 16. AUTHENTICATED TAU SELECTION
# ==================================================================================================

print()
print("-" * 174)
print("[8] AUTHENTICATED PINGAN TAU SELECTION")
print("-" * 174)


selection_result = (
    select_experiment_b_la_tau(
        "Pingan"
    )
)


if not pingan_selection_path.is_file():
    raise FileNotFoundError(
        "STOP: Pingan selection manifest was not created."
    )


selection_sha = sha256_file(
    pingan_selection_path
)


with pingan_selection_path.open(
    "r",
    encoding="utf-8",
) as f:

    selection_disk = json.load(f)


print(
    "Selection function result:"
)

print(
    selection_result
)

print()
print(
    "Selection manifest:"
)

print(
    selection_disk
)

print()
print(
    "Selection SHA-256:"
)

print(
    selection_sha
)


manifest_selected_tau = (
    selection_disk.get(
        "selected_tau"
    )
)


if manifest_selected_tau is None:
    raise RuntimeError(
        "STOP: selection manifest has no selected_tau field."
    )


if (
    abs(
        float(
            manifest_selected_tau
        )
        -
        float(
            independent_selected_tau
        )
    )
    >
    METRIC_TOL
):
    raise RuntimeError(
        "STOP: authenticated selection disagrees with independent frozen rule."
    )


print()
print(
    "Authenticated selected tau:",
    manifest_selected_tau
)

print(
    "Independent agreement   : PASS"
)


# ==================================================================================================
# 17. POST-SELECTION LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[9] POST-SELECTION SCIENTIFIC LEDGER")
print("-" * 174)


post_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_post = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 2,
}


for method, expected in expected_post.items():

    observed = post_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: post-job {method} count changed."
        )


if sum(
    post_ledger.values()
) != 30:
    raise RuntimeError(
        "STOP: experiment is not 30/36."
    )


print()
print(
    "Experiment-B progress : 30/36 PASS"
)

print(
    "LA-loss progress       : 2/8 PASS"
)

print(
    "Pingan tuning          : COMPLETE PASS"
)


# ==================================================================================================
# 18. FILE-FOOTPRINT AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[10] PROJECT-FOOTPRINT AUDIT")
print("-" * 174)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if (
    len(project_files_after)
    !=
    EXPECTED_POST_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 464 files after Pingan tuning.\n"
        f"Observed: {len(project_files_after)}"
    )


after_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_after
}


new_since_29 = sorted(
    after_paths
    -
    before_paths
)


if len(
    new_since_29
) != 8:
    raise RuntimeError(
        "STOP: tau=1.0 job + selection must add exactly 8 files."
    )


tau10_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/pingan/tau1p0/seed42/"
)

selection_rel = str(
    pingan_selection_path.relative_to(
        ROOT
    )
)


for path in new_since_29:

    if not (
        path.startswith(
            tau10_prefix
        )
        or
        path == selection_rel
    ):
        raise RuntimeError(
            "STOP: unauthorized new path:\n"
            f"{path}"
        )


print(
    "Files added since 29/36:",
    len(
        new_since_29
    ),
    "PASS",
)


for path in new_since_29:
    print(
        " +",
        path
    )


# ==================================================================================================
# 19. CUMULATIVE DELTA FROM 449-FILE BASE
# ==================================================================================================

cumulative_delta_paths = sorted(
    after_paths
    -
    base_paths
)


if (
    len(
        cumulative_delta_paths
    )
    !=
    EXPECTED_POST_CUMULATIVE_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: cumulative 30/36 delta must contain exactly 15 files."
    )


print()
print(
    "Migration base files   : 449"
)

print(
    "Current project files  : 464"
)

print(
    "Cumulative delta files : 15 PASS"
)


# ==================================================================================================
# 20. HASH ONLY THE 15 DELTA FILES
# ==================================================================================================

delta_records = []


for relative_path in cumulative_delta_paths:

    path = (
        ROOT
        /
        relative_path
    )

    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(
                    path.stat().st_size
                ),

            "sha256":
                sha256_file(
                    path
                ),
        }
    )


# ==================================================================================================
# 21. BUILD 30/36 PERSISTENCE MANIFEST
# ==================================================================================================

tau10_completion_sha = sha256_file(
    tau10_completion_path
)


persistence_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent_chain": {
        "base_project_file_count":
            449,

        "previous_progress":
            "29/36",

        "previous_persistence_zip_sha256":
            previous_zip_sha,

        "previous_persistence_manifest_sha256":
            previous_manifest_sha,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            29,

        "after":
            30,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            2,
    },

    "completed_job": {
        "dataset":
            "Pingan",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            42,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                tau10_completion_path.relative_to(
                    ROOT
                )
            ),

        "completion_manifest_sha256":
            tau10_completion_sha,

        "best_AA_epoch":
            int(
                tau10_best_aa_epoch
            ),

        "best_AA_metrics":
            tau10_best_aa,

        "best_OA_epoch":
            int(
                tau10_best_oa_epoch
            ),

        "best_OA_metrics":
            tau10_best_oa,
    },

    "pingan_tuning": {
        "candidate_tau_0p5_best_AA":
            tau05_best_aa,

        "candidate_tau_1p0_best_AA":
            tau10_best_aa,

        "independent_selected_tau":
            independent_selected_tau,

        "authenticated_selected_tau":
            float(
                manifest_selected_tau
            ),

        "selection_manifest_relative_path":
            selection_rel,

        "selection_manifest_sha256":
            selection_sha,

        "selection_manifest":
            selection_disk,
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(
                torch.backends.cudnn.version()
            ),

        "gpu0":
            torch.cuda.get_device_name(
                0
            ),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "full_delta": {
        "base_project_file_count":
            449,

        "delta_file_count":
            len(
                delta_records
            ),

        "resulting_project_file_count":
            len(
                project_files_after
            ),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "pingan_tau_selection_complete":
            True,
    },
}


manifest_bytes = json.dumps(
    persistence_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 22. NO-OVERWRITE PERSISTENCE
# ==================================================================================================

print()
print("-" * 174)
print("[11] IMMEDIATE 30/36 GOOGLE-DRIVE PERSISTENCE")
print("-" * 174)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if path.exists():
        raise RuntimeError(
            "STOP: 30/36 persistence artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 23. CREATE 15-FILE CUMULATIVE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(
        PERSISTENCE_ZIP
    )
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_STORED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record[
            "relative_path"
        ]

        source = (
            ROOT
            /
            relative_path
        )


        if (
            sha256_file(
                source
            )
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: delta artifact changed before archive:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=(
                "LTLC_DELTA/"
                +
                relative_path
            ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 24. VERIFY DRIVE ZIP
# ==================================================================================================

record_map = {
    (
        "LTLC_DELTA/"
        +
        record[
            "relative_path"
        ]
    ):
        record

    for record in delta_records
}


verified = 0


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:

    expected_members = (
        set(
            record_map.keys()
        )
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )


    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:
        raise RuntimeError(
            "STOP: persistence ZIP member set mismatch."
        )


    if (
        zf.read(
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        )
        !=
        manifest_bytes
    ):
        raise RuntimeError(
            "STOP: archived persistence manifest differs."
        )


    for archive_name, record in record_map.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(
                stream
            )


        if (
            archived_sha
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified += 1


if (
    verified
    !=
    EXPECTED_POST_CUMULATIVE_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: expected 15/15 archived delta verification."
    )


persistence_zip_sha = sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 25. SHA SIDECAR
# ==================================================================================================

sidecar_text = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 26. FINAL GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTED CELL 7B FINAL GATE")
print("=" * 174)

print(
    "Authorized job                           : Pingan / LA / tau=1.0 / seed42"
)

print(
    "Epochs                                   : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction      : PASS"
)

print(
    "Pingan tau=0.5 candidate                 : COMPLETE"
)

print(
    "Pingan tau=1.0 candidate                 : COMPLETE"
)

print(
    "Pingan frozen tau selection              : COMPLETE"
)

print(
    "Independent selected tau                 :",
    independent_selected_tau
)

print(
    "Authenticated selected tau               :",
    manifest_selected_tau
)

print(
    "Selection-rule agreement                 : PASS"
)

print(
    "Experiment-B progress                    : 30/36"
)

print(
    "LA-loss progress                         : 2/8"
)

print(
    "Original migration base                  : 449 files"
)

print(
    "Current project                          : 464 files"
)

print(
    "Cumulative post-migration delta          : 15 files PASS"
)

print(
    "Drive delta SHA verification             : 15/15 PASS"
)

print(
    "Scientific protocol changed              : False"
)

print(
    "Mixed precision                          : False"
)

print(
    "Multi-GPU                                : False"
)

print(
    "Scientific device                        : cuda:0 ONLY"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "tau=1.0 BEST-AA EPOCH:"
)

print(
    tau10_best_aa_epoch
)

print(
    "tau=1.0 BEST-AA METRICS:"
)

print(
    tau10_best_aa
)

print()
print(
    "PINGAN SELECTED TAU:"
)

print(
    manifest_selected_tau
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(
        training_elapsed,
        2,
    )
)

print()
print(
    "30/36 PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL CORRECTED COLAB CELL 7B PASS: True"
)

print()
print(
    "STATUS: "
    "PINGAN_LA_TAU1P0_SEED42_COMPLETE__"
    "PINGAN_TAU_SELECTION_FROZEN__"
    "30_OF_36__"
    "15_FILE_CUMULATIVE_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_NEXT_GPU_JOB"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# COLAB CELL 8
#
# PINGAN LA-LOSS tau=1.0 / SEED123
# + IMMEDIATE VERIFIED 31/36 GOOGLE-DRIVE PERSISTENCE
#
# FROZEN CONFIGURATION:
#   Dataset : Pingan
#   Method  : LA_Loss
#   tau     : 1.0   <-- already selected/frozen from seed42 tuning
#   seed    : 123
#   epochs  : 100
#
# NO:
#   - retuning
#   - tau reselection
#   - mixed precision
#   - torch.compile
#   - multi-GPU
#   - protocol changes
#   - official-test access
#
# EXPECTED AFTER SUCCESS:
#   Experiment-B : 31/36
#   LA-loss      : 3/8
#   Project      : 471 files
#   Cumulative delta from 449-file migration base : 22 files
#
# NEXT AFTER EXTERNAL SAVE ACKNOWLEDGEMENT:
#   Pingan / LA-loss / tau=1.0 / seed3407
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 8")
print("PINGAN LA-LOSS tau=1.0 / SEED123 + IMMEDIATE 31/36 DRIVE PERSISTENCE")
print("=" * 174)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

ADAPTER_DIR = (
    DRIVE_ROOT
    / "migrations"
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 2. AUTHENTICATED 30/36 PARENT
# ==================================================================================================

PARENT_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "30of36_Post_Pingan_LA_Tuning_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PARENT_ZIP = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}.zip"
)

PARENT_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}__MANIFEST.json"
)


EXPECTED_PARENT_ZIP_SHA256 = (
    "ef8170cd501bdff44b7448f8e88cadaa"
    "464bff37223a0898053b3ab9fff69c0f"
)

EXPECTED_PARENT_MANIFEST_SHA256 = (
    "0debe3b433e7c4748ea0c2ec17482655"
    "6d0837aa6ecfc1fea7d683fdf6d9ef41"
)


# ==================================================================================================
# 3. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "4a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)


BASE_PROJECT_FILE_COUNT = 449
PRE_JOB_PROJECT_FILE_COUNT = 464
POST_JOB_PROJECT_FILE_COUNT = 471

PRE_JOB_DELTA_COUNT = 15
POST_JOB_DELTA_COUNT = 22


INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 4. NEW 31/36 PERSISTENCE PATHS
# ==================================================================================================

PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "31of36_Post_Pingan_LA_tau1p0_Seed123_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}.zip"
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__MANIFEST.json"
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__SHA256.txt"
)


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for chunk in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(chunk)

    return h.hexdigest()


def stream_sha256(
    stream,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    while True:

        block = stream.read(chunk_size)

        if not block:
            break

        h.update(block)

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(
        tmp,
        path,
    )


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. T4 SCIENTIFIC RUNTIME GATE
# ==================================================================================================

print()
print("-" * 174)
print("[1] T4 SCIENTIFIC RUNTIME GATE")
print("-" * 174)


if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError(
        "STOP: PyTorch runtime changed."
    )

if torch.version.cuda != "12.8":
    raise RuntimeError(
        "STOP: CUDA build changed."
    )

if not torch.cuda.is_available():
    raise RuntimeError(
        "STOP: CUDA unavailable."
    )

if int(
    torch.backends.cudnn.version()
) != 91900:
    raise RuntimeError(
        "STOP: cuDNN changed."
    )

if torch.cuda.device_count() < 1:
    raise RuntimeError(
        "STOP: no GPU visible."
    )


gpu0 = torch.cuda.get_device_name(
    0
)


if "T4" not in gpu0:
    raise RuntimeError(
        "STOP: GPU0 is not Tesla T4."
    )

if str(DEVICE) != "cuda:0":
    raise RuntimeError(
        "STOP: scientific DEVICE is not cuda:0."
    )

if not T4_GATE_PATH.is_file():
    raise FileNotFoundError(
        "STOP: T4 runtime gate missing."
    )

if (
    sha256_file(T4_GATE_PATH)
    !=
    EXPECTED_T4_GATE_SHA256
):
    raise RuntimeError(
        "STOP: T4 runtime gate SHA changed."
    )


print("PyTorch           : 2.11.0+cu128 PASS")
print("CUDA              : 12.8 PASS")
print("cuDNN             : 91900 PASS")
print("GPU0              :", gpu0, "PASS")
print("Scientific DEVICE : cuda:0 ONLY PASS")
print("T4 runtime gate   : PASS")


# ==================================================================================================
# 7. EXTERNALLY SAVED 30/36 PARENT
# ==================================================================================================

print()
print("-" * 174)
print("[2] EXTERNALLY SAVED 30/36 PARENT")
print("-" * 174)


if not PARENT_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: saved 30/36 persistence ZIP missing."
    )

if not PARENT_MANIFEST.is_file():
    raise FileNotFoundError(
        "STOP: saved 30/36 persistence manifest missing."
    )


parent_zip_sha = sha256_file(
    PARENT_ZIP
)

parent_manifest_sha = sha256_file(
    PARENT_MANIFEST
)


if (
    parent_zip_sha
    !=
    EXPECTED_PARENT_ZIP_SHA256
):
    raise RuntimeError(
        "STOP: saved 30/36 ZIP SHA changed."
    )

if (
    parent_manifest_sha
    !=
    EXPECTED_PARENT_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: saved 30/36 manifest SHA changed."
    )


print("30/36 ZIP      : PASS")
print("30/36 manifest : PASS")
print("Save boundary  : CONFIRMED")


# ==================================================================================================
# 8. PRE-JOB SCIENTIFIC BOUNDARY
# ==================================================================================================

print()
print("-" * 174)
print("[3] PRE-JOB SCIENTIFIC BOUNDARY")
print("-" * 174)


pre_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_pre = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 2,
}


for method, expected in expected_pre.items():

    observed = pre_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: pre-job {method} count changed."
        )


if sum(
    pre_ledger.values()
) != 30:
    raise RuntimeError(
        "STOP: experiment is not at 30/36."
    )


project_files_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(project_files_before)
    !=
    PRE_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 464 files before seed123."
    )


selection_path = la_selection_path(
    "Pingan"
)


if not selection_path.is_file():
    raise FileNotFoundError(
        "STOP: frozen Pingan tau-selection manifest missing."
    )


selection_sha_before = sha256_file(
    selection_path
)


if (
    selection_sha_before
    !=
    EXPECTED_PINGAN_SELECTION_SHA256
):
    raise RuntimeError(
        "STOP: Pingan tau-selection SHA changed."
    )


with selection_path.open(
    "r",
    encoding="utf-8",
) as f:

    selection_disk = json.load(f)


if float(
    selection_disk.get(
        "selected_tau"
    )
) != 1.0:
    raise RuntimeError(
        "STOP: Pingan selected tau is not frozen at 1.0."
    )


seed123_completion = la_completion_path(
    "Pingan",
    1.0,
    123,
)


if seed123_completion.exists():
    raise RuntimeError(
        "STOP: Pingan tau=1.0 seed123 already exists."
    )


print()
print("Experiment-B          : 30/36 PASS")
print("LA-loss               : 2/8 PASS")
print("Project files         : 464 PASS")
print("Frozen selected tau   : 1.0 PASS")
print("Selection SHA         : PASS")
print("Seed123 target absent : PASS")


# ==================================================================================================
# 9. FROZEN SCIENTIFIC SOURCE GATE
# ==================================================================================================

print()
print("-" * 174)
print("[4] FROZEN SCIENTIFIC SOURCE IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    actual_sha = sha256_file(
        path
    )

    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL",
    )

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP: frozen SHA mismatch: {label}"
        )


# ==================================================================================================
# 10. AUTHENTICATED 449-FILE MIGRATION BASE
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATED 28/36 MIGRATION BASE")
print("-" * 174)


if not LOCAL_MIGRATION_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: local authenticated migration ZIP missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


migration_manifest_sha = hashlib.sha256(
    migration_manifest_bytes
).hexdigest()


if (
    migration_manifest_sha
    !=
    EXPECTED_MIGRATION_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: migration manifest identity changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    migration_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(
    base_records
) != BASE_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: migration baseline is not 449 files."
    )


base_paths = {
    str(
        PurePosixPath(
            record[
                "relative_path"
            ]
        )
    )
    for record in base_records
}


before_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_before
}


pre_delta_paths = sorted(
    before_paths
    -
    base_paths
)


if (
    len(pre_delta_paths)
    !=
    PRE_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: current 30/36 cumulative delta is not 15 files."
    )


print("Migration manifest      : PASS")
print("Migration base files    : 449")
print("Current cumulative delta: 15 files PASS")


# ==================================================================================================
# 11. START AUTHORIZED GPU JOB
# ==================================================================================================

print()
print("=" * 174)
print("STARTING AUTHORIZED GPU JOB")
print("Pingan / LA-loss / tau=1.0 / seed123")
print("=" * 174)


training_start = time.time()


completion_result = (
    run_experiment_b_la_loss_job(
        "Pingan",
        1.0,
        123,
    )
)


training_elapsed = (
    time.time()
    -
    training_start
)


print()
print(
    "Authorized runner returned successfully."
)

print(
    "Runner elapsed seconds:",
    round(
        training_elapsed,
        2,
    )
)


# ==================================================================================================
# 12. COMPLETION AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[6] SEED123 COMPLETION AUDIT")
print("-" * 174)


if not seed123_completion.is_file():
    raise FileNotFoundError(
        "STOP: seed123 completion manifest missing."
    )


with seed123_completion.open(
    "r",
    encoding="utf-8",
) as f:

    completion_disk = json.load(f)


identity_checks = {
    "status":
        completion_disk.get(
            "status"
        ) == "COMPLETE",

    "dataset":
        completion_disk.get(
            "dataset"
        ) == "Pingan",

    "method":
        completion_disk.get(
            "method"
        ) == "LA_Loss",

    "tau":
        float(
            completion_disk.get(
                "tau"
            )
        ) == 1.0,

    "seed":
        int(
            completion_disk.get(
                "seed"
            )
        ) == 123,

    "epochs":
        int(
            completion_disk.get(
                "epochs_completed"
            )
        ) == 100,
}


for label, passed in identity_checks.items():

    print(
        f"{label:14s}:",
        "PASS"
        if passed
        else "FAIL",
    )


if not all(
    identity_checks.values()
):
    raise RuntimeError(
        "STOP: seed123 completion identity failed."
    )


safety = completion_disk[
    "safety"
]


if not (
    safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    safety.get(
        "official_test_values_accessed"
    ) is False
    and
    safety.get(
        "test_dataset_created"
    ) is False
    and
    safety.get(
        "test_dataloader_created"
    ) is False
    and
    safety.get(
        "test_inference_performed"
    ) is False
):
    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


completion_files = completion_disk[
    "files"
]


if len(
    completion_files
) != 6:
    raise RuntimeError(
        "STOP: seed123 completion must record six run artifacts."
    )


for role, record in completion_files.items():

    path = (
        ROOT
        /
        record[
            "path"
        ]
    )

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: seed123 run artifact missing: {role}"
        )

    if (
        sha256_file(
            path
        )
        !=
        record[
            "sha256"
        ]
    ):
        raise RuntimeError(
            f"STOP: seed123 artifact SHA mismatch: {role}"
        )

    print(
        f"{role:24s}: PASS"
    )


best_aa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

best_aa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)

best_oa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_epoch"
    ]
)

best_oa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_metrics"
    ]
)


print()
print(
    "Seed123 Best-AA epoch:",
    best_aa_epoch
)

print(
    "Seed123 Best-AA metrics:"
)

print(
    best_aa_metrics
)

print()
print(
    "Seed123 Best-OA epoch:",
    best_oa_epoch
)

print(
    "Seed123 Best-OA metrics:"
)

print(
    best_oa_metrics
)


# ==================================================================================================
# 13. VERIFY FROZEN TAU SELECTION WAS NOT MODIFIED
# ==================================================================================================

print()
print("-" * 174)
print("[7] FROZEN PINGAN TAU-SELECTION IMMUTABILITY")
print("-" * 174)


selection_sha_after = sha256_file(
    selection_path
)


if (
    selection_sha_after
    !=
    selection_sha_before
):
    raise RuntimeError(
        "STOP: seed123 job modified the frozen Pingan selection manifest."
    )


print(
    "Selected tau               : 1.0"
)

print(
    "Selection SHA before/after : IDENTICAL PASS"
)

print(
    "Retuning performed         : False"
)


# ==================================================================================================
# 14. POST-JOB LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[8] POST-JOB SCIENTIFIC LEDGER")
print("-" * 174)


post_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_post = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 3,
}


for method, expected in expected_post.items():

    observed = post_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: post-job {method} count changed."
        )


if sum(
    post_ledger.values()
) != 31:
    raise RuntimeError(
        "STOP: Experiment-B state is not 31/36."
    )


print()
print(
    "Experiment-B progress : 31/36 PASS"
)

print(
    "LA-loss progress       : 3/8 PASS"
)


# ==================================================================================================
# 15. PROJECT FOOTPRINT
# ==================================================================================================

print()
print("-" * 174)
print("[9] 31/36 PROJECT-FOOTPRINT AUDIT")
print("-" * 174)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if (
    len(project_files_after)
    !=
    POST_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 471 project files after seed123.\n"
        f"Observed: {len(project_files_after)}"
    )


after_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_after
}


new_since_30 = sorted(
    after_paths
    -
    before_paths
)


if len(
    new_since_30
) != 7:
    raise RuntimeError(
        "STOP: seed123 must add exactly seven project files."
    )


expected_seed123_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/pingan/tau1p0/seed123/"
)


if not all(
    path.startswith(
        expected_seed123_prefix
    )
    for path in new_since_30
):
    raise RuntimeError(
        "STOP: seed123 added an unauthorized project path."
    )


print(
    "New seed123 files:",
    len(
        new_since_30
    ),
    "PASS",
)


for path in new_since_30:
    print(
        " +",
        path
    )


cumulative_delta_paths = sorted(
    after_paths
    -
    base_paths
)


if (
    len(
        cumulative_delta_paths
    )
    !=
    POST_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: expected 22-file cumulative delta at 31/36."
    )


print()
print(
    "Migration base files   : 449"
)

print(
    "Current project files  : 471"
)

print(
    "Cumulative delta files : 22 PASS"
)


# ==================================================================================================
# 16. HASH ONLY THE 22 DELTA FILES
# ==================================================================================================

delta_records = []


for relative_path in cumulative_delta_paths:

    path = (
        ROOT
        /
        relative_path
    )

    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(
                    path.stat().st_size
                ),

            "sha256":
                sha256_file(
                    path
                ),
        }
    )


completion_sha = sha256_file(
    seed123_completion
)


# ==================================================================================================
# 17. BUILD 31/36 PERSISTENCE MANIFEST
# ==================================================================================================

persistence_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent_chain": {
        "previous_progress":
            "30/36",

        "previous_persistence_zip_sha256":
            parent_zip_sha,

        "previous_persistence_manifest_sha256":
            parent_manifest_sha,

        "base_project_file_count":
            449,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            30,

        "after":
            31,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            3,
    },

    "completed_job": {
        "dataset":
            "Pingan",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            123,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                seed123_completion.relative_to(
                    ROOT
                )
            ),

        "completion_manifest_sha256":
            completion_sha,

        "best_AA_epoch":
            int(
                best_aa_epoch
            ),

        "best_AA_metrics":
            best_aa_metrics,

        "best_OA_epoch":
            int(
                best_oa_epoch
            ),

        "best_OA_metrics":
            best_oa_metrics,
    },

    "frozen_pingan_selection": {
        "selected_tau":
            1.0,

        "selection_manifest_relative_path":
            str(
                selection_path.relative_to(
                    ROOT
                )
            ),

        "selection_manifest_sha256":
            selection_sha_after,

        "unchanged_during_seed123":
            True,
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(
                torch.backends.cudnn.version()
            ),

        "gpu0":
            torch.cuda.get_device_name(
                0
            ),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "full_delta": {
        "base_project_file_count":
            449,

        "delta_file_count":
            len(
                delta_records
            ),

        "resulting_project_file_count":
            len(
                project_files_after
            ),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "retuning_performed":
            False,
    },

    "next_authorized_job_after_external_save_acknowledgement": {
        "dataset":
            "Pingan",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            3407,
    },
}


manifest_bytes = json.dumps(
    persistence_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 18. IMMEDIATE GOOGLE-DRIVE PERSISTENCE
# ==================================================================================================

print()
print("-" * 174)
print("[10] IMMEDIATE 31/36 GOOGLE-DRIVE PERSISTENCE")
print("-" * 174)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if path.exists():
        raise RuntimeError(
            "STOP: 31/36 persistence artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 19. CREATE 22-FILE CUMULATIVE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(
        PERSISTENCE_ZIP
    )
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_STORED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record[
            "relative_path"
        ]

        source = (
            ROOT
            /
            relative_path
        )


        if (
            sha256_file(
                source
            )
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: delta artifact changed before archive:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=(
                "LTLC_DELTA/"
                +
                relative_path
            ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 20. VERIFY DRIVE ZIP
# ==================================================================================================

record_map = {
    (
        "LTLC_DELTA/"
        +
        record[
            "relative_path"
        ]
    ):
        record

    for record in delta_records
}


verified = 0


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:

    expected_members = (
        set(
            record_map.keys()
        )
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )


    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:
        raise RuntimeError(
            "STOP: 31/36 persistence ZIP member-set mismatch."
        )


    if (
        zf.read(
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        )
        !=
        manifest_bytes
    ):
        raise RuntimeError(
            "STOP: archived 31/36 manifest differs."
        )


    for archive_name, record in record_map.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(
                stream
            )


        if (
            archived_sha
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified += 1


if verified != POST_JOB_DELTA_COUNT:
    raise RuntimeError(
        "STOP: expected 22/22 delta verification."
    )


persistence_zip_sha = sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 21. SHA SIDECAR
# ==================================================================================================

sidecar_text = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 22. FINAL GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 8 FINAL GATE")
print("=" * 174)

print(
    "Authorized job                           : Pingan / LA / tau=1.0 / seed123"
)

print(
    "Epochs                                   : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction      : PASS"
)

print(
    "Frozen selected tau                      : 1.0"
)

print(
    "Tau selection manifest unchanged         : PASS"
)

print(
    "Retuning performed                       : False"
)

print(
    "Experiment-B progress                    : 31/36"
)

print(
    "LA-loss progress                         : 3/8"
)

print(
    "Original migration base                  : 449 files"
)

print(
    "Current project                          : 471 files"
)

print(
    "Cumulative post-migration delta          : 22 files PASS"
)

print(
    "Drive delta SHA verification             : 22/22 PASS"
)

print(
    "Scientific protocol changed              : False"
)

print(
    "Mixed precision                          : False"
)

print(
    "Multi-GPU                                : False"
)

print(
    "Scientific device                        : cuda:0 ONLY"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "SEED123 BEST-AA EPOCH:"
)

print(
    best_aa_epoch
)

print(
    "SEED123 BEST-AA METRICS:"
)

print(
    best_aa_metrics
)

print()
print(
    "SEED123 BEST-OA EPOCH:"
)

print(
    best_oa_epoch
)

print(
    "SEED123 BEST-OA METRICS:"
)

print(
    best_oa_metrics
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(
        training_elapsed,
        2,
    )
)

print()
print(
    "31/36 PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL COLAB CELL 8 PASS: True"
)

print()
print(
    "STATUS: "
    "PINGAN_LA_TAU1P0_SEED123_COMPLETE__"
    "31_OF_36__"
    "22_FILE_CUMULATIVE_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_SEED3407"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# COLAB CELL 9
#
# PINGAN LA-LOSS tau=1.0 / SEED3407
# + IMMEDIATE VERIFIED 32/36 GOOGLE-DRIVE PERSISTENCE
#
# FROZEN CONFIGURATION:
#   Dataset : Pingan
#   Method  : LA_Loss
#   tau     : 1.0
#   seed    : 3407
#   epochs  : 100
#
# EXPECTED AFTER SUCCESS:
#   Experiment-B : 32/36
#   LA-loss      : 4/8
#   Pingan LA    : COMPLETE
#   Project      : 478 files
#   Cumulative delta from 449-file migration base : 29 files
#
# NEXT AFTER EXTERNAL SAVE ACKNOWLEDGEMENT:
#   Qingyun / LA-loss / tau=0.5 / seed42
#
# SCIENTIFIC PROTOCOL REMAINS UNCHANGED.
# OFFICIAL TEST REMAINS SEALED.
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 9")
print("PINGAN LA-LOSS tau=1.0 / SEED3407 + IMMEDIATE 32/36 DRIVE PERSISTENCE")
print("=" * 174)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

ADAPTER_DIR = (
    DRIVE_ROOT
    / "migrations"
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 2. AUTHENTICATED 31/36 PARENT
# ==================================================================================================

PARENT_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "31of36_Post_Pingan_LA_tau1p0_Seed123_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PARENT_ZIP = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}.zip"
)

PARENT_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}__MANIFEST.json"
)


EXPECTED_PARENT_ZIP_SHA256 = (
    "9109a116c9d1a6dcbc83164c17621d21"
    "40e0e59dc483a5fb11f3b79df74e9922"
)

EXPECTED_PARENT_MANIFEST_SHA256 = (
    "a40d17da4e1527c3dfc3daade6b62de0"
    "36120f9a87651ad198468c0fb9606700"
)


# ==================================================================================================
# 3. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "4a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)


BASE_PROJECT_FILE_COUNT = 449
PRE_JOB_PROJECT_FILE_COUNT = 471
POST_JOB_PROJECT_FILE_COUNT = 478

PRE_JOB_DELTA_COUNT = 22
POST_JOB_DELTA_COUNT = 29


INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 4. NEW 32/36 PERSISTENCE PATHS
# ==================================================================================================

PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "32of36_Post_Pingan_LA_tau1p0_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}.zip"
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__MANIFEST.json"
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__SHA256.txt"
)


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for chunk in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(chunk)

    return h.hexdigest()


def stream_sha256(
    stream,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    while True:

        block = stream.read(chunk_size)

        if not block:
            break

        h.update(block)

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(
        tmp,
        path,
    )


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. T4 SCIENTIFIC RUNTIME GATE
# ==================================================================================================

print()
print("-" * 174)
print("[1] T4 SCIENTIFIC RUNTIME GATE")
print("-" * 174)


if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError(
        "STOP: PyTorch runtime changed."
    )

if torch.version.cuda != "12.8":
    raise RuntimeError(
        "STOP: CUDA build changed."
    )

if not torch.cuda.is_available():
    raise RuntimeError(
        "STOP: CUDA unavailable."
    )

if int(
    torch.backends.cudnn.version()
) != 91900:
    raise RuntimeError(
        "STOP: cuDNN changed."
    )

if torch.cuda.device_count() < 1:
    raise RuntimeError(
        "STOP: no CUDA GPU visible."
    )


gpu0 = torch.cuda.get_device_name(
    0
)


if "T4" not in gpu0:
    raise RuntimeError(
        "STOP: GPU0 is not Tesla T4."
    )

if str(DEVICE) != "cuda:0":
    raise RuntimeError(
        "STOP: scientific DEVICE is not cuda:0."
    )

if not T4_GATE_PATH.is_file():
    raise FileNotFoundError(
        "STOP: T4 runtime gate missing."
    )

if (
    sha256_file(
        T4_GATE_PATH
    )
    !=
    EXPECTED_T4_GATE_SHA256
):
    raise RuntimeError(
        "STOP: T4 runtime gate SHA changed."
    )


print("PyTorch           : 2.11.0+cu128 PASS")
print("CUDA              : 12.8 PASS")
print("cuDNN             : 91900 PASS")
print("GPU0              :", gpu0, "PASS")
print("Scientific DEVICE : cuda:0 ONLY PASS")
print("T4 runtime gate   : PASS")


# ==================================================================================================
# 7. EXTERNALLY SAVED 31/36 PARENT
# ==================================================================================================

print()
print("-" * 174)
print("[2] EXTERNALLY SAVED 31/36 PARENT")
print("-" * 174)


if not PARENT_ZIP.is_file():

    raise FileNotFoundError(
        "STOP: saved 31/36 persistence ZIP missing."
    )


if not PARENT_MANIFEST.is_file():

    raise FileNotFoundError(
        "STOP: saved 31/36 persistence manifest missing."
    )


parent_zip_sha = sha256_file(
    PARENT_ZIP
)

parent_manifest_sha = sha256_file(
    PARENT_MANIFEST
)


if (
    parent_zip_sha
    !=
    EXPECTED_PARENT_ZIP_SHA256
):

    raise RuntimeError(
        "STOP: saved 31/36 ZIP SHA changed."
    )


if (
    parent_manifest_sha
    !=
    EXPECTED_PARENT_MANIFEST_SHA256
):

    raise RuntimeError(
        "STOP: saved 31/36 manifest SHA changed."
    )


print("31/36 ZIP      : PASS")
print("31/36 manifest : PASS")
print("Save boundary  : CONFIRMED")


# ==================================================================================================
# 8. PRE-JOB SCIENTIFIC BOUNDARY
# ==================================================================================================

print()
print("-" * 174)
print("[3] PRE-JOB SCIENTIFIC BOUNDARY")
print("-" * 174)


pre_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_pre = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 3,
}


for method, expected in expected_pre.items():

    observed = pre_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:

        raise RuntimeError(
            f"STOP: pre-job {method} count changed."
        )


if sum(
    pre_ledger.values()
) != 31:

    raise RuntimeError(
        "STOP: experiment is not at 31/36."
    )


project_files_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(
        project_files_before
    )
    !=
    PRE_JOB_PROJECT_FILE_COUNT
):

    raise RuntimeError(
        "STOP: expected 471 files before seed3407."
    )


selection_path = (
    la_selection_path(
        "Pingan"
    )
)


if not selection_path.is_file():

    raise FileNotFoundError(
        "STOP: frozen Pingan tau-selection manifest missing."
    )


selection_sha_before = sha256_file(
    selection_path
)


if (
    selection_sha_before
    !=
    EXPECTED_PINGAN_SELECTION_SHA256
):

    raise RuntimeError(
        "STOP: Pingan tau-selection SHA changed."
    )


with selection_path.open(
    "r",
    encoding="utf-8",
) as f:

    selection_disk = json.load(
        f
    )


if float(
    selection_disk.get(
        "selected_tau"
    )
) != 1.0:

    raise RuntimeError(
        "STOP: Pingan selected tau is no longer 1.0."
    )


seed3407_completion = (
    la_completion_path(
        "Pingan",
        1.0,
        3407,
    )
)


if seed3407_completion.exists():

    raise RuntimeError(
        "STOP: Pingan tau=1.0 seed3407 already exists."
    )


print()
print("Experiment-B           : 31/36 PASS")
print("LA-loss                : 3/8 PASS")
print("Project files          : 471 PASS")
print("Frozen selected tau    : 1.0 PASS")
print("Selection manifest SHA : PASS")
print("Seed3407 target absent : PASS")


# ==================================================================================================
# 9. FROZEN SCIENTIFIC SOURCE GATE
# ==================================================================================================

print()
print("-" * 174)
print("[4] FROZEN SCIENTIFIC SOURCE IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    actual_sha = sha256_file(
        path
    )

    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL",
    )

    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: frozen SHA mismatch: {label}"
        )


# ==================================================================================================
# 10. AUTHENTICATED 449-FILE MIGRATION BASE
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATED 28/36 MIGRATION BASE")
print("-" * 174)


if not LOCAL_MIGRATION_ZIP.is_file():

    raise FileNotFoundError(
        "STOP: local authenticated migration ZIP missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


migration_manifest_sha = hashlib.sha256(
    migration_manifest_bytes
).hexdigest()


if (
    migration_manifest_sha
    !=
    EXPECTED_MIGRATION_MANIFEST_SHA256
):

    raise RuntimeError(
        "STOP: migration manifest identity changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    migration_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(
    base_records
) != BASE_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: migration baseline is not 449 files."
    )


base_paths = {
    str(
        PurePosixPath(
            record[
                "relative_path"
            ]
        )
    )
    for record in base_records
}


before_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_before
}


pre_delta_paths = sorted(
    before_paths
    -
    base_paths
)


if (
    len(
        pre_delta_paths
    )
    !=
    PRE_JOB_DELTA_COUNT
):

    raise RuntimeError(
        "STOP: current 31/36 cumulative delta is not 22 files."
    )


print("Migration manifest       : PASS")
print("Migration base files     : 449")
print("Current cumulative delta : 22 files PASS")


# ==================================================================================================
# 11. START AUTHORIZED GPU JOB
# ==================================================================================================

print()
print("=" * 174)
print("STARTING AUTHORIZED GPU JOB")
print("Pingan / LA-loss / tau=1.0 / seed3407")
print("=" * 174)


training_start = time.time()


completion_result = (
    run_experiment_b_la_loss_job(
        "Pingan",
        1.0,
        3407,
    )
)


training_elapsed = (
    time.time()
    -
    training_start
)


print()
print(
    "Authorized runner returned successfully."
)

print(
    "Runner elapsed seconds:",
    round(
        training_elapsed,
        2,
    )
)


# ==================================================================================================
# 12. COMPLETION AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[6] SEED3407 COMPLETION AUDIT")
print("-" * 174)


if not seed3407_completion.is_file():

    raise FileNotFoundError(
        "STOP: seed3407 completion manifest missing."
    )


with seed3407_completion.open(
    "r",
    encoding="utf-8",
) as f:

    completion_disk = json.load(
        f
    )


identity_checks = {
    "status":
        completion_disk.get(
            "status"
        ) == "COMPLETE",

    "dataset":
        completion_disk.get(
            "dataset"
        ) == "Pingan",

    "method":
        completion_disk.get(
            "method"
        ) == "LA_Loss",

    "tau":
        float(
            completion_disk.get(
                "tau"
            )
        ) == 1.0,

    "seed":
        int(
            completion_disk.get(
                "seed"
            )
        ) == 3407,

    "epochs":
        int(
            completion_disk.get(
                "epochs_completed"
            )
        ) == 100,
}


for label, passed in identity_checks.items():

    print(
        f"{label:14s}:",
        "PASS"
        if passed
        else "FAIL",
    )


if not all(
    identity_checks.values()
):

    raise RuntimeError(
        "STOP: seed3407 completion identity failed."
    )


safety = completion_disk[
    "safety"
]


if not (
    safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    safety.get(
        "official_test_values_accessed"
    ) is False
    and
    safety.get(
        "test_dataset_created"
    ) is False
    and
    safety.get(
        "test_dataloader_created"
    ) is False
    and
    safety.get(
        "test_inference_performed"
    ) is False
):

    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


completion_files = completion_disk[
    "files"
]


if len(
    completion_files
) != 6:

    raise RuntimeError(
        "STOP: seed3407 completion must record six run artifacts."
    )


for role, record in completion_files.items():

    path = (
        ROOT
        /
        record[
            "path"
        ]
    )


    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: seed3407 artifact missing: {role}"
        )


    if (
        sha256_file(
            path
        )
        !=
        record[
            "sha256"
        ]
    ):

        raise RuntimeError(
            f"STOP: seed3407 artifact SHA mismatch: {role}"
        )


    print(
        f"{role:24s}: PASS"
    )


best_aa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

best_aa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)

best_oa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_epoch"
    ]
)

best_oa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_metrics"
    ]
)


print()
print(
    "Seed3407 Best-AA epoch:",
    best_aa_epoch
)

print(
    "Seed3407 Best-AA metrics:"
)

print(
    best_aa_metrics
)

print()
print(
    "Seed3407 Best-OA epoch:",
    best_oa_epoch
)

print(
    "Seed3407 Best-OA metrics:"
)

print(
    best_oa_metrics
)


# ==================================================================================================
# 13. FROZEN PINGAN TAU-SELECTION IMMUTABILITY
# ==================================================================================================

print()
print("-" * 174)
print("[7] FROZEN PINGAN TAU-SELECTION IMMUTABILITY")
print("-" * 174)


selection_sha_after = sha256_file(
    selection_path
)


if (
    selection_sha_after
    !=
    selection_sha_before
):

    raise RuntimeError(
        "STOP: seed3407 modified the frozen Pingan selection manifest."
    )


print(
    "Selected tau               : 1.0"
)

print(
    "Selection SHA before/after : IDENTICAL PASS"
)

print(
    "Retuning performed         : False"
)


# ==================================================================================================
# 14. POST-JOB SCIENTIFIC LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[8] POST-JOB SCIENTIFIC LEDGER")
print("-" * 174)


post_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_post = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 4,
}


for method, expected in expected_post.items():

    observed = post_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:

        raise RuntimeError(
            f"STOP: post-job {method} count changed."
        )


if sum(
    post_ledger.values()
) != 32:

    raise RuntimeError(
        "STOP: Experiment-B state is not 32/36."
    )


print()
print(
    "Experiment-B progress : 32/36 PASS"
)

print(
    "LA-loss progress       : 4/8 PASS"
)

print(
    "Pingan LA-loss         : COMPLETE PASS"
)


# ==================================================================================================
# 15. PROJECT FOOTPRINT
# ==================================================================================================

print()
print("-" * 174)
print("[9] 32/36 PROJECT-FOOTPRINT AUDIT")
print("-" * 174)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if (
    len(
        project_files_after
    )
    !=
    POST_JOB_PROJECT_FILE_COUNT
):

    raise RuntimeError(
        "STOP: expected 478 project files after seed3407.\n"
        f"Observed: {len(project_files_after)}"
    )


after_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_after
}


new_since_31 = sorted(
    after_paths
    -
    before_paths
)


if len(
    new_since_31
) != 7:

    raise RuntimeError(
        "STOP: seed3407 must add exactly seven project files."
    )


expected_seed3407_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/pingan/tau1p0/seed3407/"
)


if not all(
    path.startswith(
        expected_seed3407_prefix
    )
    for path in new_since_31
):

    raise RuntimeError(
        "STOP: seed3407 added an unauthorized project path."
    )


print(
    "New seed3407 files:",
    len(
        new_since_31
    ),
    "PASS",
)


for path in new_since_31:

    print(
        " +",
        path
    )


cumulative_delta_paths = sorted(
    after_paths
    -
    base_paths
)


if (
    len(
        cumulative_delta_paths
    )
    !=
    POST_JOB_DELTA_COUNT
):

    raise RuntimeError(
        "STOP: expected 29-file cumulative delta at 32/36."
    )


print()
print(
    "Migration base files   : 449"
)

print(
    "Current project files  : 478"
)

print(
    "Cumulative delta files : 29 PASS"
)


# ==================================================================================================
# 16. HASH ONLY THE 29 DELTA FILES
# ==================================================================================================

delta_records = []


for relative_path in cumulative_delta_paths:

    path = (
        ROOT
        /
        relative_path
    )

    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(
                    path.stat().st_size
                ),

            "sha256":
                sha256_file(
                    path
                ),
        }
    )


completion_sha = sha256_file(
    seed3407_completion
)


# ==================================================================================================
# 17. BUILD 32/36 PERSISTENCE MANIFEST
# ==================================================================================================

persistence_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent_chain": {
        "previous_progress":
            "31/36",

        "previous_persistence_zip_sha256":
            parent_zip_sha,

        "previous_persistence_manifest_sha256":
            parent_manifest_sha,

        "base_project_file_count":
            449,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            31,

        "after":
            32,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            4,
    },

    "completed_job": {
        "dataset":
            "Pingan",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            3407,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                seed3407_completion.relative_to(
                    ROOT
                )
            ),

        "completion_manifest_sha256":
            completion_sha,

        "best_AA_epoch":
            int(
                best_aa_epoch
            ),

        "best_AA_metrics":
            best_aa_metrics,

        "best_OA_epoch":
            int(
                best_oa_epoch
            ),

        "best_OA_metrics":
            best_oa_metrics,
    },

    "frozen_pingan_selection": {
        "selected_tau":
            1.0,

        "selection_manifest_relative_path":
            str(
                selection_path.relative_to(
                    ROOT
                )
            ),

        "selection_manifest_sha256":
            selection_sha_after,

        "unchanged_during_seed3407":
            True,

        "pingan_la_complete":
            True,
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(
                torch.backends.cudnn.version()
            ),

        "gpu0":
            torch.cuda.get_device_name(
                0
            ),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "full_delta": {
        "base_project_file_count":
            449,

        "delta_file_count":
            len(
                delta_records
            ),

        "resulting_project_file_count":
            len(
                project_files_after
            ),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "retuning_performed":
            False,
    },

    "next_authorized_job_after_external_save_acknowledgement": {
        "dataset":
            "Qingyun",

        "method":
            "LA_Loss",

        "tau":
            0.5,

        "seed":
            42,
    },
}


manifest_bytes = json.dumps(
    persistence_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 18. IMMEDIATE 32/36 GOOGLE-DRIVE PERSISTENCE
# ==================================================================================================

print()
print("-" * 174)
print("[10] IMMEDIATE 32/36 GOOGLE-DRIVE PERSISTENCE")
print("-" * 174)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if path.exists():

        raise RuntimeError(
            "STOP: 32/36 persistence artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 19. CREATE 29-FILE CUMULATIVE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(
        PERSISTENCE_ZIP
    )
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_STORED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record[
            "relative_path"
        ]

        source = (
            ROOT
            /
            relative_path
        )


        if (
            sha256_file(
                source
            )
            !=
            record[
                "sha256"
            ]
        ):

            raise RuntimeError(
                "STOP: delta artifact changed before archive:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=(
                "LTLC_DELTA/"
                +
                relative_path
            ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 20. VERIFY DRIVE ZIP
# ==================================================================================================

record_map = {
    (
        "LTLC_DELTA/"
        +
        record[
            "relative_path"
        ]
    ):
        record

    for record in delta_records
}


verified = 0


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:

    expected_members = (
        set(
            record_map.keys()
        )
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )


    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:

        raise RuntimeError(
            "STOP: 32/36 persistence ZIP member-set mismatch."
        )


    if (
        zf.read(
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        )
        !=
        manifest_bytes
    ):

        raise RuntimeError(
            "STOP: archived 32/36 manifest differs."
        )


    for archive_name, record in record_map.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(
                stream
            )


        if (
            archived_sha
            !=
            record[
                "sha256"
            ]
        ):

            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified += 1


if verified != POST_JOB_DELTA_COUNT:

    raise RuntimeError(
        "STOP: expected 29/29 delta verification."
    )


persistence_zip_sha = sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 21. SHA SIDECAR
# ==================================================================================================

sidecar_text = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 22. FINAL GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 9 FINAL GATE")
print("=" * 174)

print(
    "Authorized job                           : Pingan / LA / tau=1.0 / seed3407"
)

print(
    "Epochs                                   : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction      : PASS"
)

print(
    "Frozen selected tau                      : 1.0"
)

print(
    "Tau selection manifest unchanged         : PASS"
)

print(
    "Retuning performed                       : False"
)

print(
    "Pingan LA-loss                           : COMPLETE"
)

print(
    "Experiment-B progress                    : 32/36"
)

print(
    "LA-loss progress                         : 4/8"
)

print(
    "Original migration base                  : 449 files"
)

print(
    "Current project                          : 478 files"
)

print(
    "Cumulative post-migration delta          : 29 files PASS"
)

print(
    "Drive delta SHA verification             : 29/29 PASS"
)

print(
    "Scientific protocol changed              : False"
)

print(
    "Mixed precision                          : False"
)

print(
    "Multi-GPU                                : False"
)

print(
    "Scientific device                        : cuda:0 ONLY"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "SEED3407 BEST-AA EPOCH:"
)

print(
    best_aa_epoch
)

print(
    "SEED3407 BEST-AA METRICS:"
)

print(
    best_aa_metrics
)

print()
print(
    "SEED3407 BEST-OA EPOCH:"
)

print(
    best_oa_epoch
)

print(
    "SEED3407 BEST-OA METRICS:"
)

print(
    best_oa_metrics
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(
        training_elapsed,
        2,
    )
)

print()
print(
    "32/36 PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL COLAB CELL 9 PASS: True"
)

print()
print(
    "STATUS: "
    "PINGAN_LA_TAU1P0_SEED3407_COMPLETE__"
    "PINGAN_LA_COMPLETE__"
    "32_OF_36__"
    "29_FILE_CUMULATIVE_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_QINGYUN_TAU0P5_SEED42"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# COLAB CELL 10
#
# QINGYUN LA-LOSS tau=0.5 / SEED42
# + IMMEDIATE VERIFIED 33/36 GOOGLE-DRIVE PERSISTENCE
#
# FROZEN CONFIGURATION:
#   Dataset : Qingyun
#   Method  : LA_Loss
#   tau     : 0.5
#   seed    : 42
#   epochs  : 100
#
# THIS IS THE FIRST QINGYUN TUNING CANDIDATE.
# NO tau selection is allowed until tau=1.0 / seed42 is also complete.
#
# EXPECTED AFTER SUCCESS:
#   Experiment-B : 33/36
#   LA-loss      : 5/8
#   Project      : 485 files
#   Cumulative delta from 449-file migration base : 36 files
#
# NEXT AFTER EXTERNAL SAVE ACKNOWLEDGEMENT:
#   Qingyun / LA-loss / tau=1.0 / seed42
#
# SCIENTIFIC PROTOCOL REMAINS UNCHANGED.
# OFFICIAL TEST REMAINS SEALED.
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 10")
print("QINGYUN LA-LOSS tau=0.5 / SEED42 + IMMEDIATE 33/36 DRIVE PERSISTENCE")
print("=" * 174)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

ADAPTER_DIR = (
    DRIVE_ROOT
    / "migrations"
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 2. AUTHENTICATED 32/36 PARENT
# ==================================================================================================

PARENT_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "32of36_Post_Pingan_LA_tau1p0_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PARENT_ZIP = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}.zip"
)

PARENT_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}__MANIFEST.json"
)


EXPECTED_PARENT_ZIP_SHA256 = (
    "73ae65d1aecfe7027edb6ade360c8479"
    "3cfded37b661668adc59a0eb097d83ef"
)

EXPECTED_PARENT_MANIFEST_SHA256 = (
    "600af79a109ce7e93d7955721cb7cf06"
    "f9243a72266f9bfc596a9716c1bf4b8f"
)


# ==================================================================================================
# 3. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "4a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)


BASE_PROJECT_FILE_COUNT = 449
PRE_JOB_PROJECT_FILE_COUNT = 478
POST_JOB_PROJECT_FILE_COUNT = 485

PRE_JOB_DELTA_COUNT = 29
POST_JOB_DELTA_COUNT = 36


INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 4. NEW 33/36 PERSISTENCE PATHS
# ==================================================================================================

PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "33of36_Post_Qingyun_LA_tau0p5_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}.zip"
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__MANIFEST.json"
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__SHA256.txt"
)


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for chunk in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(chunk)

    return h.hexdigest()


def stream_sha256(
    stream,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    while True:

        block = stream.read(chunk_size)

        if not block:
            break

        h.update(block)

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(
        tmp,
        path,
    )


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. T4 SCIENTIFIC RUNTIME GATE
# ==================================================================================================

print()
print("-" * 174)
print("[1] T4 SCIENTIFIC RUNTIME GATE")
print("-" * 174)


if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError(
        "STOP: PyTorch runtime changed."
    )

if torch.version.cuda != "12.8":
    raise RuntimeError(
        "STOP: CUDA build changed."
    )

if not torch.cuda.is_available():
    raise RuntimeError(
        "STOP: CUDA unavailable."
    )

if int(
    torch.backends.cudnn.version()
) != 91900:
    raise RuntimeError(
        "STOP: cuDNN changed."
    )

if torch.cuda.device_count() < 1:
    raise RuntimeError(
        "STOP: no CUDA GPU visible."
    )


gpu0 = torch.cuda.get_device_name(0)


if "T4" not in gpu0:
    raise RuntimeError(
        "STOP: GPU0 is not Tesla T4."
    )

if str(DEVICE) != "cuda:0":
    raise RuntimeError(
        "STOP: scientific DEVICE is not cuda:0."
    )

if not T4_GATE_PATH.is_file():
    raise FileNotFoundError(
        "STOP: T4 runtime gate missing."
    )

if (
    sha256_file(T4_GATE_PATH)
    !=
    EXPECTED_T4_GATE_SHA256
):
    raise RuntimeError(
        "STOP: T4 runtime gate SHA changed."
    )


print("PyTorch           : 2.11.0+cu128 PASS")
print("CUDA              : 12.8 PASS")
print("cuDNN             : 91900 PASS")
print("GPU0              :", gpu0, "PASS")
print("Scientific DEVICE : cuda:0 ONLY PASS")
print("T4 runtime gate   : PASS")


# ==================================================================================================
# 7. EXTERNALLY SAVED 32/36 PARENT
# ==================================================================================================

print()
print("-" * 174)
print("[2] EXTERNALLY SAVED 32/36 PARENT")
print("-" * 174)


if not PARENT_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: saved 32/36 persistence ZIP missing."
    )


if not PARENT_MANIFEST.is_file():
    raise FileNotFoundError(
        "STOP: saved 32/36 persistence manifest missing."
    )


parent_zip_sha = sha256_file(
    PARENT_ZIP
)

parent_manifest_sha = sha256_file(
    PARENT_MANIFEST
)


if (
    parent_zip_sha
    !=
    EXPECTED_PARENT_ZIP_SHA256
):
    raise RuntimeError(
        "STOP: saved 32/36 ZIP SHA changed."
    )


if (
    parent_manifest_sha
    !=
    EXPECTED_PARENT_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: saved 32/36 manifest SHA changed."
    )


print("32/36 ZIP      : PASS")
print("32/36 manifest : PASS")
print("Save boundary  : CONFIRMED")


# ==================================================================================================
# 8. PRE-JOB SCIENTIFIC BOUNDARY
# ==================================================================================================

print()
print("-" * 174)
print("[3] PRE-JOB SCIENTIFIC BOUNDARY")
print("-" * 174)


pre_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_pre = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 4,
}


for method, expected in expected_pre.items():

    observed = pre_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: pre-job {method} count changed."
        )


if sum(
    pre_ledger.values()
) != 32:
    raise RuntimeError(
        "STOP: experiment is not at 32/36."
    )


project_files_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(project_files_before)
    !=
    PRE_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 478 files before Qingyun tau=0.5."
    )


# Pingan must remain fully frozen and complete.
pingan_selection_path = la_selection_path(
    "Pingan"
)

if not pingan_selection_path.is_file():
    raise FileNotFoundError(
        "STOP: Pingan frozen selection missing."
    )


if (
    sha256_file(
        pingan_selection_path
    )
    !=
    EXPECTED_PINGAN_SELECTION_SHA256
):
    raise RuntimeError(
        "STOP: frozen Pingan selection SHA changed."
    )


with pingan_selection_path.open(
    "r",
    encoding="utf-8",
) as f:

    pingan_selection = json.load(f)


if float(
    pingan_selection.get(
        "selected_tau"
    )
) != 1.0:
    raise RuntimeError(
        "STOP: Pingan selected tau changed."
    )


qingyun_tau05_completion = la_completion_path(
    "Qingyun",
    0.5,
    42,
)

qingyun_tau10_completion = la_completion_path(
    "Qingyun",
    1.0,
    42,
)

qingyun_selection_path = la_selection_path(
    "Qingyun"
)


if qingyun_tau05_completion.exists():
    raise RuntimeError(
        "STOP: Qingyun tau=0.5 seed42 already exists."
    )

if qingyun_tau10_completion.exists():
    raise RuntimeError(
        "STOP: Qingyun tau=1.0 seed42 unexpectedly exists."
    )

if qingyun_selection_path.exists():
    raise RuntimeError(
        "STOP: Qingyun selection exists before tuning."
    )


print()
print("Experiment-B            : 32/36 PASS")
print("LA-loss                 : 4/8 PASS")
print("Project files           : 478 PASS")
print("Pingan LA-loss          : COMPLETE PASS")
print("Qingyun tau=0.5 seed42  : ABSENT PASS")
print("Qingyun tau=1.0 seed42  : ABSENT PASS")
print("Qingyun tau selection   : ABSENT PASS")


# ==================================================================================================
# 9. FROZEN SCIENTIFIC SOURCE GATE
# ==================================================================================================

print()
print("-" * 174)
print("[4] FROZEN SCIENTIFIC SOURCE IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    actual_sha = sha256_file(
        path
    )

    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL",
    )

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP: frozen SHA mismatch: {label}"
        )


# ==================================================================================================
# 10. AUTHENTICATED 449-FILE MIGRATION BASE
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATED 28/36 MIGRATION BASE")
print("-" * 174)


if not LOCAL_MIGRATION_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: local authenticated migration ZIP missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


migration_manifest_sha = hashlib.sha256(
    migration_manifest_bytes
).hexdigest()


if (
    migration_manifest_sha
    !=
    EXPECTED_MIGRATION_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: migration manifest identity changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    migration_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(
    base_records
) != BASE_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: migration baseline is not 449 files."
    )


base_paths = {
    str(
        PurePosixPath(
            record[
                "relative_path"
            ]
        )
    )
    for record in base_records
}


before_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_before
}


pre_delta_paths = sorted(
    before_paths
    -
    base_paths
)


if (
    len(pre_delta_paths)
    !=
    PRE_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: current 32/36 cumulative delta is not 29 files."
    )


print("Migration manifest       : PASS")
print("Migration base files     : 449")
print("Current cumulative delta : 29 files PASS")


# ==================================================================================================
# 11. START AUTHORIZED GPU JOB
# ==================================================================================================

print()
print("=" * 174)
print("STARTING AUTHORIZED GPU JOB")
print("Qingyun / LA-loss / tau=0.5 / seed42")
print("=" * 174)


training_start = time.time()


completion_result = (
    run_experiment_b_la_loss_job(
        "Qingyun",
        0.5,
        42,
    )
)


training_elapsed = (
    time.time()
    -
    training_start
)


print()
print(
    "Authorized runner returned successfully."
)

print(
    "Runner elapsed seconds:",
    round(
        training_elapsed,
        2,
    )
)


# ==================================================================================================
# 12. COMPLETION AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[6] QINGYUN tau=0.5 COMPLETION AUDIT")
print("-" * 174)


if not qingyun_tau05_completion.is_file():
    raise FileNotFoundError(
        "STOP: Qingyun tau=0.5 completion manifest missing."
    )


with qingyun_tau05_completion.open(
    "r",
    encoding="utf-8",
) as f:

    completion_disk = json.load(f)


identity_checks = {
    "status":
        completion_disk.get(
            "status"
        ) == "COMPLETE",

    "dataset":
        completion_disk.get(
            "dataset"
        ) == "Qingyun",

    "method":
        completion_disk.get(
            "method"
        ) == "LA_Loss",

    "tau":
        float(
            completion_disk.get(
                "tau"
            )
        ) == 0.5,

    "seed":
        int(
            completion_disk.get(
                "seed"
            )
        ) == 42,

    "epochs":
        int(
            completion_disk.get(
                "epochs_completed"
            )
        ) == 100,
}


for label, passed in identity_checks.items():

    print(
        f"{label:14s}:",
        "PASS"
        if passed
        else "FAIL",
    )


if not all(
    identity_checks.values()
):
    raise RuntimeError(
        "STOP: Qingyun tau=0.5 completion identity failed."
    )


safety = completion_disk[
    "safety"
]


if not (
    safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    safety.get(
        "official_test_values_accessed"
    ) is False
    and
    safety.get(
        "test_dataset_created"
    ) is False
    and
    safety.get(
        "test_dataloader_created"
    ) is False
    and
    safety.get(
        "test_inference_performed"
    ) is False
):
    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


completion_files = completion_disk[
    "files"
]


if len(
    completion_files
) != 6:
    raise RuntimeError(
        "STOP: Qingyun tau=0.5 completion must record six run artifacts."
    )


for role, record in completion_files.items():

    path = (
        ROOT
        /
        record[
            "path"
        ]
    )

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: Qingyun tau=0.5 artifact missing: {role}"
        )

    if (
        sha256_file(
            path
        )
        !=
        record[
            "sha256"
        ]
    ):
        raise RuntimeError(
            f"STOP: Qingyun tau=0.5 SHA mismatch: {role}"
        )

    print(
        f"{role:24s}: PASS"
    )


best_aa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

best_aa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)

best_oa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_epoch"
    ]
)

best_oa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_metrics"
    ]
)


print()
print(
    "Qingyun tau=0.5 Best-AA epoch:",
    best_aa_epoch
)

print(
    "Qingyun tau=0.5 Best-AA metrics:"
)

print(
    best_aa_metrics
)

print()
print(
    "Qingyun tau=0.5 Best-OA epoch:",
    best_oa_epoch
)

print(
    "Qingyun tau=0.5 Best-OA metrics:"
)

print(
    best_oa_metrics
)


# ==================================================================================================
# 13. NO PREMATURE TAU-SELECTION GATE
# ==================================================================================================

print()
print("-" * 174)
print("[7] QINGYUN TUNING-STATE GATE")
print("-" * 174)


if qingyun_tau10_completion.exists():
    raise RuntimeError(
        "STOP: Qingyun tau=1.0 completion unexpectedly appeared."
    )


if qingyun_selection_path.exists():
    raise RuntimeError(
        "STOP: Qingyun tau-selection was created before tau=1.0 candidate."
    )


print(
    "tau=0.5 seed42 : COMPLETE PASS"
)

print(
    "tau=1.0 seed42 : NOT YET RUN PASS"
)

print(
    "Selected tau   : NONE YET PASS"
)


# ==================================================================================================
# 14. POST-JOB SCIENTIFIC LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[8] POST-JOB SCIENTIFIC LEDGER")
print("-" * 174)


post_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_post = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 5,
}


for method, expected in expected_post.items():

    observed = post_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: post-job {method} count changed."
        )


if sum(
    post_ledger.values()
) != 33:
    raise RuntimeError(
        "STOP: Experiment-B state is not 33/36."
    )


print()
print(
    "Experiment-B progress : 33/36 PASS"
)

print(
    "LA-loss progress       : 5/8 PASS"
)


# ==================================================================================================
# 15. PROJECT FOOTPRINT
# ==================================================================================================

print()
print("-" * 174)
print("[9] 33/36 PROJECT-FOOTPRINT AUDIT")
print("-" * 174)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if (
    len(
        project_files_after
    )
    !=
    POST_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 485 project files after Qingyun tau=0.5.\n"
        f"Observed: {len(project_files_after)}"
    )


after_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_after
}


new_since_32 = sorted(
    after_paths
    -
    before_paths
)


if len(
    new_since_32
) != 7:
    raise RuntimeError(
        "STOP: Qingyun tau=0.5 seed42 must add exactly seven files."
    )


expected_qingyun_tau05_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/qingyun/tau0p5/seed42/"
)


if not all(
    path.startswith(
        expected_qingyun_tau05_prefix
    )
    for path in new_since_32
):
    raise RuntimeError(
        "STOP: Qingyun tau=0.5 added an unauthorized project path."
    )


print(
    "New Qingyun tau=0.5 files:",
    len(
        new_since_32
    ),
    "PASS",
)


for path in new_since_32:
    print(
        " +",
        path
    )


cumulative_delta_paths = sorted(
    after_paths
    -
    base_paths
)


if (
    len(
        cumulative_delta_paths
    )
    !=
    POST_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: expected 36-file cumulative delta at 33/36."
    )


print()
print(
    "Migration base files   : 449"
)

print(
    "Current project files  : 485"
)

print(
    "Cumulative delta files : 36 PASS"
)


# ==================================================================================================
# 16. HASH ONLY THE 36 DELTA FILES
# ==================================================================================================

delta_records = []


for relative_path in cumulative_delta_paths:

    path = (
        ROOT
        /
        relative_path
    )

    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(
                    path.stat().st_size
                ),

            "sha256":
                sha256_file(
                    path
                ),
        }
    )


completion_sha = sha256_file(
    qingyun_tau05_completion
)


# ==================================================================================================
# 17. BUILD 33/36 PERSISTENCE MANIFEST
# ==================================================================================================

persistence_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent_chain": {
        "previous_progress":
            "32/36",

        "previous_persistence_zip_sha256":
            parent_zip_sha,

        "previous_persistence_manifest_sha256":
            parent_manifest_sha,

        "base_project_file_count":
            449,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            32,

        "after":
            33,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            5,
    },

    "completed_job": {
        "dataset":
            "Qingyun",

        "method":
            "LA_Loss",

        "tau":
            0.5,

        "seed":
            42,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                qingyun_tau05_completion.relative_to(
                    ROOT
                )
            ),

        "completion_manifest_sha256":
            completion_sha,

        "best_AA_epoch":
            int(
                best_aa_epoch
            ),

        "best_AA_metrics":
            best_aa_metrics,

        "best_OA_epoch":
            int(
                best_oa_epoch
            ),

        "best_OA_metrics":
            best_oa_metrics,
    },

    "qingyun_tuning": {
        "candidate_tau_0p5_complete":
            True,

        "candidate_tau_1p0_complete":
            False,

        "selected_tau":
            None,

        "selection_performed":
            False,
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(
                torch.backends.cudnn.version()
            ),

        "gpu0":
            torch.cuda.get_device_name(
                0
            ),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "full_delta": {
        "base_project_file_count":
            449,

        "delta_file_count":
            len(
                delta_records
            ),

        "resulting_project_file_count":
            len(
                project_files_after
            ),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "retuning_outside_seed42":
            False,
    },

    "next_authorized_job_after_external_save_acknowledgement": {
        "dataset":
            "Qingyun",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            42,
    },
}


manifest_bytes = json.dumps(
    persistence_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 18. IMMEDIATE 33/36 GOOGLE-DRIVE PERSISTENCE
# ==================================================================================================

print()
print("-" * 174)
print("[10] IMMEDIATE 33/36 GOOGLE-DRIVE PERSISTENCE")
print("-" * 174)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if path.exists():
        raise RuntimeError(
            "STOP: 33/36 persistence artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 19. CREATE 36-FILE CUMULATIVE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(
        PERSISTENCE_ZIP
    )
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_STORED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record[
            "relative_path"
        ]

        source = (
            ROOT
            /
            relative_path
        )


        if (
            sha256_file(
                source
            )
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: delta artifact changed before archive:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=(
                "LTLC_DELTA/"
                +
                relative_path
            ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 20. VERIFY DRIVE ZIP
# ==================================================================================================

record_map = {
    (
        "LTLC_DELTA/"
        +
        record[
            "relative_path"
        ]
    ):
        record

    for record in delta_records
}


verified = 0


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:

    expected_members = (
        set(
            record_map.keys()
        )
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )


    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:
        raise RuntimeError(
            "STOP: 33/36 persistence ZIP member-set mismatch."
        )


    if (
        zf.read(
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        )
        !=
        manifest_bytes
    ):
        raise RuntimeError(
            "STOP: archived 33/36 manifest differs."
        )


    for archive_name, record in record_map.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(
                stream
            )


        if (
            archived_sha
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified += 1


if verified != POST_JOB_DELTA_COUNT:
    raise RuntimeError(
        "STOP: expected 36/36 delta verification."
    )


persistence_zip_sha = sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 21. SHA SIDECAR
# ==================================================================================================

sidecar_text = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 22. FINAL GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 10 FINAL GATE")
print("=" * 174)

print(
    "Authorized job                           : Qingyun / LA / tau=0.5 / seed42"
)

print(
    "Epochs                                   : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction      : PASS"
)

print(
    "Qingyun tau=0.5 candidate                : COMPLETE"
)

print(
    "Qingyun tau=1.0 candidate                : NOT YET RUN"
)

print(
    "Qingyun tau selected                     : False"
)

print(
    "Experiment-B progress                    : 33/36"
)

print(
    "LA-loss progress                         : 5/8"
)

print(
    "Original migration base                  : 449 files"
)

print(
    "Current project                          : 485 files"
)

print(
    "Cumulative post-migration delta          : 36 files PASS"
)

print(
    "Drive delta SHA verification             : 36/36 PASS"
)

print(
    "Scientific protocol changed              : False"
)

print(
    "Mixed precision                          : False"
)

print(
    "Multi-GPU                                : False"
)

print(
    "Scientific device                        : cuda:0 ONLY"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "QINGYUN tau=0.5 BEST-AA EPOCH:"
)

print(
    best_aa_epoch
)

print(
    "QINGYUN tau=0.5 BEST-AA METRICS:"
)

print(
    best_aa_metrics
)

print()
print(
    "QINGYUN tau=0.5 BEST-OA EPOCH:"
)

print(
    best_oa_epoch
)

print(
    "QINGYUN tau=0.5 BEST-OA METRICS:"
)

print(
    best_oa_metrics
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(
        training_elapsed,
        2,
    )
)

print()
print(
    "33/36 PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL COLAB CELL 10 PASS: True"
)

print()
print(
    "STATUS: "
    "QINGYUN_LA_TAU0P5_SEED42_COMPLETE__"
    "33_OF_36__"
    "36_FILE_CUMULATIVE_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_QINGYUN_TAU1P0_SEED42"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# COLAB CELL 11
#
# QINGYUN LA-LOSS tau=1.0 / SEED42
# + AUTHENTICATED QINGYUN TAU SELECTION
# + IMMEDIATE VERIFIED 34/36 GOOGLE-DRIVE PERSISTENCE
#
# EXPECTED AFTER SUCCESS:
#   Experiment-B : 34/36
#   LA-loss      : 6/8
#   Qingyun tuning : COMPLETE
#   Project      : 493 files
#   Cumulative delta from 449-file migration base : 44 files
#
# REMAINING AFTER THIS:
#   selected Qingyun tau / seed123
#   selected Qingyun tau / seed3407
#
# OFFICIAL TEST REMAINS SEALED.
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 11")
print("QINGYUN LA-LOSS tau=1.0 / SEED42 + TAU SELECTION + 34/36 DRIVE PERSISTENCE")
print("=" * 174)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path("/content/LTLC")

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

ADAPTER_DIR = (
    DRIVE_ROOT
    / "migrations"
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 2. AUTHENTICATED 33/36 PARENT
# ==================================================================================================

PARENT_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "33of36_Post_Qingyun_LA_tau0p5_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PARENT_ZIP = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}.zip"
)

PARENT_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}__MANIFEST.json"
)


EXPECTED_PARENT_ZIP_SHA256 = (
    "a8f03cfd59aae7aa38d5b171934615cc"
    "dac55cfd877189451a14bb2575df8bae"
)

EXPECTED_PARENT_MANIFEST_SHA256 = (
    "815a34a1e724fe6cab8d139c57fde68e"
    "584f98ac666fb426ac7259a72f0090e1"
)


# ==================================================================================================
# 3. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "4a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)


BASE_PROJECT_FILE_COUNT = 449
PRE_JOB_PROJECT_FILE_COUNT = 485
POST_JOB_PROJECT_FILE_COUNT = 493

PRE_JOB_DELTA_COUNT = 36
POST_JOB_DELTA_COUNT = 44

INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 4. NEW 34/36 PERSISTENCE PATHS
# ==================================================================================================

PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "34of36_Post_Qingyun_LA_Tuning_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}.zip"
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__MANIFEST.json"
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__SHA256.txt"
)


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for chunk in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(chunk)

    return h.hexdigest()


def stream_sha256(stream, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    while True:

        block = stream.read(chunk_size)

        if not block:
            break

        h.update(block)

    return h.hexdigest()


def atomic_write_bytes(path, payload):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(str(path) + ".tmp")

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(tmp, path)


def count_completion_manifests(method_name):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. T4 SCIENTIFIC RUNTIME GATE
# ==================================================================================================

print()
print("-" * 174)
print("[1] T4 SCIENTIFIC RUNTIME GATE")
print("-" * 174)


if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError("STOP: PyTorch runtime changed.")

if torch.version.cuda != "12.8":
    raise RuntimeError("STOP: CUDA build changed.")

if not torch.cuda.is_available():
    raise RuntimeError("STOP: CUDA unavailable.")

if int(torch.backends.cudnn.version()) != 91900:
    raise RuntimeError("STOP: cuDNN changed.")

if torch.cuda.device_count() < 1:
    raise RuntimeError("STOP: no CUDA GPU visible.")


gpu0 = torch.cuda.get_device_name(0)


if "T4" not in gpu0:
    raise RuntimeError("STOP: GPU0 is not Tesla T4.")

if str(DEVICE) != "cuda:0":
    raise RuntimeError(
        "STOP: scientific DEVICE is not cuda:0."
    )

if not T4_GATE_PATH.is_file():
    raise FileNotFoundError(
        "STOP: T4 runtime gate missing."
    )

if (
    sha256_file(T4_GATE_PATH)
    != EXPECTED_T4_GATE_SHA256
):
    raise RuntimeError(
        "STOP: T4 runtime gate SHA changed."
    )


print("PyTorch           : 2.11.0+cu128 PASS")
print("CUDA              : 12.8 PASS")
print("cuDNN             : 91900 PASS")
print("GPU0              :", gpu0, "PASS")
print("Scientific DEVICE : cuda:0 ONLY PASS")
print("T4 runtime gate   : PASS")


# ==================================================================================================
# 7. EXTERNALLY SAVED 33/36 PARENT
# ==================================================================================================

print()
print("-" * 174)
print("[2] EXTERNALLY SAVED 33/36 PARENT")
print("-" * 174)


if not PARENT_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: saved 33/36 persistence ZIP missing."
    )

if not PARENT_MANIFEST.is_file():
    raise FileNotFoundError(
        "STOP: saved 33/36 persistence manifest missing."
    )


parent_zip_sha = sha256_file(
    PARENT_ZIP
)

parent_manifest_sha = sha256_file(
    PARENT_MANIFEST
)


if parent_zip_sha != EXPECTED_PARENT_ZIP_SHA256:
    raise RuntimeError(
        "STOP: saved 33/36 ZIP SHA changed."
    )

if (
    parent_manifest_sha
    != EXPECTED_PARENT_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: saved 33/36 manifest SHA changed."
    )


print("33/36 ZIP      : PASS")
print("33/36 manifest : PASS")
print("Save boundary  : CONFIRMED")


# ==================================================================================================
# 8. PRE-JOB SCIENTIFIC BOUNDARY
# ==================================================================================================

print()
print("-" * 174)
print("[3] PRE-JOB SCIENTIFIC BOUNDARY")
print("-" * 174)


pre_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_pre = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 5,
}


for method, expected in expected_pre.items():

    observed = pre_ledger[method]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: pre-job {method} count changed."
        )


if sum(pre_ledger.values()) != 33:
    raise RuntimeError(
        "STOP: experiment is not at 33/36."
    )


project_files_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(project_files_before)
    != PRE_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 485 project files before Qingyun tau=1.0."
    )


# Pingan frozen boundary must remain unchanged.
pingan_selection_path = la_selection_path(
    "Pingan"
)


if not pingan_selection_path.is_file():
    raise FileNotFoundError(
        "STOP: Pingan selection manifest missing."
    )


if (
    sha256_file(pingan_selection_path)
    != EXPECTED_PINGAN_SELECTION_SHA256
):
    raise RuntimeError(
        "STOP: Pingan selection SHA changed."
    )


qingyun_tau05_completion = (
    la_completion_path(
        "Qingyun",
        0.5,
        42,
    )
)

qingyun_tau10_completion = (
    la_completion_path(
        "Qingyun",
        1.0,
        42,
    )
)

qingyun_selection_path = (
    la_selection_path(
        "Qingyun"
    )
)


if not qingyun_tau05_completion.is_file():
    raise RuntimeError(
        "STOP: Qingyun tau=0.5 seed42 completion missing."
    )

if qingyun_tau10_completion.exists():
    raise RuntimeError(
        "STOP: Qingyun tau=1.0 seed42 already exists."
    )

if qingyun_selection_path.exists():
    raise RuntimeError(
        "STOP: Qingyun selection exists prematurely."
    )


print()
print("Experiment-B           : 33/36 PASS")
print("LA-loss                : 5/8 PASS")
print("Project files          : 485 PASS")
print("Qingyun tau=0.5 seed42 : COMPLETE PASS")
print("Qingyun tau=1.0 seed42 : ABSENT PASS")
print("Qingyun tau selection  : ABSENT PASS")


# ==================================================================================================
# 9. FROZEN SOURCE GATE
# ==================================================================================================

print()
print("-" * 174)
print("[4] FROZEN SCIENTIFIC SOURCE IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    actual_sha = sha256_file(path)

    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL",
    )

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP: frozen SHA mismatch: {label}"
        )


# ==================================================================================================
# 10. AUTHENTICATED 449-FILE MIGRATION BASE
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATED 28/36 MIGRATION BASE")
print("-" * 174)


if not LOCAL_MIGRATION_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: local authenticated migration ZIP missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


migration_manifest_sha = hashlib.sha256(
    migration_manifest_bytes
).hexdigest()


if (
    migration_manifest_sha
    != EXPECTED_MIGRATION_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: migration manifest identity changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    migration_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(base_records) != BASE_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: migration baseline is not 449 files."
    )


base_paths = {
    str(
        PurePosixPath(
            record["relative_path"]
        )
    )
    for record in base_records
}


before_paths = {
    str(
        p.relative_to(ROOT)
    )
    for p in project_files_before
}


pre_delta_paths = sorted(
    before_paths
    -
    base_paths
)


if (
    len(pre_delta_paths)
    != PRE_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: current 33/36 cumulative delta is not 36 files."
    )


print("Migration manifest       : PASS")
print("Migration base files     : 449")
print("Current cumulative delta : 36 files PASS")


# ==================================================================================================
# 11. START AUTHORIZED GPU JOB
# ==================================================================================================

print()
print("=" * 174)
print("STARTING AUTHORIZED GPU JOB")
print("Qingyun / LA-loss / tau=1.0 / seed42")
print("=" * 174)


training_start = time.time()


completion_result = (
    run_experiment_b_la_loss_job(
        "Qingyun",
        1.0,
        42,
    )
)


training_elapsed = (
    time.time()
    -
    training_start
)


print()
print(
    "Authorized runner returned successfully."
)

print(
    "Runner elapsed seconds:",
    round(
        training_elapsed,
        2,
    )
)


# ==================================================================================================
# 12. tau=1.0 COMPLETION AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[6] QINGYUN tau=1.0 COMPLETION AUDIT")
print("-" * 174)


if not qingyun_tau10_completion.is_file():
    raise FileNotFoundError(
        "STOP: Qingyun tau=1.0 completion manifest missing."
    )


with qingyun_tau10_completion.open(
    "r",
    encoding="utf-8",
) as f:

    tau10_disk = json.load(f)


identity_checks = {
    "status":
        tau10_disk.get("status") == "COMPLETE",

    "dataset":
        tau10_disk.get("dataset") == "Qingyun",

    "method":
        tau10_disk.get("method") == "LA_Loss",

    "tau":
        float(
            tau10_disk.get("tau")
        ) == 1.0,

    "seed":
        int(
            tau10_disk.get("seed")
        ) == 42,

    "epochs":
        int(
            tau10_disk.get(
                "epochs_completed"
            )
        ) == 100,
}


for label, passed in identity_checks.items():

    print(
        f"{label:14s}:",
        "PASS"
        if passed
        else "FAIL",
    )


if not all(identity_checks.values()):
    raise RuntimeError(
        "STOP: Qingyun tau=1.0 completion identity failed."
    )


safety = tau10_disk["safety"]


if not (
    safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    safety.get(
        "official_test_values_accessed"
    ) is False
    and
    safety.get(
        "test_dataset_created"
    ) is False
    and
    safety.get(
        "test_dataloader_created"
    ) is False
    and
    safety.get(
        "test_inference_performed"
    ) is False
):
    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


tau10_files = tau10_disk["files"]


if len(tau10_files) != 6:
    raise RuntimeError(
        "STOP: Qingyun tau=1.0 completion must record six run artifacts."
    )


for role, record in tau10_files.items():

    path = ROOT / record["path"]

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: Qingyun tau=1.0 artifact missing: {role}"
        )

    if (
        sha256_file(path)
        != record["sha256"]
    ):
        raise RuntimeError(
            f"STOP: Qingyun tau=1.0 SHA mismatch: {role}"
        )

    print(
        f"{role:24s}: PASS"
    )


tau10_best_aa_epoch = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

tau10_best_aa = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)

tau10_best_oa_epoch = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_OA_epoch"
    ]
)

tau10_best_oa = (
    tau10_disk[
        "checkpoint_policy"
    ][
        "best_OA_metrics"
    ]
)


print()
print(
    "Qingyun tau=1.0 Best-AA epoch:",
    tau10_best_aa_epoch
)

print(
    "Qingyun tau=1.0 Best-AA metrics:"
)

print(
    tau10_best_aa
)

print()
print(
    "Qingyun tau=1.0 Best-OA epoch:",
    tau10_best_oa_epoch
)

print(
    "Qingyun tau=1.0 Best-OA metrics:"
)

print(
    tau10_best_oa
)


# ==================================================================================================
# 13. LOAD tau=0.5 CANDIDATE
# ==================================================================================================

with qingyun_tau05_completion.open(
    "r",
    encoding="utf-8",
) as f:

    tau05_disk = json.load(f)


tau05_best_aa_epoch = (
    tau05_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

tau05_best_aa = (
    tau05_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)


print()
print("-" * 174)
print("[7] QINGYUN SEED42 TUNING CANDIDATES")
print("-" * 174)


print("tau=0.5")
print(
    "  epoch    :",
    tau05_best_aa_epoch
)
print(
    "  AA       :",
    tau05_best_aa["AA"]
)
print(
    "  Macro-F1 :",
    tau05_best_aa["Macro_F1"]
)
print(
    "  OA       :",
    tau05_best_aa["OA"]
)


print()
print("tau=1.0")
print(
    "  epoch    :",
    tau10_best_aa_epoch
)
print(
    "  AA       :",
    tau10_best_aa["AA"]
)
print(
    "  Macro-F1 :",
    tau10_best_aa["Macro_F1"]
)
print(
    "  OA       :",
    tau10_best_aa["OA"]
)


# ==================================================================================================
# 14. INDEPENDENT FROZEN TAU-SELECTION RULE
# ==================================================================================================

METRIC_TOL = 1e-12


def independent_choose_tau(
    tau_a,
    metric_a,
    tau_b,
    metric_b,
):

    if (
        metric_a["AA"]
        >
        metric_b["AA"]
        +
        METRIC_TOL
    ):
        return tau_a

    if (
        metric_b["AA"]
        >
        metric_a["AA"]
        +
        METRIC_TOL
    ):
        return tau_b

    if (
        metric_a["Macro_F1"]
        >
        metric_b["Macro_F1"]
        +
        METRIC_TOL
    ):
        return tau_a

    if (
        metric_b["Macro_F1"]
        >
        metric_a["Macro_F1"]
        +
        METRIC_TOL
    ):
        return tau_b

    if (
        metric_a["OA"]
        >
        metric_b["OA"]
        +
        METRIC_TOL
    ):
        return tau_a

    if (
        metric_b["OA"]
        >
        metric_a["OA"]
        +
        METRIC_TOL
    ):
        return tau_b

    return min(
        tau_a,
        tau_b,
    )


independent_selected_tau = (
    independent_choose_tau(
        0.5,
        tau05_best_aa,
        1.0,
        tau10_best_aa,
    )
)


print()
print(
    "Independent frozen-rule selected tau:",
    independent_selected_tau
)


# ==================================================================================================
# 15. AUTHENTICATED RUNNER-CREATED QINGYUN SELECTION
# ==================================================================================================

print()
print("-" * 174)
print("[8] AUTHENTICATED QINGYUN TAU SELECTION")
print("-" * 174)


if not qingyun_selection_path.is_file():
    raise FileNotFoundError(
        "STOP: runner did not create Qingyun selection manifest."
    )


selection_sha = sha256_file(
    qingyun_selection_path
)


with qingyun_selection_path.open(
    "r",
    encoding="utf-8",
) as f:

    selection_disk = json.load(f)


manifest_selected_tau = (
    selection_disk.get(
        "selected_tau"
    )
)


if manifest_selected_tau is None:
    raise RuntimeError(
        "STOP: Qingyun selection manifest has no selected_tau."
    )


if (
    abs(
        float(manifest_selected_tau)
        -
        float(independent_selected_tau)
    )
    >
    METRIC_TOL
):
    raise RuntimeError(
        "STOP: authenticated Qingyun selection disagrees with independent frozen rule."
    )


if (
    selection_disk.get(
        "official_test_status"
    )
    != "SEALED"
):
    raise RuntimeError(
        "STOP: selection manifest does not preserve official-test seal."
    )


if (
    selection_disk.get(
        "official_test_values_accessed"
    )
    is not False
):
    raise RuntimeError(
        "STOP: selection manifest indicates official-test access."
    )


if (
    selection_disk.get(
        "official_test_used_for_selection"
    )
    is not False
):
    raise RuntimeError(
        "STOP: official test was used for tau selection."
    )


print("Selection manifest:")
print(selection_disk)

print()
print(
    "Selection SHA-256:"
)

print(
    selection_sha
)

print()
print(
    "Authenticated selected tau:",
    manifest_selected_tau
)

print(
    "Independent agreement   : PASS"
)


# ==================================================================================================
# 16. POST-JOB SCIENTIFIC LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[9] POST-SELECTION SCIENTIFIC LEDGER")
print("-" * 174)


post_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_post = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 6,
}


for method, expected in expected_post.items():

    observed = post_ledger[method]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: post-job {method} count changed."
        )


if sum(post_ledger.values()) != 34:
    raise RuntimeError(
        "STOP: Experiment-B state is not 34/36."
    )


print()
print(
    "Experiment-B progress : 34/36 PASS"
)

print(
    "LA-loss progress       : 6/8 PASS"
)

print(
    "Qingyun tuning         : COMPLETE PASS"
)


# ==================================================================================================
# 17. PROJECT FOOTPRINT
# ==================================================================================================

print()
print("-" * 174)
print("[10] 34/36 PROJECT-FOOTPRINT AUDIT")
print("-" * 174)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if (
    len(project_files_after)
    != POST_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 493 project files after Qingyun tuning.\n"
        f"Observed: {len(project_files_after)}"
    )


after_paths = {
    str(
        p.relative_to(ROOT)
    )
    for p in project_files_after
}


new_since_33 = sorted(
    after_paths
    -
    before_paths
)


if len(new_since_33) != 8:
    raise RuntimeError(
        "STOP: Qingyun tau=1.0 + selection must add exactly eight files."
    )


tau10_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/qingyun/tau1p0/seed42/"
)

selection_rel = str(
    qingyun_selection_path.relative_to(
        ROOT
    )
)


for path in new_since_33:

    if not (
        path.startswith(
            tau10_prefix
        )
        or
        path == selection_rel
    ):
        raise RuntimeError(
            "STOP: unauthorized new project path:\n"
            f"{path}"
        )


print(
    "Files added since 33/36:",
    len(new_since_33),
    "PASS",
)


for path in new_since_33:
    print(
        " +",
        path
    )


cumulative_delta_paths = sorted(
    after_paths
    -
    base_paths
)


if (
    len(cumulative_delta_paths)
    != POST_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: expected 44-file cumulative delta at 34/36."
    )


print()
print(
    "Migration base files   : 449"
)

print(
    "Current project files  : 493"
)

print(
    "Cumulative delta files : 44 PASS"
)


# ==================================================================================================
# 18. HASH ONLY THE 44 DELTA FILES
# ==================================================================================================

delta_records = []


for relative_path in cumulative_delta_paths:

    path = ROOT / relative_path

    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(
                    path.stat().st_size
                ),

            "sha256":
                sha256_file(path),
        }
    )


tau10_completion_sha = sha256_file(
    qingyun_tau10_completion
)


# ==================================================================================================
# 19. BUILD 34/36 PERSISTENCE MANIFEST
# ==================================================================================================

persistence_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent_chain": {
        "previous_progress":
            "33/36",

        "previous_persistence_zip_sha256":
            parent_zip_sha,

        "previous_persistence_manifest_sha256":
            parent_manifest_sha,

        "base_project_file_count":
            449,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            33,

        "after":
            34,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            6,
    },

    "completed_job": {
        "dataset":
            "Qingyun",

        "method":
            "LA_Loss",

        "tau":
            1.0,

        "seed":
            42,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                qingyun_tau10_completion.relative_to(
                    ROOT
                )
            ),

        "completion_manifest_sha256":
            tau10_completion_sha,

        "best_AA_epoch":
            int(
                tau10_best_aa_epoch
            ),

        "best_AA_metrics":
            tau10_best_aa,

        "best_OA_epoch":
            int(
                tau10_best_oa_epoch
            ),

        "best_OA_metrics":
            tau10_best_oa,
    },

    "qingyun_tuning": {
        "candidate_tau_0p5_best_AA":
            tau05_best_aa,

        "candidate_tau_1p0_best_AA":
            tau10_best_aa,

        "independent_selected_tau":
            independent_selected_tau,

        "authenticated_selected_tau":
            float(
                manifest_selected_tau
            ),

        "selection_manifest_relative_path":
            selection_rel,

        "selection_manifest_sha256":
            selection_sha,

        "selection_manifest":
            selection_disk,
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(
                torch.backends.cudnn.version()
            ),

        "gpu0":
            torch.cuda.get_device_name(0),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "full_delta": {
        "base_project_file_count":
            449,

        "delta_file_count":
            len(delta_records),

        "resulting_project_file_count":
            len(project_files_after),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "qingyun_tau_selection_complete":
            True,
    },

    "next_authorized_jobs_after_external_save_acknowledgement": {
        "dataset":
            "Qingyun",

        "method":
            "LA_Loss",

        "selected_tau":
            float(
                manifest_selected_tau
            ),

        "remaining_seeds":
            [
                123,
                3407,
            ],
    },
}


manifest_bytes = json.dumps(
    persistence_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 20. IMMEDIATE 34/36 GOOGLE-DRIVE PERSISTENCE
# ==================================================================================================

print()
print("-" * 174)
print("[11] IMMEDIATE 34/36 GOOGLE-DRIVE PERSISTENCE")
print("-" * 174)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if path.exists():
        raise RuntimeError(
            "STOP: 34/36 persistence artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 21. CREATE 44-FILE CUMULATIVE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(PERSISTENCE_ZIP)
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_STORED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record[
            "relative_path"
        ]

        source = ROOT / relative_path


        if (
            sha256_file(source)
            != record["sha256"]
        ):
            raise RuntimeError(
                "STOP: delta artifact changed before archive:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=(
                "LTLC_DELTA/"
                +
                relative_path
            ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 22. VERIFY DRIVE ZIP
# ==================================================================================================

record_map = {
    (
        "LTLC_DELTA/"
        +
        record["relative_path"]
    ):
        record

    for record in delta_records
}


verified = 0


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:

    expected_members = (
        set(
            record_map.keys()
        )
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )


    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:
        raise RuntimeError(
            "STOP: 34/36 persistence ZIP member-set mismatch."
        )


    if (
        zf.read(
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        )
        != manifest_bytes
    ):
        raise RuntimeError(
            "STOP: archived 34/36 manifest differs."
        )


    for archive_name, record in record_map.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(
                stream
            )


        if (
            archived_sha
            != record["sha256"]
        ):
            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified += 1


if verified != POST_JOB_DELTA_COUNT:
    raise RuntimeError(
        "STOP: expected 44/44 delta verification."
    )


persistence_zip_sha = sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 23. SHA SIDECAR
# ==================================================================================================

sidecar_text = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 24. FINAL GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 11 FINAL GATE")
print("=" * 174)

print(
    "Authorized job                           : Qingyun / LA / tau=1.0 / seed42"
)

print(
    "Epochs                                   : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction      : PASS"
)

print(
    "Qingyun tau=0.5 candidate                : COMPLETE"
)

print(
    "Qingyun tau=1.0 candidate                : COMPLETE"
)

print(
    "Qingyun frozen tau selection             : COMPLETE"
)

print(
    "Independent selected tau                 :",
    independent_selected_tau
)

print(
    "Authenticated selected tau               :",
    manifest_selected_tau
)

print(
    "Selection-rule agreement                 : PASS"
)

print(
    "Experiment-B progress                    : 34/36"
)

print(
    "LA-loss progress                         : 6/8"
)

print(
    "Original migration base                  : 449 files"
)

print(
    "Current project                          : 493 files"
)

print(
    "Cumulative post-migration delta          : 44 files PASS"
)

print(
    "Drive delta SHA verification             : 44/44 PASS"
)

print(
    "Scientific protocol changed              : False"
)

print(
    "Mixed precision                          : False"
)

print(
    "Multi-GPU                                : False"
)

print(
    "Scientific device                        : cuda:0 ONLY"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "QINGYUN tau=1.0 BEST-AA EPOCH:"
)

print(
    tau10_best_aa_epoch
)

print(
    "QINGYUN tau=1.0 BEST-AA METRICS:"
)

print(
    tau10_best_aa
)

print()
print(
    "QINGYUN SELECTED TAU:"
)

print(
    manifest_selected_tau
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(
        training_elapsed,
        2,
    )
)

print()
print(
    "34/36 PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL COLAB CELL 11 PASS: True"
)

print()
print(
    "STATUS: "
    "QINGYUN_LA_TAU1P0_SEED42_COMPLETE__"
    "QINGYUN_TAU_SELECTION_FROZEN__"
    "34_OF_36__"
    "44_FILE_CUMULATIVE_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_SELECTED_TAU_SEED123"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CELL 12B
#
# OPTIMIZED FRESH-COLAB RECOVERY:
#
#   authenticated 28/36 self-contained migration base
#                     +
#   authenticated 34/36 cumulative delta
#                     ↓
#             exact 34/36 state
#
# THEN:
#   - execute authenticated Colab bootstrap
#   - execute authenticated LA runtime prefix
#   - verify T4 / cuda:0
#   - verify frozen scientific artifacts
#   - verify Experiment-B = 34/36
#   - verify Pingan selected tau = 1.0
#   - verify Qingyun selected tau = 0.5
#   - verify seed123 absent
#   - verify seed3407 absent
#   - verify official test SEALED
#
# NO TRAINING
# NO OFFICIAL-TEST ACCESS
# NO PREPROCESSING
# NO PCA RECOMPUTATION
# NO PREVIOUS JOB REEXECUTION
# ==================================================================================================

from pathlib import Path, PurePosixPath
import hashlib
import json
import os
import platform
import shutil
import stat
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 12B")
print("OPTIMIZED FRESH-T4 RECOVERY TO AUTHENTICATED 34/36 STATE")
print("=" * 174)


# ==================================================================================================
# 1. MOUNT GOOGLE DRIVE
# ==================================================================================================

from google.colab import drive

drive.mount(
    "/content/drive",
    force_remount=False,
)


# ==================================================================================================
# 2. EXACT DRIVE PATHS
# ==================================================================================================

DRIVE_LTLC = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

BASE_DRIVE_ZIP = (
    DRIVE_LTLC
    / "migrations"
    / "LTLC_Notebook06A_ExperimentB_28of36_PreLA_Kaggle_to_Colab_SELF_CONTAINED_MIGRATION_v1.zip"
)

DELTA34_DRIVE_ZIP = (
    DRIVE_LTLC
    / "experiment_b_colab_persistence"
    / "LTLC_Notebook06A_ExperimentB_34of36_Post_Qingyun_LA_Tuning_Seed42_FULL_DELTA_FROM_28of36_MIGRATION_v1.zip"
)

DELTA34_EXTERNAL_MANIFEST = (
    DRIVE_LTLC
    / "experiment_b_colab_persistence"
    / "LTLC_Notebook06A_ExperimentB_34of36_Post_Qingyun_LA_Tuning_Seed42_FULL_DELTA_FROM_28of36_MIGRATION_v1__MANIFEST.json"
)

ADAPTER_DIR = (
    DRIVE_LTLC
    / "migrations"
    / "colab_runtime_adapter_v1"
)

BOOTSTRAP_PATH = (
    ADAPTER_DIR
    / "R6_SUBPROCESS_BOOTSTRAP_COLAB_v1.py"
)

LA_PREFIX_PATH = (
    ADAPTER_DIR
    / "R6_LA_RUNTIME_PREFIX_COLAB_v1.py"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 3. LOCAL PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

RECOVERY_TMP = Path(
    "/content/LTLC_RECOVERY_34OF36_TMP"
)

LOCAL_BASE_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

LOCAL_DELTA34_ZIP = Path(
    "/content/LTLC_34of36_delta_v1.zip"
)


# ==================================================================================================
# 4. EXPECTED AUTHENTICATED IDENTITIES
# ==================================================================================================

EXPECTED_BASE_ZIP_SIZE = 358222272

EXPECTED_BASE_ZIP_SHA256 = (
    "8d2798b570eefcc991308455ef5d28cea"
    "3e03c280250ccda0a8a9b356726b363"
)

EXPECTED_BASE_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_DELTA34_ZIP_SHA256 = (
    "75b917f24bd3116bcdaac91897b612c2"
    "2999ea5dfef51a7ece9b0e341acb42b0"
)

EXPECTED_DELTA34_MANIFEST_SHA256 = (
    "9ddbb0d66f030b25499f0518811b1a30"
    "a45f6e8f198ec7acfc843aebac95f061"
)

EXPECTED_BOOTSTRAP_SHA256 = (
    "608d1a883d6366876138ed65f75dd531"
    "f62cc9bd33c514333047842828ff7bc7"
)

EXPECTED_LA_PREFIX_SHA256 = (
    "39d5f2307666a1b220c2d8c4bc474769"
    "7fba78f69c806b3552231781533defae"
)

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "4a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)

EXPECTED_QINGYUN_SELECTION_SHA256 = (
    "367d6ecd113d786bf89510985e0f0ff7"
    "e6589a68db9c912d4c36fb29d02fae89"
)

BASE_PROJECT_FILE_COUNT = 449
DELTA34_FILE_COUNT = 44
RECOVERED_PROJECT_FILE_COUNT = 493

BASE_INTERNAL_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)

DELTA_INTERNAL_MANIFEST = (
    "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
)


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def copy_and_sha256(
    source,
    destination,
    chunk_size=4 * 1024 * 1024,
):

    source = Path(source)
    destination = Path(destination)

    if destination.exists():
        destination.unlink()

    h = hashlib.sha256()
    total = 0

    with source.open("rb") as src, destination.open("wb") as dst:

        while True:

            block = src.read(chunk_size)

            if not block:
                break

            dst.write(block)
            h.update(block)

            total += len(block)

    return total, h.hexdigest()


def validate_relative_path(
    relative_path,
):

    p = PurePosixPath(
        str(relative_path)
    )

    if p.is_absolute():
        raise RuntimeError(
            f"STOP: absolute archive path rejected: {relative_path}"
        )

    if ".." in p.parts:
        raise RuntimeError(
            f"STOP: parent traversal rejected: {relative_path}"
        )

    return str(p)


def zipinfo_is_symlink(
    info,
):

    mode = (
        info.external_attr
        >>
        16
    )

    return stat.S_ISLNK(
        mode
    )


def extract_member_verified(
    zf,
    info,
    destination,
    expected_sha,
    expected_size=None,
    chunk_size=1024 * 1024,
):

    if zipinfo_is_symlink(info):
        raise RuntimeError(
            f"STOP: symlink archive member rejected: {info.filename}"
        )

    destination = Path(
        destination
    )

    destination.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    h = hashlib.sha256()
    total = 0

    with zf.open(
        info,
        mode="r",
    ) as src, destination.open(
        "wb"
    ) as dst:

        while True:

            block = src.read(
                chunk_size
            )

            if not block:
                break

            dst.write(block)
            h.update(block)

            total += len(block)

    actual_sha = h.hexdigest()

    if actual_sha != expected_sha:

        raise RuntimeError(
            "STOP: extracted SHA mismatch:\n"
            f"{destination}"
        )

    if (
        expected_size is not None
        and
        int(total) != int(expected_size)
    ):

        raise RuntimeError(
            "STOP: extracted size mismatch:\n"
            f"{destination}"
        )

    return total


def resolve_project_member(
    infos,
    relative_path,
):

    relative_path = validate_relative_path(
        relative_path
    )

    matches = []

    for info in infos:

        name = info.filename.rstrip("/")

        if (
            name == relative_path
            or
            name.endswith(
                "/"
                +
                relative_path
            )
        ):
            matches.append(
                info
            )

    if len(matches) != 1:

        raise RuntimeError(
            "STOP: expected exactly one archive member for:\n"
            f"{relative_path}\n"
            f"matches={len(matches)}"
        )

    return matches[0]


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. CURRENT T4 ENVIRONMENT — BEFORE ANY RECOVERY
# ==================================================================================================

print()
print("-" * 174)
print("[1] CURRENT T4 ENVIRONMENT")
print("-" * 174)


print(
    "Python version :",
    platform.python_version()
)

print(
    "PyTorch        :",
    torch.__version__
)

print(
    "CUDA build     :",
    torch.version.cuda
)

print(
    "CUDA available :",
    torch.cuda.is_available()
)


if platform.python_version() != "3.12.13":
    raise RuntimeError(
        "STOP: Python version differs from frozen runtime."
    )

if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError(
        "STOP: PyTorch version differs from frozen runtime."
    )

if torch.version.cuda != "12.8":
    raise RuntimeError(
        "STOP: CUDA build differs from frozen runtime."
    )

if not torch.cuda.is_available():
    raise RuntimeError(
        "STOP: CUDA unavailable."
    )

if int(
    torch.backends.cudnn.version()
) != 91900:
    raise RuntimeError(
        "STOP: cuDNN differs from frozen runtime."
    )

if torch.cuda.device_count() != 1:
    raise RuntimeError(
        "STOP: expected exactly one visible GPU."
    )


gpu0 = torch.cuda.get_device_name(
    0
)


if "T4" not in gpu0:
    raise RuntimeError(
        "STOP: visible GPU is not Tesla T4."
    )


print(
    "cuDNN          :",
    torch.backends.cudnn.version()
)

print(
    "Visible GPUs   :",
    torch.cuda.device_count()
)

print(
    "GPU0           :",
    gpu0
)

print()
print(
    "Fresh T4 runtime environment: PASS"
)


# ==================================================================================================
# 7. DRIVE ARTIFACT PRESENCE
# ==================================================================================================

print()
print("-" * 174)
print("[2] REQUIRED GOOGLE-DRIVE ARTIFACTS")
print("-" * 174)


required_drive_files = [
    (
        "28/36 migration ZIP",
        BASE_DRIVE_ZIP,
    ),
    (
        "34/36 cumulative delta ZIP",
        DELTA34_DRIVE_ZIP,
    ),
    (
        "34/36 external manifest",
        DELTA34_EXTERNAL_MANIFEST,
    ),
    (
        "Colab bootstrap",
        BOOTSTRAP_PATH,
    ),
    (
        "Colab LA runtime prefix",
        LA_PREFIX_PATH,
    ),
    (
        "T4 runtime gate",
        T4_GATE_PATH,
    ),
]


for label, path in required_drive_files:

    print(
        f"{label:30s}:",
        "FOUND"
        if path.is_file()
        else "MISSING"
    )

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: required Drive artifact missing:\n{path}"
        )


# ==================================================================================================
# 8. AUTHENTICATE RUNTIME ADAPTERS
# ==================================================================================================

print()
print("-" * 174)
print("[3] COLAB RUNTIME ADAPTER AUTHENTICATION")
print("-" * 174)


adapter_contract = [
    (
        "Bootstrap",
        BOOTSTRAP_PATH,
        EXPECTED_BOOTSTRAP_SHA256,
    ),
    (
        "LA runtime prefix",
        LA_PREFIX_PATH,
        EXPECTED_LA_PREFIX_SHA256,
    ),
    (
        "T4 runtime gate",
        T4_GATE_PATH,
        EXPECTED_T4_GATE_SHA256,
    ),
]


for label, path, expected_sha in adapter_contract:

    actual_sha = sha256_file(
        path
    )

    print(
        f"{label:22s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL"
    )

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP: adapter SHA mismatch: {label}"
        )


# ==================================================================================================
# 9. COPY + AUTHENTICATE 28/36 BASE IN ONE DRIVE READ
# ==================================================================================================

print()
print("-" * 174)
print("[4] AUTHENTICATE 28/36 MIGRATION BASE")
print("-" * 174)


base_size, base_sha = copy_and_sha256(
    BASE_DRIVE_ZIP,
    LOCAL_BASE_ZIP,
)


print(
    "Expected size :",
    EXPECTED_BASE_ZIP_SIZE
)

print(
    "Actual size   :",
    base_size
)

print(
    "Expected SHA  :",
    EXPECTED_BASE_ZIP_SHA256
)

print(
    "Actual SHA    :",
    base_sha
)


if base_size != EXPECTED_BASE_ZIP_SIZE:
    raise RuntimeError(
        "STOP: migration ZIP size mismatch."
    )

if base_sha != EXPECTED_BASE_ZIP_SHA256:
    raise RuntimeError(
        "STOP: migration ZIP SHA mismatch."
    )


print()
print(
    "28/36 migration ZIP authentication: PASS"
)


# ==================================================================================================
# 10. COPY + AUTHENTICATE 34/36 DELTA IN ONE DRIVE READ
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATE 34/36 CUMULATIVE DELTA")
print("-" * 174)


delta_size, delta_sha = copy_and_sha256(
    DELTA34_DRIVE_ZIP,
    LOCAL_DELTA34_ZIP,
)


print(
    "Delta size bytes:",
    delta_size
)

print(
    "Expected SHA    :",
    EXPECTED_DELTA34_ZIP_SHA256
)

print(
    "Actual SHA      :",
    delta_sha
)


if delta_sha != EXPECTED_DELTA34_ZIP_SHA256:
    raise RuntimeError(
        "STOP: 34/36 delta ZIP SHA mismatch."
    )


external_manifest_sha = sha256_file(
    DELTA34_EXTERNAL_MANIFEST
)


print()
print(
    "Expected external manifest SHA:",
    EXPECTED_DELTA34_MANIFEST_SHA256
)

print(
    "Actual external manifest SHA  :",
    external_manifest_sha
)


if (
    external_manifest_sha
    !=
    EXPECTED_DELTA34_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: 34/36 external manifest SHA mismatch."
    )


print()
print(
    "34/36 cumulative delta authentication: PASS"
)


# ==================================================================================================
# 11. READ + AUTHENTICATE INTERNAL MANIFESTS
# ==================================================================================================

print()
print("-" * 174)
print("[6] INTERNAL MANIFEST AUTHENTICATION")
print("-" * 174)


with zipfile.ZipFile(
    LOCAL_BASE_ZIP,
    mode="r",
) as base_zf:

    base_manifest_bytes = base_zf.read(
        BASE_INTERNAL_MANIFEST
    )


base_manifest_sha = hashlib.sha256(
    base_manifest_bytes
).hexdigest()


if (
    base_manifest_sha
    !=
    EXPECTED_BASE_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: internal migration manifest SHA mismatch."
    )


base_manifest = json.loads(
    base_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    base_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(base_records) != BASE_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: migration snapshot is not 449 files."
    )


with zipfile.ZipFile(
    LOCAL_DELTA34_ZIP,
    mode="r",
) as delta_zf:

    delta_manifest_bytes = delta_zf.read(
        DELTA_INTERNAL_MANIFEST
    )


delta_manifest_sha = hashlib.sha256(
    delta_manifest_bytes
).hexdigest()


if (
    delta_manifest_sha
    !=
    EXPECTED_DELTA34_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: internal 34/36 manifest SHA mismatch."
    )


delta_manifest = json.loads(
    delta_manifest_bytes.decode(
        "utf-8"
    )
)


delta_records = (
    delta_manifest[
        "full_delta"
    ][
        "records"
    ]
)


if len(delta_records) != DELTA34_FILE_COUNT:
    raise RuntimeError(
        "STOP: 34/36 cumulative delta is not 44 files."
    )


print(
    "28/36 internal manifest : PASS"
)

print(
    "28/36 project records   : 449 PASS"
)

print(
    "34/36 internal manifest : PASS"
)

print(
    "34/36 delta records     : 44 PASS"
)


# ==================================================================================================
# 12. VALIDATE DELTA PATH SEMANTICS
# ==================================================================================================

print()
print("-" * 174)
print("[7] DELTA PATH SAFETY")
print("-" * 174)


base_relative_paths = {
    validate_relative_path(
        record[
            "relative_path"
        ]
    )
    for record in base_records
}


delta_relative_paths = {
    validate_relative_path(
        record[
            "relative_path"
        ]
    )
    for record in delta_records
}


if len(
    base_relative_paths
) != 449:
    raise RuntimeError(
        "STOP: duplicate base relative paths."
    )


if len(
    delta_relative_paths
) != 44:
    raise RuntimeError(
        "STOP: duplicate delta relative paths."
    )


if (
    base_relative_paths
    &
    delta_relative_paths
):
    raise RuntimeError(
        "STOP: 34/36 delta unexpectedly overlaps the 28/36 base."
    )


if not all(
    path.startswith(
        "runs/experiment_b_spatial_validation_v1/la_loss/"
    )
    for path in delta_relative_paths
):
    raise RuntimeError(
        "STOP: 34/36 delta contains path outside authorized LA-loss run tree."
    )


print(
    "Base unique paths       : 449 PASS"
)

print(
    "Delta unique paths      : 44 PASS"
)

print(
    "Base/delta overlap      : 0 PASS"
)

print(
    "Delta authorized prefix : PASS"
)


# ==================================================================================================
# 13. PREPARE TEMPORARY RECOVERY ROOT
# ==================================================================================================

print()
print("-" * 174)
print("[8] RECONSTRUCT PROJECT")
print("-" * 174)


if RECOVERY_TMP.exists():
    shutil.rmtree(
        RECOVERY_TMP
    )


RECOVERY_TMP.mkdir(
    parents=True,
    exist_ok=False,
)


# ==================================================================================================
# 14. EXTRACT + VERIFY ALL 449 BASE FILES IN ONE PASS
# ==================================================================================================

base_verified = 0


with zipfile.ZipFile(
    LOCAL_BASE_ZIP,
    mode="r",
) as zf:

    infos = [
        info
        for info in zf.infolist()
        if not info.is_dir()
    ]


    for record in base_records:

        relative_path = validate_relative_path(
            record[
                "relative_path"
            ]
        )

        info = resolve_project_member(
            infos,
            relative_path,
        )

        expected_size = record.get(
            "size_bytes"
        )

        extract_member_verified(
            zf,
            info,
            RECOVERY_TMP
            /
            relative_path,
            record[
                "sha256"
            ],
            expected_size,
        )

        base_verified += 1


if base_verified != 449:
    raise RuntimeError(
        "STOP: base extraction did not verify 449/449."
    )


print(
    "Base extraction + SHA verification:",
    "449/449 PASS"
)


# ==================================================================================================
# 15. APPLY + VERIFY ALL 44 CUMULATIVE DELTA FILES
# ==================================================================================================

delta_verified = 0


with zipfile.ZipFile(
    LOCAL_DELTA34_ZIP,
    mode="r",
) as zf:

    names = set(
        zf.namelist()
    )


    for record in delta_records:

        relative_path = validate_relative_path(
            record[
                "relative_path"
            ]
        )

        archive_name = (
            "LTLC_DELTA/"
            +
            relative_path
        )


        if archive_name not in names:
            raise RuntimeError(
                "STOP: delta member missing:\n"
                f"{archive_name}"
            )


        info = zf.getinfo(
            archive_name
        )


        destination = (
            RECOVERY_TMP
            /
            relative_path
        )


        if destination.exists():
            raise RuntimeError(
                "STOP: delta attempted to overwrite base artifact:\n"
                f"{relative_path}"
            )


        extract_member_verified(
            zf,
            info,
            destination,
            record[
                "sha256"
            ],
            record.get(
                "size_bytes"
            ),
        )

        delta_verified += 1


if delta_verified != 44:
    raise RuntimeError(
        "STOP: delta extraction did not verify 44/44."
    )


print(
    "34/36 delta extraction + SHA verification:",
    "44/44 PASS"
)


# ==================================================================================================
# 16. TEMPORARY PROJECT FOOTPRINT
# ==================================================================================================

recovered_tmp_files = [
    p
    for p in RECOVERY_TMP.rglob("*")
    if p.is_file()
]


if (
    len(recovered_tmp_files)
    !=
    RECOVERED_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: recovered temporary project is not 493 files.\n"
        f"Observed: {len(recovered_tmp_files)}"
    )


print(
    "Temporary recovered project:",
    "493 files PASS"
)


# ==================================================================================================
# 17. PROMOTE VERIFIED PROJECT TO /content/LTLC
# ==================================================================================================

if ROOT.exists():

    print(
        "Existing local /content/LTLC detected; "
        "replacing it only after verified reconstruction."
    )

    shutil.rmtree(
        ROOT
    )


os.replace(
    RECOVERY_TMP,
    ROOT,
)


project_files_after_promotion = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    project_files_after_promotion
) != 493:
    raise RuntimeError(
        "STOP: promoted project is not 493 files."
    )


print(
    "Promotion to /content/LTLC:",
    "493 files PASS"
)


# ==================================================================================================
# 18. CRITICAL SCIENTIFIC SOURCE IDENTITIES — BEFORE EXECUTION
# ==================================================================================================

print()
print("-" * 174)
print("[9] RECOVERED FROZEN SCIENTIFIC IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: recovered critical artifact missing: {label}"
        )

    actual_sha = sha256_file(
        path
    )

    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL"
    )

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP: recovered critical SHA mismatch: {label}"
        )


# ==================================================================================================
# 19. EXECUTE AUTHENTICATED COLAB BOOTSTRAP
# ==================================================================================================

print()
print("-" * 174)
print("[10] REHYDRATE SCIENTIFIC RUNTIME")
print("-" * 174)


bootstrap_source = BOOTSTRAP_PATH.read_text(
    encoding="utf-8"
)


exec(
    compile(
        bootstrap_source,
        str(
            BOOTSTRAP_PATH
        ),
        "exec",
    ),
    globals(),
    globals(),
)


print()
print(
    "Colab bootstrap execution: PASS"
)


# ==================================================================================================
# 20. EXECUTE AUTHENTICATED LA RUNTIME PREFIX
# ==================================================================================================

la_prefix_source = LA_PREFIX_PATH.read_text(
    encoding="utf-8"
)


exec(
    compile(
        la_prefix_source,
        str(
            LA_PREFIX_PATH
        ),
        "exec",
    ),
    globals(),
    globals(),
)


print()
print(
    "LA runtime-prefix execution: PASS"
)


# ==================================================================================================
# 21. REQUIRED RUNTIME INTERFACES
# ==================================================================================================

required_interfaces = [
    "ExperimentBLALoss",
    "validate_la_loss",
    "better_la_tau_candidate",
    "select_experiment_b_la_tau",
    "run_experiment_b_la_loss_job",
    "la_completion_path",
    "la_selection_path",
    "set_all_seeds",
    "make_loaders",
    "better_best_aa",
    "better_best_oa",
]


for name in required_interfaces:

    if name not in globals():
        raise RuntimeError(
            f"STOP: recovered runtime interface missing: {name}"
        )


if str(DEVICE) != "cuda:0":
    raise RuntimeError(
        "STOP: recovered scientific DEVICE is not cuda:0."
    )


print(
    "Required scientific interfaces:",
    f"{len(required_interfaces)}/{len(required_interfaces)} PASS"
)

print(
    "Recovered scientific DEVICE:",
    "cuda:0 PASS"
)


# ==================================================================================================
# 22. EXACT 34/36 LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[11] EXACT 34/36 SCIENTIFIC LEDGER")
print("-" * 174)


ledger = {
    "CE":
        count_completion_manifests(
            "ce"
        ),

    "Focal":
        count_completion_manifests(
            "focal"
        ),

    "LDAM-DRW":
        count_completion_manifests(
            "ldam_drw"
        ),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_ledger = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 6,
}


for method, expected in expected_ledger.items():

    observed = ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: recovered ledger mismatch: {method}"
        )


if sum(
    ledger.values()
) != 34:
    raise RuntimeError(
        "STOP: recovered Experiment-B ledger is not 34/36."
    )


print()
print(
    "Experiment-B recovered state:",
    "34/36 PASS"
)


# ==================================================================================================
# 23. PINGAN + QINGYUN FROZEN TAU SELECTIONS
# ==================================================================================================

print()
print("-" * 174)
print("[12] FROZEN LA TAU-SELECTION IDENTITIES")
print("-" * 174)


pingan_selection_path = la_selection_path(
    "Pingan"
)

qingyun_selection_path = la_selection_path(
    "Qingyun"
)


selection_contract = [
    (
        "Pingan",
        pingan_selection_path,
        EXPECTED_PINGAN_SELECTION_SHA256,
        1.0,
    ),
    (
        "Qingyun",
        qingyun_selection_path,
        EXPECTED_QINGYUN_SELECTION_SHA256,
        0.5,
    ),
]


for (
    dataset_name,
    selection_path,
    expected_sha,
    expected_tau,
) in selection_contract:

    if not selection_path.is_file():

        raise FileNotFoundError(
            f"STOP: {dataset_name} selection manifest missing."
        )


    actual_sha = sha256_file(
        selection_path
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: {dataset_name} selection SHA mismatch."
        )


    with selection_path.open(
        "r",
        encoding="utf-8",
    ) as f:

        selection = json.load(
            f
        )


    if (
        float(
            selection[
                "selected_tau"
            ]
        )
        != expected_tau
    ):

        raise RuntimeError(
            f"STOP: {dataset_name} selected tau changed."
        )


    if (
        selection.get(
            "official_test_status"
        )
        != "SEALED"
    ):

        raise RuntimeError(
            f"STOP: {dataset_name} selection does not preserve test seal."
        )


    if (
        selection.get(
            "official_test_values_accessed"
        )
        is not False
    ):

        raise RuntimeError(
            f"STOP: {dataset_name} selection indicates test access."
        )


    if (
        selection.get(
            "official_test_used_for_selection"
        )
        is not False
    ):

        raise RuntimeError(
            f"STOP: {dataset_name} selection used official test."
        )


    print(
        f"{dataset_name:10s}:",
        f"selected tau={expected_tau}",
        "| SHA PASS",
        "| official test SEALED",
    )


# ==================================================================================================
# 24. FINAL TWO TARGETS MUST BE ABSENT
# ==================================================================================================

print()
print("-" * 174)
print("[13] FINAL TWO AUTHORIZED JOB TARGETS")
print("-" * 174)


seed123_completion = la_completion_path(
    "Qingyun",
    0.5,
    123,
)

seed3407_completion = la_completion_path(
    "Qingyun",
    0.5,
    3407,
)


if seed123_completion.exists():

    raise RuntimeError(
        "STOP: Qingyun tau=0.5 seed123 unexpectedly already exists."
    )


if seed3407_completion.exists():

    raise RuntimeError(
        "STOP: Qingyun tau=0.5 seed3407 unexpectedly already exists."
    )


print(
    "Qingyun / tau=0.5 / seed123  : ABSENT PASS"
)

print(
    "Qingyun / tau=0.5 / seed3407 : ABSENT PASS"
)


# ==================================================================================================
# 25. FINAL PROJECT FOOTPRINT
# ==================================================================================================

project_files_final = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(project_files_final)
    != 493
):

    raise RuntimeError(
        "STOP: runtime rehydration unexpectedly changed project footprint.\n"
        f"Observed project files: {len(project_files_final)}"
    )


# ==================================================================================================
# 26. FINAL RECOVERY GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 12B FINAL RECOVERY GATE")
print("=" * 174)

print(
    "28/36 migration ZIP authentication      : PASS"
)

print(
    "34/36 cumulative-delta authentication   : PASS"
)

print(
    "Base extraction verification            : 449/449 PASS"
)

print(
    "Delta extraction verification           : 44/44 PASS"
)

print(
    "Recovered project files                 : 493 PASS"
)

print(
    "Frozen scientific identities            : PASS"
)

print(
    "Authenticated Colab bootstrap            : PASS"
)

print(
    "Authenticated LA runtime                : PASS"
)

print(
    "Scientific device                       : cuda:0 ONLY"
)

print(
    "Experiment-B progress                   : 34/36"
)

print(
    "LA-loss progress                        : 6/8"
)

print(
    "Pingan selected tau                     : 1.0"
)

print(
    "Qingyun selected tau                    : 0.5"
)

print(
    "Qingyun seed123                         : NOT YET RUN"
)

print(
    "Qingyun seed3407                        : NOT YET RUN"
)

print(
    "Training performed in recovery cell     : False"
)

print(
    "Previous completed jobs rerun            : False"
)

print(
    "Official test                           : SEALED"
)

print()
print(
    "NEXT AUTHORIZED GPU JOB:"
)

print(
    "Qingyun / LA-loss / tau=0.5 / seed123"
)

print()
print(
    "FINAL CELL 12B PASS: True"
)

print()
print(
    "STATUS: "
    "AUTHENTICATED_34_OF_36_RECOVERED__"
    "QINGYUN_TAU0P5_FROZEN__"
    "FINAL_TWO_JOBS_READY__"
    "NO_TRAINING_PERFORMED"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# COLAB CELL 13
#
# QINGYUN LA-LOSS tau=0.5 / SEED123
# + IMMEDIATE VERIFIED 35/36 GOOGLE-DRIVE PERSISTENCE
#
# FROZEN CONFIGURATION:
#   Dataset : Qingyun
#   Method  : LA_Loss
#   tau     : 0.5
#   seed    : 123
#   epochs  : 100
#
# EXPECTED AFTER SUCCESS:
#   Experiment-B : 35/36
#   LA-loss      : 7/8
#   Project      : 500 files
#   Cumulative delta from 449-file migration base : 51 files
#
# NEXT AFTER EXTERNAL SAVE ACKNOWLEDGEMENT:
#   Qingyun / LA-loss / tau=0.5 / seed3407
#
# NO RETUNING.
# OFFICIAL TEST REMAINS SEALED.
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import zipfile

import torch


print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 13")
print("QINGYUN LA-LOSS tau=0.5 / SEED123 + IMMEDIATE 35/36 DRIVE PERSISTENCE")
print("=" * 174)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

ADAPTER_DIR = (
    DRIVE_ROOT
    / "migrations"
    / "colab_runtime_adapter_v1"
)

T4_GATE_PATH = (
    ADAPTER_DIR
    / "COLAB_T4_SCIENTIFIC_RUNTIME_GATE_v1.json"
)


# ==================================================================================================
# 2. AUTHENTICATED 34/36 PARENT
# ==================================================================================================

PARENT_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "34of36_Post_Qingyun_LA_Tuning_Seed42_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PARENT_ZIP = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}.zip"
)

PARENT_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PARENT_BASENAME}__MANIFEST.json"
)


EXPECTED_PARENT_ZIP_SHA256 = (
    "75b917f24bd3116bcdaac91897b612c2"
    "2999ea5dfef51a7ece9b0e341acb42b0"
)

EXPECTED_PARENT_MANIFEST_SHA256 = (
    "9ddbb0d66f030b25499f0518811b1a30"
    "a45f6e8f198ec7acfc843aebac95f061"
)


# ==================================================================================================
# 3. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_T4_GATE_SHA256 = (
    "339a2c682d4c944efabfe3a75173b71b"
    "d69ec7eeefebb5f5527c7b3f34d74aa1"
)

EXPECTED_MIGRATION_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_TRAINING_PROTOCOL_SHA256 = (
    "41889867f18cc8a8655083c7906c608f"
    "154e4aae8bdc25da81a5e91c0125a9ec"
)

EXPECTED_CLASS_STATS_SHA256 = (
    "556128fb9ce1f22eb049a2531621b3779"
    "bec7069f7678210c13c1f162d6aea32"
)

EXPECTED_LA_SOURCE_SHA256 = (
    "e80747b1a26266b3adc88d7be2ec3f20"
    "4a3f50bc833546ae7a83484b67cecae9"
)

EXPECTED_LA_FREEZE_SHA256 = (
    "da51c0eddcf28c52e131224f9b5ec206"
    "ca7616dbe26bec16a0a9190b9027205b"
)

EXPECTED_LA_VALIDATOR_SHA256 = (
    "1abfaf0be3687ea09b0b018b23344645"
    "c2bcc7d285c0f88f8bc0ec817b1c2fd0"
)

EXPECTED_LA_RUNNER_SHA256 = (
    "c022bbfe0e9e2424fa4e11b0f27c5f3"
    "e6a7e8f8c91b477530418623d64be9816"
)

EXPECTED_LA_ENGINE_GATE_SHA256 = (
    "887960cf0504573c2c759ff630df9ed96"
    "33ef5d8a2f1b926744493e3adf0e841"
)

EXPECTED_QINGYUN_SELECTION_SHA256 = (
    "367d6ecd113d786bf89510985e0f0ff7"
    "e6589a68db9c912d4c36fb29d02fae89"
)


BASE_PROJECT_FILE_COUNT = 449
PRE_JOB_PROJECT_FILE_COUNT = 493
POST_JOB_PROJECT_FILE_COUNT = 500

PRE_JOB_DELTA_COUNT = 44
POST_JOB_DELTA_COUNT = 51

INTERNAL_MIGRATION_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)


# ==================================================================================================
# 4. NEW 35/36 PERSISTENCE PATHS
# ==================================================================================================

PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "35of36_Post_Qingyun_LA_tau0p5_Seed123_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}.zip"
)

PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__MANIFEST.json"
)

PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{PERSISTENCE_BASENAME}__SHA256.txt"
)


# ==================================================================================================
# 5. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def stream_sha256(
    stream,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    while True:

        block = stream.read(chunk_size)

        if not block:
            break

        h.update(block)

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(
        tmp,
        path,
    )


def count_completion_manifests(
    method_name,
):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_name
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


# ==================================================================================================
# 6. T4 RUNTIME GATE
# ==================================================================================================

print()
print("-" * 174)
print("[1] T4 SCIENTIFIC RUNTIME GATE")
print("-" * 174)


if torch.__version__ != "2.11.0+cu128":
    raise RuntimeError(
        "STOP: PyTorch runtime changed."
    )

if torch.version.cuda != "12.8":
    raise RuntimeError(
        "STOP: CUDA build changed."
    )

if not torch.cuda.is_available():
    raise RuntimeError(
        "STOP: CUDA unavailable."
    )

if int(
    torch.backends.cudnn.version()
) != 91900:
    raise RuntimeError(
        "STOP: cuDNN changed."
    )

if torch.cuda.device_count() != 1:
    raise RuntimeError(
        "STOP: expected exactly one visible GPU."
    )


gpu0 = torch.cuda.get_device_name(
    0
)


if "T4" not in gpu0:
    raise RuntimeError(
        "STOP: GPU0 is not Tesla T4."
    )

if str(DEVICE) != "cuda:0":
    raise RuntimeError(
        "STOP: scientific DEVICE is not cuda:0."
    )

if (
    sha256_file(
        T4_GATE_PATH
    )
    !=
    EXPECTED_T4_GATE_SHA256
):
    raise RuntimeError(
        "STOP: T4 runtime-gate SHA changed."
    )


print("PyTorch           : 2.11.0+cu128 PASS")
print("CUDA              : 12.8 PASS")
print("cuDNN             : 91900 PASS")
print("GPU0              :", gpu0, "PASS")
print("Scientific DEVICE : cuda:0 ONLY PASS")
print("T4 runtime gate   : PASS")


# ==================================================================================================
# 7. AUTHENTICATED 34/36 PARENT
# ==================================================================================================

print()
print("-" * 174)
print("[2] EXTERNALLY SAVED 34/36 PARENT")
print("-" * 174)


if not PARENT_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: saved 34/36 ZIP missing."
    )

if not PARENT_MANIFEST.is_file():
    raise FileNotFoundError(
        "STOP: saved 34/36 manifest missing."
    )


parent_zip_sha = sha256_file(
    PARENT_ZIP
)

parent_manifest_sha = sha256_file(
    PARENT_MANIFEST
)


if (
    parent_zip_sha
    !=
    EXPECTED_PARENT_ZIP_SHA256
):
    raise RuntimeError(
        "STOP: saved 34/36 ZIP SHA changed."
    )

if (
    parent_manifest_sha
    !=
    EXPECTED_PARENT_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: saved 34/36 manifest SHA changed."
    )


print("34/36 ZIP      : PASS")
print("34/36 manifest : PASS")
print("Save boundary  : CONFIRMED")


# ==================================================================================================
# 8. PRE-JOB SCIENTIFIC STATE
# ==================================================================================================

print()
print("-" * 174)
print("[3] PRE-JOB SCIENTIFIC BOUNDARY")
print("-" * 174)


pre_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_pre = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 6,
}


for method, expected in expected_pre.items():

    observed = pre_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: pre-job {method} count changed."
        )


if sum(
    pre_ledger.values()
) != 34:
    raise RuntimeError(
        "STOP: experiment is not at 34/36."
    )


project_files_before = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(project_files_before)
    !=
    PRE_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 493 project files before seed123."
    )


qingyun_selection_path = (
    la_selection_path(
        "Qingyun"
    )
)


if not qingyun_selection_path.is_file():
    raise FileNotFoundError(
        "STOP: frozen Qingyun selection missing."
    )


selection_sha_before = sha256_file(
    qingyun_selection_path
)


if (
    selection_sha_before
    !=
    EXPECTED_QINGYUN_SELECTION_SHA256
):
    raise RuntimeError(
        "STOP: Qingyun selection SHA changed."
    )


with qingyun_selection_path.open(
    "r",
    encoding="utf-8",
) as f:

    selection_disk = json.load(f)


if float(
    selection_disk.get(
        "selected_tau"
    )
) != 0.5:
    raise RuntimeError(
        "STOP: Qingyun selected tau is not 0.5."
    )


if not (
    selection_disk.get(
        "official_test_status"
    ) == "SEALED"
    and
    selection_disk.get(
        "official_test_values_accessed"
    ) is False
    and
    selection_disk.get(
        "official_test_used_for_selection"
    ) is False
):
    raise RuntimeError(
        "STOP: Qingyun selection safety boundary changed."
    )


seed123_completion = (
    la_completion_path(
        "Qingyun",
        0.5,
        123,
    )
)

seed3407_completion = (
    la_completion_path(
        "Qingyun",
        0.5,
        3407,
    )
)


if seed123_completion.exists():
    raise RuntimeError(
        "STOP: Qingyun seed123 already exists."
    )

if seed3407_completion.exists():
    raise RuntimeError(
        "STOP: Qingyun seed3407 unexpectedly exists."
    )


print()
print("Experiment-B           : 34/36 PASS")
print("LA-loss                : 6/8 PASS")
print("Project files          : 493 PASS")
print("Qingyun selected tau   : 0.5 PASS")
print("Seed123 target absent  : PASS")
print("Seed3407 target absent : PASS")


# ==================================================================================================
# 9. FROZEN SOURCE IDENTITIES
# ==================================================================================================

print()
print("-" * 174)
print("[4] FROZEN SCIENTIFIC SOURCE IDENTITIES")
print("-" * 174)


AUDIT_ROOT = (
    ROOT
    / "audit"
    / "notebook06a_spatial_validation"
)


critical_contract = [
    (
        "Training protocol",
        AUDIT_ROOT
        / "experiment_b_training_protocol_v1_frozen.json",
        EXPECTED_TRAINING_PROTOCOL_SHA256,
    ),
    (
        "Class statistics",
        AUDIT_ROOT
        / "experiment_b_training_class_statistics_v1_frozen.json",
        EXPECTED_CLASS_STATS_SHA256,
    ),
    (
        "LA executable",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1.py",
        EXPECTED_LA_SOURCE_SHA256,
    ),
    (
        "LA freeze",
        AUDIT_ROOT
        / "experiment_b_la_loss_executable_implementation_v1_frozen.json",
        EXPECTED_LA_FREEZE_SHA256,
    ),
    (
        "LA validator",
        AUDIT_ROOT
        / "experiment_b_la_loss_validator_runtime_v1.py",
        EXPECTED_LA_VALIDATOR_SHA256,
    ),
    (
        "LA runner",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_runner_runtime_v1.py",
        EXPECTED_LA_RUNNER_SHA256,
    ),
    (
        "LA engine gate",
        AUDIT_ROOT
        / "experiment_b_la_loss_compact_job_engine_pretraining_gate_v1.json",
        EXPECTED_LA_ENGINE_GATE_SHA256,
    ),
]


for label, path, expected_sha in critical_contract:

    actual_sha = sha256_file(
        path
    )

    print(
        f"{label:24s}:",
        "PASS"
        if actual_sha == expected_sha
        else "FAIL"
    )

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP: frozen SHA mismatch: {label}"
        )


# ==================================================================================================
# 10. LOAD ORIGINAL 449-FILE BASE RECORD SET
# ==================================================================================================

print()
print("-" * 174)
print("[5] AUTHENTICATED MIGRATION BASE RECORD SET")
print("-" * 174)


if not LOCAL_MIGRATION_ZIP.is_file():
    raise FileNotFoundError(
        "STOP: local migration ZIP missing."
    )


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    migration_manifest_bytes = zf.read(
        INTERNAL_MIGRATION_MANIFEST
    )


migration_manifest_sha = hashlib.sha256(
    migration_manifest_bytes
).hexdigest()


if (
    migration_manifest_sha
    !=
    EXPECTED_MIGRATION_MANIFEST_SHA256
):
    raise RuntimeError(
        "STOP: migration manifest identity changed."
    )


migration_manifest = json.loads(
    migration_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    migration_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(
    base_records
) != BASE_PROJECT_FILE_COUNT:
    raise RuntimeError(
        "STOP: migration baseline is not 449 files."
    )


base_paths = {
    str(
        PurePosixPath(
            record[
                "relative_path"
            ]
        )
    )
    for record in base_records
}


before_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_before
}


pre_delta_paths = sorted(
    before_paths
    -
    base_paths
)


if (
    len(pre_delta_paths)
    !=
    PRE_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: current 34/36 cumulative delta is not 44 files."
    )


print("Migration manifest       : PASS")
print("Migration base files     : 449")
print("Current cumulative delta : 44 files PASS")


# ==================================================================================================
# 11. START AUTHORIZED GPU JOB
# ==================================================================================================

print()
print("=" * 174)
print("STARTING AUTHORIZED GPU JOB")
print("Qingyun / LA-loss / tau=0.5 / seed123")
print("=" * 174)


training_start = time.time()


completion_result = (
    run_experiment_b_la_loss_job(
        "Qingyun",
        0.5,
        123,
    )
)


training_elapsed = (
    time.time()
    -
    training_start
)


print()
print(
    "Authorized runner returned successfully."
)

print(
    "Runner elapsed seconds:",
    round(
        training_elapsed,
        2,
    )
)


# ==================================================================================================
# 12. COMPLETION AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[6] QINGYUN SEED123 COMPLETION AUDIT")
print("-" * 174)


if not seed123_completion.is_file():
    raise FileNotFoundError(
        "STOP: seed123 completion manifest missing."
    )


with seed123_completion.open(
    "r",
    encoding="utf-8",
) as f:

    completion_disk = json.load(f)


identity_checks = {
    "status":
        completion_disk.get(
            "status"
        ) == "COMPLETE",

    "dataset":
        completion_disk.get(
            "dataset"
        ) == "Qingyun",

    "method":
        completion_disk.get(
            "method"
        ) == "LA_Loss",

    "tau":
        float(
            completion_disk.get(
                "tau"
            )
        ) == 0.5,

    "seed":
        int(
            completion_disk.get(
                "seed"
            )
        ) == 123,

    "epochs":
        int(
            completion_disk.get(
                "epochs_completed"
            )
        ) == 100,
}


for label, passed in identity_checks.items():

    print(
        f"{label:14s}:",
        "PASS"
        if passed
        else "FAIL"
    )


if not all(
    identity_checks.values()
):
    raise RuntimeError(
        "STOP: seed123 completion identity failed."
    )


safety = completion_disk[
    "safety"
]


if not (
    safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    safety.get(
        "official_test_values_accessed"
    ) is False
    and
    safety.get(
        "test_dataset_created"
    ) is False
    and
    safety.get(
        "test_dataloader_created"
    ) is False
    and
    safety.get(
        "test_inference_performed"
    ) is False
):
    raise RuntimeError(
        "STOP: official-test safety gate failed."
    )


completion_files = completion_disk[
    "files"
]


if len(
    completion_files
) != 6:
    raise RuntimeError(
        "STOP: seed123 completion must record six run artifacts."
    )


for role, record in completion_files.items():

    path = (
        ROOT
        /
        record[
            "path"
        ]
    )

    if not path.is_file():
        raise FileNotFoundError(
            f"STOP: seed123 artifact missing: {role}"
        )

    if (
        sha256_file(path)
        !=
        record["sha256"]
    ):
        raise RuntimeError(
            f"STOP: seed123 artifact SHA mismatch: {role}"
        )

    print(
        f"{role:24s}: PASS"
    )


best_aa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_epoch"
    ]
)

best_aa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_AA_metrics"
    ]
)

best_oa_epoch = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_epoch"
    ]
)

best_oa_metrics = (
    completion_disk[
        "checkpoint_policy"
    ][
        "best_OA_metrics"
    ]
)


print()
print(
    "Seed123 Best-AA epoch:",
    best_aa_epoch
)

print(
    "Seed123 Best-AA metrics:"
)

print(
    best_aa_metrics
)

print()
print(
    "Seed123 Best-OA epoch:",
    best_oa_epoch
)

print(
    "Seed123 Best-OA metrics:"
)

print(
    best_oa_metrics
)


# ==================================================================================================
# 13. FROZEN QINGYUN SELECTION IMMUTABILITY
# ==================================================================================================

print()
print("-" * 174)
print("[7] FROZEN QINGYUN TAU-SELECTION IMMUTABILITY")
print("-" * 174)


selection_sha_after = sha256_file(
    qingyun_selection_path
)


if (
    selection_sha_after
    !=
    selection_sha_before
):
    raise RuntimeError(
        "STOP: seed123 modified the Qingyun selection manifest."
    )


print(
    "Selected tau               : 0.5"
)

print(
    "Selection SHA before/after : IDENTICAL PASS"
)

print(
    "Retuning performed         : False"
)


# ==================================================================================================
# 14. POST-JOB LEDGER
# ==================================================================================================

print()
print("-" * 174)
print("[8] POST-JOB SCIENTIFIC LEDGER")
print("-" * 174)


post_ledger = {
    "CE":
        count_completion_manifests("ce"),

    "Focal":
        count_completion_manifests("focal"),

    "LDAM-DRW":
        count_completion_manifests("ldam_drw"),

    "Balanced Softmax":
        count_completion_manifests(
            "balanced_softmax"
        ),

    "LA-loss":
        count_completion_manifests(
            "la_loss"
        ),
}


expected_post = {
    "CE": 6,
    "Focal": 8,
    "LDAM-DRW": 8,
    "Balanced Softmax": 6,
    "LA-loss": 7,
}


for method, expected in expected_post.items():

    observed = post_ledger[
        method
    ]

    print(
        f"{method:18s}: {observed}"
    )

    if observed != expected:
        raise RuntimeError(
            f"STOP: post-job {method} count changed."
        )


if sum(
    post_ledger.values()
) != 35:
    raise RuntimeError(
        "STOP: Experiment-B state is not 35/36."
    )


print()
print(
    "Experiment-B progress : 35/36 PASS"
)

print(
    "LA-loss progress       : 7/8 PASS"
)


# ==================================================================================================
# 15. PROJECT FOOTPRINT AUDIT
# ==================================================================================================

print()
print("-" * 174)
print("[9] 35/36 PROJECT-FOOTPRINT AUDIT")
print("-" * 174)


project_files_after = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


if (
    len(project_files_after)
    !=
    POST_JOB_PROJECT_FILE_COUNT
):
    raise RuntimeError(
        "STOP: expected 500 project files after seed123.\n"
        f"Observed: {len(project_files_after)}"
    )


after_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in project_files_after
}


new_since_34 = sorted(
    after_paths
    -
    before_paths
)


if len(
    new_since_34
) != 7:
    raise RuntimeError(
        "STOP: Qingyun seed123 must add exactly seven files."
    )


expected_seed123_prefix = (
    "runs/experiment_b_spatial_validation_v1/"
    "la_loss/qingyun/tau0p5/seed123/"
)


if not all(
    path.startswith(
        expected_seed123_prefix
    )
    for path in new_since_34
):
    raise RuntimeError(
        "STOP: seed123 added an unauthorized project path."
    )


print(
    "New seed123 files:",
    len(
        new_since_34
    ),
    "PASS"
)


for path in new_since_34:
    print(
        " +",
        path
    )


cumulative_delta_paths = sorted(
    after_paths
    -
    base_paths
)


if (
    len(
        cumulative_delta_paths
    )
    !=
    POST_JOB_DELTA_COUNT
):
    raise RuntimeError(
        "STOP: expected 51-file cumulative delta at 35/36."
    )


print()
print(
    "Migration base files   : 449"
)

print(
    "Current project files  : 500"
)

print(
    "Cumulative delta files : 51 PASS"
)


# ==================================================================================================
# 16. HASH ONLY THE 51 DELTA FILES
# ==================================================================================================

delta_records = []


for relative_path in cumulative_delta_paths:

    path = (
        ROOT
        /
        relative_path
    )

    delta_records.append(
        {
            "relative_path":
                relative_path,

            "size_bytes":
                int(
                    path.stat().st_size
                ),

            "sha256":
                sha256_file(
                    path
                ),
        }
    )


completion_sha = sha256_file(
    seed123_completion
)


# ==================================================================================================
# 17. BUILD 35/36 PERSISTENCE MANIFEST
# ==================================================================================================

persistence_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_COLAB_FULL_DELTA_FROM_AUTHENTICATED_28OF36_MIGRATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent_chain": {
        "previous_progress":
            "34/36",

        "previous_persistence_zip_sha256":
            parent_zip_sha,

        "previous_persistence_manifest_sha256":
            parent_manifest_sha,

        "base_project_file_count":
            449,

        "t4_runtime_gate_sha256":
            EXPECTED_T4_GATE_SHA256,
    },

    "scientific_progress": {
        "before":
            34,

        "after":
            35,

        "frozen_total":
            36,

        "ce":
            6,

        "focal":
            8,

        "ldam_drw":
            8,

        "balanced_softmax":
            6,

        "la_loss":
            7,
    },

    "completed_job": {
        "dataset":
            "Qingyun",

        "method":
            "LA_Loss",

        "tau":
            0.5,

        "seed":
            123,

        "epochs":
            100,

        "completion_manifest_relative_path":
            str(
                seed123_completion.relative_to(
                    ROOT
                )
            ),

        "completion_manifest_sha256":
            completion_sha,

        "best_AA_epoch":
            int(
                best_aa_epoch
            ),

        "best_AA_metrics":
            best_aa_metrics,

        "best_OA_epoch":
            int(
                best_oa_epoch
            ),

        "best_OA_metrics":
            best_oa_metrics,
    },

    "frozen_qingyun_selection": {
        "selected_tau":
            0.5,

        "selection_manifest_relative_path":
            str(
                qingyun_selection_path.relative_to(
                    ROOT
                )
            ),

        "selection_manifest_sha256":
            selection_sha_after,

        "unchanged_during_seed123":
            True,
    },

    "runtime": {
        "python":
            "3.12.13",

        "torch":
            torch.__version__,

        "cuda":
            torch.version.cuda,

        "cudnn":
            int(
                torch.backends.cudnn.version()
            ),

        "gpu0":
            torch.cuda.get_device_name(
                0
            ),

        "device":
            "cuda:0",

        "mixed_precision":
            False,

        "multi_gpu":
            False,
    },

    "full_delta": {
        "base_project_file_count":
            449,

        "delta_file_count":
            len(
                delta_records
            ),

        "resulting_project_file_count":
            len(
                project_files_after
            ),

        "records":
            delta_records,
    },

    "safety": {
        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,

        "retuning_performed":
            False,
    },

    "next_authorized_job_after_external_save_acknowledgement": {
        "dataset":
            "Qingyun",

        "method":
            "LA_Loss",

        "tau":
            0.5,

        "seed":
            3407,
    },
}


manifest_bytes = json.dumps(
    persistence_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 18. IMMEDIATE 35/36 GOOGLE-DRIVE PERSISTENCE
# ==================================================================================================

print()
print("-" * 174)
print("[10] IMMEDIATE 35/36 GOOGLE-DRIVE PERSISTENCE")
print("-" * 174)


PERSISTENCE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    PERSISTENCE_ZIP,
    PERSISTENCE_MANIFEST,
    PERSISTENCE_SHA,
]:

    if path.exists():
        raise RuntimeError(
            "STOP: 35/36 persistence artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    PERSISTENCE_MANIFEST,
    manifest_bytes,
)


manifest_sha = sha256_file(
    PERSISTENCE_MANIFEST
)


# ==================================================================================================
# 19. CREATE 51-FILE CUMULATIVE DELTA ZIP
# ==================================================================================================

zip_tmp = Path(
    str(
        PERSISTENCE_ZIP
    )
    +
    ".tmp"
)


if zip_tmp.exists():
    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_STORED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json",
        manifest_bytes,
    )


    for record in delta_records:

        relative_path = record[
            "relative_path"
        ]

        source = (
            ROOT
            /
            relative_path
        )


        if (
            sha256_file(
                source
            )
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: delta artifact changed before archive:\n"
                f"{relative_path}"
            )


        zf.write(
            source,
            arcname=(
                "LTLC_DELTA/"
                +
                relative_path
            ),
        )


os.replace(
    zip_tmp,
    PERSISTENCE_ZIP,
)


# ==================================================================================================
# 20. VERIFY DRIVE ZIP
# ==================================================================================================

record_map = {
    (
        "LTLC_DELTA/"
        +
        record[
            "relative_path"
        ]
    ):
        record

    for record in delta_records
}


verified = 0


with zipfile.ZipFile(
    PERSISTENCE_ZIP,
    mode="r",
) as zf:

    expected_members = (
        set(
            record_map.keys()
        )
        |
        {
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        }
    )


    observed_members = set(
        zf.namelist()
    )


    if observed_members != expected_members:
        raise RuntimeError(
            "STOP: 35/36 persistence ZIP member-set mismatch."
        )


    if (
        zf.read(
            "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
        )
        !=
        manifest_bytes
    ):
        raise RuntimeError(
            "STOP: archived 35/36 manifest differs."
        )


    for archive_name, record in record_map.items():

        with zf.open(
            archive_name,
            mode="r",
        ) as stream:

            archived_sha = stream_sha256(
                stream
            )


        if (
            archived_sha
            !=
            record[
                "sha256"
            ]
        ):
            raise RuntimeError(
                "STOP: archived delta SHA mismatch:\n"
                f"{archive_name}"
            )


        verified += 1


if verified != POST_JOB_DELTA_COUNT:
    raise RuntimeError(
        "STOP: expected 51/51 delta verification."
    )


persistence_zip_sha = sha256_file(
    PERSISTENCE_ZIP
)

persistence_zip_size = int(
    PERSISTENCE_ZIP.stat().st_size
)


# ==================================================================================================
# 21. SHA SIDECAR
# ==================================================================================================

sidecar_text = (
    f"{persistence_zip_sha}  {PERSISTENCE_ZIP.name}\n"
    f"{manifest_sha}  {PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    PERSISTENCE_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 22. FINAL GATE
# ==================================================================================================

print()
print("=" * 174)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — COLAB CELL 13 FINAL GATE")
print("=" * 174)

print(
    "Authorized job                           : Qingyun / LA / tau=0.5 / seed123"
)

print(
    "Epochs                                   : 100/100 PASS"
)

print(
    "Independent checkpoint reproduction      : PASS"
)

print(
    "Frozen Qingyun selected tau              : 0.5"
)

print(
    "Tau selection manifest unchanged         : PASS"
)

print(
    "Retuning performed                       : False"
)

print(
    "Experiment-B progress                    : 35/36"
)

print(
    "LA-loss progress                         : 7/8"
)

print(
    "Original migration base                  : 449 files"
)

print(
    "Current project                          : 500 files"
)

print(
    "Cumulative post-migration delta          : 51 files PASS"
)

print(
    "Drive delta SHA verification             : 51/51 PASS"
)

print(
    "Scientific protocol changed              : False"
)

print(
    "Mixed precision                          : False"
)

print(
    "Multi-GPU                                : False"
)

print(
    "Scientific device                        : cuda:0 ONLY"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "SEED123 BEST-AA EPOCH:"
)

print(
    best_aa_epoch
)

print(
    "SEED123 BEST-AA METRICS:"
)

print(
    best_aa_metrics
)

print()
print(
    "SEED123 BEST-OA EPOCH:"
)

print(
    best_oa_epoch
)

print(
    "SEED123 BEST-OA METRICS:"
)

print(
    best_oa_metrics
)

print()
print(
    "TRAINING ELAPSED SECONDS:"
)

print(
    round(
        training_elapsed,
        2,
    )
)

print()
print(
    "35/36 PERSISTENCE ZIP:"
)

print(
    PERSISTENCE_ZIP
)

print()
print(
    "ZIP SIZE BYTES:"
)

print(
    persistence_zip_size
)

print()
print(
    "ZIP SHA-256:"
)

print(
    persistence_zip_sha
)

print()
print(
    "PERSISTENCE MANIFEST:"
)

print(
    PERSISTENCE_MANIFEST
)

print()
print(
    "MANIFEST SHA-256:"
)

print(
    manifest_sha
)

print()
print(
    "SHA SIDECAR:"
)

print(
    PERSISTENCE_SHA
)

print()
print(
    "FINAL COLAB CELL 13 PASS: True"
)

print()
print(
    "STATUS: "
    "QINGYUN_LA_TAU0P5_SEED123_COMPLETE__"
    "35_OF_36__"
    "51_FILE_CUMULATIVE_DELTA_VERIFIED_IN_GOOGLE_DRIVE__"
    "SAVE_ACKNOWLEDGEMENT_REQUIRED_BEFORE_FINAL_SEED3407"
)

print("=" * 174)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CELL 15
#
# FINAL 36/36 CLOSING AUDIT + EXACT SNAPSHOT VERIFICATION + EXTERNAL SEAL
#
# THIS CELL:
#   1. authenticates the externally saved 36/36 persistence package
#   2. reconstructs the exact expected 507-file path/hash contract from:
#        - authenticated 449-file migration base
#        - authenticated 58-file final cumulative delta
#   3. verifies CURRENT /content/LTLC = exact 507/507 state
#   4. verifies completion ledger = CE6 + Focal8 + LDAM8 + BS6 + LA8 = 36/36
#   5. verifies all completion manifests are COMPLETE
#   6. verifies the exact 8-run LA matrix
#   7. verifies Pingan tau=1.0 and Qingyun tau=0.5 frozen selections
#   8. scans project JSON for explicit official-test safety violations
#   9. verifies final 36/36 persistence says no GPU training remains
#  10. writes FINAL EXTERNAL closing-audit JSON + run-ledger CSV + SHA sidecar
#
# IMPORTANT:
#   - NO TRAINING
#   - NO MODEL INFERENCE
#   - NO OFFICIAL-TEST ACCESS
#   - NO PROJECT MUTATION
#   - AUDIT OUTPUTS ARE WRITTEN OUTSIDE /content/LTLC
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import csv
import hashlib
import io
import json
import os
import zipfile


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 15")
print("FINAL 36/36 CLOSING AUDIT + EXACT SNAPSHOT VERIFICATION + EXPERIMENT-B SEAL")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

FINAL_PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "36of36_Post_Qingyun_LA_tau0p5_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

FINAL_PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}.zip"
)

FINAL_PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__MANIFEST.json"
)

FINAL_PERSISTENCE_SHA_SIDECAR = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__SHA256.txt"
)


FINAL_AUDIT_DIR = (
    DRIVE_ROOT
    / "final_audits"
    / "notebook06a_experiment_b"
)

FINAL_AUDIT_JSON = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1.json"
)

FINAL_RUN_LEDGER_CSV = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_RUN_LEDGER_v1.csv"
)

FINAL_AUDIT_SHA = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1__SHA256.txt"
)


# ==================================================================================================
# 2. EXACT FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_FINAL_ZIP_SHA256 = (
    "30ba228994f0a16e756c5ade22480b00"
    "f3e2304ee3e15bd8e00409d86dc2b7b8"
)

EXPECTED_FINAL_MANIFEST_SHA256 = (
    "5f93f457f3b1510d72b366aa8a78aa65"
    "fbf59e866dadfcefd7388d80f81c79b9"
)

EXPECTED_BASE_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)

EXPECTED_QINGYUN_SELECTION_SHA256 = (
    "367d6ecd113d786bf89510985e0f0ff7"
    "e6589a68db9c912d4c36fb29d02fae89"
)

EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256 = (
    "1e9ac68021a9d01d59818c7061120bc9"
    "d511e1957f02b7a96ef3eb442933524d"
)

EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256 = (
    "d99a05fa4c349267c542fae7a8007024"
    "cdcbb2e773ad9d817909ac43d538bcd7"
)

BASE_PROJECT_FILE_COUNT = 449
FINAL_DELTA_FILE_COUNT = 58
FINAL_PROJECT_FILE_COUNT = 507

BASE_INTERNAL_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)

DELTA_INTERNAL_MANIFEST = (
    "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
)


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def atomic_write_bytes(path, payload):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(
        tmp,
        path,
    )


def canonical_rel(path):

    p = PurePosixPath(
        str(path)
    )

    if p.is_absolute():
        raise RuntimeError(
            f"STOP: absolute relative path rejected: {path}"
        )

    if ".." in p.parts:
        raise RuntimeError(
            f"STOP: path traversal rejected: {path}"
        )

    return str(p)


def count_completion_manifests(method_dir):

    path = (
        ROOT
        / "runs"
        / "experiment_b_spatial_validation_v1"
        / method_dir
    )

    if not path.is_dir():
        return 0

    return len(
        list(
            path.rglob(
                "*__completion_manifest.json"
            )
        )
    )


FALSE_SAFETY_KEYS = {
    "official_test_used",
    "official_test_used_for_selection",
    "official_test_values_accessed",
    "official_test_accessed",
    "official_test_opened",
    "official_test_may_be_opened",
    "test_dataset_created",
    "test_dataloader_created",
    "test_inference_performed",
}


STATUS_SAFETY_KEYS = {
    "official_test_status",
}


def scan_safety_tree(obj, source_rel, evidence, violations):

    if isinstance(obj, dict):

        for key, value in obj.items():

            key_l = str(key).lower()

            if key_l in FALSE_SAFETY_KEYS:

                evidence.append(
                    {
                        "source":
                            source_rel,

                        "key":
                            key_l,

                        "value":
                            value,
                    }
                )

                if value is not False:

                    violations.append(
                        {
                            "source":
                                source_rel,

                            "key":
                                key_l,

                            "value":
                                value,
                        }
                    )


            elif key_l in STATUS_SAFETY_KEYS:

                evidence.append(
                    {
                        "source":
                            source_rel,

                        "key":
                            key_l,

                        "value":
                            value,
                    }
                )

                if str(value).upper() != "SEALED":

                    violations.append(
                        {
                            "source":
                                source_rel,

                            "key":
                                key_l,

                            "value":
                                value,
                        }
                    )


            scan_safety_tree(
                value,
                source_rel,
                evidence,
                violations,
            )


    elif isinstance(obj, list):

        for item in obj:

            scan_safety_tree(
                item,
                source_rel,
                evidence,
                violations,
            )


# ==================================================================================================
# 4. REQUIRED STATE
# ==================================================================================================

print()
print("-" * 180)
print("[1] REQUIRED FINAL STATE")
print("-" * 180)


required_files = [
    (
        "Local migration ZIP",
        LOCAL_MIGRATION_ZIP,
    ),
    (
        "36/36 persistence ZIP",
        FINAL_PERSISTENCE_ZIP,
    ),
    (
        "36/36 persistence manifest",
        FINAL_PERSISTENCE_MANIFEST,
    ),
    (
        "36/36 SHA sidecar",
        FINAL_PERSISTENCE_SHA_SIDECAR,
    ),
]


for label, path in required_files:

    exists = path.is_file()

    print(
        f"{label:32s}:",
        "FOUND"
        if exists
        else "MISSING"
    )

    if not exists:

        raise FileNotFoundError(
            f"STOP: required final artifact missing:\n{path}"
        )


if not ROOT.is_dir():

    raise FileNotFoundError(
        "STOP: /content/LTLC missing."
    )


# ==================================================================================================
# 5. AUTHENTICATE EXTERNALLY SAVED 36/36 PACKAGE
# ==================================================================================================

print()
print("-" * 180)
print("[2] AUTHENTICATE EXTERNALLY SAVED 36/36 PACKAGE")
print("-" * 180)


final_zip_sha = sha256_file(
    FINAL_PERSISTENCE_ZIP
)

final_manifest_sha = sha256_file(
    FINAL_PERSISTENCE_MANIFEST
)


print(
    "36/36 ZIP SHA      :",
    final_zip_sha,
)

print(
    "36/36 manifest SHA :",
    final_manifest_sha,
)


if final_zip_sha != EXPECTED_FINAL_ZIP_SHA256:

    raise RuntimeError(
        "STOP: final 36/36 ZIP SHA mismatch."
    )


if final_manifest_sha != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: final 36/36 manifest SHA mismatch."
    )


with FINAL_PERSISTENCE_MANIFEST.open(
    "r",
    encoding="utf-8",
) as f:

    final_manifest = json.load(f)


if (
    final_manifest[
        "scientific_progress"
    ][
        "after"
    ]
    != 36
):

    raise RuntimeError(
        "STOP: final persistence does not say 36/36."
    )


if (
    final_manifest[
        "scientific_progress"
    ].get(
        "experiment_b_complete"
    )
    is not True
):

    raise RuntimeError(
        "STOP: final persistence does not mark Experiment B complete."
    )


if (
    final_manifest[
        "next_action"
    ].get(
        "gpu_training_remaining"
    )
    is not False
):

    raise RuntimeError(
        "STOP: final persistence still indicates GPU training remains."
    )


final_safety = final_manifest[
    "safety"
]


if not (
    final_safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    final_safety.get(
        "official_test_values_accessed"
    ) is False
    and
    final_safety.get(
        "test_dataset_created"
    ) is False
    and
    final_safety.get(
        "test_dataloader_created"
    ) is False
    and
    final_safety.get(
        "test_inference_performed"
    ) is False
):

    raise RuntimeError(
        "STOP: final persistence safety boundary failed."
    )


print()
print(
    "Externally saved 36/36 package : AUTHENTICATED PASS"
)

print(
    "Experiment-B complete          : PASS"
)

print(
    "GPU training remaining         : False PASS"
)

print(
    "Official test                  : SEALED PASS"
)


# ==================================================================================================
# 6. AUTHENTICATE BASE + FINAL DELTA MANIFESTS
# ==================================================================================================

print()
print("-" * 180)
print("[3] RECONSTRUCT EXACT 507-FILE SNAPSHOT CONTRACT")
print("-" * 180)


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    mode="r",
) as zf:

    base_manifest_bytes = zf.read(
        BASE_INTERNAL_MANIFEST
    )


base_manifest_sha = hashlib.sha256(
    base_manifest_bytes
).hexdigest()


if base_manifest_sha != EXPECTED_BASE_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: authenticated base manifest SHA mismatch."
    )


base_manifest = json.loads(
    base_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    base_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(base_records) != BASE_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: base snapshot is not 449 files."
    )


with zipfile.ZipFile(
    FINAL_PERSISTENCE_ZIP,
    mode="r",
) as zf:

    delta_manifest_bytes = zf.read(
        DELTA_INTERNAL_MANIFEST
    )


if hashlib.sha256(
    delta_manifest_bytes
).hexdigest() != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: internal final-delta manifest SHA mismatch."
    )


external_manifest_bytes = (
    FINAL_PERSISTENCE_MANIFEST.read_bytes()
)


if delta_manifest_bytes != external_manifest_bytes:

    raise RuntimeError(
        "STOP: internal and external final manifests differ."
    )


delta_records = (
    final_manifest[
        "full_delta"
    ][
        "records"
    ]
)


if len(delta_records) != FINAL_DELTA_FILE_COUNT:

    raise RuntimeError(
        "STOP: final cumulative delta is not 58 files."
    )


base_record_map = {}


for record in base_records:

    rel = canonical_rel(
        record[
            "relative_path"
        ]
    )

    if rel in base_record_map:

        raise RuntimeError(
            f"STOP: duplicate base path: {rel}"
        )

    base_record_map[
        rel
    ] = record


delta_record_map = {}


for record in delta_records:

    rel = canonical_rel(
        record[
            "relative_path"
        ]
    )

    if rel in delta_record_map:

        raise RuntimeError(
            f"STOP: duplicate delta path: {rel}"
        )

    delta_record_map[
        rel
    ] = record


overlap = (
    set(
        base_record_map
    )
    &
    set(
        delta_record_map
    )
)


if overlap:

    raise RuntimeError(
        "STOP: final delta overlaps authenticated base."
    )


expected_record_map = {
    **base_record_map,
    **delta_record_map,
}


if len(expected_record_map) != FINAL_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: expected snapshot contract is not 507 files."
    )


print(
    "Authenticated base records : 449 PASS"
)

print(
    "Final delta records         : 58 PASS"
)

print(
    "Base/delta overlap          : 0 PASS"
)

print(
    "Expected final snapshot     : 507 files PASS"
)


# ==================================================================================================
# 7. EXACT CURRENT PROJECT PATH SET
# ==================================================================================================

print()
print("-" * 180)
print("[4] EXACT CURRENT PROJECT PATH-SET AUDIT")
print("-" * 180)


current_files = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


current_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )
    for p in current_files
}


expected_paths = set(
    expected_record_map
)


missing_paths = sorted(
    expected_paths
    -
    current_paths
)

unexpected_paths = sorted(
    current_paths
    -
    expected_paths
)


if missing_paths:

    raise RuntimeError(
        "STOP: expected project files missing:\n"
        +
        "\n".join(
            missing_paths[:20]
        )
    )


if unexpected_paths:

    raise RuntimeError(
        "STOP: unexpected project files detected:\n"
        +
        "\n".join(
            unexpected_paths[:20]
        )
    )


if len(current_paths) != FINAL_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: current project is not exactly 507 files."
    )


print(
    "Current project files : 507"
)

print(
    "Missing expected files: 0 PASS"
)

print(
    "Unexpected files      : 0 PASS"
)

print(
    "Path-set identity     : EXACT PASS"
)


# ==================================================================================================
# 8. FULL 507/507 HASH VERIFICATION
# ==================================================================================================

print()
print("-" * 180)
print("[5] FULL 507/507 PROJECT SHA-256 VERIFICATION")
print("-" * 180)


verified_files = 0
verified_bytes = 0


for idx, relative_path in enumerate(
    sorted(
        expected_record_map
    ),
    start=1,
):

    record = expected_record_map[
        relative_path
    ]

    path = (
        ROOT
        /
        relative_path
    )


    actual_sha = sha256_file(
        path
    )

    expected_sha = record[
        "sha256"
    ]


    if actual_sha != expected_sha:

        raise RuntimeError(
            "STOP: project file SHA mismatch:\n"
            f"{relative_path}"
        )


    if (
        "size_bytes"
        in record
        and
        int(
            path.stat().st_size
        )
        !=
        int(
            record[
                "size_bytes"
            ]
        )
    ):

        raise RuntimeError(
            "STOP: project file size mismatch:\n"
            f"{relative_path}"
        )


    verified_files += 1
    verified_bytes += int(
        path.stat().st_size
    )


    if (
        idx % 100 == 0
        or
        idx == FINAL_PROJECT_FILE_COUNT
    ):

        print(
            f"Verified {idx:3d}/{FINAL_PROJECT_FILE_COUNT}"
        )


if verified_files != FINAL_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: full project verification incomplete."
    )


print()
print(
    "Full project SHA verification:",
    "507/507 PASS"
)

print(
    "Verified project bytes       :",
    verified_bytes,
)


# ==================================================================================================
# 9. FINAL 36-RUN COMPLETION LEDGER
# ==================================================================================================

print()
print("-" * 180)
print("[6] FINAL 36-RUN COMPLETION LEDGER")
print("-" * 180)


RUN_ROOT = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
)


method_contract = {
    "CE":
        (
            "ce",
            6,
        ),

    "Focal":
        (
            "focal",
            8,
        ),

    "LDAM-DRW":
        (
            "ldam_drw",
            8,
        ),

    "Balanced Softmax":
        (
            "balanced_softmax",
            6,
        ),

    "LA-loss":
        (
            "la_loss",
            8,
        ),
}


completion_paths_by_method = {}


for label, (
    method_dir,
    expected_count,
) in method_contract.items():

    method_path = (
        RUN_ROOT
        /
        method_dir
    )


    paths = sorted(
        method_path.rglob(
            "*__completion_manifest.json"
        )
    )


    completion_paths_by_method[
        label
    ] = paths


    observed_count = len(
        paths
    )


    print(
        f"{label:18s}: "
        f"{observed_count}/{expected_count}",
        "PASS"
        if observed_count == expected_count
        else "FAIL",
    )


    if observed_count != expected_count:

        raise RuntimeError(
            f"STOP: final completion count mismatch: {label}"
        )


all_completion_paths = []


for paths in completion_paths_by_method.values():

    all_completion_paths.extend(
        paths
    )


if len(all_completion_paths) != 36:

    raise RuntimeError(
        "STOP: total completion-manifest count is not 36."
    )


print()
print(
    "Experiment-B completion manifests:",
    "36/36 PASS"
)


# ==================================================================================================
# 10. AUDIT ALL 36 COMPLETION MANIFESTS
# ==================================================================================================

print()
print("-" * 180)
print("[7] ALL 36 COMPLETION-MANIFEST AUDIT")
print("-" * 180)


run_records = []


for path in sorted(
    all_completion_paths
):

    relative_path = str(
        path.relative_to(
            ROOT
        )
    )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        manifest = json.load(
            f
        )


    status = manifest.get(
        "status"
    )


    if status != "COMPLETE":

        raise RuntimeError(
            "STOP: non-COMPLETE run manifest:\n"
            f"{relative_path}\n"
            f"status={status}"
        )


    dataset = manifest.get(
        "dataset"
    )


    if dataset not in {
        "Pingan",
        "Qingyun",
    }:

        raise RuntimeError(
            "STOP: unexpected dataset in completion manifest:\n"
            f"{relative_path}"
        )


    seed = int(
        manifest.get(
            "seed"
        )
    )


    if seed not in {
        42,
        123,
        3407,
    }:

        raise RuntimeError(
            "STOP: unexpected seed in completion manifest:\n"
            f"{relative_path}"
        )


    checkpoint_policy = manifest.get(
        "checkpoint_policy",
        {},
    )


    best_aa_metrics = checkpoint_policy.get(
        "best_AA_metrics",
        {},
    )

    best_oa_metrics = checkpoint_policy.get(
        "best_OA_metrics",
        {},
    )


    method_dir = (
        PurePosixPath(
            relative_path
        ).parts[
            2
        ]
    )


    run_records.append(
        {
            "relative_path":
                relative_path,

            "dataset":
                dataset,

            "method":
                manifest.get(
                    "method",
                    ""
                ),

            "method_dir":
                method_dir,

            "seed":
                seed,

            "status":
                status,

            "tau":
                manifest.get(
                    "tau",
                    ""
                ),

            "gamma":
                manifest.get(
                    "gamma",
                    ""
                ),

            "C":
                manifest.get(
                    "C",
                    manifest.get(
                        "c",
                        ""
                    ),
                ),

            "best_AA_epoch":
                checkpoint_policy.get(
                    "best_AA_epoch",
                    ""
                ),

            "best_AA":
                best_aa_metrics.get(
                    "AA",
                    ""
                ),

            "best_AA_Macro_F1":
                best_aa_metrics.get(
                    "Macro_F1",
                    ""
                ),

            "best_AA_OA":
                best_aa_metrics.get(
                    "OA",
                    ""
                ),

            "best_OA_epoch":
                checkpoint_policy.get(
                    "best_OA_epoch",
                    ""
                ),

            "best_OA":
                best_oa_metrics.get(
                    "OA",
                    ""
                ),

            "completion_sha256":
                sha256_file(
                    path
                ),
        }
    )


if len(
    {
        record[
            "relative_path"
        ]
        for record in run_records
    }
) != 36:

    raise RuntimeError(
        "STOP: duplicate completion-manifest path."
    )


print(
    "Completion status COMPLETE:",
    "36/36 PASS"
)

print(
    "Datasets restricted to Pingan/Qingyun:",
    "PASS"
)

print(
    "Seeds restricted to 42/123/3407:",
    "PASS"
)

print(
    "Unique completion records:",
    "36/36 PASS"
)


# ==================================================================================================
# 11. EXACT LA 8-RUN MATRIX
# ==================================================================================================

print()
print("-" * 180)
print("[8] EXACT LA-LOSS 8-RUN MATRIX")
print("-" * 180)


expected_la_completion_paths = {
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau0p5/seed42/"
        "la_loss__pingan__tau0p5__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau1p0/seed42/"
        "la_loss__pingan__tau1p0__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau1p0/seed123/"
        "la_loss__pingan__tau1p0__seed123__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau1p0/seed3407/"
        "la_loss__pingan__tau1p0__seed3407__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed42/"
        "la_loss__qingyun__tau0p5__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau1p0/seed42/"
        "la_loss__qingyun__tau1p0__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed123/"
        "la_loss__qingyun__tau0p5__seed123__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed3407/"
        "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json"
    ),
}


observed_la_completion_paths = {
    str(
        path.relative_to(
            ROOT
        )
    )
    for path in completion_paths_by_method[
        "LA-loss"
    ]
}


if observed_la_completion_paths != expected_la_completion_paths:

    missing = sorted(
        expected_la_completion_paths
        -
        observed_la_completion_paths
    )

    unexpected = sorted(
        observed_la_completion_paths
        -
        expected_la_completion_paths
    )

    raise RuntimeError(
        "STOP: LA-loss 8-run matrix mismatch.\n"
        f"Missing: {missing}\n"
        f"Unexpected: {unexpected}"
    )


print(
    "Pingan tau=0.5 seed42        : PASS"
)

print(
    "Pingan tau=1.0 seeds42/123/3407 : PASS"
)

print(
    "Qingyun tau=1.0 seed42       : PASS"
)

print(
    "Qingyun tau=0.5 seeds42/123/3407: PASS"
)

print(
    "Exact LA-loss matrix         : 8/8 PASS"
)


# ==================================================================================================
# 12. FROZEN LA SELECTIONS
# ==================================================================================================

print()
print("-" * 180)
print("[9] FROZEN LA TAU-SELECTION SEAL")
print("-" * 180)


PINGAN_SELECTION = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
    / "la_loss"
    / "pingan"
    / "la_loss__pingan__selected_tau_v1.json"
)

QINGYUN_SELECTION = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
    / "la_loss"
    / "qingyun"
    / "la_loss__qingyun__selected_tau_v1.json"
)


selection_contract = [
    (
        "Pingan",
        PINGAN_SELECTION,
        EXPECTED_PINGAN_SELECTION_SHA256,
        1.0,
    ),
    (
        "Qingyun",
        QINGYUN_SELECTION,
        EXPECTED_QINGYUN_SELECTION_SHA256,
        0.5,
    ),
]


selection_summary = {}


for (
    dataset,
    path,
    expected_sha,
    expected_tau,
) in selection_contract:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: {dataset} selection manifest missing."
        )


    actual_sha = sha256_file(
        path
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: {dataset} selection SHA mismatch."
        )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        selection = json.load(
            f
        )


    selected_tau = float(
        selection[
            "selected_tau"
        ]
    )


    if selected_tau != expected_tau:

        raise RuntimeError(
            f"STOP: {dataset} selected tau changed."
        )


    if not (
        selection.get(
            "official_test_status"
        ) == "SEALED"
        and
        selection.get(
            "official_test_values_accessed"
        ) is False
        and
        selection.get(
            "official_test_used_for_selection"
        ) is False
    ):

        raise RuntimeError(
            f"STOP: {dataset} selection safety boundary failed."
        )


    selection_summary[
        dataset
    ] = {
        "selected_tau":
            selected_tau,

        "selection_sha256":
            actual_sha,
    }


    print(
        f"{dataset:10s}: "
        f"tau={selected_tau} | SHA PASS | OFFICIAL TEST SEALED"
    )


# ==================================================================================================
# 13. FINAL TWO COMPLETION IDENTITIES
# ==================================================================================================

print()
print("-" * 180)
print("[10] FINAL QINGYUN SELECTED-TAU RUN IDENTITIES")
print("-" * 180)


qingyun_seed123_completion = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed123"
    / "la_loss__qingyun__tau0p5__seed123__completion_manifest.json"
)

qingyun_seed3407_completion = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed3407"
    / "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json"
)


if (
    sha256_file(
        qingyun_seed123_completion
    )
    !=
    EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed123 completion SHA changed."
    )


if (
    sha256_file(
        qingyun_seed3407_completion
    )
    !=
    EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed3407 completion SHA changed."
    )


print(
    "Qingyun tau=0.5 seed123 completion : AUTHENTICATED PASS"
)

print(
    "Qingyun tau=0.5 seed3407 completion: AUTHENTICATED PASS"
)


# ==================================================================================================
# 14. PROJECT-WIDE EXPLICIT OFFICIAL-TEST SAFETY SCAN
# ==================================================================================================

print()
print("-" * 180)
print("[11] PROJECT-WIDE OFFICIAL-TEST SAFETY SCAN")
print("-" * 180)


safety_evidence = []
safety_violations = []
json_files_scanned = 0


for path in sorted(
    ROOT.rglob(
        "*.json"
    )
):

    relative_path = str(
        path.relative_to(
            ROOT
        )
    )


    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as f:

            payload = json.load(
                f
            )

    except Exception as exc:

        raise RuntimeError(
            "STOP: project JSON could not be parsed:\n"
            f"{relative_path}\n"
            f"{type(exc).__name__}: {exc}"
        )


    json_files_scanned += 1


    scan_safety_tree(
        payload,
        relative_path,
        safety_evidence,
        safety_violations,
    )


if safety_violations:

    print(
        json.dumps(
            safety_violations[:20],
            indent=2,
            sort_keys=True,
        )
    )

    raise RuntimeError(
        "STOP: explicit official-test safety violation detected."
    )


if len(
    safety_evidence
) == 0:

    raise RuntimeError(
        "STOP: no explicit official-test safety evidence found."
    )


print(
    "JSON files scanned        :",
    json_files_scanned
)

print(
    "Explicit safety assertions:",
    len(
        safety_evidence
    )
)

print(
    "Explicit violations       : 0 PASS"
)

print(
    "Official test             : SEALED PASS"
)


# ==================================================================================================
# 15. BUILD EXTERNAL FINAL RUN-LEDGER CSV
# ==================================================================================================

print()
print("-" * 180)
print("[12] BUILD FINAL EXTERNAL RUN LEDGER")
print("-" * 180)


ledger_buffer = io.StringIO()


ledger_fields = [
    "relative_path",
    "dataset",
    "method",
    "method_dir",
    "seed",
    "status",
    "tau",
    "gamma",
    "C",
    "best_AA_epoch",
    "best_AA",
    "best_AA_Macro_F1",
    "best_AA_OA",
    "best_OA_epoch",
    "best_OA",
    "completion_sha256",
]


writer = csv.DictWriter(
    ledger_buffer,
    fieldnames=ledger_fields,
    lineterminator="\n",
)


writer.writeheader()


for record in sorted(
    run_records,
    key=lambda r: (
        r[
            "method_dir"
        ],
        r[
            "dataset"
        ],
        r[
            "seed"
        ],
        r[
            "relative_path"
        ],
    ),
):

    writer.writerow(
        {
            key:
                record.get(
                    key,
                    ""
                )
            for key in ledger_fields
        }
    )


ledger_bytes = ledger_buffer.getvalue().encode(
    "utf-8"
)

ledger_sha = hashlib.sha256(
    ledger_bytes
).hexdigest()


print(
    "Run-ledger rows:",
    len(
        run_records
    ),
    "PASS"
)

print(
    "Run-ledger SHA-256:",
    ledger_sha
)


# ==================================================================================================
# 16. BUILD FINAL EXTERNAL CLOSING-AUDIT RECORD
# ==================================================================================================

print()
print("-" * 180)
print("[13] BUILD FINAL CLOSING-AUDIT SEAL")
print("-" * 180)


method_counts = {
    label:
        len(
            completion_paths_by_method[
                label
            ]
        )
    for label in method_contract
}


audit_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_NOTEBOOK06A_EXPERIMENT_B_FINAL_CLOSING_AUDIT",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "final_state": {
        "experiment_b_progress":
            "36/36",

        "experiment_b_complete":
            True,

        "gpu_training_remaining":
            False,

        "project_file_count":
            FINAL_PROJECT_FILE_COUNT,

        "base_project_file_count":
            BASE_PROJECT_FILE_COUNT,

        "final_cumulative_delta_file_count":
            FINAL_DELTA_FILE_COUNT,
    },

    "snapshot_authentication": {
        "final_36of36_persistence_zip":
            str(
                FINAL_PERSISTENCE_ZIP
            ),

        "final_36of36_persistence_zip_sha256":
            final_zip_sha,

        "final_36of36_persistence_manifest":
            str(
                FINAL_PERSISTENCE_MANIFEST
            ),

        "final_36of36_persistence_manifest_sha256":
            final_manifest_sha,

        "base_internal_manifest_sha256":
            base_manifest_sha,

        "full_current_project_sha_verification":
            "507/507 PASS",

        "missing_expected_paths":
            0,

        "unexpected_paths":
            0,
    },

    "completion_ledger": {
        "CE":
            method_counts[
                "CE"
            ],

        "Focal":
            method_counts[
                "Focal"
            ],

        "LDAM-DRW":
            method_counts[
                "LDAM-DRW"
            ],

        "Balanced Softmax":
            method_counts[
                "Balanced Softmax"
            ],

        "LA-loss":
            method_counts[
                "LA-loss"
            ],

        "total":
            len(
                all_completion_paths
            ),
    },

    "la_loss": {
        "run_matrix":
            "8/8 EXACT PASS",

        "Pingan":
            selection_summary[
                "Pingan"
            ],

        "Qingyun":
            selection_summary[
                "Qingyun"
            ],

        "qingyun_seed123_completion_sha256":
            EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256,

        "qingyun_seed3407_completion_sha256":
            EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256,
    },

    "official_test": {
        "status":
            "SEALED",

        "project_json_files_scanned":
            json_files_scanned,

        "explicit_safety_assertions":
            len(
                safety_evidence
            ),

        "explicit_safety_violations":
            0,

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,
    },

    "protocol_integrity": {
        "training_performed_in_closing_audit":
            False,

        "model_inference_performed_in_closing_audit":
            False,

        "project_mutated_by_closing_audit":
            False,

        "retuning_performed":
            False,

        "mixed_precision_changed":
            False,

        "multi_gpu_used":
            False,
    },

    "external_run_ledger": {
        "file":
            str(
                FINAL_RUN_LEDGER_CSV
            ),

        "rows":
            len(
                run_records
            ),

        "sha256":
            ledger_sha,
    },

    "closing_status":
        "PASS",

    "seal":
        (
            "EXPERIMENT_B_36_OF_36_COMPLETE__"
            "FULL_507_FILE_SNAPSHOT_VERIFIED__"
            "36_COMPLETION_RECORDS_AUDITED__"
            "LA_SELECTIONS_FROZEN__"
            "OFFICIAL_TEST_SEALED__"
            "NO_GPU_TRAINING_REMAINING"
        ),
}


audit_bytes = json.dumps(
    audit_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


audit_sha = hashlib.sha256(
    audit_bytes
).hexdigest()


print(
    "Closing-audit SHA-256:",
    audit_sha
)


# ==================================================================================================
# 17. WRITE EXTERNAL AUDIT PACKAGE — OUTSIDE SCIENTIFIC PROJECT ROOT
# ==================================================================================================

print()
print("-" * 180)
print("[14] WRITE FINAL EXTERNAL AUDIT PACKAGE TO GOOGLE DRIVE")
print("-" * 180)


FINAL_AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for final_path in [
    FINAL_AUDIT_JSON,
    FINAL_RUN_LEDGER_CSV,
    FINAL_AUDIT_SHA,
]:

    if final_path.exists():

        raise RuntimeError(
            "STOP: final closing-audit artifact already exists:\n"
            f"{final_path}"
        )


atomic_write_bytes(
    FINAL_RUN_LEDGER_CSV,
    ledger_bytes,
)

atomic_write_bytes(
    FINAL_AUDIT_JSON,
    audit_bytes,
)


sidecar_text = (
    f"{audit_sha}  {FINAL_AUDIT_JSON.name}\n"
    f"{ledger_sha}  {FINAL_RUN_LEDGER_CSV.name}\n"
    f"{final_zip_sha}  {FINAL_PERSISTENCE_ZIP.name}\n"
    f"{final_manifest_sha}  {FINAL_PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    FINAL_AUDIT_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 18. VERIFY EXTERNAL AUDIT PACKAGE
# ==================================================================================================

if (
    sha256_file(
        FINAL_AUDIT_JSON
    )
    != audit_sha
):

    raise RuntimeError(
        "STOP: written final audit SHA verification failed."
    )


if (
    sha256_file(
        FINAL_RUN_LEDGER_CSV
    )
    != ledger_sha
):

    raise RuntimeError(
        "STOP: written run-ledger SHA verification failed."
    )


if not FINAL_AUDIT_SHA.is_file():

    raise RuntimeError(
        "STOP: closing-audit SHA sidecar missing."
    )


# ==================================================================================================
# 19. PROVE CLOSING AUDIT DID NOT MUTATE /content/LTLC
# ==================================================================================================

project_files_after_audit = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(
        project_files_after_audit
    )
    !=
    FINAL_PROJECT_FILE_COUNT
):

    raise RuntimeError(
        "STOP: closing audit mutated scientific project footprint."
    )


# ==================================================================================================
# 20. FINAL TERMINAL GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 15 FINAL TERMINAL GATE")
print("=" * 180)

print(
    "Externally saved 36/36 persistence       : AUTHENTICATED PASS"
)

print(
    "Authenticated base snapshot              : 449 files"
)

print(
    "Final cumulative delta                   : 58 files"
)

print(
    "Exact final scientific project           : 507 files"
)

print(
    "Full project SHA verification            : 507/507 PASS"
)

print(
    "Missing expected project files           : 0"
)

print(
    "Unexpected project files                 : 0"
)

print()
print(
    "CE                                       : 6/6 PASS"
)

print(
    "Focal                                    : 8/8 PASS"
)

print(
    "LDAM-DRW                                 : 8/8 PASS"
)

print(
    "Balanced Softmax                         : 6/6 PASS"
)

print(
    "LA-loss                                  : 8/8 PASS"
)

print(
    "Experiment-B                             : 36/36 COMPLETE"
)

print()
print(
    "Pingan frozen selected tau               : 1.0 PASS"
)

print(
    "Qingyun frozen selected tau              : 0.5 PASS"
)

print(
    "Exact LA-loss run matrix                 : 8/8 PASS"
)

print()
print(
    "Project-wide JSON safety violations      : 0 PASS"
)

print(
    "Official test                            : SEALED"
)

print(
    "Official-test inference                  : NOT PERFORMED"
)

print(
    "Training in closing audit                : False"
)

print(
    "GPU training remaining                   : False"
)

print(
    "Scientific project mutated by audit      : False"
)

print()
print(
    "FINAL CLOSING AUDIT:"
)

print(
    FINAL_AUDIT_JSON
)

print()
print(
    "FINAL CLOSING AUDIT SHA-256:"
)

print(
    audit_sha
)

print()
print(
    "FINAL 36-RUN LEDGER:"
)

print(
    FINAL_RUN_LEDGER_CSV
)

print()
print(
    "FINAL 36-RUN LEDGER SHA-256:"
)

print(
    ledger_sha
)

print()
print(
    "FINAL AUDIT SHA SIDECAR:"
)

print(
    FINAL_AUDIT_SHA
)

print()
print(
    "FINAL CELL 15 PASS: True"
)

print()
print(
    "STATUS: "
    "EXPERIMENT_B_36_OF_36_COMPLETE__"
    "FULL_507_FILE_SNAPSHOT_VERIFIED__"
    "36_COMPLETION_RECORDS_AUDITED__"
    "LA_SELECTIONS_FROZEN__"
    "OFFICIAL_TEST_SEALED__"
    "NO_GPU_TRAINING_REMAINING__"
    "FINAL_CLOSING_AUDIT_WRITTEN_TO_GOOGLE_DRIVE"
)

print("=" * 180)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CORRECTION CELL 15A
#
# CORRECTION OF CELL 15 AUDIT-ONLY DEFECT
#
# ROOT CAUSE:
#   Cell 15 treated completion_manifest["dataset"] as case-sensitive and required
#   exactly "Pingan" or "Qingyun".
#
#   Historical completion manifests may canonically store:
#       "pingan"
#       "qingyun"
#
# CORRECTION:
#   - normalize manifest dataset token case-insensitively
#   - map to canonical display names Pingan/Qingyun
#   - independently cross-check normalized dataset against dataset slug in path
#
# SCIENTIFIC STATE:
#   - NOT MODIFIED
#   - NO TRAINING
#   - NO MODEL INFERENCE
#   - NO RETUNING
#   - OFFICIAL TEST REMAINS SEALED
#
# THIS CELL REPERFORMS THE COMPLETE TERMINAL AUDIT AND WRITES THE FINAL EXTERNAL SEAL.
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import csv
import hashlib
import io
import json
import os
import zipfile


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 15A")
print("CORRECTED FINAL 36/36 CLOSING AUDIT — CANONICAL DATASET NORMALIZATION")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path("/content/LTLC")

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

FINAL_PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "36of36_Post_Qingyun_LA_tau0p5_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

FINAL_PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}.zip"
)

FINAL_PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__MANIFEST.json"
)

FINAL_PERSISTENCE_SHA_SIDECAR = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__SHA256.txt"
)


FINAL_AUDIT_DIR = (
    DRIVE_ROOT
    / "final_audits"
    / "notebook06a_experiment_b"
)

FINAL_AUDIT_JSON = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1.json"
)

FINAL_RUN_LEDGER_CSV = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_RUN_LEDGER_v1.csv"
)

FINAL_AUDIT_SHA = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1__SHA256.txt"
)


# ==================================================================================================
# 2. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_FINAL_ZIP_SHA256 = (
    "30ba228994f0a16e756c5ade22480b00"
    "f3e2304ee3e15bd8e00409d86dc2b7b8"
)

EXPECTED_FINAL_MANIFEST_SHA256 = (
    "5f93f457f3b1510d72b366aa8a78aa65"
    "fbf59e866dadfcefd7388d80f81c79b9"
)

EXPECTED_BASE_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)

EXPECTED_QINGYUN_SELECTION_SHA256 = (
    "367d6ecd113d786bf89510985e0f0ff7"
    "e6589a68db9c912d4c36fb29d02fae89"
)

EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256 = (
    "1e9ac68021a9d01d59818c7061120bc9"
    "d511e1957f02b7a96ef3eb442933524d"
)

EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256 = (
    "d99a05fa4c349267c542fae7a8007024"
    "cdcbb2e773ad9d817909ac43d538bcd7"
)


BASE_PROJECT_FILE_COUNT = 449
FINAL_DELTA_FILE_COUNT = 58
FINAL_PROJECT_FILE_COUNT = 507


BASE_INTERNAL_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)

DELTA_INTERNAL_MANIFEST = (
    "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
)


DATASET_CANONICAL = {
    "pingan": "Pingan",
    "qingyun": "Qingyun",
}


METHOD_CONTRACT = {
    "CE": ("ce", 6),
    "Focal": ("focal", 8),
    "LDAM-DRW": ("ldam_drw", 8),
    "Balanced Softmax": ("balanced_softmax", 6),
    "LA-loss": ("la_loss", 8),
}


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def atomic_write_bytes(path, payload):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(str(path) + ".tmp")

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(tmp, path)


def canonical_rel(path):

    p = PurePosixPath(str(path))

    if p.is_absolute():
        raise RuntimeError(
            f"STOP: absolute relative path rejected: {path}"
        )

    if ".." in p.parts:
        raise RuntimeError(
            f"STOP: path traversal rejected: {path}"
        )

    return str(p)


def canonical_dataset(raw_value):

    token = str(raw_value).strip().lower()

    if token not in DATASET_CANONICAL:
        raise RuntimeError(
            "STOP: unsupported completion-manifest dataset token:\n"
            f"{raw_value!r}"
        )

    return token, DATASET_CANONICAL[token]


FALSE_SAFETY_KEYS = {
    "official_test_used",
    "official_test_used_for_selection",
    "official_test_values_accessed",
    "official_test_accessed",
    "official_test_opened",
    "official_test_may_be_opened",
    "test_dataset_created",
    "test_dataloader_created",
    "test_inference_performed",
}


def scan_safety_tree(
    obj,
    source_rel,
    evidence,
    violations,
):

    if isinstance(obj, dict):

        for key, value in obj.items():

            key_l = str(key).lower()

            if key_l in FALSE_SAFETY_KEYS:

                evidence.append(
                    {
                        "source": source_rel,
                        "key": key_l,
                        "value": value,
                    }
                )

                if value is not False:

                    violations.append(
                        {
                            "source": source_rel,
                            "key": key_l,
                            "value": value,
                        }
                    )


            elif key_l == "official_test_status":

                evidence.append(
                    {
                        "source": source_rel,
                        "key": key_l,
                        "value": value,
                    }
                )

                if str(value).upper() != "SEALED":

                    violations.append(
                        {
                            "source": source_rel,
                            "key": key_l,
                            "value": value,
                        }
                    )


            scan_safety_tree(
                value,
                source_rel,
                evidence,
                violations,
            )


    elif isinstance(obj, list):

        for item in obj:

            scan_safety_tree(
                item,
                source_rel,
                evidence,
                violations,
            )


# ==================================================================================================
# 4. AUTHENTICATE FINAL SAVED PACKAGE
# ==================================================================================================

print()
print("-" * 180)
print("[1] AUTHENTICATE FINAL 36/36 SAVED PACKAGE")
print("-" * 180)


for label, path in [
    ("Local migration ZIP", LOCAL_MIGRATION_ZIP),
    ("36/36 persistence ZIP", FINAL_PERSISTENCE_ZIP),
    ("36/36 persistence manifest", FINAL_PERSISTENCE_MANIFEST),
    ("36/36 SHA sidecar", FINAL_PERSISTENCE_SHA_SIDECAR),
]:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: required artifact missing: {label}\n{path}"
        )

    print(
        f"{label:32s}: FOUND"
    )


if not ROOT.is_dir():

    raise FileNotFoundError(
        "STOP: /content/LTLC missing."
    )


final_zip_sha = sha256_file(
    FINAL_PERSISTENCE_ZIP
)

final_manifest_sha = sha256_file(
    FINAL_PERSISTENCE_MANIFEST
)


if final_zip_sha != EXPECTED_FINAL_ZIP_SHA256:

    raise RuntimeError(
        "STOP: final ZIP SHA mismatch."
    )


if final_manifest_sha != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: final manifest SHA mismatch."
    )


with FINAL_PERSISTENCE_MANIFEST.open(
    "r",
    encoding="utf-8",
) as f:

    final_manifest = json.load(f)


if (
    final_manifest["scientific_progress"]["after"] != 36
    or
    final_manifest["scientific_progress"].get(
        "experiment_b_complete"
    ) is not True
    or
    final_manifest["next_action"].get(
        "gpu_training_remaining"
    ) is not False
):

    raise RuntimeError(
        "STOP: final persistence completion state invalid."
    )


final_safety = final_manifest["safety"]


if not (
    final_safety.get("official_test_status") == "SEALED"
    and
    final_safety.get("official_test_values_accessed") is False
    and
    final_safety.get("test_dataset_created") is False
    and
    final_safety.get("test_dataloader_created") is False
    and
    final_safety.get("test_inference_performed") is False
):

    raise RuntimeError(
        "STOP: final persistence safety boundary invalid."
    )


print(
    "36/36 persistence ZIP      : AUTHENTICATED PASS"
)

print(
    "36/36 persistence manifest : AUTHENTICATED PASS"
)

print(
    "GPU training remaining     : False PASS"
)

print(
    "Official test              : SEALED PASS"
)


# ==================================================================================================
# 5. REBUILD EXACT 507-FILE CONTRACT
# ==================================================================================================

print()
print("-" * 180)
print("[2] RECONSTRUCT EXACT 507-FILE SNAPSHOT CONTRACT")
print("-" * 180)


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    "r",
) as zf:

    base_manifest_bytes = zf.read(
        BASE_INTERNAL_MANIFEST
    )


base_manifest_sha = hashlib.sha256(
    base_manifest_bytes
).hexdigest()


if base_manifest_sha != EXPECTED_BASE_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: base internal manifest SHA mismatch."
    )


base_manifest = json.loads(
    base_manifest_bytes.decode("utf-8")
)


base_records = (
    base_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


with zipfile.ZipFile(
    FINAL_PERSISTENCE_ZIP,
    "r",
) as zf:

    delta_manifest_bytes = zf.read(
        DELTA_INTERNAL_MANIFEST
    )


if hashlib.sha256(
    delta_manifest_bytes
).hexdigest() != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: internal final delta manifest SHA mismatch."
    )


if (
    delta_manifest_bytes
    !=
    FINAL_PERSISTENCE_MANIFEST.read_bytes()
):

    raise RuntimeError(
        "STOP: internal/external final manifests differ."
    )


delta_records = (
    final_manifest[
        "full_delta"
    ][
        "records"
    ]
)


if len(base_records) != 449:

    raise RuntimeError(
        "STOP: base record count changed."
    )


if len(delta_records) != 58:

    raise RuntimeError(
        "STOP: final delta record count changed."
    )


base_map = {}


for record in base_records:

    rel = canonical_rel(
        record["relative_path"]
    )

    if rel in base_map:

        raise RuntimeError(
            f"STOP: duplicate base path: {rel}"
        )

    base_map[rel] = record


delta_map = {}


for record in delta_records:

    rel = canonical_rel(
        record["relative_path"]
    )

    if rel in delta_map:

        raise RuntimeError(
            f"STOP: duplicate delta path: {rel}"
        )

    delta_map[rel] = record


if set(base_map) & set(delta_map):

    raise RuntimeError(
        "STOP: base/delta overlap detected."
    )


expected_map = {
    **base_map,
    **delta_map,
}


if len(expected_map) != 507:

    raise RuntimeError(
        "STOP: final expected snapshot is not 507 files."
    )


print("Base records        : 449 PASS")
print("Final delta records : 58 PASS")
print("Base/delta overlap  : 0 PASS")
print("Final contract      : 507 files PASS")


# ==================================================================================================
# 6. EXACT PATH SET + 507/507 HASH AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[3] EXACT 507-FILE PROJECT VERIFICATION")
print("-" * 180)


current_files = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


current_paths = {
    str(p.relative_to(ROOT))
    for p in current_files
}


expected_paths = set(
    expected_map
)


missing = sorted(
    expected_paths
    -
    current_paths
)

unexpected = sorted(
    current_paths
    -
    expected_paths
)


if missing:

    raise RuntimeError(
        "STOP: missing final project paths:\n"
        +
        "\n".join(missing[:20])
    )


if unexpected:

    raise RuntimeError(
        "STOP: unexpected final project paths:\n"
        +
        "\n".join(unexpected[:20])
    )


if len(current_paths) != 507:

    raise RuntimeError(
        "STOP: final project count is not 507."
    )


verified = 0
verified_bytes = 0


for relative_path in sorted(
    expected_map
):

    record = expected_map[
        relative_path
    ]

    path = ROOT / relative_path


    if sha256_file(path) != record["sha256"]:

        raise RuntimeError(
            "STOP: final file SHA mismatch:\n"
            f"{relative_path}"
        )


    if (
        "size_bytes" in record
        and
        int(path.stat().st_size)
        !=
        int(record["size_bytes"])
    ):

        raise RuntimeError(
            "STOP: final file size mismatch:\n"
            f"{relative_path}"
        )


    verified += 1
    verified_bytes += int(
        path.stat().st_size
    )


print(
    "Current project files      : 507 PASS"
)

print(
    "Missing expected files     : 0 PASS"
)

print(
    "Unexpected files           : 0 PASS"
)

print(
    "Full SHA verification      : 507/507 PASS"
)

print(
    "Verified scientific bytes  :",
    verified_bytes,
)


# ==================================================================================================
# 7. EXACT METHOD LEDGER
# ==================================================================================================

print()
print("-" * 180)
print("[4] FINAL 36-RUN METHOD LEDGER")
print("-" * 180)


RUN_ROOT = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
)


completion_paths_by_method = {}


for label, (
    method_dir,
    expected_count,
) in METHOD_CONTRACT.items():

    paths = sorted(
        (
            RUN_ROOT
            /
            method_dir
        ).rglob(
            "*__completion_manifest.json"
        )
    )

    completion_paths_by_method[
        label
    ] = paths


    if len(paths) != expected_count:

        raise RuntimeError(
            f"STOP: {label} completion count mismatch."
        )


    print(
        f"{label:18s}: "
        f"{len(paths)}/{expected_count} PASS"
    )


all_completion_paths = []


for paths in completion_paths_by_method.values():

    all_completion_paths.extend(
        paths
    )


if len(all_completion_paths) != 36:

    raise RuntimeError(
        "STOP: final completion total is not 36."
    )


print()
print(
    "Experiment-B completion manifests: 36/36 PASS"
)


# ==================================================================================================
# 8. CORRECTED ALL-36 MANIFEST AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[5] CORRECTED ALL-36 COMPLETION-MANIFEST AUDIT")
print("-" * 180)


run_records = []
dataset_token_counts = {}


for path in sorted(
    all_completion_paths
):

    relative_path = str(
        path.relative_to(ROOT)
    )


    parts = PurePosixPath(
        relative_path
    ).parts


    if len(parts) < 5:

        raise RuntimeError(
            "STOP: malformed completion path:\n"
            f"{relative_path}"
        )


    if not (
        parts[0] == "runs"
        and
        parts[1]
        ==
        "experiment_b_spatial_validation_v1"
    ):

        raise RuntimeError(
            "STOP: completion manifest outside Experiment-B run tree:\n"
            f"{relative_path}"
        )


    method_dir_from_path = parts[2]
    dataset_slug_from_path = parts[3].lower()


    if dataset_slug_from_path not in DATASET_CANONICAL:

        raise RuntimeError(
            "STOP: unsupported dataset slug in path:\n"
            f"{relative_path}"
        )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        manifest = json.load(f)


    if manifest.get("status") != "COMPLETE":

        raise RuntimeError(
            "STOP: completion status is not COMPLETE:\n"
            f"{relative_path}"
        )


    raw_dataset = manifest.get(
        "dataset"
    )


    dataset_token, dataset_canonical = (
        canonical_dataset(
            raw_dataset
        )
    )


    # Stronger than the original Cell 15 check:
    # manifest dataset must agree exactly with dataset slug encoded in path.
    if dataset_token != dataset_slug_from_path:

        raise RuntimeError(
            "STOP: manifest/path dataset disagreement:\n"
            f"{relative_path}\n"
            f"manifest dataset={raw_dataset!r}\n"
            f"path dataset={dataset_slug_from_path!r}"
        )


    dataset_token_counts[
        str(raw_dataset)
    ] = (
        dataset_token_counts.get(
            str(raw_dataset),
            0,
        )
        +
        1
    )


    try:

        seed = int(
            manifest.get("seed")
        )

    except Exception:

        raise RuntimeError(
            "STOP: invalid seed in completion manifest:\n"
            f"{relative_path}"
        )


    if seed not in {
        42,
        123,
        3407,
    }:

        raise RuntimeError(
            "STOP: unexpected seed in completion manifest:\n"
            f"{relative_path}\n"
            f"seed={seed}"
        )


    checkpoint_policy = manifest.get(
        "checkpoint_policy",
        {},
    )


    best_aa_metrics = checkpoint_policy.get(
        "best_AA_metrics",
        {},
    )

    best_oa_metrics = checkpoint_policy.get(
        "best_OA_metrics",
        {},
    )


    run_records.append(
        {
            "relative_path":
                relative_path,

            "dataset_raw":
                raw_dataset,

            "dataset":
                dataset_canonical,

            "dataset_slug":
                dataset_token,

            "method":
                manifest.get(
                    "method",
                    ""
                ),

            "method_dir":
                method_dir_from_path,

            "seed":
                seed,

            "status":
                manifest.get(
                    "status"
                ),

            "tau":
                manifest.get(
                    "tau",
                    ""
                ),

            "gamma":
                manifest.get(
                    "gamma",
                    ""
                ),

            "C":
                manifest.get(
                    "C",
                    manifest.get(
                        "c",
                        ""
                    ),
                ),

            "best_AA_epoch":
                checkpoint_policy.get(
                    "best_AA_epoch",
                    ""
                ),

            "best_AA":
                best_aa_metrics.get(
                    "AA",
                    ""
                ),

            "best_AA_Macro_F1":
                best_aa_metrics.get(
                    "Macro_F1",
                    ""
                ),

            "best_AA_OA":
                best_aa_metrics.get(
                    "OA",
                    ""
                ),

            "best_OA_epoch":
                checkpoint_policy.get(
                    "best_OA_epoch",
                    ""
                ),

            "best_OA":
                best_oa_metrics.get(
                    "OA",
                    ""
                ),

            "completion_sha256":
                sha256_file(
                    path
                ),
        }
    )


if len(run_records) != 36:

    raise RuntimeError(
        "STOP: corrected completion audit did not produce 36 records."
    )


if len(
    {
        record["relative_path"]
        for record in run_records
    }
) != 36:

    raise RuntimeError(
        "STOP: duplicate completion path detected."
    )


print(
    "Completion status COMPLETE       : 36/36 PASS"
)

print(
    "Dataset token normalization      : 36/36 PASS"
)

print(
    "Manifest/path dataset agreement  : 36/36 PASS"
)

print(
    "Seed-domain audit                : 36/36 PASS"
)

print(
    "Unique completion records        : 36/36 PASS"
)

print()
print(
    "Observed raw dataset tokens:"
)

for token, count in sorted(
    dataset_token_counts.items()
):

    print(
        f"  {token!r}: {count}"
    )


# ==================================================================================================
# 9. EXACT LA 8-RUN MATRIX
# ==================================================================================================

print()
print("-" * 180)
print("[6] EXACT LA-LOSS 8-RUN MATRIX")
print("-" * 180)


expected_la_completion_paths = {
    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau0p5/seed42/"
    "la_loss__pingan__tau0p5__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau1p0/seed42/"
    "la_loss__pingan__tau1p0__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau1p0/seed123/"
    "la_loss__pingan__tau1p0__seed123__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau1p0/seed3407/"
    "la_loss__pingan__tau1p0__seed3407__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau0p5/seed42/"
    "la_loss__qingyun__tau0p5__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau1p0/seed42/"
    "la_loss__qingyun__tau1p0__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau0p5/seed123/"
    "la_loss__qingyun__tau0p5__seed123__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau0p5/seed3407/"
    "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json",
}


observed_la_completion_paths = {
    str(
        path.relative_to(ROOT)
    )
    for path in completion_paths_by_method[
        "LA-loss"
    ]
}


if observed_la_completion_paths != expected_la_completion_paths:

    raise RuntimeError(
        "STOP: exact LA-loss run matrix mismatch."
    )


print(
    "Pingan tau=0.5 seed42           : PASS"
)

print(
    "Pingan tau=1.0 seeds42/123/3407 : PASS"
)

print(
    "Qingyun tau=1.0 seed42          : PASS"
)

print(
    "Qingyun tau=0.5 seeds42/123/3407: PASS"
)

print(
    "Exact LA-loss matrix            : 8/8 PASS"
)


# ==================================================================================================
# 10. FROZEN SELECTIONS
# ==================================================================================================

print()
print("-" * 180)
print("[7] FROZEN LA TAU-SELECTION SEAL")
print("-" * 180)


PINGAN_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "pingan"
    / "la_loss__pingan__selected_tau_v1.json"
)

QINGYUN_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "la_loss__qingyun__selected_tau_v1.json"
)


selection_summary = {}


for (
    dataset,
    path,
    expected_sha,
    expected_tau,
) in [
    (
        "Pingan",
        PINGAN_SELECTION,
        EXPECTED_PINGAN_SELECTION_SHA256,
        1.0,
    ),
    (
        "Qingyun",
        QINGYUN_SELECTION,
        EXPECTED_QINGYUN_SELECTION_SHA256,
        0.5,
    ),
]:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: {dataset} selection missing."
        )


    actual_sha = sha256_file(
        path
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: {dataset} selection SHA mismatch."
        )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        selection = json.load(f)


    if float(
        selection["selected_tau"]
    ) != expected_tau:

        raise RuntimeError(
            f"STOP: {dataset} selected tau changed."
        )


    if not (
        selection.get(
            "official_test_status"
        ) == "SEALED"
        and
        selection.get(
            "official_test_values_accessed"
        ) is False
        and
        selection.get(
            "official_test_used_for_selection"
        ) is False
    ):

        raise RuntimeError(
            f"STOP: {dataset} selection safety boundary failed."
        )


    selection_summary[
        dataset
    ] = {
        "selected_tau":
            expected_tau,

        "selection_sha256":
            actual_sha,
    }


    print(
        f"{dataset:10s}: "
        f"tau={expected_tau} | SHA PASS | OFFICIAL TEST SEALED"
    )


# ==================================================================================================
# 11. FINAL SELECTED-QINGYUN COMPLETION IDENTITIES
# ==================================================================================================

print()
print("-" * 180)
print("[8] FINAL QINGYUN SELECTED-TAU COMPLETION IDENTITIES")
print("-" * 180)


QINGYUN_SEED123 = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed123"
    / "la_loss__qingyun__tau0p5__seed123__completion_manifest.json"
)

QINGYUN_SEED3407 = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed3407"
    / "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json"
)


if (
    sha256_file(QINGYUN_SEED123)
    !=
    EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed123 completion SHA mismatch."
    )


if (
    sha256_file(QINGYUN_SEED3407)
    !=
    EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed3407 completion SHA mismatch."
    )


print(
    "Qingyun tau=0.5 seed123 : AUTHENTICATED PASS"
)

print(
    "Qingyun tau=0.5 seed3407: AUTHENTICATED PASS"
)


# ==================================================================================================
# 12. PROJECT-WIDE OFFICIAL-TEST SAFETY SCAN
# ==================================================================================================

print()
print("-" * 180)
print("[9] PROJECT-WIDE OFFICIAL-TEST SAFETY SCAN")
print("-" * 180)


safety_evidence = []
safety_violations = []
json_files_scanned = 0


for path in sorted(
    ROOT.rglob("*.json")
):

    relative_path = str(
        path.relative_to(ROOT)
    )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        payload = json.load(f)


    json_files_scanned += 1


    scan_safety_tree(
        payload,
        relative_path,
        safety_evidence,
        safety_violations,
    )


if safety_violations:

    print(
        json.dumps(
            safety_violations[:20],
            indent=2,
            sort_keys=True,
        )
    )

    raise RuntimeError(
        "STOP: official-test safety violation detected."
    )


if len(safety_evidence) == 0:

    raise RuntimeError(
        "STOP: no explicit safety evidence found."
    )


print(
    "JSON files scanned         :",
    json_files_scanned
)

print(
    "Explicit safety assertions :",
    len(safety_evidence)
)

print(
    "Explicit safety violations : 0 PASS"
)

print(
    "Official test              : SEALED PASS"
)


# ==================================================================================================
# 13. BUILD FINAL 36-RUN LEDGER
# ==================================================================================================

print()
print("-" * 180)
print("[10] BUILD FINAL 36-RUN LEDGER")
print("-" * 180)


ledger_fields = [
    "relative_path",
    "dataset_raw",
    "dataset",
    "dataset_slug",
    "method",
    "method_dir",
    "seed",
    "status",
    "tau",
    "gamma",
    "C",
    "best_AA_epoch",
    "best_AA",
    "best_AA_Macro_F1",
    "best_AA_OA",
    "best_OA_epoch",
    "best_OA",
    "completion_sha256",
]


ledger_buffer = io.StringIO()


writer = csv.DictWriter(
    ledger_buffer,
    fieldnames=ledger_fields,
    lineterminator="\n",
)


writer.writeheader()


for record in sorted(
    run_records,
    key=lambda r: (
        r["method_dir"],
        r["dataset_slug"],
        r["seed"],
        r["relative_path"],
    ),
):

    writer.writerow(
        {
            key:
                record.get(
                    key,
                    ""
                )
            for key in ledger_fields
        }
    )


ledger_bytes = ledger_buffer.getvalue().encode(
    "utf-8"
)


ledger_sha = hashlib.sha256(
    ledger_bytes
).hexdigest()


print(
    "Run-ledger rows    : 36 PASS"
)

print(
    "Run-ledger SHA-256 :",
    ledger_sha
)


# ==================================================================================================
# 14. BUILD FINAL CORRECTED CLOSING AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[11] BUILD FINAL CORRECTED CLOSING-AUDIT SEAL")
print("-" * 180)


audit_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_NOTEBOOK06A_EXPERIMENT_B_FINAL_CLOSING_AUDIT",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "closing_audit_execution": {
        "cell":
            "15A",

        "correction_of":
            "Cell 15",

        "correction_reason":
            (
                "Cell 15 incorrectly treated completion-manifest "
                "dataset tokens as case-sensitive. Cell 15A uses "
                "case-insensitive canonical normalization and also "
                "cross-checks every normalized dataset token against "
                "the dataset slug encoded in its run path."
            ),

        "scientific_state_modified":
            False,
    },

    "final_state": {
        "experiment_b_progress":
            "36/36",

        "experiment_b_complete":
            True,

        "gpu_training_remaining":
            False,

        "project_file_count":
            507,

        "base_project_file_count":
            449,

        "final_cumulative_delta_file_count":
            58,
    },

    "snapshot_authentication": {
        "final_36of36_persistence_zip":
            str(
                FINAL_PERSISTENCE_ZIP
            ),

        "final_36of36_persistence_zip_sha256":
            final_zip_sha,

        "final_36of36_persistence_manifest":
            str(
                FINAL_PERSISTENCE_MANIFEST
            ),

        "final_36of36_persistence_manifest_sha256":
            final_manifest_sha,

        "base_internal_manifest_sha256":
            base_manifest_sha,

        "full_current_project_sha_verification":
            "507/507 PASS",

        "missing_expected_paths":
            0,

        "unexpected_paths":
            0,
    },

    "completion_ledger": {
        "CE":
            6,

        "Focal":
            8,

        "LDAM-DRW":
            8,

        "Balanced Softmax":
            6,

        "LA-loss":
            8,

        "total":
            36,
    },

    "completion_manifest_dataset_audit": {
        "case_insensitive_normalization":
            True,

        "manifest_path_dataset_agreement":
            "36/36 PASS",

        "canonical_datasets":
            [
                "Pingan",
                "Qingyun",
            ],

        "observed_raw_dataset_tokens":
            dataset_token_counts,
    },

    "la_loss": {
        "run_matrix":
            "8/8 EXACT PASS",

        "Pingan":
            selection_summary[
                "Pingan"
            ],

        "Qingyun":
            selection_summary[
                "Qingyun"
            ],

        "qingyun_seed123_completion_sha256":
            EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256,

        "qingyun_seed3407_completion_sha256":
            EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256,
    },

    "official_test": {
        "status":
            "SEALED",

        "project_json_files_scanned":
            json_files_scanned,

        "explicit_safety_assertions":
            len(
                safety_evidence
            ),

        "explicit_safety_violations":
            0,

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,
    },

    "protocol_integrity": {
        "training_performed_in_closing_audit":
            False,

        "model_inference_performed_in_closing_audit":
            False,

        "project_mutated_by_closing_audit":
            False,

        "retuning_performed":
            False,

        "mixed_precision_changed":
            False,

        "multi_gpu_used":
            False,
    },

    "external_run_ledger": {
        "file":
            str(
                FINAL_RUN_LEDGER_CSV
            ),

        "rows":
            36,

        "sha256":
            ledger_sha,
    },

    "closing_status":
        "PASS",

    "seal":
        (
            "EXPERIMENT_B_36_OF_36_COMPLETE__"
            "FULL_507_FILE_SNAPSHOT_VERIFIED__"
            "36_COMPLETION_RECORDS_AUDITED__"
            "DATASET_TOKENS_CANONICALLY_NORMALIZED_AND_PATH_VERIFIED__"
            "LA_SELECTIONS_FROZEN__"
            "OFFICIAL_TEST_SEALED__"
            "NO_GPU_TRAINING_REMAINING"
        ),
}


audit_bytes = json.dumps(
    audit_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


audit_sha = hashlib.sha256(
    audit_bytes
).hexdigest()


print(
    "Corrected closing-audit SHA-256:",
    audit_sha
)


# ==================================================================================================
# 15. WRITE FINAL EXTERNAL AUDIT PACKAGE
# ==================================================================================================

print()
print("-" * 180)
print("[12] WRITE FINAL EXTERNAL AUDIT PACKAGE TO GOOGLE DRIVE")
print("-" * 180)


FINAL_AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    FINAL_AUDIT_JSON,
    FINAL_RUN_LEDGER_CSV,
    FINAL_AUDIT_SHA,
]:

    if path.exists():

        raise RuntimeError(
            "STOP: final audit artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    FINAL_RUN_LEDGER_CSV,
    ledger_bytes,
)


atomic_write_bytes(
    FINAL_AUDIT_JSON,
    audit_bytes,
)


sidecar_text = (
    f"{audit_sha}  {FINAL_AUDIT_JSON.name}\n"
    f"{ledger_sha}  {FINAL_RUN_LEDGER_CSV.name}\n"
    f"{final_zip_sha}  {FINAL_PERSISTENCE_ZIP.name}\n"
    f"{final_manifest_sha}  {FINAL_PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    FINAL_AUDIT_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


if sha256_file(
    FINAL_AUDIT_JSON
) != audit_sha:

    raise RuntimeError(
        "STOP: written final audit SHA verification failed."
    )


if sha256_file(
    FINAL_RUN_LEDGER_CSV
) != ledger_sha:

    raise RuntimeError(
        "STOP: written final ledger SHA verification failed."
    )


# ==================================================================================================
# 16. PROVE NO SCIENTIFIC PROJECT MUTATION
# ==================================================================================================

final_project_files = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    final_project_files
) != 507:

    raise RuntimeError(
        "STOP: correction audit changed project footprint."
    )


# ==================================================================================================
# 17. TERMINAL GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 15A TERMINAL GATE")
print("=" * 180)

print(
    "Cell 15 failure classification             : AUDIT-CODE SEMANTIC DEFECT"
)

print(
    "Scientific-state failure                   : False"
)

print(
    "Scientific artifacts modified by correction: False"
)

print()
print(
    "Final persistence                          : AUTHENTICATED PASS"
)

print(
    "Exact final project                        : 507 files"
)

print(
    "Full project SHA verification              : 507/507 PASS"
)

print(
    "Missing expected files                     : 0"
)

print(
    "Unexpected project files                   : 0"
)

print()
print(
    "CE                                         : 6/6 PASS"
)

print(
    "Focal                                      : 8/8 PASS"
)

print(
    "LDAM-DRW                                   : 8/8 PASS"
)

print(
    "Balanced Softmax                           : 6/6 PASS"
)

print(
    "LA-loss                                    : 8/8 PASS"
)

print(
    "Experiment-B                               : 36/36 COMPLETE"
)

print()
print(
    "Completion dataset normalization           : 36/36 PASS"
)

print(
    "Manifest/path dataset agreement            : 36/36 PASS"
)

print(
    "Exact LA-loss run matrix                   : 8/8 PASS"
)

print(
    "Pingan selected tau                        : 1.0 PASS"
)

print(
    "Qingyun selected tau                       : 0.5 PASS"
)

print()
print(
    "Official-test safety violations            : 0"
)

print(
    "Official test                              : SEALED"
)

print(
    "Official-test inference                    : NOT PERFORMED"
)

print(
    "Training in correction audit               : False"
)

print(
    "GPU training remaining                     : False"
)

print()
print(
    "FINAL CLOSING AUDIT:"
)

print(
    FINAL_AUDIT_JSON
)

print()
print(
    "FINAL CLOSING AUDIT SHA-256:"
)

print(
    audit_sha
)

print()
print(
    "FINAL 36-RUN LEDGER:"
)

print(
    FINAL_RUN_LEDGER_CSV
)

print()
print(
    "FINAL 36-RUN LEDGER SHA-256:"
)

print(
    ledger_sha
)

print()
print(
    "FINAL AUDIT SHA SIDECAR:"
)

print(
    FINAL_AUDIT_SHA
)

print()
print(
    "FINAL CORRECTION CELL 15A PASS: True"
)

print()
print(
    "STATUS: "
    "CELL15_AUDIT_BUG_CORRECTED__"
    "EXPERIMENT_B_36_OF_36_COMPLETE__"
    "FULL_507_FILE_SNAPSHOT_VERIFIED__"
    "36_COMPLETION_RECORDS_AUDITED__"
    "DATASET_TOKENS_CANONICALLY_NORMALIZED_AND_PATH_VERIFIED__"
    "LA_SELECTIONS_FROZEN__"
    "OFFICIAL_TEST_SEALED__"
    "NO_GPU_TRAINING_REMAINING__"
    "FINAL_CLOSING_AUDIT_WRITTEN_TO_GOOGLE_DRIVE"
)

print("=" * 180)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CORRECTION CELL 15A
#
# CORRECTION OF CELL 15 AUDIT-ONLY DEFECT
#
# ROOT CAUSE:
#   Cell 15 treated completion_manifest["dataset"] as case-sensitive and required
#   exactly "Pingan" or "Qingyun".
#
#   Historical completion manifests may canonically store:
#       "pingan"
#       "qingyun"
#
# CORRECTION:
#   - normalize manifest dataset token case-insensitively
#   - map to canonical display names Pingan/Qingyun
#   - independently cross-check normalized dataset against dataset slug in path
#
# SCIENTIFIC STATE:
#   - NOT MODIFIED
#   - NO TRAINING
#   - NO MODEL INFERENCE
#   - NO RETUNING
#   - OFFICIAL TEST REMAINS SEALED
#
# THIS CELL REPERFORMS THE COMPLETE TERMINAL AUDIT AND WRITES THE FINAL EXTERNAL SEAL.
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import csv
import hashlib
import io
import json
import os
import zipfile


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 15A")
print("CORRECTED FINAL 36/36 CLOSING AUDIT — CANONICAL DATASET NORMALIZATION")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path("/content/LTLC")

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

FINAL_PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "36of36_Post_Qingyun_LA_tau0p5_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

FINAL_PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}.zip"
)

FINAL_PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__MANIFEST.json"
)

FINAL_PERSISTENCE_SHA_SIDECAR = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__SHA256.txt"
)


FINAL_AUDIT_DIR = (
    DRIVE_ROOT
    / "final_audits"
    / "notebook06a_experiment_b"
)

FINAL_AUDIT_JSON = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1.json"
)

FINAL_RUN_LEDGER_CSV = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_RUN_LEDGER_v1.csv"
)

FINAL_AUDIT_SHA = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1__SHA256.txt"
)


# ==================================================================================================
# 2. FROZEN IDENTITIES
# ==================================================================================================

EXPECTED_FINAL_ZIP_SHA256 = (
    "30ba228994f0a16e756c5ade22480b00"
    "f3e2304ee3e15bd8e00409d86dc2b7b8"
)

EXPECTED_FINAL_MANIFEST_SHA256 = (
    "5f93f457f3b1510d72b366aa8a78aa65"
    "fbf59e866dadfcefd7388d80f81c79b9"
)

EXPECTED_BASE_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)

EXPECTED_QINGYUN_SELECTION_SHA256 = (
    "367d6ecd113d786bf89510985e0f0ff7"
    "e6589a68db9c912d4c36fb29d02fae89"
)

EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256 = (
    "1e9ac68021a9d01d59818c7061120bc9"
    "d511e1957f02b7a96ef3eb442933524d"
)

EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256 = (
    "d99a05fa4c349267c542fae7a8007024"
    "cdcbb2e773ad9d817909ac43d538bcd7"
)


BASE_PROJECT_FILE_COUNT = 449
FINAL_DELTA_FILE_COUNT = 58
FINAL_PROJECT_FILE_COUNT = 507


BASE_INTERNAL_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)

DELTA_INTERNAL_MANIFEST = (
    "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
)


DATASET_CANONICAL = {
    "pingan": "Pingan",
    "qingyun": "Qingyun",
}


METHOD_CONTRACT = {
    "CE": ("ce", 6),
    "Focal": ("focal", 8),
    "LDAM-DRW": ("ldam_drw", 8),
    "Balanced Softmax": ("balanced_softmax", 6),
    "LA-loss": ("la_loss", 8),
}


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def atomic_write_bytes(path, payload):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(str(path) + ".tmp")

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(tmp, path)


def canonical_rel(path):

    p = PurePosixPath(str(path))

    if p.is_absolute():
        raise RuntimeError(
            f"STOP: absolute relative path rejected: {path}"
        )

    if ".." in p.parts:
        raise RuntimeError(
            f"STOP: path traversal rejected: {path}"
        )

    return str(p)


def canonical_dataset(raw_value):

    token = str(raw_value).strip().lower()

    if token not in DATASET_CANONICAL:
        raise RuntimeError(
            "STOP: unsupported completion-manifest dataset token:\n"
            f"{raw_value!r}"
        )

    return token, DATASET_CANONICAL[token]


FALSE_SAFETY_KEYS = {
    "official_test_used",
    "official_test_used_for_selection",
    "official_test_values_accessed",
    "official_test_accessed",
    "official_test_opened",
    "official_test_may_be_opened",
    "test_dataset_created",
    "test_dataloader_created",
    "test_inference_performed",
}


def scan_safety_tree(
    obj,
    source_rel,
    evidence,
    violations,
):

    if isinstance(obj, dict):

        for key, value in obj.items():

            key_l = str(key).lower()

            if key_l in FALSE_SAFETY_KEYS:

                evidence.append(
                    {
                        "source": source_rel,
                        "key": key_l,
                        "value": value,
                    }
                )

                if value is not False:

                    violations.append(
                        {
                            "source": source_rel,
                            "key": key_l,
                            "value": value,
                        }
                    )


            elif key_l == "official_test_status":

                evidence.append(
                    {
                        "source": source_rel,
                        "key": key_l,
                        "value": value,
                    }
                )

                if str(value).upper() != "SEALED":

                    violations.append(
                        {
                            "source": source_rel,
                            "key": key_l,
                            "value": value,
                        }
                    )


            scan_safety_tree(
                value,
                source_rel,
                evidence,
                violations,
            )


    elif isinstance(obj, list):

        for item in obj:

            scan_safety_tree(
                item,
                source_rel,
                evidence,
                violations,
            )


# ==================================================================================================
# 4. AUTHENTICATE FINAL SAVED PACKAGE
# ==================================================================================================

print()
print("-" * 180)
print("[1] AUTHENTICATE FINAL 36/36 SAVED PACKAGE")
print("-" * 180)


for label, path in [
    ("Local migration ZIP", LOCAL_MIGRATION_ZIP),
    ("36/36 persistence ZIP", FINAL_PERSISTENCE_ZIP),
    ("36/36 persistence manifest", FINAL_PERSISTENCE_MANIFEST),
    ("36/36 SHA sidecar", FINAL_PERSISTENCE_SHA_SIDECAR),
]:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: required artifact missing: {label}\n{path}"
        )

    print(
        f"{label:32s}: FOUND"
    )


if not ROOT.is_dir():

    raise FileNotFoundError(
        "STOP: /content/LTLC missing."
    )


final_zip_sha = sha256_file(
    FINAL_PERSISTENCE_ZIP
)

final_manifest_sha = sha256_file(
    FINAL_PERSISTENCE_MANIFEST
)


if final_zip_sha != EXPECTED_FINAL_ZIP_SHA256:

    raise RuntimeError(
        "STOP: final ZIP SHA mismatch."
    )


if final_manifest_sha != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: final manifest SHA mismatch."
    )


with FINAL_PERSISTENCE_MANIFEST.open(
    "r",
    encoding="utf-8",
) as f:

    final_manifest = json.load(f)


if (
    final_manifest["scientific_progress"]["after"] != 36
    or
    final_manifest["scientific_progress"].get(
        "experiment_b_complete"
    ) is not True
    or
    final_manifest["next_action"].get(
        "gpu_training_remaining"
    ) is not False
):

    raise RuntimeError(
        "STOP: final persistence completion state invalid."
    )


final_safety = final_manifest["safety"]


if not (
    final_safety.get("official_test_status") == "SEALED"
    and
    final_safety.get("official_test_values_accessed") is False
    and
    final_safety.get("test_dataset_created") is False
    and
    final_safety.get("test_dataloader_created") is False
    and
    final_safety.get("test_inference_performed") is False
):

    raise RuntimeError(
        "STOP: final persistence safety boundary invalid."
    )


print(
    "36/36 persistence ZIP      : AUTHENTICATED PASS"
)

print(
    "36/36 persistence manifest : AUTHENTICATED PASS"
)

print(
    "GPU training remaining     : False PASS"
)

print(
    "Official test              : SEALED PASS"
)


# ==================================================================================================
# 5. REBUILD EXACT 507-FILE CONTRACT
# ==================================================================================================

print()
print("-" * 180)
print("[2] RECONSTRUCT EXACT 507-FILE SNAPSHOT CONTRACT")
print("-" * 180)


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    "r",
) as zf:

    base_manifest_bytes = zf.read(
        BASE_INTERNAL_MANIFEST
    )


base_manifest_sha = hashlib.sha256(
    base_manifest_bytes
).hexdigest()


if base_manifest_sha != EXPECTED_BASE_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: base internal manifest SHA mismatch."
    )


base_manifest = json.loads(
    base_manifest_bytes.decode("utf-8")
)


base_records = (
    base_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


with zipfile.ZipFile(
    FINAL_PERSISTENCE_ZIP,
    "r",
) as zf:

    delta_manifest_bytes = zf.read(
        DELTA_INTERNAL_MANIFEST
    )


if hashlib.sha256(
    delta_manifest_bytes
).hexdigest() != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: internal final delta manifest SHA mismatch."
    )


if (
    delta_manifest_bytes
    !=
    FINAL_PERSISTENCE_MANIFEST.read_bytes()
):

    raise RuntimeError(
        "STOP: internal/external final manifests differ."
    )


delta_records = (
    final_manifest[
        "full_delta"
    ][
        "records"
    ]
)


if len(base_records) != 449:

    raise RuntimeError(
        "STOP: base record count changed."
    )


if len(delta_records) != 58:

    raise RuntimeError(
        "STOP: final delta record count changed."
    )


base_map = {}


for record in base_records:

    rel = canonical_rel(
        record["relative_path"]
    )

    if rel in base_map:

        raise RuntimeError(
            f"STOP: duplicate base path: {rel}"
        )

    base_map[rel] = record


delta_map = {}


for record in delta_records:

    rel = canonical_rel(
        record["relative_path"]
    )

    if rel in delta_map:

        raise RuntimeError(
            f"STOP: duplicate delta path: {rel}"
        )

    delta_map[rel] = record


if set(base_map) & set(delta_map):

    raise RuntimeError(
        "STOP: base/delta overlap detected."
    )


expected_map = {
    **base_map,
    **delta_map,
}


if len(expected_map) != 507:

    raise RuntimeError(
        "STOP: final expected snapshot is not 507 files."
    )


print("Base records        : 449 PASS")
print("Final delta records : 58 PASS")
print("Base/delta overlap  : 0 PASS")
print("Final contract      : 507 files PASS")


# ==================================================================================================
# 6. EXACT PATH SET + 507/507 HASH AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[3] EXACT 507-FILE PROJECT VERIFICATION")
print("-" * 180)


current_files = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


current_paths = {
    str(p.relative_to(ROOT))
    for p in current_files
}


expected_paths = set(
    expected_map
)


missing = sorted(
    expected_paths
    -
    current_paths
)

unexpected = sorted(
    current_paths
    -
    expected_paths
)


if missing:

    raise RuntimeError(
        "STOP: missing final project paths:\n"
        +
        "\n".join(missing[:20])
    )


if unexpected:

    raise RuntimeError(
        "STOP: unexpected final project paths:\n"
        +
        "\n".join(unexpected[:20])
    )


if len(current_paths) != 507:

    raise RuntimeError(
        "STOP: final project count is not 507."
    )


verified = 0
verified_bytes = 0


for relative_path in sorted(
    expected_map
):

    record = expected_map[
        relative_path
    ]

    path = ROOT / relative_path


    if sha256_file(path) != record["sha256"]:

        raise RuntimeError(
            "STOP: final file SHA mismatch:\n"
            f"{relative_path}"
        )


    if (
        "size_bytes" in record
        and
        int(path.stat().st_size)
        !=
        int(record["size_bytes"])
    ):

        raise RuntimeError(
            "STOP: final file size mismatch:\n"
            f"{relative_path}"
        )


    verified += 1
    verified_bytes += int(
        path.stat().st_size
    )


print(
    "Current project files      : 507 PASS"
)

print(
    "Missing expected files     : 0 PASS"
)

print(
    "Unexpected files           : 0 PASS"
)

print(
    "Full SHA verification      : 507/507 PASS"
)

print(
    "Verified scientific bytes  :",
    verified_bytes,
)


# ==================================================================================================
# 7. EXACT METHOD LEDGER
# ==================================================================================================

print()
print("-" * 180)
print("[4] FINAL 36-RUN METHOD LEDGER")
print("-" * 180)


RUN_ROOT = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
)


completion_paths_by_method = {}


for label, (
    method_dir,
    expected_count,
) in METHOD_CONTRACT.items():

    paths = sorted(
        (
            RUN_ROOT
            /
            method_dir
        ).rglob(
            "*__completion_manifest.json"
        )
    )

    completion_paths_by_method[
        label
    ] = paths


    if len(paths) != expected_count:

        raise RuntimeError(
            f"STOP: {label} completion count mismatch."
        )


    print(
        f"{label:18s}: "
        f"{len(paths)}/{expected_count} PASS"
    )


all_completion_paths = []


for paths in completion_paths_by_method.values():

    all_completion_paths.extend(
        paths
    )


if len(all_completion_paths) != 36:

    raise RuntimeError(
        "STOP: final completion total is not 36."
    )


print()
print(
    "Experiment-B completion manifests: 36/36 PASS"
)


# ==================================================================================================
# 8. CORRECTED ALL-36 MANIFEST AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[5] CORRECTED ALL-36 COMPLETION-MANIFEST AUDIT")
print("-" * 180)


run_records = []
dataset_token_counts = {}


for path in sorted(
    all_completion_paths
):

    relative_path = str(
        path.relative_to(ROOT)
    )


    parts = PurePosixPath(
        relative_path
    ).parts


    if len(parts) < 5:

        raise RuntimeError(
            "STOP: malformed completion path:\n"
            f"{relative_path}"
        )


    if not (
        parts[0] == "runs"
        and
        parts[1]
        ==
        "experiment_b_spatial_validation_v1"
    ):

        raise RuntimeError(
            "STOP: completion manifest outside Experiment-B run tree:\n"
            f"{relative_path}"
        )


    method_dir_from_path = parts[2]
    dataset_slug_from_path = parts[3].lower()


    if dataset_slug_from_path not in DATASET_CANONICAL:

        raise RuntimeError(
            "STOP: unsupported dataset slug in path:\n"
            f"{relative_path}"
        )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        manifest = json.load(f)


    if manifest.get("status") != "COMPLETE":

        raise RuntimeError(
            "STOP: completion status is not COMPLETE:\n"
            f"{relative_path}"
        )


    raw_dataset = manifest.get(
        "dataset"
    )


    dataset_token, dataset_canonical = (
        canonical_dataset(
            raw_dataset
        )
    )


    # Stronger than the original Cell 15 check:
    # manifest dataset must agree exactly with dataset slug encoded in path.
    if dataset_token != dataset_slug_from_path:

        raise RuntimeError(
            "STOP: manifest/path dataset disagreement:\n"
            f"{relative_path}\n"
            f"manifest dataset={raw_dataset!r}\n"
            f"path dataset={dataset_slug_from_path!r}"
        )


    dataset_token_counts[
        str(raw_dataset)
    ] = (
        dataset_token_counts.get(
            str(raw_dataset),
            0,
        )
        +
        1
    )


    try:

        seed = int(
            manifest.get("seed")
        )

    except Exception:

        raise RuntimeError(
            "STOP: invalid seed in completion manifest:\n"
            f"{relative_path}"
        )


    if seed not in {
        42,
        123,
        3407,
    }:

        raise RuntimeError(
            "STOP: unexpected seed in completion manifest:\n"
            f"{relative_path}\n"
            f"seed={seed}"
        )


    checkpoint_policy = manifest.get(
        "checkpoint_policy",
        {},
    )


    best_aa_metrics = checkpoint_policy.get(
        "best_AA_metrics",
        {},
    )

    best_oa_metrics = checkpoint_policy.get(
        "best_OA_metrics",
        {},
    )


    run_records.append(
        {
            "relative_path":
                relative_path,

            "dataset_raw":
                raw_dataset,

            "dataset":
                dataset_canonical,

            "dataset_slug":
                dataset_token,

            "method":
                manifest.get(
                    "method",
                    ""
                ),

            "method_dir":
                method_dir_from_path,

            "seed":
                seed,

            "status":
                manifest.get(
                    "status"
                ),

            "tau":
                manifest.get(
                    "tau",
                    ""
                ),

            "gamma":
                manifest.get(
                    "gamma",
                    ""
                ),

            "C":
                manifest.get(
                    "C",
                    manifest.get(
                        "c",
                        ""
                    ),
                ),

            "best_AA_epoch":
                checkpoint_policy.get(
                    "best_AA_epoch",
                    ""
                ),

            "best_AA":
                best_aa_metrics.get(
                    "AA",
                    ""
                ),

            "best_AA_Macro_F1":
                best_aa_metrics.get(
                    "Macro_F1",
                    ""
                ),

            "best_AA_OA":
                best_aa_metrics.get(
                    "OA",
                    ""
                ),

            "best_OA_epoch":
                checkpoint_policy.get(
                    "best_OA_epoch",
                    ""
                ),

            "best_OA":
                best_oa_metrics.get(
                    "OA",
                    ""
                ),

            "completion_sha256":
                sha256_file(
                    path
                ),
        }
    )


if len(run_records) != 36:

    raise RuntimeError(
        "STOP: corrected completion audit did not produce 36 records."
    )


if len(
    {
        record["relative_path"]
        for record in run_records
    }
) != 36:

    raise RuntimeError(
        "STOP: duplicate completion path detected."
    )


print(
    "Completion status COMPLETE       : 36/36 PASS"
)

print(
    "Dataset token normalization      : 36/36 PASS"
)

print(
    "Manifest/path dataset agreement  : 36/36 PASS"
)

print(
    "Seed-domain audit                : 36/36 PASS"
)

print(
    "Unique completion records        : 36/36 PASS"
)

print()
print(
    "Observed raw dataset tokens:"
)

for token, count in sorted(
    dataset_token_counts.items()
):

    print(
        f"  {token!r}: {count}"
    )


# ==================================================================================================
# 9. EXACT LA 8-RUN MATRIX
# ==================================================================================================

print()
print("-" * 180)
print("[6] EXACT LA-LOSS 8-RUN MATRIX")
print("-" * 180)


expected_la_completion_paths = {
    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau0p5/seed42/"
    "la_loss__pingan__tau0p5__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau1p0/seed42/"
    "la_loss__pingan__tau1p0__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau1p0/seed123/"
    "la_loss__pingan__tau1p0__seed123__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/pingan/tau1p0/seed3407/"
    "la_loss__pingan__tau1p0__seed3407__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau0p5/seed42/"
    "la_loss__qingyun__tau0p5__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau1p0/seed42/"
    "la_loss__qingyun__tau1p0__seed42__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau0p5/seed123/"
    "la_loss__qingyun__tau0p5__seed123__completion_manifest.json",

    "runs/experiment_b_spatial_validation_v1/la_loss/qingyun/tau0p5/seed3407/"
    "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json",
}


observed_la_completion_paths = {
    str(
        path.relative_to(ROOT)
    )
    for path in completion_paths_by_method[
        "LA-loss"
    ]
}


if observed_la_completion_paths != expected_la_completion_paths:

    raise RuntimeError(
        "STOP: exact LA-loss run matrix mismatch."
    )


print(
    "Pingan tau=0.5 seed42           : PASS"
)

print(
    "Pingan tau=1.0 seeds42/123/3407 : PASS"
)

print(
    "Qingyun tau=1.0 seed42          : PASS"
)

print(
    "Qingyun tau=0.5 seeds42/123/3407: PASS"
)

print(
    "Exact LA-loss matrix            : 8/8 PASS"
)


# ==================================================================================================
# 10. FROZEN SELECTIONS
# ==================================================================================================

print()
print("-" * 180)
print("[7] FROZEN LA TAU-SELECTION SEAL")
print("-" * 180)


PINGAN_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "pingan"
    / "la_loss__pingan__selected_tau_v1.json"
)

QINGYUN_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "la_loss__qingyun__selected_tau_v1.json"
)


selection_summary = {}


for (
    dataset,
    path,
    expected_sha,
    expected_tau,
) in [
    (
        "Pingan",
        PINGAN_SELECTION,
        EXPECTED_PINGAN_SELECTION_SHA256,
        1.0,
    ),
    (
        "Qingyun",
        QINGYUN_SELECTION,
        EXPECTED_QINGYUN_SELECTION_SHA256,
        0.5,
    ),
]:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: {dataset} selection missing."
        )


    actual_sha = sha256_file(
        path
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: {dataset} selection SHA mismatch."
        )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        selection = json.load(f)


    if float(
        selection["selected_tau"]
    ) != expected_tau:

        raise RuntimeError(
            f"STOP: {dataset} selected tau changed."
        )


    if not (
        selection.get(
            "official_test_status"
        ) == "SEALED"
        and
        selection.get(
            "official_test_values_accessed"
        ) is False
        and
        selection.get(
            "official_test_used_for_selection"
        ) is False
    ):

        raise RuntimeError(
            f"STOP: {dataset} selection safety boundary failed."
        )


    selection_summary[
        dataset
    ] = {
        "selected_tau":
            expected_tau,

        "selection_sha256":
            actual_sha,
    }


    print(
        f"{dataset:10s}: "
        f"tau={expected_tau} | SHA PASS | OFFICIAL TEST SEALED"
    )


# ==================================================================================================
# 11. FINAL SELECTED-QINGYUN COMPLETION IDENTITIES
# ==================================================================================================

print()
print("-" * 180)
print("[8] FINAL QINGYUN SELECTED-TAU COMPLETION IDENTITIES")
print("-" * 180)


QINGYUN_SEED123 = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed123"
    / "la_loss__qingyun__tau0p5__seed123__completion_manifest.json"
)

QINGYUN_SEED3407 = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed3407"
    / "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json"
)


if (
    sha256_file(QINGYUN_SEED123)
    !=
    EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed123 completion SHA mismatch."
    )


if (
    sha256_file(QINGYUN_SEED3407)
    !=
    EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed3407 completion SHA mismatch."
    )


print(
    "Qingyun tau=0.5 seed123 : AUTHENTICATED PASS"
)

print(
    "Qingyun tau=0.5 seed3407: AUTHENTICATED PASS"
)


# ==================================================================================================
# 12. PROJECT-WIDE OFFICIAL-TEST SAFETY SCAN
# ==================================================================================================

print()
print("-" * 180)
print("[9] PROJECT-WIDE OFFICIAL-TEST SAFETY SCAN")
print("-" * 180)


safety_evidence = []
safety_violations = []
json_files_scanned = 0


for path in sorted(
    ROOT.rglob("*.json")
):

    relative_path = str(
        path.relative_to(ROOT)
    )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        payload = json.load(f)


    json_files_scanned += 1


    scan_safety_tree(
        payload,
        relative_path,
        safety_evidence,
        safety_violations,
    )


if safety_violations:

    print(
        json.dumps(
            safety_violations[:20],
            indent=2,
            sort_keys=True,
        )
    )

    raise RuntimeError(
        "STOP: official-test safety violation detected."
    )


if len(safety_evidence) == 0:

    raise RuntimeError(
        "STOP: no explicit safety evidence found."
    )


print(
    "JSON files scanned         :",
    json_files_scanned
)

print(
    "Explicit safety assertions :",
    len(safety_evidence)
)

print(
    "Explicit safety violations : 0 PASS"
)

print(
    "Official test              : SEALED PASS"
)


# ==================================================================================================
# 13. BUILD FINAL 36-RUN LEDGER
# ==================================================================================================

print()
print("-" * 180)
print("[10] BUILD FINAL 36-RUN LEDGER")
print("-" * 180)


ledger_fields = [
    "relative_path",
    "dataset_raw",
    "dataset",
    "dataset_slug",
    "method",
    "method_dir",
    "seed",
    "status",
    "tau",
    "gamma",
    "C",
    "best_AA_epoch",
    "best_AA",
    "best_AA_Macro_F1",
    "best_AA_OA",
    "best_OA_epoch",
    "best_OA",
    "completion_sha256",
]


ledger_buffer = io.StringIO()


writer = csv.DictWriter(
    ledger_buffer,
    fieldnames=ledger_fields,
    lineterminator="\n",
)


writer.writeheader()


for record in sorted(
    run_records,
    key=lambda r: (
        r["method_dir"],
        r["dataset_slug"],
        r["seed"],
        r["relative_path"],
    ),
):

    writer.writerow(
        {
            key:
                record.get(
                    key,
                    ""
                )
            for key in ledger_fields
        }
    )


ledger_bytes = ledger_buffer.getvalue().encode(
    "utf-8"
)


ledger_sha = hashlib.sha256(
    ledger_bytes
).hexdigest()


print(
    "Run-ledger rows    : 36 PASS"
)

print(
    "Run-ledger SHA-256 :",
    ledger_sha
)


# ==================================================================================================
# 14. BUILD FINAL CORRECTED CLOSING AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[11] BUILD FINAL CORRECTED CLOSING-AUDIT SEAL")
print("-" * 180)


audit_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_NOTEBOOK06A_EXPERIMENT_B_FINAL_CLOSING_AUDIT",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "closing_audit_execution": {
        "cell":
            "15A",

        "correction_of":
            "Cell 15",

        "correction_reason":
            (
                "Cell 15 incorrectly treated completion-manifest "
                "dataset tokens as case-sensitive. Cell 15A uses "
                "case-insensitive canonical normalization and also "
                "cross-checks every normalized dataset token against "
                "the dataset slug encoded in its run path."
            ),

        "scientific_state_modified":
            False,
    },

    "final_state": {
        "experiment_b_progress":
            "36/36",

        "experiment_b_complete":
            True,

        "gpu_training_remaining":
            False,

        "project_file_count":
            507,

        "base_project_file_count":
            449,

        "final_cumulative_delta_file_count":
            58,
    },

    "snapshot_authentication": {
        "final_36of36_persistence_zip":
            str(
                FINAL_PERSISTENCE_ZIP
            ),

        "final_36of36_persistence_zip_sha256":
            final_zip_sha,

        "final_36of36_persistence_manifest":
            str(
                FINAL_PERSISTENCE_MANIFEST
            ),

        "final_36of36_persistence_manifest_sha256":
            final_manifest_sha,

        "base_internal_manifest_sha256":
            base_manifest_sha,

        "full_current_project_sha_verification":
            "507/507 PASS",

        "missing_expected_paths":
            0,

        "unexpected_paths":
            0,
    },

    "completion_ledger": {
        "CE":
            6,

        "Focal":
            8,

        "LDAM-DRW":
            8,

        "Balanced Softmax":
            6,

        "LA-loss":
            8,

        "total":
            36,
    },

    "completion_manifest_dataset_audit": {
        "case_insensitive_normalization":
            True,

        "manifest_path_dataset_agreement":
            "36/36 PASS",

        "canonical_datasets":
            [
                "Pingan",
                "Qingyun",
            ],

        "observed_raw_dataset_tokens":
            dataset_token_counts,
    },

    "la_loss": {
        "run_matrix":
            "8/8 EXACT PASS",

        "Pingan":
            selection_summary[
                "Pingan"
            ],

        "Qingyun":
            selection_summary[
                "Qingyun"
            ],

        "qingyun_seed123_completion_sha256":
            EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256,

        "qingyun_seed3407_completion_sha256":
            EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256,
    },

    "official_test": {
        "status":
            "SEALED",

        "project_json_files_scanned":
            json_files_scanned,

        "explicit_safety_assertions":
            len(
                safety_evidence
            ),

        "explicit_safety_violations":
            0,

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,
    },

    "protocol_integrity": {
        "training_performed_in_closing_audit":
            False,

        "model_inference_performed_in_closing_audit":
            False,

        "project_mutated_by_closing_audit":
            False,

        "retuning_performed":
            False,

        "mixed_precision_changed":
            False,

        "multi_gpu_used":
            False,
    },

    "external_run_ledger": {
        "file":
            str(
                FINAL_RUN_LEDGER_CSV
            ),

        "rows":
            36,

        "sha256":
            ledger_sha,
    },

    "closing_status":
        "PASS",

    "seal":
        (
            "EXPERIMENT_B_36_OF_36_COMPLETE__"
            "FULL_507_FILE_SNAPSHOT_VERIFIED__"
            "36_COMPLETION_RECORDS_AUDITED__"
            "DATASET_TOKENS_CANONICALLY_NORMALIZED_AND_PATH_VERIFIED__"
            "LA_SELECTIONS_FROZEN__"
            "OFFICIAL_TEST_SEALED__"
            "NO_GPU_TRAINING_REMAINING"
        ),
}


audit_bytes = json.dumps(
    audit_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


audit_sha = hashlib.sha256(
    audit_bytes
).hexdigest()


print(
    "Corrected closing-audit SHA-256:",
    audit_sha
)


# ==================================================================================================
# 15. WRITE FINAL EXTERNAL AUDIT PACKAGE
# ==================================================================================================

print()
print("-" * 180)
print("[12] WRITE FINAL EXTERNAL AUDIT PACKAGE TO GOOGLE DRIVE")
print("-" * 180)


FINAL_AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    FINAL_AUDIT_JSON,
    FINAL_RUN_LEDGER_CSV,
    FINAL_AUDIT_SHA,
]:

    if path.exists():

        raise RuntimeError(
            "STOP: final audit artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    FINAL_RUN_LEDGER_CSV,
    ledger_bytes,
)


atomic_write_bytes(
    FINAL_AUDIT_JSON,
    audit_bytes,
)


sidecar_text = (
    f"{audit_sha}  {FINAL_AUDIT_JSON.name}\n"
    f"{ledger_sha}  {FINAL_RUN_LEDGER_CSV.name}\n"
    f"{final_zip_sha}  {FINAL_PERSISTENCE_ZIP.name}\n"
    f"{final_manifest_sha}  {FINAL_PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    FINAL_AUDIT_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


if sha256_file(
    FINAL_AUDIT_JSON
) != audit_sha:

    raise RuntimeError(
        "STOP: written final audit SHA verification failed."
    )


if sha256_file(
    FINAL_RUN_LEDGER_CSV
) != ledger_sha:

    raise RuntimeError(
        "STOP: written final ledger SHA verification failed."
    )


# ==================================================================================================
# 16. PROVE NO SCIENTIFIC PROJECT MUTATION
# ==================================================================================================

final_project_files = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    final_project_files
) != 507:

    raise RuntimeError(
        "STOP: correction audit changed project footprint."
    )


# ==================================================================================================
# 17. TERMINAL GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 15A TERMINAL GATE")
print("=" * 180)

print(
    "Cell 15 failure classification             : AUDIT-CODE SEMANTIC DEFECT"
)

print(
    "Scientific-state failure                   : False"
)

print(
    "Scientific artifacts modified by correction: False"
)

print()
print(
    "Final persistence                          : AUTHENTICATED PASS"
)

print(
    "Exact final project                        : 507 files"
)

print(
    "Full project SHA verification              : 507/507 PASS"
)

print(
    "Missing expected files                     : 0"
)

print(
    "Unexpected project files                   : 0"
)

print()
print(
    "CE                                         : 6/6 PASS"
)

print(
    "Focal                                      : 8/8 PASS"
)

print(
    "LDAM-DRW                                   : 8/8 PASS"
)

print(
    "Balanced Softmax                           : 6/6 PASS"
)

print(
    "LA-loss                                    : 8/8 PASS"
)

print(
    "Experiment-B                               : 36/36 COMPLETE"
)

print()
print(
    "Completion dataset normalization           : 36/36 PASS"
)

print(
    "Manifest/path dataset agreement            : 36/36 PASS"
)

print(
    "Exact LA-loss run matrix                   : 8/8 PASS"
)

print(
    "Pingan selected tau                        : 1.0 PASS"
)

print(
    "Qingyun selected tau                       : 0.5 PASS"
)

print()
print(
    "Official-test safety violations            : 0"
)

print(
    "Official test                              : SEALED"
)

print(
    "Official-test inference                    : NOT PERFORMED"
)

print(
    "Training in correction audit               : False"
)

print(
    "GPU training remaining                     : False"
)

print()
print(
    "FINAL CLOSING AUDIT:"
)

print(
    FINAL_AUDIT_JSON
)

print()
print(
    "FINAL CLOSING AUDIT SHA-256:"
)

print(
    audit_sha
)

print()
print(
    "FINAL 36-RUN LEDGER:"
)

print(
    FINAL_RUN_LEDGER_CSV
)

print()
print(
    "FINAL 36-RUN LEDGER SHA-256:"
)

print(
    ledger_sha
)

print()
print(
    "FINAL AUDIT SHA SIDECAR:"
)

print(
    FINAL_AUDIT_SHA
)

print()
print(
    "FINAL CORRECTION CELL 15A PASS: True"
)

print()
print(
    "STATUS: "
    "CELL15_AUDIT_BUG_CORRECTED__"
    "EXPERIMENT_B_36_OF_36_COMPLETE__"
    "FULL_507_FILE_SNAPSHOT_VERIFIED__"
    "36_COMPLETION_RECORDS_AUDITED__"
    "DATASET_TOKENS_CANONICALLY_NORMALIZED_AND_PATH_VERIFIED__"
    "LA_SELECTIONS_FROZEN__"
    "OFFICIAL_TEST_SEALED__"
    "NO_GPU_TRAINING_REMAINING__"
    "FINAL_CLOSING_AUDIT_WRITTEN_TO_GOOGLE_DRIVE"
)

print("=" * 180)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CORRECTION CELL 15B
#
# FINAL SCHEMA-TOLERANT CLOSING AUDIT
#
# Cell 15 defect:
#   Assumed manifest["dataset"] used exact display-case strings.
#
# Cell 15A defect:
#   Assumed manifest["dataset"] was mandatory.
#
# Historical manifests are allowed to omit redundant dataset/seed metadata because
# their authenticated run paths already encode those identities.
#
# AUTHORITATIVE IDENTITY FOR THIS AUDIT:
#   runs/experiment_b_spatial_validation_v1/
#       <method>/<dataset>/.../<seed>/...completion_manifest.json
#
# RULES:
#   - dataset is derived from authenticated path
#   - seed is derived from authenticated path
#   - if manifest dataset exists, it MUST agree with the path
#   - if manifest seed exists, it MUST agree with the path
#   - missing optional redundant metadata does NOT invalidate historical runs
#
# NO TRAINING
# NO MODEL INFERENCE
# NO RETUNING
# NO OFFICIAL-TEST ACCESS
# NO SCIENTIFIC PROJECT MUTATION
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import csv
import hashlib
import io
import json
import os
import re
import zipfile


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 15B")
print("SCHEMA-TOLERANT FINAL 36/36 CLOSING AUDIT")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path("/content/LTLC")

LOCAL_MIGRATION_ZIP = Path(
    "/content/LTLC_Kaggle_Migration_v1.zip"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

FINAL_PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "36of36_Post_Qingyun_LA_tau0p5_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

FINAL_PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}.zip"
)

FINAL_PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__MANIFEST.json"
)

FINAL_PERSISTENCE_SHA_SIDECAR = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__SHA256.txt"
)


FINAL_AUDIT_DIR = (
    DRIVE_ROOT
    / "final_audits"
    / "notebook06a_experiment_b"
)

FINAL_AUDIT_JSON = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1.json"
)

FINAL_RUN_LEDGER_CSV = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_RUN_LEDGER_v1.csv"
)

FINAL_AUDIT_SHA = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1__SHA256.txt"
)


# ==================================================================================================
# 2. AUTHENTICATED IDENTITIES
# ==================================================================================================

EXPECTED_FINAL_ZIP_SHA256 = (
    "30ba228994f0a16e756c5ade22480b00"
    "f3e2304ee3e15bd8e00409d86dc2b7b8"
)

EXPECTED_FINAL_MANIFEST_SHA256 = (
    "5f93f457f3b1510d72b366aa8a78aa65"
    "fbf59e866dadfcefd7388d80f81c79b9"
)

EXPECTED_BASE_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)

EXPECTED_PINGAN_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)

EXPECTED_QINGYUN_SELECTION_SHA256 = (
    "367d6ecd113d786bf89510985e0f0ff7"
    "e6589a68db9c912d4c36fb29d02fae89"
)

EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256 = (
    "1e9ac68021a9d01d59818c7061120bc9"
    "d511e1957f02b7a96ef3eb442933524d"
)

EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256 = (
    "d99a05fa4c349267c542fae7a8007024"
    "cdcbb2e773ad9d817909ac43d538bcd7"
)


BASE_PROJECT_FILE_COUNT = 449
FINAL_DELTA_FILE_COUNT = 58
FINAL_PROJECT_FILE_COUNT = 507

BASE_INTERNAL_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)

DELTA_INTERNAL_MANIFEST = (
    "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
)


DATASET_CANONICAL = {
    "pingan": "Pingan",
    "qingyun": "Qingyun",
}


METHOD_CONTRACT = {
    "CE": ("ce", 6),
    "Focal": ("focal", 8),
    "LDAM-DRW": ("ldam_drw", 8),
    "Balanced Softmax": ("balanced_softmax", 6),
    "LA-loss": ("la_loss", 8),
}


ALLOWED_SEEDS = {
    42,
    123,
    3407,
}


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def atomic_write_bytes(path, payload):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(str(path) + ".tmp")

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:
        f.write(payload)

    os.replace(tmp, path)


def canonical_rel(value):

    p = PurePosixPath(str(value))

    if p.is_absolute():
        raise RuntimeError(
            f"STOP: absolute project-relative path rejected: {value}"
        )

    if ".." in p.parts:
        raise RuntimeError(
            f"STOP: parent traversal rejected: {value}"
        )

    return str(p)


def extract_path_identity(relative_path):

    parts = PurePosixPath(
        relative_path
    ).parts

    if len(parts) < 6:

        raise RuntimeError(
            "STOP: malformed completion path:\n"
            f"{relative_path}"
        )

    if (
        parts[0] != "runs"
        or
        parts[1] != "experiment_b_spatial_validation_v1"
    ):

        raise RuntimeError(
            "STOP: completion outside Experiment-B run tree:\n"
            f"{relative_path}"
        )


    method_dir = parts[2]
    dataset_slug = parts[3].lower()


    if dataset_slug not in DATASET_CANONICAL:

        raise RuntimeError(
            "STOP: unsupported dataset slug in completion path:\n"
            f"{relative_path}"
        )


    seed_matches = []

    for part in parts:

        match = re.fullmatch(
            r"seed(\d+)",
            part,
        )

        if match:

            seed_matches.append(
                int(
                    match.group(1)
                )
            )


    if len(seed_matches) != 1:

        raise RuntimeError(
            "STOP: expected exactly one seed directory in completion path:\n"
            f"{relative_path}\n"
            f"seed_matches={seed_matches}"
        )


    seed = seed_matches[0]


    if seed not in ALLOWED_SEEDS:

        raise RuntimeError(
            "STOP: unauthorized seed encoded in path:\n"
            f"{relative_path}"
        )


    return {
        "method_dir":
            method_dir,

        "dataset_slug":
            dataset_slug,

        "dataset":
            DATASET_CANONICAL[
                dataset_slug
            ],

        "seed":
            seed,
    }


FALSE_SAFETY_KEYS = {
    "official_test_used",
    "official_test_used_for_selection",
    "official_test_values_accessed",
    "official_test_accessed",
    "official_test_opened",
    "official_test_may_be_opened",
    "test_dataset_created",
    "test_dataloader_created",
    "test_inference_performed",
}


def scan_safety_tree(
    obj,
    source,
    evidence,
    violations,
):

    if isinstance(obj, dict):

        for key, value in obj.items():

            key_l = str(key).lower()


            if key_l in FALSE_SAFETY_KEYS:

                evidence.append(
                    {
                        "source":
                            source,

                        "key":
                            key_l,

                        "value":
                            value,
                    }
                )


                if value is not False:

                    violations.append(
                        {
                            "source":
                                source,

                            "key":
                                key_l,

                            "value":
                                value,
                        }
                    )


            elif key_l == "official_test_status":

                evidence.append(
                    {
                        "source":
                            source,

                        "key":
                            key_l,

                        "value":
                            value,
                    }
                )


                if str(value).upper() != "SEALED":

                    violations.append(
                        {
                            "source":
                                source,

                            "key":
                                key_l,

                            "value":
                                value,
                        }
                    )


            scan_safety_tree(
                value,
                source,
                evidence,
                violations,
            )


    elif isinstance(obj, list):

        for item in obj:

            scan_safety_tree(
                item,
                source,
                evidence,
                violations,
            )


# ==================================================================================================
# 4. CLASSIFY EARLIER CLOSING-AUDIT FAILURES
# ==================================================================================================

print()
print("-" * 180)
print("[1] CLOSING-AUDIT CORRECTION SEMANTICS")
print("-" * 180)

print(
    "Cell 15 failure  : case-sensitive redundant dataset-field assumption"
)

print(
    "Cell 15A failure : mandatory redundant dataset-field assumption"
)

print(
    "Scientific-state failure : False"
)

print(
    "Correction principle     : authenticated run path is authoritative"
)

print(
    "Training required        : False"
)

print(
    "Project mutation required: False"
)


# ==================================================================================================
# 5. AUTHENTICATE FINAL SAVED 36/36 PACKAGE
# ==================================================================================================

print()
print("-" * 180)
print("[2] AUTHENTICATE EXTERNALLY SAVED 36/36 PACKAGE")
print("-" * 180)


required_artifacts = [
    (
        "Local migration ZIP",
        LOCAL_MIGRATION_ZIP,
    ),
    (
        "36/36 persistence ZIP",
        FINAL_PERSISTENCE_ZIP,
    ),
    (
        "36/36 persistence manifest",
        FINAL_PERSISTENCE_MANIFEST,
    ),
    (
        "36/36 SHA sidecar",
        FINAL_PERSISTENCE_SHA_SIDECAR,
    ),
]


for label, path in required_artifacts:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: missing required artifact: {label}\n{path}"
        )

    print(
        f"{label:32s}: FOUND"
    )


if not ROOT.is_dir():

    raise FileNotFoundError(
        "STOP: /content/LTLC missing."
    )


final_zip_sha = sha256_file(
    FINAL_PERSISTENCE_ZIP
)

final_manifest_sha = sha256_file(
    FINAL_PERSISTENCE_MANIFEST
)


if final_zip_sha != EXPECTED_FINAL_ZIP_SHA256:

    raise RuntimeError(
        "STOP: final 36/36 ZIP SHA mismatch."
    )


if final_manifest_sha != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: final 36/36 manifest SHA mismatch."
    )


with FINAL_PERSISTENCE_MANIFEST.open(
    "r",
    encoding="utf-8",
) as f:

    final_manifest = json.load(f)


if (
    final_manifest[
        "scientific_progress"
    ][
        "after"
    ]
    != 36
):

    raise RuntimeError(
        "STOP: final persistence is not 36/36."
    )


if (
    final_manifest[
        "scientific_progress"
    ].get(
        "experiment_b_complete"
    )
    is not True
):

    raise RuntimeError(
        "STOP: final persistence does not mark Experiment B complete."
    )


if (
    final_manifest[
        "next_action"
    ].get(
        "gpu_training_remaining"
    )
    is not False
):

    raise RuntimeError(
        "STOP: final persistence indicates remaining GPU training."
    )


final_safety = final_manifest[
    "safety"
]


if not (
    final_safety.get(
        "official_test_status"
    ) == "SEALED"
    and
    final_safety.get(
        "official_test_values_accessed"
    ) is False
    and
    final_safety.get(
        "test_dataset_created"
    ) is False
    and
    final_safety.get(
        "test_dataloader_created"
    ) is False
    and
    final_safety.get(
        "test_inference_performed"
    ) is False
):

    raise RuntimeError(
        "STOP: final persistence official-test gate failed."
    )


print()
print(
    "Final persistence ZIP      : AUTHENTICATED PASS"
)

print(
    "Final persistence manifest : AUTHENTICATED PASS"
)

print(
    "Experiment-B               : 36/36 COMPLETE"
)

print(
    "GPU training remaining     : False"
)

print(
    "Official test              : SEALED"
)


# ==================================================================================================
# 6. RECONSTRUCT EXACT 507-FILE CONTRACT
# ==================================================================================================

print()
print("-" * 180)
print("[3] RECONSTRUCT EXACT FINAL 507-FILE CONTRACT")
print("-" * 180)


with zipfile.ZipFile(
    LOCAL_MIGRATION_ZIP,
    "r",
) as zf:

    base_manifest_bytes = zf.read(
        BASE_INTERNAL_MANIFEST
    )


base_manifest_sha = hashlib.sha256(
    base_manifest_bytes
).hexdigest()


if base_manifest_sha != EXPECTED_BASE_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: authenticated base manifest SHA mismatch."
    )


base_manifest = json.loads(
    base_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    base_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


with zipfile.ZipFile(
    FINAL_PERSISTENCE_ZIP,
    "r",
) as zf:

    delta_manifest_bytes = zf.read(
        DELTA_INTERNAL_MANIFEST
    )


if hashlib.sha256(
    delta_manifest_bytes
).hexdigest() != EXPECTED_FINAL_MANIFEST_SHA256:

    raise RuntimeError(
        "STOP: internal final-delta manifest SHA mismatch."
    )


if (
    delta_manifest_bytes
    !=
    FINAL_PERSISTENCE_MANIFEST.read_bytes()
):

    raise RuntimeError(
        "STOP: internal/external final persistence manifests differ."
    )


delta_records = (
    final_manifest[
        "full_delta"
    ][
        "records"
    ]
)


if len(base_records) != BASE_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: base snapshot is not 449 files."
    )


if len(delta_records) != FINAL_DELTA_FILE_COUNT:

    raise RuntimeError(
        "STOP: final delta is not 58 files."
    )


base_map = {}


for record in base_records:

    rel = canonical_rel(
        record[
            "relative_path"
        ]
    )

    if rel in base_map:

        raise RuntimeError(
            f"STOP: duplicate base path: {rel}"
        )

    base_map[rel] = record


delta_map = {}


for record in delta_records:

    rel = canonical_rel(
        record[
            "relative_path"
        ]
    )

    if rel in delta_map:

        raise RuntimeError(
            f"STOP: duplicate final-delta path: {rel}"
        )

    delta_map[rel] = record


if set(base_map) & set(delta_map):

    raise RuntimeError(
        "STOP: base/final-delta path overlap detected."
    )


expected_map = {
    **base_map,
    **delta_map,
}


if len(expected_map) != FINAL_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: final expected contract is not 507 files."
    )


print(
    "Authenticated base records : 449 PASS"
)

print(
    "Final delta records         : 58 PASS"
)

print(
    "Base/delta overlap          : 0 PASS"
)

print(
    "Expected project contract   : 507 files PASS"
)


# ==================================================================================================
# 7. FULL EXACT PROJECT VERIFICATION
# ==================================================================================================

print()
print("-" * 180)
print("[4] EXACT FINAL PROJECT PATH + SHA AUDIT")
print("-" * 180)


current_files = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


current_paths = {
    str(
        p.relative_to(ROOT)
    )
    for p in current_files
}


expected_paths = set(
    expected_map
)


missing_paths = sorted(
    expected_paths
    -
    current_paths
)

unexpected_paths = sorted(
    current_paths
    -
    expected_paths
)


if missing_paths:

    raise RuntimeError(
        "STOP: missing expected final project files:\n"
        +
        "\n".join(
            missing_paths[:20]
        )
    )


if unexpected_paths:

    raise RuntimeError(
        "STOP: unexpected final project files:\n"
        +
        "\n".join(
            unexpected_paths[:20]
        )
    )


if len(current_paths) != FINAL_PROJECT_FILE_COUNT:

    raise RuntimeError(
        "STOP: final project is not 507 files."
    )


verified_files = 0
verified_bytes = 0


for relative_path in sorted(
    expected_map
):

    record = expected_map[
        relative_path
    ]

    path = (
        ROOT
        /
        relative_path
    )


    if sha256_file(
        path
    ) != record[
        "sha256"
    ]:

        raise RuntimeError(
            "STOP: final project SHA mismatch:\n"
            f"{relative_path}"
        )


    if (
        "size_bytes" in record
        and
        int(
            path.stat().st_size
        )
        !=
        int(
            record[
                "size_bytes"
            ]
        )
    ):

        raise RuntimeError(
            "STOP: final project size mismatch:\n"
            f"{relative_path}"
        )


    verified_files += 1
    verified_bytes += int(
        path.stat().st_size
    )


print(
    "Current project files     : 507 PASS"
)

print(
    "Missing expected files    : 0 PASS"
)

print(
    "Unexpected files          : 0 PASS"
)

print(
    "Full SHA verification     : 507/507 PASS"
)

print(
    "Verified scientific bytes :",
    verified_bytes,
)


# ==================================================================================================
# 8. FINAL METHOD LEDGER
# ==================================================================================================

print()
print("-" * 180)
print("[5] FINAL 36-RUN METHOD LEDGER")
print("-" * 180)


RUN_ROOT = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
)


completion_paths_by_method = {}


for label, (
    method_dir,
    expected_count,
) in METHOD_CONTRACT.items():

    method_root = (
        RUN_ROOT
        /
        method_dir
    )


    paths = sorted(
        method_root.rglob(
            "*__completion_manifest.json"
        )
    )


    completion_paths_by_method[
        label
    ] = paths


    if len(paths) != expected_count:

        raise RuntimeError(
            f"STOP: {label} completion count mismatch."
        )


    print(
        f"{label:18s}: "
        f"{len(paths)}/{expected_count} PASS"
    )


all_completion_paths = []


for paths in completion_paths_by_method.values():

    all_completion_paths.extend(
        paths
    )


if len(all_completion_paths) != 36:

    raise RuntimeError(
        "STOP: total Experiment-B completion count is not 36."
    )


print()
print(
    "Experiment-B completion manifests: 36/36 PASS"
)


# ==================================================================================================
# 9. SCHEMA-TOLERANT AUDIT OF ALL 36 COMPLETION MANIFESTS
# ==================================================================================================

print()
print("-" * 180)
print("[6] SCHEMA-TOLERANT ALL-36 COMPLETION AUDIT")
print("-" * 180)


run_records = []

dataset_field_present = 0
dataset_field_absent = 0
dataset_field_match = 0

seed_field_present = 0
seed_field_absent = 0
seed_field_match = 0

raw_dataset_tokens = {}


for completion_path in sorted(
    all_completion_paths
):

    relative_path = str(
        completion_path.relative_to(
            ROOT
        )
    )


    path_identity = extract_path_identity(
        relative_path
    )


    if (
        path_identity[
            "method_dir"
        ]
        not in {
            contract[0]
            for contract in METHOD_CONTRACT.values()
        }
    ):

        raise RuntimeError(
            "STOP: unauthorized method directory:\n"
            f"{relative_path}"
        )


    with completion_path.open(
        "r",
        encoding="utf-8",
    ) as f:

        manifest = json.load(
            f
        )


    if manifest.get(
        "status"
    ) != "COMPLETE":

        raise RuntimeError(
            "STOP: completion manifest is not COMPLETE:\n"
            f"{relative_path}"
        )


    # ----------------------------------------------------------------------------------------------
    # OPTIONAL REDUNDANT DATASET FIELD
    # ----------------------------------------------------------------------------------------------

    raw_dataset = manifest.get(
        "dataset",
        None,
    )


    if (
        raw_dataset is None
        or
        str(
            raw_dataset
        ).strip() == ""
    ):

        dataset_field_absent += 1
        manifest_dataset_normalized = None

    else:

        dataset_field_present += 1

        manifest_dataset_normalized = str(
            raw_dataset
        ).strip().lower()


        raw_dataset_tokens[
            str(
                raw_dataset
            )
        ] = (
            raw_dataset_tokens.get(
                str(
                    raw_dataset
                ),
                0,
            )
            +
            1
        )


        if (
            manifest_dataset_normalized
            not in
            DATASET_CANONICAL
        ):

            raise RuntimeError(
                "STOP: unsupported non-empty dataset field:\n"
                f"{relative_path}\n"
                f"dataset={raw_dataset!r}"
            )


        if (
            manifest_dataset_normalized
            !=
            path_identity[
                "dataset_slug"
            ]
        ):

            raise RuntimeError(
                "STOP: manifest/path dataset disagreement:\n"
                f"{relative_path}\n"
                f"manifest={raw_dataset!r}\n"
                f"path={path_identity['dataset_slug']!r}"
            )


        dataset_field_match += 1


    # ----------------------------------------------------------------------------------------------
    # OPTIONAL REDUNDANT SEED FIELD
    # ----------------------------------------------------------------------------------------------

    raw_seed = manifest.get(
        "seed",
        None,
    )


    if raw_seed is None:

        seed_field_absent += 1
        manifest_seed = None

    else:

        seed_field_present += 1


        try:

            manifest_seed = int(
                raw_seed
            )

        except Exception:

            raise RuntimeError(
                "STOP: non-numeric seed field:\n"
                f"{relative_path}\n"
                f"seed={raw_seed!r}"
            )


        if (
            manifest_seed
            !=
            path_identity[
                "seed"
            ]
        ):

            raise RuntimeError(
                "STOP: manifest/path seed disagreement:\n"
                f"{relative_path}\n"
                f"manifest={manifest_seed}\n"
                f"path={path_identity['seed']}"
            )


        seed_field_match += 1


    checkpoint_policy = manifest.get(
        "checkpoint_policy",
        {},
    )


    best_aa_metrics = checkpoint_policy.get(
        "best_AA_metrics",
        {},
    )

    best_oa_metrics = checkpoint_policy.get(
        "best_OA_metrics",
        {},
    )


    run_records.append(
        {
            "relative_path":
                relative_path,

            "method_dir":
                path_identity[
                    "method_dir"
                ],

            "dataset":
                path_identity[
                    "dataset"
                ],

            "dataset_slug":
                path_identity[
                    "dataset_slug"
                ],

            "seed":
                path_identity[
                    "seed"
                ],

            "status":
                manifest.get(
                    "status"
                ),

            "manifest_dataset_present":
                raw_dataset is not None
                and
                str(
                    raw_dataset
                ).strip() != "",

            "manifest_dataset_raw":
                ""
                if raw_dataset is None
                else raw_dataset,

            "manifest_seed_present":
                raw_seed is not None,

            "manifest_seed_raw":
                ""
                if raw_seed is None
                else raw_seed,

            "method_manifest":
                manifest.get(
                    "method",
                    ""
                ),

            "tau":
                manifest.get(
                    "tau",
                    ""
                ),

            "gamma":
                manifest.get(
                    "gamma",
                    ""
                ),

            "C":
                manifest.get(
                    "C",
                    manifest.get(
                        "c",
                        ""
                    ),
                ),

            "best_AA_epoch":
                checkpoint_policy.get(
                    "best_AA_epoch",
                    ""
                ),

            "best_AA":
                best_aa_metrics.get(
                    "AA",
                    ""
                ),

            "best_AA_Macro_F1":
                best_aa_metrics.get(
                    "Macro_F1",
                    ""
                ),

            "best_AA_OA":
                best_aa_metrics.get(
                    "OA",
                    ""
                ),

            "best_OA_epoch":
                checkpoint_policy.get(
                    "best_OA_epoch",
                    ""
                ),

            "best_OA":
                best_oa_metrics.get(
                    "OA",
                    ""
                ),

            "completion_sha256":
                sha256_file(
                    completion_path
                ),
        }
    )


if len(
    run_records
) != 36:

    raise RuntimeError(
        "STOP: schema-tolerant audit did not produce 36 records."
    )


if len(
    {
        record[
            "relative_path"
        ]
        for record in run_records
    }
) != 36:

    raise RuntimeError(
        "STOP: duplicate completion-manifest path detected."
    )


print(
    "Completion status COMPLETE       : 36/36 PASS"
)

print(
    "Path-derived dataset identities  : 36/36 PASS"
)

print(
    "Path-derived seed identities     : 36/36 PASS"
)

print(
    "Unique completion records        : 36/36 PASS"
)

print()
print(
    "Optional manifest dataset field:"
)

print(
    "  present :",
    dataset_field_present
)

print(
    "  absent  :",
    dataset_field_absent
)

print(
    "  present-and-path-matching :",
    dataset_field_match,
)


if (
    dataset_field_match
    !=
    dataset_field_present
):

    raise RuntimeError(
        "STOP: not all present dataset fields matched authenticated paths."
    )


print()
print(
    "Optional manifest seed field:"
)

print(
    "  present :",
    seed_field_present
)

print(
    "  absent  :",
    seed_field_absent
)

print(
    "  present-and-path-matching :",
    seed_field_match,
)


if seed_field_match != seed_field_present:

    raise RuntimeError(
        "STOP: not all present seed fields matched authenticated paths."
    )


# ==================================================================================================
# 10. DATASET / METHOD / SEED STRUCTURE
# ==================================================================================================

print()
print("-" * 180)
print("[7] FINAL RUN-STRUCTURE AUDIT")
print("-" * 180)


for dataset_slug in [
    "pingan",
    "qingyun",
]:

    ce_seeds = {
        record["seed"]
        for record in run_records
        if (
            record["method_dir"] == "ce"
            and
            record["dataset_slug"] == dataset_slug
        )
    }


    if ce_seeds != {
        42,
        123,
        3407,
    }:

        raise RuntimeError(
            f"STOP: CE seed matrix invalid for {dataset_slug}."
        )


    bs_seeds = {
        record["seed"]
        for record in run_records
        if (
            record["method_dir"]
            ==
            "balanced_softmax"
            and
            record["dataset_slug"]
            ==
            dataset_slug
        )
    }


    if bs_seeds != {
        42,
        123,
        3407,
    }:

        raise RuntimeError(
            f"STOP: Balanced Softmax seed matrix invalid for {dataset_slug}."
        )


    focal_records = [
        record
        for record in run_records
        if (
            record["method_dir"] == "focal"
            and
            record["dataset_slug"] == dataset_slug
        )
    ]


    if (
        len(
            focal_records
        )
        != 4
    ):

        raise RuntimeError(
            f"STOP: Focal run count invalid for {dataset_slug}."
        )


    if (
        sum(
            record["seed"] == 42
            for record in focal_records
        )
        != 2
        or
        sum(
            record["seed"] == 123
            for record in focal_records
        )
        != 1
        or
        sum(
            record["seed"] == 3407
            for record in focal_records
        )
        != 1
    ):

        raise RuntimeError(
            f"STOP: Focal tuning/seed structure invalid for {dataset_slug}."
        )


    ldam_records = [
        record
        for record in run_records
        if (
            record["method_dir"] == "ldam_drw"
            and
            record["dataset_slug"] == dataset_slug
        )
    ]


    if len(
        ldam_records
    ) != 4:

        raise RuntimeError(
            f"STOP: LDAM-DRW run count invalid for {dataset_slug}."
        )


    if (
        sum(
            record["seed"] == 42
            for record in ldam_records
        )
        != 2
        or
        sum(
            record["seed"] == 123
            for record in ldam_records
        )
        != 1
        or
        sum(
            record["seed"] == 3407
            for record in ldam_records
        )
        != 1
    ):

        raise RuntimeError(
            f"STOP: LDAM-DRW tuning/seed structure invalid for {dataset_slug}."
        )


print(
    "CE seed matrices                 : PASS"
)

print(
    "Balanced Softmax seed matrices   : PASS"
)

print(
    "Focal tuning/selected-seed shape : PASS"
)

print(
    "LDAM-DRW tuning/selected-seed shape: PASS"
)


# ==================================================================================================
# 11. EXACT LA-LOSS 8-RUN MATRIX
# ==================================================================================================

print()
print("-" * 180)
print("[8] EXACT LA-LOSS 8-RUN MATRIX")
print("-" * 180)


expected_la_paths = {
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau0p5/seed42/"
        "la_loss__pingan__tau0p5__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau1p0/seed42/"
        "la_loss__pingan__tau1p0__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau1p0/seed123/"
        "la_loss__pingan__tau1p0__seed123__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/tau1p0/seed3407/"
        "la_loss__pingan__tau1p0__seed3407__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed42/"
        "la_loss__qingyun__tau0p5__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau1p0/seed42/"
        "la_loss__qingyun__tau1p0__seed42__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed123/"
        "la_loss__qingyun__tau0p5__seed123__completion_manifest.json"
    ),
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed3407/"
        "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json"
    ),
}


observed_la_paths = {
    str(
        path.relative_to(
            ROOT
        )
    )
    for path in completion_paths_by_method[
        "LA-loss"
    ]
}


if observed_la_paths != expected_la_paths:

    raise RuntimeError(
        "STOP: exact LA-loss 8-run matrix mismatch."
    )


print(
    "Pingan tau=0.5 seed42            : PASS"
)

print(
    "Pingan tau=1.0 seeds42/123/3407  : PASS"
)

print(
    "Qingyun tau=1.0 seed42           : PASS"
)

print(
    "Qingyun tau=0.5 seeds42/123/3407 : PASS"
)

print(
    "Exact LA-loss matrix             : 8/8 PASS"
)


# ==================================================================================================
# 12. FROZEN LA SELECTIONS
# ==================================================================================================

print()
print("-" * 180)
print("[9] FROZEN LA TAU-SELECTION SEAL")
print("-" * 180)


PINGAN_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "pingan"
    / "la_loss__pingan__selected_tau_v1.json"
)

QINGYUN_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "la_loss__qingyun__selected_tau_v1.json"
)


selection_summary = {}


for (
    dataset,
    path,
    expected_sha,
    expected_tau,
) in [
    (
        "Pingan",
        PINGAN_SELECTION,
        EXPECTED_PINGAN_SELECTION_SHA256,
        1.0,
    ),
    (
        "Qingyun",
        QINGYUN_SELECTION,
        EXPECTED_QINGYUN_SELECTION_SHA256,
        0.5,
    ),
]:

    if not path.is_file():

        raise FileNotFoundError(
            f"STOP: {dataset} selection manifest missing."
        )


    actual_sha = sha256_file(
        path
    )


    if actual_sha != expected_sha:

        raise RuntimeError(
            f"STOP: {dataset} selection SHA mismatch."
        )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        selection = json.load(
            f
        )


    if float(
        selection[
            "selected_tau"
        ]
    ) != expected_tau:

        raise RuntimeError(
            f"STOP: {dataset} selected tau changed."
        )


    if not (
        selection.get(
            "official_test_status"
        )
        == "SEALED"
        and
        selection.get(
            "official_test_values_accessed"
        )
        is False
        and
        selection.get(
            "official_test_used_for_selection"
        )
        is False
    ):

        raise RuntimeError(
            f"STOP: {dataset} selection official-test gate failed."
        )


    selection_summary[
        dataset
    ] = {
        "selected_tau":
            expected_tau,

        "selection_sha256":
            actual_sha,
    }


    print(
        f"{dataset:10s}: "
        f"tau={expected_tau} | SHA PASS | OFFICIAL TEST SEALED"
    )


# ==================================================================================================
# 13. FINAL QINGYUN COMPLETION IDENTITIES
# ==================================================================================================

print()
print("-" * 180)
print("[10] FINAL QINGYUN SELECTED-TAU COMPLETION IDENTITIES")
print("-" * 180)


QINGYUN_SEED123 = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed123"
    / "la_loss__qingyun__tau0p5__seed123__completion_manifest.json"
)

QINGYUN_SEED3407 = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "tau0p5"
    / "seed3407"
    / "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json"
)


if (
    sha256_file(
        QINGYUN_SEED123
    )
    !=
    EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed123 completion SHA mismatch."
    )


if (
    sha256_file(
        QINGYUN_SEED3407
    )
    !=
    EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256
):

    raise RuntimeError(
        "STOP: Qingyun seed3407 completion SHA mismatch."
    )


print(
    "Qingyun tau=0.5 seed123  : AUTHENTICATED PASS"
)

print(
    "Qingyun tau=0.5 seed3407 : AUTHENTICATED PASS"
)


# ==================================================================================================
# 14. PROJECT-WIDE EXPLICIT OFFICIAL-TEST SAFETY SCAN
# ==================================================================================================

print()
print("-" * 180)
print("[11] PROJECT-WIDE OFFICIAL-TEST SAFETY SCAN")
print("-" * 180)


safety_evidence = []
safety_violations = []
json_files_scanned = 0


for path in sorted(
    ROOT.rglob(
        "*.json"
    )
):

    relative_path = str(
        path.relative_to(
            ROOT
        )
    )


    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        payload = json.load(
            f
        )


    json_files_scanned += 1


    scan_safety_tree(
        payload,
        relative_path,
        safety_evidence,
        safety_violations,
    )


if safety_violations:

    print(
        json.dumps(
            safety_violations[:20],
            indent=2,
            sort_keys=True,
        )
    )

    raise RuntimeError(
        "STOP: explicit official-test safety violation detected."
    )


if not safety_evidence:

    raise RuntimeError(
        "STOP: no explicit official-test safety evidence found."
    )


print(
    "JSON files scanned         :",
    json_files_scanned
)

print(
    "Explicit safety assertions :",
    len(
        safety_evidence
    )
)

print(
    "Explicit safety violations : 0 PASS"
)

print(
    "Official test              : SEALED PASS"
)


# ==================================================================================================
# 15. BUILD EXTERNAL FINAL RUN LEDGER
# ==================================================================================================

print()
print("-" * 180)
print("[12] BUILD FINAL EXTERNAL 36-RUN LEDGER")
print("-" * 180)


ledger_fields = [
    "relative_path",
    "method_dir",
    "dataset",
    "dataset_slug",
    "seed",
    "status",
    "manifest_dataset_present",
    "manifest_dataset_raw",
    "manifest_seed_present",
    "manifest_seed_raw",
    "method_manifest",
    "tau",
    "gamma",
    "C",
    "best_AA_epoch",
    "best_AA",
    "best_AA_Macro_F1",
    "best_AA_OA",
    "best_OA_epoch",
    "best_OA",
    "completion_sha256",
]


ledger_buffer = io.StringIO()


writer = csv.DictWriter(
    ledger_buffer,
    fieldnames=ledger_fields,
    lineterminator="\n",
)


writer.writeheader()


for record in sorted(
    run_records,
    key=lambda r: (
        r["method_dir"],
        r["dataset_slug"],
        r["seed"],
        r["relative_path"],
    ),
):

    writer.writerow(
        {
            key:
                record.get(
                    key,
                    ""
                )
            for key in ledger_fields
        }
    )


ledger_bytes = ledger_buffer.getvalue().encode(
    "utf-8"
)


ledger_sha = hashlib.sha256(
    ledger_bytes
).hexdigest()


print(
    "Run-ledger rows    : 36 PASS"
)

print(
    "Run-ledger SHA-256 :",
    ledger_sha
)


# ==================================================================================================
# 16. BUILD FINAL CLOSING AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[13] BUILD FINAL NOTEBOOK06A EXPERIMENT-B CLOSING SEAL")
print("-" * 180)


audit_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_NOTEBOOK06A_EXPERIMENT_B_FINAL_CLOSING_AUDIT",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "closing_audit_execution": {
        "successful_cell":
            "15B",

        "prior_audit_corrections":
            [
                {
                    "cell":
                        "15",

                    "classification":
                        "AUDIT_CODE_SCHEMA_ASSUMPTION_DEFECT",

                    "issue":
                        "case-sensitive completion-manifest dataset field",
                },
                {
                    "cell":
                        "15A",

                    "classification":
                        "AUDIT_CODE_SCHEMA_ASSUMPTION_DEFECT",

                    "issue":
                        "assumed completion-manifest dataset field was mandatory",
                },
            ],

        "authoritative_identity_policy":
            (
                "Authenticated run path is authoritative for method, dataset, "
                "and seed identity. Redundant completion-manifest dataset and "
                "seed fields are consistency-checked when present and may be "
                "absent in historical manifest schemas."
            ),

        "scientific_state_modified":
            False,
    },

    "final_state": {
        "experiment_b_progress":
            "36/36",

        "experiment_b_complete":
            True,

        "gpu_training_remaining":
            False,

        "project_file_count":
            507,

        "base_project_file_count":
            449,

        "final_cumulative_delta_file_count":
            58,
    },

    "snapshot_authentication": {
        "final_persistence_zip":
            str(
                FINAL_PERSISTENCE_ZIP
            ),

        "final_persistence_zip_sha256":
            final_zip_sha,

        "final_persistence_manifest":
            str(
                FINAL_PERSISTENCE_MANIFEST
            ),

        "final_persistence_manifest_sha256":
            final_manifest_sha,

        "base_internal_manifest_sha256":
            base_manifest_sha,

        "current_project_sha_verification":
            "507/507 PASS",

        "missing_expected_paths":
            0,

        "unexpected_paths":
            0,

        "verified_project_bytes":
            verified_bytes,
    },

    "completion_ledger": {
        "CE":
            6,

        "Focal":
            8,

        "LDAM-DRW":
            8,

        "Balanced Softmax":
            6,

        "LA-loss":
            8,

        "total":
            36,
    },

    "historical_manifest_schema_audit": {
        "path_authoritative_identity":
            True,

        "dataset_field_present":
            dataset_field_present,

        "dataset_field_absent":
            dataset_field_absent,

        "present_dataset_fields_matching_path":
            dataset_field_match,

        "seed_field_present":
            seed_field_present,

        "seed_field_absent":
            seed_field_absent,

        "present_seed_fields_matching_path":
            seed_field_match,

        "path_derived_completion_identity":
            "36/36 PASS",
    },

    "run_structure": {
        "ce_seed_matrix":
            "PASS",

        "balanced_softmax_seed_matrix":
            "PASS",

        "focal_tuning_selected_seed_structure":
            "PASS",

        "ldam_drw_tuning_selected_seed_structure":
            "PASS",

        "la_loss_run_matrix":
            "8/8 EXACT PASS",
    },

    "la_loss": {
        "Pingan":
            selection_summary[
                "Pingan"
            ],

        "Qingyun":
            selection_summary[
                "Qingyun"
            ],

        "qingyun_seed123_completion_sha256":
            EXPECTED_QINGYUN_SEED123_COMPLETION_SHA256,

        "qingyun_seed3407_completion_sha256":
            EXPECTED_QINGYUN_SEED3407_COMPLETION_SHA256,
    },

    "official_test": {
        "status":
            "SEALED",

        "project_json_files_scanned":
            json_files_scanned,

        "explicit_safety_assertions":
            len(
                safety_evidence
            ),

        "explicit_safety_violations":
            0,

        "official_test_values_accessed":
            False,

        "test_dataset_created":
            False,

        "test_dataloader_created":
            False,

        "test_inference_performed":
            False,
    },

    "protocol_integrity": {
        "training_performed_in_closing_audit":
            False,

        "model_inference_performed_in_closing_audit":
            False,

        "project_mutated_by_closing_audit":
            False,

        "retuning_performed":
            False,

        "gpu_training_remaining":
            False,
    },

    "external_run_ledger": {
        "file":
            str(
                FINAL_RUN_LEDGER_CSV
            ),

        "rows":
            36,

        "sha256":
            ledger_sha,
    },

    "closing_status":
        "PASS",

    "seal":
        (
            "EXPERIMENT_B_36_OF_36_COMPLETE__"
            "FULL_507_FILE_SNAPSHOT_VERIFIED__"
            "36_COMPLETION_RECORDS_PATH_AUTHENTICATED__"
            "HISTORICAL_MANIFEST_SCHEMAS_TOLERATED_WITH_CONSISTENCY_CHECKS__"
            "EXACT_LA_8_OF_8_MATRIX__"
            "LA_SELECTIONS_FROZEN__"
            "OFFICIAL_TEST_SEALED__"
            "NO_GPU_TRAINING_REMAINING"
        ),
}


audit_bytes = json.dumps(
    audit_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


audit_sha = hashlib.sha256(
    audit_bytes
).hexdigest()


print(
    "Closing-audit SHA-256:",
    audit_sha
)


# ==================================================================================================
# 17. WRITE EXTERNAL FINAL AUDIT PACKAGE
# ==================================================================================================

print()
print("-" * 180)
print("[14] WRITE FINAL AUDIT PACKAGE TO GOOGLE DRIVE")
print("-" * 180)


FINAL_AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    FINAL_AUDIT_JSON,
    FINAL_RUN_LEDGER_CSV,
    FINAL_AUDIT_SHA,
]:

    if path.exists():

        raise RuntimeError(
            "STOP: final audit artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    FINAL_RUN_LEDGER_CSV,
    ledger_bytes,
)


atomic_write_bytes(
    FINAL_AUDIT_JSON,
    audit_bytes,
)


sidecar_text = (
    f"{audit_sha}  {FINAL_AUDIT_JSON.name}\n"
    f"{ledger_sha}  {FINAL_RUN_LEDGER_CSV.name}\n"
    f"{final_zip_sha}  {FINAL_PERSISTENCE_ZIP.name}\n"
    f"{final_manifest_sha}  {FINAL_PERSISTENCE_MANIFEST.name}\n"
)


atomic_write_bytes(
    FINAL_AUDIT_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


if (
    sha256_file(
        FINAL_AUDIT_JSON
    )
    !=
    audit_sha
):

    raise RuntimeError(
        "STOP: written final closing-audit SHA mismatch."
    )


if (
    sha256_file(
        FINAL_RUN_LEDGER_CSV
    )
    !=
    ledger_sha
):

    raise RuntimeError(
        "STOP: written final run-ledger SHA mismatch."
    )


if not FINAL_AUDIT_SHA.is_file():

    raise RuntimeError(
        "STOP: final audit SHA sidecar missing."
    )


# ==================================================================================================
# 18. PROVE SCIENTIFIC PROJECT REMAINED UNCHANGED
# ==================================================================================================

project_files_after_audit = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if (
    len(
        project_files_after_audit
    )
    != 507
):

    raise RuntimeError(
        "STOP: closing audit changed scientific project footprint."
    )


# ==================================================================================================
# 19. TERMINAL GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 15B TERMINAL GATE")
print("=" * 180)

print(
    "Cell 15 classification                       : AUDIT-CODE SCHEMA ASSUMPTION DEFECT"
)

print(
    "Cell 15A classification                      : AUDIT-CODE SCHEMA ASSUMPTION DEFECT"
)

print(
    "Scientific-state failure                     : False"
)

print(
    "Scientific artifacts modified                : False"
)

print()
print(
    "Externally saved final persistence            : AUTHENTICATED PASS"
)

print(
    "Final scientific project                      : 507 files"
)

print(
    "Full project SHA verification                : 507/507 PASS"
)

print(
    "Missing expected files                       : 0"
)

print(
    "Unexpected files                             : 0"
)

print()
print(
    "CE                                           : 6/6 PASS"
)

print(
    "Focal                                        : 8/8 PASS"
)

print(
    "LDAM-DRW                                     : 8/8 PASS"
)

print(
    "Balanced Softmax                             : 6/6 PASS"
)

print(
    "LA-loss                                      : 8/8 PASS"
)

print(
    "Experiment-B                                 : 36/36 COMPLETE"
)

print()
print(
    "Completion identities from authenticated paths: 36/36 PASS"
)

print(
    "Present dataset fields/path consistency       : PASS"
)

print(
    "Present seed fields/path consistency          : PASS"
)

print(
    "Exact LA-loss matrix                         : 8/8 PASS"
)

print(
    "Pingan selected tau                          : 1.0 PASS"
)

print(
    "Qingyun selected tau                         : 0.5 PASS"
)

print()
print(
    "Official-test safety violations              : 0"
)

print(
    "Official test                                : SEALED"
)

print(
    "Official-test inference                      : NOT PERFORMED"
)

print(
    "Training in Cell 15B                         : False"
)

print(
    "GPU training remaining                       : False"
)

print(
    "Scientific project mutated by Cell 15B       : False"
)

print()
print(
    "FINAL CLOSING AUDIT:"
)

print(
    FINAL_AUDIT_JSON
)

print()
print(
    "FINAL CLOSING AUDIT SHA-256:"
)

print(
    audit_sha
)

print()
print(
    "FINAL 36-RUN LEDGER:"
)

print(
    FINAL_RUN_LEDGER_CSV
)

print()
print(
    "FINAL 36-RUN LEDGER SHA-256:"
)

print(
    ledger_sha
)

print()
print(
    "FINAL AUDIT SHA SIDECAR:"
)

print(
    FINAL_AUDIT_SHA
)

print()
print(
    "FINAL CORRECTION CELL 15B PASS: True"
)

print()
print(
    "STATUS: "
    "CELL15_AND_15A_AUDIT_SCHEMA_BUGS_CORRECTED__"
    "EXPERIMENT_B_36_OF_36_COMPLETE__"
    "FULL_507_FILE_SNAPSHOT_VERIFIED__"
    "36_COMPLETION_RECORDS_PATH_AUTHENTICATED__"
    "HISTORICAL_MANIFEST_SCHEMAS_TOLERATED_WITH_CONSISTENCY_CHECKS__"
    "EXACT_LA_8_OF_8_MATRIX__"
    "LA_SELECTIONS_FROZEN__"
    "OFFICIAL_TEST_SEALED__"
    "NO_GPU_TRAINING_REMAINING__"
    "FINAL_CLOSING_AUDIT_WRITTEN_TO_GOOGLE_DRIVE"
)

print("=" * 180)

print()
print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CELL 16
#
# FINAL 36-RUN VALIDATION RESULTS CONSOLIDATION
#
# INPUT STATE:
#   Experiment-B 36/36 COMPLETE
#   Correction Cell 15B PASS
#   Scientific project = exact 507-file authenticated snapshot
#
# OUTPUT:
#   1. all 36 run validation records
#   2. selected 30-run comparison set
#      = 5 methods × 2 datasets × 3 seeds
#   3. validation-only tuning table for Focal / LDAM-DRW / LA-loss
#   4. authenticated consolidation JSON + SHA sidecar
#
# NO TRAINING
# NO CHECKPOINT INFERENCE
# NO OFFICIAL-TEST ACCESS
# NO RETUNING
# NO SCIENTIFIC PROJECT MUTATION
# ==================================================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import csv
import hashlib
import json
import math
import os
import re


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 16")
print("FINAL 36-RUN VALIDATION RESULTS CONSOLIDATION")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

FINAL_AUDIT_DIR = (
    DRIVE_ROOT
    / "final_audits"
    / "notebook06a_experiment_b"
)

FINAL_AUDIT_JSON = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1.json"
)

FINAL_RUN_LEDGER = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_RUN_LEDGER_v1.csv"
)

RESULTS_DIR = (
    DRIVE_ROOT
    / "results"
    / "notebook06a_experiment_b"
)

ALL36_CSV = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_ALL36_VALIDATION_RUNS_v1.csv"
)

SELECTED30_CSV = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_SELECTED30_VALIDATION_RUNS_v1.csv"
)

BASELINE_SELECTION_CSV = (
    RESULTS_DIR
    / "baseline_selection.csv"
)

CONSOLIDATION_JSON = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_RESULTS_CONSOLIDATION_v1.json"
)

CONSOLIDATION_SHA = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_RESULTS_CONSOLIDATION_v1__SHA256.txt"
)


# ==================================================================================================
# 2. AUTHENTICATED PARENT IDENTITIES
# ==================================================================================================

EXPECTED_FINAL_AUDIT_SHA256 = (
    "2a24fdbaba212890443413c39e2873b2"
    "9862e4810dda91a2ed5879a7b6d59768"
)

EXPECTED_FINAL_LEDGER_SHA256 = (
    "6335262b572a389a33c26b877c7c7f78"
    "5934bee1ab53844f8fa6b5f61848b07d"
)

EXPECTED_PINGAN_LA_SELECTION_SHA256 = (
    "3d6d0c064081e9a18cc14217d0f9835b"
    "23f7bdb99e1002362ffec31a933f4bf8"
)

EXPECTED_QINGYUN_LA_SELECTION_SHA256 = (
    "367d6ecd113d786bf89510985e0f0ff7"
    "e6589a68db9c912d4c36fb29d02fae89"
)


DATASETS = {
    "pingan": "Pingan",
    "qingyun": "Qingyun",
}

ALLOWED_SEEDS = {
    42,
    123,
    3407,
}

METHOD_LABELS = {
    "ce": "CE",
    "focal": "Focal",
    "ldam_drw": "LDAM-DRW",
    "balanced_softmax": "Balanced Softmax",
    "la_loss": "LA-loss",
}

METHOD_EXPECTED_COUNTS = {
    "ce": 6,
    "focal": 8,
    "ldam_drw": 8,
    "balanced_softmax": 6,
    "la_loss": 8,
}


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )

    if tmp.exists():
        tmp.unlink()

    with tmp.open("wb") as f:

        f.write(
            payload
        )

    os.replace(
        tmp,
        path,
    )


def is_number(
    value,
):

    return (
        isinstance(
            value,
            (int, float),
        )
        and
        not isinstance(
            value,
            bool,
        )
        and
        math.isfinite(
            float(
                value
            )
        )
    )


def first_numeric(
    mapping,
    aliases,
):

    if not isinstance(
        mapping,
        dict,
    ):

        return None

    for key in aliases:

        if (
            key in mapping
            and
            is_number(
                mapping[
                    key
                ]
            )
        ):

            return float(
                mapping[
                    key
                ]
            )

    return None


def first_integer(
    mapping,
    aliases,
):

    if not isinstance(
        mapping,
        dict,
    ):

        return None

    for key in aliases:

        if key in mapping:

            try:

                return int(
                    mapping[
                        key
                    ]
                )

            except Exception:

                pass

    return None


METRIC_ALIASES = {
    "AA": [
        "AA",
        "aa",
        "val_AA",
        "val_aa",
        "validation_AA",
        "validation_aa",
    ],

    "Macro_F1": [
        "Macro_F1",
        "macro_f1",
        "MacroF1",
        "macroF1",
        "val_Macro_F1",
        "val_macro_f1",
        "validation_Macro_F1",
        "validation_macro_f1",
    ],

    "OA": [
        "OA",
        "oa",
        "val_OA",
        "val_oa",
        "validation_OA",
        "validation_oa",
    ],

    "Kappa": [
        "Kappa",
        "kappa",
        "val_Kappa",
        "val_kappa",
        "validation_Kappa",
        "validation_kappa",
    ],

    "loss": [
        "loss",
        "val_loss",
        "validation_loss",
    ],
}


def normalize_metric_dict(
    mapping,
):

    if not isinstance(
        mapping,
        dict,
    ):

        return None


    result = {
        key:
            first_numeric(
                mapping,
                aliases,
            )

        for key, aliases in METRIC_ALIASES.items()
    }


    if not all(
        result[
            key
        ]
        is not None

        for key in (
            "AA",
            "Macro_F1",
            "OA",
        )
    ):

        return None


    epoch = first_integer(
        mapping,
        [
            "epoch",
            "best_epoch",
            "best_AA_epoch",
            "best_aa_epoch",
            "selected_epoch",
            "checkpoint_epoch",
        ],
    )


    result[
        "epoch"
    ] = epoch


    return result


def walk_dicts(
    obj,
    path=(),
):

    if isinstance(
        obj,
        dict,
    ):

        yield path, obj

        for key, value in obj.items():

            yield from walk_dicts(
                value,
                path
                +
                (
                    str(
                        key
                    ),
                ),
            )


    elif isinstance(
        obj,
        list,
    ):

        for idx, value in enumerate(
            obj
        ):

            yield from walk_dicts(
                value,
                path
                +
                (
                    str(
                        idx
                    ),
                ),
            )


def score_best_aa_candidate(
    path,
):

    text = "/".join(
        str(
            item
        ).lower()

        for item in path
    )


    score = 0


    if (
        "best_validation_aa"
        in text
        or
        "best_aa"
        in text
    ):

        score += 100


    if "authoritative" in text:

        score += 50


    if "checkpoint_selection" in text:

        score += 25


    if "independent_validation_confirmation" in text:

        score += 20


    if (
        "best_validation_oa"
        in text
        or
        "best_oa"
        in text
    ):

        score -= 80


    return score


def find_best_aa_metrics_in_object(
    obj,
):

    candidates = []


    for path, mapping in walk_dicts(
        obj
    ):

        normalized = normalize_metric_dict(
            mapping
        )


        if normalized is None:

            continue


        candidates.append(
            (
                score_best_aa_candidate(
                    path
                ),
                path,
                normalized,
            )
        )


    if not candidates:

        return None


    candidates.sort(
        key=lambda item: (
            item[
                0
            ],
            -len(
                item[
                    1
                ]
            ),
        ),
        reverse=True,
    )


    best = candidates[
        0
    ]


    # Avoid silently accepting an obviously best-OA-only record.
    if best[
        0
    ] < 0:

        return None


    return {
        "metrics":
            best[
                2
            ],

        "object_path":
            "/".join(
                best[
                    1
                ]
            ),
    }


def better_best_aa_record(
    candidate,
    incumbent,
    tol=1e-12,
):

    if incumbent is None:

        return True


    if (
        candidate["AA"]
        >
        incumbent["AA"]
        +
        tol
    ):

        return True


    if (
        incumbent["AA"]
        >
        candidate["AA"]
        +
        tol
    ):

        return False


    if (
        candidate["Macro_F1"]
        >
        incumbent["Macro_F1"]
        +
        tol
    ):

        return True


    if (
        incumbent["Macro_F1"]
        >
        candidate["Macro_F1"]
        +
        tol
    ):

        return False


    if (
        candidate["OA"]
        >
        incumbent["OA"]
        +
        tol
    ):

        return True


    if (
        incumbent["OA"]
        >
        candidate["OA"]
        +
        tol
    ):

        return False


    candidate_epoch = (
        candidate.get(
            "epoch"
        )
    )

    incumbent_epoch = (
        incumbent.get(
            "epoch"
        )
    )


    if (
        candidate_epoch
        is None
        or
        incumbent_epoch
        is None
    ):

        return False


    return (
        int(
            candidate_epoch
        )
        <
        int(
            incumbent_epoch
        )
    )


def recover_best_aa_from_history_json(
    run_dir,
):

    candidates = []


    for path in sorted(
        run_dir.glob(
            "*history*.json"
        )
    ):

        try:

            with path.open(
                "r",
                encoding="utf-8",
            ) as f:

                payload = json.load(
                    f
                )

        except Exception:

            continue


        for _, mapping in walk_dicts(
            payload
        ):

            normalized = normalize_metric_dict(
                mapping
            )


            if normalized is None:

                continue


            if normalized[
                "epoch"
            ] is None:

                continue


            candidates.append(
                (
                    path,
                    normalized,
                )
            )


    best = None
    best_path = None


    for path, candidate in candidates:

        if better_best_aa_record(
            candidate,
            best,
        ):

            best = candidate
            best_path = path


    if best is None:

        return None


    return {
        "metrics":
            best,

        "source_path":
            str(
                best_path.relative_to(
                    ROOT
                )
            ),

        "source_type":
            "epoch_history_json_recomputed_best_AA",
    }


def recover_best_aa_from_history_csv(
    run_dir,
):

    candidates = []


    for path in sorted(
        run_dir.glob(
            "*history*.csv"
        )
    ):

        try:

            with path.open(
                "r",
                encoding="utf-8",
                newline="",
            ) as f:

                rows = list(
                    csv.DictReader(
                        f
                    )
                )

        except Exception:

            continue


        for row in rows:

            numeric_row = {}


            for key, value in row.items():

                if value is None:

                    continue


                text = str(
                    value
                ).strip()


                if text == "":

                    continue


                try:

                    numeric_row[
                        key
                    ] = float(
                        text
                    )

                except Exception:

                    numeric_row[
                        key
                    ] = value


            normalized = normalize_metric_dict(
                numeric_row
            )


            if normalized is None:

                continue


            if normalized[
                "epoch"
            ] is None:

                epoch = first_integer(
                    numeric_row,
                    [
                        "epoch",
                        "Epoch",
                    ],
                )

                normalized[
                    "epoch"
                ] = epoch


            if normalized[
                "epoch"
            ] is None:

                continue


            candidates.append(
                (
                    path,
                    normalized,
                )
            )


    best = None
    best_path = None


    for path, candidate in candidates:

        if better_best_aa_record(
            candidate,
            best,
        ):

            best = candidate
            best_path = path


    if best is None:

        return None


    return {
        "metrics":
            best,

        "source_path":
            str(
                best_path.relative_to(
                    ROOT
                )
            ),

        "source_type":
            "epoch_history_csv_recomputed_best_AA",
    }


def extract_best_aa_result(
    completion_path,
):

    with completion_path.open(
        "r",
        encoding="utf-8",
    ) as f:

        manifest = json.load(
            f
        )


    # ----------------------------------------------------------------------------------------------
    # STANDARD SCHEMA
    # ----------------------------------------------------------------------------------------------

    checkpoint_policy = manifest.get(
        "checkpoint_policy"
    )


    if isinstance(
        checkpoint_policy,
        dict,
    ):

        metrics = normalize_metric_dict(
            checkpoint_policy.get(
                "best_AA_metrics"
            )
        )


        if metrics is not None:

            metrics[
                "epoch"
            ] = int(
                checkpoint_policy[
                    "best_AA_epoch"
                ]
            )


            return {
                "metrics":
                    metrics,

                "metric_source":
                    "completion_manifest.checkpoint_policy.best_AA_metrics",

                "metric_source_path":
                    str(
                        completion_path.relative_to(
                            ROOT
                        )
                    ),
            }


    # ----------------------------------------------------------------------------------------------
    # RECOVERY / HISTORICAL COMPLETION SCHEMA
    # ----------------------------------------------------------------------------------------------

    fallback = find_best_aa_metrics_in_object(
        manifest
    )


    if fallback is not None:

        metrics = fallback[
            "metrics"
        ]


        return {
            "metrics":
                metrics,

            "metric_source":
                (
                    "completion_manifest_recursive_best_AA:"
                    +
                    fallback[
                        "object_path"
                    ]
                ),

            "metric_source_path":
                str(
                    completion_path.relative_to(
                        ROOT
                    )
                ),
        }


    # ----------------------------------------------------------------------------------------------
    # INDEPENDENT AUDIT JSON FALLBACK
    # ----------------------------------------------------------------------------------------------

    run_dir = completion_path.parent


    for audit_path in sorted(
        run_dir.glob(
            "*audit*.json"
        )
    ):

        try:

            with audit_path.open(
                "r",
                encoding="utf-8",
            ) as f:

                audit_payload = json.load(
                    f
                )

        except Exception:

            continue


        audit_fallback = find_best_aa_metrics_in_object(
            audit_payload
        )


        if audit_fallback is not None:

            return {
                "metrics":
                    audit_fallback[
                        "metrics"
                    ],

                "metric_source":
                    (
                        "independent_audit_recursive_best_AA:"
                        +
                        audit_fallback[
                            "object_path"
                        ]
                    ),

                "metric_source_path":
                    str(
                        audit_path.relative_to(
                            ROOT
                        )
                    ),
            }


    # ----------------------------------------------------------------------------------------------
    # HISTORY JSON FALLBACK
    # ----------------------------------------------------------------------------------------------

    history_json_result = (
        recover_best_aa_from_history_json(
            run_dir
        )
    )


    if history_json_result is not None:

        return {
            "metrics":
                history_json_result[
                    "metrics"
                ],

            "metric_source":
                history_json_result[
                    "source_type"
                ],

            "metric_source_path":
                history_json_result[
                    "source_path"
                ],
        }


    # ----------------------------------------------------------------------------------------------
    # HISTORY CSV FALLBACK
    # ----------------------------------------------------------------------------------------------

    history_csv_result = (
        recover_best_aa_from_history_csv(
            run_dir
        )
    )


    if history_csv_result is not None:

        return {
            "metrics":
                history_csv_result[
                    "metrics"
                ],

            "metric_source":
                history_csv_result[
                    "source_type"
                ],

            "metric_source_path":
                history_csv_result[
                    "source_path"
                ],
        }


    raise RuntimeError(
        "STOP: unable to recover authoritative best-AA metrics for:\n"
        f"{completion_path}"
    )


def extract_run_identity_from_path(
    completion_path,
):

    rel = str(
        completion_path.relative_to(
            ROOT
        )
    )


    parts = PurePosixPath(
        rel
    ).parts


    if len(
        parts
    ) < 6:

        raise RuntimeError(
            f"STOP: malformed completion path: {rel}"
        )


    if (
        parts[
            0
        ] != "runs"
        or
        parts[
            1
        ] != "experiment_b_spatial_validation_v1"
    ):

        raise RuntimeError(
            f"STOP: completion outside Experiment-B tree: {rel}"
        )


    method_dir = parts[
        2
    ]

    dataset_slug = parts[
        3
    ].lower()


    if method_dir not in METHOD_LABELS:

        raise RuntimeError(
            f"STOP: unsupported method path: {rel}"
        )


    if dataset_slug not in DATASETS:

        raise RuntimeError(
            f"STOP: unsupported dataset path: {rel}"
        )


    seed_indices = []


    for idx, part in enumerate(
        parts
    ):

        match = re.fullmatch(
            r"seed(\d+)",
            part,
        )


        if match:

            seed_indices.append(
                (
                    idx,
                    int(
                        match.group(
                            1
                        )
                    ),
                )
            )


    if len(
        seed_indices
    ) != 1:

        raise RuntimeError(
            f"STOP: expected exactly one seed directory: {rel}"
        )


    seed_index, seed = seed_indices[
        0
    ]


    if seed not in ALLOWED_SEEDS:

        raise RuntimeError(
            f"STOP: unexpected seed {seed}: {rel}"
        )


    parameter_parts = parts[
        4:
        seed_index
    ]


    if method_dir in {
        "ce",
        "balanced_softmax",
    }:

        if parameter_parts:

            raise RuntimeError(
                "STOP: fixed-objective method unexpectedly has "
                f"parameter directory: {rel}"
            )


        parameter_token = ""


    else:

        if len(
            parameter_parts
        ) != 1:

            raise RuntimeError(
                "STOP: expected exactly one tuning parameter directory:\n"
                f"{rel}\n"
                f"parameter_parts={parameter_parts}"
            )


        parameter_token = parameter_parts[
            0
        ]


    return {
        "relative_path":
            rel,

        "method_dir":
            method_dir,

        "method":
            METHOD_LABELS[
                method_dir
            ],

        "dataset_slug":
            dataset_slug,

        "dataset":
            DATASETS[
                dataset_slug
            ],

        "seed":
            seed,

        "parameter_token":
            parameter_token,
    }


def parameter_token_to_value(
    method_dir,
    token,
):

    if method_dir in {
        "ce",
        "balanced_softmax",
    }:

        return None


    low = str(
        token
    ).lower()


    prefix_map = {
        "focal":
            "gamma",

        "ldam_drw":
            "c",

        "la_loss":
            "tau",
    }


    prefix = prefix_map[
        method_dir
    ]


    if not low.startswith(
        prefix
    ):

        raise RuntimeError(
            "STOP: parameter token does not match method:\n"
            f"method={method_dir}, token={token}"
        )


    raw = low[
        len(
            prefix
        ):
    ]


    raw = raw.replace(
        "p",
        ".",
    )


    try:

        return float(
            raw
        )

    except Exception:

        raise RuntimeError(
            f"STOP: could not parse parameter token: {token}"
        )


def selected_config_label(
    method_dir,
    parameter_value,
):

    if method_dir == "ce":

        return "CrossEntropy"


    if method_dir == "balanced_softmax":

        return "training_counts"


    if method_dir == "focal":

        return (
            f"gamma={parameter_value:g}"
        )


    if method_dir == "ldam_drw":

        return (
            f"C={parameter_value:g}"
        )


    if method_dir == "la_loss":

        return (
            f"tau={parameter_value:g}"
        )


    raise RuntimeError(
        f"STOP: unknown method: {method_dir}"
    )


# ==================================================================================================
# 4. PARENT TERMINAL AUDIT GATE
# ==================================================================================================

print()
print("-" * 180)
print("[1] AUTHENTICATED CELL 15B PARENT")
print("-" * 180)


if not FINAL_AUDIT_JSON.is_file():

    raise FileNotFoundError(
        "STOP: final Cell15B audit JSON missing."
    )


if not FINAL_RUN_LEDGER.is_file():

    raise FileNotFoundError(
        "STOP: final Cell15B run ledger missing."
    )


final_audit_sha = sha256_file(
    FINAL_AUDIT_JSON
)

final_ledger_sha = sha256_file(
    FINAL_RUN_LEDGER
)


if (
    final_audit_sha
    !=
    EXPECTED_FINAL_AUDIT_SHA256
):

    raise RuntimeError(
        "STOP: Cell15B final audit SHA changed."
    )


if (
    final_ledger_sha
    !=
    EXPECTED_FINAL_LEDGER_SHA256
):

    raise RuntimeError(
        "STOP: Cell15B final ledger SHA changed."
    )


with FINAL_AUDIT_JSON.open(
    "r",
    encoding="utf-8",
) as f:

    parent_audit = json.load(
        f
    )


if (
    parent_audit.get(
        "closing_status"
    )
    != "PASS"
):

    raise RuntimeError(
        "STOP: Cell15B closing status is not PASS."
    )


if (
    parent_audit[
        "final_state"
    ][
        "experiment_b_complete"
    ]
    is not True
):

    raise RuntimeError(
        "STOP: parent audit does not mark Experiment B complete."
    )


if (
    parent_audit[
        "official_test"
    ][
        "status"
    ]
    != "SEALED"
):

    raise RuntimeError(
        "STOP: parent audit official test is not SEALED."
    )


print(
    "Final closing audit : AUTHENTICATED PASS"
)

print(
    "Final 36-run ledger : AUTHENTICATED PASS"
)

print(
    "Experiment-B        : 36/36 COMPLETE"
)

print(
    "Official test       : SEALED"
)


# ==================================================================================================
# 5. PROJECT FOOTPRINT GATE
# ==================================================================================================

print()
print("-" * 180)
print("[2] SCIENTIFIC PROJECT FOOTPRINT")
print("-" * 180)


project_files = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    project_files
) != 507:

    raise RuntimeError(
        "STOP: scientific project is no longer exactly 507 files."
    )


print(
    "Scientific project files : 507 PASS"
)


# ==================================================================================================
# 6. DISCOVER ALL 36 COMPLETION MANIFESTS
# ==================================================================================================

print()
print("-" * 180)
print("[3] DISCOVER ALL 36 COMPLETION RECORDS")
print("-" * 180)


RUN_ROOT = (
    ROOT
    / "runs"
    / "experiment_b_spatial_validation_v1"
)


completion_paths = []


for method_dir, expected_count in METHOD_EXPECTED_COUNTS.items():

    paths = sorted(
        (
            RUN_ROOT
            /
            method_dir
        ).rglob(
            "*__completion_manifest.json"
        )
    )


    if len(
        paths
    ) != expected_count:

        raise RuntimeError(
            "STOP: completion count mismatch:\n"
            f"{method_dir}: observed={len(paths)}, "
            f"expected={expected_count}"
        )


    completion_paths.extend(
        paths
    )


if len(
    completion_paths
) != 36:

    raise RuntimeError(
        "STOP: Experiment-B completion count is not 36."
    )


print(
    "Completion manifests : 36/36 PASS"
)


# ==================================================================================================
# 7. EXTRACT ALL 36 AUTHORITATIVE BEST-AA RESULTS
# ==================================================================================================

print()
print("-" * 180)
print("[4] EXTRACT AUTHORITATIVE BEST-VALIDATION-AA RESULTS")
print("-" * 180)


all_records = []


for completion_path in sorted(
    completion_paths
):

    identity = extract_run_identity_from_path(
        completion_path
    )


    extracted = extract_best_aa_result(
        completion_path
    )


    metrics = extracted[
        "metrics"
    ]


    # Main reporting metrics must exist.
    for required_metric in (
        "AA",
        "Macro_F1",
        "OA",
        "Kappa",
    ):

        if metrics.get(
            required_metric
        ) is None:

            raise RuntimeError(
                "STOP: required validation metric missing:\n"
                f"{identity['relative_path']}\n"
                f"metric={required_metric}"
            )


    parameter_value = parameter_token_to_value(
        identity[
            "method_dir"
        ],
        identity[
            "parameter_token"
        ],
    )


    record = {
        **identity,

        "parameter_value":
            parameter_value,

        "best_AA_epoch":
            metrics.get(
                "epoch"
            ),

        "val_AA":
            float(
                metrics[
                    "AA"
                ]
            ),

        "val_Macro_F1":
            float(
                metrics[
                    "Macro_F1"
                ]
            ),

        "val_OA":
            float(
                metrics[
                    "OA"
                ]
            ),

        "val_Kappa":
            float(
                metrics[
                    "Kappa"
                ]
            ),

        "val_loss":
            (
                ""
                if metrics.get(
                    "loss"
                ) is None
                else float(
                    metrics[
                        "loss"
                    ]
                )
            ),

        "completion_sha256":
            sha256_file(
                completion_path
            ),

        "metric_source":
            extracted[
                "metric_source"
            ],

        "metric_source_path":
            extracted[
                "metric_source_path"
            ],
    }


    all_records.append(
        record
    )


if len(
    all_records
) != 36:

    raise RuntimeError(
        "STOP: expected 36 extracted result records."
    )


print(
    "Best-AA result extraction : 36/36 PASS"
)


source_counts = {}


for record in all_records:

    source = record[
        "metric_source"
    ]


    source_counts[
        source
    ] = (
        source_counts.get(
            source,
            0,
        )
        +
        1
    )


print()
print(
    "Metric-source schema distribution:"
)


for source, count in sorted(
    source_counts.items()
):

    print(
        f"  {count:2d} × {source}"
    )


# ==================================================================================================
# 8. INFER FROZEN SELECTED CONFIGURATIONS FROM AUTHENTICATED RUN MATRIX
# ==================================================================================================

print()
print("-" * 180)
print("[5] FROZEN CONFIGURATION SELECTION RECONSTRUCTION")
print("-" * 180)


selection_map = {}


# Fixed methods
for dataset_slug in DATASETS:

    selection_map[
        (
            dataset_slug,
            "ce",
        )
    ] = {
        "parameter_token":
            "",

        "parameter_value":
            None,

        "selection_basis":
            "fixed_objective",
    }


    selection_map[
        (
            dataset_slug,
            "balanced_softmax",
        )
    ] = {
        "parameter_token":
            "",

        "parameter_value":
            None,

        "selection_basis":
            "fixed_training_counts_objective",
    }


# Tuned methods:
# selected parameter is the one replicated at seeds 123 and 3407 after seed42 tuning.
for method_dir in (
    "focal",
    "ldam_drw",
    "la_loss",
):

    for dataset_slug in DATASETS:

        subset = [
            record

            for record in all_records

            if (
                record[
                    "method_dir"
                ] == method_dir
                and
                record[
                    "dataset_slug"
                ] == dataset_slug
            )
        ]


        seed42_tokens = {
            record[
                "parameter_token"
            ]

            for record in subset

            if record[
                "seed"
            ] == 42
        }


        if len(
            seed42_tokens
        ) != 2:

            raise RuntimeError(
                "STOP: tuned method must have exactly two seed42 candidates:\n"
                f"{dataset_slug} / {method_dir}\n"
                f"observed={sorted(seed42_tokens)}"
            )


        seed123_tokens = {
            record[
                "parameter_token"
            ]

            for record in subset

            if record[
                "seed"
            ] == 123
        }


        seed3407_tokens = {
            record[
                "parameter_token"
            ]

            for record in subset

            if record[
                "seed"
            ] == 3407
        }


        if (
            len(
                seed123_tokens
            ) != 1
            or
            len(
                seed3407_tokens
            ) != 1
        ):

            raise RuntimeError(
                "STOP: replication seed configuration is not unique:\n"
                f"{dataset_slug} / {method_dir}"
            )


        selected_token_123 = next(
            iter(
                seed123_tokens
            )
        )

        selected_token_3407 = next(
            iter(
                seed3407_tokens
            )
        )


        if (
            selected_token_123
            !=
            selected_token_3407
        ):

            raise RuntimeError(
                "STOP: seeds123/3407 were not trained with the same "
                "frozen selected hyperparameter:\n"
                f"{dataset_slug} / {method_dir}"
            )


        if (
            selected_token_123
            not in
            seed42_tokens
        ):

            raise RuntimeError(
                "STOP: replicated selected parameter was not one of "
                "the seed42 candidates."
            )


        selected_value = parameter_token_to_value(
            method_dir,
            selected_token_123,
        )


        selection_map[
            (
                dataset_slug,
                method_dir,
            )
        ] = {
            "parameter_token":
                selected_token_123,

            "parameter_value":
                selected_value,

            "candidate_tokens":
                sorted(
                    seed42_tokens
                ),

            "selection_basis":
                (
                    "seed42_validation_tuning_then_authenticated_"
                    "replication_at_seeds123_3407"
                ),
        }


# ==================================================================================================
# 9. AUTHENTICATE LA SELECTION MANIFESTS AGAINST RECONSTRUCTED RUN MATRIX
# ==================================================================================================

PINGAN_LA_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "pingan"
    / "la_loss__pingan__selected_tau_v1.json"
)

QINGYUN_LA_SELECTION = (
    RUN_ROOT
    / "la_loss"
    / "qingyun"
    / "la_loss__qingyun__selected_tau_v1.json"
)


for (
    dataset_slug,
    selection_path,
    expected_sha,
    expected_tau,
) in [
    (
        "pingan",
        PINGAN_LA_SELECTION,
        EXPECTED_PINGAN_LA_SELECTION_SHA256,
        1.0,
    ),
    (
        "qingyun",
        QINGYUN_LA_SELECTION,
        EXPECTED_QINGYUN_LA_SELECTION_SHA256,
        0.5,
    ),
]:

    if not selection_path.is_file():

        raise FileNotFoundError(
            f"STOP: LA selection manifest missing: {dataset_slug}"
        )


    if sha256_file(
        selection_path
    ) != expected_sha:

        raise RuntimeError(
            f"STOP: LA selection SHA mismatch: {dataset_slug}"
        )


    with selection_path.open(
        "r",
        encoding="utf-8",
    ) as f:

        payload = json.load(
            f
        )


    if float(
        payload[
            "selected_tau"
        ]
    ) != expected_tau:

        raise RuntimeError(
            f"STOP: LA selected tau changed: {dataset_slug}"
        )


    reconstructed_tau = (
        selection_map[
            (
                dataset_slug,
                "la_loss",
            )
        ][
            "parameter_value"
        ]
    )


    if float(
        reconstructed_tau
    ) != float(
        expected_tau
    ):

        raise RuntimeError(
            f"STOP: LA selection/run-matrix disagreement: {dataset_slug}"
        )


print(
    "Pingan LA selected tau  : 1.0 PASS"
)

print(
    "Qingyun LA selected tau : 0.5 PASS"
)


print()
print(
    "Reconstructed selected configurations:"
)


for dataset_slug in DATASETS:

    print()
    print(
        DATASETS[
            dataset_slug
        ]
    )


    for method_dir in (
        "ce",
        "focal",
        "ldam_drw",
        "balanced_softmax",
        "la_loss",
    ):

        selection = selection_map[
            (
                dataset_slug,
                method_dir,
            )
        ]


        print(
            "  "
            f"{METHOD_LABELS[method_dir]:18s}"
            " -> "
            +
            selected_config_label(
                method_dir,
                selection[
                    "parameter_value"
                ],
            )
        )


# ==================================================================================================
# 10. MARK FINAL SELECTED 30-RUN COMPARISON SET
# ==================================================================================================

selected_records = []


for record in all_records:

    selection = selection_map[
        (
            record[
                "dataset_slug"
            ],
            record[
                "method_dir"
            ],
        )
    ]


    if record[
        "method_dir"
    ] in {
        "ce",
        "balanced_softmax",
    }:

        selected = True


    else:

        selected = (
            record[
                "parameter_token"
            ]
            ==
            selection[
                "parameter_token"
            ]
        )


    record[
        "selected_final_config"
    ] = bool(
        selected
    )


    record[
        "selected_config"
    ] = selected_config_label(
        record[
            "method_dir"
        ],
        selection[
            "parameter_value"
        ],
    )


    if selected:

        selected_records.append(
            record
        )


if len(
    selected_records
) != 30:

    raise RuntimeError(
        "STOP: final selected comparison set is not 30 runs."
    )


# Each dataset/method must have exactly 3 selected seeds.
for dataset_slug in DATASETS:

    for method_dir in METHOD_LABELS:

        subset = [
            record

            for record in selected_records

            if (
                record[
                    "dataset_slug"
                ] == dataset_slug
                and
                record[
                    "method_dir"
                ] == method_dir
            )
        ]


        seeds = {
            record[
                "seed"
            ]

            for record in subset
        }


        if seeds != ALLOWED_SEEDS:

            raise RuntimeError(
                "STOP: selected comparison set does not contain exact "
                "three-seed matrix:\n"
                f"{dataset_slug} / {method_dir} / seeds={sorted(seeds)}"
            )


print()
print(
    "Selected final comparison runs : 30/30 PASS"
)

print(
    "Method × dataset groups        : 10"
)

print(
    "Seeds per selected group       : {42,123,3407} PASS"
)


# ==================================================================================================
# 11. BUILD VALIDATION-ONLY TUNING TABLE
# ==================================================================================================

tuning_records = []


for record in all_records:

    if (
        record[
            "method_dir"
        ]
        not in {
            "focal",
            "ldam_drw",
            "la_loss",
        }
    ):

        continue


    if record[
        "seed"
    ] != 42:

        continue


    selection = selection_map[
        (
            record[
                "dataset_slug"
            ],
            record[
                "method_dir"
            ],
        )
    ]


    tuning_records.append(
        {
            "dataset":
                record[
                    "dataset"
                ],

            "method":
                record[
                    "method"
                ],

            "candidate_id":
                record[
                    "parameter_token"
                ],

            "candidate_value":
                record[
                    "parameter_value"
                ],

            "seed":
                42,

            "val_oa":
                record[
                    "val_OA"
                ],

            "val_aa":
                record[
                    "val_AA"
                ],

            "val_macro_f1":
                record[
                    "val_Macro_F1"
                ],

            "best_epoch":
                record[
                    "best_AA_epoch"
                ],

            "selected":
                (
                    record[
                        "parameter_token"
                    ]
                    ==
                    selection[
                        "parameter_token"
                    ]
                ),

            "completion_sha256":
                record[
                    "completion_sha256"
                ],
        }
    )


if len(
    tuning_records
) != 12:

    raise RuntimeError(
        "STOP: expected exactly 12 validation-only tuning records."
    )


if sum(
    bool(
        record[
            "selected"
        ]
    )
    for record in tuning_records
) != 6:

    raise RuntimeError(
        "STOP: expected exactly six selected tuning candidates."
    )


print(
    "Tuning candidate records : 12/12 PASS"
)

print(
    "Selected candidates      : 6/6 PASS"
)


# ==================================================================================================
# 12. WRITE CSV HELPERS
# ==================================================================================================

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    ALL36_CSV,
    SELECTED30_CSV,
    BASELINE_SELECTION_CSV,
    CONSOLIDATION_JSON,
    CONSOLIDATION_SHA,
]:

    if path.exists():

        raise RuntimeError(
            "STOP: Cell16 output already exists:\n"
            f"{path}"
        )


all36_fields = [
    "dataset",
    "dataset_slug",
    "method",
    "method_dir",
    "parameter_token",
    "parameter_value",
    "seed",
    "selected_final_config",
    "selected_config",
    "best_AA_epoch",
    "val_AA",
    "val_Macro_F1",
    "val_OA",
    "val_Kappa",
    "val_loss",
    "metric_source",
    "metric_source_path",
    "relative_path",
    "completion_sha256",
]


def csv_bytes_from_records(
    records,
    fields,
):

    import io

    buffer = io.StringIO()


    writer = csv.DictWriter(
        buffer,
        fieldnames=fields,
        lineterminator="\n",
    )


    writer.writeheader()


    for record in records:

        writer.writerow(
            {
                field:
                    record.get(
                        field,
                        ""
                    )

                for field in fields
            }
        )


    return buffer.getvalue().encode(
        "utf-8"
    )


all36_sorted = sorted(
    all_records,
    key=lambda record: (
        record[
            "dataset_slug"
        ],
        record[
            "method_dir"
        ],
        record[
            "parameter_token"
        ],
        record[
            "seed"
        ],
    ),
)


selected30_sorted = sorted(
    selected_records,
    key=lambda record: (
        record[
            "dataset_slug"
        ],
        record[
            "method_dir"
        ],
        record[
            "seed"
        ],
    ),
)


tuning_sorted = sorted(
    tuning_records,
    key=lambda record: (
        record[
            "dataset"
        ],
        record[
            "method"
        ],
        float(
            record[
                "candidate_value"
            ]
        ),
    ),
)


all36_bytes = csv_bytes_from_records(
    all36_sorted,
    all36_fields,
)


selected30_bytes = csv_bytes_from_records(
    selected30_sorted,
    all36_fields,
)


baseline_selection_fields = [
    "dataset",
    "method",
    "candidate_id",
    "candidate_value",
    "seed",
    "val_oa",
    "val_aa",
    "val_macro_f1",
    "best_epoch",
    "selected",
    "completion_sha256",
]


baseline_selection_bytes = csv_bytes_from_records(
    tuning_sorted,
    baseline_selection_fields,
)


atomic_write_bytes(
    ALL36_CSV,
    all36_bytes,
)

atomic_write_bytes(
    SELECTED30_CSV,
    selected30_bytes,
)

atomic_write_bytes(
    BASELINE_SELECTION_CSV,
    baseline_selection_bytes,
)


# ==================================================================================================
# 13. BUILD CONSOLIDATION MANIFEST
# ==================================================================================================

all36_sha = sha256_file(
    ALL36_CSV
)

selected30_sha = sha256_file(
    SELECTED30_CSV
)

baseline_selection_sha = sha256_file(
    BASELINE_SELECTION_CSV
)


selection_serializable = {}


for (
    dataset_slug,
    method_dir,
), value in selection_map.items():

    selection_serializable[
        f"{dataset_slug}/{method_dir}"
    ] = value


consolidation = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_NOTEBOOK06A_EXPERIMENT_B_VALIDATION_RESULTS_CONSOLIDATION",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent": {
        "final_closing_audit_path":
            str(
                FINAL_AUDIT_JSON
            ),

        "final_closing_audit_sha256":
            final_audit_sha,

        "final_36_run_ledger_sha256":
            final_ledger_sha,
    },

    "scientific_state": {
        "training_complete":
            True,

        "experiment_b_runs":
            36,

        "selected_comparison_runs":
            30,

        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "training_performed_in_cell16":
            False,

        "model_inference_performed_in_cell16":
            False,

        "retuning_performed_in_cell16":
            False,

        "scientific_project_mutated":
            False,
    },

    "selection_map":
        selection_serializable,

    "result_files": {
        "all36":
            {
                "path":
                    str(
                        ALL36_CSV
                    ),

                "rows":
                    36,

                "sha256":
                    all36_sha,
            },

        "selected30":
            {
                "path":
                    str(
                        SELECTED30_CSV
                    ),

                "rows":
                    30,

                "sha256":
                    selected30_sha,
            },

        "baseline_selection":
            {
                "path":
                    str(
                        BASELINE_SELECTION_CSV
                    ),

                "rows":
                    12,

                "sha256":
                    baseline_selection_sha,
            },
    },

    "metric_policy": {
        "authoritative_checkpoint":
            "best_validation_AA",

        "checkpoint_tie_break":
            [
                "higher validation AA",
                "higher validation Macro-F1",
                "higher validation OA",
                "earlier epoch",
            ],

        "primary_reporting_metrics":
            [
                "AA",
                "Macro_F1",
                "OA",
                "Kappa",
            ],

        "official_test_used":
            False,
    },

    "status":
        "PASS",
}


consolidation_bytes = json.dumps(
    consolidation,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


atomic_write_bytes(
    CONSOLIDATION_JSON,
    consolidation_bytes,
)


consolidation_sha = sha256_file(
    CONSOLIDATION_JSON
)


sidecar_text = (
    f"{consolidation_sha}  {CONSOLIDATION_JSON.name}\n"
    f"{all36_sha}  {ALL36_CSV.name}\n"
    f"{selected30_sha}  {SELECTED30_CSV.name}\n"
    f"{baseline_selection_sha}  {BASELINE_SELECTION_CSV.name}\n"
)


atomic_write_bytes(
    CONSOLIDATION_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 14. READ-BACK VERIFICATION
# ==================================================================================================

if sha256_file(
    ALL36_CSV
) != all36_sha:

    raise RuntimeError(
        "STOP: all36 CSV read-back failed."
    )


if sha256_file(
    SELECTED30_CSV
) != selected30_sha:

    raise RuntimeError(
        "STOP: selected30 CSV read-back failed."
    )


if sha256_file(
    BASELINE_SELECTION_CSV
) != baseline_selection_sha:

    raise RuntimeError(
        "STOP: baseline-selection CSV read-back failed."
    )


if sha256_file(
    CONSOLIDATION_JSON
) != consolidation_sha:

    raise RuntimeError(
        "STOP: consolidation JSON read-back failed."
    )


# Project must still be 507 files.
if len(
    [
        p
        for p in ROOT.rglob("*")
        if p.is_file()
    ]
) != 507:

    raise RuntimeError(
        "STOP: Cell16 mutated scientific project footprint."
    )


# ==================================================================================================
# 15. FINAL GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 16 FINAL GATE")
print("=" * 180)

print(
    "Experiment-B training                : 36/36 COMPLETE"
)

print(
    "All completion records               : 36/36 CONSOLIDATED"
)

print(
    "Final selected comparison runs       : 30/30 PASS"
)

print(
    "Validation tuning records            : 12/12 PASS"
)

print(
    "Selected tuning candidates           : 6/6 PASS"
)

print(
    "Authoritative metric checkpoint      : BEST VALIDATION AA"
)

print(
    "Official test                        : SEALED"
)

print(
    "Training performed                   : False"
)

print(
    "Model inference performed            : False"
)

print(
    "Scientific project mutated           : False"
)

print()
print(
    "ALL 36 RESULTS:"
)

print(
    ALL36_CSV
)

print(
    "SHA-256:",
    all36_sha
)

print()
print(
    "SELECTED 30 RESULTS:"
)

print(
    SELECTED30_CSV
)

print(
    "SHA-256:",
    selected30_sha
)

print()
print(
    "BASELINE SELECTION:"
)

print(
    BASELINE_SELECTION_CSV
)

print(
    "SHA-256:",
    baseline_selection_sha
)

print()
print(
    "CONSOLIDATION MANIFEST:"
)

print(
    CONSOLIDATION_JSON
)

print(
    "SHA-256:",
    consolidation_sha
)

print()
print(
    "CELL 16 PASS: True"
)

print(
    "STATUS: "
    "EXPERIMENT_B_36_RUN_RESULTS_CONSOLIDATED__"
    "SELECTED_30_RUN_COMPARISON_SET_FROZEN__"
    "BASELINE_SELECTION_LOG_WRITTEN__"
    "OFFICIAL_TEST_SEALED"
)

print("=" * 180)
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CELL 17
#
# FINAL THREE-SEED VALIDATION SUMMARY + FREEZE
#
# INPUT:
#   Cell16 selected 30-run comparison table
#
# OUTPUT:
#   10 rows = 2 datasets × 5 methods
#   mean ± sample SD over seeds 42,123,3407
#
# IMPORTANT:
#   This is VALIDATION-ONLY reporting.
#   It is NOT an official-test result table.
#
# NO TRAINING
# NO INFERENCE
# NO OFFICIAL-TEST ACCESS
# ==================================================================================================

from pathlib import Path
from datetime import datetime, timezone
import csv
import hashlib
import json
import os
import statistics


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 17")
print("FINAL THREE-SEED VALIDATION SUMMARY + FREEZE")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

RESULTS_DIR = (
    DRIVE_ROOT
    / "results"
    / "notebook06a_experiment_b"
)

SELECTED30_CSV = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_SELECTED30_VALIDATION_RUNS_v1.csv"
)

BASELINE_SELECTION_CSV = (
    RESULTS_DIR
    / "baseline_selection.csv"
)

CONSOLIDATION_JSON = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_RESULTS_CONSOLIDATION_v1.json"
)

SUMMARY_CSV = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_MEAN_STD_v1.csv"
)

SUMMARY_MARKDOWN = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_SUMMARY_TABLE_v1.md"
)

SUMMARY_JSON = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_SUMMARY_FREEZE_v1.json"
)

SUMMARY_SHA = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_SUMMARY_FREEZE_v1__SHA256.txt"
)


# ==================================================================================================
# 2. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):

            h.update(
                block
            )

    return h.hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    tmp = Path(
        str(path)
        +
        ".tmp"
    )


    if tmp.exists():

        tmp.unlink()


    with tmp.open(
        "wb"
    ) as f:

        f.write(
            payload
        )


    os.replace(
        tmp,
        path,
    )


def fmean(
    values,
):

    return float(
        statistics.fmean(
            values
        )
    )


def sample_std(
    values,
):

    if len(
        values
    ) < 2:

        return 0.0


    return float(
        statistics.stdev(
            values
        )
    )


def fmt_mean_std(
    mean,
    std,
):

    return (
        f"{mean:.6f} ± {std:.6f}"
    )


# ==================================================================================================
# 3. AUTHENTICATE CELL16 OUTPUT
# ==================================================================================================

print()
print("-" * 180)
print("[1] CELL16 CONSOLIDATION PARENT GATE")
print("-" * 180)


for path in [
    SELECTED30_CSV,
    BASELINE_SELECTION_CSV,
    CONSOLIDATION_JSON,
]:

    if not path.is_file():

        raise FileNotFoundError(
            "STOP: required Cell16 artifact missing:\n"
            f"{path}"
        )


with CONSOLIDATION_JSON.open(
    "r",
    encoding="utf-8",
) as f:

    consolidation = json.load(
        f
    )


if consolidation.get(
    "status"
) != "PASS":

    raise RuntimeError(
        "STOP: Cell16 consolidation status is not PASS."
    )


if (
    consolidation[
        "scientific_state"
    ][
        "official_test_status"
    ]
    != "SEALED"
):

    raise RuntimeError(
        "STOP: Cell16 did not preserve official-test seal."
    )


selected30_expected_sha = (
    consolidation[
        "result_files"
    ][
        "selected30"
    ][
        "sha256"
    ]
)


baseline_selection_expected_sha = (
    consolidation[
        "result_files"
    ][
        "baseline_selection"
    ][
        "sha256"
    ]
)


if sha256_file(
    SELECTED30_CSV
) != selected30_expected_sha:

    raise RuntimeError(
        "STOP: selected30 CSV SHA mismatch."
    )


if sha256_file(
    BASELINE_SELECTION_CSV
) != baseline_selection_expected_sha:

    raise RuntimeError(
        "STOP: baseline-selection CSV SHA mismatch."
    )


consolidation_sha = sha256_file(
    CONSOLIDATION_JSON
)


print(
    "Cell16 consolidation      : AUTHENTICATED PASS"
)

print(
    "Selected30 result table   : AUTHENTICATED PASS"
)

print(
    "Baseline-selection table  : AUTHENTICATED PASS"
)

print(
    "Official test             : SEALED"
)


# ==================================================================================================
# 4. LOAD SELECTED 30 RUNS
# ==================================================================================================

print()
print("-" * 180)
print("[2] LOAD FINAL SELECTED 30-RUN COMPARISON SET")
print("-" * 180)


with SELECTED30_CSV.open(
    "r",
    encoding="utf-8",
    newline="",
) as f:

    raw_rows = list(
        csv.DictReader(
            f
        )
    )


if len(
    raw_rows
) != 30:

    raise RuntimeError(
        "STOP: selected comparison file does not contain 30 rows."
    )


rows = []


for row in raw_rows:

    converted = dict(
        row
    )


    converted[
        "seed"
    ] = int(
        row[
            "seed"
        ]
    )


    for field in (
        "val_AA",
        "val_Macro_F1",
        "val_OA",
        "val_Kappa",
    ):

        converted[
            field
        ] = float(
            row[
                field
            ]
        )


    if (
        row.get(
            "val_loss",
            ""
        )
        != ""
    ):

        converted[
            "val_loss"
        ] = float(
            row[
                "val_loss"
            ]
        )


    else:

        converted[
            "val_loss"
        ] = None


    rows.append(
        converted
    )


print(
    "Selected result rows : 30/30 PASS"
)


# ==================================================================================================
# 5. GROUP INTO 10 DATASET × METHOD THREE-SEED RESULTS
# ==================================================================================================

print()
print("-" * 180)
print("[3] THREE-SEED GROUP AUDIT")
print("-" * 180)


grouped = {}


for row in rows:

    key = (
        row[
            "dataset"
        ],
        row[
            "method"
        ],
    )


    grouped.setdefault(
        key,
        [],
    ).append(
        row
    )


if len(
    grouped
) != 10:

    raise RuntimeError(
        "STOP: expected exactly 10 dataset×method groups."
    )


for key, group_rows in grouped.items():

    seeds = {
        row[
            "seed"
        ]
        for row in group_rows
    }


    if seeds != {
        42,
        123,
        3407,
    }:

        raise RuntimeError(
            "STOP: incomplete three-seed group:\n"
            f"{key}: seeds={sorted(seeds)}"
        )


    configs = {
        row[
            "selected_config"
        ]
        for row in group_rows
    }


    if len(
        configs
    ) != 1:

        raise RuntimeError(
            "STOP: selected configuration differs across replication seeds:\n"
            f"{key}: {configs}"
        )


print(
    "Dataset × method groups : 10/10 PASS"
)

print(
    "Three seeds per group   : PASS"
)

print(
    "Frozen config per group : PASS"
)


# ==================================================================================================
# 6. COMPUTE MEAN ± SAMPLE SD
# ==================================================================================================

summary_rows = []


for (
    dataset,
    method,
), group_rows in sorted(
    grouped.items()
):

    config = group_rows[
        0
    ][
        "selected_config"
    ]


    aa = [
        row[
            "val_AA"
        ]
        for row in group_rows
    ]

    macro_f1 = [
        row[
            "val_Macro_F1"
        ]
        for row in group_rows
    ]

    oa = [
        row[
            "val_OA"
        ]
        for row in group_rows
    ]

    kappa = [
        row[
            "val_Kappa"
        ]
        for row in group_rows
    ]


    summary_rows.append(
        {
            "dataset":
                dataset,

            "method":
                method,

            "selected_config":
                config,

            "n_seeds":
                3,

            "val_AA_mean":
                fmean(
                    aa
                ),

            "val_AA_std":
                sample_std(
                    aa
                ),

            "val_Macro_F1_mean":
                fmean(
                    macro_f1
                ),

            "val_Macro_F1_std":
                sample_std(
                    macro_f1
                ),

            "val_OA_mean":
                fmean(
                    oa
                ),

            "val_OA_std":
                sample_std(
                    oa
                ),

            "val_Kappa_mean":
                fmean(
                    kappa
                ),

            "val_Kappa_std":
                sample_std(
                    kappa
                ),
        }
    )


if len(
    summary_rows
) != 10:

    raise RuntimeError(
        "STOP: expected 10 summary rows."
    )


# ==================================================================================================
# 7. VALIDATION-ONLY DESCRIPTIVE RANKING
# ==================================================================================================

for dataset in (
    "Pingan",
    "Qingyun",
):

    dataset_rows = [
        row

        for row in summary_rows

        if row[
            "dataset"
        ] == dataset
    ]


    ranked = sorted(
        dataset_rows,
        key=lambda row: (
            -row[
                "val_AA_mean"
            ],
            -row[
                "val_Macro_F1_mean"
            ],
            -row[
                "val_OA_mean"
            ],
            row[
                "method"
            ],
        ),
    )


    for rank, row in enumerate(
        ranked,
        start=1,
    ):

        row[
            "validation_rank_by_AA"
        ] = rank


# ==================================================================================================
# 8. DELTA VERSUS CE
# ==================================================================================================

for dataset in (
    "Pingan",
    "Qingyun",
):

    ce_row = next(
        row

        for row in summary_rows

        if (
            row[
                "dataset"
            ] == dataset
            and
            row[
                "method"
            ] == "CE"
        )
    )


    for row in summary_rows:

        if row[
            "dataset"
        ] != dataset:

            continue


        row[
            "delta_AA_vs_CE"
        ] = (
            row[
                "val_AA_mean"
            ]
            -
            ce_row[
                "val_AA_mean"
            ]
        )


        row[
            "delta_Macro_F1_vs_CE"
        ] = (
            row[
                "val_Macro_F1_mean"
            ]
            -
            ce_row[
                "val_Macro_F1_mean"
            ]
        )


        row[
            "delta_OA_vs_CE"
        ] = (
            row[
                "val_OA_mean"
            ]
            -
            ce_row[
                "val_OA_mean"
            ]
        )


# ==================================================================================================
# 9. PRINT SUMMARY
# ==================================================================================================

print()
print("-" * 180)
print("[4] FINAL VALIDATION SUMMARY")
print("-" * 180)


for dataset in (
    "Pingan",
    "Qingyun",
):

    print()
    print(
        dataset.upper()
    )

    print(
        "-" * 130
    )


    dataset_rows = sorted(
        [
            row
            for row in summary_rows
            if row[
                "dataset"
            ] == dataset
        ],
        key=lambda row: row[
            "validation_rank_by_AA"
        ],
    )


    for row in dataset_rows:

        print(
            f"#{row['validation_rank_by_AA']} "
            f"{row['method']:18s} "
            f"{row['selected_config']:16s} | "
            f"AA {fmt_mean_std(row['val_AA_mean'], row['val_AA_std'])} | "
            f"F1 {fmt_mean_std(row['val_Macro_F1_mean'], row['val_Macro_F1_std'])} | "
            f"OA {fmt_mean_std(row['val_OA_mean'], row['val_OA_std'])} | "
            f"Kappa {fmt_mean_std(row['val_Kappa_mean'], row['val_Kappa_std'])}"
        )


# ==================================================================================================
# 10. WRITE SUMMARY CSV
# ==================================================================================================

for path in [
    SUMMARY_CSV,
    SUMMARY_MARKDOWN,
    SUMMARY_JSON,
    SUMMARY_SHA,
]:

    if path.exists():

        raise RuntimeError(
            "STOP: Cell17 output already exists:\n"
            f"{path}"
        )


summary_fields = [
    "dataset",
    "method",
    "selected_config",
    "n_seeds",
    "validation_rank_by_AA",
    "val_AA_mean",
    "val_AA_std",
    "val_Macro_F1_mean",
    "val_Macro_F1_std",
    "val_OA_mean",
    "val_OA_std",
    "val_Kappa_mean",
    "val_Kappa_std",
    "delta_AA_vs_CE",
    "delta_Macro_F1_vs_CE",
    "delta_OA_vs_CE",
]


import io

csv_buffer = io.StringIO()


writer = csv.DictWriter(
    csv_buffer,
    fieldnames=summary_fields,
    lineterminator="\n",
)


writer.writeheader()


for row in sorted(
    summary_rows,
    key=lambda row: (
        row[
            "dataset"
        ],
        row[
            "validation_rank_by_AA"
        ],
    ),
):

    writer.writerow(
        {
            field:
                row.get(
                    field,
                    ""
                )

            for field in summary_fields
        }
    )


summary_csv_bytes = csv_buffer.getvalue().encode(
    "utf-8"
)


atomic_write_bytes(
    SUMMARY_CSV,
    summary_csv_bytes,
)


summary_csv_sha = sha256_file(
    SUMMARY_CSV
)


# ==================================================================================================
# 11. BUILD MARKDOWN TABLE
# ==================================================================================================

markdown_lines = [
    "# LTLC Notebook 06A — Experiment B Validation Summary",
    "",
    "> Validation-only results. Official test remains SEALED.",
    "",
]


for dataset in (
    "Pingan",
    "Qingyun",
):

    markdown_lines.extend(
        [
            f"## {dataset}",
            "",
            "| Rank | Method | Selected configuration | AA (mean ± SD) | Macro-F1 (mean ± SD) | OA (mean ± SD) | Kappa (mean ± SD) | ΔAA vs CE |",
            "|---:|---|---|---:|---:|---:|---:|---:|",
        ]
    )


    dataset_rows = sorted(
        [
            row
            for row in summary_rows
            if row[
                "dataset"
            ] == dataset
        ],
        key=lambda row: row[
            "validation_rank_by_AA"
        ],
    )


    for row in dataset_rows:

        markdown_lines.append(
            "| "
            f"{row['validation_rank_by_AA']} | "
            f"{row['method']} | "
            f"{row['selected_config']} | "
            f"{fmt_mean_std(row['val_AA_mean'], row['val_AA_std'])} | "
            f"{fmt_mean_std(row['val_Macro_F1_mean'], row['val_Macro_F1_std'])} | "
            f"{fmt_mean_std(row['val_OA_mean'], row['val_OA_std'])} | "
            f"{fmt_mean_std(row['val_Kappa_mean'], row['val_Kappa_std'])} | "
            f"{row['delta_AA_vs_CE']:+.6f} |"
        )


    markdown_lines.append(
        ""
    )


markdown_bytes = (
    "\n".join(
        markdown_lines
    )
    +
    "\n"
).encode(
    "utf-8"
)


atomic_write_bytes(
    SUMMARY_MARKDOWN,
    markdown_bytes,
)


summary_markdown_sha = sha256_file(
    SUMMARY_MARKDOWN
)


# ==================================================================================================
# 12. BUILD VALIDATION SUMMARY FREEZE
# ==================================================================================================

summary_freeze = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_NOTEBOOK06A_EXPERIMENT_B_VALIDATION_SUMMARY_FREEZE",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "parent": {
        "cell16_consolidation_path":
            str(
                CONSOLIDATION_JSON
            ),

        "cell16_consolidation_sha256":
            consolidation_sha,

        "selected30_sha256":
            selected30_expected_sha,

        "baseline_selection_sha256":
            baseline_selection_expected_sha,
    },

    "statistics": {
        "replication_seeds":
            [
                42,
                123,
                3407,
            ],

        "n_per_method_dataset":
            3,

        "mean":
            "arithmetic_mean",

        "standard_deviation":
            "sample_standard_deviation_ddof_1",

        "ranking":
            (
                "validation AA mean descending; "
                "then Macro-F1 mean; then OA mean"
            ),
    },

    "results":
        sorted(
            summary_rows,
            key=lambda row: (
                row[
                    "dataset"
                ],
                row[
                    "validation_rank_by_AA"
                ],
            ),
        ),

    "outputs": {
        "summary_csv":
            {
                "path":
                    str(
                        SUMMARY_CSV
                    ),

                "sha256":
                    summary_csv_sha,
            },

        "summary_markdown":
            {
                "path":
                    str(
                        SUMMARY_MARKDOWN
                    ),

                "sha256":
                    summary_markdown_sha,
            },
    },

    "safety": {
        "validation_only":
            True,

        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "training_performed":
            False,

        "model_inference_performed":
            False,

        "retuning_performed":
            False,
    },

    "interpretation_boundary":
        (
            "These are Experiment-B spatial-validation robustness results. "
            "They must not be described as official-test performance."
        ),

    "status":
        "PASS",
}


summary_json_bytes = json.dumps(
    summary_freeze,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


atomic_write_bytes(
    SUMMARY_JSON,
    summary_json_bytes,
)


summary_json_sha = sha256_file(
    SUMMARY_JSON
)


sidecar_text = (
    f"{summary_json_sha}  {SUMMARY_JSON.name}\n"
    f"{summary_csv_sha}  {SUMMARY_CSV.name}\n"
    f"{summary_markdown_sha}  {SUMMARY_MARKDOWN.name}\n"
    f"{consolidation_sha}  {CONSOLIDATION_JSON.name}\n"
)


atomic_write_bytes(
    SUMMARY_SHA,
    sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 13. VERIFY OUTPUTS + PROJECT IMMUTABILITY
# ==================================================================================================

if sha256_file(
    SUMMARY_JSON
) != summary_json_sha:

    raise RuntimeError(
        "STOP: summary freeze read-back failed."
    )


if sha256_file(
    SUMMARY_CSV
) != summary_csv_sha:

    raise RuntimeError(
        "STOP: summary CSV read-back failed."
    )


if sha256_file(
    SUMMARY_MARKDOWN
) != summary_markdown_sha:

    raise RuntimeError(
        "STOP: summary markdown read-back failed."
    )


if len(
    [
        p
        for p in ROOT.rglob("*")
        if p.is_file()
    ]
) != 507:

    raise RuntimeError(
        "STOP: Cell17 mutated scientific project footprint."
    )


# ==================================================================================================
# 14. FINAL GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 17 FINAL GATE")
print("=" * 180)

print(
    "Selected comparison runs               : 30/30 PASS"
)

print(
    "Dataset × method summaries             : 10/10 PASS"
)

print(
    "Replications per summary               : 3"
)

print(
    "Replication seeds                      : 42, 123, 3407"
)

print(
    "Statistics                             : mean ± sample SD"
)

print(
    "Primary descriptive ranking            : validation AA"
)

print(
    "Official test                          : SEALED"
)

print(
    "Training performed                     : False"
)

print(
    "Model inference performed              : False"
)

print(
    "Retuning performed                     : False"
)

print(
    "Scientific project mutated             : False"
)

print()
print(
    "VALIDATION SUMMARY CSV:"
)

print(
    SUMMARY_CSV
)

print(
    "SHA-256:",
    summary_csv_sha
)

print()
print(
    "VALIDATION SUMMARY MARKDOWN:"
)

print(
    SUMMARY_MARKDOWN
)

print(
    "SHA-256:",
    summary_markdown_sha
)

print()
print(
    "VALIDATION SUMMARY FREEZE:"
)

print(
    SUMMARY_JSON
)

print(
    "SHA-256:",
    summary_json_sha
)

print()
print(
    "CELL 17 PASS: True"
)

print(
    "STATUS: "
    "EXPERIMENT_B_VALIDATION_RESULTS_AGGREGATED__"
    "10_THREE_SEED_SUMMARIES_FROZEN__"
    "MEAN_PLUS_SAMPLE_SD_COMPLETE__"
    "OFFICIAL_TEST_SEALED"
)

print("=" * 180)
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CELL 18
#
# FINAL RELEASE / ARCHIVAL PACKAGE
#
# This is the terminal Notebook06A Experiment-B packaging cell.
#
# The large 36/36 persistence ZIP is NOT duplicated inside this release archive.
# Instead, its exact Drive path / size / SHA-256 are frozen in the release manifest.
#
# NO TRAINING
# NO INFERENCE
# NO TEST ACCESS
# NO PROJECT MUTATION
# ==================================================================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import os
import zipfile


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 18")
print("FINAL NOTEBOOK06A RELEASE / ARCHIVE PACKAGE")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

ROOT = Path(
    "/content/LTLC"
)

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)

FINAL_PERSISTENCE_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "36of36_Post_Qingyun_LA_tau0p5_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

FINAL_PERSISTENCE_ZIP = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}.zip"
)

FINAL_PERSISTENCE_MANIFEST = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__MANIFEST.json"
)

FINAL_PERSISTENCE_SHA = (
    PERSISTENCE_DIR
    / f"{FINAL_PERSISTENCE_BASENAME}__SHA256.txt"
)


FINAL_AUDIT_DIR = (
    DRIVE_ROOT
    / "final_audits"
    / "notebook06a_experiment_b"
)

FINAL_AUDIT_JSON = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1.json"
)

FINAL_RUN_LEDGER = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_RUN_LEDGER_v1.csv"
)

FINAL_AUDIT_SHA = (
    FINAL_AUDIT_DIR
    / "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1__SHA256.txt"
)


RESULTS_DIR = (
    DRIVE_ROOT
    / "results"
    / "notebook06a_experiment_b"
)

ALL36_CSV = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_ALL36_VALIDATION_RUNS_v1.csv"
)

SELECTED30_CSV = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_SELECTED30_VALIDATION_RUNS_v1.csv"
)

BASELINE_SELECTION_CSV = (
    RESULTS_DIR
    / "baseline_selection.csv"
)

CONSOLIDATION_JSON = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_RESULTS_CONSOLIDATION_v1.json"
)

CONSOLIDATION_SHA = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_RESULTS_CONSOLIDATION_v1__SHA256.txt"
)

SUMMARY_CSV = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_MEAN_STD_v1.csv"
)

SUMMARY_MARKDOWN = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_SUMMARY_TABLE_v1.md"
)

SUMMARY_JSON = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_SUMMARY_FREEZE_v1.json"
)

SUMMARY_SHA = (
    RESULTS_DIR
    / "LTLC_Notebook06A_ExperimentB_VALIDATION_SUMMARY_FREEZE_v1__SHA256.txt"
)


RELEASE_DIR = (
    DRIVE_ROOT
    / "releases"
    / "notebook06a_experiment_b"
)

RELEASE_MANIFEST = (
    RELEASE_DIR
    / "LTLC_Notebook06A_ExperimentB_FINAL_RELEASE_v1.json"
)

RELEASE_ZIP = (
    RELEASE_DIR
    / "LTLC_Notebook06A_ExperimentB_FINAL_RELEASE_v1.zip"
)

RELEASE_SHA = (
    RELEASE_DIR
    / "LTLC_Notebook06A_ExperimentB_FINAL_RELEASE_v1__SHA256.txt"
)


# ==================================================================================================
# 2. EXPECTED AUTHENTICATED FINAL SCIENTIFIC STATE
# ==================================================================================================

EXPECTED_FINAL_PERSISTENCE_ZIP_SHA256 = (
    "30ba228994f0a16e756c5ade22480b00"
    "f3e2304ee3e15bd8e00409d86dc2b7b8"
)

EXPECTED_FINAL_PERSISTENCE_MANIFEST_SHA256 = (
    "5f93f457f3b1510d72b366aa8a78aa65"
    "fbf59e866dadfcefd7388d80f81c79b9"
)

EXPECTED_FINAL_AUDIT_SHA256 = (
    "2a24fdbaba212890443413c39e2873b2"
    "9862e4810dda91a2ed5879a7b6d59768"
)

EXPECTED_FINAL_RUN_LEDGER_SHA256 = (
    "6335262b572a389a33c26b877c7c7f78"
    "5934bee1ab53844f8fa6b5f61848b07d"
)


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as f:

        for block in iter(
            lambda: f.read(
                chunk_size
            ),
            b"",
        ):

            h.update(
                block
            )


    return h.hexdigest()


def sha256_bytes(
    payload,
):

    return hashlib.sha256(
        payload
    ).hexdigest()


def atomic_write_bytes(
    path,
    payload,
):

    path = Path(
        path
    )


    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    tmp = Path(
        str(
            path
        )
        +
        ".tmp"
    )


    if tmp.exists():

        tmp.unlink()


    with tmp.open(
        "wb"
    ) as f:

        f.write(
            payload
        )


    os.replace(
        tmp,
        path,
    )


# ==================================================================================================
# 4. REQUIRED ARTIFACT GATE
# ==================================================================================================

print()
print("-" * 180)
print("[1] REQUIRED TERMINAL ARTIFACTS")
print("-" * 180)


required_files = [
    FINAL_PERSISTENCE_ZIP,
    FINAL_PERSISTENCE_MANIFEST,
    FINAL_PERSISTENCE_SHA,
    FINAL_AUDIT_JSON,
    FINAL_RUN_LEDGER,
    FINAL_AUDIT_SHA,
    ALL36_CSV,
    SELECTED30_CSV,
    BASELINE_SELECTION_CSV,
    CONSOLIDATION_JSON,
    CONSOLIDATION_SHA,
    SUMMARY_CSV,
    SUMMARY_MARKDOWN,
    SUMMARY_JSON,
    SUMMARY_SHA,
]


for path in required_files:

    if not path.is_file():

        raise FileNotFoundError(
            "STOP: required release artifact missing:\n"
            f"{path}"
        )


print(
    "Required terminal artifacts:",
    f"{len(required_files)}/{len(required_files)} FOUND"
)


# ==================================================================================================
# 5. AUTHENTICATE TERMINAL SCIENTIFIC PARENTS
# ==================================================================================================

print()
print("-" * 180)
print("[2] AUTHENTICATE TERMINAL SCIENTIFIC PARENTS")
print("-" * 180)


# The final persistence ZIP is 63 MB; one terminal SHA pass is intentional.
final_persistence_zip_sha = sha256_file(
    FINAL_PERSISTENCE_ZIP
)


if (
    final_persistence_zip_sha
    !=
    EXPECTED_FINAL_PERSISTENCE_ZIP_SHA256
):

    raise RuntimeError(
        "STOP: final 36/36 persistence ZIP SHA mismatch."
    )


final_persistence_manifest_sha = sha256_file(
    FINAL_PERSISTENCE_MANIFEST
)


if (
    final_persistence_manifest_sha
    !=
    EXPECTED_FINAL_PERSISTENCE_MANIFEST_SHA256
):

    raise RuntimeError(
        "STOP: final persistence manifest SHA mismatch."
    )


final_audit_sha = sha256_file(
    FINAL_AUDIT_JSON
)


if (
    final_audit_sha
    !=
    EXPECTED_FINAL_AUDIT_SHA256
):

    raise RuntimeError(
        "STOP: final closing-audit SHA mismatch."
    )


final_ledger_sha = sha256_file(
    FINAL_RUN_LEDGER
)


if (
    final_ledger_sha
    !=
    EXPECTED_FINAL_RUN_LEDGER_SHA256
):

    raise RuntimeError(
        "STOP: final 36-run ledger SHA mismatch."
    )


with FINAL_AUDIT_JSON.open(
    "r",
    encoding="utf-8",
) as f:

    final_audit_payload = json.load(
        f
    )


if final_audit_payload.get(
    "closing_status"
) != "PASS":

    raise RuntimeError(
        "STOP: final closing audit does not say PASS."
    )


if (
    final_audit_payload[
        "official_test"
    ][
        "status"
    ]
    != "SEALED"
):

    raise RuntimeError(
        "STOP: official test is not SEALED in final audit."
    )


print(
    "Final 36/36 persistence ZIP      : AUTHENTICATED PASS"
)

print(
    "Final persistence manifest       : AUTHENTICATED PASS"
)

print(
    "Final closing audit              : AUTHENTICATED PASS"
)

print(
    "Final 36-run ledger              : AUTHENTICATED PASS"
)

print(
    "Official test                    : SEALED"
)


# ==================================================================================================
# 6. AUTHENTICATE CELL16 / CELL17 ARTIFACT CHAINS
# ==================================================================================================

print()
print("-" * 180)
print("[3] AUTHENTICATE RESULTS-CONSOLIDATION CHAIN")
print("-" * 180)


with CONSOLIDATION_JSON.open(
    "r",
    encoding="utf-8",
) as f:

    consolidation = json.load(
        f
    )


if consolidation.get(
    "status"
) != "PASS":

    raise RuntimeError(
        "STOP: Cell16 consolidation is not PASS."
    )


if (
    consolidation[
        "parent"
    ][
        "final_closing_audit_sha256"
    ]
    !=
    final_audit_sha
):

    raise RuntimeError(
        "STOP: Cell16 parent audit does not match terminal audit."
    )


if (
    sha256_file(
        ALL36_CSV
    )
    !=
    consolidation[
        "result_files"
    ][
        "all36"
    ][
        "sha256"
    ]
):

    raise RuntimeError(
        "STOP: all36 results SHA mismatch."
    )


if (
    sha256_file(
        SELECTED30_CSV
    )
    !=
    consolidation[
        "result_files"
    ][
        "selected30"
    ][
        "sha256"
    ]
):

    raise RuntimeError(
        "STOP: selected30 results SHA mismatch."
    )


if (
    sha256_file(
        BASELINE_SELECTION_CSV
    )
    !=
    consolidation[
        "result_files"
    ][
        "baseline_selection"
    ][
        "sha256"
    ]
):

    raise RuntimeError(
        "STOP: baseline-selection SHA mismatch."
    )


consolidation_sha = sha256_file(
    CONSOLIDATION_JSON
)


with SUMMARY_JSON.open(
    "r",
    encoding="utf-8",
) as f:

    summary = json.load(
        f
    )


if summary.get(
    "status"
) != "PASS":

    raise RuntimeError(
        "STOP: Cell17 summary freeze is not PASS."
    )


if (
    summary[
        "parent"
    ][
        "cell16_consolidation_sha256"
    ]
    !=
    consolidation_sha
):

    raise RuntimeError(
        "STOP: Cell17 summary parent does not match Cell16."
    )


if (
    summary[
        "safety"
    ][
        "official_test_status"
    ]
    != "SEALED"
):

    raise RuntimeError(
        "STOP: Cell17 summary does not preserve test seal."
    )


if (
    sha256_file(
        SUMMARY_CSV
    )
    !=
    summary[
        "outputs"
    ][
        "summary_csv"
    ][
        "sha256"
    ]
):

    raise RuntimeError(
        "STOP: summary CSV SHA mismatch."
    )


if (
    sha256_file(
        SUMMARY_MARKDOWN
    )
    !=
    summary[
        "outputs"
    ][
        "summary_markdown"
    ][
        "sha256"
    ]
):

    raise RuntimeError(
        "STOP: summary markdown SHA mismatch."
    )


summary_sha = sha256_file(
    SUMMARY_JSON
)


print(
    "Cell16 consolidation chain : PASS"
)

print(
    "Cell17 summary-freeze chain: PASS"
)

print(
    "Validation-only boundary   : PASS"
)

print(
    "Official test              : SEALED"
)


# ==================================================================================================
# 7. RELEASE MEMBER INVENTORY
# ==================================================================================================

# The large persistence ZIP is deliberately excluded to avoid a duplicate ~63 MB copy.
release_source_files = [
    (
        "audit/final_closing_audit.json",
        FINAL_AUDIT_JSON,
    ),
    (
        "audit/final_36_run_ledger.csv",
        FINAL_RUN_LEDGER,
    ),
    (
        "audit/final_audit_sha256.txt",
        FINAL_AUDIT_SHA,
    ),
    (
        "persistence/final_36of36_manifest.json",
        FINAL_PERSISTENCE_MANIFEST,
    ),
    (
        "persistence/final_36of36_sha256.txt",
        FINAL_PERSISTENCE_SHA,
    ),
    (
        "results/all36_validation_runs.csv",
        ALL36_CSV,
    ),
    (
        "results/selected30_validation_runs.csv",
        SELECTED30_CSV,
    ),
    (
        "results/baseline_selection.csv",
        BASELINE_SELECTION_CSV,
    ),
    (
        "results/results_consolidation.json",
        CONSOLIDATION_JSON,
    ),
    (
        "results/results_consolidation_sha256.txt",
        CONSOLIDATION_SHA,
    ),
    (
        "results/validation_mean_std.csv",
        SUMMARY_CSV,
    ),
    (
        "results/validation_summary_table.md",
        SUMMARY_MARKDOWN,
    ),
    (
        "results/validation_summary_freeze.json",
        SUMMARY_JSON,
    ),
    (
        "results/validation_summary_freeze_sha256.txt",
        SUMMARY_SHA,
    ),
]


release_records = []


for archive_name, source_path in release_source_files:

    release_records.append(
        {
            "archive_name":
                archive_name,

            "source_path":
                str(
                    source_path
                ),

            "size_bytes":
                int(
                    source_path.stat().st_size
                ),

            "sha256":
                sha256_file(
                    source_path
                ),
        }
    )


# ==================================================================================================
# 8. BUILD PERSISTENCE POINTER
# ==================================================================================================

persistence_pointer = {
    "artifact_type":
        "LTLC_NOTEBOOK06A_FINAL_36OF36_PERSISTENCE_POINTER",

    "path":
        str(
            FINAL_PERSISTENCE_ZIP
        ),

    "filename":
        FINAL_PERSISTENCE_ZIP.name,

    "size_bytes":
        int(
            FINAL_PERSISTENCE_ZIP.stat().st_size
        ),

    "sha256":
        final_persistence_zip_sha,

    "manifest_path":
        str(
            FINAL_PERSISTENCE_MANIFEST
        ),

    "manifest_sha256":
        final_persistence_manifest_sha,

    "note":
        (
            "The large authenticated 36/36 cumulative-delta ZIP is stored "
            "separately in Google Drive and is intentionally not duplicated "
            "inside the compact final release archive."
        ),
}


persistence_pointer_bytes = json.dumps(
    persistence_pointer,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


# ==================================================================================================
# 9. BUILD README
# ==================================================================================================

readme_text = f"""LTLC NOTEBOOK 06A / EXPERIMENT B — FINAL RELEASE

Notebook:
06A_Spatially_Disjoint_Validation_Robustness_Study

Final scientific status:
- Experiment B: 36/36 COMPLETE
- CE: 6/6
- Focal: 8/8
- LDAM-DRW: 8/8
- Balanced Softmax: 6/6
- LA-loss: 8/8
- Pingan selected LA tau: 1.0
- Qingyun selected LA tau: 0.5
- Official test: SEALED
- Official-test inference: NOT PERFORMED
- GPU training remaining: NONE

Scientific snapshot:
- 507 authenticated project files
- 507/507 SHA verification PASS

Final 36/36 persistence ZIP:
{FINAL_PERSISTENCE_ZIP}

Final persistence SHA-256:
{final_persistence_zip_sha}

Final closing audit SHA-256:
{final_audit_sha}

This compact release contains the audit records, validation-result
consolidation, selected-run results, baseline-selection log, validation
mean±SD table, and SHA records.

The large 36/36 persistence ZIP remains stored separately and is linked by
persistence/PERSISTENCE_POINTER.json.

IMPORTANT:
The Experiment-B tables in this release are validation-only robustness
results. They must not be described as official-test results.

Colab notebook file:
The live .ipynb itself is not automatically captured by this runtime
archive. Ensure the notebook is saved in Google Drive from the Colab UI
after this cell completes.
"""


readme_bytes = readme_text.encode(
    "utf-8"
)


# ==================================================================================================
# 10. BUILD RELEASE MANIFEST
# ==================================================================================================

release_manifest_payload = {
    "schema_version":
        1,

    "artifact_type":
        "LTLC_NOTEBOOK06A_EXPERIMENT_B_FINAL_RELEASE",

    "project":
        "LTLC",

    "notebook":
        "06A_Spatially_Disjoint_Validation_Robustness_Study",

    "experiment":
        "Experiment B",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "terminal_state": {
        "experiment_b_complete":
            True,

        "experiment_b_runs":
            "36/36",

        "scientific_project_files":
            507,

        "full_project_sha_verification":
            "507/507 PASS",

        "gpu_training_remaining":
            False,

        "official_test_status":
            "SEALED",

        "official_test_values_accessed":
            False,

        "official_test_inference_performed":
            False,
    },

    "authoritative_parents": {
        "final_persistence_zip_sha256":
            final_persistence_zip_sha,

        "final_persistence_manifest_sha256":
            final_persistence_manifest_sha,

        "final_closing_audit_sha256":
            final_audit_sha,

        "final_36_run_ledger_sha256":
            final_ledger_sha,

        "results_consolidation_sha256":
            consolidation_sha,

        "validation_summary_freeze_sha256":
            summary_sha,
    },

    "large_persistence_artifact":
        persistence_pointer,

    "release_members":
        release_records,

    "release_notes": {
        "scientific_project_mutated":
            False,

        "training_performed":
            False,

        "model_inference_performed":
            False,

        "retuning_performed":
            False,

        "validation_results_are_official_test_results":
            False,
    },

    "status":
        "PASS",
}


release_manifest_bytes = json.dumps(
    release_manifest_payload,
    indent=2,
    sort_keys=True,
).encode(
    "utf-8"
)


release_manifest_sha = hashlib.sha256(
    release_manifest_bytes
).hexdigest()


# ==================================================================================================
# 11. CREATE FINAL RELEASE DIRECTORY
# ==================================================================================================

RELEASE_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


for path in [
    RELEASE_MANIFEST,
    RELEASE_ZIP,
    RELEASE_SHA,
]:

    if path.exists():

        raise RuntimeError(
            "STOP: final release artifact already exists:\n"
            f"{path}"
        )


atomic_write_bytes(
    RELEASE_MANIFEST,
    release_manifest_bytes,
)


if sha256_file(
    RELEASE_MANIFEST
) != release_manifest_sha:

    raise RuntimeError(
        "STOP: release-manifest write verification failed."
    )


# ==================================================================================================
# 12. CREATE COMPACT FINAL RELEASE ZIP
# ==================================================================================================

zip_tmp = Path(
    str(
        RELEASE_ZIP
    )
    +
    ".tmp"
)


if zip_tmp.exists():

    zip_tmp.unlink()


with zipfile.ZipFile(
    zip_tmp,
    mode="w",
    compression=zipfile.ZIP_DEFLATED,
    allowZip64=True,
) as zf:

    zf.writestr(
        "README.txt",
        readme_bytes,
    )


    zf.writestr(
        "FINAL_RELEASE_MANIFEST.json",
        release_manifest_bytes,
    )


    zf.writestr(
        "persistence/PERSISTENCE_POINTER.json",
        persistence_pointer_bytes,
    )


    for record in release_records:

        source_path = Path(
            record[
                "source_path"
            ]
        )


        if sha256_file(
            source_path
        ) != record[
            "sha256"
        ]:

            raise RuntimeError(
                "STOP: release member changed before archive:\n"
                f"{source_path}"
            )


        zf.write(
            source_path,
            arcname=record[
                "archive_name"
            ],
        )


os.replace(
    zip_tmp,
    RELEASE_ZIP,
)


# ==================================================================================================
# 13. VERIFY RELEASE ZIP CONTENT
# ==================================================================================================

with zipfile.ZipFile(
    RELEASE_ZIP,
    mode="r",
) as zf:

    observed_members = set(
        zf.namelist()
    )


    expected_members = {
        "README.txt",
        "FINAL_RELEASE_MANIFEST.json",
        "persistence/PERSISTENCE_POINTER.json",
    }


    expected_members.update(
        record[
            "archive_name"
        ]
        for record in release_records
    )


    if observed_members != expected_members:

        raise RuntimeError(
            "STOP: release ZIP member set mismatch."
        )


    if (
        zf.read(
            "FINAL_RELEASE_MANIFEST.json"
        )
        !=
        release_manifest_bytes
    ):

        raise RuntimeError(
            "STOP: archived release manifest differs."
        )


    if (
        zf.read(
            "persistence/PERSISTENCE_POINTER.json"
        )
        !=
        persistence_pointer_bytes
    ):

        raise RuntimeError(
            "STOP: archived persistence pointer differs."
        )


    if (
        zf.read(
            "README.txt"
        )
        !=
        readme_bytes
    ):

        raise RuntimeError(
            "STOP: archived README differs."
        )


    for record in release_records:

        archived_bytes = zf.read(
            record[
                "archive_name"
            ]
        )


        if hashlib.sha256(
            archived_bytes
        ).hexdigest() != record[
            "sha256"
        ]:

            raise RuntimeError(
                "STOP: archived release-member SHA mismatch:\n"
                f"{record['archive_name']}"
            )


# ==================================================================================================
# 14. FINAL RELEASE SHA SIDECAR
# ==================================================================================================

release_zip_sha = sha256_file(
    RELEASE_ZIP
)

release_zip_size = int(
    RELEASE_ZIP.stat().st_size
)


release_sidecar_text = (
    f"{release_zip_sha}  {RELEASE_ZIP.name}\n"
    f"{release_manifest_sha}  {RELEASE_MANIFEST.name}\n"
    f"{final_persistence_zip_sha}  {FINAL_PERSISTENCE_ZIP.name}\n"
    f"{final_audit_sha}  {FINAL_AUDIT_JSON.name}\n"
    f"{consolidation_sha}  {CONSOLIDATION_JSON.name}\n"
    f"{summary_sha}  {SUMMARY_JSON.name}\n"
)


atomic_write_bytes(
    RELEASE_SHA,
    release_sidecar_text.encode(
        "utf-8"
    ),
)


os.sync()


# ==================================================================================================
# 15. FINAL SCIENTIFIC IMMUTABILITY GATE
# ==================================================================================================

project_files_after_release = [
    p
    for p in ROOT.rglob("*")
    if p.is_file()
]


if len(
    project_files_after_release
) != 507:

    raise RuntimeError(
        "STOP: final release packaging mutated scientific project."
    )


# ==================================================================================================
# 16. NOTEBOOK06A TERMINAL GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CELL 18 FINAL TERMINAL GATE")
print("=" * 180)

print(
    "Experiment-B                             : 36/36 COMPLETE"
)

print(
    "Scientific snapshot                      : 507 FILES"
)

print(
    "Full scientific snapshot audit           : 507/507 PASS"
)

print(
    "Final closing audit                      : AUTHENTICATED PASS"
)

print(
    "36-run results consolidation             : PASS"
)

print(
    "Selected 30-run comparison set           : PASS"
)

print(
    "Validation three-seed summary             : 10/10 PASS"
)

print(
    "Baseline-selection log                   : PASS"
)

print(
    "Final 36/36 persistence                  : AUTHENTICATED PASS"
)

print(
    "Final release archive                    : VERIFIED PASS"
)

print(
    "Scientific project mutated by Cell18     : False"
)

print(
    "Training performed in Cell18              : False"
)

print(
    "Model inference performed in Cell18       : False"
)

print(
    "GPU training remaining                    : False"
)

print(
    "Official test                             : SEALED"
)

print(
    "Official-test inference                   : NOT PERFORMED"
)

print()
print(
    "FINAL NOTEBOOK06A RELEASE ZIP:"
)

print(
    RELEASE_ZIP
)

print()
print(
    "RELEASE ZIP SIZE BYTES:"
)

print(
    release_zip_size
)

print()
print(
    "RELEASE ZIP SHA-256:"
)

print(
    release_zip_sha
)

print()
print(
    "FINAL RELEASE MANIFEST:"
)

print(
    RELEASE_MANIFEST
)

print()
print(
    "RELEASE MANIFEST SHA-256:"
)

print(
    release_manifest_sha
)

print()
print(
    "FINAL RELEASE SHA SIDECAR:"
)

print(
    RELEASE_SHA
)

print()
print(
    "CELL 18 PASS: True"
)

print()
print(
    "STATUS: "
    "NOTEBOOK06A_EXPERIMENT_B_COMPLETE__"
    "36_OF_36_TRAINING_COMPLETE__"
    "FINAL_507_FILE_SNAPSHOT_AUTHENTICATED__"
    "VALIDATION_RESULTS_CONSOLIDATED__"
    "THREE_SEED_SUMMARY_FROZEN__"
    "FINAL_RELEASE_ARCHIVE_VERIFIED__"
    "OFFICIAL_TEST_SEALED__"
    "NO_GPU_WORK_REMAINING"
)

print("=" * 180)

print()
print("NOTEBOOK 06A / EXPERIMENT B — TERMINAL SCIENTIFIC WORKFLOW COMPLETE")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CORRECTION CELL 16A
#
# RESOLVE CELL15B FINAL-AUDIT PARENT + DETECT CURRENT RUNTIME STATE
#
# NO TRAINING
# NO MODEL INFERENCE
# NO OFFICIAL-TEST ACCESS
# NO SCIENTIFIC MUTATION
# ==================================================================================================

from pathlib import Path
import hashlib
import json
import os


print("=" * 170)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 16A")
print("RESOLVE CELL15B PARENT + CURRENT RUNTIME STATE")
print("=" * 170)


# --------------------------------------------------------------------------------------------------
# 1. MOUNT GOOGLE DRIVE IF NEEDED
# --------------------------------------------------------------------------------------------------

from google.colab import drive

DRIVE_MOUNT = Path("/content/drive")
MYDRIVE = DRIVE_MOUNT / "MyDrive"


if not MYDRIVE.is_dir():

    print()
    print("Google Drive currently not mounted.")
    print("Mounting now...")

    drive.mount(
        "/content/drive",
        force_remount=False,
    )

else:

    print()
    print("Google Drive already mounted: PASS")


# --------------------------------------------------------------------------------------------------
# 2. EXPECTED FINAL CELL15B ARTIFACTS
# --------------------------------------------------------------------------------------------------

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

FINAL_AUDIT_DIR = (
    DRIVE_ROOT
    / "final_audits"
    / "notebook06a_experiment_b"
)

EXPECTED_AUDIT_NAME = (
    "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1.json"
)

EXPECTED_LEDGER_NAME = (
    "LTLC_Notebook06A_ExperimentB_36of36_FINAL_RUN_LEDGER_v1.csv"
)

EXPECTED_SIDECAR_NAME = (
    "LTLC_Notebook06A_ExperimentB_36of36_FINAL_CLOSING_AUDIT_v1__SHA256.txt"
)


EXPECTED_AUDIT_SHA = (
    "2a24fdbaba212890443413c39e2873b2"
    "9862e4810dda91a2ed5879a7b6d59768"
)

EXPECTED_LEDGER_SHA = (
    "6335262b572a389a33c26b877c7c7f78"
    "5934bee1ab53844f8fa6b5f61848b07d"
)


def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


expected_audit = (
    FINAL_AUDIT_DIR
    / EXPECTED_AUDIT_NAME
)

expected_ledger = (
    FINAL_AUDIT_DIR
    / EXPECTED_LEDGER_NAME
)

expected_sidecar = (
    FINAL_AUDIT_DIR
    / EXPECTED_SIDECAR_NAME
)


# --------------------------------------------------------------------------------------------------
# 3. LOCATE FINAL AUDIT ARTIFACTS
# --------------------------------------------------------------------------------------------------

print()
print("-" * 170)
print("[1] CELL15B FINAL-AUDIT ARTIFACT LOCATION")
print("-" * 170)


def find_unique(
    root,
    filename,
):

    if not root.is_dir():

        return None, []


    matches = list(
        root.rglob(filename)
    )


    if len(matches) == 1:

        return matches[0], matches


    return None, matches


if expected_audit.is_file():

    audit_path = expected_audit

else:

    audit_path, audit_matches = find_unique(
        DRIVE_ROOT,
        EXPECTED_AUDIT_NAME,
    )


    if audit_path is None:

        print(
            "Expected audit path:",
            expected_audit
        )

        print(
            "Matching audit files found:",
            len(audit_matches)
        )

        raise FileNotFoundError(
            "STOP: final Cell15B closing-audit JSON could not be found in Google Drive."
        )


if expected_ledger.is_file():

    ledger_path = expected_ledger

else:

    ledger_path, ledger_matches = find_unique(
        DRIVE_ROOT,
        EXPECTED_LEDGER_NAME,
    )


    if ledger_path is None:

        print(
            "Expected ledger path:",
            expected_ledger
        )

        print(
            "Matching ledger files found:",
            len(ledger_matches)
        )

        raise FileNotFoundError(
            "STOP: final Cell15B run-ledger CSV could not be found in Google Drive."
        )


if expected_sidecar.is_file():

    sidecar_path = expected_sidecar

else:

    sidecar_path, sidecar_matches = find_unique(
        DRIVE_ROOT,
        EXPECTED_SIDECAR_NAME,
    )


    if sidecar_path is None:

        print(
            "Expected sidecar path:",
            expected_sidecar
        )

        print(
            "Matching sidecars found:",
            len(sidecar_matches)
        )

        raise FileNotFoundError(
            "STOP: final Cell15B SHA sidecar could not be found in Google Drive."
        )


print(
    "Final audit  :",
    audit_path
)

print(
    "Final ledger :",
    ledger_path
)

print(
    "SHA sidecar  :",
    sidecar_path
)


# --------------------------------------------------------------------------------------------------
# 4. AUTHENTICATE CELL15B OUTPUTS
# --------------------------------------------------------------------------------------------------

print()
print("-" * 170)
print("[2] CELL15B PARENT AUTHENTICATION")
print("-" * 170)


audit_sha = sha256_file(
    audit_path
)

ledger_sha = sha256_file(
    ledger_path
)


print(
    "Audit SHA expected :",
    EXPECTED_AUDIT_SHA
)

print(
    "Audit SHA actual   :",
    audit_sha
)

print(
    "Ledger SHA expected:",
    EXPECTED_LEDGER_SHA
)

print(
    "Ledger SHA actual  :",
    ledger_sha
)


if audit_sha != EXPECTED_AUDIT_SHA:

    raise RuntimeError(
        "STOP: Cell15B final audit SHA mismatch."
    )


if ledger_sha != EXPECTED_LEDGER_SHA:

    raise RuntimeError(
        "STOP: Cell15B final ledger SHA mismatch."
    )


with audit_path.open(
    "r",
    encoding="utf-8",
) as f:

    audit_payload = json.load(f)


if audit_payload.get(
    "closing_status"
) != "PASS":

    raise RuntimeError(
        "STOP: Cell15B final closing audit does not say PASS."
    )


if (
    audit_payload[
        "final_state"
    ][
        "experiment_b_complete"
    ]
    is not True
):

    raise RuntimeError(
        "STOP: Cell15B does not mark Experiment B complete."
    )


if (
    audit_payload[
        "official_test"
    ][
        "status"
    ]
    != "SEALED"
):

    raise RuntimeError(
        "STOP: Cell15B official-test status is not SEALED."
    )


print()
print(
    "Final closing audit : AUTHENTICATED PASS"
)

print(
    "Final 36-run ledger : AUTHENTICATED PASS"
)

print(
    "Experiment-B        : 36/36 COMPLETE"
)

print(
    "Official test       : SEALED"
)


# --------------------------------------------------------------------------------------------------
# 5. DETECT LOCAL SCIENTIFIC RUNTIME STATE
# --------------------------------------------------------------------------------------------------

print()
print("-" * 170)
print("[3] CURRENT /content/LTLC STATE")
print("-" * 170)


ROOT = Path(
    "/content/LTLC"
)


if not ROOT.is_dir():

    print(
        "/content/LTLC : MISSING"
    )

    print()
    print(
        "Runtime classification : FRESH/RESET RUNTIME"
    )

    print(
        "34/36 or 36/36 recovery needed before Cell16 : YES"
    )

    print(
        "Training required : NO"
    )

    print(
        "Official test     : SEALED"
    )

    print()
    print(
        "STATUS: CELL15B_PARENT_AUTHENTICATED__LOCAL_PROJECT_MISSING__36OF36_RECOVERY_REQUIRED"
    )

    print("=" * 170)
    print()
    print("OUTPUT CHECKPOINT — STOP")

else:

    local_files = [
        p
        for p in ROOT.rglob("*")
        if p.is_file()
    ]


    print(
        "/content/LTLC exists : YES"
    )

    print(
        "Local project files :",
        len(local_files)
    )


    if len(local_files) != 507:

        print()
        print(
            "Runtime classification : LOCAL PROJECT NOT EXACT"
        )

        print(
            "36/36 recovery needed before Cell16 : YES"
        )

        print(
            "Training required : NO"
        )

        print(
            "Official test     : SEALED"
        )

        print()
        print(
            "STATUS: CELL15B_PARENT_AUTHENTICATED__LOCAL_PROJECT_NOT_507__36OF36_RECOVERY_REQUIRED"
        )

    else:

        print()
        print(
            "Runtime classification : LIVE FINAL 36/36 STATE"
        )

        print(
            "Scientific project      : 507 files PASS"
        )

        print(
            "36/36 recovery required : NO"
        )

        print(
            "Cell16 may be rerun      : YES"
        )

        print(
            "Official test            : SEALED"
        )

        print()
        print(
            "STATUS: CELL15B_PARENT_AUTHENTICATED__LIVE_507_FILE_STATE_PRESENT__CELL16_READY"
        )


    print("=" * 170)

    print()
    print("OUTPUT CHECKPOINT — STOP")
```

```python
# ==================================================================================================
# LTLC NOTEBOOK 06A / EXPERIMENT B
# CORRECTION CELL 16B
#
# AUTHENTICATED FINAL 36/36 RUNTIME RECOVERY
#
# SOURCE A:
#   Authenticated self-contained 28/36 migration
#   449 scientific project files
#
# SOURCE B:
#   Authenticated final 36/36 cumulative delta
#   58 post-migration scientific project files
#
# RESULT:
#   /content/LTLC
#   exact 507-file final Experiment-B scientific state
#
# NO TRAINING
# NO MODEL INFERENCE
# NO OFFICIAL-TEST ACCESS
# NO RETUNING
# ==================================================================================================

from pathlib import Path, PurePosixPath
import hashlib
import json
import os
import shutil
import zipfile


print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 16B")
print("AUTHENTICATED FINAL 36/36 RUNTIME RECOVERY")
print("=" * 180)


# ==================================================================================================
# 1. PATHS
# ==================================================================================================

DRIVE_ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

MIGRATION_DIR = (
    DRIVE_ROOT
    / "migrations"
)

PERSISTENCE_DIR = (
    DRIVE_ROOT
    / "experiment_b_colab_persistence"
)


BASE_ZIP = (
    MIGRATION_DIR
    / "LTLC_Notebook06A_ExperimentB_28of36_PreLA_Kaggle_to_Colab_SELF_CONTAINED_MIGRATION_v1.zip"
)


FINAL_BASENAME = (
    "LTLC_Notebook06A_ExperimentB_"
    "36of36_Post_Qingyun_LA_tau0p5_Seed3407_"
    "FULL_DELTA_FROM_28of36_MIGRATION_v1"
)

FINAL_DELTA_ZIP = (
    PERSISTENCE_DIR
    / f"{FINAL_BASENAME}.zip"
)

FINAL_DELTA_MANIFEST = (
    PERSISTENCE_DIR
    / f"{FINAL_BASENAME}__MANIFEST.json"
)


ROOT = Path(
    "/content/LTLC"
)

TMP_ROOT = Path(
    "/content/LTLC_RECOVERY_36OF36_TMP"
)


BASE_INTERNAL_MANIFEST = (
    "__LTLC_KAGGLE_TO_COLAB_MIGRATION_MANIFEST__.json"
)

DELTA_INTERNAL_MANIFEST = (
    "__LTLC_COLAB_FULL_DELTA_MANIFEST__.json"
)


# ==================================================================================================
# 2. EXPECTED AUTHENTICATED IDENTITIES
# ==================================================================================================

EXPECTED_BASE_ZIP_SIZE = 358222272

EXPECTED_BASE_ZIP_SHA256 = (
    "8d2798b570eefcc991308455ef5d28cea"
    "3e03c280250ccda0a8a9b356726b363"
)

EXPECTED_BASE_MANIFEST_SHA256 = (
    "72ca2939fb2cd158c2f50bfac0fd9a494"
    "7973fd5fbe9c8c95625f75ee32da604"
)


EXPECTED_FINAL_ZIP_SIZE = 63121690

EXPECTED_FINAL_ZIP_SHA256 = (
    "30ba228994f0a16e756c5ade22480b00"
    "f3e2304ee3e15bd8e00409d86dc2b7b8"
)

EXPECTED_FINAL_MANIFEST_SHA256 = (
    "5f93f457f3b1510d72b366aa8a78aa65"
    "fbf59e866dadfcefd7388d80f81c79b9"
)


EXPECTED_BASE_FILE_COUNT = 449
EXPECTED_DELTA_FILE_COUNT = 58
EXPECTED_FINAL_FILE_COUNT = 507

EXPECTED_FINAL_PROJECT_BYTES = 506445635


# ==================================================================================================
# 3. HELPERS
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):

    h = hashlib.sha256()

    with Path(path).open("rb") as f:

        for block in iter(
            lambda: f.read(chunk_size),
            b"",
        ):
            h.update(block)

    return h.hexdigest()


def safe_relative_path(
    value,
):

    p = PurePosixPath(
        str(value)
    )

    if p.is_absolute():

        raise RuntimeError(
            f"STOP: absolute project path rejected: {value}"
        )


    if ".." in p.parts:

        raise RuntimeError(
            f"STOP: path traversal rejected: {value}"
        )


    return str(p)


def extract_verified(
    zf,
    member_name,
    destination,
    expected_sha,
    expected_size=None,
    chunk_size=1024 * 1024,
):

    destination = Path(
        destination
    )

    destination.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    h = hashlib.sha256()

    byte_count = 0


    with zf.open(
        member_name,
        "r",
    ) as src:

        with destination.open(
            "wb"
        ) as dst:

            while True:

                block = src.read(
                    chunk_size
                )

                if not block:

                    break


                dst.write(
                    block
                )

                h.update(
                    block
                )

                byte_count += len(
                    block
                )


    actual_sha = h.hexdigest()


    if actual_sha != expected_sha:

        raise RuntimeError(
            "STOP: recovered file SHA mismatch:\n"
            f"{destination}"
        )


    if (
        expected_size is not None
        and
        byte_count != int(
            expected_size
        )
    ):

        raise RuntimeError(
            "STOP: recovered file size mismatch:\n"
            f"{destination}\n"
            f"expected={expected_size}\n"
            f"actual={byte_count}"
        )


    return byte_count


def resolve_base_member(
    all_members,
    relative_path,
):

    rel = safe_relative_path(
        relative_path
    )


    exact_candidates = [
        rel,
        f"LTLC/{rel}",
    ]


    for candidate in exact_candidates:

        if candidate in all_members:

            return candidate


    suffix = (
        "/"
        +
        rel
    )


    candidates = [
        name

        for name in all_members

        if name.endswith(
            suffix
        )
    ]


    if len(
        candidates
    ) != 1:

        raise RuntimeError(
            "STOP: could not uniquely resolve migration member:\n"
            f"{rel}\n"
            f"candidates={candidates[:10]}"
        )


    return candidates[
        0
    ]


# ==================================================================================================
# 4. PRE-RECOVERY BOUNDARY
# ==================================================================================================

print()
print("-" * 180)
print("[1] PRE-RECOVERY BOUNDARY")
print("-" * 180)


if ROOT.exists():

    raise RuntimeError(
        "STOP: /content/LTLC already exists. "
        "This recovery cell is only for the confirmed reset runtime."
    )


if TMP_ROOT.exists():

    shutil.rmtree(
        TMP_ROOT
    )


if not BASE_ZIP.is_file():

    raise FileNotFoundError(
        "STOP: authenticated migration ZIP missing:\n"
        f"{BASE_ZIP}"
    )


if not FINAL_DELTA_ZIP.is_file():

    raise FileNotFoundError(
        "STOP: final 36/36 persistence ZIP missing:\n"
        f"{FINAL_DELTA_ZIP}"
    )


if not FINAL_DELTA_MANIFEST.is_file():

    raise FileNotFoundError(
        "STOP: final 36/36 external manifest missing:\n"
        f"{FINAL_DELTA_MANIFEST}"
    )


print(
    "Local scientific project : MISSING as expected"
)

print(
    "28/36 migration ZIP      : FOUND"
)

print(
    "36/36 cumulative delta   : FOUND"
)

print(
    "36/36 external manifest  : FOUND"
)


# ==================================================================================================
# 5. AUTHENTICATE SOURCE ARCHIVES
# ==================================================================================================

print()
print("-" * 180)
print("[2] AUTHENTICATE RECOVERY SOURCES")
print("-" * 180)


base_size = int(
    BASE_ZIP.stat().st_size
)

final_zip_size = int(
    FINAL_DELTA_ZIP.stat().st_size
)


if base_size != EXPECTED_BASE_ZIP_SIZE:

    raise RuntimeError(
        "STOP: migration ZIP size mismatch."
    )


if final_zip_size != EXPECTED_FINAL_ZIP_SIZE:

    raise RuntimeError(
        "STOP: final 36/36 ZIP size mismatch."
    )


base_zip_sha = sha256_file(
    BASE_ZIP
)

final_zip_sha = sha256_file(
    FINAL_DELTA_ZIP
)

final_external_manifest_sha = sha256_file(
    FINAL_DELTA_MANIFEST
)


if base_zip_sha != EXPECTED_BASE_ZIP_SHA256:

    raise RuntimeError(
        "STOP: migration ZIP SHA mismatch."
    )


if final_zip_sha != EXPECTED_FINAL_ZIP_SHA256:

    raise RuntimeError(
        "STOP: final 36/36 ZIP SHA mismatch."
    )


if (
    final_external_manifest_sha
    !=
    EXPECTED_FINAL_MANIFEST_SHA256
):

    raise RuntimeError(
        "STOP: final external manifest SHA mismatch."
    )


print(
    "Migration ZIP size/SHA        : AUTHENTICATED PASS"
)

print(
    "Final 36/36 ZIP size/SHA      : AUTHENTICATED PASS"
)

print(
    "Final external manifest SHA   : AUTHENTICATED PASS"
)


# ==================================================================================================
# 6. LOAD AND AUTHENTICATE MANIFESTS
# ==================================================================================================

print()
print("-" * 180)
print("[3] AUTHENTICATE INTERNAL MANIFEST CONTRACTS")
print("-" * 180)


with zipfile.ZipFile(
    BASE_ZIP,
    "r",
) as zf:

    base_manifest_bytes = zf.read(
        BASE_INTERNAL_MANIFEST
    )


base_manifest_sha = hashlib.sha256(
    base_manifest_bytes
).hexdigest()


if (
    base_manifest_sha
    !=
    EXPECTED_BASE_MANIFEST_SHA256
):

    raise RuntimeError(
        "STOP: internal migration manifest SHA mismatch."
    )


base_manifest = json.loads(
    base_manifest_bytes.decode(
        "utf-8"
    )
)


base_records = (
    base_manifest[
        "project_snapshot"
    ][
        "records"
    ]
)


if len(
    base_records
) != EXPECTED_BASE_FILE_COUNT:

    raise RuntimeError(
        "STOP: migration base does not contain 449 records."
    )


external_manifest_bytes = (
    FINAL_DELTA_MANIFEST.read_bytes()
)


with zipfile.ZipFile(
    FINAL_DELTA_ZIP,
    "r",
) as zf:

    internal_delta_manifest_bytes = zf.read(
        DELTA_INTERNAL_MANIFEST
    )


if (
    internal_delta_manifest_bytes
    !=
    external_manifest_bytes
):

    raise RuntimeError(
        "STOP: final internal/external manifests differ."
    )


if (
    hashlib.sha256(
        internal_delta_manifest_bytes
    ).hexdigest()
    !=
    EXPECTED_FINAL_MANIFEST_SHA256
):

    raise RuntimeError(
        "STOP: internal final manifest SHA mismatch."
    )


final_manifest = json.loads(
    external_manifest_bytes.decode(
        "utf-8"
    )
)


delta_records = (
    final_manifest[
        "full_delta"
    ][
        "records"
    ]
)


if len(
    delta_records
) != EXPECTED_DELTA_FILE_COUNT:

    raise RuntimeError(
        "STOP: final cumulative delta does not contain 58 records."
    )


if (
    final_manifest[
        "scientific_progress"
    ][
        "after"
    ]
    != 36
):

    raise RuntimeError(
        "STOP: final persistence does not say 36/36."
    )


if (
    final_manifest[
        "scientific_progress"
    ].get(
        "experiment_b_complete"
    )
    is not True
):

    raise RuntimeError(
        "STOP: final persistence does not mark Experiment B complete."
    )


if (
    final_manifest[
        "safety"
    ].get(
        "official_test_status"
    )
    != "SEALED"
):

    raise RuntimeError(
        "STOP: official-test seal changed."
    )


print(
    "Migration internal manifest : PASS"
)

print(
    "Migration base records      : 449 PASS"
)

print(
    "Final internal manifest     : PASS"
)

print(
    "Final delta records         : 58 PASS"
)

print(
    "Experiment-B state          : 36/36 COMPLETE"
)

print(
    "Official test               : SEALED"
)


# ==================================================================================================
# 7. BUILD PATH CONTRACT
# ==================================================================================================

print()
print("-" * 180)
print("[4] BUILD EXACT 507-FILE RECOVERY CONTRACT")
print("-" * 180)


base_map = {}


for record in base_records:

    rel = safe_relative_path(
        record[
            "relative_path"
        ]
    )


    if rel in base_map:

        raise RuntimeError(
            f"STOP: duplicate migration path: {rel}"
        )


    base_map[
        rel
    ] = record


delta_map = {}


for record in delta_records:

    rel = safe_relative_path(
        record[
            "relative_path"
        ]
    )


    if rel in delta_map:

        raise RuntimeError(
            f"STOP: duplicate delta path: {rel}"
        )


    delta_map[
        rel
    ] = record


overlap = (
    set(
        base_map
    )
    &
    set(
        delta_map
    )
)


if overlap:

    raise RuntimeError(
        "STOP: migration base and final delta overlap."
    )


expected_paths = (
    set(
        base_map
    )
    |
    set(
        delta_map
    )
)


if len(
    expected_paths
) != EXPECTED_FINAL_FILE_COUNT:

    raise RuntimeError(
        "STOP: final recovery contract is not exactly 507 files."
    )


print(
    "Base paths         : 449 PASS"
)

print(
    "Delta paths        : 58 PASS"
)

print(
    "Base/delta overlap : 0 PASS"
)

print(
    "Final path contract: 507 PASS"
)


# ==================================================================================================
# 8. EXTRACT + VERIFY AUTHENTICATED 449-FILE BASE
# ==================================================================================================

print()
print("-" * 180)
print("[5] RECOVER AUTHENTICATED 449-FILE MIGRATION BASE")
print("-" * 180)


TMP_ROOT.mkdir(
    parents=True,
    exist_ok=False,
)


base_verified = 0
base_bytes = 0


with zipfile.ZipFile(
    BASE_ZIP,
    "r",
) as zf:

    member_names = {
        info.filename

        for info in zf.infolist()

        if not info.is_dir()
    }


    for idx, rel in enumerate(
        sorted(
            base_map
        ),
        start=1,
    ):

        record = base_map[
            rel
        ]


        member = resolve_base_member(
            member_names,
            rel,
        )


        recovered_size = extract_verified(
            zf=zf,
            member_name=member,
            destination=TMP_ROOT / rel,
            expected_sha=record[
                "sha256"
            ],
            expected_size=record.get(
                "size_bytes"
            ),
        )


        base_verified += 1
        base_bytes += recovered_size


        if (
            idx % 100 == 0
            or
            idx == EXPECTED_BASE_FILE_COUNT
        ):

            print(
                f"Base recovery verified: "
                f"{idx}/{EXPECTED_BASE_FILE_COUNT}"
            )


if base_verified != EXPECTED_BASE_FILE_COUNT:

    raise RuntimeError(
        "STOP: base recovery verification incomplete."
    )


print()
print(
    "Base extraction SHA verification : 449/449 PASS"
)


# ==================================================================================================
# 9. EXTRACT + VERIFY FINAL 58-FILE CUMULATIVE DELTA
# ==================================================================================================

print()
print("-" * 180)
print("[6] RECOVER AUTHENTICATED FINAL 58-FILE DELTA")
print("-" * 180)


delta_verified = 0
delta_bytes = 0


with zipfile.ZipFile(
    FINAL_DELTA_ZIP,
    "r",
) as zf:

    archive_names = set(
        zf.namelist()
    )


    for idx, rel in enumerate(
        sorted(
            delta_map
        ),
        start=1,
    ):

        record = delta_map[
            rel
        ]


        member = (
            "LTLC_DELTA/"
            +
            rel
        )


        if member not in archive_names:

            raise FileNotFoundError(
                "STOP: expected final-delta archive member missing:\n"
                f"{member}"
            )


        recovered_size = extract_verified(
            zf=zf,
            member_name=member,
            destination=TMP_ROOT / rel,
            expected_sha=record[
                "sha256"
            ],
            expected_size=record.get(
                "size_bytes"
            ),
        )


        delta_verified += 1
        delta_bytes += recovered_size


        if (
            idx % 20 == 0
            or
            idx == EXPECTED_DELTA_FILE_COUNT
        ):

            print(
                f"Delta recovery verified: "
                f"{idx}/{EXPECTED_DELTA_FILE_COUNT}"
            )


if delta_verified != EXPECTED_DELTA_FILE_COUNT:

    raise RuntimeError(
        "STOP: final delta verification incomplete."
    )


print()
print(
    "Delta extraction SHA verification: 58/58 PASS"
)


# ==================================================================================================
# 10. FINAL TEMPORARY SNAPSHOT AUDIT
# ==================================================================================================

print()
print("-" * 180)
print("[7] EXACT TEMPORARY 507-FILE SNAPSHOT AUDIT")
print("-" * 180)


temp_files = sorted(
    p
    for p in TMP_ROOT.rglob("*")
    if p.is_file()
)


temp_paths = {
    str(
        p.relative_to(
            TMP_ROOT
        )
    )

    for p in temp_files
}


missing = sorted(
    expected_paths
    -
    temp_paths
)

unexpected = sorted(
    temp_paths
    -
    expected_paths
)


if missing:

    raise RuntimeError(
        "STOP: recovered snapshot is missing files:\n"
        +
        "\n".join(
            missing[:20]
        )
    )


if unexpected:

    raise RuntimeError(
        "STOP: recovered snapshot contains unexpected files:\n"
        +
        "\n".join(
            unexpected[:20]
        )
    )


if len(
    temp_files
) != EXPECTED_FINAL_FILE_COUNT:

    raise RuntimeError(
        "STOP: recovered temporary project is not 507 files."
    )


total_recovered_bytes = sum(
    int(
        p.stat().st_size
    )
    for p in temp_files
)


if (
    total_recovered_bytes
    !=
    EXPECTED_FINAL_PROJECT_BYTES
):

    raise RuntimeError(
        "STOP: recovered project byte count mismatch.\n"
        f"Expected: {EXPECTED_FINAL_PROJECT_BYTES}\n"
        f"Actual  : {total_recovered_bytes}"
    )


print(
    "Recovered project files : 507 PASS"
)

print(
    "Missing files           : 0 PASS"
)

print(
    "Unexpected files        : 0 PASS"
)

print(
    "Recovered project bytes :",
    total_recovered_bytes,
    "PASS"
)

print(
    "Per-file SHA verified   : 507/507 PASS"
)


# ==================================================================================================
# 11. PROMOTE RECOVERED PROJECT
# ==================================================================================================

print()
print("-" * 180)
print("[8] PROMOTE AUTHENTICATED FINAL PROJECT")
print("-" * 180)


if ROOT.exists():

    raise RuntimeError(
        "STOP: /content/LTLC appeared before promotion."
    )


os.replace(
    TMP_ROOT,
    ROOT,
)


if not ROOT.is_dir():

    raise RuntimeError(
        "STOP: recovered project promotion failed."
    )


print(
    "Recovered project promoted to:"
)

print(
    ROOT
)


# ==================================================================================================
# 12. POST-PROMOTION PATH + HASH SPOT CONTRACT
# ==================================================================================================

print()
print("-" * 180)
print("[9] POST-PROMOTION FINAL STATE")
print("-" * 180)


final_files = sorted(
    p
    for p in ROOT.rglob("*")
    if p.is_file()
)


final_paths = {
    str(
        p.relative_to(
            ROOT
        )
    )

    for p in final_files
}


if final_paths != expected_paths:

    raise RuntimeError(
        "STOP: promoted project path contract changed."
    )


if len(
    final_files
) != 507:

    raise RuntimeError(
        "STOP: promoted project file count changed."
    )


# Verify critical final LA artifacts after promotion.
critical_paths = [
    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/pingan/"
        "la_loss__pingan__selected_tau_v1.json"
    ),

    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/"
        "la_loss__qingyun__selected_tau_v1.json"
    ),

    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed123/"
        "la_loss__qingyun__tau0p5__seed123__completion_manifest.json"
    ),

    (
        "runs/experiment_b_spatial_validation_v1/"
        "la_loss/qingyun/tau0p5/seed3407/"
        "la_loss__qingyun__tau0p5__seed3407__completion_manifest.json"
    ),
]


for rel in critical_paths:

    if rel not in delta_map:

        raise RuntimeError(
            "STOP: critical final artifact absent from authenticated delta:\n"
            f"{rel}"
        )


    if (
        sha256_file(
            ROOT / rel
        )
        !=
        delta_map[
            rel
        ][
            "sha256"
        ]
    ):

        raise RuntimeError(
            "STOP: critical final artifact changed after promotion:\n"
            f"{rel}"
        )


print(
    "Final project files       : 507 PASS"
)

print(
    "Final path-set identity   : EXACT PASS"
)

print(
    "Critical final identities : PASS"
)

print(
    "Experiment-B              : 36/36 COMPLETE"
)

print(
    "Official test             : SEALED"
)

print(
    "Training performed        : False"
)

print(
    "Model inference performed : False"
)


# ==================================================================================================
# 13. TERMINAL RECOVERY GATE
# ==================================================================================================

print()
print("=" * 180)
print("LTLC NOTEBOOK 06A / EXPERIMENT B — CORRECTION CELL 16B FINAL GATE")
print("=" * 180)

print(
    "Recovery source base                    : AUTHENTICATED 449-FILE MIGRATION"
)

print(
    "Recovery source delta                   : AUTHENTICATED FINAL 58-FILE DELTA"
)

print(
    "Base extraction verification            : 449/449 PASS"
)

print(
    "Delta extraction verification           : 58/58 PASS"
)

print(
    "Recovered final scientific project      : 507 FILES"
)

print(
    "Recovered scientific project bytes      : 506445635"
)

print(
    "Recovered path-set identity              : EXACT PASS"
)

print(
    "Recovered per-file SHA verification      : 507/507 PASS"
)

print(
    "Experiment-B                             : 36/36 COMPLETE"
)

print(
    "GPU training remaining                   : False"
)

print(
    "Training during recovery                 : False"
)

print(
    "Model inference during recovery          : False"
)

print(
    "Official test                            : SEALED"
)

print()
print(
    "RECOVERED PROJECT ROOT:"
)

print(
    ROOT
)

print()
print(
    "CORRECTION CELL 16B PASS: True"
)

print()
print(
    "STATUS: "
    "FINAL_36OF36_RUNTIME_RECOVERED__"
    "449_BASE_PLUS_58_DELTA__"
    "507_FILE_PROJECT_VERIFIED__"
    "NO_RETRAINING__"
    "OFFICIAL_TEST_SEALED__"
    "CELL16_READY"
)

print("=" * 180)

print()
print("OUTPUT CHECKPOINT — STOP")
```
