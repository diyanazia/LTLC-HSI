```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 1
# LTLC POST-HOC CALIBRATION + LTLC DEVELOPMENT
# PROJECT MOUNT + NOTEBOOK 04 FREEZE VERIFICATION
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

from google.colab import drive

from pathlib import Path
import json
import hashlib


# --------------------------------------------------------------------------------------------------
# 1. MOUNT GOOGLE DRIVE
# --------------------------------------------------------------------------------------------------

drive.mount(
    "/content/drive"
)


# --------------------------------------------------------------------------------------------------
# 2. PROJECT ROOT
# --------------------------------------------------------------------------------------------------

ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

assert ROOT.exists(), (
    f"LTLC project root not found:\n{ROOT}"
)


# --------------------------------------------------------------------------------------------------
# 3. CORE PROJECT DIRECTORIES
# --------------------------------------------------------------------------------------------------

NOTEBOOK_DIR = (
    ROOT /
    "notebooks"
)

AUDIT_DIR = (
    ROOT /
    "audit"
)

RUNS_DIR = (
    ROOT /
    "runs"
)

DATA_CACHE_DIR = (
    ROOT /
    "data_cache"
)

SPLIT_DIR = (
    ROOT /
    "split_indices"
)


for path in [
    NOTEBOOK_DIR,
    AUDIT_DIR,
    RUNS_DIR,
    DATA_CACHE_DIR,
    SPLIT_DIR,
]:

    assert path.exists(), (
        f"Required LTLC directory missing:\n{path}"
    )


# --------------------------------------------------------------------------------------------------
# 4. NOTEBOOK 04 FINAL MASTER AUDIT
# --------------------------------------------------------------------------------------------------

NOTEBOOK04_AUDIT_DIR = (
    AUDIT_DIR /
    "notebook04_longtail"
)

NOTEBOOK04_MASTER_AUDIT_PATH = (
    NOTEBOOK04_AUDIT_DIR /
    "notebook04_final_master_audit.json"
)


assert NOTEBOOK04_MASTER_AUDIT_PATH.exists(), (
    "Notebook 04 final master audit is missing."
)


# --------------------------------------------------------------------------------------------------
# 5. SHA-256 HELPER
# --------------------------------------------------------------------------------------------------

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


# --------------------------------------------------------------------------------------------------
# 6. VERIFY EXACT NOTEBOOK 04 MASTER AUDIT
# --------------------------------------------------------------------------------------------------

EXPECTED_NOTEBOOK04_MASTER_SHA = (
    "ad45760aded00dfb6c76d9eb3c757736"
    "4a67f75873cda94121dc9fe0b2465363"
)


actual_notebook04_master_sha = (
    sha256_file(
        NOTEBOOK04_MASTER_AUDIT_PATH
    )
)


assert (
    actual_notebook04_master_sha
    ==
    EXPECTED_NOTEBOOK04_MASTER_SHA
), (
    "Notebook 04 master audit SHA mismatch.\n"
    "STOP before Notebook 05."
)


with open(
    NOTEBOOK04_MASTER_AUDIT_PATH,
    "r",
    encoding="utf-8"
) as f:

    notebook04_master = json.load(
        f
    )


# --------------------------------------------------------------------------------------------------
# 7. NOTEBOOK 04 FREEZE / TEST-SEAL GATES
# --------------------------------------------------------------------------------------------------

assert (
    notebook04_master[
        "status"
    ]
    ==
    "FINAL_FROZEN"
)


assert (
    notebook04_master[
        "training_stage_complete"
    ]
    is True
)


assert (
    notebook04_master[
        "official_test_dataset_created"
    ]
    is False
)


assert (
    notebook04_master[
        "official_test_dataloader_created"
    ]
    is False
)


assert (
    notebook04_master[
        "official_test_used_for_training_or_selection"
    ]
    is False
)


assert (
    notebook04_master[
        "cryptographic_checkpoints_verified"
    ]
    ==
    45
)


assert (
    notebook04_master[
        "selected_checkpoints_verified"
    ]
    ==
    36
)


# --------------------------------------------------------------------------------------------------
# 8. NOTEBOOK 05 DIRECTORIES
# --------------------------------------------------------------------------------------------------

NOTEBOOK05_AUDIT_DIR = (
    AUDIT_DIR /
    "notebook05_posthoc_ltlc"
)

NOTEBOOK05_RUN_DIR = (
    RUNS_DIR /
    "posthoc_ltlc"
)


NOTEBOOK05_AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

NOTEBOOK05_RUN_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# --------------------------------------------------------------------------------------------------
# 9. NOTEBOOK 05 SAFETY FLAGS
# --------------------------------------------------------------------------------------------------

NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED = False

NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED = False

NOTEBOOK05_OFFICIAL_TEST_USED = False

NOTEBOOK05_TRAINING_PERFORMED = False


# --------------------------------------------------------------------------------------------------
# 10. STARTUP REPORT
# --------------------------------------------------------------------------------------------------

print("=" * 110)
print("LTLC NOTEBOOK 05 — STARTUP GATE")
print("=" * 110)

print(
    "Project root:",
    ROOT
)

print()

print(
    "Notebook 04 master audit SHA:",
    actual_notebook04_master_sha
)

print(
    "Notebook 04 status:",
    notebook04_master[
        "status"
    ]
)

print(
    "Notebook 04 checkpoints verified:",
    notebook04_master[
        "cryptographic_checkpoints_verified"
    ]
)

print(
    "Notebook 04 selected checkpoints verified:",
    notebook04_master[
        "selected_checkpoints_verified"
    ]
)

print()

print(
    "Notebook 05 runtime:",
    "CPU"
)

print(
    "Training performed:",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset created:",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created:",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "Notebook 05 audit directory:",
    NOTEBOOK05_AUDIT_DIR
)

print(
    "Notebook 05 run directory:",
    NOTEBOOK05_RUN_DIR
)

print("=" * 110)

print(
    "NOTEBOOK 05 STARTUP GATE: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 110)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 2
# FROZEN NOTEBOOK 03 CE + VALIDATION-LOGIT INVENTORY
#
# READ ONLY
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST ACCESS
# ==================================================================================================

from pathlib import Path
import re
import numpy as np


# --------------------------------------------------------------------------------------------------
# 1. NOTEBOOK 03 CE RUN ROOT
# --------------------------------------------------------------------------------------------------

CE_RUN_ROOT = (
    ROOT /
    "runs/hybridsn_ce_baseline"
)

assert CE_RUN_ROOT.exists(), (
    f"Notebook 03 CE run root missing:\n{CE_RUN_ROOT}"
)


# --------------------------------------------------------------------------------------------------
# 2. VALIDATION LOGIT DIRECTORY
#
# IMPORTANT:
# We explicitly point ONLY to validation/.
# We do NOT inspect official_test/.
# --------------------------------------------------------------------------------------------------

CE_VALIDATION_LOGIT_DIR = (
    CE_RUN_ROOT /
    "logits/validation"
)

assert CE_VALIDATION_LOGIT_DIR.exists(), (
    f"Validation-logit directory missing:\n{CE_VALIDATION_LOGIT_DIR}"
)


# --------------------------------------------------------------------------------------------------
# 3. EXPECTED DATASETS / SEEDS
# --------------------------------------------------------------------------------------------------

DATASET_ORDER = [
    "Pingan",
    "Qingyun",
    "Tangdaowan",
]

EXPECTED_SEEDS = [
    42,
    123,
    3407,
]


# --------------------------------------------------------------------------------------------------
# 4. INVENTORY VALIDATION OUTPUT FILES ONLY
# --------------------------------------------------------------------------------------------------

validation_files = sorted(
    CE_VALIDATION_LOGIT_DIR.glob(
        "*_validation_outputs.npz"
    )
)


print("=" * 120)
print("NOTEBOOK 05 — FROZEN CE VALIDATION ARTIFACT INVENTORY")
print("=" * 120)

print(
    "CE run root:",
    CE_RUN_ROOT
)

print(
    "Validation-logit directory:",
    CE_VALIDATION_LOGIT_DIR
)

print()

print(
    "Validation output files found:",
    len(validation_files)
)

assert len(validation_files) == 9, (
    f"Expected 9 validation-output files, found {len(validation_files)}."
)


# --------------------------------------------------------------------------------------------------
# 5. PARSE + VERIFY ALL 9 VALIDATION FILES
# --------------------------------------------------------------------------------------------------

VALIDATION_OUTPUTS = {}

inventory_rows = []


pattern = re.compile(
    r"^(pingan|qingyun|tangdaowan)_seed(42|123|3407)_validation_outputs\.npz$"
)


for path in validation_files:

    match = pattern.match(
        path.name
    )

    assert match is not None, (
        f"Unexpected validation filename:\n{path.name}"
    )


    dataset_slug = match.group(1)

    seed = int(
        match.group(2)
    )


    dataset = {
        "pingan": "Pingan",
        "qingyun": "Qingyun",
        "tangdaowan": "Tangdaowan",
    }[
        dataset_slug
    ]


    assert seed in EXPECTED_SEEDS


    # ----------------------------------------------------------------------------------------------
    # Read ONLY this validation NPZ
    # ----------------------------------------------------------------------------------------------

    with np.load(
        path,
        allow_pickle=False
    ) as data:

        keys = list(
            data.files
        )


        required_keys = [
            "logits",
            "predictions",
            "labels",
            "indices",
        ]


        for key in required_keys:

            assert key in keys, (
                f"{path.name}: missing key '{key}'"
            )


        logits = data[
            "logits"
        ]

        predictions = data[
            "predictions"
        ]

        labels = data[
            "labels"
        ]

        indices = data[
            "indices"
        ]


        # ------------------------------------------------------------------------------------------
        # BASIC CONSISTENCY
        # ------------------------------------------------------------------------------------------

        n = int(
            logits.shape[0]
        )


        assert predictions.shape[0] == n
        assert labels.shape[0] == n
        assert indices.shape[0] == n

        assert logits.ndim == 2
        assert predictions.ndim == 1
        assert labels.ndim == 1

        assert np.isfinite(
            logits
        ).all(), (
            f"{path.name}: non-finite logits detected."
        )


        inventory_rows.append({

            "dataset":
                dataset,

            "seed":
                seed,

            "filename":
                path.name,

            "samples":
                n,

            "classes":
                int(
                    logits.shape[1]
                ),

            "logits_shape":
                tuple(
                    logits.shape
                ),

            "labels_shape":
                tuple(
                    labels.shape
                ),

            "indices_shape":
                tuple(
                    indices.shape
                ),

            "sha256":
                sha256_file(
                    path
                ),
        })


    VALIDATION_OUTPUTS[
        (
            dataset,
            seed
        )
    ] = path


# --------------------------------------------------------------------------------------------------
# 6. COMPLETE DATASET × SEED COVERAGE
# --------------------------------------------------------------------------------------------------

expected_pairs = {

    (
        dataset,
        seed
    )

    for dataset in DATASET_ORDER

    for seed in EXPECTED_SEEDS
}


actual_pairs = set(
    VALIDATION_OUTPUTS.keys()
)


assert (
    actual_pairs
    ==
    expected_pairs
), (
    "Validation dataset/seed coverage mismatch."
)


# --------------------------------------------------------------------------------------------------
# 7. PRINT VALIDATION INVENTORY
# --------------------------------------------------------------------------------------------------

for dataset in DATASET_ORDER:

    print()
    print("-" * 120)
    print(dataset.upper())
    print("-" * 120)


    for seed in EXPECTED_SEEDS:

        row = next(

            r

            for r in inventory_rows

            if (
                r[
                    "dataset"
                ]
                ==
                dataset
                and
                r[
                    "seed"
                ]
                ==
                seed
            )
        )


        print(
            f"seed={seed:4d} | "
            f"samples={row['samples']:6d} | "
            f"classes={row['classes']:2d} | "
            f"logits={row['logits_shape']} | "
            f"SHA={row['sha256'][:12]}..."
        )


# --------------------------------------------------------------------------------------------------
# 8. VERIFY EXPECTED CLASS DIMENSIONS
#
# Tangdaowan public train/test-mask protocol uses K=16.
# --------------------------------------------------------------------------------------------------

EXPECTED_CLASS_COUNT = {

    "Pingan":
        10,

    "Qingyun":
        6,

    "Tangdaowan":
        16,
}


for row in inventory_rows:

    assert (
        row[
            "classes"
        ]
        ==
        EXPECTED_CLASS_COUNT[
            row[
                "dataset"
            ]
        ]
    ), (
        f"{row['dataset']} seed {row['seed']}: "
        "unexpected class dimension."
    )


# --------------------------------------------------------------------------------------------------
# 9. INVENTORY FROZEN FINAL CE CHECKPOINTS
#
# IMPORTANT:
# Only the canonical final checkpoint directory is used.
#
# The historical:
#   runs/hybridsn_ce_baseline/_pilot/
#
# directory is intentionally excluded.
# --------------------------------------------------------------------------------------------------

CE_CHECKPOINT_DIR = (
    CE_RUN_ROOT /
    "checkpoints"
)

assert CE_CHECKPOINT_DIR.exists(), (
    f"Final CE checkpoint directory missing:\n{CE_CHECKPOINT_DIR}"
)


checkpoint_candidates = sorted(
    CE_CHECKPOINT_DIR.glob(
        "*_best_valOA.pt"
    )
)


print()
print("=" * 120)
print("FROZEN FINAL CE CHECKPOINT INVENTORY")
print("=" * 120)

print(
    "Final checkpoint directory:",
    CE_CHECKPOINT_DIR
)

print()

print(
    "Final CE checkpoints found:",
    len(checkpoint_candidates)
)


for path in checkpoint_candidates:

    print(
        path.relative_to(
            ROOT
        )
    )


assert len(checkpoint_candidates) == 9, (
    "Expected exactly 9 FINAL CE checkpoints "
    f"(3 datasets × 3 seeds), found {len(checkpoint_candidates)}."
)


# --------------------------------------------------------------------------------------------------
# 9A. VERIFY EXACT DATASET × SEED CHECKPOINT COVERAGE
# --------------------------------------------------------------------------------------------------

checkpoint_pattern = re.compile(
    r"^(pingan|qingyun|tangdaowan)_seed(42|123|3407)_best_valOA\.pt$"
)


CE_CHECKPOINTS = {}


for path in checkpoint_candidates:

    match = checkpoint_pattern.match(
        path.name
    )

    assert match is not None, (
        f"Unexpected final checkpoint filename:\n{path.name}"
    )


    dataset_slug = match.group(1)

    seed = int(
        match.group(2)
    )


    dataset = {
        "pingan": "Pingan",
        "qingyun": "Qingyun",
        "tangdaowan": "Tangdaowan",
    }[
        dataset_slug
    ]


    key = (
        dataset,
        seed
    )


    assert key not in CE_CHECKPOINTS, (
        f"Duplicate final checkpoint for {key}"
    )


    CE_CHECKPOINTS[
        key
    ] = path


assert (
    set(
        CE_CHECKPOINTS.keys()
    )
    ==
    expected_pairs
), (
    "Final CE checkpoint dataset/seed coverage mismatch."
)


# --------------------------------------------------------------------------------------------------
# 9B. REPORT HISTORICAL PILOT CHECKPOINTS SEPARATELY
#
# READ ONLY.
# They are NOT part of the frozen final 9-checkpoint CE baseline.
# --------------------------------------------------------------------------------------------------

PILOT_DIR = (
    CE_RUN_ROOT /
    "_pilot"
)


pilot_checkpoints = []


if PILOT_DIR.exists():

    pilot_checkpoints = sorted(
        PILOT_DIR.glob(
            "*.pt"
        )
    )


print()
print(
    "Historical pilot checkpoints excluded:",
    len(pilot_checkpoints)
)


for path in pilot_checkpoints:

    print(
        "EXCLUDED |",
        path.relative_to(
            ROOT
        )
    )


# --------------------------------------------------------------------------------------------------
# 10. NOTEBOOK 05 SAFETY GATE
# --------------------------------------------------------------------------------------------------

assert (
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
    is False
)

assert (
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
    is False
)

assert (
    NOTEBOOK05_OFFICIAL_TEST_USED
    is False
)

assert (
    NOTEBOOK05_TRAINING_PERFORMED
    is False
)


NOTEBOOK05_VALIDATION_INVENTORY_PASS = True


print()
print("=" * 120)

print(
    "Validation logits verified : 9/9 PASS"
)

print(
    "Frozen FINAL CE checkpoints: 9/9 PASS"
)

print(
    "Historical pilot excluded  :",
    len(pilot_checkpoints)
)

print(
    "Training performed         :",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset      :",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader   :",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used         :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 VALIDATION-INVENTORY GATE: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 120)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 3
# CE VALIDATION CACHE INTEGRITY + LABEL-CONVENTION AUDIT
#
# READ ONLY
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import numpy as np


# --------------------------------------------------------------------------------------------------
# 1. EXPECTED VALIDATION SIZES / CLASS COUNTS
# --------------------------------------------------------------------------------------------------

EXPECTED_VALIDATION_SIZE = {

    "Pingan":
        34230,

    "Qingyun":
        28648,

    "Tangdaowan":
        10612,
}


EXPECTED_CLASS_COUNT = {

    "Pingan":
        10,

    "Qingyun":
        6,

    "Tangdaowan":
        16,
}


# --------------------------------------------------------------------------------------------------
# 2. LOAD + AUDIT EACH DATASET/SEED VALIDATION CACHE
# --------------------------------------------------------------------------------------------------

VALIDATION_CACHE = {}

VALIDATION_CLASS_COUNTS = {}

VALIDATION_LABEL_RANGE = {}


print("=" * 130)
print("NOTEBOOK 05 — VALIDATION CACHE INTEGRITY + LABEL-CONVENTION AUDIT")
print("=" * 130)


for dataset in DATASET_ORDER:

    print()
    print("-" * 130)
    print(dataset.upper())
    print("-" * 130)


    reference_indices = None
    reference_labels = None
    reference_class_counts = None


    for seed in EXPECTED_SEEDS:

        path = (
            VALIDATION_OUTPUTS[
                (
                    dataset,
                    seed
                )
            ]
        )


        with np.load(
            path,
            allow_pickle=False
        ) as data:

            logits = np.asarray(
                data[
                    "logits"
                ]
            )

            predictions = np.asarray(
                data[
                    "predictions"
                ]
            )

            labels = np.asarray(
                data[
                    "labels"
                ]
            )

            indices = np.asarray(
                data[
                    "indices"
                ]
            )


        # ------------------------------------------------------------------------------------------
        # SHAPE
        # ------------------------------------------------------------------------------------------

        n = int(
            logits.shape[
                0
            ]
        )

        k = int(
            logits.shape[
                1
            ]
        )


        assert (
            n
            ==
            EXPECTED_VALIDATION_SIZE[
                dataset
            ]
        ), (
            f"{dataset} seed {seed}: "
            f"unexpected validation size {n}"
        )


        assert (
            k
            ==
            EXPECTED_CLASS_COUNT[
                dataset
            ]
        ), (
            f"{dataset} seed {seed}: "
            f"unexpected class count {k}"
        )


        assert (
            predictions.shape
            ==
            (
                n,
            )
        )


        assert (
            labels.shape
            ==
            (
                n,
            )
        )


        assert (
            indices.shape
            ==
            (
                n,
            )
        )


        # ------------------------------------------------------------------------------------------
        # FINITE LOGITS
        # ------------------------------------------------------------------------------------------

        assert np.isfinite(
            logits
        ).all(), (
            f"{dataset} seed {seed}: "
            "non-finite logits detected."
        )


        # ------------------------------------------------------------------------------------------
        # LABEL / PREDICTION INTEGER CONVENTION
        # ------------------------------------------------------------------------------------------

        assert np.issubdtype(
            labels.dtype,
            np.integer
        ), (
            f"{dataset} seed {seed}: "
            f"labels dtype is not integer: {labels.dtype}"
        )


        assert np.issubdtype(
            predictions.dtype,
            np.integer
        ), (
            f"{dataset} seed {seed}: "
            f"predictions dtype is not integer: {predictions.dtype}"
        )


        assert np.issubdtype(
            indices.dtype,
            np.integer
        ), (
            f"{dataset} seed {seed}: "
            f"indices dtype is not integer: {indices.dtype}"
        )


        label_min = int(
            labels.min()
        )

        label_max = int(
            labels.max()
        )


        pred_min = int(
            predictions.min()
        )

        pred_max = int(
            predictions.max()
        )


        # ------------------------------------------------------------------------------------------
        # FROZEN CE INTERNAL LABEL CONVENTION MUST BE 0 ... K-1
        # ------------------------------------------------------------------------------------------

        assert (
            label_min
            ==
            0
        ), (
            f"{dataset} seed {seed}: "
            f"expected 0-based labels, min={label_min}"
        )


        assert (
            label_max
            ==
            k - 1
        ), (
            f"{dataset} seed {seed}: "
            f"expected max label {k - 1}, got {label_max}"
        )


        assert (
            pred_min
            >=
            0
        )


        assert (
            pred_max
            <
            k
        )


        # ------------------------------------------------------------------------------------------
        # SAVED PREDICTIONS MUST EXACTLY MATCH ARGMAX(LOGITS)
        # ------------------------------------------------------------------------------------------

        recomputed_predictions = (
            np.argmax(
                logits,
                axis=1
            )
            .astype(
                predictions.dtype,
                copy=False
            )
        )


        assert np.array_equal(
            predictions,
            recomputed_predictions
        ), (
            f"{dataset} seed {seed}: "
            "saved predictions do not equal argmax(logits)."
        )


        # ------------------------------------------------------------------------------------------
        # VALIDATION INDICES MUST BE UNIQUE
        # ------------------------------------------------------------------------------------------

        unique_index_count = int(
            np.unique(
                indices
            ).size
        )


        assert (
            unique_index_count
            ==
            n
        ), (
            f"{dataset} seed {seed}: "
            "duplicate validation indices detected."
        )


        # ------------------------------------------------------------------------------------------
        # CLASS SUPPORT
        # ------------------------------------------------------------------------------------------

        class_counts = (
            np.bincount(
                labels,
                minlength=k
            )
            .astype(
                np.int64
            )
        )


        assert (
            class_counts.shape[
                0
            ]
            ==
            k
        )


        assert (
            class_counts
            >
            0
        ).all(), (
            f"{dataset} seed {seed}: "
            "one or more validation classes have zero support."
        )


        assert (
            int(
                class_counts.sum()
            )
            ==
            n
        )


        # ------------------------------------------------------------------------------------------
        # CROSS-SEED ALIGNMENT
        #
        # All three CE models must have been evaluated on exactly
        # the same frozen validation samples in the same order.
        # ------------------------------------------------------------------------------------------

        if (
            reference_indices
            is None
        ):

            reference_indices = (
                indices.copy()
            )

            reference_labels = (
                labels.copy()
            )

            reference_class_counts = (
                class_counts.copy()
            )


        else:

            assert np.array_equal(
                indices,
                reference_indices
            ), (
                f"{dataset} seed {seed}: "
                "validation indices/order differ across seeds."
            )


            assert np.array_equal(
                labels,
                reference_labels
            ), (
                f"{dataset} seed {seed}: "
                "validation labels differ across seeds."
            )


            assert np.array_equal(
                class_counts,
                reference_class_counts
            )


        # ------------------------------------------------------------------------------------------
        # STORE IN MEMORY FOR NOTEBOOK 05
        # ------------------------------------------------------------------------------------------

        VALIDATION_CACHE[
            (
                dataset,
                seed
            )
        ] = {

            "path":
                path,

            "logits":
                logits,

            "predictions":
                predictions,

            "labels":
                labels,

            "indices":
                indices,

            "sha256":
                sha256_file(
                    path
                ),
        }


        print(
            f"seed={seed:4d} | "
            f"N={n:6d} | "
            f"K={k:2d} | "
            f"labels={label_min}..{label_max} | "
            f"unique_indices={unique_index_count:6d} | "
            "argmax=PASS"
        )


    # ----------------------------------------------------------------------------------------------
    # STORE CANONICAL DATASET-LEVEL VALIDATION LABEL INFORMATION
    # ----------------------------------------------------------------------------------------------

    VALIDATION_CLASS_COUNTS[
        dataset
    ] = (
        reference_class_counts
    )


    VALIDATION_LABEL_RANGE[
        dataset
    ] = (
        int(
            reference_labels.min()
        ),
        int(
            reference_labels.max()
        )
    )


    print()
    print(
        "Class support counts:"
    )


    for class_index, count in enumerate(
        reference_class_counts
    ):

        print(
            f"  internal class {class_index:2d} "
            f"(public label {class_index + 1:2d}) "
            f": {int(count):6d}"
        )


# --------------------------------------------------------------------------------------------------
# 3. COMPLETE CACHE COVERAGE
# --------------------------------------------------------------------------------------------------

assert (
    len(
        VALIDATION_CACHE
    )
    ==
    9
)


# --------------------------------------------------------------------------------------------------
# 4. DATASET-LEVEL LABEL CONVENTION GATE
# --------------------------------------------------------------------------------------------------

for dataset in DATASET_ORDER:

    expected_k = (
        EXPECTED_CLASS_COUNT[
            dataset
        ]
    )


    assert (
        VALIDATION_LABEL_RANGE[
            dataset
        ]
        ==
        (
            0,
            expected_k - 1
        )
    )


# --------------------------------------------------------------------------------------------------
# 5. OFFICIAL-TEST SAFETY GATE
# --------------------------------------------------------------------------------------------------

assert (
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
    is False
)

assert (
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
    is False
)

assert (
    NOTEBOOK05_OFFICIAL_TEST_USED
    is False
)

assert (
    NOTEBOOK05_TRAINING_PERFORMED
    is False
)


NOTEBOOK05_VALIDATION_CACHE_INTEGRITY_PASS = True


# --------------------------------------------------------------------------------------------------
# 6. FINAL REPORT
# --------------------------------------------------------------------------------------------------

print()
print("=" * 130)

print(
    "Validation caches verified          : 9/9 PASS"
)

print(
    "Cross-seed validation alignment     : 3/3 datasets PASS"
)

print(
    "Saved prediction == argmax(logits)  : 9/9 PASS"
)

print(
    "Validation indices unique           : 9/9 PASS"
)

print(
    "Internal label convention           : 0-based PASS"
)

print(
    "Public reporting convention         : internal k -> public label k+1"
)

print(
    "Training performed                  :",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset               :",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader            :",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used                  :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 VALIDATION-CACHE INTEGRITY GATE: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 130)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 4
# FROZEN TRAINING PRIORS + RARITY + FREQUENCY-GROUP AUDIT
#
# SOURCES:
#   audit/<dataset>_ltlc_class_statistics.csv
#   audit/<dataset>_fixed_70_30_counts.csv
#
# READS ACTUAL 70% TRAIN COUNTS ONLY.
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
import math
import numpy as np
import pandas as pd


# --------------------------------------------------------------------------------------------------
# 1. LOCKED CONSTANTS
# --------------------------------------------------------------------------------------------------

RARITY_EPS = 1e-12


EXPECTED_TRAIN_TOTAL = {

    "Pingan":
        79869,

    "Qingyun":
        66844,

    "Tangdaowan":
        24760,
}


EXPECTED_CLASS_COUNT = {

    "Pingan":
        10,

    "Qingyun":
        6,

    "Tangdaowan":
        16,
}


# --------------------------------------------------------------------------------------------------
# 2. HELPERS
# --------------------------------------------------------------------------------------------------

def compute_rarity(
    class_counts,
    eps=1e-12
):

    counts = np.asarray(
        class_counts,
        dtype=np.float64
    )

    assert (
        counts.ndim
        ==
        1
    )

    assert (
        counts
        >
        0
    ).all()


    n_max = float(
        counts.max()
    )

    n_min = float(
        counts.min()
    )


    if math.isclose(
        n_max,
        n_min,
        rel_tol=0.0,
        abs_tol=0.0
    ):

        return np.zeros_like(
            counts,
            dtype=np.float64
        )


    numerator = (
        np.log(
            n_max + eps
        )
        -
        np.log(
            counts + eps
        )
    )


    denominator = (
        np.log(
            n_max + eps
        )
        -
        np.log(
            n_min + eps
        )
    )


    return (
        numerator
        /
        denominator
    )


def compute_frequency_groups(
    class_ids,
    class_counts
):
    """
    Locked grouping rule:

        q = floor(K / 3)

        Tail:
            lowest-frequency q classes

        Head:
            highest-frequency q classes

        Medium:
            all remaining classes

    Deterministic tie handling:
        sort by (train_count, class_id).
    """

    class_ids = np.asarray(
        class_ids,
        dtype=np.int64
    )

    class_counts = np.asarray(
        class_counts,
        dtype=np.int64
    )


    k = len(
        class_ids
    )

    q = int(
        np.floor(
            k / 3
        )
    )


    ordering = sorted(

        range(
            k
        ),

        key=lambda i: (
            int(
                class_counts[
                    i
                ]
            ),
            int(
                class_ids[
                    i
                ]
            )
        )
    )


    tail_positions = set(
        ordering[
            :q
        ]
    )


    head_positions = set(
        ordering[
            -q:
        ]
    )


    groups = []


    for i in range(
        k
    ):

        if (
            i
            in
            tail_positions
        ):

            groups.append(
                "Tail"
            )


        elif (
            i
            in
            head_positions
        ):

            groups.append(
                "Head"
            )


        else:

            groups.append(
                "Medium"
            )


    return (
        np.asarray(
            groups,
            dtype=object
        ),
        q
    )


# --------------------------------------------------------------------------------------------------
# 3. LOAD + VERIFY FROZEN CLASS STATISTICS
# --------------------------------------------------------------------------------------------------

NOTEBOOK05_CLASS_STATS = {}

CLASS_PRIORS = {}

CLASS_RARITY = {}

CLASS_TRAIN_COUNTS = {}

CLASS_FREQUENCY_GROUPS = {}

CLASS_STATS_SOURCE_SHA = {}

FIXED_COUNT_SOURCE_SHA = {}


print("=" * 135)
print("NOTEBOOK 05 — FROZEN TRAINING PRIOR + RARITY AUDIT")
print("=" * 135)


for dataset in DATASET_ORDER:

    slug = (
        dataset.lower()
    )


    class_stats_path = (
        AUDIT_DIR /
        f"{slug}_ltlc_class_statistics.csv"
    )


    fixed_counts_path = (
        AUDIT_DIR /
        f"{slug}_fixed_70_30_counts.csv"
    )


    assert class_stats_path.exists(), (
        f"Missing class-statistics file:\n"
        f"{class_stats_path}"
    )


    assert fixed_counts_path.exists(), (
        f"Missing fixed-count file:\n"
        f"{fixed_counts_path}"
    )


    class_df = pd.read_csv(
        class_stats_path
    )


    fixed_df = pd.read_csv(
        fixed_counts_path
    )


    # ----------------------------------------------------------------------------------------------
    # 3A. REQUIRED SCHEMA
    # ----------------------------------------------------------------------------------------------

    required_class_columns = [
        "class_id",
        "train_count",
        "prior",
        "rarity",
        "frequency_group",
    ]


    required_fixed_columns = [
        "class_id",
        "official_train",
        "actual_train_70pct",
        "validation_30pct",
        "reconstructed_total",
    ]


    assert (
        list(
            class_df.columns
        )
        ==
        required_class_columns
    ), (
        f"{dataset}: unexpected class-statistics schema."
    )


    assert all(

        column
        in fixed_df.columns

        for column
        in required_fixed_columns
    ), (
        f"{dataset}: unexpected fixed-count schema."
    )


    k = (
        EXPECTED_CLASS_COUNT[
            dataset
        ]
    )


    assert (
        len(
            class_df
        )
        ==
        k
    )


    assert (
        len(
            fixed_df
        )
        ==
        k
    )


    # ----------------------------------------------------------------------------------------------
    # 3B. CLASS IDS MUST BE PUBLIC LABELS 1 ... K
    # ----------------------------------------------------------------------------------------------

    expected_class_ids = np.arange(
        1,
        k + 1,
        dtype=np.int64
    )


    class_ids = (
        class_df[
            "class_id"
        ]
        .to_numpy(
            dtype=np.int64
        )
    )


    fixed_class_ids = (
        fixed_df[
            "class_id"
        ]
        .to_numpy(
            dtype=np.int64
        )
    )


    assert np.array_equal(
        class_ids,
        expected_class_ids
    )


    assert np.array_equal(
        fixed_class_ids,
        expected_class_ids
    )


    # ----------------------------------------------------------------------------------------------
    # 3C. TRAIN COUNTS MUST MATCH THE ACTUAL 70% TRAINING SUBSET
    # ----------------------------------------------------------------------------------------------

    train_counts = (
        class_df[
            "train_count"
        ]
        .to_numpy(
            dtype=np.int64
        )
    )


    fixed_train_counts = (
        fixed_df[
            "actual_train_70pct"
        ]
        .to_numpy(
            dtype=np.int64
        )
    )


    assert np.array_equal(
        train_counts,
        fixed_train_counts
    ), (
        f"{dataset}: LTLC train_count does not match "
        "actual_train_70pct."
    )


    assert (
        int(
            train_counts.sum()
        )
        ==
        EXPECTED_TRAIN_TOTAL[
            dataset
        ]
    ), (
        f"{dataset}: unexpected 70% training total."
    )


    assert (
        train_counts
        >
        0
    ).all()


    # ----------------------------------------------------------------------------------------------
    # 3D. VALIDATION COUNTS MUST MATCH CELL 3
    #
    # This confirms that:
    #
    # actual 70% train + frozen 30% validation
    # =
    # original official-train mask
    # ----------------------------------------------------------------------------------------------

    fixed_val_counts = (
        fixed_df[
            "validation_30pct"
        ]
        .to_numpy(
            dtype=np.int64
        )
    )


    assert np.array_equal(
        fixed_val_counts,
        VALIDATION_CLASS_COUNTS[
            dataset
        ]
    ), (
        f"{dataset}: frozen validation class counts "
        "do not match Cell 3 validation caches."
    )


    assert np.array_equal(

        (
            fixed_train_counts
            +
            fixed_val_counts
        ),

        fixed_df[
            "official_train"
        ].to_numpy(
            dtype=np.int64
        )

    ), (
        f"{dataset}: train + validation does not "
        "reconstruct official training mask."
    )


    assert np.array_equal(

        fixed_df[
            "reconstructed_total"
        ].to_numpy(
            dtype=np.int64
        ),

        fixed_df[
            "official_train"
        ].to_numpy(
            dtype=np.int64
        )

    )


    # ----------------------------------------------------------------------------------------------
    # 3E. RECOMPUTE EMPIRICAL TRAINING PRIOR
    #
    # pi_k = n_k / sum_j n_j
    # ----------------------------------------------------------------------------------------------

    recomputed_prior = (
        train_counts.astype(
            np.float64
        )
        /
        float(
            train_counts.sum()
        )
    )


    stored_prior = (
        class_df[
            "prior"
        ]
        .to_numpy(
            dtype=np.float64
        )
    )


    assert np.allclose(
        stored_prior,
        recomputed_prior,
        rtol=0.0,
        atol=1e-6
    ), (
        f"{dataset}: stored training prior mismatch."
    )


    assert np.isclose(
        recomputed_prior.sum(),
        1.0,
        rtol=0.0,
        atol=1e-12
    )


    # ----------------------------------------------------------------------------------------------
    # 3F. RECOMPUTE LOCKED LOG-FREQUENCY RARITY
    # ----------------------------------------------------------------------------------------------

    recomputed_rarity = compute_rarity(
        train_counts,
        eps=RARITY_EPS
    )


    stored_rarity = (
        class_df[
            "rarity"
        ]
        .to_numpy(
            dtype=np.float64
        )
    )


    assert np.allclose(
        stored_rarity,
        recomputed_rarity,
        rtol=0.0,
        atol=1e-6
    ), (
        f"{dataset}: stored rarity mismatch."
    )


    assert (
        recomputed_rarity
        >=
        -1e-12
    ).all()


    assert (
        recomputed_rarity
        <=
        1.0 + 1e-12
    ).all()


    # ----------------------------------------------------------------------------------------------
    # 3G. RECOMPUTE LOCKED HEAD / MEDIUM / TAIL GROUPS
    # ----------------------------------------------------------------------------------------------

    (
        recomputed_groups,
        q
    ) = compute_frequency_groups(
        class_ids,
        train_counts
    )


    stored_groups = (
        class_df[
            "frequency_group"
        ]
        .astype(
            str
        )
        .to_numpy()
    )


    assert np.array_equal(
        stored_groups,
        recomputed_groups
    ), (
        f"{dataset}: stored Head/Medium/Tail groups "
        "do not match locked training-count rule."
    )


    tail_count = int(
        np.sum(
            recomputed_groups
            ==
            "Tail"
        )
    )


    medium_count = int(
        np.sum(
            recomputed_groups
            ==
            "Medium"
        )
    )


    head_count = int(
        np.sum(
            recomputed_groups
            ==
            "Head"
        )
    )


    assert (
        tail_count
        ==
        q
    )


    assert (
        head_count
        ==
        q
    )


    assert (
        tail_count
        +
        medium_count
        +
        head_count
        ==
        k
    )


    # ----------------------------------------------------------------------------------------------
    # 3H. STORE CANONICAL NOTEBOOK 05 CLASS STATISTICS
    #
    # IMPORTANT:
    # model internal class 0 corresponds to public class_id 1.
    # ----------------------------------------------------------------------------------------------

    NOTEBOOK05_CLASS_STATS[
        dataset
    ] = {

        "class_ids_public":
            class_ids.copy(),

        "class_indices_internal":
            np.arange(
                k,
                dtype=np.int64
            ),

        "train_counts":
            train_counts.copy(),

        "prior":
            recomputed_prior.copy(),

        "rarity":
            recomputed_rarity.copy(),

        "frequency_group":
            recomputed_groups.copy(),

        "q":
            q,
    }


    CLASS_TRAIN_COUNTS[
        dataset
    ] = (
        train_counts.copy()
    )


    CLASS_PRIORS[
        dataset
    ] = (
        recomputed_prior.copy()
    )


    CLASS_RARITY[
        dataset
    ] = (
        recomputed_rarity.copy()
    )


    CLASS_FREQUENCY_GROUPS[
        dataset
    ] = (
        recomputed_groups.copy()
    )


    CLASS_STATS_SOURCE_SHA[
        dataset
    ] = (
        sha256_file(
            class_stats_path
        )
    )


    FIXED_COUNT_SOURCE_SHA[
        dataset
    ] = (
        sha256_file(
            fixed_counts_path
        )
    )


    # ----------------------------------------------------------------------------------------------
    # 3I. REPORT
    # ----------------------------------------------------------------------------------------------

    print()
    print("-" * 135)

    print(
        dataset.upper()
    )

    print("-" * 135)


    print(
        f"K={k} | "
        f"train_total={int(train_counts.sum())} | "
        f"n_min={int(train_counts.min())} | "
        f"n_max={int(train_counts.max())} | "
        f"IR={float(train_counts.max()/train_counts.min()):.6f}"
    )


    print(
        f"Groups | "
        f"Tail={tail_count} | "
        f"Medium={medium_count} | "
        f"Head={head_count}"
    )


    print()


    for i in range(
        k
    ):

        print(
            f"internal={i:2d} | "
            f"public={int(class_ids[i]):2d} | "
            f"n={int(train_counts[i]):6d} | "
            f"prior={recomputed_prior[i]:.8f} | "
            f"rarity={recomputed_rarity[i]:.6f} | "
            f"group={recomputed_groups[i]}"
        )


# --------------------------------------------------------------------------------------------------
# 4. EXPECTED IMBALANCE-RATIO SANITY CHECK
# --------------------------------------------------------------------------------------------------

EXPECTED_IR = {

    "Pingan":
        70.99649122807018,

    "Qingyun":
        28.46783625730994,

    "Tangdaowan":
        191.77777777777777,
}


for dataset in DATASET_ORDER:

    counts = (
        CLASS_TRAIN_COUNTS[
            dataset
        ]
    )


    actual_ir = float(
        counts.max()
        /
        counts.min()
    )


    assert math.isclose(
        actual_ir,
        EXPECTED_IR[
            dataset
        ],
        rel_tol=0.0,
        abs_tol=1e-10
    )


# --------------------------------------------------------------------------------------------------
# 5. CREATE NOTEBOOK 05 PRIOR / RARITY FREEZE MANIFEST
#
# This is a NEW Notebook 05 audit artifact.
# Existing source CSV files are NOT modified.
# --------------------------------------------------------------------------------------------------

NOTEBOOK05_PRIOR_MANIFEST_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "notebook05_training_prior_rarity_manifest.json"
)


manifest_object = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "status":
        "FROZEN_FOR_VALIDATION_TUNING",

    "source":
        "actual_70pct_training_subset",

    "rarity_epsilon":
        RARITY_EPS,

    "prior_definition":
        "pi_k = n_k / sum_j(n_j)",

    "rarity_definition":
        (
            "(log(n_max+eps)-log(n_k+eps)) / "
            "(log(n_max+eps)-log(n_min+eps))"
        ),

    "frequency_group_definition":
        (
            "sort classes by training count; "
            "q=floor(K/3); lowest q=Tail; "
            "highest q=Head; remainder=Medium"
        ),

    "datasets":
        {},

    "official_test_used":
        False,

    "training_performed":
        False,
}


for dataset in DATASET_ORDER:

    stats = (
        NOTEBOOK05_CLASS_STATS[
            dataset
        ]
    )


    manifest_object[
        "datasets"
    ][
        dataset
    ] = {

        "class_count":
            int(
                len(
                    stats[
                        "train_counts"
                    ]
                )
            ),

        "train_total":
            int(
                stats[
                    "train_counts"
                ].sum()
            ),

        "class_ids_public":
            [
                int(x)

                for x
                in stats[
                    "class_ids_public"
                ]
            ],

        "class_indices_internal":
            [
                int(x)

                for x
                in stats[
                    "class_indices_internal"
                ]
            ],

        "train_counts":
            [
                int(x)

                for x
                in stats[
                    "train_counts"
                ]
            ],

        "prior":
            [
                float(x)

                for x
                in stats[
                    "prior"
                ]
            ],

        "rarity":
            [
                float(x)

                for x
                in stats[
                    "rarity"
                ]
            ],

        "frequency_group":
            [
                str(x)

                for x
                in stats[
                    "frequency_group"
                ]
            ],

        "q":
            int(
                stats[
                    "q"
                ]
            ),

        "class_statistics_source_sha256":
            CLASS_STATS_SOURCE_SHA[
                dataset
            ],

        "fixed_70_30_counts_source_sha256":
            FIXED_COUNT_SOURCE_SHA[
                dataset
            ],
    }


# --------------------------------------------------------------------------------------------------
# 6. RESTART-SAFE WRITE
#
# Existing identical artifact -> keep.
# Existing different artifact -> STOP.
# --------------------------------------------------------------------------------------------------

if NOTEBOOK05_PRIOR_MANIFEST_PATH.exists():

    with open(
        NOTEBOOK05_PRIOR_MANIFEST_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing_manifest = (
            json.load(
                f
            )
        )


    assert (
        existing_manifest
        ==
        manifest_object
    ), (
        "Existing Notebook 05 prior/rarity manifest differs "
        "from current recomputation. STOP."
    )


else:

    with open(
        NOTEBOOK05_PRIOR_MANIFEST_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            manifest_object,
            f,
            indent=2,
            sort_keys=True
        )


NOTEBOOK05_PRIOR_MANIFEST_SHA = (
    sha256_file(
        NOTEBOOK05_PRIOR_MANIFEST_PATH
    )
)


# --------------------------------------------------------------------------------------------------
# 7. SAFETY GATE
# --------------------------------------------------------------------------------------------------

assert (
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
    is False
)

assert (
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
    is False
)

assert (
    NOTEBOOK05_OFFICIAL_TEST_USED
    is False
)

assert (
    NOTEBOOK05_TRAINING_PERFORMED
    is False
)


NOTEBOOK05_PRIOR_RARITY_GATE_PASS = True


# --------------------------------------------------------------------------------------------------
# 8. FINAL REPORT
# --------------------------------------------------------------------------------------------------

print()
print("=" * 135)

print(
    "Training-count sources verified : 3/3 PASS"
)

print(
    "70/30 count reconstruction       : 3/3 PASS"
)

print(
    "Training priors                  : 3/3 PASS"
)

print(
    "Log-frequency rarity             : 3/3 PASS"
)

print(
    "Head/Medium/Tail groups          : 3/3 PASS"
)

print(
    "Validation counts cross-checked  : 3/3 PASS"
)

print()

print(
    "Prior/rarity manifest:"
)

print(
    NOTEBOOK05_PRIOR_MANIFEST_PATH
)

print()

print(
    "Manifest SHA-256:"
)

print(
    NOTEBOOK05_PRIOR_MANIFEST_SHA
)

print()

print(
    "Training performed:",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset:",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader:",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 PRIOR + RARITY FREEZE GATE: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 135)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 5
# POST-HOC LA + LTLC MATHEMATICAL ENGINE
# METRIC UTILITIES + MANDATORY NUMERICAL SANITY TESTS
#
# CPU ONLY
# NO TRAINING
# NO PARAMETER TUNING YET
# NO OFFICIAL TEST
# ==================================================================================================

import json
import math
import numpy as np


# ==================================================================================================
# 1. LOCKED POST-HOC CONSTANTS
# ==================================================================================================

POSTHOC_EPS = 1e-12


TAU_GRID = [
    0.00,
    0.25,
    0.50,
    0.75,
    1.00,
    1.25,
    1.50,
    1.75,
    2.00,
]


ALPHA_GRID = [
    0.00,
    0.25,
    0.50,
    0.75,
    1.00,
    1.50,
    2.00,
]


LTLC_TEMPERATURE_BOUNDS = {

    "a":
        [
            -2.0,
            2.0
        ],

    "b":
        [
            -2.0,
            2.0
        ],
}


LTLC_TEMPERATURE_INITIAL = {

    "a":
        0.0,

    "b":
        0.0,
}


# ==================================================================================================
# 2. INPUT VALIDATION HELPERS
# ==================================================================================================

def validate_logits(
    logits
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )


    assert (
        logits.ndim
        ==
        2
    ), (
        "Logits must have shape [N, K]."
    )


    assert (
        logits.shape[
            0
        ]
        >
        0
    )


    assert (
        logits.shape[
            1
        ]
        >
        1
    )


    assert np.isfinite(
        logits
    ).all(), (
        "Non-finite logits detected."
    )


    return logits


def validate_class_vector(
    vector,
    k,
    name,
    strictly_positive=False
):

    vector = np.asarray(
        vector,
        dtype=np.float64
    )


    assert (
        vector.ndim
        ==
        1
    )


    assert (
        len(
            vector
        )
        ==
        k
    ), (
        f"{name} length does not match K."
    )


    assert np.isfinite(
        vector
    ).all(), (
        f"{name} contains non-finite values."
    )


    if strictly_positive:

        assert (
            vector
            >
            0
        ).all(), (
            f"{name} must be strictly positive."
        )


    return vector


# ==================================================================================================
# 3. NUMERICALLY STABLE SOFTMAX
# ==================================================================================================

def stable_softmax(
    logits
):

    logits = validate_logits(
        logits
    )


    shifted = (
        logits
        -
        np.max(
            logits,
            axis=1,
            keepdims=True
        )
    )


    exp_logits = np.exp(
        shifted
    )


    probabilities = (
        exp_logits
        /
        np.sum(
            exp_logits,
            axis=1,
            keepdims=True
        )
    )


    assert np.isfinite(
        probabilities
    ).all()


    return probabilities


# ==================================================================================================
# 4. NUMERICALLY STABLE MULTICLASS NLL
# ==================================================================================================

def multiclass_nll_from_logits(
    logits,
    labels
):

    logits = validate_logits(
        logits
    )


    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    n, k = (
        logits.shape
    )


    assert (
        labels.shape
        ==
        (
            n,
        )
    )


    assert (
        labels.min()
        >=
        0
    )


    assert (
        labels.max()
        <
        k
    )


    row_max = np.max(
        logits,
        axis=1,
        keepdims=True
    )


    shifted = (
        logits
        -
        row_max
    )


    log_sum_exp = (
        np.log(
            np.sum(
                np.exp(
                    shifted
                ),
                axis=1
            )
        )
        +
        row_max[
            :,
            0
        ]
    )


    correct_logits = (
        logits[
            np.arange(
                n
            ),
            labels
        ]
    )


    nll = np.mean(
        log_sum_exp
        -
        correct_logits
    )


    assert np.isfinite(
        nll
    )


    return float(
        nll
    )


# ==================================================================================================
# 5. CONFUSION MATRIX
# ==================================================================================================

def confusion_matrix_numpy(
    labels,
    predictions,
    k
):

    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    predictions = np.asarray(
        predictions,
        dtype=np.int64
    )


    assert (
        labels.shape
        ==
        predictions.shape
    )


    assert (
        labels.ndim
        ==
        1
    )


    assert (
        labels.min()
        >=
        0
    )


    assert (
        labels.max()
        <
        k
    )


    assert (
        predictions.min()
        >=
        0
    )


    assert (
        predictions.max()
        <
        k
    )


    encoded = (
        labels
        *
        k
        +
        predictions
    )


    matrix = (
        np.bincount(
            encoded,
            minlength=
                k * k
        )
        .reshape(
            k,
            k
        )
        .astype(
            np.int64
        )
    )


    return matrix


# ==================================================================================================
# 6. CORE CLASSIFICATION METRICS
# ==================================================================================================

def classification_metrics_from_predictions(
    labels,
    predictions,
    k
):

    cm = confusion_matrix_numpy(
        labels,
        predictions,
        k
    )


    total = int(
        cm.sum()
    )


    true_positive = (
        np.diag(
            cm
        )
        .astype(
            np.float64
        )
    )


    support = (
        cm.sum(
            axis=1
        )
        .astype(
            np.float64
        )
    )


    predicted_support = (
        cm.sum(
            axis=0
        )
        .astype(
            np.float64
        )
    )


    assert (
        support
        >
        0
    ).all(), (
        "Every evaluation class must have support."
    )


    # ----------------------------------------------------------------------------------------------
    # OA
    # ----------------------------------------------------------------------------------------------

    oa = float(
        true_positive.sum()
        /
        total
    )


    # ----------------------------------------------------------------------------------------------
    # Per-class recall / AA
    # ----------------------------------------------------------------------------------------------

    per_class_accuracy = (
        true_positive
        /
        support
    )


    aa = float(
        np.mean(
            per_class_accuracy
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Per-class precision / F1
    # ----------------------------------------------------------------------------------------------

    precision = np.divide(

        true_positive,

        predicted_support,

        out=np.zeros_like(
            true_positive,
            dtype=np.float64
        ),

        where=(
            predicted_support
            >
            0
        )
    )


    recall = (
        per_class_accuracy
    )


    f1 = np.divide(

        2.0
        *
        precision
        *
        recall,

        precision
        +
        recall,

        out=np.zeros_like(
            precision,
            dtype=np.float64
        ),

        where=(
            precision
            +
            recall
            >
            0
        )
    )


    macro_f1 = float(
        np.mean(
            f1
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Cohen's Kappa
    # ----------------------------------------------------------------------------------------------

    expected_agreement = float(

        np.sum(
            support
            *
            predicted_support
        )

        /
        (
            total
            *
            total
        )
    )


    if math.isclose(
        1.0
        -
        expected_agreement,
        0.0,
        abs_tol=1e-15
    ):

        kappa = 0.0


    else:

        kappa = float(

            (
                oa
                -
                expected_agreement
            )

            /
            (
                1.0
                -
                expected_agreement
            )
        )


    return {

        "oa":
            oa,

        "aa":
            aa,

        "macro_f1":
            macro_f1,

        "kappa":
            kappa,

        "per_class_accuracy":
            per_class_accuracy,

        "per_class_f1":
            f1,

        "confusion_matrix":
            cm,
    }


# ==================================================================================================
# 7. NLL / BRIER / ECE
# ==================================================================================================

def multiclass_brier_score(
    probabilities,
    labels
):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )


    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    assert (
        probabilities.ndim
        ==
        2
    )


    n, k = (
        probabilities.shape
    )


    assert (
        labels.shape
        ==
        (
            n,
        )
    )


    one_hot = np.zeros(
        (
            n,
            k
        ),
        dtype=np.float64
    )


    one_hot[
        np.arange(
            n
        ),
        labels
    ] = 1.0


    return float(
        np.mean(
            np.sum(
                (
                    probabilities
                    -
                    one_hot
                )
                ** 2,
                axis=1
            )
        )
    )


def expected_calibration_error(
    probabilities,
    labels,
    n_bins=15
):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )


    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    assert (
        probabilities.ndim
        ==
        2
    )


    confidence = np.max(
        probabilities,
        axis=1
    )


    predictions = np.argmax(
        probabilities,
        axis=1
    )


    correct = (
        predictions
        ==
        labels
    ).astype(
        np.float64
    )


    boundaries = np.linspace(
        0.0,
        1.0,
        n_bins + 1
    )


    ece = 0.0


    for bin_index in range(
        n_bins
    ):

        lower = boundaries[
            bin_index
        ]

        upper = boundaries[
            bin_index + 1
        ]


        if (
            bin_index
            ==
            0
        ):

            in_bin = (
                confidence
                >=
                lower
            ) & (
                confidence
                <=
                upper
            )


        else:

            in_bin = (
                confidence
                >
                lower
            ) & (
                confidence
                <=
                upper
            )


        count = int(
            in_bin.sum()
        )


        if (
            count
            ==
            0
        ):

            continue


        bin_accuracy = float(
            correct[
                in_bin
            ].mean()
        )


        bin_confidence = float(
            confidence[
                in_bin
            ].mean()
        )


        ece += (
            count
            /
            len(
                labels
            )
        ) * abs(
            bin_accuracy
            -
            bin_confidence
        )


    return float(
        ece
    )


# ==================================================================================================
# 8. FREQUENCY-GROUP METRICS
# ==================================================================================================

def frequency_group_metrics(
    per_class_accuracy,
    frequency_groups
):

    per_class_accuracy = np.asarray(
        per_class_accuracy,
        dtype=np.float64
    )


    frequency_groups = np.asarray(
        frequency_groups,
        dtype=object
    )


    assert (
        len(
            per_class_accuracy
        )
        ==
        len(
            frequency_groups
        )
    )


    result = {}


    for group in [
        "Tail",
        "Medium",
        "Head",
    ]:

        mask = (
            frequency_groups
            ==
            group
        )


        assert (
            mask.sum()
            >
            0
        )


        result[
            group.lower()
            +
            "_accuracy"
        ] = float(
            np.mean(
                per_class_accuracy[
                    mask
                ]
            )
        )


    result[
        "worst_class_accuracy"
    ] = float(
        np.min(
            per_class_accuracy
        )
    )


    return result


# ==================================================================================================
# 9. COMPONENT A — STANDARD POST-HOC LOGIT ADJUSTMENT
#
# z_A = z - tau * log(pi + eps)
# ==================================================================================================

def standard_posthoc_la(
    logits,
    prior,
    tau,
    eps=POSTHOC_EPS
):

    logits = validate_logits(
        logits
    )


    n, k = (
        logits.shape
    )


    prior = validate_class_vector(
        prior,
        k,
        "prior",
        strictly_positive=True
    )


    tau = float(
        tau
    )


    assert (
        tau
        >=
        0.0
    )


    adjustment = (
        tau
        *
        np.log(
            prior
            +
            eps
        )
    )


    adjusted_logits = (
        logits
        -
        adjustment[
            None,
            :
        ]
    )


    assert np.isfinite(
        adjusted_logits
    ).all()


    return adjusted_logits


# ==================================================================================================
# 10. COMPONENT B — LTLC RARITY-ADAPTIVE SHIFT
#
# z_B =
# z - tau * (1 + alpha*r_k) * log(pi_k + eps)
# ==================================================================================================

def ltlc_shift(
    logits,
    prior,
    rarity,
    tau,
    alpha,
    eps=POSTHOC_EPS
):

    logits = validate_logits(
        logits
    )


    n, k = (
        logits.shape
    )


    prior = validate_class_vector(
        prior,
        k,
        "prior",
        strictly_positive=True
    )


    rarity = validate_class_vector(
        rarity,
        k,
        "rarity",
        strictly_positive=False
    )


    assert (
        rarity
        >=
        -1e-12
    ).all()


    assert (
        rarity
        <=
        1.0 + 1e-12
    ).all()


    tau = float(
        tau
    )


    alpha = float(
        alpha
    )


    assert (
        tau
        >=
        0.0
    )


    assert (
        alpha
        >=
        0.0
    )


    strength = (
        tau
        *
        (
            1.0
            +
            alpha
            *
            rarity
        )
    )


    adjustment = (
        strength
        *
        np.log(
            prior
            +
            eps
        )
    )


    shifted_logits = (
        logits
        -
        adjustment[
            None,
            :
        ]
    )


    assert np.isfinite(
        shifted_logits
    ).all()


    return shifted_logits


# ==================================================================================================
# 11. GLOBAL TEMPERATURE SCALING
#
# Parameterized in log-temperature space:
#
# T = exp(a)
# ==================================================================================================

def apply_global_temperature(
    logits,
    a
):

    logits = validate_logits(
        logits
    )


    a = float(
        a
    )


    temperature = float(
        np.exp(
            a
        )
    )


    assert (
        temperature
        >
        0.0
    )


    calibrated_logits = (
        logits
        /
        temperature
    )


    return (
        calibrated_logits,
        temperature
    )


# ==================================================================================================
# 12. COMPONENT C — RARITY-CONDITIONED SAMPLE-WISE TEMPERATURE
#
# predicted class:
#   y_hat = argmax(z_B)
#
# T(x) =
#   exp(a + b * r_yhat)
#
# same scalar temperature divides every class logit
# for one sample.
# ==================================================================================================

def apply_ltlc_temperature(
    shifted_logits,
    rarity,
    a,
    b
):

    shifted_logits = validate_logits(
        shifted_logits
    )


    n, k = (
        shifted_logits.shape
    )


    rarity = validate_class_vector(
        rarity,
        k,
        "rarity",
        strictly_positive=False
    )


    a = float(
        a
    )


    b = float(
        b
    )


    predicted_before_temperature = (
        np.argmax(
            shifted_logits,
            axis=1
        )
    )


    predicted_rarity = (
        rarity[
            predicted_before_temperature
        ]
    )


    temperatures = np.exp(
        a
        +
        b
        *
        predicted_rarity
    )


    assert (
        temperatures.shape
        ==
        (
            n,
        )
    )


    assert np.isfinite(
        temperatures
    ).all()


    assert (
        temperatures
        >
        0.0
    ).all()


    calibrated_logits = (
        shifted_logits
        /
        temperatures[
            :,
            None
        ]
    )


    predicted_after_temperature = (
        np.argmax(
            calibrated_logits,
            axis=1
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Mathematical requirement:
    # positive scalar-per-sample temperature MUST preserve argmax.
    # ----------------------------------------------------------------------------------------------

    assert np.array_equal(
        predicted_before_temperature,
        predicted_after_temperature
    ), (
        "LTLC temperature changed predicted class."
    )


    return {

        "calibrated_logits":
            calibrated_logits,

        "temperature":
            temperatures,

        "prediction":
            predicted_before_temperature,

        "predicted_rarity":
            predicted_rarity,
    }


# ==================================================================================================
# 13. COMPLETE LTLC FORWARD FUNCTION
# ==================================================================================================

def ltlc_forward(
    logits,
    prior,
    rarity,
    tau,
    alpha,
    a,
    b,
    eps=POSTHOC_EPS
):

    shifted_logits = ltlc_shift(

        logits=
            logits,

        prior=
            prior,

        rarity=
            rarity,

        tau=
            tau,

        alpha=
            alpha,

        eps=
            eps
    )


    temperature_result = (
        apply_ltlc_temperature(

            shifted_logits=
                shifted_logits,

            rarity=
                rarity,

            a=
                a,

            b=
                b
        )
    )


    probabilities = stable_softmax(

        temperature_result[
            "calibrated_logits"
        ]
    )


    return {

        "shifted_logits":
            shifted_logits,

        "calibrated_logits":
            temperature_result[
                "calibrated_logits"
            ],

        "probabilities":
            probabilities,

        "temperature":
            temperature_result[
                "temperature"
            ],

        "prediction":
            temperature_result[
                "prediction"
            ],

        "predicted_rarity":
            temperature_result[
                "predicted_rarity"
            ],
    }


# ==================================================================================================
# 14. MANDATORY SANITY TESTS ON ALL 9 REAL VALIDATION CACHES
# ==================================================================================================

SANITY_ROWS = []


print("=" * 135)
print("NOTEBOOK 05 — LTLC MATHEMATICAL SANITY AUDIT")
print("=" * 135)


for dataset in DATASET_ORDER:

    prior = (
        CLASS_PRIORS[
            dataset
        ]
    )


    rarity = (
        CLASS_RARITY[
            dataset
        ]
    )


    k = len(
        prior
    )


    print()
    print("-" * 135)
    print(dataset.upper())
    print("-" * 135)


    for seed in EXPECTED_SEEDS:

        cache = (
            VALIDATION_CACHE[
                (
                    dataset,
                    seed
                )
            ]
        )


        logits = np.asarray(
            cache[
                "logits"
            ],
            dtype=np.float64
        )


        labels = np.asarray(
            cache[
                "labels"
            ],
            dtype=np.int64
        )


        # ------------------------------------------------------------------------------------------
        # TEST 1
        # Standard LA tau=0 must reduce to original logits.
        # ------------------------------------------------------------------------------------------

        la_tau_zero = (
            standard_posthoc_la(
                logits,
                prior,
                tau=0.0
            )
        )


        assert np.allclose(
            la_tau_zero,
            logits,
            rtol=0.0,
            atol=1e-12
        )


        # ------------------------------------------------------------------------------------------
        # TEST 2
        # LTLC tau=0 must reduce to original logits,
        # regardless of alpha.
        # ------------------------------------------------------------------------------------------

        ltlc_tau_zero = (
            ltlc_shift(
                logits,
                prior,
                rarity,
                tau=0.0,
                alpha=2.0
            )
        )


        assert np.allclose(
            ltlc_tau_zero,
            logits,
            rtol=0.0,
            atol=1e-12
        )


        # ------------------------------------------------------------------------------------------
        # TEST 3 — MANDATORY
        #
        # alpha=0 LTLC-Shift must reproduce standard LA.
        # ------------------------------------------------------------------------------------------

        la_reference = (
            standard_posthoc_la(
                logits,
                prior,
                tau=1.0
            )
        )


        ltlc_alpha_zero = (
            ltlc_shift(
                logits,
                prior,
                rarity,
                tau=1.0,
                alpha=0.0
            )
        )


        alpha_zero_max_diff = float(

            np.max(
                np.abs(
                    la_reference
                    -
                    ltlc_alpha_zero
                )
            )
        )


        assert (
            alpha_zero_max_diff
            <=
            1e-12
        ), (
            f"{dataset} seed {seed}: "
            "alpha=0 does not reproduce standard LA."
        )


        # ------------------------------------------------------------------------------------------
        # TEST 4
        # Positive rarity-conditioned temperature must preserve argmax.
        # ------------------------------------------------------------------------------------------

        shift_test = (
            ltlc_shift(
                logits,
                prior,
                rarity,
                tau=1.0,
                alpha=0.75
            )
        )


        temp_test = (
            apply_ltlc_temperature(
                shifted_logits=
                    shift_test,

                rarity=
                    rarity,

                a=
                    0.20,

                b=
                    0.40
            )
        )


        pred_before = np.argmax(
            shift_test,
            axis=1
        )


        pred_after = np.argmax(

            temp_test[
                "calibrated_logits"
            ],

            axis=1
        )


        assert np.array_equal(
            pred_before,
            pred_after
        )


        # ------------------------------------------------------------------------------------------
        # TEST 5
        # a=0, b=0 -> T=1 -> temperature stage is identity.
        # ------------------------------------------------------------------------------------------

        identity_temperature = (
            apply_ltlc_temperature(

                shifted_logits=
                    shift_test,

                rarity=
                    rarity,

                a=
                    0.0,

                b=
                    0.0
            )
        )


        assert np.allclose(

            identity_temperature[
                "temperature"
            ],

            1.0,

            rtol=0.0,
            atol=1e-12
        )


        assert np.allclose(

            identity_temperature[
                "calibrated_logits"
            ],

            shift_test,

            rtol=0.0,
            atol=1e-12
        )


        # ------------------------------------------------------------------------------------------
        # TEST 6
        # b=0 must reproduce ordinary global TS.
        # ------------------------------------------------------------------------------------------

        samplewise_b_zero = (
            apply_ltlc_temperature(

                shifted_logits=
                    shift_test,

                rarity=
                    rarity,

                a=
                    0.35,

                b=
                    0.0
            )
        )


        (
            global_ts_logits,
            global_temperature
        ) = apply_global_temperature(

            shift_test,
            a=0.35
        )


        b_zero_max_diff = float(

            np.max(
                np.abs(

                    samplewise_b_zero[
                        "calibrated_logits"
                    ]

                    -

                    global_ts_logits
                )
            )
        )


        assert (
            b_zero_max_diff
            <=
            1e-12
        ), (
            f"{dataset} seed {seed}: "
            "b=0 does not reproduce global TS."
        )


        assert np.allclose(

            samplewise_b_zero[
                "temperature"
            ],

            global_temperature,

            rtol=0.0,
            atol=1e-12
        )


        # ------------------------------------------------------------------------------------------
        # TEST 7
        # Softmax probabilities must be finite and sum to 1.
        # ------------------------------------------------------------------------------------------

        probabilities = stable_softmax(

            temp_test[
                "calibrated_logits"
            ]
        )


        probability_row_sum_error = float(

            np.max(

                np.abs(

                    probabilities.sum(
                        axis=1
                    )

                    -
                    1.0
                )
            )
        )


        assert (
            probability_row_sum_error
            <=
            1e-12
        )


        assert (
            probabilities
            >=
            0.0
        ).all()


        assert (
            probabilities
            <=
            1.0
        ).all()


        # ------------------------------------------------------------------------------------------
        # TEST 8
        # Baseline metrics should reproduce saved CE predictions.
        # ------------------------------------------------------------------------------------------

        baseline_predictions = np.argmax(
            logits,
            axis=1
        )


        assert np.array_equal(

            baseline_predictions,

            cache[
                "predictions"
            ]
        )


        baseline_metrics = (
            classification_metrics_from_predictions(

                labels=
                    labels,

                predictions=
                    baseline_predictions,

                k=
                    k
            )
        )


        baseline_nll = (
            multiclass_nll_from_logits(
                logits,
                labels
            )
        )


        baseline_probabilities = (
            stable_softmax(
                logits
            )
        )


        baseline_ece = (
            expected_calibration_error(

                baseline_probabilities,

                labels,
                n_bins=15
            )
        )


        baseline_brier = (
            multiclass_brier_score(

                baseline_probabilities,

                labels
            )
        )


        group_metrics = (
            frequency_group_metrics(

                baseline_metrics[
                    "per_class_accuracy"
                ],

                CLASS_FREQUENCY_GROUPS[
                    dataset
                ]
            )
        )


        # ------------------------------------------------------------------------------------------
        # TEST 9
        # Temperature bounds imply strictly positive finite T.
        #
        # Test extreme corners of allowed a,b bounds.
        # ------------------------------------------------------------------------------------------

        for test_a in [
            -2.0,
            2.0
        ]:

            for test_b in [
                -2.0,
                2.0
            ]:

                extreme_temp = (
                    apply_ltlc_temperature(

                        shifted_logits=
                            shift_test,

                        rarity=
                            rarity,

                        a=
                            test_a,

                        b=
                            test_b
                    )
                )


                assert np.isfinite(

                    extreme_temp[
                        "temperature"
                    ]

                ).all()


                assert (

                    extreme_temp[
                        "temperature"
                    ]

                    >
                    0.0

                ).all()


        SANITY_ROWS.append({

            "dataset":
                dataset,

            "seed":
                seed,

            "alpha_zero_max_diff":
                alpha_zero_max_diff,

            "b_zero_global_ts_max_diff":
                b_zero_max_diff,

            "probability_row_sum_max_error":
                probability_row_sum_error,

            "baseline_oa":
                baseline_metrics[
                    "oa"
                ],

            "baseline_aa":
                baseline_metrics[
                    "aa"
                ],

            "baseline_macro_f1":
                baseline_metrics[
                    "macro_f1"
                ],

            "baseline_kappa":
                baseline_metrics[
                    "kappa"
                ],

            "baseline_nll":
                baseline_nll,

            "baseline_ece":
                baseline_ece,

            "baseline_brier":
                baseline_brier,

            "baseline_tail_accuracy":
                group_metrics[
                    "tail_accuracy"
                ],

            "baseline_medium_accuracy":
                group_metrics[
                    "medium_accuracy"
                ],

            "baseline_head_accuracy":
                group_metrics[
                    "head_accuracy"
                ],

            "baseline_worst_class_accuracy":
                group_metrics[
                    "worst_class_accuracy"
                ],
        })


        print(
            f"seed={seed:4d} | "
            f"alpha0_diff={alpha_zero_max_diff:.3e} | "
            f"b0_diff={b_zero_max_diff:.3e} | "
            f"argmax=PASS | "
            f"prob_sum=PASS | "
            f"OA={baseline_metrics['oa']:.6f} | "
            f"AA={baseline_metrics['aa']:.6f} | "
            f"NLL={baseline_nll:.6f}"
        )


# ==================================================================================================
# 15. SANITY AUDIT TABLE
# ==================================================================================================

SANITY_DF = (
    pd.DataFrame(
        SANITY_ROWS
    )
)


assert (
    len(
        SANITY_DF
    )
    ==
    9
)


SANITY_CSV_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "notebook05_posthoc_mathematical_sanity_audit.csv"
)


SANITY_DF.to_csv(
    SANITY_CSV_PATH,
    index=False
)


SANITY_CSV_SHA = (
    sha256_file(
        SANITY_CSV_PATH
    )
)


# ==================================================================================================
# 16. FREEZE NOTEBOOK 05 POST-HOC IMPLEMENTATION SPEC
# ==================================================================================================

POSTHOC_SPEC_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "notebook05_posthoc_ltlc_implementation_spec.json"
)


POSTHOC_SPEC = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "status":
        "IMPLEMENTATION_LOCKED_BEFORE_TUNING",

    "epsilon":
        POSTHOC_EPS,

    "standard_posthoc_LA":
        (
            "z_A[k] = z[k] - "
            "tau * log(pi[k] + eps)"
        ),

    "LTLC_shift":
        (
            "z_B[k] = z[k] - "
            "tau * (1 + alpha * rarity[k]) "
            "* log(pi[k] + eps)"
        ),

    "LTLC_temperature":
        (
            "T(x) = exp(a + b * "
            "rarity[argmax(z_B(x))])"
        ),

    "LTLC_calibrated_logits":
        (
            "z_C(x) = z_B(x) / T(x), "
            "same positive scalar T(x) for every "
            "class of one sample"
        ),

    "tau_grid":
        TAU_GRID,

    "alpha_grid":
        ALPHA_GRID,

    "shift_selection_rule": [

        "highest_validation_AA",

        "tie_highest_validation_Macro_F1",

        "tie_lowest_validation_NLL",

        "tie_smaller_alpha",
    ],

    "temperature_optimizer":
        "L-BFGS-B",

    "temperature_initial":
        LTLC_TEMPERATURE_INITIAL,

    "temperature_bounds":
        LTLC_TEMPERATURE_BOUNDS,

    "temperature_objective":
        "validation_multiclass_NLL",

    "mandatory_sanity_checks": {

        "tau_zero_standard_LA_identity":
            True,

        "tau_zero_LTLC_shift_identity":
            True,

        "alpha_zero_reproduces_standard_LA":
            True,

        "temperature_strictly_positive":
            True,

        "samplewise_scalar_temperature_preserves_argmax":
            True,

        "a_zero_b_zero_temperature_identity":
            True,

        "b_zero_reproduces_global_temperature_scaling":
            True,

        "softmax_probability_rows_sum_to_one":
            True,
    },

    "training_prior_rarity_manifest_sha256":
        NOTEBOOK05_PRIOR_MANIFEST_SHA,

    "mathematical_sanity_csv_sha256":
        SANITY_CSV_SHA,

    "validation_output_sha256": {

        f"{dataset}_seed{seed}":
            VALIDATION_CACHE[
                (
                    dataset,
                    seed
                )
            ][
                "sha256"
            ]

        for dataset
        in DATASET_ORDER

        for seed
        in EXPECTED_SEEDS
    },

    "training_performed":
        False,

    "official_test_used":
        False,
}


# ==================================================================================================
# 17. RESTART-SAFE SPEC WRITE
# ==================================================================================================

if (
    POSTHOC_SPEC_PATH.exists()
):

    with open(
        POSTHOC_SPEC_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing_spec = (
            json.load(
                f
            )
        )


    assert (
        existing_spec
        ==
        POSTHOC_SPEC
    ), (
        "Existing Notebook 05 post-hoc implementation "
        "spec differs from the current locked implementation."
    )


else:

    with open(
        POSTHOC_SPEC_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            POSTHOC_SPEC,
            f,
            indent=2,
            sort_keys=True
        )


POSTHOC_SPEC_SHA = (
    sha256_file(
        POSTHOC_SPEC_PATH
    )
)


# ==================================================================================================
# 18. OFFICIAL-TEST SAFETY GATE
# ==================================================================================================

assert (
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
    is False
)


assert (
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
    is False
)


assert (
    NOTEBOOK05_OFFICIAL_TEST_USED
    is False
)


assert (
    NOTEBOOK05_TRAINING_PERFORMED
    is False
)


NOTEBOOK05_POSTHOC_ENGINE_GATE_PASS = True


# ==================================================================================================
# 19. FINAL REPORT
# ==================================================================================================

print()
print("=" * 135)

print(
    "Standard LA implementation              : PASS"
)

print(
    "LTLC rarity-adaptive shift              : PASS"
)

print(
    "alpha=0 reproduces standard LA          : 9/9 PASS"
)

print(
    "tau=0 identity                          : 9/9 PASS"
)

print(
    "Scalar temperature preserves argmax     : 9/9 PASS"
)

print(
    "a=0, b=0 temperature identity           : 9/9 PASS"
)

print(
    "b=0 reproduces Global TS                : 9/9 PASS"
)

print(
    "Stable softmax probability check        : 9/9 PASS"
)

print(
    "Baseline metric engine                  : 9/9 PASS"
)

print()

print(
    "Sanity audit CSV:"
)

print(
    SANITY_CSV_PATH
)

print(
    "SHA-256:"
)

print(
    SANITY_CSV_SHA
)

print()

print(
    "Locked implementation spec:"
)

print(
    POSTHOC_SPEC_PATH
)

print(
    "SHA-256:"
)

print(
    POSTHOC_SPEC_SHA
)

print()

print(
    "Training performed:",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset:",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader:",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 POST-HOC MATHEMATICAL ENGINE GATE: PASS"
)

print(
    "IMPLEMENTATION LOCKED BEFORE PARAMETER TUNING"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 135)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 6
# TANGDAOWAN VALIDATION-ONLY STANDARD LA + LTLC-SHIFT GRID SEARCH
#
# PRIMARY STRESS-TEST ONLY
#
# For each frozen CE seed:
#
#   Standard LA:
#       alpha = 0
#       tau in TAU_GRID
#
#   LTLC-Shift:
#       tau   in TAU_GRID
#       alpha in ALPHA_GRID
#
# Selection rule:
#   1. highest validation AA
#   2. tie -> highest validation Macro-F1
#   3. tie -> lowest validation NLL
#   4. tie -> smaller alpha
#
# If the declared rule still leaves >1 candidate,
# STOP rather than inventing an extra scientific criterion.
#
# CPU ONLY
# NO TRAINING
# NO TEMPERATURE FITTING YET
# NO OFFICIAL TEST
# ==================================================================================================

import json
import math
from pathlib import Path

import numpy as np
import pandas as pd


# ==================================================================================================
# 1. HARD PRECONDITION GATES
# ==================================================================================================

assert NOTEBOOK05_VALIDATION_INVENTORY_PASS is True
assert NOTEBOOK05_VALIDATION_CACHE_INTEGRITY_PASS is True
assert NOTEBOOK05_PRIOR_RARITY_GATE_PASS is True
assert NOTEBOOK05_POSTHOC_ENGINE_GATE_PASS is True

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False
assert NOTEBOOK05_TRAINING_PERFORMED is False


# ==================================================================================================
# 2. LOCKED TARGET
# ==================================================================================================

TARGET_DATASET = "Tangdaowan"

TARGET_SEEDS = [
    42,
    123,
    3407,
]

GRID_TOL = 1e-12


prior = np.asarray(
    CLASS_PRIORS[
        TARGET_DATASET
    ],
    dtype=np.float64
)

rarity = np.asarray(
    CLASS_RARITY[
        TARGET_DATASET
    ],
    dtype=np.float64
)

frequency_groups = np.asarray(
    CLASS_FREQUENCY_GROUPS[
        TARGET_DATASET
    ],
    dtype=object
)

K = int(
    len(
        prior
    )
)


assert K == 16


# ==================================================================================================
# 3. SELECTION HELPER
# ==================================================================================================

def select_by_locked_shift_rule(
    candidate_df,
    label
):
    """
    Locked selection:

        1. highest validation AA
        2. highest validation Macro-F1
        3. lowest validation NLL
        4. smaller alpha

    If more than one candidate remains after all declared
    criteria, STOP rather than silently adding another rule.
    """

    df = candidate_df.copy()


    assert (
        len(
            df
        )
        >
        0
    )


    # ----------------------------------------------------------------------------------------------
    # 1. MAX AA
    # ----------------------------------------------------------------------------------------------

    best_aa = float(
        df[
            "validation_aa"
        ].max()
    )


    df = df[
        np.isclose(
            df[
                "validation_aa"
            ].to_numpy(
                dtype=np.float64
            ),
            best_aa,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 2. MAX MACRO-F1
    # ----------------------------------------------------------------------------------------------

    best_f1 = float(
        df[
            "validation_macro_f1"
        ].max()
    )


    df = df[
        np.isclose(
            df[
                "validation_macro_f1"
            ].to_numpy(
                dtype=np.float64
            ),
            best_f1,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 3. MIN NLL
    # ----------------------------------------------------------------------------------------------

    best_nll = float(
        df[
            "validation_nll"
        ].min()
    )


    df = df[
        np.isclose(
            df[
                "validation_nll"
            ].to_numpy(
                dtype=np.float64
            ),
            best_nll,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 4. SMALLER ALPHA
    # ----------------------------------------------------------------------------------------------

    best_alpha = float(
        df[
            "alpha"
        ].min()
    )


    df = df[
        np.isclose(
            df[
                "alpha"
            ].to_numpy(
                dtype=np.float64
            ),
            best_alpha,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # NO UNDECLARED TIE-BREAK
    # ----------------------------------------------------------------------------------------------

    if (
        len(
            df
        )
        !=
        1
    ):

        print()
        print("=" * 120)
        print("UNRESOLVED LOCKED-PROTOCOL TIE")
        print("=" * 120)

        print(
            label
        )

        print()

        print(
            df[
                [
                    "tau",
                    "alpha",
                    "validation_aa",
                    "validation_macro_f1",
                    "validation_nll",
                ]
            ].to_string(
                index=False
            )
        )

        print("=" * 120)


        raise RuntimeError(
            f"{label}: the predeclared selection rule "
            "did not yield a unique candidate. "
            "STOP before adding any new criterion."
        )


    return (
        df.iloc[
            0
        ]
        .to_dict()
    )


# ==================================================================================================
# 4. FULL 9 × 7 GRID FOR EACH TANGDAOWAN CE SEED
# ==================================================================================================

GRID_ROWS = []

CE_BASELINE_ROWS = []


print("=" * 140)
print("NOTEBOOK 05 — TANGDAOWAN VALIDATION-ONLY LA / LTLC-SHIFT SEARCH")
print("=" * 140)

print(
    "Tau grid:",
    TAU_GRID
)

print(
    "Alpha grid:",
    ALPHA_GRID
)

print(
    "Grid points per seed:",
    len(
        TAU_GRID
    )
    *
    len(
        ALPHA_GRID
    )
)

print(
    "CE seeds:",
    TARGET_SEEDS
)

print(
    "Official test used:",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print("=" * 140)


for seed in TARGET_SEEDS:

    cache = (
        VALIDATION_CACHE[
            (
                TARGET_DATASET,
                seed
            )
        ]
    )


    logits = np.asarray(
        cache[
            "logits"
        ],
        dtype=np.float64
    )


    labels = np.asarray(
        cache[
            "labels"
        ],
        dtype=np.int64
    )


    # ==============================================================================================
    # 4A. CE VALIDATION REFERENCE
    # ==============================================================================================

    ce_predictions = np.argmax(
        logits,
        axis=1
    )


    ce_metrics = (
        classification_metrics_from_predictions(

            labels=
                labels,

            predictions=
                ce_predictions,

            k=
                K
        )
    )


    ce_group = (
        frequency_group_metrics(

            ce_metrics[
                "per_class_accuracy"
            ],

            frequency_groups
        )
    )


    ce_nll = (
        multiclass_nll_from_logits(
            logits,
            labels
        )
    )


    ce_probabilities = (
        stable_softmax(
            logits
        )
    )


    ce_ece = (
        expected_calibration_error(

            ce_probabilities,

            labels,

            n_bins=
                15
        )
    )


    CE_BASELINE_ROWS.append({

        "dataset":
            TARGET_DATASET,

        "seed":
            int(
                seed
            ),

        "method":
            "CE",

        "tau":
            0.0,

        "alpha":
            0.0,

        "validation_oa":
            float(
                ce_metrics[
                    "oa"
                ]
            ),

        "validation_aa":
            float(
                ce_metrics[
                    "aa"
                ]
            ),

        "validation_macro_f1":
            float(
                ce_metrics[
                    "macro_f1"
                ]
            ),

        "validation_nll":
            float(
                ce_nll
            ),

        "validation_ece":
            float(
                ce_ece
            ),

        "validation_tail_accuracy":
            float(
                ce_group[
                    "tail_accuracy"
                ]
            ),

        "validation_medium_accuracy":
            float(
                ce_group[
                    "medium_accuracy"
                ]
            ),

        "validation_head_accuracy":
            float(
                ce_group[
                    "head_accuracy"
                ]
            ),

        "validation_worst_class_accuracy":
            float(
                ce_group[
                    "worst_class_accuracy"
                ]
            ),
    })


    # ==============================================================================================
    # 4B. LTLC-SHIFT GRID
    # ==============================================================================================

    print()
    print("-" * 140)

    print(
        f"TANGDAOWAN — SEED {seed}"
    )

    print("-" * 140)


    for tau in TAU_GRID:

        for alpha in ALPHA_GRID:

            shifted_logits = (
                ltlc_shift(

                    logits=
                        logits,

                    prior=
                        prior,

                    rarity=
                        rarity,

                    tau=
                        tau,

                    alpha=
                        alpha,

                    eps=
                        POSTHOC_EPS
                )
            )


            predictions = np.argmax(
                shifted_logits,
                axis=1
            )


            metrics = (
                classification_metrics_from_predictions(

                    labels=
                        labels,

                    predictions=
                        predictions,

                    k=
                        K
                )
            )


            group_metrics = (
                frequency_group_metrics(

                    metrics[
                        "per_class_accuracy"
                    ],

                    frequency_groups
                )
            )


            nll = (
                multiclass_nll_from_logits(

                    shifted_logits,

                    labels
                )
            )


            probabilities = (
                stable_softmax(
                    shifted_logits
                )
            )


            ece = (
                expected_calibration_error(

                    probabilities,

                    labels,

                    n_bins=
                        15
                )
            )


            GRID_ROWS.append({

                "dataset":
                    TARGET_DATASET,

                "seed":
                    int(
                        seed
                    ),

                "tau":
                    float(
                        tau
                    ),

                "alpha":
                    float(
                        alpha
                    ),

                "validation_oa":
                    float(
                        metrics[
                            "oa"
                        ]
                    ),

                "validation_aa":
                    float(
                        metrics[
                            "aa"
                        ]
                    ),

                "validation_macro_f1":
                    float(
                        metrics[
                            "macro_f1"
                        ]
                    ),

                "validation_kappa":
                    float(
                        metrics[
                            "kappa"
                        ]
                    ),

                "validation_nll":
                    float(
                        nll
                    ),

                "validation_ece":
                    float(
                        ece
                    ),

                "validation_tail_accuracy":
                    float(
                        group_metrics[
                            "tail_accuracy"
                        ]
                    ),

                "validation_medium_accuracy":
                    float(
                        group_metrics[
                            "medium_accuracy"
                        ]
                    ),

                "validation_head_accuracy":
                    float(
                        group_metrics[
                            "head_accuracy"
                        ]
                    ),

                "validation_worst_class_accuracy":
                    float(
                        group_metrics[
                            "worst_class_accuracy"
                        ]
                    ),
            })


    print(
        "Grid evaluation: "
        f"{len(TAU_GRID) * len(ALPHA_GRID)}/"
        f"{len(TAU_GRID) * len(ALPHA_GRID)} PASS"
    )


# ==================================================================================================
# 5. COMPLETE GRID TABLE
# ==================================================================================================

GRID_DF = (
    pd.DataFrame(
        GRID_ROWS
    )
)


EXPECTED_GRID_ROWS = (
    len(
        TARGET_SEEDS
    )
    *
    len(
        TAU_GRID
    )
    *
    len(
        ALPHA_GRID
    )
)


assert (
    len(
        GRID_DF
    )
    ==
    EXPECTED_GRID_ROWS
)


assert (
    len(
        GRID_DF
    )
    ==
    189
)


CE_BASELINE_DF = (
    pd.DataFrame(
        CE_BASELINE_ROWS
    )
)


assert (
    len(
        CE_BASELINE_DF
    )
    ==
    3
)


# ==================================================================================================
# 6. SELECT STANDARD LA AND LTLC-SHIFT FOR EACH SEED
# ==================================================================================================

STANDARD_LA_SELECTIONS = {}

LTLC_SHIFT_SELECTIONS = {}

SELECTION_SUMMARY_ROWS = []


print()
print("=" * 140)
print("TANGDAOWAN — VALIDATION-ONLY SELECTION RESULTS")
print("=" * 140)


for seed in TARGET_SEEDS:

    seed_df = (
        GRID_DF[
            GRID_DF[
                "seed"
            ]
            ==
            seed
        ]
        .copy()
    )


    assert (
        len(
            seed_df
        )
        ==
        63
    )


    # ==============================================================================================
    # 6A. STANDARD LA
    #
    # Alpha = 0 is exactly Component A.
    # ==============================================================================================

    la_df = (
        seed_df[
            np.isclose(
                seed_df[
                    "alpha"
                ],
                0.0,
                rtol=0.0,
                atol=GRID_TOL
            )
        ]
        .copy()
    )


    assert (
        len(
            la_df
        )
        ==
        len(
            TAU_GRID
        )
    )


    la_selected = (
        select_by_locked_shift_rule(

            la_df,

            label=(
                f"Tangdaowan seed {seed} "
                "Standard LA"
            )
        )
    )


    assert math.isclose(
        float(
            la_selected[
                "alpha"
            ]
        ),
        0.0,
        rel_tol=0.0,
        abs_tol=GRID_TOL
    )


    # ==============================================================================================
    # 6B. LTLC-SHIFT
    # ==============================================================================================

    ltlc_selected = (
        select_by_locked_shift_rule(

            seed_df,

            label=(
                f"Tangdaowan seed {seed} "
                "LTLC-Shift"
            )
        )
    )


    STANDARD_LA_SELECTIONS[
        seed
    ] = (
        la_selected
    )


    LTLC_SHIFT_SELECTIONS[
        seed
    ] = (
        ltlc_selected
    )


    ce_row = (

        CE_BASELINE_DF[
            CE_BASELINE_DF[
                "seed"
            ]
            ==
            seed
        ]

        .iloc[
            0
        ]

        .to_dict()
    )


    SELECTION_SUMMARY_ROWS.extend([

        {

            "dataset":
                TARGET_DATASET,

            "seed":
                seed,

            "method":
                "CE",

            "tau":
                0.0,

            "alpha":
                0.0,

            "validation_oa":
                ce_row[
                    "validation_oa"
                ],

            "validation_aa":
                ce_row[
                    "validation_aa"
                ],

            "validation_macro_f1":
                ce_row[
                    "validation_macro_f1"
                ],

            "validation_nll":
                ce_row[
                    "validation_nll"
                ],

            "validation_ece":
                ce_row[
                    "validation_ece"
                ],

            "validation_tail_accuracy":
                ce_row[
                    "validation_tail_accuracy"
                ],

            "validation_worst_class_accuracy":
                ce_row[
                    "validation_worst_class_accuracy"
                ],
        },

        {

            "dataset":
                TARGET_DATASET,

            "seed":
                seed,

            "method":
                "Standard_LA",

            "tau":
                float(
                    la_selected[
                        "tau"
                    ]
                ),

            "alpha":
                0.0,

            "validation_oa":
                float(
                    la_selected[
                        "validation_oa"
                    ]
                ),

            "validation_aa":
                float(
                    la_selected[
                        "validation_aa"
                    ]
                ),

            "validation_macro_f1":
                float(
                    la_selected[
                        "validation_macro_f1"
                    ]
                ),

            "validation_nll":
                float(
                    la_selected[
                        "validation_nll"
                    ]
                ),

            "validation_ece":
                float(
                    la_selected[
                        "validation_ece"
                    ]
                ),

            "validation_tail_accuracy":
                float(
                    la_selected[
                        "validation_tail_accuracy"
                    ]
                ),

            "validation_worst_class_accuracy":
                float(
                    la_selected[
                        "validation_worst_class_accuracy"
                    ]
                ),
        },

        {

            "dataset":
                TARGET_DATASET,

            "seed":
                seed,

            "method":
                "LTLC_Shift",

            "tau":
                float(
                    ltlc_selected[
                        "tau"
                    ]
                ),

            "alpha":
                float(
                    ltlc_selected[
                        "alpha"
                    ]
                ),

            "validation_oa":
                float(
                    ltlc_selected[
                        "validation_oa"
                    ]
                ),

            "validation_aa":
                float(
                    ltlc_selected[
                        "validation_aa"
                    ]
                ),

            "validation_macro_f1":
                float(
                    ltlc_selected[
                        "validation_macro_f1"
                    ]
                ),

            "validation_nll":
                float(
                    ltlc_selected[
                        "validation_nll"
                    ]
                ),

            "validation_ece":
                float(
                    ltlc_selected[
                        "validation_ece"
                    ]
                ),

            "validation_tail_accuracy":
                float(
                    ltlc_selected[
                        "validation_tail_accuracy"
                    ]
                ),

            "validation_worst_class_accuracy":
                float(
                    ltlc_selected[
                        "validation_worst_class_accuracy"
                    ]
                ),
        },
    ])


    print()
    print(
        f"Seed {seed}"
    )

    print(
        "  Standard LA | "
        f"tau={float(la_selected['tau']):.2f} | "
        f"AA={float(la_selected['validation_aa']):.6f} | "
        f"F1={float(la_selected['validation_macro_f1']):.6f} | "
        f"NLL={float(la_selected['validation_nll']):.6f}"
    )

    print(
        "  LTLC-Shift  | "
        f"tau={float(ltlc_selected['tau']):.2f} | "
        f"alpha={float(ltlc_selected['alpha']):.2f} | "
        f"AA={float(ltlc_selected['validation_aa']):.6f} | "
        f"F1={float(ltlc_selected['validation_macro_f1']):.6f} | "
        f"NLL={float(ltlc_selected['validation_nll']):.6f}"
    )


# ==================================================================================================
# 7. SELECTION SUMMARY TABLE
# ==================================================================================================

SELECTION_SUMMARY_DF = (
    pd.DataFrame(
        SELECTION_SUMMARY_ROWS
    )
)


assert (
    len(
        SELECTION_SUMMARY_DF
    )
    ==
    9
)


print()
print("=" * 140)
print("TANGDAOWAN — CE vs STANDARD LA vs LTLC-SHIFT")
print("=" * 140)


print(

    SELECTION_SUMMARY_DF[
        [
            "seed",
            "method",
            "tau",
            "alpha",
            "validation_oa",
            "validation_aa",
            "validation_macro_f1",
            "validation_tail_accuracy",
            "validation_worst_class_accuracy",
            "validation_ece",
            "validation_nll",
        ]
    ].to_string(
        index=False
    )
)


# ==================================================================================================
# 8. CLASSIFICATION-SIGNAL DIAGNOSTIC
#
# This is descriptive only.
# It does NOT alter selection.
# ==================================================================================================

SIGNAL_ROWS = []


for seed in TARGET_SEEDS:

    ce = (
        SELECTION_SUMMARY_DF[
            (
                SELECTION_SUMMARY_DF[
                    "seed"
                ]
                ==
                seed
            )
            &
            (
                SELECTION_SUMMARY_DF[
                    "method"
                ]
                ==
                "CE"
            )
        ]
        .iloc[
            0
        ]
    )


    la = (
        SELECTION_SUMMARY_DF[
            (
                SELECTION_SUMMARY_DF[
                    "seed"
                ]
                ==
                seed
            )
            &
            (
                SELECTION_SUMMARY_DF[
                    "method"
                ]
                ==
                "Standard_LA"
            )
        ]
        .iloc[
            0
        ]
    )


    ltlc = (
        SELECTION_SUMMARY_DF[
            (
                SELECTION_SUMMARY_DF[
                    "seed"
                ]
                ==
                seed
            )
            &
            (
                SELECTION_SUMMARY_DF[
                    "method"
                ]
                ==
                "LTLC_Shift"
            )
        ]
        .iloc[
            0
        ]
    )


    SIGNAL_ROWS.append({

        "seed":
            seed,

        "delta_aa_vs_ce":
            float(
                ltlc[
                    "validation_aa"
                ]
                -
                ce[
                    "validation_aa"
                ]
            ),

        "delta_aa_vs_la":
            float(
                ltlc[
                    "validation_aa"
                ]
                -
                la[
                    "validation_aa"
                ]
            ),

        "delta_macro_f1_vs_la":
            float(
                ltlc[
                    "validation_macro_f1"
                ]
                -
                la[
                    "validation_macro_f1"
                ]
            ),

        "delta_tail_vs_la":
            float(
                ltlc[
                    "validation_tail_accuracy"
                ]
                -
                la[
                    "validation_tail_accuracy"
                ]
            ),

        "delta_worst_vs_la":
            float(
                ltlc[
                    "validation_worst_class_accuracy"
                ]
                -
                la[
                    "validation_worst_class_accuracy"
                ]
            ),

        "delta_oa_vs_la":
            float(
                ltlc[
                    "validation_oa"
                ]
                -
                la[
                    "validation_oa"
                ]
            ),
    })


SIGNAL_DF = (
    pd.DataFrame(
        SIGNAL_ROWS
    )
)


print()
print("=" * 140)
print("LTLC-SHIFT CLASSIFICATION SIGNAL VS STANDARD LA")
print("=" * 140)

print(
    SIGNAL_DF.to_string(
        index=False
    )
)


# ==================================================================================================
# 9. SAVE GRID + SUMMARY
# ==================================================================================================

GRID_CSV_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_ltlc_shift_validation_grid.csv"
)


SUMMARY_CSV_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_ce_la_ltlc_shift_validation_summary.csv"
)


GRID_DF.to_csv(
    GRID_CSV_PATH,
    index=False
)


SELECTION_SUMMARY_DF.to_csv(
    SUMMARY_CSV_PATH,
    index=False
)


GRID_CSV_SHA = (
    sha256_file(
        GRID_CSV_PATH
    )
)


SUMMARY_CSV_SHA = (
    sha256_file(
        SUMMARY_CSV_PATH
    )
)


# ==================================================================================================
# 10. FREEZE TANGDAOWAN SHIFT-SELECTION MANIFEST
# ==================================================================================================

SHIFT_SELECTION_MANIFEST_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_ltlc_shift_validation_selection_manifest.json"
)


SHIFT_SELECTION_MANIFEST = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "dataset":
        TARGET_DATASET,

    "status":
        "SHIFT_PARAMETERS_FROZEN_ON_VALIDATION",

    "source_model":
        "HybridSN_CE",

    "ce_seeds":
        TARGET_SEEDS,

    "tau_grid":
        [
            float(x)
            for x in
            TAU_GRID
        ],

    "alpha_grid":
        [
            float(x)
            for x in
            ALPHA_GRID
        ],

    "grid_points_per_seed":
        int(
            len(
                TAU_GRID
            )
            *
            len(
                ALPHA_GRID
            )
        ),

    "selection_rule": [

        "highest_validation_AA",

        "tie_highest_validation_Macro_F1",

        "tie_lowest_validation_NLL",

        "tie_smaller_alpha",
    ],

    "standard_la_definition":
        "LTLC shift grid restricted to alpha=0",

    "standard_la_selection":
        {},

    "ltlc_shift_selection":
        {},

    "validation_grid_sha256":
        GRID_CSV_SHA,

    "validation_summary_sha256":
        SUMMARY_CSV_SHA,

    "implementation_spec_sha256":
        POSTHOC_SPEC_SHA,

    "training_prior_rarity_manifest_sha256":
        NOTEBOOK05_PRIOR_MANIFEST_SHA,

    "training_performed":
        False,

    "official_test_used":
        False,
}


for seed in TARGET_SEEDS:

    la = (
        STANDARD_LA_SELECTIONS[
            seed
        ]
    )


    ltlc = (
        LTLC_SHIFT_SELECTIONS[
            seed
        ]
    )


    SHIFT_SELECTION_MANIFEST[
        "standard_la_selection"
    ][
        str(seed)
    ] = {

        "tau":
            float(
                la[
                    "tau"
                ]
            ),

        "alpha":
            0.0,

        "validation_oa":
            float(
                la[
                    "validation_oa"
                ]
            ),

        "validation_aa":
            float(
                la[
                    "validation_aa"
                ]
            ),

        "validation_macro_f1":
            float(
                la[
                    "validation_macro_f1"
                ]
            ),

        "validation_nll":
            float(
                la[
                    "validation_nll"
                ]
            ),

        "validation_ece":
            float(
                la[
                    "validation_ece"
                ]
            ),

        "validation_tail_accuracy":
            float(
                la[
                    "validation_tail_accuracy"
                ]
            ),

        "validation_worst_class_accuracy":
            float(
                la[
                    "validation_worst_class_accuracy"
                ]
            ),
    }


    SHIFT_SELECTION_MANIFEST[
        "ltlc_shift_selection"
    ][
        str(seed)
    ] = {

        "tau":
            float(
                ltlc[
                    "tau"
                ]
            ),

        "alpha":
            float(
                ltlc[
                    "alpha"
                ]
            ),

        "validation_oa":
            float(
                ltlc[
                    "validation_oa"
                ]
            ),

        "validation_aa":
            float(
                ltlc[
                    "validation_aa"
                ]
            ),

        "validation_macro_f1":
            float(
                ltlc[
                    "validation_macro_f1"
                ]
            ),

        "validation_nll":
            float(
                ltlc[
                    "validation_nll"
                ]
            ),

        "validation_ece":
            float(
                ltlc[
                    "validation_ece"
                ]
            ),

        "validation_tail_accuracy":
            float(
                ltlc[
                    "validation_tail_accuracy"
                ]
            ),

        "validation_worst_class_accuracy":
            float(
                ltlc[
                    "validation_worst_class_accuracy"
                ]
            ),
    }


# ==================================================================================================
# 11. RESTART-SAFE MANIFEST WRITE
# ==================================================================================================

if (
    SHIFT_SELECTION_MANIFEST_PATH.exists()
):

    with open(
        SHIFT_SELECTION_MANIFEST_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing_manifest = (
            json.load(
                f
            )
        )


    assert (
        existing_manifest
        ==
        SHIFT_SELECTION_MANIFEST
    ), (
        "Existing Tangdaowan LTLC-Shift selection manifest "
        "differs from the current validation-only selection. STOP."
    )


else:

    with open(
        SHIFT_SELECTION_MANIFEST_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            SHIFT_SELECTION_MANIFEST,
            f,
            indent=2,
            sort_keys=True
        )


SHIFT_SELECTION_MANIFEST_SHA = (
    sha256_file(
        SHIFT_SELECTION_MANIFEST_PATH
    )
)


# ==================================================================================================
# 12. FINAL SAFETY GATE
# ==================================================================================================

assert (
    len(
        STANDARD_LA_SELECTIONS
    )
    ==
    3
)


assert (
    len(
        LTLC_SHIFT_SELECTIONS
    )
    ==
    3
)


assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False
assert NOTEBOOK05_TRAINING_PERFORMED is False


NOTEBOOK05_TANGDAOWAN_SHIFT_SELECTION_GATE_PASS = True


# ==================================================================================================
# 13. FINAL REPORT
# ==================================================================================================

print()
print("=" * 140)

print(
    "Tangdaowan CE seeds evaluated        : 3/3 PASS"
)

print(
    "LTLC shift grid points evaluated     : "
    f"{len(GRID_DF)}/{EXPECTED_GRID_ROWS} PASS"
)

print(
    "Standard LA selections              : 3/3 PASS"
)

print(
    "LTLC-Shift selections               : 3/3 PASS"
)

print()

for seed in TARGET_SEEDS:

    la = (
        STANDARD_LA_SELECTIONS[
            seed
        ]
    )

    ltlc = (
        LTLC_SHIFT_SELECTIONS[
            seed
        ]
    )


    print(
        f"Seed {seed:4d} | "
        f"LA tau={float(la['tau']):.2f} | "
        f"LTLC tau={float(ltlc['tau']):.2f}, "
        f"alpha={float(ltlc['alpha']):.2f}"
    )


print()

print(
    "Grid CSV:"
)

print(
    GRID_CSV_PATH
)

print(
    "SHA-256:"
)

print(
    GRID_CSV_SHA
)

print()

print(
    "Selection manifest:"
)

print(
    SHIFT_SELECTION_MANIFEST_PATH
)

print(
    "SHA-256:"
)

print(
    SHIFT_SELECTION_MANIFEST_SHA
)

print()

print(
    "Training performed:",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset:",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader:",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 TANGDAOWAN SHIFT-SELECTION GATE: PASS"
)

print(
    "TANGDAOWAN tau/alpha FROZEN ON VALIDATION"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 140)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 7
# TANGDAOWAN VALIDATION-ONLY TEMPERATURE FITTING
#
# Fits:
#   1. Global TS on CE
#   2. Global TS after Standard LA
#   3. Global TS after LTLC-Shift
#   4. LTLC Full rarity-conditioned temperature
#
# Then evaluates Tangdaowan development GO / REVISE gate.
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
import math
import numpy as np
import pandas as pd

from scipy.optimize import minimize


# ==================================================================================================
# 1. HARD PRECONDITION GATES
# ==================================================================================================

assert NOTEBOOK05_POSTHOC_ENGINE_GATE_PASS is True
assert NOTEBOOK05_TANGDAOWAN_SHIFT_SELECTION_GATE_PASS is True

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False
assert NOTEBOOK05_TRAINING_PERFORMED is False


DATASET = "Tangdaowan"

SEEDS = [
    42,
    123,
    3407,
]

BOUND_LOW = -2.0
BOUND_HIGH = 2.0

BOUND_TOL = 1e-6
SIGNAL_TOL = 1e-12


# ==================================================================================================
# 2. TAIL-ECE
# ==================================================================================================

def tail_ece(
    probabilities,
    labels,
    frequency_groups,
    n_bins=15
):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )

    labels = np.asarray(
        labels,
        dtype=np.int64
    )

    frequency_groups = np.asarray(
        frequency_groups,
        dtype=object
    )


    tail_class_mask = (
        frequency_groups
        ==
        "Tail"
    )


    sample_mask = (
        tail_class_mask[
            labels
        ]
    )


    assert (
        sample_mask.sum()
        >
        0
    )


    return expected_calibration_error(

        probabilities[
            sample_mask
        ],

        labels[
            sample_mask
        ],

        n_bins=n_bins
    )


# ==================================================================================================
# 3. COMPLETE METRIC HELPER
# ==================================================================================================

def evaluate_logits(
    logits,
    labels,
    frequency_groups
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    k = logits.shape[
        1
    ]


    predictions = np.argmax(
        logits,
        axis=1
    )


    classification = (
        classification_metrics_from_predictions(

            labels=
                labels,

            predictions=
                predictions,

            k=
                k
        )
    )


    group = (
        frequency_group_metrics(

            classification[
                "per_class_accuracy"
            ],

            frequency_groups
        )
    )


    probabilities = stable_softmax(
        logits
    )


    return {

        "oa":
            float(
                classification[
                    "oa"
                ]
            ),

        "aa":
            float(
                classification[
                    "aa"
                ]
            ),

        "macro_f1":
            float(
                classification[
                    "macro_f1"
                ]
            ),

        "kappa":
            float(
                classification[
                    "kappa"
                ]
            ),

        "tail_accuracy":
            float(
                group[
                    "tail_accuracy"
                ]
            ),

        "medium_accuracy":
            float(
                group[
                    "medium_accuracy"
                ]
            ),

        "head_accuracy":
            float(
                group[
                    "head_accuracy"
                ]
            ),

        "worst_class_accuracy":
            float(
                group[
                    "worst_class_accuracy"
                ]
            ),

        "nll":
            float(
                multiclass_nll_from_logits(
                    logits,
                    labels
                )
            ),

        "ece":
            float(
                expected_calibration_error(
                    probabilities,
                    labels,
                    n_bins=15
                )
            ),

        "tail_ece":
            float(
                tail_ece(
                    probabilities,
                    labels,
                    frequency_groups,
                    n_bins=15
                )
            ),

        "brier":
            float(
                multiclass_brier_score(
                    probabilities,
                    labels
                )
            ),

        "predictions":
            predictions,
    }


# ==================================================================================================
# 4. GLOBAL TEMPERATURE FIT
#
# T = exp(a)
# a in [-2, 2]
# ==================================================================================================

def fit_global_temperature(
    logits,
    labels
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    initial_nll = (
        multiclass_nll_from_logits(
            logits,
            labels
        )
    )


    def objective(
        x
    ):

        a = float(
            x[
                0
            ]
        )


        temperature = float(
            np.exp(
                a
            )
        )


        calibrated = (
            logits
            /
            temperature
        )


        return (
            multiclass_nll_from_logits(
                calibrated,
                labels
            )
        )


    result = minimize(

        objective,

        x0=np.array(
            [
                0.0
            ],
            dtype=np.float64
        ),

        method="L-BFGS-B",

        bounds=[
            (
                BOUND_LOW,
                BOUND_HIGH
            )
        ],

        options={
            "maxiter": 500,
            "ftol": 1e-15,
        }
    )


    assert np.isfinite(
        result.fun
    )


    a = float(
        result.x[
            0
        ]
    )


    temperature = float(
        np.exp(
            a
        )
    )


    calibrated_logits = (
        logits
        /
        temperature
    )


    final_nll = (
        multiclass_nll_from_logits(
            calibrated_logits,
            labels
        )
    )


    # Identity a=0 was available to optimizer.
    assert (
        final_nll
        <=
        initial_nll
        +
        1e-10
    )


    # Positive scalar temperature cannot change argmax.
    assert np.array_equal(

        np.argmax(
            logits,
            axis=1
        ),

        np.argmax(
            calibrated_logits,
            axis=1
        )
    )


    return {

        "success":
            bool(
                result.success
            ),

        "message":
            str(
                result.message
            ),

        "a":
            a,

        "temperature":
            temperature,

        "a_at_bound":
            bool(
                abs(
                    a
                    -
                    BOUND_LOW
                )
                <=
                BOUND_TOL

                or

                abs(
                    a
                    -
                    BOUND_HIGH
                )
                <=
                BOUND_TOL
            ),

        "initial_nll":
            float(
                initial_nll
            ),

        "final_nll":
            float(
                final_nll
            ),

        "calibrated_logits":
            calibrated_logits,
    }


# ==================================================================================================
# 5. LTLC RARITY-CONDITIONED TEMPERATURE FIT
#
# T(x) = exp(a + b * rarity[predicted_class])
#
# a,b in [-2,2]
# ==================================================================================================

def fit_ltlc_temperature(
    shifted_logits,
    labels,
    rarity
):

    shifted_logits = np.asarray(
        shifted_logits,
        dtype=np.float64
    )

    labels = np.asarray(
        labels,
        dtype=np.int64
    )

    rarity = np.asarray(
        rarity,
        dtype=np.float64
    )


    predicted_class = np.argmax(
        shifted_logits,
        axis=1
    )


    predicted_rarity = (
        rarity[
            predicted_class
        ]
    )


    initial_nll = (
        multiclass_nll_from_logits(
            shifted_logits,
            labels
        )
    )


    def objective(
        x
    ):

        a = float(
            x[
                0
            ]
        )

        b = float(
            x[
                1
            ]
        )


        temperatures = np.exp(

            a
            +
            b
            *
            predicted_rarity
        )


        calibrated = (

            shifted_logits
            /
            temperatures[
                :,
                None
            ]
        )


        return (
            multiclass_nll_from_logits(
                calibrated,
                labels
            )
        )


    result = minimize(

        objective,

        x0=np.array(
            [
                0.0,
                0.0
            ],
            dtype=np.float64
        ),

        method="L-BFGS-B",

        bounds=[
            (
                BOUND_LOW,
                BOUND_HIGH
            ),
            (
                BOUND_LOW,
                BOUND_HIGH
            ),
        ],

        options={
            "maxiter": 500,
            "ftol": 1e-15,
        }
    )


    assert np.isfinite(
        result.fun
    )


    a = float(
        result.x[
            0
        ]
    )

    b = float(
        result.x[
            1
        ]
    )


    temperatures = np.exp(

        a
        +
        b
        *
        predicted_rarity
    )


    assert np.isfinite(
        temperatures
    ).all()


    assert (
        temperatures
        >
        0
    ).all()


    calibrated_logits = (

        shifted_logits
        /
        temperatures[
            :,
            None
        ]
    )


    final_nll = (
        multiclass_nll_from_logits(
            calibrated_logits,
            labels
        )
    )


    assert (
        final_nll
        <=
        initial_nll
        +
        1e-10
    )


    # Mandatory order-preservation test
    assert np.array_equal(

        predicted_class,

        np.argmax(
            calibrated_logits,
            axis=1
        )
    )


    return {

        "success":
            bool(
                result.success
            ),

        "message":
            str(
                result.message
            ),

        "a":
            a,

        "b":
            b,

        "temperature_min":
            float(
                np.min(
                    temperatures
                )
            ),

        "temperature_median":
            float(
                np.median(
                    temperatures
                )
            ),

        "temperature_mean":
            float(
                np.mean(
                    temperatures
                )
            ),

        "temperature_max":
            float(
                np.max(
                    temperatures
                )
            ),

        "a_at_bound":
            bool(
                abs(
                    a
                    -
                    BOUND_LOW
                )
                <=
                BOUND_TOL

                or

                abs(
                    a
                    -
                    BOUND_HIGH
                )
                <=
                BOUND_TOL
            ),

        "b_at_bound":
            bool(
                abs(
                    b
                    -
                    BOUND_LOW
                )
                <=
                BOUND_TOL

                or

                abs(
                    b
                    -
                    BOUND_HIGH
                )
                <=
                BOUND_TOL
            ),

        "initial_nll":
            float(
                initial_nll
            ),

        "final_nll":
            float(
                final_nll
            ),

        "calibrated_logits":
            calibrated_logits,
    }


# ==================================================================================================
# 6. FIT ALL TANGDAOWAN VALIDATION CALIBRATION VARIANTS
# ==================================================================================================

RESULT_ROWS = []

PARAMETER_ROWS = []


prior = np.asarray(
    CLASS_PRIORS[
        DATASET
    ],
    dtype=np.float64
)


rarity = np.asarray(
    CLASS_RARITY[
        DATASET
    ],
    dtype=np.float64
)


frequency_groups = np.asarray(
    CLASS_FREQUENCY_GROUPS[
        DATASET
    ],
    dtype=object
)


print("=" * 145)
print("NOTEBOOK 05 — TANGDAOWAN TEMPERATURE FITTING")
print("=" * 145)


for seed in SEEDS:

    cache = (
        VALIDATION_CACHE[
            (
                DATASET,
                seed
            )
        ]
    )


    logits = np.asarray(
        cache[
            "logits"
        ],
        dtype=np.float64
    )


    labels = np.asarray(
        cache[
            "labels"
        ],
        dtype=np.int64
    )


    # ----------------------------------------------------------------------------------------------
    # Frozen Standard LA parameters
    # ----------------------------------------------------------------------------------------------

    la_selection = (
        STANDARD_LA_SELECTIONS[
            seed
        ]
    )


    la_tau = float(
        la_selection[
            "tau"
        ]
    )


    la_logits = (
        standard_posthoc_la(

            logits=
                logits,

            prior=
                prior,

            tau=
                la_tau
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Frozen LTLC-Shift parameters
    # ----------------------------------------------------------------------------------------------

    shift_selection = (
        LTLC_SHIFT_SELECTIONS[
            seed
        ]
    )


    shift_tau = float(
        shift_selection[
            "tau"
        ]
    )


    shift_alpha = float(
        shift_selection[
            "alpha"
        ]
    )


    shifted_logits = (
        ltlc_shift(

            logits=
                logits,

            prior=
                prior,

            rarity=
                rarity,

            tau=
                shift_tau,

            alpha=
                shift_alpha
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Temperature fitting
    # ----------------------------------------------------------------------------------------------

    ce_global_ts = (
        fit_global_temperature(
            logits,
            labels
        )
    )


    la_global_ts = (
        fit_global_temperature(
            la_logits,
            labels
        )
    )


    shift_global_ts = (
        fit_global_temperature(
            shifted_logits,
            labels
        )
    )


    ltlc_full_fit = (
        fit_ltlc_temperature(

            shifted_logits=
                shifted_logits,

            labels=
                labels,

            rarity=
                rarity
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Evaluation variants
    # ----------------------------------------------------------------------------------------------

    variants = {

        "CE":
            logits,

        "Standard_LA":
            la_logits,

        "Global_TS":
            ce_global_ts[
                "calibrated_logits"
            ],

        "LA_Global_TS":
            la_global_ts[
                "calibrated_logits"
            ],

        "LTLC_Shift":
            shifted_logits,

        "LTLC_Shift_Global_TS":
            shift_global_ts[
                "calibrated_logits"
            ],

        "LTLC_Full":
            ltlc_full_fit[
                "calibrated_logits"
            ],
    }


    base_predictions = np.argmax(
        shifted_logits,
        axis=1
    )


    # Component C MUST preserve LTLC-Shift hard predictions.
    assert np.array_equal(

        base_predictions,

        np.argmax(
            ltlc_full_fit[
                "calibrated_logits"
            ],
            axis=1
        )
    )


    for method, method_logits in variants.items():

        metrics = (
            evaluate_logits(

                logits=
                    method_logits,

                labels=
                    labels,

                frequency_groups=
                    frequency_groups
            )
        )


        RESULT_ROWS.append({

            "dataset":
                DATASET,

            "seed":
                int(
                    seed
                ),

            "method":
                method,

            "tau":
                (
                    shift_tau
                    if method.startswith(
                        "LTLC"
                    )
                    else
                    la_tau
                    if method.startswith(
                        "Standard_LA"
                    )
                    or
                    method.startswith(
                        "LA_Global"
                    )
                    else
                    0.0
                ),

            "alpha":
                (
                    shift_alpha
                    if method.startswith(
                        "LTLC"
                    )
                    else
                    0.0
                ),

            "oa":
                metrics[
                    "oa"
                ],

            "aa":
                metrics[
                    "aa"
                ],

            "macro_f1":
                metrics[
                    "macro_f1"
                ],

            "kappa":
                metrics[
                    "kappa"
                ],

            "tail_accuracy":
                metrics[
                    "tail_accuracy"
                ],

            "medium_accuracy":
                metrics[
                    "medium_accuracy"
                ],

            "head_accuracy":
                metrics[
                    "head_accuracy"
                ],

            "worst_class_accuracy":
                metrics[
                    "worst_class_accuracy"
                ],

            "ece":
                metrics[
                    "ece"
                ],

            "tail_ece":
                metrics[
                    "tail_ece"
                ],

            "nll":
                metrics[
                    "nll"
                ],

            "brier":
                metrics[
                    "brier"
                ],
        })


    PARAMETER_ROWS.extend([

        {
            "dataset":
                DATASET,

            "seed":
                seed,

            "method":
                "Global_TS",

            "tau":
                0.0,

            "alpha":
                0.0,

            "a":
                ce_global_ts[
                    "a"
                ],

            "b":
                0.0,

            "temperature_min":
                ce_global_ts[
                    "temperature"
                ],

            "temperature_median":
                ce_global_ts[
                    "temperature"
                ],

            "temperature_mean":
                ce_global_ts[
                    "temperature"
                ],

            "temperature_max":
                ce_global_ts[
                    "temperature"
                ],

            "a_at_bound":
                ce_global_ts[
                    "a_at_bound"
                ],

            "b_at_bound":
                False,

            "optimizer_success":
                ce_global_ts[
                    "success"
                ],
        },

        {
            "dataset":
                DATASET,

            "seed":
                seed,

            "method":
                "LA_Global_TS",

            "tau":
                la_tau,

            "alpha":
                0.0,

            "a":
                la_global_ts[
                    "a"
                ],

            "b":
                0.0,

            "temperature_min":
                la_global_ts[
                    "temperature"
                ],

            "temperature_median":
                la_global_ts[
                    "temperature"
                ],

            "temperature_mean":
                la_global_ts[
                    "temperature"
                ],

            "temperature_max":
                la_global_ts[
                    "temperature"
                ],

            "a_at_bound":
                la_global_ts[
                    "a_at_bound"
                ],

            "b_at_bound":
                False,

            "optimizer_success":
                la_global_ts[
                    "success"
                ],
        },

        {
            "dataset":
                DATASET,

            "seed":
                seed,

            "method":
                "LTLC_Shift_Global_TS",

            "tau":
                shift_tau,

            "alpha":
                shift_alpha,

            "a":
                shift_global_ts[
                    "a"
                ],

            "b":
                0.0,

            "temperature_min":
                shift_global_ts[
                    "temperature"
                ],

            "temperature_median":
                shift_global_ts[
                    "temperature"
                ],

            "temperature_mean":
                shift_global_ts[
                    "temperature"
                ],

            "temperature_max":
                shift_global_ts[
                    "temperature"
                ],

            "a_at_bound":
                shift_global_ts[
                    "a_at_bound"
                ],

            "b_at_bound":
                False,

            "optimizer_success":
                shift_global_ts[
                    "success"
                ],
        },

        {
            "dataset":
                DATASET,

            "seed":
                seed,

            "method":
                "LTLC_Full",

            "tau":
                shift_tau,

            "alpha":
                shift_alpha,

            "a":
                ltlc_full_fit[
                    "a"
                ],

            "b":
                ltlc_full_fit[
                    "b"
                ],

            "temperature_min":
                ltlc_full_fit[
                    "temperature_min"
                ],

            "temperature_median":
                ltlc_full_fit[
                    "temperature_median"
                ],

            "temperature_mean":
                ltlc_full_fit[
                    "temperature_mean"
                ],

            "temperature_max":
                ltlc_full_fit[
                    "temperature_max"
                ],

            "a_at_bound":
                ltlc_full_fit[
                    "a_at_bound"
                ],

            "b_at_bound":
                ltlc_full_fit[
                    "b_at_bound"
                ],

            "optimizer_success":
                ltlc_full_fit[
                    "success"
                ],
        },
    ])


    print()
    print("-" * 145)

    print(
        f"Seed {seed}"
    )

    print("-" * 145)

    print(
        f"Frozen LA             : tau={la_tau:.2f}"
    )

    print(
        f"Frozen LTLC-Shift     : "
        f"tau={shift_tau:.2f}, alpha={shift_alpha:.2f}"
    )

    print(
        f"Global TS             : "
        f"a={ce_global_ts['a']:.6f}, "
        f"T={ce_global_ts['temperature']:.6f}"
    )

    print(
        f"LTLC Full             : "
        f"a={ltlc_full_fit['a']:.6f}, "
        f"b={ltlc_full_fit['b']:.6f}"
    )

    print(
        "LTLC temperature      : "
        f"min={ltlc_full_fit['temperature_min']:.6f}, "
        f"median={ltlc_full_fit['temperature_median']:.6f}, "
        f"mean={ltlc_full_fit['temperature_mean']:.6f}, "
        f"max={ltlc_full_fit['temperature_max']:.6f}"
    )

    print(
        "Bound diagnostics     : "
        f"a_at_bound={ltlc_full_fit['a_at_bound']}, "
        f"b_at_bound={ltlc_full_fit['b_at_bound']}"
    )


# ==================================================================================================
# 7. TABLES
# ==================================================================================================

TEMPERATURE_RESULT_DF = (
    pd.DataFrame(
        RESULT_ROWS
    )
)


CALIBRATION_PARAMETER_DF = (
    pd.DataFrame(
        PARAMETER_ROWS
    )
)


assert (
    len(
        TEMPERATURE_RESULT_DF
    )
    ==
    21
)


assert (
    len(
        CALIBRATION_PARAMETER_DF
    )
    ==
    12
)


print()
print("=" * 145)
print("TANGDAOWAN VALIDATION CALIBRATION RESULTS")
print("=" * 145)


print(

    TEMPERATURE_RESULT_DF[
        [
            "seed",
            "method",
            "tau",
            "alpha",
            "oa",
            "aa",
            "macro_f1",
            "tail_accuracy",
            "worst_class_accuracy",
            "ece",
            "tail_ece",
            "nll",
            "brier",
        ]
    ].to_string(
        index=False
    )
)


# ==================================================================================================
# 8. MEAN RESULTS BY METHOD
# ==================================================================================================

MEAN_RESULT_DF = (

    TEMPERATURE_RESULT_DF

    .groupby(
        "method",
        as_index=False
    )[
        [
            "oa",
            "aa",
            "macro_f1",
            "tail_accuracy",
            "worst_class_accuracy",
            "ece",
            "tail_ece",
            "nll",
            "brier",
        ]
    ]

    .mean()
)


print()
print("=" * 145)
print("TANGDAOWAN 3-SEED VALIDATION MEANS")
print("=" * 145)

print(
    MEAN_RESULT_DF.to_string(
        index=False
    )
)


# ==================================================================================================
# 9. DEVELOPMENT SIGNAL CHECK
#
# Gate comparison:
#
# classification:
#   LTLC-Shift vs Standard LA
#
# calibration:
#   LTLC Full vs Standard LA
# ==================================================================================================

def get_mean_row(
    method
):

    rows = (
        MEAN_RESULT_DF[
            MEAN_RESULT_DF[
                "method"
            ]
            ==
            method
        ]
    )


    assert (
        len(
            rows
        )
        ==
        1
    )


    return (
        rows.iloc[
            0
        ]
    )


LA_MEAN = get_mean_row(
    "Standard_LA"
)


SHIFT_MEAN = get_mean_row(
    "LTLC_Shift"
)


FULL_MEAN = get_mean_row(
    "LTLC_Full"
)


CLASSIFICATION_DELTAS = {

    "AA":
        float(
            SHIFT_MEAN[
                "aa"
            ]
            -
            LA_MEAN[
                "aa"
            ]
        ),

    "Macro_F1":
        float(
            SHIFT_MEAN[
                "macro_f1"
            ]
            -
            LA_MEAN[
                "macro_f1"
            ]
        ),

    "Tail_Accuracy":
        float(
            SHIFT_MEAN[
                "tail_accuracy"
            ]
            -
            LA_MEAN[
                "tail_accuracy"
            ]
        ),

    "Worst_Class_Accuracy":
        float(
            SHIFT_MEAN[
                "worst_class_accuracy"
            ]
            -
            LA_MEAN[
                "worst_class_accuracy"
            ]
        ),

    "OA":
        float(
            SHIFT_MEAN[
                "oa"
            ]
            -
            LA_MEAN[
                "oa"
            ]
        ),
}


CALIBRATION_IMPROVEMENTS = {

    "ECE":
        float(
            LA_MEAN[
                "ece"
            ]
            -
            FULL_MEAN[
                "ece"
            ]
        ),

    "Tail_ECE":
        float(
            LA_MEAN[
                "tail_ece"
            ]
            -
            FULL_MEAN[
                "tail_ece"
            ]
        ),

    "NLL":
        float(
            LA_MEAN[
                "nll"
            ]
            -
            FULL_MEAN[
                "nll"
            ]
        ),

    "Brier":
        float(
            LA_MEAN[
                "brier"
            ]
            -
            FULL_MEAN[
                "brier"
            ]
        ),
}


CLASSIFICATION_SIGNAL = any(

    CLASSIFICATION_DELTAS[
        key
    ]
    >
    SIGNAL_TOL

    for key
    in [
        "AA",
        "Macro_F1",
        "Tail_Accuracy",
        "Worst_Class_Accuracy",
    ]
)


CALIBRATION_SIGNAL = any(

    value
    >
    SIGNAL_TOL

    for value
    in CALIBRATION_IMPROVEMENTS.values()
)


# No large OA collapse test.
OA_COLLAPSE = (
    CLASSIFICATION_DELTAS[
        "OA"
    ]
    <
    -0.01
)


TANGDAOWAN_GO = bool(

    CLASSIFICATION_SIGNAL

    and

    CALIBRATION_SIGNAL

    and

    not OA_COLLAPSE
)


print()
print("=" * 145)
print("TANGDAOWAN LTLC DEVELOPMENT SIGNAL")
print("=" * 145)

print(
    "Classification deltas "
    "(LTLC-Shift minus Standard LA):"
)

for key, value in (
    CLASSIFICATION_DELTAS.items()
):

    print(
        f"  {key:22s}: {value:+.12f}"
    )


print()

print(
    "Calibration improvements "
    "(Standard LA minus LTLC Full; positive is better):"
)

for key, value in (
    CALIBRATION_IMPROVEMENTS.items()
):

    print(
        f"  {key:22s}: {value:+.12f}"
    )


print()

print(
    "Classification signal:",
    CLASSIFICATION_SIGNAL
)

print(
    "Calibration signal   :",
    CALIBRATION_SIGNAL
)

print(
    "Large OA collapse     :",
    OA_COLLAPSE
)

print()


if TANGDAOWAN_GO:

    print(
        "TANGDAOWAN DEVELOPMENT DECISION: GO"
    )

    print(
        "LTLC may proceed to Pingan and Qingyun."
    )

else:

    print(
        "TANGDAOWAN DEVELOPMENT DECISION: REVISE"
    )

    print(
        "Do NOT expand LTLC tuning to Pingan/Qingyun yet."
    )


print("=" * 145)


# ==================================================================================================
# 10. SAVE VALIDATION RESULTS
# ==================================================================================================

RESULT_CSV_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_temperature_validation_results.csv"
)


PARAMETER_CSV_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_temperature_validation_parameters.csv"
)


TEMPERATURE_RESULT_DF.to_csv(
    RESULT_CSV_PATH,
    index=False
)


CALIBRATION_PARAMETER_DF.to_csv(
    PARAMETER_CSV_PATH,
    index=False
)


RESULT_CSV_SHA = (
    sha256_file(
        RESULT_CSV_PATH
    )
)


PARAMETER_CSV_SHA = (
    sha256_file(
        PARAMETER_CSV_PATH
    )
)


# ==================================================================================================
# 11. SAVE DEVELOPMENT-GATE MANIFEST
# ==================================================================================================

DEVELOPMENT_GATE_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_ltlc_development_gate.json"
)


DEVELOPMENT_GATE_OBJECT = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "dataset":
        "Tangdaowan",

    "status":
        (
            "GO"
            if TANGDAOWAN_GO
            else
            "REVISE"
        ),

    "classification_signal":
        bool(
            CLASSIFICATION_SIGNAL
        ),

    "calibration_signal":
        bool(
            CALIBRATION_SIGNAL
        ),

    "large_oa_collapse":
        bool(
            OA_COLLAPSE
        ),

    "classification_deltas_ltlc_shift_minus_standard_la":
        {

            key:
                float(
                    value
                )

            for key, value
            in CLASSIFICATION_DELTAS.items()
        },

    "calibration_improvements_standard_la_minus_ltlc_full":
        {

            key:
                float(
                    value
                )

            for key, value
            in CALIBRATION_IMPROVEMENTS.items()
        },

    "temperature_bounds":
        {
            "a":
                [
                    -2.0,
                    2.0
                ],

            "b":
                [
                    -2.0,
                    2.0
                ],
        },

    "temperature_validation_results_sha256":
        RESULT_CSV_SHA,

    "temperature_parameter_sha256":
        PARAMETER_CSV_SHA,

    "shift_selection_manifest_sha256":
        SHIFT_SELECTION_MANIFEST_SHA,

    "implementation_spec_sha256":
        POSTHOC_SPEC_SHA,

    "training_performed":
        False,

    "official_test_used":
        False,
}


# Restart-safe
if DEVELOPMENT_GATE_PATH.exists():

    with open(
        DEVELOPMENT_GATE_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing = json.load(
            f
        )


    assert (
        existing
        ==
        DEVELOPMENT_GATE_OBJECT
    ), (
        "Existing Tangdaowan development gate "
        "differs from current validation result."
    )


else:

    with open(
        DEVELOPMENT_GATE_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            DEVELOPMENT_GATE_OBJECT,
            f,
            indent=2,
            sort_keys=True
        )


DEVELOPMENT_GATE_SHA = (
    sha256_file(
        DEVELOPMENT_GATE_PATH
    )
)


# ==================================================================================================
# 12. FINAL SAFETY REPORT
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False
assert NOTEBOOK05_TRAINING_PERFORMED is False


NOTEBOOK05_TANGDAOWAN_TEMPERATURE_GATE_COMPLETE = True


print()
print("=" * 145)

print(
    "Temperature fits completed          : 3/3 seeds PASS"
)

print(
    "LTLC Full argmax preservation       : 3/3 PASS"
)

print(
    "Temperature diagnostics saved       : PASS"
)

print(
    "Bound diagnostics saved             : PASS"
)

print()

print(
    "Result CSV SHA:"
)

print(
    RESULT_CSV_SHA
)

print()

print(
    "Parameter CSV SHA:"
)

print(
    PARAMETER_CSV_SHA
)

print()

print(
    "Development gate manifest:"
)

print(
    DEVELOPMENT_GATE_PATH
)

print(
    "SHA-256:"
)

print(
    DEVELOPMENT_GATE_SHA
)

print()

print(
    "Training performed:",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset:",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader:",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 TANGDAOWAN TEMPERATURE/SANITY AUDIT: COMPLETE"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 8
# CE CHECKPOINT-PURPOSE / BEST-VAL-AA AVAILABILITY AUDIT
#
# PURPOSE:
#   v3.1 requires:
#
#       Table I         -> best validation OA
#       Long-tail work  -> best validation AA
#
# Before revising LTLC after the Tangdaowan REVISE result,
# verify whether the required best-val-AA CE artifacts already exist.
#
# READ ONLY
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST ACCESS
# ==================================================================================================

from pathlib import Path
import json
import re
import pandas as pd


# --------------------------------------------------------------------------------------------------
# 1. ROOT
# --------------------------------------------------------------------------------------------------

CE_ROOT = (
    ROOT /
    "runs" /
    "hybridsn_ce_baseline"
)

assert CE_ROOT.exists()


print("=" * 135)
print("NOTEBOOK 05 — CE CHECKPOINT-PURPOSE AUDIT")
print("=" * 135)

print(
    "CE root:",
    CE_ROOT
)


# --------------------------------------------------------------------------------------------------
# 2. INVENTORY ALL CHECKPOINT-LIKE FILES
#
# Explicitly exclude:
#   _pilot
#   official_test
# --------------------------------------------------------------------------------------------------

checkpoint_files = sorted([

    p

    for p in CE_ROOT.rglob(
        "*.pt"
    )

    if (
        "_pilot"
        not in str(
            p
        ).lower()
    )

    and

    (
        "official_test"
        not in str(
            p
        ).lower()
    )
])


print()
print("-" * 135)
print("CHECKPOINT FILES")
print("-" * 135)

print(
    "Checkpoint count:",
    len(
        checkpoint_files
    )
)


for p in checkpoint_files:

    lower_name = (
        p.name.lower()
    )


    if (
        "valaa"
        in lower_name
        or
        "val_aa"
        in lower_name
    ):

        purpose = (
            "BEST_VAL_AA"
        )


    elif (
        "valoa"
        in lower_name
        or
        "val_oa"
        in lower_name
    ):

        purpose = (
            "BEST_VAL_OA"
        )


    else:

        purpose = (
            "UNSPECIFIED"
        )


    print(
        f"{purpose:14s} | "
        f"{p.relative_to(ROOT)}"
    )


# --------------------------------------------------------------------------------------------------
# 3. COUNT PURPOSES
# --------------------------------------------------------------------------------------------------

best_aa_checkpoint_files = [

    p

    for p in checkpoint_files

    if (
        "valaa"
        in p.name.lower()
        or
        "val_aa"
        in p.name.lower()
    )
]


best_oa_checkpoint_files = [

    p

    for p in checkpoint_files

    if (
        "valoa"
        in p.name.lower()
        or
        "val_oa"
        in p.name.lower()
    )
]


print()
print(
    "Best-val-AA checkpoint files:",
    len(
        best_aa_checkpoint_files
    )
)

print(
    "Best-val-OA checkpoint files:",
    len(
        best_oa_checkpoint_files
    )
)


# --------------------------------------------------------------------------------------------------
# 4. INVENTORY VALIDATION LOGIT / OUTPUT DIRECTORIES
#
# NO official_test directory is opened.
# --------------------------------------------------------------------------------------------------

validation_artifacts = []


for p in CE_ROOT.rglob(
    "*"
):

    if (
        not p.is_file()
    ):

        continue


    path_lower = str(
        p
    ).lower()


    if (
        "official_test"
        in path_lower
    ):

        continue


    name_lower = (
        p.name.lower()
    )


    if (
        "validation"
        in name_lower
        or
        "valaa"
        in name_lower
        or
        "val_aa"
        in name_lower
        or
        "valoa"
        in name_lower
        or
        "val_oa"
        in name_lower
    ):

        validation_artifacts.append(
            p
        )


validation_artifacts = sorted(
    set(
        validation_artifacts
    )
)


print()
print("-" * 135)
print("VALIDATION-SIDE ARTIFACTS")
print("-" * 135)

print(
    "Files found:",
    len(
        validation_artifacts
    )
)


for p in validation_artifacts:

    print(
        p.relative_to(
            ROOT
        )
    )


# --------------------------------------------------------------------------------------------------
# 5. INVENTORY JSON METADATA / MANIFESTS
#
# Read JSON only outside official_test.
# Search recursively for checkpoint-selection fields.
# --------------------------------------------------------------------------------------------------

json_files = sorted([

    p

    for p in CE_ROOT.rglob(
        "*.json"
    )

    if (
        "official_test"
        not in str(
            p
        ).lower()
    )
])


METADATA_FINDINGS = []


interesting_tokens = [

    "best",
    "checkpoint",
    "epoch",
    "validation",
    "val_aa",
    "val_oa",
    "selection",
    "metric",
    "purpose",
]


def flatten_json(
    obj,
    prefix=""
):

    items = {}


    if isinstance(
        obj,
        dict
    ):

        for key, value in (
            obj.items()
        ):

            new_prefix = (
                f"{prefix}.{key}"
                if prefix
                else str(
                    key
                )
            )


            items.update(
                flatten_json(
                    value,
                    new_prefix
                )
            )


    elif isinstance(
        obj,
        list
    ):

        # Do not explode large lists.
        if (
            len(
                obj
            )
            <=
            20
        ):

            for i, value in enumerate(
                obj
            ):

                items.update(
                    flatten_json(
                        value,
                        f"{prefix}[{i}]"
                    )
                )


    else:

        items[
            prefix
        ] = obj


    return items


print()
print("-" * 135)
print("METADATA / MANIFEST CHECKPOINT-SELECTION FINDINGS")
print("-" * 135)

print(
    "JSON files inspected:",
    len(
        json_files
    )
)


for path in json_files:

    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:

            obj = json.load(
                f
            )


    except Exception:

        continue


    flat = flatten_json(
        obj
    )


    matched = {}


    for key, value in (
        flat.items()
    ):

        key_lower = (
            key.lower()
        )


        if any(

            token
            in key_lower

            for token in
            interesting_tokens
        ):

            if isinstance(
                value,
                (
                    str,
                    int,
                    float,
                    bool
                )
            ):

                matched[
                    key
                ] = value


    if matched:

        METADATA_FINDINGS.append({

            "path":
                path,

            "fields":
                matched,
        })


        print()
        print(
            path.relative_to(
                ROOT
            )
        )


        for key, value in (
            matched.items()
        ):

            if any(

                token
                in key.lower()

                for token in [

                    "best",
                    "checkpoint",
                    "selection",
                    "metric",
                    "val_aa",
                    "val_oa",
                ]
            ):

                print(
                    f"  {key}: {value}"
                )


# --------------------------------------------------------------------------------------------------
# 6. SEARCH HISTORY CSV FILES
#
# Histories may allow us to identify whether a separate best-AA epoch
# existed even if only one checkpoint file was saved.
# --------------------------------------------------------------------------------------------------

history_csv_files = sorted([

    p

    for p in CE_ROOT.rglob(
        "*.csv"
    )

    if (
        "official_test"
        not in str(
            p
        ).lower()
    )

    and

    (
        "history"
        in p.name.lower()
        or
        "train"
        in p.name.lower()
    )
])


HISTORY_ROWS = []


print()
print("-" * 135)
print("CE HISTORY AUDIT")
print("-" * 135)

print(
    "History CSV files found:",
    len(
        history_csv_files
    )
)


for path in history_csv_files:

    try:

        df = pd.read_csv(
            path
        )

    except Exception:

        continue


    lower_columns = {

        str(
            column
        ).lower():
            column

        for column in
        df.columns
    }


    aa_column = None
    oa_column = None
    epoch_column = None


    for candidate in [

        "val_aa",
        "validation_aa",
        "aa",
    ]:

        if candidate in lower_columns:

            aa_column = (
                lower_columns[
                    candidate
                ]
            )

            break


    for candidate in [

        "val_oa",
        "validation_oa",
        "oa",
    ]:

        if candidate in lower_columns:

            oa_column = (
                lower_columns[
                    candidate
                ]
            )

            break


    for candidate in [

        "epoch",
        "epochs",
    ]:

        if candidate in lower_columns:

            epoch_column = (
                lower_columns[
                    candidate
                ]
            )

            break


    if (
        aa_column
        is None
        or
        oa_column
        is None
        or
        epoch_column
        is None
    ):

        print(
            "SCHEMA UNKNOWN |",
            path.relative_to(
                ROOT
            ),
            "| columns =",
            list(
                df.columns
            )
        )

        continue


    best_aa_row = (
        df.loc[
            df[
                aa_column
            ].astype(
                float
            ).idxmax()
        ]
    )


    best_oa_row = (
        df.loc[
            df[
                oa_column
            ].astype(
                float
            ).idxmax()
        ]
    )


    best_aa_epoch = int(
        best_aa_row[
            epoch_column
        ]
    )


    best_oa_epoch = int(
        best_oa_row[
            epoch_column
        ]
    )


    same_epoch = bool(
        best_aa_epoch
        ==
        best_oa_epoch
    )


    HISTORY_ROWS.append({

        "path":
            str(
                path.relative_to(
                    ROOT
                )
            ),

        "best_aa_epoch":
            best_aa_epoch,

        "best_oa_epoch":
            best_oa_epoch,

        "same_epoch":
            same_epoch,

        "best_aa":
            float(
                best_aa_row[
                    aa_column
                ]
            ),

        "best_oa":
            float(
                best_oa_row[
                    oa_column
                ]
            ),
    })


    print(
        f"{path.name:55s} | "
        f"best-AA epoch={best_aa_epoch:3d} | "
        f"best-OA epoch={best_oa_epoch:3d} | "
        f"same={'YES' if same_epoch else 'NO'}"
    )


# --------------------------------------------------------------------------------------------------
# 7. SUMMARY
# --------------------------------------------------------------------------------------------------

HISTORY_AUDIT_DF = (
    pd.DataFrame(
        HISTORY_ROWS
    )
)


different_epoch_count = (

    int(
        (
            ~HISTORY_AUDIT_DF[
                "same_epoch"
            ]
        ).sum()
    )

    if len(
        HISTORY_AUDIT_DF
    )
    >
    0

    else None
)


BEST_VAL_AA_ARTIFACT_AVAILABLE = bool(

    len(
        best_aa_checkpoint_files
    )
    >
    0
)


print()
print("=" * 135)
print("CE CHECKPOINT-PURPOSE AUDIT SUMMARY")
print("=" * 135)

print(
    "Canonical best-val-OA checkpoints found :",
    len(
        best_oa_checkpoint_files
    )
)

print(
    "Explicit best-val-AA checkpoints found  :",
    len(
        best_aa_checkpoint_files
    )
)

print(
    "CE history files successfully audited   :",
    len(
        HISTORY_AUDIT_DF
    )
)

print(
    "Histories where best-AA != best-OA epoch:",
    different_epoch_count
)

print()

print(
    "v3.1 long-tail checkpoint requirement:"
)

print(
    "BEST VALIDATION AA"
)

print()

print(
    "Explicit required checkpoint available:",
    BEST_VAL_AA_ARTIFACT_AVAILABLE
)

print()

print(
    "Training performed:",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official-test dataset:",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader:",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 CE CHECKPOINT-PURPOSE AUDIT: COMPLETE"
)

print(
    "DO NOT REVISE LTLC FORMULATION UNTIL THIS AUDIT IS REVIEWED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 135)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 9
# TANGDAOWAN VALIDATION-SATURATION / LTLC-SHIFT HEADROOM DIAGNOSTIC
#
# PURPOSE:
#   Diagnose the Tangdaowan REVISE result before changing
#   the locked LTLC formulation.
#
# CPU ONLY
# READ / COMPUTE ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
import numpy as np
import pandas as pd


# ==================================================================================================
# 1. PRECONDITION GATES
# ==================================================================================================

assert NOTEBOOK05_TANGDAOWAN_SHIFT_SELECTION_GATE_PASS is True
assert NOTEBOOK05_TANGDAOWAN_TEMPERATURE_GATE_COMPLETE is True

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False
assert NOTEBOOK05_TRAINING_PERFORMED is False


DATASET = "Tangdaowan"

SEEDS = [
    42,
    123,
    3407,
]

DIAG_TOL = 1e-12


prior = np.asarray(
    CLASS_PRIORS[
        DATASET
    ],
    dtype=np.float64
)


rarity = np.asarray(
    CLASS_RARITY[
        DATASET
    ],
    dtype=np.float64
)


frequency_groups = np.asarray(
    CLASS_FREQUENCY_GROUPS[
        DATASET
    ],
    dtype=object
)


K = len(
    prior
)


assert K == 16


# ==================================================================================================
# 2. DIAGNOSTIC STORAGE
# ==================================================================================================

DIAGNOSTIC_ROWS = []

BASELINE_ROWS = []

PER_CLASS_ERROR_ROWS = []


print("=" * 145)
print("NOTEBOOK 05 — TANGDAOWAN VALIDATION-SATURATION DIAGNOSTIC")
print("=" * 145)


# ==================================================================================================
# 3. SEED-BY-SEED BASELINE ERROR ANALYSIS
# ==================================================================================================

for seed in SEEDS:

    cache = (
        VALIDATION_CACHE[
            (
                DATASET,
                seed
            )
        ]
    )


    logits = np.asarray(
        cache[
            "logits"
        ],
        dtype=np.float64
    )


    labels = np.asarray(
        cache[
            "labels"
        ],
        dtype=np.int64
    )


    ce_predictions = np.argmax(
        logits,
        axis=1
    )


    ce_correct = (
        ce_predictions
        ==
        labels
    )


    ce_wrong = (
        ~ce_correct
    )


    total_errors = int(
        ce_wrong.sum()
    )


    # --------------------------------------------------------------------------------------------------
    # Baseline classification metrics
    # --------------------------------------------------------------------------------------------------

    ce_eval = evaluate_logits(

        logits=
            logits,

        labels=
            labels,

        frequency_groups=
            frequency_groups
    )


    # --------------------------------------------------------------------------------------------------
    # Error counts by true class
    # --------------------------------------------------------------------------------------------------

    support = np.bincount(
        labels,
        minlength=K
    )


    error_counts = np.bincount(
        labels[
            ce_wrong
        ],
        minlength=K
    )


    tail_error_count = 0
    medium_error_count = 0
    head_error_count = 0


    for class_index in range(
        K
    ):

        group = str(
            frequency_groups[
                class_index
            ]
        )


        errors = int(
            error_counts[
                class_index
            ]
        )


        if group == "Tail":

            tail_error_count += errors


        elif group == "Medium":

            medium_error_count += errors


        elif group == "Head":

            head_error_count += errors


        PER_CLASS_ERROR_ROWS.append({

            "seed":
                seed,

            "internal_class":
                class_index,

            "public_class":
                class_index + 1,

            "frequency_group":
                group,

            "validation_support":
                int(
                    support[
                        class_index
                    ]
                ),

            "validation_errors":
                errors,

            "validation_accuracy":
                float(
                    1.0
                    -
                    (
                        errors
                        /
                        support[
                            class_index
                        ]
                    )
                ),
        })


    BASELINE_ROWS.append({

        "seed":
            seed,

        "samples":
            int(
                len(
                    labels
                )
            ),

        "total_errors":
            total_errors,

        "tail_errors":
            tail_error_count,

        "medium_errors":
            medium_error_count,

        "head_errors":
            head_error_count,

        "oa":
            ce_eval[
                "oa"
            ],

        "aa":
            ce_eval[
                "aa"
            ],

        "macro_f1":
            ce_eval[
                "macro_f1"
            ],

        "tail_accuracy":
            ce_eval[
                "tail_accuracy"
            ],

        "worst_class_accuracy":
            ce_eval[
                "worst_class_accuracy"
            ],
    })


    print()
    print("-" * 145)

    print(
        f"TANGDAOWAN — SEED {seed}"
    )

    print("-" * 145)

    print(
        f"Validation samples : {len(labels)}"
    )

    print(
        f"Total CE errors    : {total_errors}"
    )

    print(
        f"Tail errors        : {tail_error_count}"
    )

    print(
        f"Medium errors      : {medium_error_count}"
    )

    print(
        f"Head errors        : {head_error_count}"
    )

    print()

    print(
        f"CE AA              : {ce_eval['aa']:.9f}"
    )

    print(
        f"CE Macro-F1        : {ce_eval['macro_f1']:.9f}"
    )

    print(
        f"CE Tail Accuracy   : {ce_eval['tail_accuracy']:.9f}"
    )

    print(
        f"CE Worst-Class Acc : {ce_eval['worst_class_accuracy']:.9f}"
    )


    wrong_classes = np.where(
        error_counts
        >
        0
    )[0]


    print()

    print(
        "Classes containing validation errors:"
    )


    if len(
        wrong_classes
    ) == 0:

        print(
            "  NONE"
        )


    else:

        for class_index in wrong_classes:

            print(
                f"  public={class_index + 1:2d} | "
                f"group={frequency_groups[class_index]:6s} | "
                f"errors={int(error_counts[class_index]):2d} | "
                f"support={int(support[class_index]):4d}"
            )


    # ==================================================================================================
    # 4. RE-EVALUATE EVERY NONZERO SHIFT
    #
    # This is DIAGNOSTIC ONLY.
    # It does NOT replace the frozen selection from Cell 6.
    # ==================================================================================================

    for tau in TAU_GRID:

        for alpha in ALPHA_GRID:

            # tau=0 is always the identity regardless of alpha.
            if np.isclose(
                tau,
                0.0,
                rtol=0.0,
                atol=DIAG_TOL
            ):

                continue


            shifted_logits = ltlc_shift(

                logits=
                    logits,

                prior=
                    prior,

                rarity=
                    rarity,

                tau=
                    tau,

                alpha=
                    alpha
            )


            shifted_predictions = np.argmax(
                shifted_logits,
                axis=1
            )


            shifted_correct = (
                shifted_predictions
                ==
                labels
            )


            changed = (
                shifted_predictions
                !=
                ce_predictions
            )


            # CE wrong -> shifted correct
            corrected_errors = int(

                (
                    ce_wrong
                    &
                    shifted_correct
                ).sum()
            )


            # CE correct -> shifted wrong
            introduced_errors = int(

                (
                    ce_correct
                    &
                    (~shifted_correct)
                ).sum()
            )


            # Wrong under both, but predicted class changed
            wrong_to_wrong_changes = int(

                (
                    ce_wrong
                    &
                    (~shifted_correct)
                    &
                    changed
                ).sum()
            )


            shift_eval = evaluate_logits(

                logits=
                    shifted_logits,

                labels=
                    labels,

                frequency_groups=
                    frequency_groups
            )


            DIAGNOSTIC_ROWS.append({

                "seed":
                    seed,

                "tau":
                    float(
                        tau
                    ),

                "alpha":
                    float(
                        alpha
                    ),

                "adaptive_ltlc":
                    bool(
                        alpha
                        >
                        0.0
                    ),

                "changed_predictions":
                    int(
                        changed.sum()
                    ),

                "corrected_errors":
                    corrected_errors,

                "introduced_errors":
                    introduced_errors,

                "wrong_to_wrong_changes":
                    wrong_to_wrong_changes,

                "net_error_change":
                    int(
                        introduced_errors
                        -
                        corrected_errors
                    ),

                "aa":
                    shift_eval[
                        "aa"
                    ],

                "macro_f1":
                    shift_eval[
                        "macro_f1"
                    ],

                "tail_accuracy":
                    shift_eval[
                        "tail_accuracy"
                    ],

                "worst_class_accuracy":
                    shift_eval[
                        "worst_class_accuracy"
                    ],

                "oa":
                    shift_eval[
                        "oa"
                    ],

                "nll":
                    shift_eval[
                        "nll"
                    ],

                "delta_aa_vs_ce":
                    float(
                        shift_eval[
                            "aa"
                        ]
                        -
                        ce_eval[
                            "aa"
                        ]
                    ),

                "delta_macro_f1_vs_ce":
                    float(
                        shift_eval[
                            "macro_f1"
                        ]
                        -
                        ce_eval[
                            "macro_f1"
                        ]
                    ),

                "delta_tail_vs_ce":
                    float(
                        shift_eval[
                            "tail_accuracy"
                        ]
                        -
                        ce_eval[
                            "tail_accuracy"
                        ]
                    ),

                "delta_worst_vs_ce":
                    float(
                        shift_eval[
                            "worst_class_accuracy"
                        ]
                        -
                        ce_eval[
                            "worst_class_accuracy"
                        ]
                    ),

                "delta_oa_vs_ce":
                    float(
                        shift_eval[
                            "oa"
                        ]
                        -
                        ce_eval[
                            "oa"
                        ]
                    ),
            })


# ==================================================================================================
# 5. TABLES
# ==================================================================================================

BASELINE_DIAGNOSTIC_DF = pd.DataFrame(
    BASELINE_ROWS
)


PER_CLASS_ERROR_DF = pd.DataFrame(
    PER_CLASS_ERROR_ROWS
)


SHIFT_DIAGNOSTIC_DF = pd.DataFrame(
    DIAGNOSTIC_ROWS
)


assert len(
    BASELINE_DIAGNOSTIC_DF
) == 3


# tau has 8 nonzero values × 7 alpha × 3 seeds
assert len(
    SHIFT_DIAGNOSTIC_DF
) == (
    8
    *
    7
    *
    3
)


# ==================================================================================================
# 6. ADAPTIVE-LTLC-ONLY SIGNAL
#
# alpha > 0 and tau > 0
# ==================================================================================================

ADAPTIVE_DF = (
    SHIFT_DIAGNOSTIC_DF[
        SHIFT_DIAGNOSTIC_DF[
            "adaptive_ltlc"
        ]
        ==
        True
    ]
    .copy()
)


assert len(
    ADAPTIVE_DF
) == (
    8
    *
    6
    *
    3
)


signal_columns = [

    "delta_aa_vs_ce",

    "delta_macro_f1_vs_ce",

    "delta_tail_vs_ce",

    "delta_worst_vs_ce",
]


MAX_ADAPTIVE_SIGNAL = {

    column:
        float(
            ADAPTIVE_DF[
                column
            ].max()
        )

    for column
    in signal_columns
}


ANY_ADAPTIVE_CLASSIFICATION_GAIN = any(

    value
    >
    DIAG_TOL

    for value
    in MAX_ADAPTIVE_SIGNAL.values()
)


# ==================================================================================================
# 7. VALIDATION-SATURATION CHECK
# ==================================================================================================

ALL_SEEDS_TAIL_PERFECT = bool(

    np.allclose(

        BASELINE_DIAGNOSTIC_DF[
            "tail_accuracy"
        ].to_numpy(
            dtype=np.float64
        ),

        1.0,

        rtol=0.0,
        atol=DIAG_TOL
    )
)


ALL_SEEDS_ZERO_TAIL_ERRORS = bool(

    (
        BASELINE_DIAGNOSTIC_DF[
            "tail_errors"
        ]
        ==
        0
    ).all()
)


MAX_ERRORS_PER_SEED = int(

    BASELINE_DIAGNOSTIC_DF[
        "total_errors"
    ].max()
)


VALIDATION_SATURATED = bool(

    ALL_SEEDS_TAIL_PERFECT

    and

    ALL_SEEDS_ZERO_TAIL_ERRORS

    and

    MAX_ERRORS_PER_SEED
    <=
    3
)


# ==================================================================================================
# 8. BEST NONZERO ADAPTIVE CANDIDATE — DIAGNOSTIC ONLY
#
# Uses the original locked ranking for descriptive inspection.
# DOES NOT replace Cell 6 selection.
# ==================================================================================================

BEST_NONZERO_ROWS = []


for seed in SEEDS:

    seed_df = (
        ADAPTIVE_DF[
            ADAPTIVE_DF[
                "seed"
            ]
            ==
            seed
        ]
        .copy()
    )


    seed_df = seed_df.sort_values(

        by=[
            "aa",
            "macro_f1",
            "nll",
            "alpha",
        ],

        ascending=[
            False,
            False,
            True,
            True,
        ],

        kind="mergesort"
    )


    BEST_NONZERO_ROWS.append(
        seed_df.iloc[
            0
        ].to_dict()
    )


BEST_NONZERO_DF = pd.DataFrame(
    BEST_NONZERO_ROWS
)


print()
print("=" * 145)
print("BEST NONZERO ADAPTIVE LTLC CANDIDATES — DIAGNOSTIC ONLY")
print("=" * 145)


print(

    BEST_NONZERO_DF[
        [
            "seed",
            "tau",
            "alpha",
            "changed_predictions",
            "corrected_errors",
            "introduced_errors",
            "aa",
            "macro_f1",
            "tail_accuracy",
            "worst_class_accuracy",
            "delta_aa_vs_ce",
            "delta_macro_f1_vs_ce",
            "delta_tail_vs_ce",
            "delta_worst_vs_ce",
        ]
    ].to_string(
        index=False
    )
)


# ==================================================================================================
# 9. OVERALL DIAGNOSIS
# ==================================================================================================

print()
print("=" * 145)
print("TANGDAOWAN VALIDATION HEADROOM DIAGNOSIS")
print("=" * 145)


print(
    "CE validation errors by seed:"
)

for _, row in (
    BASELINE_DIAGNOSTIC_DF.iterrows()
):

    print(
        f"  seed {int(row['seed']):4d} | "
        f"errors={int(row['total_errors'])} | "
        f"tail_errors={int(row['tail_errors'])} | "
        f"tail_acc={row['tail_accuracy']:.6f}"
    )


print()

print(
    "Maximum adaptive-LTLC classification deltas "
    "over ALL tau>0, alpha>0 candidates:"
)


for key, value in (
    MAX_ADAPTIVE_SIGNAL.items()
):

    print(
        f"  {key:28s}: {value:+.12f}"
    )


print()

print(
    "All seeds have perfect Tail Accuracy :",
    ALL_SEEDS_TAIL_PERFECT
)

print(
    "All seeds have zero Tail errors      :",
    ALL_SEEDS_ZERO_TAIL_ERRORS
)

print(
    "Maximum CE errors in any seed        :",
    MAX_ERRORS_PER_SEED
)

print(
    "Any adaptive classification gain     :",
    ANY_ADAPTIVE_CLASSIFICATION_GAIN
)

print(
    "Validation saturation diagnostic     :",
    VALIDATION_SATURATED
)


if VALIDATION_SATURATED:

    print()

    print(
        "DIAGNOSIS: TANGDAOWAN VALIDATION CLASSIFICATION IS SATURATED."
    )

    print(
        "The frozen validation split provides essentially no "
        "tail-class decision headroom for LTLC-Shift."
    )


print("=" * 145)


# ==================================================================================================
# 10. SAVE DIAGNOSTIC ARTIFACTS
# ==================================================================================================

SHIFT_DIAGNOSTIC_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_validation_saturation_shift_diagnostic.csv"
)


PER_CLASS_ERROR_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_validation_saturation_per_class_errors.csv"
)


DIAGNOSTIC_MANIFEST_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "tangdaowan_validation_saturation_diagnostic.json"
)


SHIFT_DIAGNOSTIC_DF.to_csv(
    SHIFT_DIAGNOSTIC_PATH,
    index=False
)


PER_CLASS_ERROR_DF.to_csv(
    PER_CLASS_ERROR_PATH,
    index=False
)


SHIFT_DIAGNOSTIC_SHA = sha256_file(
    SHIFT_DIAGNOSTIC_PATH
)


PER_CLASS_ERROR_SHA = sha256_file(
    PER_CLASS_ERROR_PATH
)


DIAGNOSTIC_MANIFEST = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "dataset":
        "Tangdaowan",

    "purpose":
        "diagnose_REVISE_before_method_change",

    "best_aa_equals_best_oa_epoch_for_all_tangdaowan_ce_seeds":
        True,

    "ce_validation_errors_by_seed":
        {

            str(
                int(
                    row[
                        "seed"
                    ]
                )
            ):
                int(
                    row[
                        "total_errors"
                    ]
                )

            for _, row
            in BASELINE_DIAGNOSTIC_DF.iterrows()
        },

    "tail_errors_by_seed":
        {

            str(
                int(
                    row[
                        "seed"
                    ]
                )
            ):
                int(
                    row[
                        "tail_errors"
                    ]
                )

            for _, row
            in BASELINE_DIAGNOSTIC_DF.iterrows()
        },

    "all_seeds_tail_accuracy_one":
        ALL_SEEDS_TAIL_PERFECT,

    "all_seeds_zero_tail_errors":
        ALL_SEEDS_ZERO_TAIL_ERRORS,

    "max_errors_per_seed":
        MAX_ERRORS_PER_SEED,

    "any_nonzero_adaptive_classification_gain":
        ANY_ADAPTIVE_CLASSIFICATION_GAIN,

    "maximum_adaptive_classification_deltas":
        MAX_ADAPTIVE_SIGNAL,

    "validation_saturated":
        VALIDATION_SATURATED,

    "frozen_cell6_selection_changed":
        False,

    "ltlc_formula_changed":
        False,

    "training_performed":
        False,

    "official_test_used":
        False,

    "shift_diagnostic_csv_sha256":
        SHIFT_DIAGNOSTIC_SHA,

    "per_class_error_csv_sha256":
        PER_CLASS_ERROR_SHA,
}


if DIAGNOSTIC_MANIFEST_PATH.exists():

    with open(
        DIAGNOSTIC_MANIFEST_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing = json.load(
            f
        )


    assert (
        existing
        ==
        DIAGNOSTIC_MANIFEST
    ), (
        "Existing saturation diagnostic differs. STOP."
    )


else:

    with open(
        DIAGNOSTIC_MANIFEST_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            DIAGNOSTIC_MANIFEST,
            f,
            indent=2,
            sort_keys=True
        )


DIAGNOSTIC_MANIFEST_SHA = sha256_file(
    DIAGNOSTIC_MANIFEST_PATH
)


# ==================================================================================================
# 11. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False
assert NOTEBOOK05_TRAINING_PERFORMED is False


NOTEBOOK05_TANGDAOWAN_SATURATION_DIAGNOSTIC_PASS = True


print()
print("=" * 145)

print(
    "Saturation diagnostic                 : COMPLETE"
)

print(
    "Frozen Cell 6 selection altered       : False"
)

print(
    "LTLC formulation altered              : False"
)

print(
    "Training performed                    :",
    NOTEBOOK05_TRAINING_PERFORMED
)

print(
    "Official test used                    :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "Diagnostic manifest SHA:"
)

print(
    DIAGNOSTIC_MANIFEST_SHA
)

print()

print(
    "NOTEBOOK 05 TANGDAOWAN SATURATION DIAGNOSTIC: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 11
# T4 RECOVERY STARTUP + FROZEN AMENDMENT RELOAD
#
# PURPOSE:
#   Restart Notebook 05 after switching runtime to T4.
#   Reload the frozen saturation amendment from Google Drive.
#   Verify the exact 5 missing best-val-AA CE recovery targets.
#
# IMPORTANT:
#   - NO TRAINING IN THIS CELL
#   - NO OFFICIAL TEST
#   - DO NOT RERUN NOTEBOOK 03
# ==================================================================================================

from google.colab import drive

from pathlib import Path
import hashlib
import json
import platform

import pandas as pd
import torch


# ==================================================================================================
# 1. MOUNT GOOGLE DRIVE
# ==================================================================================================

drive.mount(
    "/content/drive"
)


# ==================================================================================================
# 2. PROJECT ROOT
# ==================================================================================================

ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)

assert ROOT.exists(), (
    f"Project root missing:\n{ROOT}"
)


AUDIT_DIR = (
    ROOT /
    "audit"
)

RUNS_DIR = (
    ROOT /
    "runs"
)

DATA_CACHE_DIR = (
    ROOT /
    "data_cache"
)

SPLIT_DIR = (
    ROOT /
    "split_indices"
)


for path in [
    AUDIT_DIR,
    RUNS_DIR,
    DATA_CACHE_DIR,
    SPLIT_DIR,
]:

    assert path.exists(), (
        f"Required directory missing:\n{path}"
    )


# ==================================================================================================
# 3. SHA-256 HELPER
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


# ==================================================================================================
# 4. T4 GPU GATE
# ==================================================================================================

assert torch.cuda.is_available(), (
    "CUDA is not available. STOP."
)


GPU_NAME = (
    torch.cuda.get_device_name(
        0
    )
)


GPU_MEMORY_GB = (
    torch.cuda.get_device_properties(
        0
    ).total_memory
    /
    (
        1024 ** 3
    )
)


assert (
    "T4"
    in
    GPU_NAME.upper()
), (
    f"Expected T4 GPU, got: {GPU_NAME}"
)


DEVICE = torch.device(
    "cuda:0"
)


# ==================================================================================================
# 5. NOTEBOOK 05 AUDIT DIRECTORY
# ==================================================================================================

NOTEBOOK05_AUDIT_DIR = (
    AUDIT_DIR /
    "notebook05_posthoc_ltlc"
)

assert NOTEBOOK05_AUDIT_DIR.exists()


# ==================================================================================================
# 6. RELOAD FROZEN SATURATION AMENDMENT
# ==================================================================================================

AMENDMENT_PATH = (
    NOTEBOOK05_AUDIT_DIR /
    "notebook05_tangdaowan_saturation_protocol_amendment.json"
)


EXPECTED_AMENDMENT_SHA = (
    "12ebdc1a2b158f4aba9f6775d671c22b"
    "8f22565e24c57a9b138f89ed9931b88a"
)


assert AMENDMENT_PATH.exists(), (
    "Frozen Notebook 05 amendment artifact missing."
)


ACTUAL_AMENDMENT_SHA = (
    sha256_file(
        AMENDMENT_PATH
    )
)


assert (
    ACTUAL_AMENDMENT_SHA
    ==
    EXPECTED_AMENDMENT_SHA
), (
    "Notebook 05 amendment SHA mismatch. STOP."
)


with open(
    AMENDMENT_PATH,
    "r",
    encoding="utf-8"
) as f:

    AMENDMENT = json.load(
        f
    )


assert (
    AMENDMENT[
        "status"
    ]
    ==
    "VALIDATION_ONLY_AMENDMENT_FROZEN"
)


assert (
    AMENDMENT[
        "required_recovery_run_count"
    ]
    ==
    5
)


assert (
    AMENDMENT[
        "official_test_used"
    ]
    is False
)


RECOVERY_RUNS = (
    AMENDMENT[
        "required_recovery_runs"
    ]
)


# ==================================================================================================
# 7. EXACT EXPECTED RECOVERY TARGETS
# ==================================================================================================

EXPECTED_RECOVERY_TARGETS = {

    (
        "Pingan",
        123
    ):
        97,

    (
        "Pingan",
        3407
    ):
        100,

    (
        "Qingyun",
        42
    ):
        100,

    (
        "Qingyun",
        123
    ):
        65,

    (
        "Qingyun",
        3407
    ):
        85,
}


ACTUAL_RECOVERY_TARGETS = {

    (
        run[
            "dataset"
        ],
        int(
            run[
                "seed"
            ]
        )
    ):
        int(
            run[
                "best_aa_epoch"
            ]
        )

    for run
    in RECOVERY_RUNS
}


assert (
    ACTUAL_RECOVERY_TARGETS
    ==
    EXPECTED_RECOVERY_TARGETS
), (
    "Frozen recovery target list mismatch."
)


# ==================================================================================================
# 8. NOTEBOOK 03 CE ROOT
# ==================================================================================================

CE_ROOT = (
    RUNS_DIR /
    "hybridsn_ce_baseline"
)


CE_CHECKPOINT_DIR = (
    CE_ROOT /
    "checkpoints"
)


CE_HISTORY_DIR = (
    CE_ROOT /
    "history"
)


CE_METRICS_DIR = (
    CE_ROOT /
    "metrics"
)


assert CE_ROOT.exists()
assert CE_CHECKPOINT_DIR.exists()


# ==================================================================================================
# 9. LOCATE EXACT HISTORY FILES
# ==================================================================================================

RECOVERY_HISTORY = {}


for (
    dataset,
    seed
), target_epoch in (
    EXPECTED_RECOVERY_TARGETS.items()
):

    slug = (
        dataset.lower()
    )


    matches = sorted(
        CE_ROOT.rglob(
            f"{slug}_seed{seed}_history.csv"
        )
    )


    assert (
        len(
            matches
        )
        ==
        1
    ), (
        f"{dataset} seed {seed}: "
        f"expected exactly 1 history CSV, found {len(matches)}."
    )


    history_path = (
        matches[
            0
        ]
    )


    history_df = pd.read_csv(
        history_path
    )


    assert (
        "epoch"
        in
        history_df.columns
    )


    assert (
        target_epoch
        in
        set(
            history_df[
                "epoch"
            ].astype(
                int
            )
        )
    ), (
        f"{dataset} seed {seed}: target epoch "
        f"{target_epoch} absent from history."
    )


    RECOVERY_HISTORY[
        (
            dataset,
            seed
        )
    ] = (
        history_path
    )


# ==================================================================================================
# 10. EXISTING BEST-OA CHECKPOINTS MUST REMAIN PRESENT
# ==================================================================================================

EXISTING_BEST_OA = {}


for (
    dataset,
    seed
), target_epoch in (
    EXPECTED_RECOVERY_TARGETS.items()
):

    slug = (
        dataset.lower()
    )


    checkpoint_path = (
        CE_CHECKPOINT_DIR /
        f"{slug}_seed{seed}_best_valOA.pt"
    )


    assert checkpoint_path.exists(), (
        f"Existing best-OA checkpoint missing:\n{checkpoint_path}"
    )


    EXISTING_BEST_OA[
        (
            dataset,
            seed
        )
    ] = (
        checkpoint_path
    )


# ==================================================================================================
# 11. DEFINE NEW RECOVERY OUTPUT DIRECTORY
#
# Existing Notebook 03 artifacts are NOT overwritten.
# ==================================================================================================

BEST_AA_RECOVERY_ROOT = (
    RUNS_DIR /
    "hybridsn_ce_best_valAA_recovery"
)


BEST_AA_CHECKPOINT_DIR = (
    BEST_AA_RECOVERY_ROOT /
    "checkpoints"
)


BEST_AA_HISTORY_DIR = (
    BEST_AA_RECOVERY_ROOT /
    "history"
)


BEST_AA_METADATA_DIR = (
    BEST_AA_RECOVERY_ROOT /
    "metadata"
)


BEST_AA_VALIDATION_LOGIT_DIR = (
    BEST_AA_RECOVERY_ROOT /
    "logits" /
    "validation"
)


for path in [

    BEST_AA_CHECKPOINT_DIR,

    BEST_AA_HISTORY_DIR,

    BEST_AA_METADATA_DIR,

    BEST_AA_VALIDATION_LOGIT_DIR,

]:

    path.mkdir(
        parents=True,
        exist_ok=True
    )


# ==================================================================================================
# 12. OFFICIAL-TEST SAFETY FLAGS
# ==================================================================================================

NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED = False

NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED = False

NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED = False

NOTEBOOK05_RECOVERY_TRAINING_STARTED = False


# ==================================================================================================
# 13. STARTUP REPORT
# ==================================================================================================

print("=" * 140)
print("NOTEBOOK 05 — T4 BEST-VAL-AA RECOVERY STARTUP")
print("=" * 140)

print(
    "Python:",
    platform.python_version()
)

print(
    "PyTorch:",
    torch.__version__
)

print(
    "CUDA available:",
    torch.cuda.is_available()
)

print(
    "CUDA runtime:",
    torch.version.cuda
)

print(
    "GPU:",
    GPU_NAME
)

print(
    f"GPU memory: {GPU_MEMORY_GB:.2f} GB"
)

print()

print(
    "Frozen amendment SHA:"
)

print(
    ACTUAL_AMENDMENT_SHA
)

print()

print(
    "Recovery runs required:",
    len(
        EXPECTED_RECOVERY_TARGETS
    )
)

print()


for (
    dataset,
    seed
), target_epoch in (
    EXPECTED_RECOVERY_TARGETS.items()
):

    history_path = (
        RECOVERY_HISTORY[
            (
                dataset,
                seed
            )
        ]
    )


    oa_path = (
        EXISTING_BEST_OA[
            (
                dataset,
                seed
            )
        ]
    )


    print(
        f"RECOVER | "
        f"{dataset:10s} | "
        f"seed={seed:4d} | "
        f"target best-AA epoch={target_epoch:3d}"
    )

    print(
        "         history:",
        history_path.relative_to(
            ROOT
        )
    )

    print(
        "         existing best-OA:",
        oa_path.relative_to(
            ROOT
        )
    )


print()
print(
    "Recovery output root:"
)

print(
    BEST_AA_RECOVERY_ROOT
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official-test dataset created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 T4 RECOVERY STARTUP GATE: PASS"
)

print(
    "READY TO BUILD EXACT NOTEBOOK 03 TRAINING RECOVERY ENVIRONMENT"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 140)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 12
# NOTEBOOK 03 RECOVERY-SOURCE + ENVIRONMENT AUDIT
#
# PURPOSE:
#   Before retraining any missing best-val-AA checkpoint:
#   - inspect frozen Notebook 03 audit/config artifacts
#   - verify target epochs from original histories
#   - locate canonical PCA caches
#   - locate train/validation split artifacts
#   - record current T4 software environment
#
# READ ONLY
# NO TRAINING
# NO OFFICIAL TEST ACCESS
# ==================================================================================================

from pathlib import Path
import json
import hashlib
import platform

import numpy as np
import pandas as pd
import torch


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert torch.cuda.is_available()
assert "T4" in torch.cuda.get_device_name(0).upper()

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


# ==================================================================================================
# 2. NOTEBOOK 03 AUDIT ROOT
# ==================================================================================================

NB03_AUDIT_ROOT = (
    AUDIT_DIR /
    "notebook03_hybridsn"
)

assert NB03_AUDIT_ROOT.exists(), (
    f"Notebook 03 audit root missing:\n{NB03_AUDIT_ROOT}"
)


print("=" * 145)
print("NOTEBOOK 05 — NOTEBOOK 03 RECOVERY-SOURCE AUDIT")
print("=" * 145)

print(
    "Notebook 03 audit root:",
    NB03_AUDIT_ROOT
)


# ==================================================================================================
# 3. INVENTORY NOTEBOOK 03 AUDIT FILES
# ==================================================================================================

nb03_files = sorted([
    p
    for p in NB03_AUDIT_ROOT.rglob("*")
    if p.is_file()
])


print()
print("-" * 145)
print("NOTEBOOK 03 AUDIT FILE INVENTORY")
print("-" * 145)

print(
    "Files found:",
    len(nb03_files)
)


for p in nb03_files:

    print(
        p.relative_to(
            ROOT
        )
    )


assert len(nb03_files) > 0


# ==================================================================================================
# 4. INSPECT JSON AUDIT / CONFIG FILES
#
# We print only compact protocol/environment fields.
# ==================================================================================================

json_files = sorted([
    p
    for p in nb03_files
    if p.suffix.lower() == ".json"
])


INTERESTING_TOKENS = [

    "python",
    "torch",
    "cuda",
    "cudnn",

    "seed",
    "determin",

    "optimizer",
    "adam",
    "learning_rate",
    "lr",
    "decay",

    "batch",
    "epoch",

    "checkpoint",
    "selection",

    "patch",
    "pca",

    "hybridsn",
    "conv",
    "dropout",

    "train",
    "validation",
]


def flatten_json_small(
    obj,
    prefix=""
):

    result = {}


    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            new_key = (
                f"{prefix}.{key}"
                if prefix
                else str(key)
            )

            result.update(
                flatten_json_small(
                    value,
                    new_key
                )
            )


    elif isinstance(
        obj,
        list
    ):

        # Keep only small scalar lists.
        if (
            len(obj) <= 30
            and
            all(
                isinstance(
                    x,
                    (
                        str,
                        int,
                        float,
                        bool,
                        type(None)
                    )
                )
                for x in obj
            )
        ):

            result[prefix] = obj


    elif isinstance(
        obj,
        (
            str,
            int,
            float,
            bool
        )
    ):

        result[prefix] = obj


    return result


JSON_FINDINGS = []


print()
print("-" * 145)
print("NOTEBOOK 03 JSON PROTOCOL / ENVIRONMENT FINDINGS")
print("-" * 145)

print(
    "JSON files:",
    len(json_files)
)


for path in json_files:

    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:

            obj = json.load(f)

    except Exception:

        continue


    flat = flatten_json_small(
        obj
    )


    matched = {}


    for key, value in flat.items():

        key_lower = key.lower()

        if any(
            token in key_lower
            for token in INTERESTING_TOKENS
        ):

            matched[key] = value


    if matched:

        JSON_FINDINGS.append({
            "path":
                path,
            "fields":
                matched,
        })


        print()
        print(
            path.relative_to(
                ROOT
            )
        )


        shown = 0

        for key, value in matched.items():

            print(
                f"  {key}: {value}"
            )

            shown += 1

            if shown >= 40:

                print(
                    "  ... additional fields omitted ..."
                )

                break


# ==================================================================================================
# 5. VERIFY ORIGINAL HISTORY TARGET EPOCHS
# ==================================================================================================

TARGET_HISTORY_ROWS = []


print()
print("-" * 145)
print("RECOVERY TARGET HISTORY VERIFICATION")
print("-" * 145)


for (
    dataset,
    seed
), target_epoch in (
    EXPECTED_RECOVERY_TARGETS.items()
):

    history_path = (
        RECOVERY_HISTORY[
            (
                dataset,
                seed
            )
        ]
    )


    df = pd.read_csv(
        history_path
    )


    assert "epoch" in df.columns


    row = (
        df[
            df[
                "epoch"
            ].astype(int)
            ==
            target_epoch
        ]
    )


    assert len(row) == 1


    row = row.iloc[0]


    # Detect likely metric-column names robustly.
    lower_cols = {
        str(c).lower():
            c
        for c in df.columns
    }


    def get_optional_value(
        candidates
    ):

        for candidate in candidates:

            if candidate in lower_cols:

                return float(
                    row[
                        lower_cols[
                            candidate
                        ]
                    ]
                )

        return None


    val_aa = get_optional_value(
        [
            "val_aa",
            "validation_aa",
            "aa",
        ]
    )


    val_oa = get_optional_value(
        [
            "val_oa",
            "validation_oa",
            "oa",
        ]
    )


    val_f1 = get_optional_value(
        [
            "val_macro_f1",
            "validation_macro_f1",
            "macro_f1",
        ]
    )


    val_loss = get_optional_value(
        [
            "val_loss",
            "validation_loss",
            "loss",
        ]
    )


    TARGET_HISTORY_ROWS.append({

        "dataset":
            dataset,

        "seed":
            seed,

        "target_epoch":
            target_epoch,

        "validation_aa":
            val_aa,

        "validation_oa":
            val_oa,

        "validation_macro_f1":
            val_f1,

        "validation_loss":
            val_loss,

        "history_path":
            str(
                history_path.relative_to(
                    ROOT
                )
            ),
    })


    print(
        f"{dataset:10s} | "
        f"seed={seed:4d} | "
        f"target_epoch={target_epoch:3d} | "
        f"val_AA={val_aa} | "
        f"val_OA={val_oa} | "
        f"val_F1={val_f1} | "
        f"val_loss={val_loss}"
    )


TARGET_HISTORY_DF = pd.DataFrame(
    TARGET_HISTORY_ROWS
)


assert len(
    TARGET_HISTORY_DF
) == 5


# ==================================================================================================
# 6. CANONICAL PCA CACHE VERIFICATION
#
# Only Pingan + Qingyun are needed for recovery.
# ==================================================================================================

EXPECTED_PCA_CACHE = {

    "Pingan":
        (
            DATA_CACHE_DIR /
            "pca_normalized_cubes" /
            "pingan_pca15_rowminmax_float32.npy"
        ),

    "Qingyun":
        (
            DATA_CACHE_DIR /
            "pca_normalized_cubes" /
            "qingyun_pca15_rowminmax_float32.npy"
        ),
}


PCA_CACHE_INFO = {}


print()
print("-" * 145)
print("CANONICAL PCA CACHE VERIFICATION")
print("-" * 145)


for dataset, path in (
    EXPECTED_PCA_CACHE.items()
):

    assert path.exists(), (
        f"{dataset}: PCA cache missing:\n{path}"
    )


    arr = np.load(
        path,
        mmap_mode="r"
    )


    assert arr.dtype == np.float32
    assert arr.ndim == 3
    assert arr.shape[2] == 15


    PCA_CACHE_INFO[
        dataset
    ] = {

        "path":
            path,

        "shape":
            tuple(
                int(x)
                for x in arr.shape
            ),

        "dtype":
            str(
                arr.dtype
            ),

        "sha256":
            sha256_file(
                path
            ),
    }


    print(
        f"{dataset:10s} | "
        f"shape={arr.shape} | "
        f"dtype={arr.dtype} | "
        f"SHA={PCA_CACHE_INFO[dataset]['sha256'][:16]}..."
    )


# ==================================================================================================
# 7. TRAIN / VALIDATION SPLIT ARTIFACT INVENTORY
#
# IMPORTANT:
#   We inventory only filenames containing train or validation/val.
#   Any file containing 'test' is explicitly excluded.
# ==================================================================================================

split_files = sorted([
    p
    for p in SPLIT_DIR.rglob("*")
    if (
        p.is_file()
        and
        "test" not in p.name.lower()
        and
        (
            "train" in p.name.lower()
            or
            "validation" in p.name.lower()
            or
            "_val" in p.name.lower()
            or
            "val_" in p.name.lower()
        )
    )
])


print()
print("-" * 145)
print("TRAIN / VALIDATION SPLIT ARTIFACT INVENTORY")
print("-" * 145)

print(
    "Eligible split files found:",
    len(split_files)
)


for path in split_files:

    print(
        path.relative_to(
            ROOT
        )
    )


assert len(
    split_files
) > 0


# ==================================================================================================
# 8. CONFIG DIRECTORY INVENTORY
# ==================================================================================================

CONFIG_DIR = (
    ROOT /
    "configs"
)


config_files = []


if CONFIG_DIR.exists():

    config_files = sorted([
        p
        for p in CONFIG_DIR.rglob("*")
        if p.is_file()
    ])


print()
print("-" * 145)
print("PROJECT CONFIG INVENTORY")
print("-" * 145)

print(
    "Config files found:",
    len(
        config_files
    )
)


for path in config_files:

    print(
        path.relative_to(
            ROOT
        )
    )


# ==================================================================================================
# 9. CURRENT RECOVERY ENVIRONMENT
# ==================================================================================================

CURRENT_ENVIRONMENT = {

    "python":
        platform.python_version(),

    "pytorch":
        torch.__version__,

    "cuda_runtime":
        torch.version.cuda,

    "cudnn":
        (
            int(
                torch.backends.cudnn.version()
            )
            if torch.backends.cudnn.is_available()
            else None
        ),

    "gpu":
        torch.cuda.get_device_name(
            0
        ),

    "gpu_memory_gb":
        float(
            torch.cuda.get_device_properties(
                0
            ).total_memory
            /
            (
                1024 ** 3
            )
        ),
}


print()
print("-" * 145)
print("CURRENT RECOVERY ENVIRONMENT")
print("-" * 145)


for key, value in (
    CURRENT_ENVIRONMENT.items()
):

    print(
        f"{key:20s}: {value}"
    )


# ==================================================================================================
# 10. WRITE READ-ONLY RECOVERY-SOURCE AUDIT
# ==================================================================================================

RECOVERY_AUDIT_DIR = (
    NOTEBOOK05_AUDIT_DIR /
    "best_valAA_recovery"
)


RECOVERY_AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


RECOVERY_SOURCE_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR /
    "notebook05_best_valAA_recovery_source_audit.json"
)


RECOVERY_SOURCE_AUDIT = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "purpose":
        "best_validation_AA_CE_checkpoint_recovery",

    "recovery_run_count":
        5,

    "recovery_targets":
        TARGET_HISTORY_ROWS,

    "current_environment":
        CURRENT_ENVIRONMENT,

    "notebook03_audit_root":
        str(
            NB03_AUDIT_ROOT.relative_to(
                ROOT
            )
        ),

    "notebook03_audit_file_count":
        len(
            nb03_files
        ),

    "notebook03_json_file_count":
        len(
            json_files
        ),

    "pca_caches":
        {

            dataset: {

                "path":
                    str(
                        info[
                            "path"
                        ].relative_to(
                            ROOT
                        )
                    ),

                "shape":
                    list(
                        info[
                            "shape"
                        ]
                    ),

                "dtype":
                    info[
                        "dtype"
                    ],

                "sha256":
                    info[
                        "sha256"
                    ],
            }

            for dataset, info in (
                PCA_CACHE_INFO.items()
            )
        },

    "train_validation_split_files":
        [
            str(
                p.relative_to(
                    ROOT
                )
            )
            for p in split_files
        ],

    "official_test_accessed":
        False,

    "training_started":
        False,
}


with open(
    RECOVERY_SOURCE_AUDIT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        RECOVERY_SOURCE_AUDIT,
        f,
        indent=2,
        sort_keys=True
    )


RECOVERY_SOURCE_AUDIT_SHA = (
    sha256_file(
        RECOVERY_SOURCE_AUDIT_PATH
    )
)


# ==================================================================================================
# 11. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_RECOVERY_SOURCE_AUDIT_PASS = True


print()
print("=" * 145)

print(
    "Notebook 03 audit source located       : PASS"
)

print(
    "Five target history epochs verified    : 5/5 PASS"
)

print(
    "Pingan/Qingyun PCA caches verified     : 2/2 PASS"
)

print(
    "Train/validation split inventory       : PASS"
)

print(
    "Current T4 environment recorded        : PASS"
)

print()

print(
    "Recovery source audit:"
)

print(
    RECOVERY_SOURCE_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    RECOVERY_SOURCE_AUDIT_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 RECOVERY-SOURCE AUDIT GATE: PASS"
)

print(
    "NO RECOVERY TRAINING HAS STARTED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 12A
# SPLIT-INDICES DIRECTORY NAME AUDIT
#
# METADATA / FILENAMES ONLY
# DOES NOT LOAD ANY SPLIT ARRAY
# NO TRAINING
# NO OFFICIAL TEST DATA ACCESS
# ==================================================================================================

from pathlib import Path


assert SPLIT_DIR.exists()


all_split_files = sorted([
    p
    for p in SPLIT_DIR.rglob("*")
    if p.is_file()
])


print("=" * 130)
print("NOTEBOOK 05 — SPLIT-INDICES DIRECTORY NAME AUDIT")
print("=" * 130)

print(
    "Split directory:",
    SPLIT_DIR
)

print()

print(
    "Total files found:",
    len(all_split_files)
)

print()


for path in all_split_files:

    relative_path = (
        path.relative_to(
            ROOT
        )
    )

    name_lower = (
        path.name.lower()
    )


    # Filename classification only.
    # NO FILE IS OPENED.
    if "test" in name_lower:

        category = "TEST-NAMED — NOT OPENED"

    elif (
        "train" in name_lower
        or
        "val" in name_lower
    ):

        category = "TRAIN/VAL-NAMED"

    else:

        category = "OTHER / COMBINED"


    print(
        f"{category:24s} | "
        f"{relative_path} | "
        f"{path.stat().st_size:,} bytes"
    )


print()
print("=" * 130)

print(
    "Files opened:",
    0
)

print(
    "Split arrays loaded:",
    0
)

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 SPLIT-FILENAME AUDIT: COMPLETE"
)

print(
    "NO SPLIT CONTENT HAS BEEN READ"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 130)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 12B
# COMBINED FIXED-SPLIT NPZ MEMBER-NAME AUDIT
#
# PURPOSE:
#   Inspect the names stored inside each fixed-split .npz archive
#   WITHOUT loading any NumPy arrays.
#
# IMPORTANT:
#   - archive structure only
#   - no train/val/test array content read
#   - official test remains sealed
# ==================================================================================================

from pathlib import Path
import zipfile


# ==================================================================================================
# 1. EXPECTED COMBINED SPLIT ARCHIVES
# ==================================================================================================

SPLIT_ARCHIVES = {

    "Pingan":
        SPLIT_DIR /
        "pingan_fixed_split_seed2026.npz",

    "Qingyun":
        SPLIT_DIR /
        "qingyun_fixed_split_seed2026.npz",

    "Tangdaowan":
        SPLIT_DIR /
        "tangdaowan_fixed_split_seed2026.npz",
}


for dataset, path in SPLIT_ARCHIVES.items():

    assert path.exists(), (
        f"{dataset}: split archive missing:\n{path}"
    )


# ==================================================================================================
# 2. INSPECT ZIP MEMBER NAMES ONLY
#
# NPZ is a ZIP archive containing .npy members.
# We DO NOT load any .npy member.
# ==================================================================================================

SPLIT_ARCHIVE_MEMBERS = {}


print("=" * 135)
print("NOTEBOOK 05 — COMBINED FIXED-SPLIT ARCHIVE KEY AUDIT")
print("=" * 135)


for dataset, path in SPLIT_ARCHIVES.items():

    with zipfile.ZipFile(
        path,
        mode="r"
    ) as zf:

        members = sorted(
            zf.namelist()
        )


    SPLIT_ARCHIVE_MEMBERS[
        dataset
    ] = members


    print()
    print("-" * 135)
    print(dataset.upper())
    print("-" * 135)

    print(
        "Archive:",
        path.relative_to(
            ROOT
        )
    )

    print(
        "Members found:",
        len(
            members
        )
    )

    print()


    for member in members:

        member_lower = (
            member.lower()
        )


        if "test" in member_lower:

            category = (
                "TEST-NAMED — CONTENT NOT READ"
            )


        elif (
            "train" in member_lower
            or
            "val" in member_lower
        ):

            category = (
                "TRAIN/VAL-NAMED"
            )


        else:

            category = (
                "OTHER"
            )


        print(
            f"{category:31s} | "
            f"{member}"
        )


# ==================================================================================================
# 3. SAFETY CHECK
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_SPLIT_ARCHIVE_KEY_AUDIT_PASS = True


# ==================================================================================================
# 4. FINAL REPORT
# ==================================================================================================

print()
print("=" * 135)

print(
    "Split archives verified     : 3/3 PASS"
)

print(
    "Archive member names read   : PASS"
)

print(
    "NumPy arrays loaded         : 0"
)

print(
    "Training started            :",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used          :",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 SPLIT-ARCHIVE KEY AUDIT: PASS"
)

print(
    "NO SPLIT ARRAY CONTENT HAS BEEN LOADED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 135)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 12C
# FROZEN TRAIN / VALIDATION SPLIT LOAD + INTEGRITY AUDIT
#
# PURPOSE:
#   Load ONLY:
#       train_indices
#       train_labels_model
#       train_labels_original
#       val_indices
#       val_labels_model
#       val_labels_original
#
#   Plus safe split metadata.
#
# IMPORTANT:
#   test_indices / test_labels_* ARE NEVER REQUESTED.
#
# CPU/GPU INDEPENDENT
# NO TRAINING
# NO OFFICIAL TEST ACCESS
# ==================================================================================================

import json
import numpy as np


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_SPLIT_ARCHIVE_KEY_AUDIT_PASS is True

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


# ==================================================================================================
# 2. ONLY DATASETS REQUIRED FOR RECOVERY
# ==================================================================================================

RECOVERY_DATASETS = [
    "Pingan",
    "Qingyun",
]


EXPECTED_SPLIT_INFO = {

    "Pingan": {

        "K":
            10,

        "train_size":
            79869,

        "val_size":
            34230,

        "train_counts":
            np.array(
                [
                    3425,
                    40468,
                    585,
                    6229,
                    1453,
                    986,
                    979,
                    5818,
                    570,
                    19356,
                ],
                dtype=np.int64
            ),

        "val_counts":
            np.array(
                [
                    1468,
                    17343,
                    250,
                    2670,
                    623,
                    423,
                    419,
                    2494,
                    244,
                    8296,
                ],
                dtype=np.int64
            ),

        "spatial_shape":
            (
                1230,
                1000
            ),
    },


    "Qingyun": {

        "K":
            6,

        "train_size":
            66844,

        "val_size":
            28648,

        "train_counts":
            np.array(
                [
                    19472,
                    12565,
                    966,
                    684,
                    15242,
                    17915,
                ],
                dtype=np.int64
            ),

        "val_counts":
            np.array(
                [
                    8345,
                    5386,
                    414,
                    293,
                    6532,
                    7678,
                ],
                dtype=np.int64
            ),

        "spatial_shape":
            (
                880,
                1360
            ),
    },
}


# ==================================================================================================
# 3. EXACT ALLOWED MEMBERS
#
# No test member is present here.
# ==================================================================================================

ALLOWED_SPLIT_KEYS = [

    "evaluation_class_ids",

    "spatial_shape",

    "spectral_bands",

    "train_indices",

    "train_labels_model",

    "train_labels_original",

    "val_indices",

    "val_labels_model",

    "val_labels_original",

    "validation_fraction",

    "validation_split_seed",
]


# ==================================================================================================
# 4. LOAD ONLY ALLOWED MEMBERS
# ==================================================================================================

RECOVERY_SPLITS = {}

SPLIT_AUDIT_ROWS = []


print("=" * 145)
print("NOTEBOOK 05 — FROZEN TRAIN / VALIDATION SPLIT INTEGRITY AUDIT")
print("=" * 145)


for dataset in RECOVERY_DATASETS:

    archive_path = (
        SPLIT_ARCHIVES[
            dataset
        ]
    )


    expected = (
        EXPECTED_SPLIT_INFO[
            dataset
        ]
    )


    # ----------------------------------------------------------------------------------------------
    # Open archive, but explicitly request ONLY allowed train/validation metadata members.
    # test_* arrays are never indexed.
    # ----------------------------------------------------------------------------------------------

    with np.load(
        archive_path,
        allow_pickle=False
    ) as split_npz:

        # Confirm required keys exist.
        for key in ALLOWED_SPLIT_KEYS:

            assert key in split_npz.files, (
                f"{dataset}: missing required split key '{key}'."
            )


        evaluation_class_ids = np.asarray(
            split_npz[
                "evaluation_class_ids"
            ]
        ).copy()


        spatial_shape = np.asarray(
            split_npz[
                "spatial_shape"
            ]
        ).copy()


        spectral_bands = np.asarray(
            split_npz[
                "spectral_bands"
            ]
        ).copy()


        train_indices = np.asarray(
            split_npz[
                "train_indices"
            ],
            dtype=np.int64
        ).copy()


        train_labels_model = np.asarray(
            split_npz[
                "train_labels_model"
            ],
            dtype=np.int64
        ).copy()


        train_labels_original = np.asarray(
            split_npz[
                "train_labels_original"
            ],
            dtype=np.int64
        ).copy()


        val_indices = np.asarray(
            split_npz[
                "val_indices"
            ],
            dtype=np.int64
        ).copy()


        val_labels_model = np.asarray(
            split_npz[
                "val_labels_model"
            ],
            dtype=np.int64
        ).copy()


        val_labels_original = np.asarray(
            split_npz[
                "val_labels_original"
            ],
            dtype=np.int64
        ).copy()


        validation_fraction = float(
            np.asarray(
                split_npz[
                    "validation_fraction"
                ]
            ).reshape(
                -1
            )[0]
        )


        validation_split_seed = int(
            np.asarray(
                split_npz[
                    "validation_split_seed"
                ]
            ).reshape(
                -1
            )[0]
        )


    # ==============================================================================================
    # 5. BASIC SPLIT SIZE CHECKS
    # ==============================================================================================

    K = int(
        expected[
            "K"
        ]
    )


    assert train_indices.ndim == 1
    assert val_indices.ndim == 1

    assert train_labels_model.ndim == 1
    assert val_labels_model.ndim == 1

    assert train_labels_original.ndim == 1
    assert val_labels_original.ndim == 1


    assert len(
        train_indices
    ) == expected[
        "train_size"
    ]


    assert len(
        val_indices
    ) == expected[
        "val_size"
    ]


    assert (
        len(
            train_indices
        )
        ==
        len(
            train_labels_model
        )
        ==
        len(
            train_labels_original
        )
    )


    assert (
        len(
            val_indices
        )
        ==
        len(
            val_labels_model
        )
        ==
        len(
            val_labels_original
        )
    )


    # ==============================================================================================
    # 6. UNIQUE + DISJOINT TRAIN / VALIDATION INDICES
    # ==============================================================================================

    assert (
        np.unique(
            train_indices
        ).size
        ==
        len(
            train_indices
        )
    ), (
        f"{dataset}: duplicate train indices."
    )


    assert (
        np.unique(
            val_indices
        ).size
        ==
        len(
            val_indices
        )
    ), (
        f"{dataset}: duplicate validation indices."
    )


    overlap = np.intersect1d(

        train_indices,

        val_indices,

        assume_unique=True
    )


    assert (
        overlap.size
        ==
        0
    ), (
        f"{dataset}: train/validation overlap detected."
    )


    # ==============================================================================================
    # 7. MODEL LABEL CONVENTION
    # ==============================================================================================

    assert int(
        train_labels_model.min()
    ) == 0

    assert int(
        train_labels_model.max()
    ) == K - 1


    assert int(
        val_labels_model.min()
    ) == 0

    assert int(
        val_labels_model.max()
    ) == K - 1


    # ==============================================================================================
    # 8. ORIGINAL LABEL CONVENTION
    #
    # Public labels should be model labels + 1.
    # ==============================================================================================

    assert np.array_equal(

        train_labels_original,

        train_labels_model
        +
        1
    )


    assert np.array_equal(

        val_labels_original,

        val_labels_model
        +
        1
    )


    # ==============================================================================================
    # 9. EVALUATION CLASS IDs
    # ==============================================================================================

    expected_class_ids = np.arange(
        1,
        K + 1,
        dtype=np.int64
    )


    evaluation_class_ids = np.asarray(
        evaluation_class_ids,
        dtype=np.int64
    ).reshape(
        -1
    )


    assert np.array_equal(

        evaluation_class_ids,

        expected_class_ids
    )


    # ==============================================================================================
    # 10. CLASS COUNTS
    # ==============================================================================================

    train_counts = np.bincount(
        train_labels_model,
        minlength=K
    ).astype(
        np.int64
    )


    val_counts = np.bincount(
        val_labels_model,
        minlength=K
    ).astype(
        np.int64
    )


    assert np.array_equal(

        train_counts,

        expected[
            "train_counts"
        ]
    ), (
        f"{dataset}: frozen train class counts mismatch."
    )


    assert np.array_equal(

        val_counts,

        expected[
            "val_counts"
        ]
    ), (
        f"{dataset}: frozen validation class counts mismatch."
    )


    # ==============================================================================================
    # 11. FROZEN SPLIT PARAMETERS
    # ==============================================================================================

    assert np.isclose(
        validation_fraction,
        0.30,
        rtol=0.0,
        atol=1e-12
    )


    assert (
        validation_split_seed
        ==
        2026
    )


    # ==============================================================================================
    # 12. SPATIAL / SPECTRAL METADATA
    # ==============================================================================================

    spatial_shape_tuple = tuple(

        int(x)

        for x in np.asarray(
            spatial_shape
        ).reshape(
            -1
        )
    )


    assert (
        spatial_shape_tuple
        ==
        expected[
            "spatial_shape"
        ]
    )


    spectral_bands_flat = np.asarray(
        spectral_bands
    ).reshape(
        -1
    )


    # Original QUH cubes use 176 spectral bands.
    assert (
        spectral_bands_flat.size
        >=
        1
    )


    # ==============================================================================================
    # 13. INDEX RANGE CHECK
    # ==============================================================================================

    total_pixels = int(

        spatial_shape_tuple[
            0
        ]

        *

        spatial_shape_tuple[
            1
        ]
    )


    assert (
        train_indices
        >=
        0
    ).all()


    assert (
        train_indices
        <
        total_pixels
    ).all()


    assert (
        val_indices
        >=
        0
    ).all()


    assert (
        val_indices
        <
        total_pixels
    ).all()


    # ==============================================================================================
    # 14. STORE RECOVERY-SAFE SPLITS
    # ==============================================================================================

    RECOVERY_SPLITS[
        dataset
    ] = {

        "archive_path":
            archive_path,

        "evaluation_class_ids":
            evaluation_class_ids,

        "spatial_shape":
            spatial_shape_tuple,

        "train_indices":
            train_indices,

        "train_labels":
            train_labels_model,

        "val_indices":
            val_indices,

        "val_labels":
            val_labels_model,

        "validation_fraction":
            validation_fraction,

        "validation_split_seed":
            validation_split_seed,

        "archive_sha256":
            sha256_file(
                archive_path
            ),
    }


    SPLIT_AUDIT_ROWS.append({

        "dataset":
            dataset,

        "K":
            K,

        "train_size":
            len(
                train_indices
            ),

        "val_size":
            len(
                val_indices
            ),

        "train_val_overlap":
            int(
                overlap.size
            ),

        "validation_fraction":
            validation_fraction,

        "validation_split_seed":
            validation_split_seed,

        "archive_sha256":
            sha256_file(
                archive_path
            ),
    })


    # ==============================================================================================
    # 15. REPORT
    # ==============================================================================================

    print()
    print("-" * 145)
    print(dataset.upper())
    print("-" * 145)

    print(
        f"K                  : {K}"
    )

    print(
        f"Train samples      : {len(train_indices)}"
    )

    print(
        f"Validation samples : {len(val_indices)}"
    )

    print(
        f"Train/val overlap  : {overlap.size}"
    )

    print(
        f"Validation fraction: {validation_fraction}"
    )

    print(
        f"Split seed         : {validation_split_seed}"
    )

    print()

    print(
        "Train class counts:"
    )

    print(
        train_counts.tolist()
    )

    print(
        "Validation class counts:"
    )

    print(
        val_counts.tolist()
    )

    print()

    print(
        "Split archive SHA:"
    )

    print(
        sha256_file(
            archive_path
        )
    )


# ==================================================================================================
# 16. SAVE SPLIT AUDIT
# ==================================================================================================

SPLIT_AUDIT_DF = pd.DataFrame(
    SPLIT_AUDIT_ROWS
)


RECOVERY_SPLIT_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR /
    "notebook05_best_valAA_train_val_split_audit.json"
)


RECOVERY_SPLIT_AUDIT = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "purpose":
        "best_valAA_recovery_train_validation_split_verification",

    "datasets":
        {},

    "test_members_present_in_archives":
        True,

    "test_members_loaded":
        False,

    "official_test_used":
        False,

    "training_started":
        False,
}


for dataset in RECOVERY_DATASETS:

    split = RECOVERY_SPLITS[
        dataset
    ]


    RECOVERY_SPLIT_AUDIT[
        "datasets"
    ][
        dataset
    ] = {

        "archive":
            str(
                split[
                    "archive_path"
                ].relative_to(
                    ROOT
                )
            ),

        "archive_sha256":
            split[
                "archive_sha256"
            ],

        "train_size":
            int(
                len(
                    split[
                        "train_indices"
                    ]
                )
            ),

        "validation_size":
            int(
                len(
                    split[
                        "val_indices"
                    ]
                )
            ),

        "validation_fraction":
            float(
                split[
                    "validation_fraction"
                ]
            ),

        "validation_split_seed":
            int(
                split[
                    "validation_split_seed"
                ]
            ),

        "train_validation_overlap":
            0,
    }


with open(
    RECOVERY_SPLIT_AUDIT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        RECOVERY_SPLIT_AUDIT,
        f,
        indent=2,
        sort_keys=True
    )


RECOVERY_SPLIT_AUDIT_SHA = sha256_file(
    RECOVERY_SPLIT_AUDIT_PATH
)


# ==================================================================================================
# 17. FINAL SAFETY GATE
# ==================================================================================================

assert len(
    RECOVERY_SPLITS
) == 2

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_RECOVERY_SPLIT_LOAD_PASS = True


print()
print("=" * 145)

print(
    "Frozen train splits verified         : 2/2 PASS"
)

print(
    "Frozen validation splits verified    : 2/2 PASS"
)

print(
    "Train/validation disjointness        : 2/2 PASS"
)

print(
    "Class-count reconstruction           : 2/2 PASS"
)

print(
    "Validation seed/fraction             : 2/2 PASS"
)

print(
    "Test members loaded                  : 0"
)

print()

print(
    "Split audit artifact:"
)

print(
    RECOVERY_SPLIT_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    RECOVERY_SPLIT_AUDIT_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 TRAIN/VALIDATION SPLIT GATE: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 12D
# REPAIR RECOVERY AUDIT DIRECTORY + COMPLETE TRAIN/VALIDATION SPLIT GATE
#
# NO SPLITS ARE RELOADED
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

from pathlib import Path
import json


# ==================================================================================================
# 1. VERIFY CELL 12C ALREADY COMPLETED ITS DATA CHECKS
# ==================================================================================================

assert "RECOVERY_SPLITS" in globals(), (
    "RECOVERY_SPLITS not found. Do not continue."
)

assert len(
    RECOVERY_SPLITS
) == 2

assert set(
    RECOVERY_SPLITS.keys()
) == {
    "Pingan",
    "Qingyun",
}


# ==================================================================================================
# 2. CREATE MISSING RECOVERY AUDIT DIRECTORY
# ==================================================================================================

RECOVERY_AUDIT_DIR = (
    NOTEBOOK05_AUDIT_DIR /
    "best_valAA_recovery"
)


RECOVERY_AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


assert RECOVERY_AUDIT_DIR.exists()


# ==================================================================================================
# 3. BUILD SPLIT AUDIT ARTIFACT
# ==================================================================================================

RECOVERY_SPLIT_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR /
    "notebook05_best_valAA_train_val_split_audit.json"
)


RECOVERY_SPLIT_AUDIT = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "purpose":
        "best_valAA_recovery_train_validation_split_verification",

    "datasets":
        {},

    "test_members_present_in_archives":
        True,

    "test_members_loaded":
        False,

    "official_test_used":
        False,

    "training_started":
        False,
}


for dataset in [
    "Pingan",
    "Qingyun",
]:

    split = (
        RECOVERY_SPLITS[
            dataset
        ]
    )


    RECOVERY_SPLIT_AUDIT[
        "datasets"
    ][
        dataset
    ] = {

        "archive":
            str(
                split[
                    "archive_path"
                ].relative_to(
                    ROOT
                )
            ),

        "archive_sha256":
            split[
                "archive_sha256"
            ],

        "train_size":
            int(
                len(
                    split[
                        "train_indices"
                    ]
                )
            ),

        "validation_size":
            int(
                len(
                    split[
                        "val_indices"
                    ]
                )
            ),

        "validation_fraction":
            float(
                split[
                    "validation_fraction"
                ]
            ),

        "validation_split_seed":
            int(
                split[
                    "validation_split_seed"
                ]
            ),

        "train_validation_overlap":
            0,
    }


# ==================================================================================================
# 4. RESTART-SAFE WRITE
# ==================================================================================================

if RECOVERY_SPLIT_AUDIT_PATH.exists():

    with open(
        RECOVERY_SPLIT_AUDIT_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing = json.load(
            f
        )


    assert (
        existing
        ==
        RECOVERY_SPLIT_AUDIT
    ), (
        "Existing recovery split audit differs. STOP."
    )


else:

    with open(
        RECOVERY_SPLIT_AUDIT_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            RECOVERY_SPLIT_AUDIT,
            f,
            indent=2,
            sort_keys=True
        )


RECOVERY_SPLIT_AUDIT_SHA = (
    sha256_file(
        RECOVERY_SPLIT_AUDIT_PATH
    )
)


# ==================================================================================================
# 5. RECONFIRM SPLIT STATE
# ==================================================================================================

assert (
    len(
        RECOVERY_SPLITS[
            "Pingan"
        ][
            "train_indices"
        ]
    )
    ==
    79869
)

assert (
    len(
        RECOVERY_SPLITS[
            "Pingan"
        ][
            "val_indices"
        ]
    )
    ==
    34230
)

assert (
    len(
        RECOVERY_SPLITS[
            "Qingyun"
        ][
            "train_indices"
        ]
    )
    ==
    66844
)

assert (
    len(
        RECOVERY_SPLITS[
            "Qingyun"
        ][
            "val_indices"
        ]
    )
    ==
    28648
)


# ==================================================================================================
# 6. SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_RECOVERY_SPLIT_LOAD_PASS = True


# ==================================================================================================
# 7. FINAL REPORT
# ==================================================================================================

print("=" * 140)
print("NOTEBOOK 05 — TRAIN/VALIDATION SPLIT GATE REPAIR")
print("=" * 140)

print(
    "Pingan train samples             :",
    len(
        RECOVERY_SPLITS[
            "Pingan"
        ][
            "train_indices"
        ]
    )
)

print(
    "Pingan validation samples        :",
    len(
        RECOVERY_SPLITS[
            "Pingan"
        ][
            "val_indices"
        ]
    )
)

print(
    "Qingyun train samples            :",
    len(
        RECOVERY_SPLITS[
            "Qingyun"
        ][
            "train_indices"
        ]
    )
)

print(
    "Qingyun validation samples       :",
    len(
        RECOVERY_SPLITS[
            "Qingyun"
        ][
            "val_indices"
        ]
    )
)

print()

print(
    "Frozen train splits verified     : 2/2 PASS"
)

print(
    "Frozen validation splits verified: 2/2 PASS"
)

print(
    "Train/validation disjointness    : 2/2 PASS"
)

print(
    "Class-count reconstruction       : 2/2 PASS"
)

print(
    "Test members loaded              : 0"
)

print()

print(
    "Split audit artifact:"
)

print(
    RECOVERY_SPLIT_AUDIT_PATH
)

print()

print(
    "SHA-256:"
)

print(
    RECOVERY_SPLIT_AUDIT_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 TRAIN/VALIDATION SPLIT GATE: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 140)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 13
# EXACT HYBRIDSN RECOVERY DATA/MODEL PIPELINE + GPU SMOKE TEST
#
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import os
import json
import math
import random
import shutil
import hashlib

from pathlib import Path

import numpy as np
import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_SPLIT_LOAD_PASS is True

assert torch.cuda.is_available()
assert "T4" in torch.cuda.get_device_name(0).upper()

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


DEVICE = torch.device(
    "cuda:0"
)


# ==================================================================================================
# 2. LOCKED RECOVERY DATASET CONFIGURATION
# ==================================================================================================

RECOVERY_CONFIG = {

    "Pingan": {

        "patch_size":
            13,

        "margin":
            6,

        "pca_channels":
            15,

        "classes":
            10,

        "expected_parameters":
            519546,

        "source_cube":
            (
                DATA_CACHE_DIR
                /
                "pca_normalized_cubes"
                /
                "pingan_pca15_rowminmax_float32.npy"
            ),
    },


    "Qingyun": {

        "patch_size":
            11,

        "margin":
            5,

        "pca_channels":
            15,

        "classes":
            6,

        "expected_parameters":
            256886,

        "source_cube":
            (
                DATA_CACHE_DIR
                /
                "pca_normalized_cubes"
                /
                "qingyun_pca15_rowminmax_float32.npy"
            ),
    },
}


# ==================================================================================================
# 3. CURRENT ENVIRONMENT MUST MATCH NOTEBOOK 03
# ==================================================================================================

NB03_MASTER_AUDIT_PATH = (
    AUDIT_DIR
    /
    "notebook03_hybridsn"
    /
    "notebook03_master_audit.json"
)


assert NB03_MASTER_AUDIT_PATH.exists()


with open(
    NB03_MASTER_AUDIT_PATH,
    "r",
    encoding="utf-8"
) as f:

    NB03_MASTER_AUDIT = json.load(
        f
    )


nb03_environment = (
    NB03_MASTER_AUDIT[
        "environment"
    ]
)


assert (
    str(
        torch.__version__
    )
    ==
    str(
        nb03_environment[
            "pytorch"
        ]
    )
), (
    "PyTorch version differs from Notebook 03."
)


assert (
    str(
        torch.version.cuda
    )
    ==
    str(
        nb03_environment[
            "cuda"
        ]
    )
), (
    "CUDA runtime differs from Notebook 03."
)


assert (
    str(
        nb03_environment[
            "python"
        ]
    ).startswith(
        "3.12.13"
    )
)


# ==================================================================================================
# 4. COPY CANONICAL PCA CUBES TO LOCAL COLAB STORAGE
# ==================================================================================================

LOCAL_CACHE_ROOT = Path(
    "/content/ltlc_notebook05_recovery_cache"
)


LOCAL_CACHE_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


LOCAL_CUBE_PATHS = {}


print("=" * 145)
print("NOTEBOOK 05 — EXACT HYBRIDSN RECOVERY PIPELINE")
print("=" * 145)

print()
print("-" * 145)
print("LOCAL PCA CACHE PREPARATION")
print("-" * 145)


for dataset, config in (
    RECOVERY_CONFIG.items()
):

    source = (
        config[
            "source_cube"
        ]
    )


    assert source.exists(), (
        f"{dataset}: source PCA cube missing."
    )


    local_path = (
        LOCAL_CACHE_ROOT
        /
        source.name
    )


    source_sha = sha256_file(
        source
    )


    # ----------------------------------------------------------------------------------------------
    # Copy only when absent or incorrect.
    # ----------------------------------------------------------------------------------------------

    if local_path.exists():

        local_sha = sha256_file(
            local_path
        )


        if local_sha != source_sha:

            local_path.unlink()

            shutil.copy2(
                source,
                local_path
            )


    else:

        shutil.copy2(
            source,
            local_path
        )


    local_sha = sha256_file(
        local_path
    )


    assert (
        local_sha
        ==
        source_sha
    ), (
        f"{dataset}: local PCA cache SHA mismatch."
    )


    LOCAL_CUBE_PATHS[
        dataset
    ] = local_path


    print(
        f"{dataset:10s} | "
        f"{source.name} | "
        f"SHA={local_sha[:16]}... | PASS"
    )


# ==================================================================================================
# 5. LOAD LOCAL CUBES AS MEMORY MAPS
# ==================================================================================================

LOCAL_CUBE_MMAPS = {}


for dataset, local_path in (
    LOCAL_CUBE_PATHS.items()
):

    cube = np.load(
        local_path,
        mmap_mode="r"
    )


    config = (
        RECOVERY_CONFIG[
            dataset
        ]
    )


    assert cube.dtype == np.float32

    assert (
        cube.ndim
        ==
        3
    )

    assert (
        cube.shape[
            2
        ]
        ==
        config[
            "pca_channels"
        ]
    )


    LOCAL_CUBE_MMAPS[
        dataset
    ] = cube


# ==================================================================================================
# 6. ZERO-PAD CUBES
#
# Canonical Notebook 02/03 behavior:
# symmetric zero padding using patch margin.
# ==================================================================================================

PADDED_CUBES = {}


print()
print("-" * 145)
print("ZERO-PADDED PCA CUBES")
print("-" * 145)


for dataset, cube in (
    LOCAL_CUBE_MMAPS.items()
):

    margin = (
        RECOVERY_CONFIG[
            dataset
        ][
            "margin"
        ]
    )


    padded = np.pad(

        np.asarray(
            cube
        ),

        pad_width=(
            (
                margin,
                margin
            ),
            (
                margin,
                margin
            ),
            (
                0,
                0
            ),
        ),

        mode="constant",

        constant_values=0,
    )


    assert padded.dtype == np.float32


    expected_shape = (

        cube.shape[
            0
        ]
        +
        2 * margin,

        cube.shape[
            1
        ]
        +
        2 * margin,

        cube.shape[
            2
        ],
    )


    assert (
        padded.shape
        ==
        expected_shape
    )


    PADDED_CUBES[
        dataset
    ] = padded


    print(
        f"{dataset:10s} | "
        f"original={cube.shape} | "
        f"padded={padded.shape} | "
        f"margin={margin}"
    )


# ==================================================================================================
# 7. LAZY HYPERSPECTRAL PATCH DATASET
#
# Input center index is flattened row-major index from the
# original unpadded spatial cube.
#
# Returned tensor:
#
#     [1, spectral_channels, patch_height, patch_width]
#
# DataLoader therefore produces:
#
#     [B, 1, spectral_channels, H, W]
# ==================================================================================================

class LazyHSIPatchDataset(
    Dataset
):

    def __init__(
        self,
        padded_cube,
        original_spatial_shape,
        indices,
        labels,
        patch_size
    ):

        self.padded_cube = (
            padded_cube
        )


        self.height = int(
            original_spatial_shape[
                0
            ]
        )


        self.width = int(
            original_spatial_shape[
                1
            ]
        )


        self.indices = np.asarray(
            indices,
            dtype=np.int64
        )


        self.labels = np.asarray(
            labels,
            dtype=np.int64
        )


        self.patch_size = int(
            patch_size
        )


        assert (
            self.patch_size
            %
            2
            ==
            1
        )


        assert (
            len(
                self.indices
            )
            ==
            len(
                self.labels
            )
        )


    def __len__(
        self
    ):

        return len(
            self.indices
        )


    def __getitem__(
        self,
        item
    ):

        flat_index = int(
            self.indices[
                item
            ]
        )


        row = (
            flat_index
            //
            self.width
        )


        col = (
            flat_index
            %
            self.width
        )


        # ------------------------------------------------------------------------------------------
        # Because the cube is padded by margin=(patch_size//2),
        # the correct centered patch begins at [row, col].
        # ------------------------------------------------------------------------------------------

        patch = (

            self.padded_cube[
                row:
                    row
                    +
                    self.patch_size,

                col:
                    col
                    +
                    self.patch_size,

                :
            ]
        )


        assert (
            patch.shape[
                0
            ]
            ==
            self.patch_size
        )


        assert (
            patch.shape[
                1
            ]
            ==
            self.patch_size
        )


        # H,W,C -> C,H,W
        patch = np.transpose(
            patch,
            (
                2,
                0,
                1
            )
        )


        # Make writable contiguous copy for PyTorch.
        patch = np.ascontiguousarray(
            patch,
            dtype=np.float32
        )


        patch_tensor = torch.from_numpy(
            patch
        ).unsqueeze(
            0
        )


        label_tensor = torch.tensor(
            int(
                self.labels[
                    item
                ]
            ),
            dtype=torch.long
        )


        return (
            patch_tensor,
            label_tensor
        )


# ==================================================================================================
# 8. BUILD TRAIN / VALIDATION DATASETS
# ==================================================================================================

RECOVERY_TRAIN_DATASETS = {}

RECOVERY_VAL_DATASETS = {}


print()
print("-" * 145)
print("LAZY PATCH DATASETS")
print("-" * 145)


for dataset in [
    "Pingan",
    "Qingyun",
]:

    config = (
        RECOVERY_CONFIG[
            dataset
        ]
    )


    split = (
        RECOVERY_SPLITS[
            dataset
        ]
    )


    spatial_shape = (
        split[
            "spatial_shape"
        ]
    )


    RECOVERY_TRAIN_DATASETS[
        dataset
    ] = LazyHSIPatchDataset(

        padded_cube=
            PADDED_CUBES[
                dataset
            ],

        original_spatial_shape=
            spatial_shape,

        indices=
            split[
                "train_indices"
            ],

        labels=
            split[
                "train_labels"
            ],

        patch_size=
            config[
                "patch_size"
            ],
    )


    RECOVERY_VAL_DATASETS[
        dataset
    ] = LazyHSIPatchDataset(

        padded_cube=
            PADDED_CUBES[
                dataset
            ],

        original_spatial_shape=
            spatial_shape,

        indices=
            split[
                "val_indices"
            ],

        labels=
            split[
                "val_labels"
            ],

        patch_size=
            config[
                "patch_size"
            ],
    )


    assert (
        len(
            RECOVERY_TRAIN_DATASETS[
                dataset
            ]
        )
        ==
        len(
            split[
                "train_indices"
            ]
        )
    )


    assert (
        len(
            RECOVERY_VAL_DATASETS[
                dataset
            ]
        )
        ==
        len(
            split[
                "val_indices"
            ]
        )
    )


    print(
        f"{dataset:10s} | "
        f"train={len(RECOVERY_TRAIN_DATASETS[dataset])} | "
        f"val={len(RECOVERY_VAL_DATASETS[dataset])} | "
        f"patch={config['patch_size']}"
    )


# ==================================================================================================
# 9. CANONICAL HYBRIDSN
#
# 3D:
#   1  -> 8   kernel=(7,3,3)
#   8  -> 16  kernel=(5,3,3)
#   16 -> 32  kernel=(3,3,3)
#
# reshape spectral-depth into channels
#
# 2D:
#   -> 64 kernel=3
#
# Dense:
#   256 -> dropout 0.4
#   128 -> dropout 0.4
#   logits
# ==================================================================================================

class HybridSN(
    nn.Module
):

    def __init__(
        self,
        patch_size,
        spectral_channels,
        num_classes
    ):

        super().__init__()


        self.conv3d_1 = nn.Conv3d(

            in_channels=1,

            out_channels=8,

            kernel_size=(
                7,
                3,
                3
            ),

            stride=1,

            padding=0,
        )


        self.conv3d_2 = nn.Conv3d(

            in_channels=8,

            out_channels=16,

            kernel_size=(
                5,
                3,
                3
            ),

            stride=1,

            padding=0,
        )


        self.conv3d_3 = nn.Conv3d(

            in_channels=16,

            out_channels=32,

            kernel_size=(
                3,
                3,
                3
            ),

            stride=1,

            padding=0,
        )


        self.relu = nn.ReLU(
            inplace=False
        )


        # ------------------------------------------------------------------------------------------
        # Determine reshape dimensions exactly.
        # ------------------------------------------------------------------------------------------

        depth_after_3d = (
            spectral_channels
            -
            6
            -
            4
            -
            2
        )


        spatial_after_3d = (
            patch_size
            -
            2
            -
            2
            -
            2
        )


        assert (
            depth_after_3d
            >
            0
        )


        assert (
            spatial_after_3d
            >
            2
        )


        conv2d_in_channels = (
            32
            *
            depth_after_3d
        )


        self.conv2d = nn.Conv2d(

            in_channels=
                conv2d_in_channels,

            out_channels=
                64,

            kernel_size=
                3,

            stride=
                1,

            padding=
                0,
        )


        spatial_after_2d = (
            spatial_after_3d
            -
            2
        )


        flatten_features = (
            64
            *
            spatial_after_2d
            *
            spatial_after_2d
        )


        self.fc1 = nn.Linear(

            flatten_features,

            256
        )


        self.dropout1 = nn.Dropout(
            p=0.4
        )


        self.fc2 = nn.Linear(

            256,

            128
        )


        self.dropout2 = nn.Dropout(
            p=0.4
        )


        self.classifier = nn.Linear(

            128,

            num_classes
        )


    def forward(
        self,
        x
    ):

        x = self.relu(
            self.conv3d_1(
                x
            )
        )


        x = self.relu(
            self.conv3d_2(
                x
            )
        )


        x = self.relu(
            self.conv3d_3(
                x
            )
        )


        # [B,C,D,H,W]
        batch_size = (
            x.shape[
                0
            ]
        )


        x = x.reshape(

            batch_size,

            x.shape[
                1
            ]
            *
            x.shape[
                2
            ],

            x.shape[
                3
            ],

            x.shape[
                4
            ]
        )


        x = self.relu(
            self.conv2d(
                x
            )
        )


        x = torch.flatten(
            x,
            start_dim=1
        )


        x = self.relu(
            self.fc1(
                x
            )
        )


        x = self.dropout1(
            x
        )


        x = self.relu(
            self.fc2(
                x
            )
        )


        x = self.dropout2(
            x
        )


        logits = self.classifier(
            x
        )


        return logits


# ==================================================================================================
# 10. PARAMETER-COUNT HELPER
# ==================================================================================================

def count_parameters(
    model
):

    return int(
        sum(
            p.numel()
            for p in model.parameters()
        )
    )


# ==================================================================================================
# 11. BUILD MODELS + VERIFY EXACT PARAMETER COUNTS
# ==================================================================================================

RECOVERY_MODEL_PROTOTYPES = {}


print()
print("-" * 145)
print("HYBRIDSN ARCHITECTURE AUDIT")
print("-" * 145)


for dataset in [
    "Pingan",
    "Qingyun",
]:

    config = (
        RECOVERY_CONFIG[
            dataset
        ]
    )


    model = HybridSN(

        patch_size=
            config[
                "patch_size"
            ],

        spectral_channels=
            config[
                "pca_channels"
            ],

        num_classes=
            config[
                "classes"
            ],
    )


    parameter_count = count_parameters(
        model
    )


    assert (
        parameter_count
        ==
        config[
            "expected_parameters"
        ]
    ), (
        f"{dataset}: parameter count mismatch. "
        f"Got {parameter_count}, "
        f"expected {config['expected_parameters']}."
    )


    RECOVERY_MODEL_PROTOTYPES[
        dataset
    ] = model


    print(
        f"{dataset:10s} | "
        f"parameters={parameter_count:,} | PASS"
    )


# ==================================================================================================
# 12. VALIDATION GPU SMOKE LOADERS
#
# shuffle=False
# These are ONLY for the pipeline smoke test.
# They are NOT the future training DataLoaders.
# ==================================================================================================

SMOKE_LOADERS = {}


for dataset in [
    "Pingan",
    "Qingyun",
]:

    SMOKE_LOADERS[
        dataset
    ] = DataLoader(

        RECOVERY_VAL_DATASETS[
            dataset
        ],

        batch_size=256,

        shuffle=False,

        num_workers=0,

        pin_memory=True,

        drop_last=False,
    )


# ==================================================================================================
# 13. GPU FORWARD-PASS SMOKE TEST
# ==================================================================================================

SMOKE_RESULTS = {}


print()
print("-" * 145)
print("T4 GPU PIPELINE SMOKE TEST")
print("-" * 145)


for dataset in [
    "Pingan",
    "Qingyun",
]:

    config = (
        RECOVERY_CONFIG[
            dataset
        ]
    )


    model = (
        RECOVERY_MODEL_PROTOTYPES[
            dataset
        ]
    ).to(
        DEVICE
    )


    model.eval()


    batch_x, batch_y = next(
        iter(
            SMOKE_LOADERS[
                dataset
            ]
        )
    )


    expected_input_shape = (

        batch_x.shape[
            0
        ],

        1,

        config[
            "pca_channels"
        ],

        config[
            "patch_size"
        ],

        config[
            "patch_size"
        ],
    )


    assert (
        tuple(
            batch_x.shape
        )
        ==
        expected_input_shape
    )


    assert (
        batch_y.ndim
        ==
        1
    )


    batch_x_gpu = batch_x.to(

        DEVICE,

        non_blocking=True
    )


    batch_y_gpu = batch_y.to(

        DEVICE,

        non_blocking=True
    )


    with torch.no_grad():

        logits = model(
            batch_x_gpu
        )


    assert (
        tuple(
            logits.shape
        )
        ==
        (
            batch_x.shape[
                0
            ],
            config[
                "classes"
            ],
        )
    )


    assert torch.isfinite(
        logits
    ).all()


    SMOKE_RESULTS[
        dataset
    ] = {

        "input_shape":
            tuple(
                int(x)
                for x in batch_x.shape
            ),

        "label_shape":
            tuple(
                int(x)
                for x in batch_y.shape
            ),

        "logit_shape":
            tuple(
                int(x)
                for x in logits.shape
            ),

        "parameters":
            count_parameters(
                model
            ),
    }


    print(
        f"{dataset:10s} | "
        f"input={tuple(batch_x.shape)} | "
        f"logits={tuple(logits.shape)} | "
        f"parameters={count_parameters(model):,} | PASS"
    )


    # Return prototype to CPU.
    model.to(
        "cpu"
    )


    del batch_x_gpu
    del batch_y_gpu
    del logits

    torch.cuda.empty_cache()


# ==================================================================================================
# 14. VERIFY EXPECTED NOTEBOOK 03 BATCH SHAPES
# ==================================================================================================

assert (
    SMOKE_RESULTS[
        "Pingan"
    ][
        "input_shape"
    ]
    ==
    (
        256,
        1,
        15,
        13,
        13
    )
)


assert (
    SMOKE_RESULTS[
        "Qingyun"
    ][
        "input_shape"
    ]
    ==
    (
        256,
        1,
        15,
        11,
        11
    )
)


# ==================================================================================================
# 15. FREEZE RECOVERY PIPELINE SPEC
# ==================================================================================================

RECOVERY_PIPELINE_SPEC_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_best_valAA_recovery_pipeline_spec.json"
)


RECOVERY_PIPELINE_SPEC = {

    "project":
        "LTLC",

    "purpose":
        "recover_missing_HybridSN_CE_best_validation_AA_checkpoints",

    "environment": {

        "python":
            "3.12.13",

        "pytorch":
            str(
                torch.__version__
            ),

        "cuda":
            str(
                torch.version.cuda
            ),

        "gpu":
            torch.cuda.get_device_name(
                0
            ),
    },

    "batch_size":
        256,

    "num_workers":
        0,

    "mixed_precision":
        False,

    "padding":
        "zero",

    "datasets": {},

    "architecture": {

        "conv3d_1":
            "1->8 kernel=(7,3,3)",

        "conv3d_2":
            "8->16 kernel=(5,3,3)",

        "conv3d_3":
            "16->32 kernel=(3,3,3)",

        "conv2d":
            "reshape spectral-depth into channels; ->64 kernel=3",

        "fc1":
            256,

        "dropout1":
            0.4,

        "fc2":
            128,

        "dropout2":
            0.4,

        "output":
            "dataset_class_count",
    },

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,

    "training_started":
        False,
}


for dataset in [
    "Pingan",
    "Qingyun",
]:

    config = (
        RECOVERY_CONFIG[
            dataset
        ]
    )


    RECOVERY_PIPELINE_SPEC[
        "datasets"
    ][
        dataset
    ] = {

        "patch_size":
            int(
                config[
                    "patch_size"
                ]
            ),

        "margin":
            int(
                config[
                    "margin"
                ]
            ),

        "pca_channels":
            int(
                config[
                    "pca_channels"
                ]
            ),

        "classes":
            int(
                config[
                    "classes"
                ]
            ),

        "parameter_count":
            int(
                config[
                    "expected_parameters"
                ]
            ),

        "local_cube":
            str(
                LOCAL_CUBE_PATHS[
                    dataset
                ]
            ),

        "local_cube_sha256":
            sha256_file(
                LOCAL_CUBE_PATHS[
                    dataset
                ]
            ),

        "train_samples":
            int(
                len(
                    RECOVERY_TRAIN_DATASETS[
                        dataset
                    ]
                )
            ),

        "validation_samples":
            int(
                len(
                    RECOVERY_VAL_DATASETS[
                        dataset
                    ]
                )
            ),
    }


# ==================================================================================================
# 16. RESTART-SAFE WRITE
# ==================================================================================================

if RECOVERY_PIPELINE_SPEC_PATH.exists():

    with open(
        RECOVERY_PIPELINE_SPEC_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing = json.load(
            f
        )


    assert (
        existing
        ==
        RECOVERY_PIPELINE_SPEC
    ), (
        "Existing recovery pipeline spec differs. STOP."
    )


else:

    with open(
        RECOVERY_PIPELINE_SPEC_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            RECOVERY_PIPELINE_SPEC,
            f,
            indent=2,
            sort_keys=True
        )


RECOVERY_PIPELINE_SPEC_SHA = sha256_file(
    RECOVERY_PIPELINE_SPEC_PATH
)


# ==================================================================================================
# 17. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_RECOVERY_PIPELINE_GATE_PASS = True


print()
print("=" * 145)

print(
    "Notebook 03 software match             : PASS"
)

print(
    "Local PCA cache verification           : 2/2 PASS"
)

print(
    "Zero-padding reconstruction            : 2/2 PASS"
)

print(
    "Lazy train/validation datasets         : 2/2 PASS"
)

print(
    "Pingan HybridSN parameters             : 519,546 PASS"
)

print(
    "Qingyun HybridSN parameters            : 256,886 PASS"
)

print(
    "T4 forward-pass smoke tests            : 2/2 PASS"
)

print(
    "Expected Notebook 03 batch shapes      : 2/2 PASS"
)

print()

print(
    "Pipeline spec:"
)

print(
    RECOVERY_PIPELINE_SPEC_PATH
)

print(
    "SHA-256:"
)

print(
    RECOVERY_PIPELINE_SPEC_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official-test dataset created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 HYBRIDSN RECOVERY PIPELINE GATE: PASS"
)

print(
    "READY TO LOCK TRAINING ENGINE"
)

print(
    "NO RECOVERY TRAINING HAS STARTED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 14
# EXECUTED NOTEBOOK 03 CE TRAINING-ENGINE EVIDENCE AUDIT
#
# PURPOSE:
#   Determine the ACTUAL training engine used by completed Notebook 03
#   before attempting deterministic best-val-AA checkpoint recovery.
#
# READ ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
from pathlib import Path

import pandas as pd
import torch


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_PIPELINE_GATE_PASS is True

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


print("=" * 145)
print("NOTEBOOK 05 — EXECUTED NOTEBOOK 03 TRAINING-ENGINE AUDIT")
print("=" * 145)


# ==================================================================================================
# 2. ORIGINAL HISTORY SCHEMA + LR BEHAVIOR
# ==================================================================================================

print()
print("-" * 145)
print("ORIGINAL CE HISTORY SCHEMA / LEARNING-RATE TRACE")
print("-" * 145)


HISTORY_ENGINE_FINDINGS = {}


for (
    dataset,
    seed
), target_epoch in EXPECTED_RECOVERY_TARGETS.items():

    history_path = RECOVERY_HISTORY[
        (
            dataset,
            seed
        )
    ]


    df = pd.read_csv(
        history_path
    )


    print()
    print(
        f"{dataset} | seed={seed}"
    )

    print(
        "History:",
        history_path.relative_to(
            ROOT
        )
    )

    print(
        "Columns:",
        list(
            df.columns
        )
    )


    lower_cols = {
        str(c).lower():
            c
        for c in df.columns
    }


    lr_column = None


    for candidate in [
        "lr",
        "learning_rate",
        "learningrate",
    ]:

        if candidate in lower_cols:

            lr_column = lower_cols[
                candidate
            ]

            break


    if lr_column is not None:

        lr_values = (
            df[
                lr_column
            ]
            .astype(float)
            .to_numpy()
        )


        unique_lr = sorted(
            set(
                float(x)
                for x in lr_values
            )
        )


        print(
            "LR column:",
            lr_column
        )

        print(
            "First LR:",
            float(
                lr_values[
                    0
                ]
            )
        )

        print(
            "Last LR:",
            float(
                lr_values[
                    -1
                ]
            )
        )

        print(
            "Unique LR count:",
            len(
                unique_lr
            )
        )

        print(
            "First unique LR values:",
            unique_lr[
                :10
            ]
        )


    else:

        print(
            "LR column: NOT STORED"
        )


    # Show selected epochs around target.
    epoch_col = lower_cols.get(
        "epoch"
    )


    assert epoch_col is not None


    inspect_epochs = sorted(
        set(
            [
                1,
                2,
                max(
                    1,
                    target_epoch - 1
                ),
                target_epoch,
                min(
                    100,
                    target_epoch + 1
                ),
                100,
            ]
        )
    )


    display_rows = (
        df[
            df[
                epoch_col
            ].astype(int)
            .isin(
                inspect_epochs
            )
        ]
        .copy()
    )


    print()
    print(
        "Selected history rows:"
    )

    print(
        display_rows.to_string(
            index=False
        )
    )


    HISTORY_ENGINE_FINDINGS[
        (
            dataset,
            seed
        )
    ] = {

        "columns":
            list(
                df.columns
            ),

        "lr_column":
            lr_column,

        "rows":
            len(
                df
            ),
    }


# ==================================================================================================
# 3. INSPECT EXISTING NOTEBOOK 03 BEST-OA CHECKPOINT STRUCTURE
#
# No state is loaded into a model.
# ==================================================================================================

print()
print("-" * 145)
print("EXISTING NOTEBOOK 03 CHECKPOINT STRUCTURE")
print("-" * 145)


CHECKPOINT_STRUCTURE = {}


for (
    dataset,
    seed
), checkpoint_path in EXISTING_BEST_OA.items():

    checkpoint = torch.load(

        checkpoint_path,

        map_location="cpu",

        weights_only=False
    )


    print()
    print(
        f"{dataset} | seed={seed}"
    )

    print(
        "Checkpoint:",
        checkpoint_path.relative_to(
            ROOT
        )
    )

    print(
        "Loaded type:",
        type(
            checkpoint
        ).__name__
    )


    if isinstance(
        checkpoint,
        dict
    ):

        top_keys = list(
            checkpoint.keys()
        )


        print(
            "Top-level keys:",
            top_keys
        )


        # ------------------------------------------------------------------------------------------
        # Print safe scalar/string/list metadata only.
        # Do NOT print tensor contents.
        # ------------------------------------------------------------------------------------------

        scalar_metadata = {}


        for key, value in checkpoint.items():

            if isinstance(
                value,
                (
                    str,
                    int,
                    float,
                    bool,
                    type(None)
                )
            ):

                scalar_metadata[
                    key
                ] = value


            elif isinstance(
                value,
                (
                    list,
                    tuple
                )
            ):

                if (
                    len(value) <= 20
                    and
                    all(
                        isinstance(
                            x,
                            (
                                str,
                                int,
                                float,
                                bool,
                                type(None)
                            )
                        )
                        for x in value
                    )
                ):

                    scalar_metadata[
                        key
                    ] = value


        print(
            "Scalar metadata:"
        )

        for key, value in scalar_metadata.items():

            print(
                f"  {key}: {value}"
            )


        # ------------------------------------------------------------------------------------------
        # Optimizer / scheduler structure, if stored.
        # ------------------------------------------------------------------------------------------

        for possible_key in [

            "optimizer_state_dict",

            "optimizer",

            "scheduler_state_dict",

            "scheduler",

        ]:

            if possible_key in checkpoint:

                value = checkpoint[
                    possible_key
                ]


                if isinstance(
                    value,
                    dict
                ):

                    print(
                        f"{possible_key} keys:",
                        list(
                            value.keys()
                        )
                    )


                    if (
                        "param_groups"
                        in value
                    ):

                        param_groups = (
                            value[
                                "param_groups"
                            ]
                        )


                        print(
                            f"{possible_key} param_groups:",
                            len(
                                param_groups
                            )
                        )


                        for i, group in enumerate(
                            param_groups
                        ):

                            safe_group = {

                                k:
                                    v

                                for k, v in group.items()

                                if isinstance(
                                    v,
                                    (
                                        str,
                                        int,
                                        float,
                                        bool,
                                        type(None)
                                    )
                                )
                            }


                            print(
                                f"  group {i}:",
                                safe_group
                            )


        CHECKPOINT_STRUCTURE[
            (
                dataset,
                seed
            )
        ] = {

            "type":
                "dict",

            "keys":
                top_keys,

            "scalar_metadata":
                scalar_metadata,
        }


    else:

        print(
            "Checkpoint appears to be a raw state_dict/object."
        )


        CHECKPOINT_STRUCTURE[
            (
                dataset,
                seed
            )
        ] = {

            "type":
                type(
                    checkpoint
                ).__name__,
        }


    del checkpoint


# ==================================================================================================
# 4. PROJECT CONFIG INVENTORY
# ==================================================================================================

CONFIG_DIR = (
    ROOT /
    "configs"
)


print()
print("-" * 145)
print("PROJECT CONFIG INVENTORY")
print("-" * 145)


CONFIG_FILES = []


if CONFIG_DIR.exists():

    CONFIG_FILES = sorted([
        p
        for p in CONFIG_DIR.rglob("*")
        if p.is_file()
    ])


print(
    "Config files found:",
    len(
        CONFIG_FILES
    )
)


for path in CONFIG_FILES:

    print(
        path.relative_to(
            ROOT
        )
    )


# ==================================================================================================
# 5. READ SMALL TEXT/JSON/YAML CONFIG FILES
# ==================================================================================================

print()
print("-" * 145)
print("SMALL CONFIG CONTENTS")
print("-" * 145)


for path in CONFIG_FILES:

    if path.stat().st_size > 200_000:

        continue


    if path.suffix.lower() not in [
        ".json",
        ".yaml",
        ".yml",
        ".txt",
        ".md",
    ]:

        continue


    print()
    print(
        "FILE:",
        path.relative_to(
            ROOT
        )
    )

    print(
        "-" * 100
    )


    try:

        text = path.read_text(
            encoding="utf-8"
        )


        # Prevent huge notebook output.
        print(
            text[
                :8000
            ]
        )


        if len(
            text
        ) > 8000:

            print(
                "... CONTENT TRUNCATED ..."
            )


    except Exception as e:

        print(
            "Could not read:",
            repr(
                e
            )
        )


# ==================================================================================================
# 6. NOTEBOOK 03 MASTER-AUDIT TRAINING SECTION
# ==================================================================================================

print()
print("-" * 145)
print("NOTEBOOK 03 MASTER AUDIT — TRAINING SECTION")
print("-" * 145)


print(
    json.dumps(

        NB03_MASTER_AUDIT.get(
            "training",
            {}
        ),

        indent=2,

        sort_keys=True
    )
)


# ==================================================================================================
# 7. IMPLEMENTATION-LOCK DISCREPANCY RECORD
#
# We do not decide the engine here.
# We only document that training must not start until
# executed evidence is reviewed.
# ==================================================================================================

ENGINE_EVIDENCE_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_executed_ce_engine_evidence_audit.json"
)


ENGINE_EVIDENCE_AUDIT = {

    "project":
        "LTLC",

    "purpose":
        "identify_actual_executed_Notebook03_CE_training_engine",

    "environment_match": {

        "python_3_12_13":
            True,

        "pytorch":
            str(
                torch.__version__
            ),

        "cuda":
            str(
                torch.version.cuda
            ),
    },

    "recovery_targets":
        {

            f"{dataset}_seed{seed}":
                int(
                    epoch
                )

            for (
                dataset,
                seed
            ), epoch
            in EXPECTED_RECOVERY_TARGETS.items()
        },

    "history_columns":
        {

            f"{dataset}_seed{seed}":
                info[
                    "columns"
                ]

            for (
                dataset,
                seed
            ), info
            in HISTORY_ENGINE_FINDINGS.items()
        },

    "checkpoint_structures":
        {

            f"{dataset}_seed{seed}":
                info

            for (
                dataset,
                seed
            ), info
            in CHECKPOINT_STRUCTURE.items()
        },

    "config_files":
        [
            str(
                p.relative_to(
                    ROOT
                )
            )
            for p in CONFIG_FILES
        ],

    "training_started":
        False,

    "official_test_used":
        False,
}


with open(
    ENGINE_EVIDENCE_AUDIT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        ENGINE_EVIDENCE_AUDIT,
        f,
        indent=2,
        sort_keys=True
    )


ENGINE_EVIDENCE_AUDIT_SHA = (
    sha256_file(
        ENGINE_EVIDENCE_AUDIT_PATH
    )
)


# ==================================================================================================
# 8. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_EXECUTED_ENGINE_EVIDENCE_AUDIT_PASS = True


print()
print("=" * 145)

print(
    "Original CE histories inspected        : 5/5 PASS"
)

print(
    "Existing CE checkpoints inspected      : 5/5 PASS"
)

print(
    "Project config inventory               : PASS"
)

print(
    "Notebook 03 training audit displayed   : PASS"
)

print()

print(
    "Engine evidence audit:"
)

print(
    ENGINE_EVIDENCE_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    ENGINE_EVIDENCE_AUDIT_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 EXECUTED CE ENGINE EVIDENCE AUDIT: PASS"
)

print(
    "DO NOT START RECOVERY TRAINING UNTIL ENGINE EVIDENCE IS REVIEWED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 14A
# EXACT EXECUTED NOTEBOOK 03 TRAINING-CONFIG EXTRACTION
#
# READ ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
import math
import torch


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_EXECUTED_ENGINE_EVIDENCE_AUDIT_PASS is True
assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


# ==================================================================================================
# 2. USE ALL FIVE EXISTING CHECKPOINTS AS CONSISTENCY EVIDENCE
# ==================================================================================================

CONFIG_RECORDS = {}


print("=" * 145)
print("NOTEBOOK 05 — EXACT EXECUTED CE TRAINING-CONFIG EXTRACTION")
print("=" * 145)


for (
    dataset,
    seed
), checkpoint_path in EXISTING_BEST_OA.items():

    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )


    assert isinstance(
        checkpoint,
        dict
    )


    assert (
        "training_config"
        in checkpoint
    )


    assert (
        "model_configuration"
        in checkpoint
    )


    assert (
        "optimizer_state_dict"
        in checkpoint
    )


    training_config = checkpoint[
        "training_config"
    ]


    model_configuration = checkpoint[
        "model_configuration"
    ]


    optimizer_state = checkpoint[
        "optimizer_state_dict"
    ]


    assert isinstance(
        training_config,
        dict
    )

    assert isinstance(
        model_configuration,
        dict
    )


    param_groups = optimizer_state[
        "param_groups"
    ]


    assert len(
        param_groups
    ) == 1


    optimizer_group = param_groups[
        0
    ]


    print()
    print("-" * 145)

    print(
        f"{dataset.upper()} — SEED {seed}"
    )

    print("-" * 145)

    print(
        "training_config:"
    )

    print(
        json.dumps(
            training_config,
            indent=2,
            sort_keys=True,
            default=str
        )
    )


    print()

    print(
        "model_configuration:"
    )

    print(
        json.dumps(
            model_configuration,
            indent=2,
            sort_keys=True,
            default=str
        )
    )


    print()

    print(
        "optimizer param group:"
    )


    for key in [

        "lr",

        "betas",

        "eps",

        "weight_decay",

        "amsgrad",

        "maximize",

        "foreach",

        "capturable",

        "differentiable",

        "fused",

        "decoupled_weight_decay",
    ]:

        if key in optimizer_group:

            print(
                f"  {key}: "
                f"{optimizer_group[key]}"
            )


    CONFIG_RECORDS[
        (
            dataset,
            seed
        )
    ] = {

        "training_config":
            training_config,

        "model_configuration":
            model_configuration,

        "optimizer_betas":
            tuple(
                float(x)

                for x in optimizer_group[
                    "betas"
                ]
            ),

        "optimizer_eps":
            float(
                optimizer_group[
                    "eps"
                ]
            ),

        "optimizer_weight_decay":
            float(
                optimizer_group[
                    "weight_decay"
                ]
            ),

        "checkpoint_epoch":
            int(
                checkpoint[
                    "epoch"
                ]
            ),

        "global_step":
            int(
                checkpoint[
                    "global_step"
                ]
            ),
    }


    del checkpoint


# ==================================================================================================
# 3. VERIFY CONFIG CONSISTENCY ACROSS ALL FIVE CHECKPOINTS
# ==================================================================================================

records = list(
    CONFIG_RECORDS.values()
)


reference_training_config = (
    records[
        0
    ][
        "training_config"
    ]
)


reference_betas = (
    records[
        0
    ][
        "optimizer_betas"
    ]
)


reference_eps = (
    records[
        0
    ][
        "optimizer_eps"
    ]
)


reference_weight_decay = (
    records[
        0
    ][
        "optimizer_weight_decay"
    ]
)


for record in records[1:]:

    assert (
        record[
            "training_config"
        ]
        ==
        reference_training_config
    ), (
        "Training config differs across recovered CE runs."
    )


    assert (
        record[
            "optimizer_betas"
        ]
        ==
        reference_betas
    )


    assert math.isclose(
        record[
            "optimizer_eps"
        ],
        reference_eps,
        rel_tol=0.0,
        abs_tol=0.0
    )


    assert math.isclose(
        record[
            "optimizer_weight_decay"
        ],
        reference_weight_decay,
        rel_tol=0.0,
        abs_tol=0.0
    )


# ==================================================================================================
# 4. VERIFY EXECUTED BATCH SIZE FROM GLOBAL-STEP HISTORY
# ==================================================================================================

EXPECTED_STEPS_PER_EPOCH = {

    "Pingan":
        312,

    "Qingyun":
        262,
}


for dataset in [
    "Pingan",
    "Qingyun",
]:

    train_size = len(
        RECOVERY_SPLITS[
            dataset
        ][
            "train_indices"
        ]
    )


    calculated = math.ceil(
        train_size
        /
        256
    )


    assert (
        calculated
        ==
        EXPECTED_STEPS_PER_EPOCH[
            dataset
        ]
    )


print()
print("=" * 145)
print("EXECUTED ENGINE CONSISTENCY SUMMARY")
print("=" * 145)

print(
    "Five checkpoint training configs identical : PASS"
)

print(
    "Optimizer betas identical                  : PASS"
)

print(
    "Optimizer epsilon identical                : PASS"
)

print(
    "Optimizer weight decay identical           : PASS"
)

print(
    "Executed Pingan steps/epoch                : 312"
)

print(
    "Executed Qingyun steps/epoch               : 262"
)

print(
    "Executed batch size inference              : 256 PASS"
)

print()

print(
    "Canonical optimizer betas:"
)

print(
    reference_betas
)

print(
    "Canonical optimizer epsilon:"
)

print(
    reference_eps
)

print(
    "Canonical optimizer weight decay:"
)

print(
    reference_weight_decay
)

print()


# ==================================================================================================
# 5. VERIFY EXACT LEGACY LR FORMULA AGAINST ALL FIVE HISTORIES
#
# History epoch-end LR:
#
#     lr = base_lr / (1 + decay * global_step)
#
# Optimizer checkpoint LR:
#
#     last LR actually used =
#     base_lr / (1 + decay * (global_step - 1))
# ==================================================================================================

BASE_LR = 1e-3

LEGACY_DECAY = 1e-6


for (
    dataset,
    seed
), history_path in RECOVERY_HISTORY.items():

    df = pd.read_csv(
        history_path
    )


    expected_epoch_end_lr = (

        BASE_LR

        /

        (
            1.0
            +
            LEGACY_DECAY
            *
            df[
                "global_step"
            ].to_numpy(
                dtype=np.float64
            )
        )
    )


    actual_lr = df[
        "learning_rate"
    ].to_numpy(
        dtype=np.float64
    )


    max_difference = float(

        np.max(

            np.abs(
                expected_epoch_end_lr
                -
                actual_lr
            )
        )
    )


    assert (
        max_difference
        <=
        1e-15
    ), (
        f"{dataset} seed {seed}: LR schedule mismatch."
    )


print(
    "Legacy inverse-time LR schedule           : 5/5 PASS"
)


# ==================================================================================================
# 6. FREEZE EXACT EXECUTED ENGINE EVIDENCE
# ==================================================================================================

EXECUTED_CONFIG_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_exact_executed_ce_training_config.json"
)


EXECUTED_CONFIG_OBJECT = {

    "project":
        "LTLC",

    "purpose":
        "best_valAA_recovery",

    "source":
        "executed_Notebook03_checkpoint_and_history_evidence",

    "training_config":
        reference_training_config,

    "optimizer": {

        "name":
            "Adam",

        "base_learning_rate":
            BASE_LR,

        "betas":
            list(
                reference_betas
            ),

        "epsilon":
            reference_eps,

        "weight_decay":
            reference_weight_decay,

        "legacy_inverse_time_decay":
            LEGACY_DECAY,

        "lr_definition":
            (
                "lr(global_step) = "
                "0.001 / (1 + 1e-6 * global_step)"
            ),
    },

    "executed_batch_size":
        256,

    "steps_per_epoch": {

        "Pingan":
            312,

        "Qingyun":
            262,
    },

    "epochs":
        100,

    "mixed_precision":
        False,

    "environment": {

        "python":
            "3.12.13",

        "pytorch":
            str(
                torch.__version__
            ),

        "cuda":
            str(
                torch.version.cuda
            ),
    },

    "training_started":
        False,

    "official_test_used":
        False,
}


# Restart-safe write
if EXECUTED_CONFIG_PATH.exists():

    with open(
        EXECUTED_CONFIG_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing = json.load(
            f
        )


    assert (
        existing
        ==
        EXECUTED_CONFIG_OBJECT
    ), (
        "Existing exact executed config differs. STOP."
    )


else:

    with open(
        EXECUTED_CONFIG_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            EXECUTED_CONFIG_OBJECT,
            f,
            indent=2,
            sort_keys=True
        )


EXECUTED_CONFIG_SHA = sha256_file(
    EXECUTED_CONFIG_PATH
)


# ==================================================================================================
# 7. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_EXACT_EXECUTED_CONFIG_GATE_PASS = True


print()
print("=" * 145)

print(
    "Exact executed CE configuration          : RECOVERED"
)

print(
    "Training started                         :",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used                       :",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "Exact executed config artifact:"
)

print(
    EXECUTED_CONFIG_PATH
)

print(
    "SHA-256:"
)

print(
    EXECUTED_CONFIG_SHA
)

print()

print(
    "NOTEBOOK 05 EXACT EXECUTED CE CONFIG GATE: PASS"
)

print(
    "READY TO BUILD DETERMINISTIC RECOVERY ENGINE"
)

print(
    "NO RECOVERY TRAINING HAS STARTED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 14B
# ORIGINAL NOTEBOOK 03 SOURCE AUDIT
#
# PURPOSE:
#   Recover exact:
#       - seed function
#       - deterministic settings
#       - DataLoader construction
#       - shuffle behavior
#       - generator usage
#       - optimizer creation
#       - LR update logic
#
# READ ONLY
# NO TRAINING
# NO OFFICIAL TEST DATA ACCESS
# ==================================================================================================

from pathlib import Path
import json
import re


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_EXACT_EXECUTED_CONFIG_GATE_PASS is True
assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


# ==================================================================================================
# 2. LOCATE NOTEBOOK FILES
# ==================================================================================================

NOTEBOOK_DIR = (
    ROOT /
    "notebooks"
)


assert NOTEBOOK_DIR.exists()


ipynb_files = sorted(
    NOTEBOOK_DIR.glob(
        "*.ipynb"
    )
)


print("=" * 145)
print("NOTEBOOK 05 — ORIGINAL NOTEBOOK 03 SOURCE AUDIT")
print("=" * 145)

print(
    "Notebook directory:",
    NOTEBOOK_DIR
)

print()

print(
    "Notebook files found:",
    len(
        ipynb_files
    )
)


for path in ipynb_files:

    print(
        path.name
    )


# ==================================================================================================
# 3. IDENTIFY LIKELY NOTEBOOK 03 FILE(S)
# ==================================================================================================

candidate_files = []


for path in ipynb_files:

    name_lower = (
        path.name.lower()
    )


    if (
        "03"
        in name_lower
        or
        "hybridsn"
        in name_lower
        or
        "baseline"
        in name_lower
    ):

        candidate_files.append(
            path
        )


print()
print("-" * 145)
print("LIKELY NOTEBOOK 03 SOURCE FILES")
print("-" * 145)


for path in candidate_files:

    print(
        path.name
    )


assert (
    len(
        candidate_files
    )
    >
    0
), (
    "No likely Notebook 03 .ipynb file found."
)


# ==================================================================================================
# 4. SEARCH TERMS
# ==================================================================================================

SEARCH_TERMS = [

    "random.seed",

    "np.random.seed",

    "numpy.random.seed",

    "torch.manual_seed",

    "torch.cuda.manual_seed",

    "torch.cuda.manual_seed_all",

    "use_deterministic_algorithms",

    "cudnn.deterministic",

    "cudnn.benchmark",

    "DataLoader",

    "shuffle=",

    "generator=",

    "worker_init_fn",

    "num_workers",

    "pin_memory",

    "drop_last",

    "torch.optim.Adam",

    "Adam(",

    "legacy_lr",

    "global_step",

    "learning_rate",

    "param_groups",

    "CrossEntropyLoss",

]


# ==================================================================================================
# 5. READ CODE CELLS + PRINT ONLY RELEVANT CELLS
# ==================================================================================================

SOURCE_FINDINGS = []


for notebook_path in candidate_files:

    print()
    print("=" * 145)

    print(
        "SOURCE:",
        notebook_path.name
    )

    print("=" * 145)


    with open(
        notebook_path,
        "r",
        encoding="utf-8"
    ) as f:

        notebook = json.load(
            f
        )


    cells = notebook.get(
        "cells",
        []
    )


    matching_cell_count = 0


    for cell_index, cell in enumerate(
        cells,
        start=1
    ):

        if (
            cell.get(
                "cell_type"
            )
            !=
            "code"
        ):

            continue


        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        matched_terms = [

            term

            for term in
            SEARCH_TERMS

            if term.lower()
            in source.lower()
        ]


        if (
            len(
                matched_terms
            )
            ==
            0
        ):

            continue


        matching_cell_count += 1


        print()
        print("-" * 145)

        print(
            f"CELL {cell_index}"
        )

        print(
            "Matched:",
            matched_terms
        )

        print("-" * 145)


        # Avoid gigantic output.
        if (
            len(
                source
            )
            >
            12000
        ):

            print(
                source[
                    :12000
                ]
            )

            print(
                "\n... SOURCE TRUNCATED ..."
            )


        else:

            print(
                source
            )


        SOURCE_FINDINGS.append({

            "notebook":
                notebook_path.name,

            "cell":
                int(
                    cell_index
                ),

            "matched_terms":
                matched_terms,
        })


    print()

    print(
        "Matching code cells:",
        matching_cell_count
    )


# ==================================================================================================
# 6. SAVE SOURCE-AUDIT RECORD
# ==================================================================================================

SOURCE_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_notebook03_source_engine_audit.json"
)


SOURCE_AUDIT_OBJECT = {

    "project":
        "LTLC",

    "purpose":
        "recover_exact_Notebook03_seed_and_dataloader_behavior",

    "candidate_notebooks":
        [
            str(
                p.relative_to(
                    ROOT
                )
            )

            for p in
            candidate_files
        ],

    "matching_cells":
        SOURCE_FINDINGS,

    "training_started":
        False,

    "official_test_used":
        False,
}


with open(
    SOURCE_AUDIT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        SOURCE_AUDIT_OBJECT,
        f,
        indent=2,
        sort_keys=True
    )


SOURCE_AUDIT_SHA = (
    sha256_file(
        SOURCE_AUDIT_PATH
    )
)


# ==================================================================================================
# 7. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_NOTEBOOK03_SOURCE_AUDIT_PASS = True


print()
print("=" * 145)

print(
    "Notebook 03 candidate source located : PASS"
)

print(
    "Relevant code cells extracted        :",
    len(
        SOURCE_FINDINGS
    )
)

print()

print(
    "Source audit artifact:"
)

print(
    SOURCE_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    SOURCE_AUDIT_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 NOTEBOOK03 SOURCE ENGINE AUDIT: PASS"
)

print(
    "NO RECOVERY TRAINING HAS STARTED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 14C
# TARGETED NOTEBOOK 03 DETERMINISM / DATALOADER / OPTIMIZER SOURCE EXTRACTION
#
# PURPOSE:
#   Identify the actual Notebook 03 source and print only small snippets around:
#       - random seeding
#       - deterministic CUDA/cuDNN settings
#       - training DataLoader construction
#       - shuffle / generator / workers
#       - Adam creation
#       - global_step / legacy LR update
#
# READ ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

from pathlib import Path
import json
import re


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_NOTEBOOK03_SOURCE_AUDIT_PASS is True
assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK_DIR = ROOT / "notebooks"

assert NOTEBOOK_DIR.exists()


# ==================================================================================================
# 2. SCORE NOTEBOOKS TO IDENTIFY ACTUAL NOTEBOOK 03
# ==================================================================================================

ipynb_files = sorted(
    NOTEBOOK_DIR.glob("*.ipynb")
)


NOTEBOOK_SCORES = []


for path in ipynb_files:

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        nb = json.load(f)


    all_source = "\n".join(

        "".join(
            cell.get("source", [])
        )

        for cell in nb.get("cells", [])
    )


    upper = all_source.upper()

    score = 0


    if "NOTEBOOK 03" in upper:
        score += 50

    if "HYBRIDSN_CE" in upper:
        score += 20

    if "HYBRIDSN CE" in upper:
        score += 15

    if "CE_CHECKPOINT_MANIFEST" in upper:
        score += 15

    if "BEST_VALIDATION_OA" in upper:
        score += 10

    if "HYBRIDSN_CE_BASELINE" in upper:
        score += 10


    # Penalize obvious Notebook 04 long-tail source.
    if "NOTEBOOK 04" in upper:
        score -= 40

    if "LDAM_DRW" in upper:
        score -= 15

    if "BALANCEDSOFTMAX" in upper:
        score -= 15


    NOTEBOOK_SCORES.append(
        (
            score,
            path,
            nb
        )
    )


NOTEBOOK_SCORES.sort(
    key=lambda x: x[0],
    reverse=True
)


print("=" * 145)
print("NOTEBOOK 05 — TARGETED NOTEBOOK 03 SOURCE EXTRACTION")
print("=" * 145)

print()
print("Notebook source scores:")

for score, path, _ in NOTEBOOK_SCORES:

    print(
        f"{score:4d} | {path.name}"
    )


assert len(
    NOTEBOOK_SCORES
) > 0


best_score, NB03_SOURCE_PATH, NB03_SOURCE = (
    NOTEBOOK_SCORES[
        0
    ]
)


assert best_score > 0, (
    "Could not confidently identify Notebook 03."
)


# Require unique winner.
if len(
    NOTEBOOK_SCORES
) > 1:

    assert (
        best_score
        >
        NOTEBOOK_SCORES[
            1
        ][
            0
        ]
    ), (
        "Notebook 03 source identification is ambiguous. STOP."
    )


print()
print(
    "Selected Notebook 03 source:"
)

print(
    NB03_SOURCE_PATH.name
)

print(
    "Selection score:",
    best_score
)


# ==================================================================================================
# 3. ENGINE-CRITICAL SEARCH PATTERNS
# ==================================================================================================

PATTERNS = {

    "PYTHON_RANDOM_SEED":
        r"random\.seed",

    "NUMPY_RANDOM_SEED":
        r"(np|numpy)\.random\.seed",

    "TORCH_MANUAL_SEED":
        r"torch\.manual_seed",

    "TORCH_CUDA_SEED":
        r"torch\.cuda\.manual_seed",

    "TORCH_CUDA_SEED_ALL":
        r"torch\.cuda\.manual_seed_all",

    "DETERMINISTIC_ALGORITHMS":
        r"use_deterministic_algorithms",

    "CUDNN_DETERMINISTIC":
        r"cudnn\.deterministic",

    "CUDNN_BENCHMARK":
        r"cudnn\.benchmark",

    "DATALOADER":
        r"DataLoader\s*\(",

    "SHUFFLE":
        r"shuffle\s*=",

    "GENERATOR":
        r"generator\s*=",

    "WORKER_INIT":
        r"worker_init_fn\s*=",

    "NUM_WORKERS":
        r"num_workers\s*=",

    "PIN_MEMORY":
        r"pin_memory\s*=",

    "DROP_LAST":
        r"drop_last\s*=",

    "ADAM":
        r"(torch\.optim\.Adam|optim\.Adam)\s*\(",

    "CROSS_ENTROPY":
        r"CrossEntropyLoss\s*\(",

    "GLOBAL_STEP":
        r"global_step",

    "LEGACY_DECAY":
        r"legacy_lr_decay",

    "PARAM_GROUP_LR":
        r"param_groups.*[\"']lr[\"']",
}


# ==================================================================================================
# 4. PRINT SMALL CONTEXT WINDOWS ONLY
# ==================================================================================================

MATCH_RECORDS = []


cells = NB03_SOURCE.get(
    "cells",
    []
)


for cell_number, cell in enumerate(
    cells,
    start=1
):

    if (
        cell.get(
            "cell_type"
        )
        !=
        "code"
    ):

        continue


    source = "".join(
        cell.get(
            "source",
            []
        )
    )


    lines = source.splitlines()


    matched_line_numbers = []


    for line_index, line in enumerate(
        lines
    ):

        matched_names = []


        for name, pattern in PATTERNS.items():

            if re.search(
                pattern,
                line,
                flags=re.IGNORECASE
            ):

                matched_names.append(
                    name
                )


        if matched_names:

            matched_line_numbers.append(
                (
                    line_index,
                    matched_names
                )
            )


    if not matched_line_numbers:

        continue


    # ----------------------------------------------------------------------------------------------
    # Merge overlapping ±8-line windows.
    # ----------------------------------------------------------------------------------------------

    ranges = []


    for line_index, _ in matched_line_numbers:

        start = max(
            0,
            line_index - 8
        )

        end = min(
            len(lines),
            line_index + 9
        )


        if (
            ranges
            and
            start <= ranges[-1][1]
        ):

            ranges[-1] = (
                ranges[-1][0],
                max(
                    ranges[-1][1],
                    end
                )
            )

        else:

            ranges.append(
                (
                    start,
                    end
                )
            )


    print()
    print("=" * 145)
    print(
        f"NOTEBOOK 03 CODE CELL {cell_number}"
    )
    print("=" * 145)


    for start, end in ranges:

        print(
            f"\n--- source lines {start + 1} to {end} ---"
        )


        for i in range(
            start,
            end
        ):

            marker = "   "


            for matched_index, _ in matched_line_numbers:

                if i == matched_index:

                    marker = ">>>"
                    break


            print(
                f"{marker} {i + 1:4d}: {lines[i]}"
            )


    MATCH_RECORDS.append({

        "cell":
            int(
                cell_number
            ),

        "matched_patterns":
            sorted(
                set(

                    name

                    for _, names in matched_line_numbers

                    for name in names
                )
            ),
    })


# ==================================================================================================
# 5. REQUIRED EVIDENCE SUMMARY
# ==================================================================================================

FOUND_PATTERNS = sorted(
    set(

        pattern

        for record in MATCH_RECORDS

        for pattern in record[
            "matched_patterns"
        ]
    )
)


print()
print("=" * 145)
print("ENGINE SOURCE EVIDENCE SUMMARY")
print("=" * 145)


for name in PATTERNS:

    print(
        f"{name:28s}: "
        f"{'FOUND' if name in FOUND_PATTERNS else 'NOT FOUND'}"
    )


print()
print(
    "Relevant Notebook 03 code cells:",
    len(
        MATCH_RECORDS
    )
)


# ==================================================================================================
# 6. SAVE TARGETED SOURCE AUDIT
# ==================================================================================================

TARGETED_SOURCE_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_notebook03_targeted_engine_source_audit.json"
)


TARGETED_SOURCE_AUDIT = {

    "selected_notebook":
        str(
            NB03_SOURCE_PATH.relative_to(
                ROOT
            )
        ),

    "selection_score":
        int(
            best_score
        ),

    "matching_cells":
        MATCH_RECORDS,

    "found_patterns":
        FOUND_PATTERNS,

    "training_started":
        False,

    "official_test_used":
        False,
}


with open(
    TARGETED_SOURCE_AUDIT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        TARGETED_SOURCE_AUDIT,
        f,
        indent=2,
        sort_keys=True
    )


TARGETED_SOURCE_AUDIT_SHA = sha256_file(
    TARGETED_SOURCE_AUDIT_PATH
)


# ==================================================================================================
# 7. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_TARGETED_ENGINE_SOURCE_AUDIT_PASS = True


print()
print("=" * 145)

print(
    "Target Notebook 03 identified : PASS"
)

print(
    "Targeted engine extraction    : PASS"
)

print()

print(
    "Audit artifact:"
)

print(
    TARGETED_SOURCE_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    TARGETED_SOURCE_AUDIT_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 TARGETED ENGINE SOURCE AUDIT: PASS"
)

print(
    "NO RECOVERY TRAINING HAS STARTED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 14D
# FINAL NOTEBOOK 03 SOURCE EXTRACTION
#
# EXACTLY RECOVER:
#   1. full-run setup order
#   2. validation evaluator
#   3. checkpoint comparison logic
#
# READ ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
import re


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_TARGETED_ENGINE_SOURCE_AUDIT_PASS is True

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


assert "NB03_SOURCE" in globals()

assert "NB03_SOURCE_PATH" in globals()


print("=" * 145)
print("NOTEBOOK 05 — FINAL NOTEBOOK 03 ENGINE SOURCE EXTRACTION")
print("=" * 145)

print(
    "Source:",
    NB03_SOURCE_PATH.name
)


# ==================================================================================================
# 2. COLLECT NOTEBOOK 03 CODE CELLS
# ==================================================================================================

CODE_CELLS = []


for notebook_cell_number, cell in enumerate(
    NB03_SOURCE.get(
        "cells",
        []
    ),
    start=1
):

    if (
        cell.get(
            "cell_type"
        )
        !=
        "code"
    ):

        continue


    source = "".join(
        cell.get(
            "source",
            []
        )
    )


    CODE_CELLS.append({

        "cell":
            notebook_cell_number,

        "source":
            source,

        "lines":
            source.splitlines(),
    })


# ==================================================================================================
# 3. HELPER — PRINT SMALL CONTEXT AROUND MATCHES
# ==================================================================================================

def print_match_context(
    title,
    patterns,
    before=12,
    after=16
):

    print()
    print("=" * 145)

    print(
        title
    )

    print("=" * 145)


    total_matches = 0


    for record in CODE_CELLS:

        lines = record[
            "lines"
        ]


        matches = []


        for i, line in enumerate(
            lines
        ):

            if any(
                re.search(
                    pattern,
                    line,
                    flags=re.IGNORECASE
                )
                for pattern in patterns
            ):

                matches.append(
                    i
                )


        if not matches:

            continue


        total_matches += len(
            matches
        )


        # Build and merge context windows.
        ranges = []


        for i in matches:

            start = max(
                0,
                i - before
            )

            end = min(
                len(lines),
                i + after + 1
            )


            if (
                ranges
                and
                start <= ranges[-1][1]
            ):

                ranges[-1] = (
                    ranges[-1][0],
                    max(
                        ranges[-1][1],
                        end
                    )
                )

            else:

                ranges.append(
                    (
                        start,
                        end
                    )
                )


        print()
        print(
            f"NOTEBOOK SOURCE CELL {record['cell']}"
        )


        for start, end in ranges:

            print(
                f"--- lines {start + 1} to {end} ---"
            )


            for i in range(
                start,
                end
            ):

                is_match = any(

                    re.search(
                        pattern,
                        lines[i],
                        flags=re.IGNORECASE
                    )

                    for pattern in patterns
                )


                marker = (
                    ">>>"
                    if is_match
                    else "   "
                )


                print(
                    f"{marker} {i + 1:4d}: {lines[i]}"
                )


    print()

    print(
        "Matches:",
        total_matches
    )


    return total_matches


# ==================================================================================================
# 4. FULL-RUN SETUP ORDER
#
# Need to know exactly where:
#   seed
#   dataloaders
#   model
#   criterion
#   optimizer
#   global_step
# occur relative to one another.
# ==================================================================================================

RUN_ORDER_MATCHES = print_match_context(

    title=
        "A. FULL CE RUN — INITIALIZATION ORDER",

    patterns=[

        r"set_global_seed\s*\(",

        r"build_dataloaders\s*\(",

        r"HybridSN\s*\(",

        r"criterion\s*=\s*nn\.CrossEntropyLoss",

        r"optimizer\s*=\s*build_ce_optimizer",

        r"global_step\s*=\s*0",

        r"for\s+epoch\s+in",
    ],

    before=18,

    after=20,
)


assert RUN_ORDER_MATCHES > 0


# ==================================================================================================
# 5. LOCATE EXACT VALIDATION EVALUATOR
# ==================================================================================================

EVALUATOR_RECORDS = []


for record in CODE_CELLS:

    if re.search(
        r"def\s+evaluate_model\s*\(",
        record[
            "source"
        ]
    ):

        EVALUATOR_RECORDS.append(
            record
        )


print()
print("=" * 145)
print("B. EXACT VALIDATION EVALUATOR")
print("=" * 145)


assert (
    len(
        EVALUATOR_RECORDS
    )
    ==
    1
), (
    "Expected exactly one evaluate_model definition."
)


eval_record = (
    EVALUATOR_RECORDS[
        0
    ]
)


eval_lines = (
    eval_record[
        "lines"
    ]
)


# Find start of evaluator.
eval_start = next(

    i

    for i, line in enumerate(
        eval_lines
    )

    if re.search(
        r"def\s+evaluate_model\s*\(",
        line
    )
)


# Find next top-level function definition.
eval_end = len(
    eval_lines
)


for i in range(
    eval_start + 1,
    len(
        eval_lines
    )
):

    if re.match(
        r"^def\s+\w+\s*\(",
        eval_lines[
            i
        ]
    ):

        eval_end = i
        break


print(
    f"Notebook source cell: {eval_record['cell']}"
)

print(
    f"Source lines: {eval_start + 1} to {eval_end}"
)

print()


for i in range(
    eval_start,
    eval_end
):

    print(
        f"{i + 1:4d}: {eval_lines[i]}"
    )


# ==================================================================================================
# 6. CHECKPOINT / IMPROVEMENT RULE
# ==================================================================================================

CHECKPOINT_MATCHES = print_match_context(

    title=
        "C. CE CHECKPOINT IMPROVEMENT / SELECTION LOGIC",

    patterns=[

        r"def\s+.*better.*checkpoint",

        r"def\s+.*checkpoint.*better",

        r"improved\s*=",

        r"best_record",

        r"validation_OA",

        r"checkpoint_saved",
    ],

    before=15,

    after=22,
)


assert CHECKPOINT_MATCHES > 0


# ==================================================================================================
# 7. TRAIN-EPOCH DATA TRANSFER / ZERO-GRAD / LOSS ORDER
# ==================================================================================================

TRAIN_STEP_MATCHES = print_match_context(

    title=
        "D. TRAINING STEP ORDER",

    patterns=[

        r"batch_x\s*=\s*batch_x\.to",

        r"batch_y\s*=\s*batch_y\.to",

        r"optimizer\.zero_grad",

        r"model\s*\(\s*batch_x",

        r"criterion\s*\(",

        r"loss\.backward",

        r"optimizer\.step",

        r"global_step\s*\+=",
    ],

    before=10,

    after=14,
)


assert TRAIN_STEP_MATCHES > 0


# ==================================================================================================
# 8. SAVE FINAL SOURCE-EVIDENCE AUDIT
# ==================================================================================================

FINAL_SOURCE_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_notebook03_final_recovery_engine_source_audit.json"
)


FINAL_SOURCE_AUDIT = {

    "project":
        "LTLC",

    "purpose":
        "final_source_evidence_before_best_valAA_recovery",

    "notebook03_source":
        str(
            NB03_SOURCE_PATH.relative_to(
                ROOT
            )
        ),

    "full_run_order_matches":
        int(
            RUN_ORDER_MATCHES
        ),

    "evaluate_model_source_cell":
        int(
            eval_record[
                "cell"
            ]
        ),

    "checkpoint_logic_matches":
        int(
            CHECKPOINT_MATCHES
        ),

    "training_step_matches":
        int(
            TRAIN_STEP_MATCHES
        ),

    "training_started":
        False,

    "official_test_used":
        False,
}


with open(
    FINAL_SOURCE_AUDIT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        FINAL_SOURCE_AUDIT,
        f,
        indent=2,
        sort_keys=True
    )


FINAL_SOURCE_AUDIT_SHA = (
    sha256_file(
        FINAL_SOURCE_AUDIT_PATH
    )
)


# ==================================================================================================
# 9. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_TRAINING_STARTED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_FINAL_ENGINE_SOURCE_GATE_PASS = True


print()
print("=" * 145)

print(
    "Full CE run initialization order : FOUND"
)

print(
    "Exact validation evaluator       : FOUND"
)

print(
    "Checkpoint comparison logic      : FOUND"
)

print(
    "Training-step order              : FOUND"
)

print()

print(
    "Final source audit:"
)

print(
    FINAL_SOURCE_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    FINAL_SOURCE_AUDIT_SHA
)

print()

print(
    "Training started:",
    NOTEBOOK05_RECOVERY_TRAINING_STARTED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 FINAL RECOVERY ENGINE SOURCE GATE: PASS"
)

print(
    "READY FOR ONE-EPOCH HISTORICAL REPRODUCTION TEST"
)

print(
    "NO FULL RECOVERY TRAINING HAS STARTED"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 15
# ONE-EPOCH HISTORICAL REPRODUCTION TEST
#
# DATASET : Pingan
# SEED    : 123
# EPOCHS  : 1
#
# PURPOSE:
#   Reproduce the ORIGINAL Notebook 03 epoch-1 trajectory
#   before allowing the 5 best-val-AA recovery runs.
#
# IMPORTANT:
#   - TRAIN + VALIDATION ONLY
#   - OFFICIAL TEST IS NOT CREATED OR USED
#   - NO MODEL CHECKPOINT IS SAVED
# ==================================================================================================

import ast
import copy
import json
import math
import os
import random
import gc

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader

# Import common metric names in case the exact Notebook 03
# compute_classification_metrics() function references them.
from sklearn.metrics import (
    confusion_matrix,
    cohen_kappa_score,
    f1_score,
    accuracy_score,
    recall_score,
)


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_FINAL_ENGINE_SOURCE_GATE_PASS is True
assert NOTEBOOK05_EXACT_EXECUTED_CONFIG_GATE_PASS is True
assert NOTEBOOK05_RECOVERY_PIPELINE_GATE_PASS is True
assert NOTEBOOK05_RECOVERY_SPLIT_LOAD_PASS is True

assert torch.cuda.is_available()
assert "T4" in torch.cuda.get_device_name(0).upper()

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


DEVICE = torch.device(
    "cuda:0"
)


# ==================================================================================================
# 2. EXTRACT EXACT EXECUTED NOTEBOOK 03 DEFINITIONS
#
# Instead of rewriting them manually, recover the actual source definitions:
#
#   set_global_seed
#   HybridSN
#   compute_classification_metrics
#   evaluate_model
#   legacy_keras_learning_rate
#   train_one_epoch
#   build_ce_optimizer
#
# This minimizes implementation drift.
# ==================================================================================================

REQUIRED_DEFINITIONS = {

    "set_global_seed",

    "HybridSN",

    "compute_classification_metrics",

    "evaluate_model",

    "legacy_keras_learning_rate",

    "train_one_epoch",

    "build_ce_optimizer",
}


FOUND_DEFINITIONS = {}


for cell in NB03_SOURCE.get(
    "cells",
    []
):

    if (
        cell.get(
            "cell_type"
        )
        !=
        "code"
    ):

        continue


    source = "".join(
        cell.get(
            "source",
            []
        )
    )


    try:

        tree = ast.parse(
            source
        )

    except SyntaxError:

        continue


    for node in tree.body:

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef
            )
        ):

            if (
                node.name
                in
                REQUIRED_DEFINITIONS
            ):

                if (
                    node.name
                    in
                    FOUND_DEFINITIONS
                ):

                    raise RuntimeError(
                        f"Duplicate Notebook 03 definition found: "
                        f"{node.name}"
                    )


                module = ast.Module(
                    body=[
                        node
                    ],
                    type_ignores=[]
                )


                ast.fix_missing_locations(
                    module
                )


                FOUND_DEFINITIONS[
                    node.name
                ] = module


assert (
    set(
        FOUND_DEFINITIONS.keys()
    )
    ==
    REQUIRED_DEFINITIONS
), (
    "Could not recover all required Notebook 03 definitions.\n"
    f"Found: {sorted(FOUND_DEFINITIONS.keys())}"
)


# ==================================================================================================
# 3. EXACT EXECUTED CE CONFIG
# ==================================================================================================

CE_TRAINING_CONFIG = copy.deepcopy(
    reference_training_config
)


assert CE_TRAINING_CONFIG == {

    "batch_size":
        256,

    "beta1":
        0.9,

    "beta2":
        0.999,

    "checkpoint_primary":
        "validation_OA",

    "checkpoint_tiebreak_1":
        "validation_AA",

    "checkpoint_tiebreak_2":
        "validation_loss",

    "dropout":
        0.4,

    "epochs":
        100,

    "epsilon":
        1e-7,

    "learning_rate":
        1e-3,

    "legacy_lr_decay":
        1e-6,

    "mixed_precision":
        False,

    "optimizer":
        "Adam",
}


# ==================================================================================================
# 4. EXECUTE EXACT NOTEBOOK 03 DEFINITIONS
# ==================================================================================================

for name in [

    "set_global_seed",

    "HybridSN",

    "compute_classification_metrics",

    "evaluate_model",

    "legacy_keras_learning_rate",

    "train_one_epoch",

    "build_ce_optimizer",
]:

    exec(
        compile(
            FOUND_DEFINITIONS[
                name
            ],
            filename=
                (
                    "03_HybridSN_Baseline_Training_"
                    "and_Benchmark_Reproduction.ipynb"
                ),
            mode="exec"
        ),
        globals()
    )


print("=" * 145)
print("NOTEBOOK 05 — ONE-EPOCH HISTORICAL REPRODUCTION")
print("=" * 145)

print(
    "Exact Notebook 03 definitions recovered :",
    len(
        FOUND_DEFINITIONS
    ),
    "/",
    len(
        REQUIRED_DEFINITIONS
    )
)


# ==================================================================================================
# 5. WRAPPER TO RETURN THE ORIGINAL CENTER INDEX
#
# Current Cell 13 dataset returns:
#       patch, label
#
# Original Notebook 03 evaluator expects:
#       patch, label, center_index
#
# This wrapper changes metadata output only.
# Patch and label generation remain unchanged.
# ==================================================================================================

class RecoveryIndexedDataset(
    Dataset
):

    def __init__(
        self,
        base_dataset,
        center_indices
    ):

        self.base_dataset = (
            base_dataset
        )


        self.center_indices = np.asarray(
            center_indices,
            dtype=np.int64
        )


        assert (
            len(
                self.base_dataset
            )
            ==
            len(
                self.center_indices
            )
        )


    def __len__(
        self
    ):

        return len(
            self.base_dataset
        )


    def __getitem__(
        self,
        item
    ):

        patch, label = (
            self.base_dataset[
                item
            ]
        )


        center_index = int(
            self.center_indices[
                item
            ]
        )


        return (
            patch,
            label,
            center_index
        )


# ==================================================================================================
# 6. PINGAN SEED-123 RECOVERY DATASETS
# ==================================================================================================

REPRO_DATASET = "Pingan"
REPRO_SEED = 123


train_dataset = RecoveryIndexedDataset(

    RECOVERY_TRAIN_DATASETS[
        REPRO_DATASET
    ],

    RECOVERY_SPLITS[
        REPRO_DATASET
    ][
        "train_indices"
    ]
)


val_dataset = RecoveryIndexedDataset(

    RECOVERY_VAL_DATASETS[
        REPRO_DATASET
    ],

    RECOVERY_SPLITS[
        REPRO_DATASET
    ][
        "val_indices"
    ]
)


assert len(
    train_dataset
) == 79869

assert len(
    val_dataset
) == 34230


# ==================================================================================================
# 7. EXACT ORIGINAL INITIALIZATION ORDER
#
# Notebook 03:
#
#   1. set_global_seed(seed)
#   2. build DataLoaders
#   3. initialize fresh model
#   4. CrossEntropyLoss
#   5. Adam
#   6. global_step = 0
# ==================================================================================================

set_global_seed(
    REPRO_SEED
)


# --------------------------------------------------------------------------------------------------
# Exact original training loader:
# Generator(seed), shuffle=True, workers=0, pin_memory=True, drop_last=False
# --------------------------------------------------------------------------------------------------

train_generator = torch.Generator()

train_generator.manual_seed(
    REPRO_SEED
)


train_loader = DataLoader(

    train_dataset,

    batch_size=
        256,

    shuffle=
        True,

    generator=
        train_generator,

    num_workers=
        0,

    pin_memory=
        True,

    drop_last=
        False,
)


# --------------------------------------------------------------------------------------------------
# Exact validation loader branch.
#
# We deliberately DO NOT reproduce the original factory's test-loader branch.
# --------------------------------------------------------------------------------------------------

val_loader = DataLoader(

    val_dataset,

    batch_size=
        256,

    shuffle=
        False,

    num_workers=
        0,

    pin_memory=
        True,

    drop_last=
        False,
)


# ==================================================================================================
# 8. VERIFY LOADER STEP COUNTS
# ==================================================================================================

assert len(
    train_loader
) == 312


assert len(
    val_loader
) == math.ceil(
    34230
    /
    256
)


# ==================================================================================================
# 9. EXACT NOTEBOOK 03 MODEL INITIALIZATION
# ==================================================================================================

model = HybridSN(

    spectral_depth=
        15,

    patch_size=
        13,

    num_classes=
        10,

    dropout=
        CE_TRAINING_CONFIG[
            "dropout"
        ]
).to(
    DEVICE
)


parameter_count = int(
    sum(
        parameter.numel()
        for parameter
        in model.parameters()
    )
)


assert (
    parameter_count
    ==
    519546
)


criterion = nn.CrossEntropyLoss()


optimizer = build_ce_optimizer(
    model
)


global_step = 0


# ==================================================================================================
# 10. HISTORICAL NOTEBOOK 03 EPOCH-1 TARGET
# ==================================================================================================

historical_history_path = (
    RECOVERY_HISTORY[
        (
            "Pingan",
            123
        )
    ]
)


historical_df = pd.read_csv(
    historical_history_path
)


historical_row = (

    historical_df[
        historical_df[
            "epoch"
        ].astype(
            int
        )
        ==
        1
    ]
    .iloc[
        0
    ]
)


HISTORICAL = {

    "train_loss":
        float(
            historical_row[
                "train_loss"
            ]
        ),

    "train_accuracy":
        float(
            historical_row[
                "train_accuracy"
            ]
        ),

    "val_loss":
        float(
            historical_row[
                "val_loss"
            ]
        ),

    "val_oa":
        float(
            historical_row[
                "val_oa"
            ]
        ),

    "val_aa":
        float(
            historical_row[
                "val_aa"
            ]
        ),

    "val_kappa":
        float(
            historical_row[
                "val_kappa"
            ]
        ),

    "val_macro_f1":
        float(
            historical_row[
                "val_macro_f1"
            ]
        ),

    "learning_rate":
        float(
            historical_row[
                "learning_rate"
            ]
        ),

    "global_step":
        int(
            historical_row[
                "global_step"
            ]
        ),
}


# ==================================================================================================
# 11. MARK TRUTHFUL TRAINING STATE
#
# From this point, a one-epoch smoke/reproduction training run really occurs.
# ==================================================================================================

NOTEBOOK05_RECOVERY_TRAINING_STARTED = True

NOTEBOOK05_RECOVERY_SMOKE_TRAINING_PERFORMED = True

NOTEBOOK05_RECOVERY_FULL_TRAINING_STARTED = False


# ==================================================================================================
# 12. RUN EXACTLY ONE TRAINING EPOCH
# ==================================================================================================

train_result = train_one_epoch(

    model=
        model,

    loader=
        train_loader,

    criterion=
        criterion,

    optimizer=
        optimizer,

    device=
        DEVICE,

    base_lr=
        CE_TRAINING_CONFIG[
            "learning_rate"
        ],

    legacy_decay=
        CE_TRAINING_CONFIG[
            "legacy_lr_decay"
        ],

    global_step=
        global_step
)


global_step = int(
    train_result[
        "global_step"
    ]
)


assert (
    global_step
    ==
    312
)


# ==================================================================================================
# 13. EXACT NOTEBOOK 03 VALIDATION EVALUATION
# ==================================================================================================

val_result = evaluate_model(

    model=
        model,

    loader=
        val_loader,

    criterion=
        criterion,

    num_classes=
        10,

    device=
        DEVICE
)


val_metrics = (
    val_result[
        "metrics"
    ]
)


# ==================================================================================================
# 14. VERIFY VALIDATION ORDER / INDICES
# ==================================================================================================

assert np.array_equal(

    np.asarray(
        val_result[
            "indices"
        ],
        dtype=np.int64
    ),

    RECOVERY_SPLITS[
        "Pingan"
    ][
        "val_indices"
    ]
)


assert np.array_equal(

    np.asarray(
        val_result[
            "labels"
        ],
        dtype=np.int64
    ),

    RECOVERY_SPLITS[
        "Pingan"
    ][
        "val_labels"
    ]
)


# ==================================================================================================
# 15. ACTUAL ONE-EPOCH RESULT
# ==================================================================================================

ACTUAL = {

    "train_loss":
        float(
            train_result[
                "loss"
            ]
        ),

    "train_accuracy":
        float(
            train_result[
                "accuracy"
            ]
        ),

    "val_loss":
        float(
            val_metrics[
                "loss"
            ]
        ),

    "val_oa":
        float(
            val_metrics[
                "oa"
            ]
        ),

    "val_aa":
        float(
            val_metrics[
                "aa"
            ]
        ),

    "val_kappa":
        float(
            val_metrics[
                "kappa"
            ]
        ),

    "val_macro_f1":
        float(
            val_metrics[
                "macro_f1"
            ]
        ),

    "learning_rate":
        float(
            train_result[
                "learning_rate"
            ]
        ),

    "global_step":
        int(
            global_step
        ),
}


# ==================================================================================================
# 16. STRICT REPRODUCTION TOLERANCES
#
# Loss uses GPU floating-point accumulation, so tolerance is 1e-7.
# Classification metrics should reproduce to near machine precision.
# ==================================================================================================

TOLERANCES = {

    "train_loss":
        1e-7,

    "train_accuracy":
        1e-12,

    "val_loss":
        1e-7,

    "val_oa":
        1e-12,

    "val_aa":
        1e-12,

    "val_kappa":
        1e-12,

    "val_macro_f1":
        1e-12,

    "learning_rate":
        1e-15,
}


REPRODUCTION_COMPARISON = {}


ALL_REPRO_METRICS_PASS = True


for metric in [

    "train_loss",

    "train_accuracy",

    "val_loss",

    "val_oa",

    "val_aa",

    "val_kappa",

    "val_macro_f1",

    "learning_rate",
]:

    historical_value = (
        HISTORICAL[
            metric
        ]
    )


    actual_value = (
        ACTUAL[
            metric
        ]
    )


    absolute_difference = abs(

        actual_value
        -
        historical_value
    )


    tolerance = (
        TOLERANCES[
            metric
        ]
    )


    passed = (
        absolute_difference
        <=
        tolerance
    )


    ALL_REPRO_METRICS_PASS = (

        ALL_REPRO_METRICS_PASS

        and

        passed
    )


    REPRODUCTION_COMPARISON[
        metric
    ] = {

        "historical":
            historical_value,

        "reproduced":
            actual_value,

        "absolute_difference":
            absolute_difference,

        "tolerance":
            tolerance,

        "pass":
            bool(
                passed
            ),
    }


GLOBAL_STEP_PASS = (

    ACTUAL[
        "global_step"
    ]
    ==
    HISTORICAL[
        "global_step"
    ]
)


ALL_REPRODUCTION_PASS = bool(

    ALL_REPRO_METRICS_PASS

    and

    GLOBAL_STEP_PASS
)


# ==================================================================================================
# 17. DISPLAY COMPARISON
# ==================================================================================================

print()
print("-" * 145)
print("PINGAN SEED 123 — EPOCH 1 HISTORICAL REPRODUCTION")
print("-" * 145)

print(
    f"{'Metric':18s} | "
    f"{'Historical':>18s} | "
    f"{'Reproduced':>18s} | "
    f"{'Abs Diff':>14s} | "
    f"{'Gate':>6s}"
)

print(
    "-" * 90
)


for metric in [

    "train_loss",

    "train_accuracy",

    "val_loss",

    "val_oa",

    "val_aa",

    "val_kappa",

    "val_macro_f1",

    "learning_rate",
]:

    item = (
        REPRODUCTION_COMPARISON[
            metric
        ]
    )


    print(

        f"{metric:18s} | "

        f"{item['historical']:18.12f} | "

        f"{item['reproduced']:18.12f} | "

        f"{item['absolute_difference']:14.3e} | "

        f"{'PASS' if item['pass'] else 'FAIL':>6s}"
    )


print()

print(
    "Historical global step :",
    HISTORICAL[
        "global_step"
    ]
)

print(
    "Reproduced global step :",
    ACTUAL[
        "global_step"
    ]
)

print(
    "Global-step gate       :",
    "PASS"
    if GLOBAL_STEP_PASS
    else "FAIL"
)


# ==================================================================================================
# 18. SAVE REPRODUCTION AUDIT
# ==================================================================================================

ONE_EPOCH_AUDIT_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_pingan_seed123_epoch1_historical_reproduction.json"
)


ONE_EPOCH_AUDIT = {

    "project":
        "LTLC",

    "purpose":
        "best_valAA_checkpoint_recovery_preflight",

    "dataset":
        "Pingan",

    "seed":
        123,

    "epochs_run":
        1,

    "source_notebook":
        str(
            NB03_SOURCE_PATH.relative_to(
                ROOT
            )
        ),

    "source_history":
        str(
            historical_history_path.relative_to(
                ROOT
            )
        ),

    "parameter_count":
        int(
            parameter_count
        ),

    "historical":
        HISTORICAL,

    "reproduced":
        ACTUAL,

    "comparison":
        REPRODUCTION_COMPARISON,

    "global_step_pass":
        bool(
            GLOBAL_STEP_PASS
        ),

    "all_metrics_pass":
        bool(
            ALL_REPRO_METRICS_PASS
        ),

    "overall_pass":
        bool(
            ALL_REPRODUCTION_PASS
        ),

    "model_checkpoint_saved":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


with open(
    ONE_EPOCH_AUDIT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        ONE_EPOCH_AUDIT,
        f,
        indent=2,
        sort_keys=True
    )


ONE_EPOCH_AUDIT_SHA = (
    sha256_file(
        ONE_EPOCH_AUDIT_PATH
    )
)


# ==================================================================================================
# 19. CLEANUP
# ==================================================================================================

del val_result
del model
del optimizer
del criterion

torch.cuda.empty_cache()

gc.collect()


# ==================================================================================================
# 20. FINAL SAFETY / REPRODUCTION GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


print()
print("=" * 145)

print(
    "One-epoch training actually performed : YES"
)

print(
    "Full recovery training started        :",
    NOTEBOOK05_RECOVERY_FULL_TRAINING_STARTED
)

print(
    "Model checkpoint saved                : NO"
)

print(
    "Official-test dataset created         :",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created      :",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used                    :",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "Reproduction audit:"
)

print(
    ONE_EPOCH_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    ONE_EPOCH_AUDIT_SHA
)

print()

print(
    "ONE-EPOCH HISTORICAL REPRODUCTION:",
    "PASS"
    if ALL_REPRODUCTION_PASS
    else "FAIL"
)

print("=" * 145)


if not ALL_REPRODUCTION_PASS:

    raise RuntimeError(
        "Historical epoch-1 reproduction failed. "
        "STOP — do not begin best-val-AA recovery."
    )


NOTEBOOK05_ONE_EPOCH_REPRODUCTION_GATE_PASS = True


print(
    "NOTEBOOK 05 ONE-EPOCH REPRODUCTION GATE: PASS"
)

print(
    "DETERMINISTIC RECOVERY ENGINE VERIFIED"
)

print(
    "READY FOR 5 BEST-VAL-AA RECOVERY RUNS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 16A
# BEST-VAL-AA CE CHECKPOINT RECOVERY
#
# DATASET : Pingan
# SEED    : 123
# TARGET  : Historical best-validation-AA epoch 97
#
# SAFETY:
#   - Fresh deterministic 100-epoch rerun
#   - Every epoch checked against frozen Notebook 03 history
#   - Target checkpoint held in memory
#   - NOTHING is written as final recovery checkpoint unless ALL 100 epochs reproduce
#   - NO OFFICIAL TEST
# ==================================================================================================

import copy
import gc
import json
import math
import time

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import DataLoader


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_ONE_EPOCH_REPRODUCTION_GATE_PASS is True

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False

assert torch.cuda.is_available()
assert "T4" in torch.cuda.get_device_name(0).upper()


DEVICE = torch.device(
    "cuda:0"
)


# ==================================================================================================
# 2. RUN IDENTITY
# ==================================================================================================

DATASET = "Pingan"
SEED = 123
TARGET_EPOCH = 97
TOTAL_EPOCHS = 100


assert (
    EXPECTED_RECOVERY_TARGETS[
        (
            DATASET,
            SEED
        )
    ]
    ==
    TARGET_EPOCH
)


cfg = RECOVERY_CONFIG[
    DATASET
]


split = RECOVERY_SPLITS[
    DATASET
]


# ==================================================================================================
# 3. OUTPUT PATHS
# ==================================================================================================

RECOVERY_RUN_NAME = (
    f"{DATASET.lower()}_seed{SEED}"
)


RECOVERY_CHECKPOINT_PATH = (
    BEST_AA_CHECKPOINT_DIR
    /
    f"{RECOVERY_RUN_NAME}_best_valAA.pt"
)


RECOVERY_HISTORY_PATH = (
    BEST_AA_HISTORY_DIR
    /
    f"{RECOVERY_RUN_NAME}_recovery_history.csv"
)


RECOVERY_METADATA_PATH = (
    BEST_AA_METADATA_DIR
    /
    f"{RECOVERY_RUN_NAME}_recovery_metadata.json"
)


# Final artifacts must not already partially exist.
artifact_existence = {

    "checkpoint":
        RECOVERY_CHECKPOINT_PATH.exists(),

    "history":
        RECOVERY_HISTORY_PATH.exists(),

    "metadata":
        RECOVERY_METADATA_PATH.exists(),
}


if any(
    artifact_existence.values()
):

    if not all(
        artifact_existence.values()
    ):

        raise RuntimeError(
            "PARTIAL Pingan seed123 recovery artifacts detected. "
            "STOP — do not overwrite automatically."
        )

    else:

        raise RuntimeError(
            "Complete Pingan seed123 recovery artifacts already exist. "
            "STOP — audit them instead of retraining."
        )


# ==================================================================================================
# 4. ORIGINAL FROZEN HISTORY
# ==================================================================================================

ORIGINAL_HISTORY_PATH = (
    RECOVERY_HISTORY[
        (
            DATASET,
            SEED
        )
    ]
)


original_history = pd.read_csv(
    ORIGINAL_HISTORY_PATH
)


assert len(
    original_history
) == 100


assert np.array_equal(

    original_history[
        "epoch"
    ].to_numpy(
        dtype=np.int64
    ),

    np.arange(
        1,
        101,
        dtype=np.int64
    )
)


# ==================================================================================================
# 5. VERIFY FROZEN TARGET REALLY IS BEST AA
# ==================================================================================================

historical_max_aa = float(

    original_history[
        "val_aa"
    ].max()
)


target_row = (

    original_history[
        original_history[
            "epoch"
        ]
        ==
        TARGET_EPOCH
    ]
    .iloc[
        0
    ]
)


assert math.isclose(

    float(
        target_row[
            "val_aa"
        ]
    ),

    historical_max_aa,

    rel_tol=0.0,

    abs_tol=1e-15
), (
    "Frozen recovery target is not historical maximum validation AA."
)


# ==================================================================================================
# 6. DATASETS WITH ORIGINAL CENTER-INDEX OUTPUT
# ==================================================================================================

train_dataset = RecoveryIndexedDataset(

    RECOVERY_TRAIN_DATASETS[
        DATASET
    ],

    split[
        "train_indices"
    ]
)


val_dataset = RecoveryIndexedDataset(

    RECOVERY_VAL_DATASETS[
        DATASET
    ],

    split[
        "val_indices"
    ]
)


assert len(
    train_dataset
) == 79869

assert len(
    val_dataset
) == 34230


# ==================================================================================================
# 7. EXACT ORIGINAL INITIALIZATION ORDER
# ==================================================================================================

set_global_seed(
    SEED
)


train_generator = torch.Generator()

train_generator.manual_seed(
    SEED
)


train_loader = DataLoader(

    train_dataset,

    batch_size=
        CE_TRAINING_CONFIG[
            "batch_size"
        ],

    shuffle=True,

    generator=
        train_generator,

    num_workers=0,

    pin_memory=True,

    drop_last=False,
)


val_loader = DataLoader(

    val_dataset,

    batch_size=
        CE_TRAINING_CONFIG[
            "batch_size"
        ],

    shuffle=False,

    num_workers=0,

    pin_memory=True,

    drop_last=False,
)


assert len(
    train_loader
) == 312


model = HybridSN(

    spectral_depth=
        cfg[
            "pca_channels"
        ],

    patch_size=
        cfg[
            "patch_size"
        ],

    num_classes=
        cfg[
            "classes"
        ],

    dropout=
        CE_TRAINING_CONFIG[
            "dropout"
        ]
).to(
    DEVICE
)


assert int(
    sum(
        p.numel()
        for p in model.parameters()
    )
) == 519546


criterion = nn.CrossEntropyLoss()


optimizer = build_ce_optimizer(
    model
)


global_step = 0


# ==================================================================================================
# 8. HISTORICAL REPRODUCTION TOLERANCES
# ==================================================================================================

TOLERANCES = {

    "train_loss":
        1e-7,

    "train_accuracy":
        1e-12,

    "val_loss":
        1e-7,

    "val_oa":
        1e-12,

    "val_aa":
        1e-12,

    "val_kappa":
        1e-12,

    "val_macro_f1":
        1e-12,

    "learning_rate":
        1e-15,
}


# ==================================================================================================
# 9. CPU-SNAPSHOT HELPER
# ==================================================================================================

def recursive_to_cpu(
    obj
):

    if torch.is_tensor(
        obj
    ):

        return (
            obj.detach()
            .cpu()
            .clone()
        )


    if isinstance(
        obj,
        dict
    ):

        return {

            key:
                recursive_to_cpu(
                    value
                )

            for key, value in obj.items()
        }


    if isinstance(
        obj,
        list
    ):

        return [

            recursive_to_cpu(
                value
            )

            for value in obj
        ]


    if isinstance(
        obj,
        tuple
    ):

        return tuple(

            recursive_to_cpu(
                value
            )

            for value in obj
        )


    return copy.deepcopy(
        obj
    )


# ==================================================================================================
# 10. RUN STATE
# ==================================================================================================

NOTEBOOK05_RECOVERY_FULL_TRAINING_STARTED = True


recovery_records = []


target_model_state = None
target_optimizer_state = None
target_validation_metrics = None
target_global_step = None


maximum_differences = {

    metric:
        0.0

    for metric in
    TOLERANCES
}


full_start_time = time.time()


print("=" * 145)
print("NOTEBOOK 05 — BEST-VAL-AA RECOVERY")
print("=" * 145)

print(
    "Dataset             :",
    DATASET
)

print(
    "Seed                :",
    SEED
)

print(
    "Target best-AA epoch:",
    TARGET_EPOCH
)

print(
    "Epochs to reproduce :",
    TOTAL_EPOCHS
)

print(
    "Official test       : SEALED"
)

print("=" * 145)


# ==================================================================================================
# 11. FULL 100-EPOCH DETERMINISTIC REPRODUCTION
# ==================================================================================================

for epoch in range(
    1,
    TOTAL_EPOCHS + 1
):

    epoch_start = time.time()


    # ----------------------------------------------------------------------------------------------
    # Train
    # ----------------------------------------------------------------------------------------------

    train_result = train_one_epoch(

        model=model,

        loader=train_loader,

        criterion=criterion,

        optimizer=optimizer,

        device=DEVICE,

        base_lr=
            CE_TRAINING_CONFIG[
                "learning_rate"
            ],

        legacy_decay=
            CE_TRAINING_CONFIG[
                "legacy_lr_decay"
            ],

        global_step=
            global_step
    )


    global_step = int(
        train_result[
            "global_step"
        ]
    )


    # ----------------------------------------------------------------------------------------------
    # Validation only
    # ----------------------------------------------------------------------------------------------

    val_result = evaluate_model(

        model=model,

        loader=val_loader,

        criterion=criterion,

        num_classes=
            cfg[
                "classes"
            ],

        device=DEVICE
    )


    val_metrics = val_result[
        "metrics"
    ]


    # ----------------------------------------------------------------------------------------------
    # Verify ordered frozen validation set
    # ----------------------------------------------------------------------------------------------

    assert np.array_equal(

        np.asarray(
            val_result[
                "indices"
            ],
            dtype=np.int64
        ),

        split[
            "val_indices"
        ]
    )


    assert np.array_equal(

        np.asarray(
            val_result[
                "labels"
            ],
            dtype=np.int64
        ),

        split[
            "val_labels"
        ]
    )


    # ----------------------------------------------------------------------------------------------
    # Actual epoch record
    # ----------------------------------------------------------------------------------------------

    actual = {

        "train_loss":
            float(
                train_result[
                    "loss"
                ]
            ),

        "train_accuracy":
            float(
                train_result[
                    "accuracy"
                ]
            ),

        "val_loss":
            float(
                val_metrics[
                    "loss"
                ]
            ),

        "val_oa":
            float(
                val_metrics[
                    "oa"
                ]
            ),

        "val_aa":
            float(
                val_metrics[
                    "aa"
                ]
            ),

        "val_kappa":
            float(
                val_metrics[
                    "kappa"
                ]
            ),

        "val_macro_f1":
            float(
                val_metrics[
                    "macro_f1"
                ]
            ),

        "learning_rate":
            float(
                train_result[
                    "learning_rate"
                ]
            ),
    }


    historical_row = (
        original_history.iloc[
            epoch - 1
        ]
    )


    # ----------------------------------------------------------------------------------------------
    # Compare EVERY metric against historical Notebook 03 trajectory
    # ----------------------------------------------------------------------------------------------

    epoch_pass = True


    for metric, tolerance in (
        TOLERANCES.items()
    ):

        historical_value = float(
            historical_row[
                metric
            ]
        )


        difference = abs(

            actual[
                metric
            ]
            -
            historical_value
        )


        maximum_differences[
            metric
        ] = max(

            maximum_differences[
                metric
            ],

            difference
        )


        if difference > tolerance:

            epoch_pass = False


            raise RuntimeError(

                f"Historical trajectory mismatch.\n"
                f"Dataset={DATASET}\n"
                f"Seed={SEED}\n"
                f"Epoch={epoch}\n"
                f"Metric={metric}\n"
                f"Historical={historical_value:.15f}\n"
                f"Recovered={actual[metric]:.15f}\n"
                f"Difference={difference:.3e}\n"
                f"Tolerance={tolerance:.3e}\n\n"
                "STOP — recovery checkpoint will not be written."
            )


    historical_global_step = int(
        historical_row[
            "global_step"
        ]
    )


    if (
        global_step
        !=
        historical_global_step
    ):

        raise RuntimeError(

            f"Global-step mismatch at epoch {epoch}: "
            f"{global_step} != {historical_global_step}"
        )


    # ----------------------------------------------------------------------------------------------
    # Capture target best-AA state IN MEMORY ONLY
    # ----------------------------------------------------------------------------------------------

    if epoch == TARGET_EPOCH:

        target_model_state = (

            recursive_to_cpu(
                model.state_dict()
            )
        )


        target_optimizer_state = (

            recursive_to_cpu(
                optimizer.state_dict()
            )
        )


        target_global_step = int(
            global_step
        )


        target_validation_metrics = {

            "epoch":
                int(
                    epoch
                ),

            "oa":
                float(
                    val_metrics[
                        "oa"
                    ]
                ),

            "aa":
                float(
                    val_metrics[
                        "aa"
                    ]
                ),

            "kappa":
                float(
                    val_metrics[
                        "kappa"
                    ]
                ),

            "macro_f1":
                float(
                    val_metrics[
                        "macro_f1"
                    ]
                ),

            "loss":
                float(
                    val_metrics[
                        "loss"
                    ]
                ),
        }


    # ----------------------------------------------------------------------------------------------
    # Recovery history
    # ----------------------------------------------------------------------------------------------

    epoch_seconds = (
        time.time()
        -
        epoch_start
    )


    recovery_records.append({

        "epoch":
            int(
                epoch
            ),

        "train_loss":
            actual[
                "train_loss"
            ],

        "train_accuracy":
            actual[
                "train_accuracy"
            ],

        "val_loss":
            actual[
                "val_loss"
            ],

        "val_oa":
            actual[
                "val_oa"
            ],

        "val_aa":
            actual[
                "val_aa"
            ],

        "val_kappa":
            actual[
                "val_kappa"
            ],

        "val_macro_f1":
            actual[
                "val_macro_f1"
            ],

        "learning_rate":
            actual[
                "learning_rate"
            ],

        "global_step":
            int(
                global_step
            ),

        "epoch_seconds":
            float(
                epoch_seconds
            ),

        "historical_match":
            bool(
                epoch_pass
            ),

        "target_best_valAA_epoch":
            bool(
                epoch
                ==
                TARGET_EPOCH
            ),
    })


    # ----------------------------------------------------------------------------------------------
    # Compact progress
    # ----------------------------------------------------------------------------------------------

    if (
        epoch == 1
        or
        epoch % 10 == 0
        or
        epoch == TARGET_EPOCH
        or
        epoch == TOTAL_EPOCHS
    ):

        marker = (
            " <TARGET BEST-AA>"
            if epoch == TARGET_EPOCH
            else ""
        )


        print(

            f"Epoch {epoch:03d}/100 | "
            f"TrainLoss={actual['train_loss']:.6f} | "
            f"ValOA={actual['val_oa']:.6f} | "
            f"ValAA={actual['val_aa']:.6f} | "
            f"F1={actual['val_macro_f1']:.6f} | "
            f"HISTORICAL MATCH{marker}"
        )


    del val_result


# ==================================================================================================
# 12. FULL TRAJECTORY PASSED
# ==================================================================================================

assert len(
    recovery_records
) == 100


assert target_model_state is not None
assert target_optimizer_state is not None
assert target_validation_metrics is not None
assert target_global_step is not None


recovery_history_df = pd.DataFrame(
    recovery_records
)


assert recovery_history_df[
    "historical_match"
].all()


# ==================================================================================================
# 13. TARGET METRIC EXACTNESS
# ==================================================================================================

assert (
    target_validation_metrics[
        "epoch"
    ]
    ==
    TARGET_EPOCH
)


assert math.isclose(

    target_validation_metrics[
        "aa"
    ],

    float(
        target_row[
            "val_aa"
        ]
    ),

    rel_tol=0.0,

    abs_tol=1e-12
)


# ==================================================================================================
# 14. ORIGINAL SOURCE CHECKPOINT METADATA
# ==================================================================================================

source_best_oa_path = (
    EXISTING_BEST_OA[
        (
            DATASET,
            SEED
        )
    ]
)


source_checkpoint = torch.load(

    source_best_oa_path,

    map_location="cpu",

    weights_only=False
)


source_checkpoint_sha = sha256_file(
    source_best_oa_path
)


source_history_sha = sha256_file(
    ORIGINAL_HISTORY_PATH
)


# ==================================================================================================
# 15. BUILD RECOVERED BEST-VAL-AA CHECKPOINT
# ==================================================================================================

RECOVERED_CHECKPOINT = {

    "project":
        "CSE498R LTLC",

    "method":
        "HybridSN_CE",

    "dataset":
        DATASET,

    "seed":
        int(
            SEED
        ),

    "epoch":
        int(
            TARGET_EPOCH
        ),

    "model_state_dict":
        target_model_state,

    "optimizer_state_dict":
        target_optimizer_state,

    "global_step":
        int(
            target_global_step
        ),

    "validation_metrics":
        target_validation_metrics,

    "training_config":
        copy.deepcopy(
            CE_TRAINING_CONFIG
        ),

    "model_configuration": {

        "spectral_depth":
            int(
                cfg[
                    "pca_channels"
                ]
            ),

        "patch_size":
            int(
                cfg[
                    "patch_size"
                ]
            ),

        "num_classes":
            int(
                cfg[
                    "classes"
                ]
            ),

        "dropout":
            float(
                CE_TRAINING_CONFIG[
                    "dropout"
                ]
            ),
    },

    "split_sha256":
        source_checkpoint[
            "split_sha256"
        ],

    "cube_sha256":
        source_checkpoint[
            "cube_sha256"
        ],

    "checkpoint_policy":
        "best_validation_AA_recovered",

    "test_used_for_selection":
        False,

    "recovery": {

        "reason":
            (
                "Implementation-locked protocol requires "
                "best-validation-AA checkpoint; original "
                "Notebook 03 retained only best-validation-OA."
            ),

        "historical_target_epoch":
            int(
                TARGET_EPOCH
            ),

        "full_100_epoch_trajectory_reproduced":
            True,

        "all_100_epochs_match_original_history":
            True,

        "source_history":
            str(
                ORIGINAL_HISTORY_PATH.relative_to(
                    ROOT
                )
            ),

        "source_history_sha256":
            source_history_sha,

        "source_best_valOA_checkpoint":
            str(
                source_best_oa_path.relative_to(
                    ROOT
                )
            ),

        "source_best_valOA_checkpoint_sha256":
            source_checkpoint_sha,

        "one_epoch_reproduction_gate_sha256":
            ONE_EPOCH_AUDIT_SHA,

        "official_test_used":
            False,
    },
}


# ==================================================================================================
# 16. ONLY NOW WRITE FINAL ARTIFACTS
# ==================================================================================================

torch.save(

    RECOVERED_CHECKPOINT,

    RECOVERY_CHECKPOINT_PATH
)


recovery_history_df.to_csv(

    RECOVERY_HISTORY_PATH,

    index=False
)


total_seconds = (
    time.time()
    -
    full_start_time
)


RECOVERY_METADATA = {

    "dataset":
        DATASET,

    "seed":
        int(
            SEED
        ),

    "target_best_valAA_epoch":
        int(
            TARGET_EPOCH
        ),

    "target_validation_metrics":
        target_validation_metrics,

    "full_epochs_reproduced":
        100,

    "all_epochs_historical_match":
        True,

    "maximum_absolute_differences":
        {

            key:
                float(
                    value
                )

            for key, value in
            maximum_differences.items()
        },

    "checkpoint":
        str(
            RECOVERY_CHECKPOINT_PATH.relative_to(
                ROOT
            )
        ),

    "checkpoint_sha256":
        sha256_file(
            RECOVERY_CHECKPOINT_PATH
        ),

    "history":
        str(
            RECOVERY_HISTORY_PATH.relative_to(
                ROOT
            )
        ),

    "history_sha256":
        sha256_file(
            RECOVERY_HISTORY_PATH
        ),

    "source_history_sha256":
        source_history_sha,

    "source_best_valOA_checkpoint_sha256":
        source_checkpoint_sha,

    "total_recovery_seconds":
        float(
            total_seconds
        ),

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


with open(
    RECOVERY_METADATA_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        RECOVERY_METADATA,
        f,
        indent=2,
        sort_keys=True
    )


# ==================================================================================================
# 17. FINAL ARTIFACT AUDIT
# ==================================================================================================

assert RECOVERY_CHECKPOINT_PATH.exists()
assert RECOVERY_HISTORY_PATH.exists()
assert RECOVERY_METADATA_PATH.exists()


audit_checkpoint = torch.load(

    RECOVERY_CHECKPOINT_PATH,

    map_location="cpu",

    weights_only=False
)


assert (
    audit_checkpoint[
        "dataset"
    ]
    ==
    DATASET
)


assert (
    int(
        audit_checkpoint[
            "seed"
        ]
    )
    ==
    SEED
)


assert (
    int(
        audit_checkpoint[
            "epoch"
        ]
    )
    ==
    TARGET_EPOCH
)


assert (
    audit_checkpoint[
        "checkpoint_policy"
    ]
    ==
    "best_validation_AA_recovered"
)


assert (
    audit_checkpoint[
        "test_used_for_selection"
    ]
    is False
)


assert (
    audit_checkpoint[
        "recovery"
    ][
        "all_100_epochs_match_original_history"
    ]
    is True
)


RECOVERY_CHECKPOINT_SHA = sha256_file(
    RECOVERY_CHECKPOINT_PATH
)


RECOVERY_HISTORY_SHA = sha256_file(
    RECOVERY_HISTORY_PATH
)


RECOVERY_METADATA_SHA = sha256_file(
    RECOVERY_METADATA_PATH
)


# ==================================================================================================
# 18. STATE UPDATE + CLEANUP
# ==================================================================================================

NOTEBOOK05_RECOVERY_COMPLETED_RUNS = globals().get(

    "NOTEBOOK05_RECOVERY_COMPLETED_RUNS",

    []
)


if (
    (
        DATASET,
        SEED
    )
    not in
    NOTEBOOK05_RECOVERY_COMPLETED_RUNS
):

    NOTEBOOK05_RECOVERY_COMPLETED_RUNS.append(
        (
            DATASET,
            SEED
        )
    )


del model
del optimizer
del criterion
del source_checkpoint
del audit_checkpoint

torch.cuda.empty_cache()

gc.collect()


# ==================================================================================================
# 19. FINAL REPORT
# ==================================================================================================

print()
print("=" * 145)

print(
    "PINGAN SEED 123 BEST-VAL-AA RECOVERY: PASS"
)

print("=" * 145)

print(
    "Historical trajectory reproduced : 100/100 epochs PASS"
)

print(
    "Recovered target epoch           :",
    TARGET_EPOCH
)

print(
    "Recovered Val AA                 :",
    f"{target_validation_metrics['aa']:.12f}"
)

print(
    "Recovered Val OA                 :",
    f"{target_validation_metrics['oa']:.12f}"
)

print(
    "Recovered Macro-F1               :",
    f"{target_validation_metrics['macro_f1']:.12f}"
)

print()

print(
    "Maximum historical differences:"
)


for key, value in (
    maximum_differences.items()
):

    print(
        f"  {key:18s}: {value:.3e}"
    )


print()

print(
    "Checkpoint:"
)

print(
    RECOVERY_CHECKPOINT_PATH
)

print(
    "SHA-256:"
)

print(
    RECOVERY_CHECKPOINT_SHA
)

print()

print(
    "History SHA-256:"
)

print(
    RECOVERY_HISTORY_SHA
)

print(
    "Metadata SHA-256:"
)

print(
    RECOVERY_METADATA_SHA
)

print()

print(
    "Completed recovery runs:",
    len(
        NOTEBOOK05_RECOVERY_COMPLETED_RUNS
    ),
    "/ 5"
)

print(
    "Full recovery training started:",
    NOTEBOOK05_RECOVERY_FULL_TRAINING_STARTED
)

print(
    "Official-test dataset created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 BEST-VAL-AA RECOVERY RUN 1/5: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 16B
# GENERIC DETERMINISTIC BEST-VAL-AA RECOVERY ENGINE
# + RUN 2/5: PINGAN SEED 3407
#
# TARGET HISTORICAL BEST-VAL-AA EPOCH: 100
#
# TRAIN + VALIDATION ONLY
# OFFICIAL TEST REMAINS SEALED
# ==================================================================================================

import copy
import gc
import json
import math
import time

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import DataLoader


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_ONE_EPOCH_REPRODUCTION_GATE_PASS is True

assert "recursive_to_cpu" in globals()

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False

assert torch.cuda.is_available()
assert "T4" in torch.cuda.get_device_name(0).upper()


DEVICE = torch.device(
    "cuda:0"
)


# ==================================================================================================
# 2. HISTORICAL REPRODUCTION TOLERANCES
# ==================================================================================================

RECOVERY_TOLERANCES = {

    "train_loss":
        1e-7,

    "train_accuracy":
        1e-12,

    "val_loss":
        1e-7,

    "val_oa":
        1e-12,

    "val_aa":
        1e-12,

    "val_kappa":
        1e-12,

    "val_macro_f1":
        1e-12,

    "learning_rate":
        1e-15,
}


# ==================================================================================================
# 3. GENERIC RECOVERY FUNCTION
# ==================================================================================================

def recover_best_valAA_ce_run(
    dataset,
    seed,
    target_epoch
):

    global NOTEBOOK05_RECOVERY_FULL_TRAINING_STARTED
    global NOTEBOOK05_RECOVERY_COMPLETED_RUNS


    # ==============================================================================================
    # A. IDENTITY / FROZEN TARGET
    # ==============================================================================================

    dataset = str(
        dataset
    )

    seed = int(
        seed
    )

    target_epoch = int(
        target_epoch
    )


    assert (
        EXPECTED_RECOVERY_TARGETS[
            (
                dataset,
                seed
            )
        ]
        ==
        target_epoch
    )


    cfg = RECOVERY_CONFIG[
        dataset
    ]


    split = RECOVERY_SPLITS[
        dataset
    ]


    total_epochs = 100


    # ==============================================================================================
    # B. ARTIFACT PATHS
    # ==============================================================================================

    run_name = (
        f"{dataset.lower()}_seed{seed}"
    )


    checkpoint_path = (
        BEST_AA_CHECKPOINT_DIR
        /
        f"{run_name}_best_valAA.pt"
    )


    history_path = (
        BEST_AA_HISTORY_DIR
        /
        f"{run_name}_recovery_history.csv"
    )


    metadata_path = (
        BEST_AA_METADATA_DIR
        /
        f"{run_name}_recovery_metadata.json"
    )


    existence = {

        "checkpoint":
            checkpoint_path.exists(),

        "history":
            history_path.exists(),

        "metadata":
            metadata_path.exists(),
    }


    if any(
        existence.values()
    ):

        if not all(
            existence.values()
        ):

            raise RuntimeError(
                f"{dataset} seed {seed}: "
                "PARTIAL recovery artifacts exist. "
                "STOP — no automatic overwrite."
            )


        raise RuntimeError(
            f"{dataset} seed {seed}: "
            "complete recovery artifacts already exist. "
            "Audit instead of retraining."
        )


    # ==============================================================================================
    # C. ORIGINAL FROZEN NOTEBOOK 03 HISTORY
    # ==============================================================================================

    original_history_path = (
        RECOVERY_HISTORY[
            (
                dataset,
                seed
            )
        ]
    )


    original_history = pd.read_csv(
        original_history_path
    )


    assert len(
        original_history
    ) == 100


    assert np.array_equal(

        original_history[
            "epoch"
        ].to_numpy(
            dtype=np.int64
        ),

        np.arange(
            1,
            101,
            dtype=np.int64
        )
    )


    target_row = (

        original_history[
            original_history[
                "epoch"
            ].astype(
                int
            )
            ==
            target_epoch
        ]
        .iloc[
            0
        ]
    )


    historical_max_aa = float(

        original_history[
            "val_aa"
        ].max()
    )


    assert math.isclose(

        float(
            target_row[
                "val_aa"
            ]
        ),

        historical_max_aa,

        rel_tol=0.0,

        abs_tol=1e-15
    ), (
        f"{dataset} seed {seed}: "
        "target epoch is not historical maximum Val AA."
    )


    # ==============================================================================================
    # D. TRAIN / VALIDATION DATASETS ONLY
    # ==============================================================================================

    train_dataset = RecoveryIndexedDataset(

        RECOVERY_TRAIN_DATASETS[
            dataset
        ],

        split[
            "train_indices"
        ]
    )


    val_dataset = RecoveryIndexedDataset(

        RECOVERY_VAL_DATASETS[
            dataset
        ],

        split[
            "val_indices"
        ]
    )


    # ==============================================================================================
    # E. EXACT NOTEBOOK 03 INITIALIZATION ORDER
    # ==============================================================================================

    set_global_seed(
        seed
    )


    train_generator = torch.Generator()


    train_generator.manual_seed(
        seed
    )


    train_loader = DataLoader(

        train_dataset,

        batch_size=
            CE_TRAINING_CONFIG[
                "batch_size"
            ],

        shuffle=True,

        generator=
            train_generator,

        num_workers=0,

        pin_memory=True,

        drop_last=False,
    )


    val_loader = DataLoader(

        val_dataset,

        batch_size=
            CE_TRAINING_CONFIG[
                "batch_size"
            ],

        shuffle=False,

        num_workers=0,

        pin_memory=True,

        drop_last=False,
    )


    expected_steps_per_epoch = math.ceil(

        len(
            train_dataset
        )

        /

        CE_TRAINING_CONFIG[
            "batch_size"
        ]
    )


    assert (
        len(
            train_loader
        )
        ==
        expected_steps_per_epoch
    )


    model = HybridSN(

        spectral_depth=
            cfg[
                "pca_channels"
            ],

        patch_size=
            cfg[
                "patch_size"
            ],

        num_classes=
            cfg[
                "classes"
            ],

        dropout=
            CE_TRAINING_CONFIG[
                "dropout"
            ]
    ).to(
        DEVICE
    )


    parameter_count = int(

        sum(

            parameter.numel()

            for parameter
            in model.parameters()
        )
    )


    assert (
        parameter_count
        ==
        cfg[
            "expected_parameters"
        ]
    )


    criterion = nn.CrossEntropyLoss()


    optimizer = build_ce_optimizer(
        model
    )


    global_step = 0


    # ==============================================================================================
    # F. RUN STATE
    # ==============================================================================================

    NOTEBOOK05_RECOVERY_FULL_TRAINING_STARTED = True


    recovery_records = []


    target_model_state = None
    target_optimizer_state = None
    target_validation_metrics = None
    target_global_step = None


    maximum_differences = {

        key:
            0.0

        for key in
        RECOVERY_TOLERANCES
    }


    start_time = time.time()


    print()
    print("=" * 145)
    print("NOTEBOOK 05 — BEST-VAL-AA RECOVERY")
    print("=" * 145)

    print(
        "Dataset             :",
        dataset
    )

    print(
        "Seed                :",
        seed
    )

    print(
        "Target best-AA epoch:",
        target_epoch
    )

    print(
        "Epochs to reproduce :",
        total_epochs
    )

    print(
        "Steps per epoch     :",
        expected_steps_per_epoch
    )

    print(
        "Official test       : SEALED"
    )

    print("=" * 145)


    # ==============================================================================================
    # G. FULL 100-EPOCH REPRODUCTION
    # ==============================================================================================

    for epoch in range(
        1,
        total_epochs + 1
    ):

        epoch_start = time.time()


        # ------------------------------------------------------------------------------------------
        # Training
        # ------------------------------------------------------------------------------------------

        train_result = train_one_epoch(

            model=model,

            loader=train_loader,

            criterion=criterion,

            optimizer=optimizer,

            device=DEVICE,

            base_lr=
                CE_TRAINING_CONFIG[
                    "learning_rate"
                ],

            legacy_decay=
                CE_TRAINING_CONFIG[
                    "legacy_lr_decay"
                ],

            global_step=
                global_step
        )


        global_step = int(
            train_result[
                "global_step"
            ]
        )


        # ------------------------------------------------------------------------------------------
        # Validation
        # ------------------------------------------------------------------------------------------

        val_result = evaluate_model(

            model=model,

            loader=val_loader,

            criterion=criterion,

            num_classes=
                cfg[
                    "classes"
                ],

            device=DEVICE
        )


        val_metrics = (
            val_result[
                "metrics"
            ]
        )


        # ------------------------------------------------------------------------------------------
        # Frozen validation order
        # ------------------------------------------------------------------------------------------

        assert np.array_equal(

            np.asarray(
                val_result[
                    "indices"
                ],
                dtype=np.int64
            ),

            split[
                "val_indices"
            ]
        )


        assert np.array_equal(

            np.asarray(
                val_result[
                    "labels"
                ],
                dtype=np.int64
            ),

            split[
                "val_labels"
            ]
        )


        # ------------------------------------------------------------------------------------------
        # Actual metrics
        # ------------------------------------------------------------------------------------------

        actual = {

            "train_loss":
                float(
                    train_result[
                        "loss"
                    ]
                ),

            "train_accuracy":
                float(
                    train_result[
                        "accuracy"
                    ]
                ),

            "val_loss":
                float(
                    val_metrics[
                        "loss"
                    ]
                ),

            "val_oa":
                float(
                    val_metrics[
                        "oa"
                    ]
                ),

            "val_aa":
                float(
                    val_metrics[
                        "aa"
                    ]
                ),

            "val_kappa":
                float(
                    val_metrics[
                        "kappa"
                    ]
                ),

            "val_macro_f1":
                float(
                    val_metrics[
                        "macro_f1"
                    ]
                ),

            "learning_rate":
                float(
                    train_result[
                        "learning_rate"
                    ]
                ),
        }


        historical_row = (
            original_history.iloc[
                epoch - 1
            ]
        )


        # ------------------------------------------------------------------------------------------
        # Every epoch must match Notebook 03
        # ------------------------------------------------------------------------------------------

        for metric, tolerance in (
            RECOVERY_TOLERANCES.items()
        ):

            historical_value = float(
                historical_row[
                    metric
                ]
            )


            difference = abs(

                actual[
                    metric
                ]
                -
                historical_value
            )


            maximum_differences[
                metric
            ] = max(

                maximum_differences[
                    metric
                ],

                difference
            )


            if (
                difference
                >
                tolerance
            ):

                raise RuntimeError(

                    "Historical trajectory mismatch.\n"
                    f"Dataset={dataset}\n"
                    f"Seed={seed}\n"
                    f"Epoch={epoch}\n"
                    f"Metric={metric}\n"
                    f"Historical={historical_value:.15f}\n"
                    f"Recovered={actual[metric]:.15f}\n"
                    f"Difference={difference:.3e}\n"
                    f"Tolerance={tolerance:.3e}\n\n"
                    "STOP — no recovered checkpoint will be written."
                )


        historical_global_step = int(
            historical_row[
                "global_step"
            ]
        )


        assert (
            global_step
            ==
            historical_global_step
        ), (
            f"{dataset} seed {seed}: "
            f"global-step mismatch at epoch {epoch}."
        )


        # ------------------------------------------------------------------------------------------
        # Capture frozen best-AA target in CPU memory
        # ------------------------------------------------------------------------------------------

        if (
            epoch
            ==
            target_epoch
        ):

            target_model_state = recursive_to_cpu(
                model.state_dict()
            )


            target_optimizer_state = recursive_to_cpu(
                optimizer.state_dict()
            )


            target_global_step = int(
                global_step
            )


            target_validation_metrics = {

                "epoch":
                    int(
                        epoch
                    ),

                "oa":
                    float(
                        val_metrics[
                            "oa"
                        ]
                    ),

                "aa":
                    float(
                        val_metrics[
                            "aa"
                        ]
                    ),

                "kappa":
                    float(
                        val_metrics[
                            "kappa"
                        ]
                    ),

                "macro_f1":
                    float(
                        val_metrics[
                            "macro_f1"
                        ]
                    ),

                "loss":
                    float(
                        val_metrics[
                            "loss"
                        ]
                    ),
            }


        epoch_seconds = (
            time.time()
            -
            epoch_start
        )


        recovery_records.append({

            "epoch":
                int(
                    epoch
                ),

            "train_loss":
                actual[
                    "train_loss"
                ],

            "train_accuracy":
                actual[
                    "train_accuracy"
                ],

            "val_loss":
                actual[
                    "val_loss"
                ],

            "val_oa":
                actual[
                    "val_oa"
                ],

            "val_aa":
                actual[
                    "val_aa"
                ],

            "val_kappa":
                actual[
                    "val_kappa"
                ],

            "val_macro_f1":
                actual[
                    "val_macro_f1"
                ],

            "learning_rate":
                actual[
                    "learning_rate"
                ],

            "global_step":
                int(
                    global_step
                ),

            "epoch_seconds":
                float(
                    epoch_seconds
                ),

            "historical_match":
                True,

            "target_best_valAA_epoch":
                bool(
                    epoch
                    ==
                    target_epoch
                ),
        })


        # ------------------------------------------------------------------------------------------
        # Compact progress
        # ------------------------------------------------------------------------------------------

        if (
            epoch == 1
            or
            epoch % 10 == 0
            or
            epoch == target_epoch
            or
            epoch == total_epochs
        ):

            marker = (

                " <TARGET BEST-AA>"

                if (
                    epoch
                    ==
                    target_epoch
                )

                else ""
            )


            print(

                f"Epoch {epoch:03d}/100 | "

                f"TrainLoss="
                f"{actual['train_loss']:.6f} | "

                f"ValOA="
                f"{actual['val_oa']:.6f} | "

                f"ValAA="
                f"{actual['val_aa']:.6f} | "

                f"F1="
                f"{actual['val_macro_f1']:.6f} | "

                f"HISTORICAL MATCH"
                f"{marker}"
            )


        del val_result


    # ==============================================================================================
    # H. ALL 100 EPOCHS MUST PASS BEFORE WRITING ANY FINAL RECOVERY ARTIFACT
    # ==============================================================================================

    assert len(
        recovery_records
    ) == 100


    assert target_model_state is not None
    assert target_optimizer_state is not None
    assert target_validation_metrics is not None
    assert target_global_step is not None


    recovery_history_df = pd.DataFrame(
        recovery_records
    )


    assert (
        recovery_history_df[
            "historical_match"
        ].all()
    )


    assert math.isclose(

        target_validation_metrics[
            "aa"
        ],

        float(
            target_row[
                "val_aa"
            ]
        ),

        rel_tol=0.0,

        abs_tol=1e-12
    )


    # ==============================================================================================
    # I. ORIGINAL BEST-OA CHECKPOINT
    # ==============================================================================================

    source_best_oa_path = (
        EXISTING_BEST_OA[
            (
                dataset,
                seed
            )
        ]
    )


    source_checkpoint = torch.load(

        source_best_oa_path,

        map_location="cpu",

        weights_only=False
    )


    source_checkpoint_sha = (
        sha256_file(
            source_best_oa_path
        )
    )


    source_history_sha = (
        sha256_file(
            original_history_path
        )
    )


    # ==============================================================================================
    # J. BUILD BEST-AA CHECKPOINT
    # ==============================================================================================

    recovered_checkpoint = {

        "project":
            "CSE498R LTLC",

        "method":
            "HybridSN_CE",

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "epoch":
            int(
                target_epoch
            ),

        "model_state_dict":
            target_model_state,

        "optimizer_state_dict":
            target_optimizer_state,

        "global_step":
            int(
                target_global_step
            ),

        "validation_metrics":
            target_validation_metrics,

        "training_config":
            copy.deepcopy(
                CE_TRAINING_CONFIG
            ),

        "model_configuration": {

            "spectral_depth":
                int(
                    cfg[
                        "pca_channels"
                    ]
                ),

            "patch_size":
                int(
                    cfg[
                        "patch_size"
                    ]
                ),

            "num_classes":
                int(
                    cfg[
                        "classes"
                    ]
                ),

            "dropout":
                float(
                    CE_TRAINING_CONFIG[
                        "dropout"
                    ]
                ),
        },

        "split_sha256":
            source_checkpoint[
                "split_sha256"
            ],

        "cube_sha256":
            source_checkpoint[
                "cube_sha256"
            ],

        "checkpoint_policy":
            "best_validation_AA_recovered",

        "test_used_for_selection":
            False,

        "recovery": {

            "reason":
                (
                    "Implementation-locked protocol requires "
                    "best-validation-AA checkpoint; original "
                    "Notebook 03 retained only best-validation-OA."
                ),

            "historical_target_epoch":
                int(
                    target_epoch
                ),

            "full_100_epoch_trajectory_reproduced":
                True,

            "all_100_epochs_match_original_history":
                True,

            "source_history":
                str(
                    original_history_path.relative_to(
                        ROOT
                    )
                ),

            "source_history_sha256":
                source_history_sha,

            "source_best_valOA_checkpoint":
                str(
                    source_best_oa_path.relative_to(
                        ROOT
                    )
                ),

            "source_best_valOA_checkpoint_sha256":
                source_checkpoint_sha,

            "one_epoch_reproduction_gate_sha256":
                ONE_EPOCH_AUDIT_SHA,

            "official_test_used":
                False,
        },
    }


    # ==============================================================================================
    # K. WRITE ONLY AFTER COMPLETE REPRODUCTION
    # ==============================================================================================

    torch.save(
        recovered_checkpoint,
        checkpoint_path
    )


    recovery_history_df.to_csv(
        history_path,
        index=False
    )


    total_seconds = (
        time.time()
        -
        start_time
    )


    metadata = {

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "target_best_valAA_epoch":
            int(
                target_epoch
            ),

        "target_validation_metrics":
            target_validation_metrics,

        "full_epochs_reproduced":
            100,

        "all_epochs_historical_match":
            True,

        "maximum_absolute_differences": {

            key:
                float(
                    value
                )

            for key, value
            in maximum_differences.items()
        },

        "checkpoint":
            str(
                checkpoint_path.relative_to(
                    ROOT
                )
            ),

        "checkpoint_sha256":
            sha256_file(
                checkpoint_path
            ),

        "history":
            str(
                history_path.relative_to(
                    ROOT
                )
            ),

        "history_sha256":
            sha256_file(
                history_path
            ),

        "source_history_sha256":
            source_history_sha,

        "source_best_valOA_checkpoint_sha256":
            source_checkpoint_sha,

        "total_recovery_seconds":
            float(
                total_seconds
            ),

        "official_test_dataset_created":
            False,

        "official_test_dataloader_created":
            False,

        "official_test_used":
            False,
    }


    with open(
        metadata_path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            metadata,
            f,
            indent=2,
            sort_keys=True
        )


    # ==============================================================================================
    # L. CLEAN-ROOM ARTIFACT RELOAD AUDIT
    # ==============================================================================================

    audit_checkpoint = torch.load(

        checkpoint_path,

        map_location="cpu",

        weights_only=False
    )


    assert audit_checkpoint[
        "dataset"
    ] == dataset


    assert int(
        audit_checkpoint[
            "seed"
        ]
    ) == seed


    assert int(
        audit_checkpoint[
            "epoch"
        ]
    ) == target_epoch


    assert (
        audit_checkpoint[
            "checkpoint_policy"
        ]
        ==
        "best_validation_AA_recovered"
    )


    assert (
        audit_checkpoint[
            "test_used_for_selection"
        ]
        is False
    )


    assert (
        audit_checkpoint[
            "recovery"
        ][
            "all_100_epochs_match_original_history"
        ]
        is True
    )


    checkpoint_sha = sha256_file(
        checkpoint_path
    )


    history_sha = sha256_file(
        history_path
    )


    metadata_sha = sha256_file(
        metadata_path
    )


    # ==============================================================================================
    # M. COMPLETION STATE
    # ==============================================================================================

    NOTEBOOK05_RECOVERY_COMPLETED_RUNS = globals().get(

        "NOTEBOOK05_RECOVERY_COMPLETED_RUNS",

        []
    )


    identity = (
        dataset,
        seed
    )


    if (
        identity
        not in
        NOTEBOOK05_RECOVERY_COMPLETED_RUNS
    ):

        NOTEBOOK05_RECOVERY_COMPLETED_RUNS.append(
            identity
        )


    # ==============================================================================================
    # N. CLEANUP
    # ==============================================================================================

    del model
    del optimizer
    del criterion
    del source_checkpoint
    del audit_checkpoint

    torch.cuda.empty_cache()

    gc.collect()


    # ==============================================================================================
    # O. REPORT
    # ==============================================================================================

    print()
    print("=" * 145)

    print(
        f"{dataset.upper()} SEED {seed} "
        "BEST-VAL-AA RECOVERY: PASS"
    )

    print("=" * 145)

    print(
        "Historical trajectory reproduced : 100/100 epochs PASS"
    )

    print(
        "Recovered target epoch           :",
        target_epoch
    )

    print(
        "Recovered Val AA                 :",
        f"{target_validation_metrics['aa']:.12f}"
    )

    print(
        "Recovered Val OA                 :",
        f"{target_validation_metrics['oa']:.12f}"
    )

    print(
        "Recovered Macro-F1               :",
        f"{target_validation_metrics['macro_f1']:.12f}"
    )

    print()

    print(
        "Maximum historical differences:"
    )


    for key, value in (
        maximum_differences.items()
    ):

        print(
            f"  {key:18s}: "
            f"{value:.3e}"
        )


    print()

    print(
        "Checkpoint:"
    )

    print(
        checkpoint_path
    )

    print(
        "SHA-256:"
    )

    print(
        checkpoint_sha
    )

    print()

    print(
        "History SHA-256:"
    )

    print(
        history_sha
    )

    print(
        "Metadata SHA-256:"
    )

    print(
        metadata_sha
    )

    print()

    print(
        "Completed recovery runs:",
        len(
            NOTEBOOK05_RECOVERY_COMPLETED_RUNS
        ),
        "/ 5"
    )

    print(
        "Official-test dataset created:",
        NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED
    )

    print(
        "Official-test DataLoader created:",
        NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED
    )

    print(
        "Official test used:",
        NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
    )

    print()

    print(
        "OFFICIAL TEST REMAINS SEALED"
    )

    print("=" * 145)


    return {

        "dataset":
            dataset,

        "seed":
            seed,

        "target_epoch":
            target_epoch,

        "validation_metrics":
            target_validation_metrics,

        "checkpoint_path":
            checkpoint_path,

        "checkpoint_sha256":
            checkpoint_sha,

        "history_sha256":
            history_sha,

        "metadata_sha256":
            metadata_sha,

        "maximum_differences":
            maximum_differences,

        "pass":
            True,
    }


# ==================================================================================================
# 4. RUN 2/5 — PINGAN SEED 3407
# ==================================================================================================

PINGAN_SEED3407_RECOVERY = (
    recover_best_valAA_ce_run(

        dataset=
            "Pingan",

        seed=
            3407,

        target_epoch=
            100
    )
)


assert PINGAN_SEED3407_RECOVERY[
    "pass"
] is True


assert (
    PINGAN_SEED3407_RECOVERY[
        "target_epoch"
    ]
    ==
    100
)


assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


print()
print("=" * 145)
print("NOTEBOOK 05 BEST-VAL-AA RECOVERY RUN 2/5: PASS")
print("NEXT TARGET: QINGYUN SEED 42 — EPOCH 100")
print("OFFICIAL TEST REMAINS SEALED")
print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 17A
# FINAL 9-CHECKPOINT BEST-VAL-AA PREFLIGHT
# + NOTEBOOK 03 VALIDATION-CACHE INVENTORY
#
# NO TRAINING
# NO OFFICIAL-TEST ARRAYS LOADED
# ==================================================================================================

import json
import math

import numpy as np
import pandas as pd
import torch


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_ALL_FIVE_BEST_AA_RECOVERY_RUNS_PASS is True

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


# ==================================================================================================
# 2. FROZEN BEST-AA EPOCHS FOR ALL 9 CE RUNS
# ==================================================================================================

FINAL_BEST_AA_EPOCHS = {

    ("Pingan", 42):
        62,

    ("Pingan", 123):
        97,

    ("Pingan", 3407):
        100,


    ("Qingyun", 42):
        100,

    ("Qingyun", 123):
        65,

    ("Qingyun", 3407):
        85,


    ("Tangdaowan", 42):
        83,

    ("Tangdaowan", 123):
        70,

    ("Tangdaowan", 3407):
        72,
}


assert len(
    FINAL_BEST_AA_EPOCHS
) == 9


RECOVERED_IDENTITIES = {

    ("Pingan", 123),

    ("Pingan", 3407),

    ("Qingyun", 42),

    ("Qingyun", 123),

    ("Qingyun", 3407),
}


SAME_OA_AA_IDENTITIES = {

    ("Pingan", 42),

    ("Tangdaowan", 42),

    ("Tangdaowan", 123),

    ("Tangdaowan", 3407),
}


assert (
    RECOVERED_IDENTITIES
    |
    SAME_OA_AA_IDENTITIES
) == set(
    FINAL_BEST_AA_EPOCHS.keys()
)


assert (
    RECOVERED_IDENTITIES
    &
    SAME_OA_AA_IDENTITIES
) == set()


# ==================================================================================================
# 3. BUILD FINAL CHECKPOINT PATH MAP
# ==================================================================================================

FINAL_BEST_AA_CHECKPOINTS = {}


for identity in sorted(
    FINAL_BEST_AA_EPOCHS.keys()
):

    dataset, seed = identity


    if identity in RECOVERED_IDENTITIES:

        checkpoint_path = (

            BEST_AA_CHECKPOINT_DIR
            /
            f"{dataset.lower()}_seed{seed}_best_valAA.pt"
        )


        source_type = (
            "recovered_best_validation_AA"
        )


    else:

        checkpoint_path = (
            EXISTING_BEST_OA[
                identity
            ]
        )


        source_type = (
            "original_bestOA_same_epoch_as_bestAA"
        )


    assert checkpoint_path.exists(), (
        f"Missing final best-AA checkpoint: "
        f"{dataset} seed {seed}"
    )


    FINAL_BEST_AA_CHECKPOINTS[
        identity
    ] = {

        "path":
            checkpoint_path,

        "source_type":
            source_type,
    }


# ==================================================================================================
# 4. CLEAN-ROOM CHECKPOINT AUDIT
# ==================================================================================================

CHECKPOINT_AUDIT_RECORDS = []


print("=" * 150)
print("NOTEBOOK 05 — FINAL 9-CHECKPOINT BEST-VAL-AA PREFLIGHT")
print("=" * 150)


for identity in [

    ("Pingan", 42),
    ("Pingan", 123),
    ("Pingan", 3407),

    ("Qingyun", 42),
    ("Qingyun", 123),
    ("Qingyun", 3407),

    ("Tangdaowan", 42),
    ("Tangdaowan", 123),
    ("Tangdaowan", 3407),
]:

    dataset, seed = identity


    expected_epoch = int(
        FINAL_BEST_AA_EPOCHS[
            identity
        ]
    )


    info = (
        FINAL_BEST_AA_CHECKPOINTS[
            identity
        ]
    )


    checkpoint_path = (
        info[
            "path"
        ]
    )


    checkpoint = torch.load(

        checkpoint_path,

        map_location="cpu",

        weights_only=False
    )


    assert checkpoint[
        "dataset"
    ] == dataset


    assert int(
        checkpoint[
            "seed"
        ]
    ) == seed


    actual_epoch = int(
        checkpoint[
            "epoch"
        ]
    )


    assert (
        actual_epoch
        ==
        expected_epoch
    ), (
        f"{dataset} seed {seed}: "
        f"checkpoint epoch {actual_epoch} "
        f"!= expected best-AA epoch {expected_epoch}"
    )


    assert (
        checkpoint[
            "test_used_for_selection"
        ]
        is False
    )


    if identity in RECOVERED_IDENTITIES:

        assert (
            checkpoint[
                "checkpoint_policy"
            ]
            ==
            "best_validation_AA_recovered"
        )


        assert (
            checkpoint[
                "recovery"
            ][
                "all_100_epochs_match_original_history"
            ]
            is True
        )


    else:

        assert (
            checkpoint[
                "checkpoint_policy"
            ]
            ==
            "best_validation_OA"
        )


        # Verify from original history that this same epoch
        # is also the frozen maximum validation-AA epoch.
        history_path = (
            RECOVERY_HISTORY[
                identity
            ]
        )


        history_df = pd.read_csv(
            history_path
        )


        max_aa = float(
            history_df[
                "val_aa"
            ].max()
        )


        epoch_row = (

            history_df[
                history_df[
                    "epoch"
                ].astype(
                    int
                )
                ==
                expected_epoch
            ]
            .iloc[
                0
            ]
        )


        assert math.isclose(

            float(
                epoch_row[
                    "val_aa"
                ]
            ),

            max_aa,

            rel_tol=0.0,

            abs_tol=1e-15
        )


    checkpoint_sha = (
        sha256_file(
            checkpoint_path
        )
    )


    CHECKPOINT_AUDIT_RECORDS.append({

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "best_valAA_epoch":
            expected_epoch,

        "checkpoint_policy":
            checkpoint[
                "checkpoint_policy"
            ],

        "source_type":
            info[
                "source_type"
            ],

        "checkpoint":
            str(
                checkpoint_path.relative_to(
                    ROOT
                )
            ),

        "checkpoint_sha256":
            checkpoint_sha,

        "test_used_for_selection":
            False,

        "pass":
            True,
    })


    print(

        f"{dataset:11s} | "
        f"seed={seed:4d} | "
        f"epoch={expected_epoch:3d} | "
        f"{info['source_type']:39s} | "
        f"PASS"
    )


    del checkpoint


FINAL_BEST_AA_CHECKPOINT_AUDIT_DF = (
    pd.DataFrame(
        CHECKPOINT_AUDIT_RECORDS
    )
)


assert len(
    FINAL_BEST_AA_CHECKPOINT_AUDIT_DF
) == 9


assert (
    FINAL_BEST_AA_CHECKPOINT_AUDIT_DF[
        "pass"
    ].all()
)


# ==================================================================================================
# 5. NOTEBOOK 03 ORIGINAL FROZEN-INFERENCE MANIFEST
# ==================================================================================================

NB03_INFERENCE_MANIFEST_CSV = (

    ROOT
    /
    "audit"
    /
    "notebook03_hybridsn"
    /
    "frozen_inference"
    /
    "hybridsn_ce_inference_output_manifest.csv"
)


NB03_INFERENCE_MANIFEST_JSON = (

    ROOT
    /
    "audit"
    /
    "notebook03_hybridsn"
    /
    "frozen_inference"
    /
    "hybridsn_ce_inference_output_manifest.json"
)


assert NB03_INFERENCE_MANIFEST_CSV.exists()
assert NB03_INFERENCE_MANIFEST_JSON.exists()


original_inference_manifest_df = (
    pd.read_csv(
        NB03_INFERENCE_MANIFEST_CSV
    )
)


print()
print("=" * 150)
print("ORIGINAL NOTEBOOK 03 INFERENCE MANIFEST — SCHEMA")
print("=" * 150)

print(
    "Rows:",
    len(
        original_inference_manifest_df
    )
)

print(
    "Columns:"
)

print(
    list(
        original_inference_manifest_df.columns
    )
)


# ==================================================================================================
# 6. IDENTIFY SPLIT COLUMN WITHOUT LOADING ANY LOGIT ARRAYS
# ==================================================================================================

possible_split_columns = [

    column

    for column in
    original_inference_manifest_df.columns

    if (
        "split"
        in column.lower()
        or
        "subset"
        in column.lower()
    )
]


assert len(
    possible_split_columns
) >= 1, (
    "Could not identify split column in "
    "Notebook 03 inference manifest."
)


print()

print(
    "Possible split columns:",
    possible_split_columns
)


# Use first column containing recognizable val/test values.
SPLIT_COLUMN = None


for column in possible_split_columns:

    values = (

        original_inference_manifest_df[
            column
        ]
        .astype(
            str
        )
        .str.lower()
        .unique()
        .tolist()
    )


    if any(
        value
        in
        [
            "val",
            "validation",
            "official_test",
            "test"
        ]

        for value in
        values
    ):

        SPLIT_COLUMN = column
        break


assert SPLIT_COLUMN is not None


print(
    "Selected split column:",
    SPLIT_COLUMN
)


# ==================================================================================================
# 7. FILTER VALIDATION METADATA ONLY
#
# IMPORTANT:
# We read MANIFEST METADATA ONLY here.
# No official-test NPZ/logit arrays are loaded.
# ==================================================================================================

split_values = (

    original_inference_manifest_df[
        SPLIT_COLUMN
    ]
    .astype(
        str
    )
    .str.lower()
)


validation_mask = (
    split_values.isin(
        [
            "val",
            "validation"
        ]
    )
)


ORIGINAL_VALIDATION_MANIFEST_DF = (

    original_inference_manifest_df[
        validation_mask
    ]
    .copy()
    .reset_index(
        drop=True
    )
)


assert len(
    ORIGINAL_VALIDATION_MANIFEST_DF
) == 9, (
    "Expected exactly 9 Notebook 03 "
    "validation inference records."
)


print()
print("=" * 150)
print("ORIGINAL NOTEBOOK 03 VALIDATION INFERENCE RECORDS")
print("=" * 150)


display(
    ORIGINAL_VALIDATION_MANIFEST_DF
)


# ==================================================================================================
# 8. SAVE FINAL 9-CHECKPOINT PREFLIGHT MANIFEST
# ==================================================================================================

FINAL_BEST_AA_PREFLIGHT_PATH = (

    RECOVERY_AUDIT_DIR
    /
    "notebook05_final_9_best_valAA_checkpoint_preflight.json"
)


FINAL_BEST_AA_PREFLIGHT = {

    "project":
        "LTLC",

    "method":
        "HybridSN_CE",

    "checkpoint_purpose":
        "best_validation_AA",

    "total_runs":
        9,

    "recovered_runs":
        5,

    "original_same_epoch_OA_AA_runs":
        4,

    "runs":
        CHECKPOINT_AUDIT_RECORDS,

    "notebook03_inference_manifest":
        str(
            NB03_INFERENCE_MANIFEST_CSV.relative_to(
                ROOT
            )
        ),

    "notebook03_validation_manifest_rows":
        int(
            len(
                ORIGINAL_VALIDATION_MANIFEST_DF
            )
        ),

    "official_test_arrays_loaded":
        False,

    "official_test_used":
        False,
}


if FINAL_BEST_AA_PREFLIGHT_PATH.exists():

    with open(
        FINAL_BEST_AA_PREFLIGHT_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        existing = json.load(
            f
        )


    assert (
        existing
        ==
        FINAL_BEST_AA_PREFLIGHT
    ), (
        "Existing final best-AA preflight differs. STOP."
    )


else:

    with open(
        FINAL_BEST_AA_PREFLIGHT_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            FINAL_BEST_AA_PREFLIGHT,
            f,
            indent=2,
            sort_keys=True
        )


FINAL_BEST_AA_PREFLIGHT_SHA = (
    sha256_file(
        FINAL_BEST_AA_PREFLIGHT_PATH
    )
)


# ==================================================================================================
# 9. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_FINAL_9_BEST_AA_PREFLIGHT_PASS = True


print()
print("=" * 150)

print(
    "Final best-AA checkpoints audited     : 9/9 PASS"
)

print(
    "Recovered best-AA checkpoints         : 5"
)

print(
    "Original OA=AA same-epoch checkpoints : 4"
)

print(
    "Validation inference records located  : 9/9 PASS"
)

print()

print(
    "Preflight artifact:"
)

print(
    FINAL_BEST_AA_PREFLIGHT_PATH
)

print(
    "SHA-256:"
)

print(
    FINAL_BEST_AA_PREFLIGHT_SHA
)

print()

print(
    "Official-test arrays loaded:",
    False
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 FINAL 9 BEST-AA PREFLIGHT: PASS"
)

print(
    "READY TO BUILD UNIFIED BEST-AA VALIDATION LOGIT MANIFEST"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 150)
```

```python
# ============================================================
# NOTEBOOK 05 — CELL 17A-R2
# COMPLETE ORIGINAL NOTEBOOK 03 CE HISTORY MAP
# ============================================================

import pandas as pd


# ------------------------------------------------------------
# Original Notebook 03 CE history directory
# ------------------------------------------------------------

NB03_CE_HISTORY_DIR = (
    ROOT
    /
    "runs"
    /
    "hybridsn_ce_baseline"
    /
    "histories"
)


assert NB03_CE_HISTORY_DIR.exists()


ALL_CE_IDENTITIES = [

    ("Pingan", 42),
    ("Pingan", 123),
    ("Pingan", 3407),

    ("Qingyun", 42),
    ("Qingyun", 123),
    ("Qingyun", 3407),

    ("Tangdaowan", 42),
    ("Tangdaowan", 123),
    ("Tangdaowan", 3407),
]


# ------------------------------------------------------------
# Build complete 9-run history mapping
# ------------------------------------------------------------

RECOVERY_HISTORY_COMPLETE = {}


print("=" * 110)
print("NOTEBOOK 05 — COMPLETE ORIGINAL CE HISTORY MAP")
print("=" * 110)


for dataset, seed in ALL_CE_IDENTITIES:

    history_path = (
        NB03_CE_HISTORY_DIR
        /
        f"{dataset.lower()}_seed{seed}_history.csv"
    )


    assert history_path.exists(), (
        f"Missing original Notebook 03 history:\n"
        f"{history_path}"
    )


    # Basic integrity check
    history_df = pd.read_csv(
        history_path
    )


    assert len(
        history_df
    ) == 100, (
        f"{dataset} seed {seed}: "
        f"expected 100 epochs, found {len(history_df)}."
    )


    assert (
        history_df[
            "epoch"
        ].astype(
            int
        ).tolist()
        ==
        list(
            range(
                1,
                101
            )
        )
    )


    RECOVERY_HISTORY_COMPLETE[
        (
            dataset,
            seed
        )
    ] = history_path


    print(
        f"{dataset:11s} | "
        f"seed={seed:4d} | "
        f"epochs={len(history_df):3d} | "
        f"{history_path.name} | PASS"
    )


# ------------------------------------------------------------
# Replace old partial map with complete 9-run map
# ------------------------------------------------------------

RECOVERY_HISTORY = (
    RECOVERY_HISTORY_COMPLETE.copy()
)


assert len(
    RECOVERY_HISTORY
) == 9


print()
print(
    "Original CE histories located:",
    len(
        RECOVERY_HISTORY
    ),
    "/ 9 PASS"
)

print(
    "All histories contain 100 epochs: PASS"
)

print(
    "Recovery checkpoints modified     : NO"
)

print(
    "Training performed                : NO"
)

print(
    "Official test used                :",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()
print("CELL 17A-R2: PASS")
print("NOW RERUN CELL 17A FROM THE BEGINNING")
print("=" * 110)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 17B
# UNIFIED 9-RUN BEST-VAL-AA VALIDATION LOGIT SET
#
# 5 recovered checkpoints -> fresh validation inference on T4
# 4 OA=AA same-epoch checkpoints -> reuse original validation outputs
#
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import os
import gc
import json
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import DataLoader


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_FINAL_9_BEST_AA_PREFLIGHT_PASS is True

assert NOTEBOOK05_ALL_FIVE_BEST_AA_RECOVERY_RUNS_PASS is True

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False

assert torch.cuda.is_available()
assert "T4" in torch.cuda.get_device_name(0).upper()


DEVICE = torch.device(
    "cuda:0"
)


# ==================================================================================================
# 2. OUTPUT DIRECTORIES
# ==================================================================================================

FINAL_BEST_AA_VALIDATION_DIR = (

    ROOT
    /
    "runs"
    /
    "hybridsn_ce_best_valAA_recovery"
    /
    "logits"
    /
    "validation"
)


FINAL_BEST_AA_VALIDATION_METRICS_DIR = (

    ROOT
    /
    "runs"
    /
    "hybridsn_ce_best_valAA_recovery"
    /
    "metrics"
    /
    "validation"
)


FINAL_BEST_AA_VALIDATION_DIR.mkdir(
    parents=True,
    exist_ok=True
)


FINAL_BEST_AA_VALIDATION_METRICS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ==================================================================================================
# 3. RUN ORDER
# ==================================================================================================

FINAL_RUN_ORDER = [

    ("Pingan", 42),
    ("Pingan", 123),
    ("Pingan", 3407),

    ("Qingyun", 42),
    ("Qingyun", 123),
    ("Qingyun", 3407),

    ("Tangdaowan", 42),
    ("Tangdaowan", 123),
    ("Tangdaowan", 3407),
]


assert len(
    FINAL_RUN_ORDER
) == 9


# ==================================================================================================
# 4. ORIGINAL VALIDATION OUTPUT LOCATOR
#
# Uses filename + frozen SHA from Notebook 03 manifest.
# It never loads an official-test output.
# ==================================================================================================

def locate_original_validation_output(
    dataset,
    seed
):

    matching_rows = (

        ORIGINAL_VALIDATION_MANIFEST_DF[
            (
                ORIGINAL_VALIDATION_MANIFEST_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                ORIGINAL_VALIDATION_MANIFEST_DF[
                    "seed"
                ].astype(
                    int
                )
                ==
                int(
                    seed
                )
            )
        ]
    )


    assert len(
        matching_rows
    ) == 1


    row = matching_rows.iloc[
        0
    ]


    assert str(
        row[
            "split"
        ]
    ).lower() in [
        "val",
        "validation"
    ]


    filename = str(
        row[
            "output_file"
        ]
    )


    expected_sha = str(
        row[
            "output_sha256"
        ]
    )


    candidates = [

        path

        for path in ROOT.rglob(
            filename
        )

        if (
            path.is_file()
            and
            "best_valAA_recovery"
            not in str(
                path
            )
        )
    ]


    sha_matches = []


    for path in candidates:

        if (
            sha256_file(
                path
            )
            ==
            expected_sha
        ):

            sha_matches.append(
                path
            )


    assert len(
        sha_matches
    ) >= 1, (
        f"Could not locate SHA-verified original "
        f"validation output for {dataset} seed {seed}."
    )


    return (
        sorted(
            sha_matches,
            key=lambda p: str(p)
        )[0],
        expected_sha
    )


# ==================================================================================================
# 5. ORIGINAL NPZ SCHEMA NORMALIZER
# ==================================================================================================

def load_canonical_validation_arrays(
    path
):

    with np.load(
        path,
        allow_pickle=False
    ) as data:

        files = set(
            data.files
        )


        aliases = {

            "logits":
                [
                    "logits"
                ],

            "labels":
                [
                    "labels",
                    "y_true",
                    "targets"
                ],

            "predictions":
                [
                    "predictions",
                    "preds",
                    "y_pred"
                ],

            "indices":
                [
                    "indices",
                    "center_indices",
                    "sample_indices"
                ],
        }


        resolved = {}


        for semantic, candidates in (
            aliases.items()
        ):

            found = [

                candidate

                for candidate in candidates

                if candidate in files
            ]


            assert len(
                found
            ) == 1, (
                f"{path.name}: could not uniquely identify "
                f"'{semantic}' key. Files={sorted(files)}"
            )


            resolved[
                semantic
            ] = found[
                0
            ]


        arrays = {

            "logits":
                np.asarray(
                    data[
                        resolved[
                            "logits"
                        ]
                    ]
                ),

            "labels":
                np.asarray(
                    data[
                        resolved[
                            "labels"
                        ]
                    ],
                    dtype=np.int64
                ),

            "predictions":
                np.asarray(
                    data[
                        resolved[
                            "predictions"
                        ]
                    ],
                    dtype=np.int64
                ),

            "indices":
                np.asarray(
                    data[
                        resolved[
                            "indices"
                        ]
                    ],
                    dtype=np.int64
                ),
        }


    return arrays


# ==================================================================================================
# 6. RESTART-SAFE CANONICAL NPZ WRITER
# ==================================================================================================

def save_or_verify_npz(
    destination,
    arrays
):

    arrays = {

        "logits":
            np.asarray(
                arrays[
                    "logits"
                ]
            ),

        "labels":
            np.asarray(
                arrays[
                    "labels"
                ],
                dtype=np.int64
            ),

        "predictions":
            np.asarray(
                arrays[
                    "predictions"
                ],
                dtype=np.int64
            ),

        "indices":
            np.asarray(
                arrays[
                    "indices"
                ],
                dtype=np.int64
            ),
    }


    if destination.exists():

        existing = (
            load_canonical_validation_arrays(
                destination
            )
        )


        for key in arrays:

            assert np.array_equal(

                existing[
                    key
                ],

                arrays[
                    key
                ]
            ), (
                f"Existing output mismatch: "
                f"{destination.name} | {key}"
            )


        return


    temporary = destination.with_name(

        destination.stem
        +
        ".tmp.npz"
    )


    if temporary.exists():

        temporary.unlink()


    np.savez_compressed(

        temporary,

        logits=
            arrays[
                "logits"
            ],

        labels=
            arrays[
                "labels"
            ],

        predictions=
            arrays[
                "predictions"
            ],

        indices=
            arrays[
                "indices"
            ],
    )


    os.replace(
        temporary,
        destination
    )


# ==================================================================================================
# 7. RESTART-SAFE JSON WRITER
# ==================================================================================================

def save_or_verify_json(
    destination,
    payload
):

    if destination.exists():

        with open(
            destination,
            "r",
            encoding="utf-8"
        ) as f:

            existing = json.load(
                f
            )


        assert (
            existing
            ==
            payload
        ), (
            f"Existing JSON differs: "
            f"{destination.name}"
        )


    else:

        with open(
            destination,
            "w",
            encoding="utf-8"
        ) as f:

            json.dump(
                payload,
                f,
                indent=2,
                sort_keys=True
            )


# ==================================================================================================
# 8. BUILD ALL 9 VALIDATION OUTPUTS
# ==================================================================================================

FINAL_VALIDATION_RECORDS = []

GENERATED_GPU_RUNS = 0
REUSED_ORIGINAL_RUNS = 0


print("=" * 150)
print("NOTEBOOK 05 — UNIFIED BEST-AA VALIDATION LOGIT GENERATION")
print("=" * 150)

print(
    "Fresh GPU inference required:",
    len(
        RECOVERED_IDENTITIES
    )
)

print(
    "Original validation outputs reusable:",
    len(
        SAME_OA_AA_IDENTITIES
    )
)

print(
    "Official test:",
    "SEALED"
)

print("=" * 150)


for dataset, seed in FINAL_RUN_ORDER:

    identity = (
        dataset,
        seed
    )


    checkpoint_info = (
        FINAL_BEST_AA_CHECKPOINTS[
            identity
        ]
    )


    checkpoint_path = (
        checkpoint_info[
            "path"
        ]
    )


    checkpoint_sha = sha256_file(
        checkpoint_path
    )


    checkpoint = torch.load(

        checkpoint_path,

        map_location="cpu",

        weights_only=False
    )


    expected_epoch = int(
        FINAL_BEST_AA_EPOCHS[
            identity
        ]
    )


    assert int(
        checkpoint[
            "epoch"
        ]
    ) == expected_epoch


    num_classes = int(
        checkpoint[
            "model_configuration"
        ][
            "num_classes"
        ]
    )


    # ----------------------------------------------------------------------------------------------
    # Destination
    # ----------------------------------------------------------------------------------------------

    output_path = (

        FINAL_BEST_AA_VALIDATION_DIR
        /
        (
            f"{dataset.lower()}_seed{seed}_"
            "best_valAA_validation_outputs.npz"
        )
    )


    metrics_path = (

        FINAL_BEST_AA_VALIDATION_METRICS_DIR
        /
        (
            f"{dataset.lower()}_seed{seed}_"
            "best_valAA_validation_metrics.json"
        )
    )


    source_output_path = None
    source_output_sha = ""


    # ==============================================================================================
    # CASE A — BEST OA AND BEST AA WERE THE SAME ORIGINAL CHECKPOINT
    #
    # Reuse original Notebook 03 validation outputs.
    # ==============================================================================================

    if identity in SAME_OA_AA_IDENTITIES:

        original_path, original_sha = (
            locate_original_validation_output(
                dataset,
                seed
            )
        )


        # The original inference manifest's checkpoint hash
        # must equal our final checkpoint hash.
        original_manifest_row = (

            ORIGINAL_VALIDATION_MANIFEST_DF[
                (
                    ORIGINAL_VALIDATION_MANIFEST_DF[
                        "dataset"
                    ]
                    ==
                    dataset
                )
                &
                (
                    ORIGINAL_VALIDATION_MANIFEST_DF[
                        "seed"
                    ].astype(
                        int
                    )
                    ==
                    seed
                )
            ]
            .iloc[
                0
            ]
        )


        assert (
            str(
                original_manifest_row[
                    "checkpoint_sha256"
                ]
            )
            ==
            checkpoint_sha
        ), (
            f"{dataset} seed {seed}: "
            "original validation output was not produced "
            "from the final best-AA checkpoint."
        )


        arrays = (
            load_canonical_validation_arrays(
                original_path
            )
        )


        source_output_path = (
            original_path
        )

        source_output_sha = (
            original_sha
        )


        inference_source = (
            "reused_original_same_checkpoint"
        )


        REUSED_ORIGINAL_RUNS += 1


    # ==============================================================================================
    # CASE B — RECOVERED BEST-AA CHECKPOINT
    #
    # Fresh ordered validation inference on T4.
    # ==============================================================================================

    else:

        assert identity in RECOVERED_IDENTITIES


        cfg = RECOVERY_CONFIG[
            dataset
        ]


        split = RECOVERY_SPLITS[
            dataset
        ]


        val_dataset = RecoveryIndexedDataset(

            RECOVERY_VAL_DATASETS[
                dataset
            ],

            split[
                "val_indices"
            ]
        )


        val_loader = DataLoader(

            val_dataset,

            batch_size=
                CE_TRAINING_CONFIG[
                    "batch_size"
                ],

            shuffle=False,

            num_workers=0,

            pin_memory=True,

            drop_last=False,
        )


        # No stochastic inference, but keep initialization controlled.
        set_global_seed(
            seed
        )


        model = HybridSN(

            spectral_depth=
                int(
                    checkpoint[
                        "model_configuration"
                    ][
                        "spectral_depth"
                    ]
                ),

            patch_size=
                int(
                    checkpoint[
                        "model_configuration"
                    ][
                        "patch_size"
                    ]
                ),

            num_classes=
                num_classes,

            dropout=
                float(
                    checkpoint[
                        "model_configuration"
                    ][
                        "dropout"
                    ]
                ),
        ).to(
            DEVICE
        )


        model.load_state_dict(

            checkpoint[
                "model_state_dict"
            ]
        )


        criterion = nn.CrossEntropyLoss()


        validation_result = evaluate_model(

            model=model,

            loader=val_loader,

            criterion=criterion,

            num_classes=
                num_classes,

            device=DEVICE
        )


        validation_metrics = (
            validation_result[
                "metrics"
            ]
        )


        arrays = {

            "logits":
                np.asarray(
                    validation_result[
                        "logits"
                    ]
                ),

            "labels":
                np.asarray(
                    validation_result[
                        "labels"
                    ],
                    dtype=np.int64
                ),

            "predictions":
                np.asarray(
                    validation_result[
                        "predictions"
                    ],
                    dtype=np.int64
                ),

            "indices":
                np.asarray(
                    validation_result[
                        "indices"
                    ],
                    dtype=np.int64
                ),
        }


        # Exact frozen validation order.
        assert np.array_equal(

            arrays[
                "indices"
            ],

            split[
                "val_indices"
            ]
        )


        assert np.array_equal(

            arrays[
                "labels"
            ],

            split[
                "val_labels"
            ]
        )


        # Recovered checkpoint validation metrics must reproduce.
        saved_metrics = (
            checkpoint[
                "validation_metrics"
            ]
        )


        for metric in [

            "oa",
            "aa",
            "kappa",
            "macro_f1",
        ]:

            assert abs(

                float(
                    validation_metrics[
                        metric
                    ]
                )
                -
                float(
                    saved_metrics[
                        metric
                    ]
                )

            ) <= 1e-12, (
                f"{dataset} seed {seed}: "
                f"validation {metric} mismatch."
            )


        assert abs(

            float(
                validation_metrics[
                    "loss"
                ]
            )
            -
            float(
                saved_metrics[
                    "loss"
                ]
            )

        ) <= 1e-7


        inference_source = (
            "fresh_bestAA_checkpoint_inference"
        )


        GENERATED_GPU_RUNS += 1


        del validation_result
        del model
        del criterion

        torch.cuda.empty_cache()

        gc.collect()


    # ==============================================================================================
    # 9. COMMON OUTPUT INTEGRITY
    # ==============================================================================================

    assert arrays[
        "logits"
    ].ndim == 2


    assert arrays[
        "logits"
    ].shape[
        1
    ] == num_classes


    sample_count = int(
        arrays[
            "logits"
        ].shape[
            0
        ]
    )


    assert arrays[
        "labels"
    ].shape == (
        sample_count,
    )


    assert arrays[
        "predictions"
    ].shape == (
        sample_count,
    )


    assert arrays[
        "indices"
    ].shape == (
        sample_count,
    )


    assert np.isfinite(
        arrays[
            "logits"
        ]
    ).all()


    assert np.array_equal(

        arrays[
            "predictions"
        ],

        np.argmax(
            arrays[
                "logits"
            ],
            axis=1
        )
    )


    assert len(
        np.unique(
            arrays[
                "indices"
            ]
        )
    ) == sample_count


    # ==============================================================================================
    # 10. VERIFY CLASSIFICATION METRICS FROM OUTPUT ARRAYS
    # ==============================================================================================

    recomputed_metrics = (
        compute_classification_metrics(

            y_true=
                arrays[
                    "labels"
                ],

            y_pred=
                arrays[
                    "predictions"
                ],

            num_classes=
                num_classes
        )
    )


    saved_checkpoint_metrics = (
        checkpoint[
            "validation_metrics"
        ]
    )


    for metric in [

        "oa",
        "aa",
        "kappa",
        "macro_f1",
    ]:

        assert abs(

            float(
                recomputed_metrics[
                    metric
                ]
            )
            -
            float(
                saved_checkpoint_metrics[
                    metric
                ]
            )

        ) <= 1e-12, (
            f"{dataset} seed {seed}: "
            f"saved logit metric mismatch for {metric}."
        )


    # ==============================================================================================
    # 11. SAVE / VERIFY CANONICAL NPZ
    # ==============================================================================================

    save_or_verify_npz(

        output_path,

        arrays
    )


    output_sha = (
        sha256_file(
            output_path
        )
    )


    # ==============================================================================================
    # 12. SAVE VALIDATION METADATA
    # ==============================================================================================

    metrics_payload = {

        "project":
            "LTLC",

        "method":
            "HybridSN_CE",

        "checkpoint_purpose":
            "best_validation_AA",

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "checkpoint_epoch":
            int(
                expected_epoch
            ),

        "samples":
            int(
                sample_count
            ),

        "classes":
            int(
                num_classes
            ),

        "validation_metrics": {

            "oa":
                float(
                    saved_checkpoint_metrics[
                        "oa"
                    ]
                ),

            "aa":
                float(
                    saved_checkpoint_metrics[
                        "aa"
                    ]
                ),

            "kappa":
                float(
                    saved_checkpoint_metrics[
                        "kappa"
                    ]
                ),

            "macro_f1":
                float(
                    saved_checkpoint_metrics[
                        "macro_f1"
                    ]
                ),

            "loss":
                float(
                    saved_checkpoint_metrics[
                        "loss"
                    ]
                ),
        },

        "checkpoint":
            str(
                checkpoint_path.relative_to(
                    ROOT
                )
            ),

        "checkpoint_sha256":
            checkpoint_sha,

        "output":
            str(
                output_path.relative_to(
                    ROOT
                )
            ),

        "output_sha256":
            output_sha,

        "inference_source":
            inference_source,

        "source_original_validation_output":
            (
                ""
                if source_output_path is None
                else str(
                    source_output_path.relative_to(
                        ROOT
                    )
                )
            ),

        "source_original_validation_output_sha256":
            source_output_sha,

        "official_test_used":
            False,
    }


    save_or_verify_json(

        metrics_path,

        metrics_payload
    )


    metrics_sha = (
        sha256_file(
            metrics_path
        )
    )


    FINAL_VALIDATION_RECORDS.append({

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "best_valAA_epoch":
            int(
                expected_epoch
            ),

        "samples":
            int(
                sample_count
            ),

        "classes":
            int(
                num_classes
            ),

        "inference_source":
            inference_source,

        "output_file":
            str(
                output_path.relative_to(
                    ROOT
                )
            ),

        "output_sha256":
            output_sha,

        "metrics_file":
            str(
                metrics_path.relative_to(
                    ROOT
                )
            ),

        "metrics_sha256":
            metrics_sha,

        "checkpoint_file":
            str(
                checkpoint_path.relative_to(
                    ROOT
                )
            ),

        "checkpoint_sha256":
            checkpoint_sha,

        "official_test_used":
            False,

        "pass":
            True,
    })


    print(

        f"{dataset:11s} | "
        f"seed={seed:4d} | "
        f"epoch={expected_epoch:3d} | "
        f"N={sample_count:6d} | "
        f"K={num_classes:2d} | "
        f"{inference_source:37s} | "
        f"PASS"
    )


    del checkpoint
    del arrays


# ==================================================================================================
# 13. COUNTS
# ==================================================================================================

assert GENERATED_GPU_RUNS == 5

assert REUSED_ORIGINAL_RUNS == 4

assert len(
    FINAL_VALIDATION_RECORDS
) == 9


# ==================================================================================================
# 14. CROSS-SEED VALIDATION ORDER / LABEL CONSISTENCY
# ==================================================================================================

for dataset in [

    "Pingan",
    "Qingyun",
    "Tangdaowan",
]:

    dataset_records = [

        record

        for record in
        FINAL_VALIDATION_RECORDS

        if (
            record[
                "dataset"
            ]
            ==
            dataset
        )
    ]


    assert len(
        dataset_records
    ) == 3


    reference_indices = None
    reference_labels = None


    for record in dataset_records:

        output_path = (
            ROOT
            /
            record[
                "output_file"
            ]
        )


        arrays = (
            load_canonical_validation_arrays(
                output_path
            )
        )


        if reference_indices is None:

            reference_indices = (
                arrays[
                    "indices"
                ].copy()
            )


            reference_labels = (
                arrays[
                    "labels"
                ].copy()
            )


        else:

            assert np.array_equal(

                arrays[
                    "indices"
                ],

                reference_indices
            )


            assert np.array_equal(

                arrays[
                    "labels"
                ],

                reference_labels
            )


    print(

        f"{dataset:11s} | "
        "cross-seed validation indices/labels | PASS"
    )


# ==================================================================================================
# 15. MANIFEST
# ==================================================================================================

FINAL_VALIDATION_MANIFEST_DF = (
    pd.DataFrame(
        FINAL_VALIDATION_RECORDS
    )
)


FINAL_VALIDATION_MANIFEST_CSV = (

    RECOVERY_AUDIT_DIR
    /
    "notebook05_best_valAA_validation_output_manifest.csv"
)


FINAL_VALIDATION_MANIFEST_JSON = (

    RECOVERY_AUDIT_DIR
    /
    "notebook05_best_valAA_validation_output_manifest.json"
)


FINAL_VALIDATION_MANIFEST_DF.to_csv(

    FINAL_VALIDATION_MANIFEST_CSV,

    index=False
)


manifest_payload = {

    "project":
        "LTLC",

    "method":
        "HybridSN_CE",

    "checkpoint_purpose":
        "best_validation_AA",

    "total_validation_outputs":
        9,

    "fresh_gpu_inference_outputs":
        5,

    "reused_same_checkpoint_outputs":
        4,

    "runs":
        FINAL_VALIDATION_RECORDS,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


save_or_verify_json(

    FINAL_VALIDATION_MANIFEST_JSON,

    manifest_payload
)


FINAL_VALIDATION_MANIFEST_CSV_SHA = (
    sha256_file(
        FINAL_VALIDATION_MANIFEST_CSV
    )
)


FINAL_VALIDATION_MANIFEST_JSON_SHA = (
    sha256_file(
        FINAL_VALIDATION_MANIFEST_JSON
    )
)


# ==================================================================================================
# 16. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED is False


NOTEBOOK05_BEST_AA_VALIDATION_LOGITS_GATE_PASS = True


print()
print("=" * 150)

print(
    "Unified best-AA validation outputs       : 9/9 PASS"
)

print(
    "Fresh recovered-checkpoint GPU inference : 5/5 PASS"
)

print(
    "Original same-checkpoint outputs reused  : 4/4 PASS"
)

print(
    "Cross-seed indices/labels                : 3/3 datasets PASS"
)

print()

print(
    "Manifest CSV:"
)

print(
    FINAL_VALIDATION_MANIFEST_CSV
)

print(
    "SHA-256:"
)

print(
    FINAL_VALIDATION_MANIFEST_CSV_SHA
)

print()

print(
    "Manifest JSON:"
)

print(
    FINAL_VALIDATION_MANIFEST_JSON
)

print(
    "SHA-256:"
)

print(
    FINAL_VALIDATION_MANIFEST_JSON_SHA
)

print()

print(
    "Official-test dataset created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used:",
    NOTEBOOK05_RECOVERY_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 BEST-AA VALIDATION LOGIT GATE: PASS"
)

print(
    "NOTEBOOK 05 GPU-DEPENDENT VALIDATION INFERENCE: COMPLETE"
)

print(
    "SAFE TO SWITCH TO CPU FOR THE REMAINDER OF NOTEBOOK 05"
)

print(
    "NOTEBOOK 06 WILL REQUIRE T4 AGAIN FOR OFFICIAL-TEST INFERENCE"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 150)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 18A
# CPU CLEAN-ROOM RESUME
#
# PURPOSE
#   1. Reconnect Google Drive after runtime reset
#   2. Verify the frozen Cell 17A / 17B artifacts by SHA-256
#   3. Reconstruct the 9 best-val-AA checkpoint map
#   4. Reconstruct the 9 best-val-AA validation-output map
#   5. Re-establish strict OFFICIAL-TEST safety flags
#
# NO TRAINING
# NO GPU
# NO OFFICIAL-TEST DATA
# ==================================================================================================

from pathlib import Path
import hashlib
import json
import os
import sys

import numpy as np
import pandas as pd
import torch


# ==================================================================================================
# 1. GOOGLE DRIVE
# ==================================================================================================

from google.colab import drive

drive.mount(
    "/content/drive",
    force_remount=False
)


ROOT = Path(
    "/content/drive/MyDrive/CSE498/LTLC"
)


assert ROOT.exists(), (
    f"Project root not found:\n{ROOT}"
)


print("=" * 145)
print("NOTEBOOK 05 — CPU CLEAN-ROOM RESUME")
print("=" * 145)

print("Project root :", ROOT)
print("Device       : CPU")
print("PyTorch      :", torch.__version__)
print("CUDA active  :", torch.cuda.is_available())


# ==================================================================================================
# 2. HARD CPU GATE
# ==================================================================================================

DEVICE = torch.device(
    "cpu"
)


# CUDA being available is not itself a failure,
# but Notebook 05 will deliberately use CPU from this point.
assert DEVICE.type == "cpu"


# ==================================================================================================
# 3. SHA-256 UTILITY
# ==================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    path = Path(
        path
    )


    assert path.exists(), (
        f"Missing file:\n{path}"
    )


    digest = hashlib.sha256()


    with open(
        path,
        "rb"
    ) as f:

        while True:

            block = f.read(
                chunk_size
            )

            if not block:
                break

            digest.update(
                block
            )


    return digest.hexdigest()


# ==================================================================================================
# 4. FROZEN ARTIFACT PATHS FROM CELLS 17A / 17B
# ==================================================================================================

RECOVERY_AUDIT_DIR = (
    ROOT
    /
    "audit"
    /
    "notebook05_posthoc_ltlc"
    /
    "best_valAA_recovery"
)


FINAL_BEST_AA_PREFLIGHT_PATH = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_final_9_best_valAA_checkpoint_preflight.json"
)


FINAL_VALIDATION_MANIFEST_CSV = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_best_valAA_validation_output_manifest.csv"
)


FINAL_VALIDATION_MANIFEST_JSON = (
    RECOVERY_AUDIT_DIR
    /
    "notebook05_best_valAA_validation_output_manifest.json"
)


for required_path in [

    FINAL_BEST_AA_PREFLIGHT_PATH,
    FINAL_VALIDATION_MANIFEST_CSV,
    FINAL_VALIDATION_MANIFEST_JSON,

]:

    assert required_path.exists(), (
        f"Missing frozen Notebook 05 artifact:\n"
        f"{required_path}"
    )


# ==================================================================================================
# 5. EXACT FROZEN SHA-256 VALUES
# ==================================================================================================

EXPECTED_RESUME_HASHES = {

    "bestAA_preflight_json":
        "9dec9933a0665aeca1b0864aa0eff997fa21b8b1d8fe5328c1ccdffc411ec411",

    "bestAA_validation_manifest_csv":
        "d66c1efe67c9c7f6ce660ae69a3bacf61fd85a8a0c8b18a10f54d4e977d5dcaa",

    "bestAA_validation_manifest_json":
        "ab12c9b95fde876bff6a66f737f1c8c9426ec045bd2cc95ff6b34654ba18194b",
}


ACTUAL_RESUME_HASHES = {

    "bestAA_preflight_json":
        sha256_file(
            FINAL_BEST_AA_PREFLIGHT_PATH
        ),

    "bestAA_validation_manifest_csv":
        sha256_file(
            FINAL_VALIDATION_MANIFEST_CSV
        ),

    "bestAA_validation_manifest_json":
        sha256_file(
            FINAL_VALIDATION_MANIFEST_JSON
        ),
}


for key in EXPECTED_RESUME_HASHES:

    assert (
        ACTUAL_RESUME_HASHES[
            key
        ]
        ==
        EXPECTED_RESUME_HASHES[
            key
        ]
    ), (
        f"Frozen artifact hash mismatch:\n"
        f"{key}\n"
        f"Expected: {EXPECTED_RESUME_HASHES[key]}\n"
        f"Actual  : {ACTUAL_RESUME_HASHES[key]}"
    )


print()
print("Frozen Cell 17A/17B artifact hashes: 3/3 PASS")


# ==================================================================================================
# 6. LOAD PREFLIGHT + VALIDATION MANIFESTS
# ==================================================================================================

with open(
    FINAL_BEST_AA_PREFLIGHT_PATH,
    "r",
    encoding="utf-8"
) as f:

    FINAL_BEST_AA_PREFLIGHT = json.load(
        f
    )


FINAL_VALIDATION_MANIFEST_DF = pd.read_csv(
    FINAL_VALIDATION_MANIFEST_CSV
)


with open(
    FINAL_VALIDATION_MANIFEST_JSON,
    "r",
    encoding="utf-8"
) as f:

    FINAL_VALIDATION_MANIFEST = json.load(
        f
    )


assert (
    FINAL_BEST_AA_PREFLIGHT[
        "total_runs"
    ]
    ==
    9
)


assert (
    FINAL_BEST_AA_PREFLIGHT[
        "recovered_runs"
    ]
    ==
    5
)


assert (
    FINAL_BEST_AA_PREFLIGHT[
        "original_same_epoch_OA_AA_runs"
    ]
    ==
    4
)


assert len(
    FINAL_VALIDATION_MANIFEST_DF
) == 9


assert (
    FINAL_VALIDATION_MANIFEST[
        "total_validation_outputs"
    ]
    ==
    9
)


assert (
    FINAL_VALIDATION_MANIFEST[
        "fresh_gpu_inference_outputs"
    ]
    ==
    5
)


assert (
    FINAL_VALIDATION_MANIFEST[
        "reused_same_checkpoint_outputs"
    ]
    ==
    4
)


# ==================================================================================================
# 7. RECONSTRUCT FINAL 9 BEST-AA CHECKPOINT MAP
# ==================================================================================================

FINAL_BEST_AA_CHECKPOINTS = {}

FINAL_BEST_AA_EPOCHS = {}

RECOVERED_IDENTITIES = set()

SAME_OA_AA_IDENTITIES = set()


for run in FINAL_BEST_AA_PREFLIGHT[
    "runs"
]:

    identity = (
        str(
            run[
                "dataset"
            ]
        ),
        int(
            run[
                "seed"
            ]
        )
    )


    checkpoint_path = (
        ROOT
        /
        run[
            "checkpoint"
        ]
    )


    assert checkpoint_path.exists(), (
        f"Missing checkpoint:\n"
        f"{checkpoint_path}"
    )


    actual_checkpoint_sha = (
        sha256_file(
            checkpoint_path
        )
    )


    assert (
        actual_checkpoint_sha
        ==
        run[
            "checkpoint_sha256"
        ]
    ), (
        f"Checkpoint SHA mismatch: {identity}"
    )


    assert (
        run[
            "test_used_for_selection"
        ]
        is False
    )


    source_type = str(
        run[
            "source_type"
        ]
    )


    FINAL_BEST_AA_CHECKPOINTS[
        identity
    ] = {

        "path":
            checkpoint_path,

        "source_type":
            source_type,

        "sha256":
            actual_checkpoint_sha,
    }


    FINAL_BEST_AA_EPOCHS[
        identity
    ] = int(
        run[
            "best_valAA_epoch"
        ]
    )


    if (
        source_type
        ==
        "recovered_best_validation_AA"
    ):

        RECOVERED_IDENTITIES.add(
            identity
        )


    elif (
        source_type
        ==
        "original_bestOA_same_epoch_as_bestAA"
    ):

        SAME_OA_AA_IDENTITIES.add(
            identity
        )


    else:

        raise RuntimeError(
            f"Unexpected checkpoint source type: "
            f"{source_type}"
        )


assert len(
    FINAL_BEST_AA_CHECKPOINTS
) == 9


assert len(
    FINAL_BEST_AA_EPOCHS
) == 9


assert len(
    RECOVERED_IDENTITIES
) == 5


assert len(
    SAME_OA_AA_IDENTITIES
) == 4


assert (
    RECOVERED_IDENTITIES
    &
    SAME_OA_AA_IDENTITIES
) == set()


# ==================================================================================================
# 8. RECONSTRUCT FINAL 9 VALIDATION-OUTPUT MAP
# ==================================================================================================

FINAL_BEST_AA_VALIDATION_OUTPUTS = {}

FINAL_BEST_AA_VALIDATION_METRICS = {}


for _, row in FINAL_VALIDATION_MANIFEST_DF.iterrows():

    identity = (
        str(
            row[
                "dataset"
            ]
        ),
        int(
            row[
                "seed"
            ]
        )
    )


    output_path = (
        ROOT
        /
        str(
            row[
                "output_file"
            ]
        )
    )


    metrics_path = (
        ROOT
        /
        str(
            row[
                "metrics_file"
            ]
        )
    )


    checkpoint_path = (
        ROOT
        /
        str(
            row[
                "checkpoint_file"
            ]
        )
    )


    assert output_path.exists()
    assert metrics_path.exists()
    assert checkpoint_path.exists()


    # ----------------------------------------------------------------------------------------------
    # Verify all three hashes from frozen manifest
    # ----------------------------------------------------------------------------------------------

    assert (
        sha256_file(
            output_path
        )
        ==
        str(
            row[
                "output_sha256"
            ]
        )
    ), (
        f"Validation output SHA mismatch: "
        f"{identity}"
    )


    assert (
        sha256_file(
            metrics_path
        )
        ==
        str(
            row[
                "metrics_sha256"
            ]
        )
    ), (
        f"Validation metrics SHA mismatch: "
        f"{identity}"
    )


    assert (
        sha256_file(
            checkpoint_path
        )
        ==
        str(
            row[
                "checkpoint_sha256"
            ]
        )
    ), (
        f"Checkpoint SHA mismatch from validation manifest: "
        f"{identity}"
    )


    # ----------------------------------------------------------------------------------------------
    # Cross-manifest checkpoint consistency
    # ----------------------------------------------------------------------------------------------

    assert (
        checkpoint_path
        ==
        FINAL_BEST_AA_CHECKPOINTS[
            identity
        ][
            "path"
        ]
    )


    assert (
        int(
            row[
                "best_valAA_epoch"
            ]
        )
        ==
        FINAL_BEST_AA_EPOCHS[
            identity
        ]
    )


    # ----------------------------------------------------------------------------------------------
    # Validation only
    # ----------------------------------------------------------------------------------------------

    assert (
        bool(
            row[
                "official_test_used"
            ]
        )
        is False
    )


    FINAL_BEST_AA_VALIDATION_OUTPUTS[
        identity
    ] = output_path


    FINAL_BEST_AA_VALIDATION_METRICS[
        identity
    ] = metrics_path


assert len(
    FINAL_BEST_AA_VALIDATION_OUTPUTS
) == 9


assert len(
    FINAL_BEST_AA_VALIDATION_METRICS
) == 9


# ==================================================================================================
# 9. JSON MANIFEST SAFETY CHECK
# ==================================================================================================

assert (
    FINAL_VALIDATION_MANIFEST[
        "official_test_dataset_created"
    ]
    is False
)


assert (
    FINAL_VALIDATION_MANIFEST[
        "official_test_dataloader_created"
    ]
    is False
)


assert (
    FINAL_VALIDATION_MANIFEST[
        "official_test_used"
    ]
    is False
)


for run in FINAL_VALIDATION_MANIFEST[
    "runs"
]:

    assert (
        run[
            "official_test_used"
        ]
        is False
    )


# ==================================================================================================
# 10. RE-ESTABLISH NOTEBOOK 05 SAFETY FLAGS IN FRESH CPU RUNTIME
# ==================================================================================================

NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED = False
NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED = False
NOTEBOOK05_OFFICIAL_TEST_USED = False

NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE = False

NOTEBOOK05_GPU_VALIDATION_PHASE_COMPLETE = True


# ==================================================================================================
# 11. EXPECTED DATASET / SEED STRUCTURE
# ==================================================================================================

EXPECTED_IDENTITIES = {

    ("Pingan", 42),
    ("Pingan", 123),
    ("Pingan", 3407),

    ("Qingyun", 42),
    ("Qingyun", 123),
    ("Qingyun", 3407),

    ("Tangdaowan", 42),
    ("Tangdaowan", 123),
    ("Tangdaowan", 3407),
}


assert (
    set(
        FINAL_BEST_AA_CHECKPOINTS.keys()
    )
    ==
    EXPECTED_IDENTITIES
)


assert (
    set(
        FINAL_BEST_AA_VALIDATION_OUTPUTS.keys()
    )
    ==
    EXPECTED_IDENTITIES
)


# ==================================================================================================
# 12. COMPACT AUDIT TABLE
# ==================================================================================================

CPU_RESUME_RECORDS = []


print()
print("=" * 145)
print("FINAL BEST-AA ARTIFACT INVENTORY")
print("=" * 145)


for identity in [

    ("Pingan", 42),
    ("Pingan", 123),
    ("Pingan", 3407),

    ("Qingyun", 42),
    ("Qingyun", 123),
    ("Qingyun", 3407),

    ("Tangdaowan", 42),
    ("Tangdaowan", 123),
    ("Tangdaowan", 3407),

]:

    dataset, seed = identity


    checkpoint_info = (
        FINAL_BEST_AA_CHECKPOINTS[
            identity
        ]
    )


    validation_output = (
        FINAL_BEST_AA_VALIDATION_OUTPUTS[
            identity
        ]
    )


    epoch = (
        FINAL_BEST_AA_EPOCHS[
            identity
        ]
    )


    CPU_RESUME_RECORDS.append({

        "dataset":
            dataset,

        "seed":
            seed,

        "best_valAA_epoch":
            epoch,

        "checkpoint_source":
            checkpoint_info[
                "source_type"
            ],

        "validation_output":
            validation_output.name,

        "pass":
            True,
    })


    print(

        f"{dataset:11s} | "
        f"seed={seed:4d} | "
        f"epoch={epoch:3d} | "
        f"{checkpoint_info['source_type']:39s} | "
        f"PASS"
    )


CPU_RESUME_AUDIT_DF = pd.DataFrame(
    CPU_RESUME_RECORDS
)


# ==================================================================================================
# 13. FINAL CPU RESUME GATE
# ==================================================================================================

assert len(
    CPU_RESUME_AUDIT_DF
) == 9


assert (
    CPU_RESUME_AUDIT_DF[
        "pass"
    ].all()
)


assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_CPU_RESUME_GATE_PASS = True


print()
print("=" * 145)

print(
    "Frozen resume artifact hashes       : 3/3 PASS"
)

print(
    "Final best-AA checkpoints           : 9/9 SHA-VERIFIED"
)

print(
    "Final best-AA validation outputs    : 9/9 SHA-VERIFIED"
)

print(
    "Final validation metric files       : 9/9 SHA-VERIFIED"
)

print(
    "Recovered checkpoint identities     : 5/5 PASS"
)

print(
    "Original OA=AA checkpoint identities: 4/4 PASS"
)

print()

print(
    "Runtime device                      :",
    DEVICE
)

print(
    "Training performed in CPU phase     :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official-test dataset created       :",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created    :",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used                  :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 CPU CLEAN-ROOM RESUME GATE: PASS"
)

print(
    "READY FOR CPU-ONLY POST-HOC CALIBRATION / LTLC ANALYSIS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 18B
# CPU RECOVERY OF FROZEN POST-HOC / LTLC CONFIGURATION ARTIFACTS
#
# PURPOSE
#   Locate and inspect the Notebook 05 artifacts created BEFORE the best-AA recovery detour:
#     - class-prior / rarity definitions
#     - calibration grids / math-engine configuration
#     - Tangdaowan development-gate artifacts
#
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

from pathlib import Path
import json
import hashlib


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_CPU_RESUME_GATE_PASS is True

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert DEVICE.type == "cpu"


# ==================================================================================================
# 2. NOTEBOOK 05 AUDIT ROOT
# ==================================================================================================

NB05_AUDIT_ROOT = (
    ROOT
    /
    "audit"
    /
    "notebook05_posthoc_ltlc"
)


assert NB05_AUDIT_ROOT.exists()


# ==================================================================================================
# 3. INVENTORY JSON / CSV ARTIFACTS
# ==================================================================================================

candidate_files = sorted(

    [

        path

        for path in NB05_AUDIT_ROOT.rglob("*")

        if (
            path.is_file()
            and
            path.suffix.lower() in {
                ".json",
                ".csv"
            }
        )
    ],

    key=lambda p: str(p)
)


assert len(
    candidate_files
) > 0


print("=" * 145)
print("NOTEBOOK 05 — FROZEN POST-HOC ARTIFACT DISCOVERY")
print("=" * 145)

print(
    "Audit root:",
    NB05_AUDIT_ROOT
)

print(
    "Candidate JSON/CSV artifacts:",
    len(
        candidate_files
    )
)

print()


# ==================================================================================================
# 4. SHOW JSON TOP-LEVEL KEYS + SHA
# ==================================================================================================

DISCOVERED_JSON_RECORDS = []


for path in candidate_files:

    if (
        path.suffix.lower()
        !=
        ".json"
    ):

        continue


    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:

            payload = json.load(
                f
            )


        if isinstance(
            payload,
            dict
        ):

            top_keys = list(
                payload.keys()
            )

        else:

            top_keys = [
                f"<{type(payload).__name__}>"
            ]


        sha = sha256_file(
            path
        )


        relative_path = str(
            path.relative_to(
                ROOT
            )
        )


        DISCOVERED_JSON_RECORDS.append({

            "path":
                path,

            "relative_path":
                relative_path,

            "sha256":
                sha,

            "top_keys":
                top_keys,
        })


        print(
            "-" * 145
        )

        print(
            relative_path
        )

        print(
            "SHA-256:",
            sha
        )

        print(
            "Top-level keys:",
            top_keys
        )


    except Exception as exc:

        print(
            "-" * 145
        )

        print(
            path.relative_to(
                ROOT
            )
        )

        print(
            "JSON READ ERROR:",
            repr(
                exc
            )
        )


# ==================================================================================================
# 5. SEARCH ARTIFACT CONTENT FOR IMPORTANT POST-HOC TERMS
# ==================================================================================================

SEARCH_TERMS = [

    "prior",
    "rarity",
    "head",
    "medium",
    "tail",

    "tau",
    "alpha",

    "temperature",
    "lbfgs",
    "calibration",

    "tangdaowan",
    "saturation",

    "revise",
    "development",
]


MATCHED_ARTIFACTS = []


print()
print("=" * 145)
print("POST-HOC CONFIGURATION / DEVELOPMENT ARTIFACT MATCHES")
print("=" * 145)


for record in DISCOVERED_JSON_RECORDS:

    path = record[
        "path"
    ]


    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        raw_text = f.read()


    raw_lower = (
        raw_text.lower()
    )


    matched_terms = [

        term

        for term in SEARCH_TERMS

        if (
            term.lower()
            in
            raw_lower
        )
    ]


    if len(
        matched_terms
    ) == 0:

        continue


    MATCHED_ARTIFACTS.append({

        "path":
            path,

        "relative_path":
            record[
                "relative_path"
            ],

        "sha256":
            record[
                "sha256"
            ],

        "matched_terms":
            matched_terms,
    })


    print(
        f"{record['relative_path']}"
    )

    print(
        "  terms:",
        matched_terms
    )

    print(
        "  sha  :",
        record[
            "sha256"
        ]
    )


# ==================================================================================================
# 6. VERIFY KNOWN FROZEN ARTIFACT HASHES IF PRESENT
#
# These were frozen before the recovery detour.
# ==================================================================================================

KNOWN_PRE_RECOVERY_HASHES = {

    "priors_rarity_manifest":
        "9698f70003c44834b7cbb29f2c69232d2dd1900eec4b57290541fd95d4ddc478",

    "tangdaowan_saturation_amendment":
        "12ebdc1a2b158f4aba9f6775d671c22b8f22565e24c57a9b138f89ed9931b88a",
}


FOUND_KNOWN_HASHES = {}


for record in DISCOVERED_JSON_RECORDS:

    for logical_name, expected_sha in (
        KNOWN_PRE_RECOVERY_HASHES.items()
    ):

        if (
            record[
                "sha256"
            ]
            ==
            expected_sha
        ):

            FOUND_KNOWN_HASHES[
                logical_name
            ] = record[
                "path"
            ]


print()
print("=" * 145)
print("KNOWN PRE-RECOVERY FROZEN ARTIFACTS")
print("=" * 145)


for logical_name, expected_sha in (
    KNOWN_PRE_RECOVERY_HASHES.items()
):

    if logical_name in FOUND_KNOWN_HASHES:

        print(
            f"{logical_name:35s}: FOUND"
        )

        print(
            "  path:",
            FOUND_KNOWN_HASHES[
                logical_name
            ]
        )

        print(
            "  sha :",
            expected_sha
        )

    else:

        print(
            f"{logical_name:35s}: NOT LOCATED BY HASH"
        )


# ==================================================================================================
# 7. SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_FROZEN_POSTHOC_DISCOVERY_PASS = True


print()
print("=" * 145)

print(
    "JSON artifacts inspected       :",
    len(
        DISCOVERED_JSON_RECORDS
    )
)

print(
    "Relevant artifacts matched     :",
    len(
        MATCHED_ARTIFACTS
    )
)

print(
    "Prior/rarity frozen hash found :",
    "priors_rarity_manifest"
    in
    FOUND_KNOWN_HASHES
)

print(
    "Saturation amendment hash found:",
    "tangdaowan_saturation_amendment"
    in
    FOUND_KNOWN_HASHES
)

print()

print(
    "Runtime device                 :",
    DEVICE
)

print(
    "Training performed             :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used             :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 FROZEN POST-HOC ARTIFACT DISCOVERY: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 18C-v2
# RESTORE EXACT FROZEN LTLC / POST-HOC CONFIGURATION
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
from pprint import pprint

import numpy as np


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_CPU_RESUME_GATE_PASS is True
assert NOTEBOOK05_FROZEN_POSTHOC_DISCOVERY_PASS is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


# ==================================================================================================
# 2. FROZEN ARTIFACT PATHS
# ==================================================================================================

IMPLEMENTATION_SPEC_PATH = (
    NB05_AUDIT_ROOT
    /
    "notebook05_posthoc_ltlc_implementation_spec.json"
)

PRIOR_RARITY_PATH = (
    NB05_AUDIT_ROOT
    /
    "notebook05_training_prior_rarity_manifest.json"
)

SATURATION_AMENDMENT_PATH = (
    NB05_AUDIT_ROOT
    /
    "notebook05_tangdaowan_saturation_protocol_amendment.json"
)

OLD_TANGDAOWAN_SHIFT_SELECTION_PATH = (
    NB05_AUDIT_ROOT
    /
    "tangdaowan_ltlc_shift_validation_selection_manifest.json"
)


for path in [
    IMPLEMENTATION_SPEC_PATH,
    PRIOR_RARITY_PATH,
    SATURATION_AMENDMENT_PATH,
    OLD_TANGDAOWAN_SHIFT_SELECTION_PATH,
]:

    assert path.exists(), (
        f"Missing frozen artifact:\n{path}"
    )


# ==================================================================================================
# 3. VERIFY EXACT FROZEN HASHES
# ==================================================================================================

EXPECTED_FROZEN_HASHES = {

    "implementation_spec":
        "2d9b944ee4a2971684ab60b97682cc2472c389d8c982eecf7d49aa7fb53f1af8",

    "prior_rarity":
        "9698f70003c44834b7cbb29f2c69232d2dd1900eec4b57290541fd95d4ddc478",

    "saturation_amendment":
        "12ebdc1a2b158f4aba9f6775d671c22b8f22565e24c57a9b138f89ed9931b88a",

    "old_tangdaowan_shift_selection":
        "f9867a9d0e99544977370aba98925cb0f23b8dfb6a2bff5f09f3a97e8585ae47",
}


ACTUAL_FROZEN_HASHES = {

    "implementation_spec":
        sha256_file(
            IMPLEMENTATION_SPEC_PATH
        ),

    "prior_rarity":
        sha256_file(
            PRIOR_RARITY_PATH
        ),

    "saturation_amendment":
        sha256_file(
            SATURATION_AMENDMENT_PATH
        ),

    "old_tangdaowan_shift_selection":
        sha256_file(
            OLD_TANGDAOWAN_SHIFT_SELECTION_PATH
        ),
}


for key in EXPECTED_FROZEN_HASHES:

    assert (
        ACTUAL_FROZEN_HASHES[key]
        ==
        EXPECTED_FROZEN_HASHES[key]
    ), (
        f"Frozen SHA mismatch: {key}"
    )


# ==================================================================================================
# 4. LOAD FROZEN PAYLOADS
# ==================================================================================================

with open(
    IMPLEMENTATION_SPEC_PATH,
    "r",
    encoding="utf-8"
) as f:

    LTLC_IMPLEMENTATION_SPEC = json.load(f)


with open(
    PRIOR_RARITY_PATH,
    "r",
    encoding="utf-8"
) as f:

    TRAINING_PRIOR_RARITY_MANIFEST = json.load(f)


with open(
    SATURATION_AMENDMENT_PATH,
    "r",
    encoding="utf-8"
) as f:

    TANGDAOWAN_SATURATION_AMENDMENT = json.load(f)


with open(
    OLD_TANGDAOWAN_SHIFT_SELECTION_PATH,
    "r",
    encoding="utf-8"
) as f:

    OLD_TANGDAOWAN_SHIFT_SELECTION = json.load(f)


# ==================================================================================================
# 5. RESTORE EXACT SEARCH GRIDS
# ==================================================================================================

TAU_GRID = np.asarray(
    LTLC_IMPLEMENTATION_SPEC[
        "tau_grid"
    ],
    dtype=np.float64
)


ALPHA_GRID = np.asarray(
    LTLC_IMPLEMENTATION_SPEC[
        "alpha_grid"
    ],
    dtype=np.float64
)


assert np.array_equal(

    TAU_GRID,

    np.asarray(
        [
            0.00,
            0.25,
            0.50,
            0.75,
            1.00,
            1.25,
            1.50,
            1.75,
            2.00,
        ],
        dtype=np.float64
    )
)


assert np.array_equal(

    ALPHA_GRID,

    np.asarray(
        [
            0.00,
            0.25,
            0.50,
            0.75,
            1.00,
            1.50,
            2.00,
        ],
        dtype=np.float64
    )
)


# ==================================================================================================
# 6. RESTORE EXACT TEMPERATURE PARAMETER CONFIGURATION
#
# Frozen JSON structure:
#
# temperature_bounds  = {"a": [-2, 2], "b": [-2, 2]}
# temperature_initial = {"a": 0, "b": 0}
# ==================================================================================================

TEMPERATURE_BOUNDS = {

    "a":
        tuple(
            float(x)

            for x in
            LTLC_IMPLEMENTATION_SPEC[
                "temperature_bounds"
            ][
                "a"
            ]
        ),

    "b":
        tuple(
            float(x)

            for x in
            LTLC_IMPLEMENTATION_SPEC[
                "temperature_bounds"
            ][
                "b"
            ]
        ),
}


TEMPERATURE_INITIAL = {

    "a":
        float(
            LTLC_IMPLEMENTATION_SPEC[
                "temperature_initial"
            ][
                "a"
            ]
        ),

    "b":
        float(
            LTLC_IMPLEMENTATION_SPEC[
                "temperature_initial"
            ][
                "b"
            ]
        ),
}


assert TEMPERATURE_BOUNDS == {

    "a":
        (-2.0, 2.0),

    "b":
        (-2.0, 2.0),
}


assert TEMPERATURE_INITIAL == {

    "a":
        0.0,

    "b":
        0.0,
}


TEMPERATURE_OPTIMIZER = (
    LTLC_IMPLEMENTATION_SPEC[
        "temperature_optimizer"
    ]
)


TEMPERATURE_OBJECTIVE = (
    LTLC_IMPLEMENTATION_SPEC[
        "temperature_objective"
    ]
)


assert TEMPERATURE_OPTIMIZER == "L-BFGS-B"

assert (
    TEMPERATURE_OBJECTIVE
    ==
    "validation_multiclass_NLL"
)


# ==================================================================================================
# 7. RARITY EPSILON + DATASET PAYLOADS
# ==================================================================================================

RARITY_EPSILON = float(
    TRAINING_PRIOR_RARITY_MANIFEST[
        "rarity_epsilon"
    ]
)


PRIOR_RARITY_BY_DATASET = (
    TRAINING_PRIOR_RARITY_MANIFEST[
        "datasets"
    ]
)


assert set(
    PRIOR_RARITY_BY_DATASET.keys()
) == {
    "Pingan",
    "Qingyun",
    "Tangdaowan",
}


# ==================================================================================================
# 8. FROZEN SAFETY CHECKS
# ==================================================================================================

assert LTLC_IMPLEMENTATION_SPEC[
    "official_test_used"
] is False

assert TRAINING_PRIOR_RARITY_MANIFEST[
    "official_test_used"
] is False

assert TANGDAOWAN_SATURATION_AMENDMENT[
    "official_test_used"
] is False

assert OLD_TANGDAOWAN_SHIFT_SELECTION[
    "official_test_used"
] is False

assert TRAINING_PRIOR_RARITY_MANIFEST[
    "training_performed"
] is False


# ==================================================================================================
# 9. DISPLAY CORE CONFIGURATION
# ==================================================================================================

print("=" * 145)
print("NOTEBOOK 05 — FROZEN LTLC MATHEMATICAL CONFIGURATION")
print("=" * 145)

print()
print(
    "Implementation spec SHA-256:",
    ACTUAL_FROZEN_HASHES[
        "implementation_spec"
    ]
)

print(
    "Prior/rarity manifest SHA-256:",
    ACTUAL_FROZEN_HASHES[
        "prior_rarity"
    ]
)

print()

print(
    "Tau grid:",
    TAU_GRID.tolist()
)

print(
    "Alpha grid:",
    ALPHA_GRID.tolist()
)

print(
    "Temperature bounds:",
    TEMPERATURE_BOUNDS
)

print(
    "Temperature initial:",
    TEMPERATURE_INITIAL
)

print(
    "Temperature optimizer:",
    TEMPERATURE_OPTIMIZER
)

print(
    "Temperature objective:",
    TEMPERATURE_OBJECTIVE
)

print(
    "Rarity epsilon:",
    RARITY_EPSILON
)


# ==================================================================================================
# 10. EXACT FROZEN FORMULAS / SELECTION RULES
# ==================================================================================================

print()
print("=" * 145)
print("FROZEN POST-HOC DEFINITIONS")
print("=" * 145)


fields_to_show = [

    "standard_posthoc_LA",

    "LTLC_shift",

    "LTLC_temperature",

    "LTLC_calibrated_logits",

    "shift_selection_rule",

    "temperature_optimizer",

    "temperature_objective",

    "mandatory_sanity_checks",
]


for field in fields_to_show:

    print()
    print(
        f"{field}:"
    )

    pprint(
        LTLC_IMPLEMENTATION_SPEC[
            field
        ],
        sort_dicts=False
    )


# ==================================================================================================
# 11. EXACT PRIOR / RARITY DEFINITIONS
# ==================================================================================================

print()
print("=" * 145)
print("FROZEN PRIOR / RARITY DEFINITIONS")
print("=" * 145)

print()
print("Prior definition:")

pprint(
    TRAINING_PRIOR_RARITY_MANIFEST[
        "prior_definition"
    ],
    sort_dicts=False
)


print()
print("Rarity definition:")

pprint(
    TRAINING_PRIOR_RARITY_MANIFEST[
        "rarity_definition"
    ],
    sort_dicts=False
)


print()
print("Frequency-group definition:")

pprint(
    TRAINING_PRIOR_RARITY_MANIFEST[
        "frequency_group_definition"
    ],
    sort_dicts=False
)


# ==================================================================================================
# 12. EXACT DATASET PRIOR / RARITY PAYLOADS
# ==================================================================================================

for dataset in [
    "Pingan",
    "Qingyun",
    "Tangdaowan",
]:

    print()
    print("=" * 145)

    print(
        f"{dataset.upper()} — FROZEN TRAINING PRIOR / RARITY PAYLOAD"
    )

    print("=" * 145)


    pprint(
        PRIOR_RARITY_BY_DATASET[
            dataset
        ],
        sort_dicts=False
    )


# ==================================================================================================
# 13. PRE-RECOVERY TANGDAOWAN SELECTION
#     PROVENANCE ONLY — NOT FINAL
# ==================================================================================================

print()
print("=" * 145)
print("PRE-RECOVERY TANGDAOWAN DEVELOPMENT SELECTION — PROVENANCE ONLY")
print("=" * 145)

print()
print("Status:")

pprint(
    OLD_TANGDAOWAN_SHIFT_SELECTION[
        "status"
    ],
    sort_dicts=False
)


print()
print("Standard-LA selection:")

pprint(
    OLD_TANGDAOWAN_SHIFT_SELECTION[
        "standard_la_selection"
    ],
    sort_dicts=False
)


print()
print("LTLC-shift selection:")

pprint(
    OLD_TANGDAOWAN_SHIFT_SELECTION[
        "ltlc_shift_selection"
    ],
    sort_dicts=False
)


# ==================================================================================================
# 14. TANGDAOWAN SATURATION AMENDMENT
# ==================================================================================================

print()
print("=" * 145)
print("TANGDAOWAN SATURATION AMENDMENT")
print("=" * 145)

print()
print("Status:")

pprint(
    TANGDAOWAN_SATURATION_AMENDMENT[
        "status"
    ],
    sort_dicts=False
)


print()
print("Tangdaowan finding:")

pprint(
    TANGDAOWAN_SATURATION_AMENDMENT[
        "tangdaowan_finding"
    ],
    sort_dicts=False
)


print()
print("Development amendment:")

pprint(
    TANGDAOWAN_SATURATION_AMENDMENT[
        "development_amendment"
    ],
    sort_dicts=False
)


# ==================================================================================================
# 15. FINAL RESTORE GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_FROZEN_LTLC_CONFIG_RESTORED = True


print()
print("=" * 145)

print(
    "Implementation specification       : SHA-VERIFIED"
)

print(
    "Training-prior/rarity manifest     : SHA-VERIFIED"
)

print(
    "Tangdaowan saturation amendment    : SHA-VERIFIED"
)

print(
    "Old shift-selection provenance     : SHA-VERIFIED"
)

print(
    "Tau grid points                    :",
    len(
        TAU_GRID
    )
)

print(
    "Alpha grid points                  :",
    len(
        ALPHA_GRID
    )
)

print(
    "Temperature bounds                 :",
    TEMPERATURE_BOUNDS
)

print(
    "Temperature initial                :",
    TEMPERATURE_INITIAL
)

print(
    "Runtime                            :",
    DEVICE
)

print(
    "Training performed                 :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used                 :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 FROZEN LTLC CONFIGURATION RESTORE: PASS"
)

print(
    "NEXT: CORRECTED BEST-AA VALIDATION-ONLY LA / LTLC ANALYSIS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 19A
# CORRECTED BEST-VAL-AA VALIDATION-ONLY STANDARD-LA / LTLC-SHIFT GRID
#
# PURPOSE
#   - Use the corrected 9 best-validation-AA CE validation-logit files
#   - Apply the EXACT frozen Standard LA and LTLC-Shift formulas
#   - Select parameters using the EXACT frozen validation selection rule
#   - Reproduce old Tangdaowan results as a regression check
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
#
# Temperature fitting is deliberately deferred to Cell 19B.
# ==================================================================================================

from pathlib import Path
import json
import math

import numpy as np
import pandas as pd

from scipy.special import logsumexp
from sklearn.metrics import f1_score


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_CPU_RESUME_GATE_PASS is True
assert NOTEBOOK05_FROZEN_LTLC_CONFIG_RESTORED is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False

assert len(FINAL_BEST_AA_VALIDATION_OUTPUTS) == 9


# ==================================================================================================
# 2. EXACT FROZEN DEFINITIONS
# ==================================================================================================

assert (
    LTLC_IMPLEMENTATION_SPEC["standard_posthoc_LA"]
    ==
    "z_A[k] = z[k] - tau * log(pi[k] + eps)"
)

assert (
    LTLC_IMPLEMENTATION_SPEC["LTLC_shift"]
    ==
    "z_B[k] = z[k] - tau * (1 + alpha * rarity[k]) * log(pi[k] + eps)"
)

assert (
    LTLC_IMPLEMENTATION_SPEC["shift_selection_rule"]
    ==
    [
        "highest_validation_AA",
        "tie_highest_validation_Macro_F1",
        "tie_lowest_validation_NLL",
        "tie_smaller_alpha",
    ]
)


# ==================================================================================================
# 3. VALIDATION OUTPUT LOADER
# ==================================================================================================

def load_bestAA_validation_output(path):

    path = Path(path)

    assert path.exists()

    with np.load(
        path,
        allow_pickle=False
    ) as data:

        required = {
            "logits",
            "labels",
            "predictions",
            "indices",
        }

        assert required.issubset(
            set(data.files)
        ), (
            f"{path.name}: unexpected NPZ schema "
            f"{data.files}"
        )

        logits = np.asarray(
            data["logits"],
            dtype=np.float64
        )

        labels = np.asarray(
            data["labels"],
            dtype=np.int64
        )

        predictions = np.asarray(
            data["predictions"],
            dtype=np.int64
        )

        indices = np.asarray(
            data["indices"],
            dtype=np.int64
        )


    assert logits.ndim == 2

    assert labels.shape == (
        logits.shape[0],
    )

    assert predictions.shape == (
        logits.shape[0],
    )

    assert indices.shape == (
        logits.shape[0],
    )

    assert np.isfinite(
        logits
    ).all()

    assert np.array_equal(
        predictions,
        np.argmax(
            logits,
            axis=1
        )
    )


    return {
        "logits": logits,
        "labels": labels,
        "predictions": predictions,
        "indices": indices,
    }


# ==================================================================================================
# 4. EXACT SHIFT FUNCTIONS
# ==================================================================================================

def standard_la_logits(
    logits,
    prior,
    tau,
    eps
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    prior = np.asarray(
        prior,
        dtype=np.float64
    )

    tau = float(
        tau
    )


    log_prior = np.log(
        prior
        +
        float(eps)
    )


    return (
        logits
        -
        tau
        *
        log_prior[
            None,
            :
        ]
    )


def ltlc_shift_logits(
    logits,
    prior,
    rarity,
    tau,
    alpha,
    eps
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    prior = np.asarray(
        prior,
        dtype=np.float64
    )

    rarity = np.asarray(
        rarity,
        dtype=np.float64
    )

    tau = float(
        tau
    )

    alpha = float(
        alpha
    )


    log_prior = np.log(
        prior
        +
        float(eps)
    )


    adaptive_multiplier = (
        1.0
        +
        alpha
        *
        rarity
    )


    return (
        logits
        -
        tau
        *
        adaptive_multiplier[
            None,
            :
        ]
        *
        log_prior[
            None,
            :
        ]
    )


# ==================================================================================================
# 5. NLL
# ==================================================================================================

def multiclass_nll_from_logits(
    logits,
    labels
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    true_logits = logits[
        np.arange(
            len(labels)
        ),
        labels
    ]


    log_partition = logsumexp(
        logits,
        axis=1
    )


    return float(
        np.mean(
            log_partition
            -
            true_logits
        )
    )


# ==================================================================================================
# 6. CLASSIFICATION METRICS
# ==================================================================================================

def validation_metrics_from_logits(
    logits,
    labels,
    frequency_group
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    labels = np.asarray(
        labels,
        dtype=np.int64
    )


    num_classes = int(
        logits.shape[1]
    )


    predictions = np.argmax(
        logits,
        axis=1
    ).astype(
        np.int64
    )


    # ----------------------------------------------------------------------------------------------
    # OA
    # ----------------------------------------------------------------------------------------------

    oa = float(
        np.mean(
            predictions
            ==
            labels
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Per-class accuracy
    # ----------------------------------------------------------------------------------------------

    per_class_accuracy = []


    for class_index in range(
        num_classes
    ):

        mask = (
            labels
            ==
            class_index
        )


        assert mask.any(), (
            f"Validation split contains no samples "
            f"for class {class_index}."
        )


        per_class_accuracy.append(

            float(
                np.mean(
                    predictions[
                        mask
                    ]
                    ==
                    labels[
                        mask
                    ]
                )
            )
        )


    per_class_accuracy = np.asarray(
        per_class_accuracy,
        dtype=np.float64
    )


    aa = float(
        np.mean(
            per_class_accuracy
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Macro-F1
    # ----------------------------------------------------------------------------------------------

    macro_f1 = float(
        f1_score(
            labels,
            predictions,
            labels=np.arange(
                num_classes
            ),
            average="macro",
            zero_division=0,
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Frequency-group accuracies
    # ----------------------------------------------------------------------------------------------

    frequency_group = np.asarray(
        frequency_group,
        dtype=object
    )


    assert len(
        frequency_group
    ) == num_classes


    group_accuracy = {}


    for group in [
        "Tail",
        "Medium",
        "Head",
    ]:

        class_mask = (
            frequency_group
            ==
            group
        )


        assert class_mask.any()


        group_accuracy[
            group
        ] = float(
            np.mean(
                per_class_accuracy[
                    class_mask
                ]
            )
        )


    worst_class_accuracy = float(
        np.min(
            per_class_accuracy
        )
    )


    nll = multiclass_nll_from_logits(
        logits,
        labels
    )


    return {

        "oa":
            oa,

        "aa":
            aa,

        "macro_f1":
            macro_f1,

        "tail_accuracy":
            group_accuracy[
                "Tail"
            ],

        "medium_accuracy":
            group_accuracy[
                "Medium"
            ],

        "head_accuracy":
            group_accuracy[
                "Head"
            ],

        "worst_class_accuracy":
            worst_class_accuracy,

        "nll":
            nll,

        "predictions":
            predictions,

        "per_class_accuracy":
            per_class_accuracy,
    }


# ==================================================================================================
# 7. EXACT FROZEN SELECTION ORDER
#
# Rule:
#   1. Highest validation AA
#   2. Highest validation Macro-F1
#   3. Lowest validation NLL
#   4. Smaller alpha
#
# Candidate generation order is deterministic:
#   tau ascending, then alpha ascending.
#
# Python's stable sort therefore preserves the original candidate order
# if every declared criterion is exactly tied.
# ==================================================================================================

def select_shift_candidate(
    candidate_records
):

    assert len(
        candidate_records
    ) > 0


    ordered = sorted(

        candidate_records,

        key=lambda row: (

            -float(
                row[
                    "validation_aa"
                ]
            ),

            -float(
                row[
                    "validation_macro_f1"
                ]
            ),

            float(
                row[
                    "validation_nll"
                ]
            ),

            float(
                row[
                    "alpha"
                ]
            ),
        )
    )


    return ordered[
        0
    ].copy()


# ==================================================================================================
# 8. RUN ORDER
# ==================================================================================================

RUN_ORDER = [

    ("Pingan", 42),
    ("Pingan", 123),
    ("Pingan", 3407),

    ("Qingyun", 42),
    ("Qingyun", 123),
    ("Qingyun", 3407),

    ("Tangdaowan", 42),
    ("Tangdaowan", 123),
    ("Tangdaowan", 3407),
]


# ==================================================================================================
# 9. GRID SEARCH
# ==================================================================================================

ALL_SHIFT_GRID_RECORDS = []

CORRECTED_STANDARD_LA_SELECTION = {}

CORRECTED_LTLC_SHIFT_SELECTION = {}


print("=" * 155)
print("NOTEBOOK 05 — CORRECTED BEST-AA VALIDATION STANDARD-LA / LTLC-SHIFT GRID")
print("=" * 155)

print(
    "Standard-LA candidates per seed:",
    len(
        TAU_GRID
    )
)

print(
    "LTLC-Shift candidates per seed:",
    len(
        TAU_GRID
    )
    *
    len(
        ALPHA_GRID
    )
)

print(
    "Official test:",
    "SEALED"
)

print("=" * 155)


for dataset, seed in RUN_ORDER:

    identity = (
        dataset,
        seed
    )


    payload = (
        PRIOR_RARITY_BY_DATASET[
            dataset
        ]
    )


    prior = np.asarray(
        payload[
            "prior"
        ],
        dtype=np.float64
    )


    rarity = np.asarray(
        payload[
            "rarity"
        ],
        dtype=np.float64
    )


    frequency_group = np.asarray(
        payload[
            "frequency_group"
        ],
        dtype=object
    )


    num_classes = int(
        payload[
            "class_count"
        ]
    )


    validation_path = (
        FINAL_BEST_AA_VALIDATION_OUTPUTS[
            identity
        ]
    )


    validation = (
        load_bestAA_validation_output(
            validation_path
        )
    )


    logits = validation[
        "logits"
    ]


    labels = validation[
        "labels"
    ]


    assert logits.shape[
        1
    ] == num_classes


    # ==============================================================================================
    # A. MANDATORY IDENTITY SANITY CHECKS
    # ==============================================================================================

    z_standard_tau0 = standard_la_logits(

        logits,
        prior,
        tau=0.0,
        eps=RARITY_EPSILON
    )


    assert np.array_equal(
        z_standard_tau0,
        logits
    )


    z_ltlc_tau0 = ltlc_shift_logits(

        logits,
        prior,
        rarity,
        tau=0.0,
        alpha=2.0,
        eps=RARITY_EPSILON
    )


    assert np.array_equal(
        z_ltlc_tau0,
        logits
    )


    # ==============================================================================================
    # B. STANDARD POST-HOC LA GRID
    # ==============================================================================================

    standard_candidates = []


    for tau in TAU_GRID:

        adjusted_logits = standard_la_logits(

            logits,
            prior,
            tau=tau,
            eps=RARITY_EPSILON
        )


        metrics = validation_metrics_from_logits(

            adjusted_logits,
            labels,
            frequency_group
        )


        record = {

            "dataset":
                dataset,

            "seed":
                int(
                    seed
                ),

            "method":
                "Standard_LA",

            "tau":
                float(
                    tau
                ),

            "alpha":
                0.0,

            "validation_oa":
                metrics[
                    "oa"
                ],

            "validation_aa":
                metrics[
                    "aa"
                ],

            "validation_macro_f1":
                metrics[
                    "macro_f1"
                ],

            "validation_tail_accuracy":
                metrics[
                    "tail_accuracy"
                ],

            "validation_medium_accuracy":
                metrics[
                    "medium_accuracy"
                ],

            "validation_head_accuracy":
                metrics[
                    "head_accuracy"
                ],

            "validation_worst_class_accuracy":
                metrics[
                    "worst_class_accuracy"
                ],

            "validation_nll":
                metrics[
                    "nll"
                ],
        }


        standard_candidates.append(
            record
        )


        ALL_SHIFT_GRID_RECORDS.append(
            record
        )


    selected_standard = (
        select_shift_candidate(
            standard_candidates
        )
    )


    CORRECTED_STANDARD_LA_SELECTION[
        identity
    ] = selected_standard


    # ==============================================================================================
    # C. LTLC-SHIFT GRID
    # ==============================================================================================

    ltlc_candidates = []


    for tau in TAU_GRID:

        # ------------------------------------------------------------------------------------------
        # Mandatory alpha=0 == Standard LA check
        # ------------------------------------------------------------------------------------------

        standard_reference = standard_la_logits(

            logits,
            prior,
            tau=tau,
            eps=RARITY_EPSILON
        )


        alpha_zero_reference = ltlc_shift_logits(

            logits,
            prior,
            rarity,
            tau=tau,
            alpha=0.0,
            eps=RARITY_EPSILON
        )


        assert np.array_equal(

            standard_reference,

            alpha_zero_reference
        )


        for alpha in ALPHA_GRID:

            adjusted_logits = ltlc_shift_logits(

                logits,
                prior,
                rarity,
                tau=tau,
                alpha=alpha,
                eps=RARITY_EPSILON
            )


            metrics = validation_metrics_from_logits(

                adjusted_logits,
                labels,
                frequency_group
            )


            record = {

                "dataset":
                    dataset,

                "seed":
                    int(
                        seed
                    ),

                "method":
                    "LTLC_Shift",

                "tau":
                    float(
                        tau
                    ),

                "alpha":
                    float(
                        alpha
                    ),

                "validation_oa":
                    metrics[
                        "oa"
                    ],

                "validation_aa":
                    metrics[
                        "aa"
                    ],

                "validation_macro_f1":
                    metrics[
                        "macro_f1"
                    ],

                "validation_tail_accuracy":
                    metrics[
                        "tail_accuracy"
                    ],

                "validation_medium_accuracy":
                    metrics[
                        "medium_accuracy"
                    ],

                "validation_head_accuracy":
                    metrics[
                        "head_accuracy"
                    ],

                "validation_worst_class_accuracy":
                    metrics[
                        "worst_class_accuracy"
                    ],

                "validation_nll":
                    metrics[
                        "nll"
                    ],
            }


            ltlc_candidates.append(
                record
            )


            ALL_SHIFT_GRID_RECORDS.append(
                record
            )


    selected_ltlc = (
        select_shift_candidate(
            ltlc_candidates
        )
    )


    CORRECTED_LTLC_SHIFT_SELECTION[
        identity
    ] = selected_ltlc


    # ==============================================================================================
    # D. COMPACT RESULT
    # ==============================================================================================

    delta_aa = (
        selected_ltlc[
            "validation_aa"
        ]
        -
        selected_standard[
            "validation_aa"
        ]
    )


    delta_tail = (
        selected_ltlc[
            "validation_tail_accuracy"
        ]
        -
        selected_standard[
            "validation_tail_accuracy"
        ]
    )


    delta_worst = (
        selected_ltlc[
            "validation_worst_class_accuracy"
        ]
        -
        selected_standard[
            "validation_worst_class_accuracy"
        ]
    )


    delta_nll = (
        selected_ltlc[
            "validation_nll"
        ]
        -
        selected_standard[
            "validation_nll"
        ]
    )


    print(

        f"{dataset:11s} | "
        f"seed={seed:4d} | "

        f"LA("
        f"tau={selected_standard['tau']:.2f}, "
        f"AA={selected_standard['validation_aa']:.6f}) | "

        f"LTLC("
        f"tau={selected_ltlc['tau']:.2f}, "
        f"alpha={selected_ltlc['alpha']:.2f}, "
        f"AA={selected_ltlc['validation_aa']:.6f}) | "

        f"dAA={delta_aa:+.6f} | "
        f"dTail={delta_tail:+.6f} | "
        f"dWorst={delta_worst:+.6f} | "
        f"dNLL={delta_nll:+.6e}"
    )


# ==================================================================================================
# 10. EXPECTED GRID SIZE
# ==================================================================================================

EXPECTED_GRID_ROWS = (

    9
    *
    (
        len(
            TAU_GRID
        )
        +
        len(
            TAU_GRID
        )
        *
        len(
            ALPHA_GRID
        )
    )
)


assert len(
    ALL_SHIFT_GRID_RECORDS
) == EXPECTED_GRID_ROWS


assert EXPECTED_GRID_ROWS == 648


# ==================================================================================================
# 11. TANGDAOWAN REGRESSION CHECK
#
# Tangdaowan best-AA == original best-OA for all 3 seeds.
# Therefore the corrected computation MUST reproduce the frozen old
# Tangdaowan shift-selection results.
# ==================================================================================================

print()
print("=" * 155)
print("TANGDAOWAN PRE-RECOVERY RESULT REGRESSION CHECK")
print("=" * 155)


OLD_STANDARD = (
    OLD_TANGDAOWAN_SHIFT_SELECTION[
        "standard_la_selection"
    ]
)


OLD_LTLC = (
    OLD_TANGDAOWAN_SHIFT_SELECTION[
        "ltlc_shift_selection"
    ]
)


REGRESSION_METRIC_MAP = {

    "validation_oa":
        "validation_oa",

    "validation_aa":
        "validation_aa",

    "validation_macro_f1":
        "validation_macro_f1",

    "validation_tail_accuracy":
        "validation_tail_accuracy",

    "validation_worst_class_accuracy":
        "validation_worst_class_accuracy",

    "validation_nll":
        "validation_nll",
}


for seed in [
    42,
    123,
    3407,
]:

    identity = (
        "Tangdaowan",
        seed
    )


    new_standard = (
        CORRECTED_STANDARD_LA_SELECTION[
            identity
        ]
    )


    new_ltlc = (
        CORRECTED_LTLC_SHIFT_SELECTION[
            identity
        ]
    )


    old_standard = (
        OLD_STANDARD[
            str(
                seed
            )
        ]
    )


    old_ltlc = (
        OLD_LTLC[
            str(
                seed
            )
        ]
    )


    # Parameter selections must reproduce exactly.
    assert float(
        new_standard[
            "tau"
        ]
    ) == float(
        old_standard[
            "tau"
        ]
    )


    assert float(
        new_ltlc[
            "tau"
        ]
    ) == float(
        old_ltlc[
            "tau"
        ]
    )


    assert float(
        new_ltlc[
            "alpha"
        ]
    ) == float(
        old_ltlc[
            "alpha"
        ]
    )


    # Classification metrics: extremely tight.
    for new_key, old_key in (
        REGRESSION_METRIC_MAP.items()
    ):

        tolerance = (

            1e-10

            if (
                new_key
                ==
                "validation_nll"
            )

            else
            1e-12
        )


        difference = abs(

            float(
                new_standard[
                    new_key
                ]
            )
            -
            float(
                old_standard[
                    old_key
                ]
            )
        )


        assert difference <= tolerance, (
            f"Tangdaowan seed {seed} "
            f"Standard-LA regression mismatch: "
            f"{new_key} | difference={difference}"
        )


        difference = abs(

            float(
                new_ltlc[
                    new_key
                ]
            )
            -
            float(
                old_ltlc[
                    old_key
                ]
            )
        )


        assert difference <= tolerance, (
            f"Tangdaowan seed {seed} "
            f"LTLC regression mismatch: "
            f"{new_key} | difference={difference}"
        )


    print(
        f"Tangdaowan seed {seed:4d} | "
        "parameters + AA/F1/OA/Tail/Worst/NLL | PASS"
    )


# ==================================================================================================
# 12. BUILD SELECTION SUMMARY
# ==================================================================================================

SELECTION_SUMMARY_RECORDS = []


for identity in RUN_ORDER:

    dataset, seed = identity


    standard = (
        CORRECTED_STANDARD_LA_SELECTION[
            identity
        ]
    )


    ltlc = (
        CORRECTED_LTLC_SHIFT_SELECTION[
            identity
        ]
    )


    SELECTION_SUMMARY_RECORDS.append({

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "standard_la_tau":
            float(
                standard[
                    "tau"
                ]
            ),

        "standard_la_aa":
            float(
                standard[
                    "validation_aa"
                ]
            ),

        "standard_la_macro_f1":
            float(
                standard[
                    "validation_macro_f1"
                ]
            ),

        "standard_la_tail_accuracy":
            float(
                standard[
                    "validation_tail_accuracy"
                ]
            ),

        "standard_la_worst_class_accuracy":
            float(
                standard[
                    "validation_worst_class_accuracy"
                ]
            ),

        "standard_la_nll":
            float(
                standard[
                    "validation_nll"
                ]
            ),

        "ltlc_tau":
            float(
                ltlc[
                    "tau"
                ]
            ),

        "ltlc_alpha":
            float(
                ltlc[
                    "alpha"
                ]
            ),

        "ltlc_aa":
            float(
                ltlc[
                    "validation_aa"
                ]
            ),

        "ltlc_macro_f1":
            float(
                ltlc[
                    "validation_macro_f1"
                ]
            ),

        "ltlc_tail_accuracy":
            float(
                ltlc[
                    "validation_tail_accuracy"
                ]
            ),

        "ltlc_worst_class_accuracy":
            float(
                ltlc[
                    "validation_worst_class_accuracy"
                ]
            ),

        "ltlc_nll":
            float(
                ltlc[
                    "validation_nll"
                ]
            ),

        "delta_aa":
            float(
                ltlc[
                    "validation_aa"
                ]
                -
                standard[
                    "validation_aa"
                ]
            ),

        "delta_macro_f1":
            float(
                ltlc[
                    "validation_macro_f1"
                ]
                -
                standard[
                    "validation_macro_f1"
                ]
            ),

        "delta_tail_accuracy":
            float(
                ltlc[
                    "validation_tail_accuracy"
                ]
                -
                standard[
                    "validation_tail_accuracy"
                ]
            ),

        "delta_worst_class_accuracy":
            float(
                ltlc[
                    "validation_worst_class_accuracy"
                ]
                -
                standard[
                    "validation_worst_class_accuracy"
                ]
            ),

        # Negative is an NLL improvement.
        "delta_nll":
            float(
                ltlc[
                    "validation_nll"
                ]
                -
                standard[
                    "validation_nll"
                ]
            ),
    })


CORRECTED_SHIFT_SELECTION_SUMMARY_DF = pd.DataFrame(
    SELECTION_SUMMARY_RECORDS
)


CORRECTED_SHIFT_GRID_DF = pd.DataFrame(
    ALL_SHIFT_GRID_RECORDS
)


# ==================================================================================================
# 13. SAVE DETERMINISTIC CORRECTED DEVELOPMENT ARTIFACTS
# ==================================================================================================

CORRECTED_SHIFT_GRID_PATH = (
    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_validation_shift_grid.csv"
)


CORRECTED_SHIFT_SELECTION_PATH = (
    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_validation_shift_selection_manifest.json"
)


# --------------------------------------------------------------------------------------------------
# CSV
# --------------------------------------------------------------------------------------------------

csv_text = CORRECTED_SHIFT_GRID_DF.to_csv(
    index=False,
    float_format="%.17g"
)


if CORRECTED_SHIFT_GRID_PATH.exists():

    existing_text = (
        CORRECTED_SHIFT_GRID_PATH.read_text(
            encoding="utf-8"
        )
    )

    assert existing_text == csv_text, (
        "Existing corrected shift grid differs. STOP."
    )

else:

    CORRECTED_SHIFT_GRID_PATH.write_text(
        csv_text,
        encoding="utf-8"
    )


CORRECTED_SHIFT_GRID_SHA = sha256_file(
    CORRECTED_SHIFT_GRID_PATH
)


# --------------------------------------------------------------------------------------------------
# JSON
# --------------------------------------------------------------------------------------------------

def selection_dict_for_json(
    selection_map
):

    output = {}


    for (
        dataset,
        seed
    ), record in selection_map.items():

        output.setdefault(
            dataset,
            {}
        )


        output[
            dataset
        ][
            str(
                seed
            )
        ] = {

            key:
                (
                    int(value)

                    if isinstance(
                        value,
                        np.integer
                    )

                    else float(value)

                    if isinstance(
                        value,
                        (
                            np.floating,
                            float,
                        )
                    )

                    else value
                )

            for key, value
            in record.items()

            if key not in {
                "dataset",
                "seed",
                "method",
            }
        }


    return output


selection_payload = {

    "project":
        "LTLC",

    "notebook":
        "05",

    "purpose":
        (
            "Corrected validation-only Standard-LA and "
            "LTLC-Shift development analysis using final "
            "best-validation-AA CE checkpoints."
        ),

    "source_checkpoint_purpose":
        "best_validation_AA",

    "source_validation_manifest_sha256":
        ACTUAL_RESUME_HASHES[
            "bestAA_validation_manifest_json"
        ],

    "implementation_spec_sha256":
        ACTUAL_FROZEN_HASHES[
            "implementation_spec"
        ],

    "training_prior_rarity_manifest_sha256":
        ACTUAL_FROZEN_HASHES[
            "prior_rarity"
        ],

    "tau_grid":
        TAU_GRID.tolist(),

    "alpha_grid":
        ALPHA_GRID.tolist(),

    "selection_rule":
        LTLC_IMPLEMENTATION_SPEC[
            "shift_selection_rule"
        ],

    "standard_la_selection":
        selection_dict_for_json(
            CORRECTED_STANDARD_LA_SELECTION
        ),

    "ltlc_shift_selection":
        selection_dict_for_json(
            CORRECTED_LTLC_SHIFT_SELECTION
        ),

    "grid_csv":
        str(
            CORRECTED_SHIFT_GRID_PATH.relative_to(
                ROOT
            )
        ),

    "grid_csv_sha256":
        CORRECTED_SHIFT_GRID_SHA,

    "tangdaowan_old_result_regression_pass":
        True,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,

    "status":
        "CORRECTED_BEST_AA_SHIFT_SELECTION_COMPLETE",
}


json_text = (
    json.dumps(
        selection_payload,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )
    +
    "\n"
)


if CORRECTED_SHIFT_SELECTION_PATH.exists():

    existing_text = (
        CORRECTED_SHIFT_SELECTION_PATH.read_text(
            encoding="utf-8"
        )
    )

    assert existing_text == json_text, (
        "Existing corrected selection manifest differs. STOP."
    )

else:

    CORRECTED_SHIFT_SELECTION_PATH.write_text(
        json_text,
        encoding="utf-8"
    )


CORRECTED_SHIFT_SELECTION_SHA = sha256_file(
    CORRECTED_SHIFT_SELECTION_PATH
)


# ==================================================================================================
# 14. DATASET-LEVEL DEVELOPMENT SUMMARY
# ==================================================================================================

print()
print("=" * 155)
print("CORRECTED BEST-AA DEVELOPMENT SUMMARY")
print("=" * 155)


for dataset in [
    "Pingan",
    "Qingyun",
    "Tangdaowan",
]:

    subset = (
        CORRECTED_SHIFT_SELECTION_SUMMARY_DF[
            CORRECTED_SHIFT_SELECTION_SUMMARY_DF[
                "dataset"
            ]
            ==
            dataset
        ]
    )


    assert len(
        subset
    ) == 3


    print()

    print(
        dataset
    )


    print(
        "  Mean delta AA      :",
        f"{subset['delta_aa'].mean():+.9f}"
    )

    print(
        "  Mean delta F1      :",
        f"{subset['delta_macro_f1'].mean():+.9f}"
    )

    print(
        "  Mean delta Tail Acc:",
        f"{subset['delta_tail_accuracy'].mean():+.9f}"
    )

    print(
        "  Mean delta Worst   :",
        f"{subset['delta_worst_class_accuracy'].mean():+.9f}"
    )

    print(
        "  Mean delta NLL     :",
        f"{subset['delta_nll'].mean():+.9e}"
    )

    print(
        "  Nonzero alpha seeds:",
        int(
            (
                subset[
                    "ltlc_alpha"
                ]
                >
                0.0
            ).sum()
        ),
        "/ 3"
    )


# ==================================================================================================
# 15. FINAL SAFETY / COMPLETION GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_CORRECTED_SHIFT_GRID_PASS = True


print()
print("=" * 155)

print(
    "Corrected Standard-LA selections : 9/9 PASS"
)

print(
    "Corrected LTLC-Shift selections  : 9/9 PASS"
)

print(
    "Total corrected grid rows        :",
    len(
        CORRECTED_SHIFT_GRID_DF
    )
)

print(
    "Tangdaowan old-result regression : 3/3 PASS"
)

print()

print(
    "Corrected grid artifact:"
)

print(
    CORRECTED_SHIFT_GRID_PATH
)

print(
    "SHA-256:"
)

print(
    CORRECTED_SHIFT_GRID_SHA
)

print()

print(
    "Corrected selection manifest:"
)

print(
    CORRECTED_SHIFT_SELECTION_PATH
)

print(
    "SHA-256:"
)

print(
    CORRECTED_SHIFT_SELECTION_SHA
)

print()

print(
    "Runtime                           :",
    DEVICE
)

print(
    "Training performed                :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official-test dataset created     :",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created  :",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used                :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 CORRECTED BEST-AA SHIFT GRID: PASS"
)

print(
    "NEXT: FIT FROZEN LTLC TEMPERATURE MODEL ON VALIDATION ONLY"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 155)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 19B-S
# EXACT SOURCE AUDIT FOR TEMPERATURE / CALIBRATION IMPLEMENTATION
#
# PURPOSE
#   Recover the exact pre-recovery Notebook 05 source code for:
#       - ECE
#       - Tail-ECE
#       - Brier
#       - NLL
#       - softmax/probabilities
#       - Global temperature scaling
#       - LTLC samplewise temperature
#       - L-BFGS-B objective / fitting
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

from pathlib import Path
import json
import re


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_TEMPERATURE_ARTIFACT_INSPECTION_PASS is True
assert NOTEBOOK05_CORRECTED_SHIFT_GRID_PASS is True
assert NOTEBOOK05_FROZEN_LTLC_CONFIG_RESTORED is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


# ==================================================================================================
# 2. LOCATE NOTEBOOK 05 SOURCE CANDIDATES
# ==================================================================================================

notebook_candidates = []


for path in ROOT.rglob("*.ipynb"):

    path_lower = str(path).lower()


    # Ignore transient notebook checkpoints.
    if ".ipynb_checkpoints" in path_lower:
        continue


    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:

            notebook = json.load(f)

    except Exception:
        continue


    code_text = "\n".join(

        "".join(
            cell.get(
                "source",
                []
            )
        )

        for cell in notebook.get(
            "cells",
            []
        )

        if cell.get(
            "cell_type"
        ) == "code"
    )


    score = 0


    score_terms = {

        "LTLC_temperature":
            20,

        "Tail_ECE":
            20,

        "tail_ece":
            20,

        "L-BFGS-B":
            15,

        "brier":
            10,

        "expected_calibration":
            10,

        "temperature":
            5,

        "minimize(":
            5,
    }


    for term, weight in score_terms.items():

        if term.lower() in code_text.lower():

            score += weight


    if score > 0:

        notebook_candidates.append({

            "path":
                path,

            "score":
                score,

            "notebook":
                notebook,
        })


assert len(
    notebook_candidates
) > 0, (
    "No Notebook 05 calibration-source candidate found."
)


notebook_candidates = sorted(

    notebook_candidates,

    key=lambda item: (
        -item[
            "score"
        ],
        str(
            item[
                "path"
            ]
        )
    )
)


print("=" * 145)
print("NOTEBOOK 05 — CALIBRATION SOURCE CANDIDATES")
print("=" * 145)


for item in notebook_candidates[:10]:

    print(

        f"score={item['score']:3d} | "
        f"{item['path'].relative_to(ROOT)}"
    )


# ==================================================================================================
# 3. SELECT HIGHEST-SCORING SOURCE NOTEBOOK
# ==================================================================================================

selected = notebook_candidates[
    0
]


SOURCE_NOTEBOOK_PATH = (
    selected[
        "path"
    ]
)


SOURCE_NOTEBOOK = (
    selected[
        "notebook"
    ]
)


print()
print(
    "SELECTED SOURCE NOTEBOOK:"
)

print(
    SOURCE_NOTEBOOK_PATH
)

print(
    "Selection score:",
    selected[
        "score"
    ]
)


# ==================================================================================================
# 4. FIND RELEVANT CODE CELLS
# ==================================================================================================

SEARCH_GROUPS = {

    "ECE":

        [
            "expected_calibration",
            "ece",
        ],


    "TAIL_ECE":

        [
            "tail_ece",
            "Tail_ECE",
        ],


    "BRIER":

        [
            "brier",
        ],


    "NLL":

        [
            "nll",
            "logsumexp",
        ],


    "TEMPERATURE":

        [
            "temperature",
            "exp(a",
            "rarity",
        ],


    "OPTIMIZER":

        [
            "L-BFGS-B",
            "minimize(",
        ],
}


MATCHED_SOURCE_CELLS = []


for cell_index, cell in enumerate(

    SOURCE_NOTEBOOK.get(
        "cells",
        []
    )
):

    if (
        cell.get(
            "cell_type"
        )
        !=
        "code"
    ):

        continue


    source = "".join(
        cell.get(
            "source",
            []
        )
    )


    source_lower = (
        source.lower()
    )


    matched_groups = []


    for group, terms in SEARCH_GROUPS.items():

        if any(

            term.lower()
            in
            source_lower

            for term in terms
        ):

            matched_groups.append(
                group
            )


    if len(
        matched_groups
    ) > 0:

        MATCHED_SOURCE_CELLS.append({

            "cell_index":
                int(
                    cell_index
                ),

            "matched_groups":
                matched_groups,

            "source":
                source,
        })


assert len(
    MATCHED_SOURCE_CELLS
) > 0


# ==================================================================================================
# 5. PRINT RELEVANT SOURCE CELLS
# ==================================================================================================

print()
print("=" * 145)
print("EXACT MATCHED CALIBRATION SOURCE CELLS")
print("=" * 145)


for record in MATCHED_SOURCE_CELLS:

    print()
    print("#" * 145)

    print(
        "SOURCE CELL INDEX:",
        record[
            "cell_index"
        ]
    )

    print(
        "MATCHED GROUPS:",
        record[
            "matched_groups"
        ]
    )

    print("#" * 145)


    source = (
        record[
            "source"
        ]
    )


    # Print full cell, but cap pathological giant cells.
    lines = source.splitlines()


    if len(
        lines
    ) <= 350:

        print(
            source
        )

    else:

        print(
            "\n".join(
                lines[
                    :350
                ]
            )
        )

        print()
        print(
            f"... SOURCE TRUNCATED AFTER 350 / "
            f"{len(lines)} LINES ..."
        )


# ==================================================================================================
# 6. TARGETED FUNCTION EXTRACTION
#
# Search function definitions specifically so we can later reconstruct
# exactly rather than relying on whole-cell context.
# ==================================================================================================

FULL_SOURCE_TEXT = "\n\n".join(

    record[
        "source"
    ]

    for record in MATCHED_SOURCE_CELLS
)


function_names = []


for match in re.finditer(

    r"(?m)^def\s+([A-Za-z_][A-Za-z0-9_]*)\s*\(",

    FULL_SOURCE_TEXT
):

    name = match.group(
        1
    )


    relevant_tokens = [

        "ece",
        "calib",
        "brier",
        "nll",
        "temperature",
        "softmax",
        "metric",
        "fit",
    ]


    if any(

        token in name.lower()

        for token in relevant_tokens
    ):

        function_names.append(
            name
        )


function_names = list(
    dict.fromkeys(
        function_names
    )
)


print()
print("=" * 145)
print("RELEVANT FUNCTION NAMES DISCOVERED")
print("=" * 145)

print(
    function_names
)


# ==================================================================================================
# 7. SAVE SOURCE AUDIT
# ==================================================================================================

CALIBRATION_SOURCE_AUDIT_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_calibration_source_audit.json"
)


source_cell_records = []


for record in MATCHED_SOURCE_CELLS:

    source_cell_records.append({

        "cell_index":
            record[
                "cell_index"
            ],

        "matched_groups":
            record[
                "matched_groups"
            ],

        "source_sha256":
            __import__(
                "hashlib"
            ).sha256(

                record[
                    "source"
                ].encode(
                    "utf-8"
                )

            ).hexdigest(),
    })


audit_payload = {

    "project":
        "LTLC",

    "notebook":
        "05",

    "purpose":
        (
            "Recover exact frozen calibration and temperature "
            "implementation source before corrected best-AA fitting."
        ),

    "selected_source_notebook":
        str(
            SOURCE_NOTEBOOK_PATH.relative_to(
                ROOT
            )
        ),

    "selection_score":
        int(
            selected[
                "score"
            ]
        ),

    "matched_source_cells":
        source_cell_records,

    "relevant_function_names":
        function_names,

    "implementation_spec_sha256":
        ACTUAL_FROZEN_HASHES[
            "implementation_spec"
        ],

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


audit_text = (
    json.dumps(
        audit_payload,
        indent=2,
        sort_keys=True
    )
    +
    "\n"
)


if CALIBRATION_SOURCE_AUDIT_PATH.exists():

    existing_text = (
        CALIBRATION_SOURCE_AUDIT_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing_text == audit_text, (
        "Existing calibration source audit differs. STOP."
    )


else:

    CALIBRATION_SOURCE_AUDIT_PATH.write_text(
        audit_text,
        encoding="utf-8"
    )


CALIBRATION_SOURCE_AUDIT_SHA = (
    sha256_file(
        CALIBRATION_SOURCE_AUDIT_PATH
    )
)


# ==================================================================================================
# 8. SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_EXACT_CALIBRATION_SOURCE_AUDIT_PASS = True


print()
print("=" * 145)

print(
    "Selected source notebook         :",
    SOURCE_NOTEBOOK_PATH.relative_to(
        ROOT
    )
)

print(
    "Matched calibration source cells :",
    len(
        MATCHED_SOURCE_CELLS
    )
)

print(
    "Relevant functions discovered    :",
    len(
        function_names
    )
)

print()

print(
    "Source-audit artifact:"
)

print(
    CALIBRATION_SOURCE_AUDIT_PATH
)

print(
    "SHA-256:"
)

print(
    CALIBRATION_SOURCE_AUDIT_SHA
)

print()

print(
    "Training performed               :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used               :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 EXACT CALIBRATION SOURCE AUDIT: PASS"
)

print(
    "NEXT: EXACT CORRECTED BEST-AA TEMPERATURE FITTING"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 19A-V
# EXACT FROZEN SHIFT-SELECTION RULE VERIFICATION
#
# Uses the already-computed corrected best-AA grid.
#
# NO GRID RECOMPUTATION
# NO TRAINING
# NO OFFICIAL TEST
# CPU ONLY
# ==================================================================================================

import numpy as np
import pandas as pd


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_CORRECTED_SHIFT_GRID_PASS is True
assert NOTEBOOK05_EXACT_CALIBRATION_SOURCE_AUDIT_PASS is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


GRID_TOL = 1e-12


# ==================================================================================================
# 2. EXACT ORIGINAL FROZEN SELECTOR
# ==================================================================================================

def select_by_exact_locked_shift_rule(
    candidate_df,
    label
):

    df = candidate_df.copy()


    assert len(df) > 0


    # ----------------------------------------------------------------------------------------------
    # 1. Highest validation AA
    # ----------------------------------------------------------------------------------------------

    best_aa = float(
        df[
            "validation_aa"
        ].max()
    )


    df = df[
        np.isclose(
            df[
                "validation_aa"
            ].to_numpy(
                dtype=np.float64
            ),
            best_aa,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 2. Highest validation Macro-F1
    # ----------------------------------------------------------------------------------------------

    best_f1 = float(
        df[
            "validation_macro_f1"
        ].max()
    )


    df = df[
        np.isclose(
            df[
                "validation_macro_f1"
            ].to_numpy(
                dtype=np.float64
            ),
            best_f1,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 3. Lowest validation NLL
    # ----------------------------------------------------------------------------------------------

    best_nll = float(
        df[
            "validation_nll"
        ].min()
    )


    df = df[
        np.isclose(
            df[
                "validation_nll"
            ].to_numpy(
                dtype=np.float64
            ),
            best_nll,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 4. Smaller alpha
    # ----------------------------------------------------------------------------------------------

    best_alpha = float(
        df[
            "alpha"
        ].min()
    )


    df = df[
        np.isclose(
            df[
                "alpha"
            ].to_numpy(
                dtype=np.float64
            ),
            best_alpha,
            rtol=0.0,
            atol=GRID_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # NO UNDECLARED FINAL TIE-BREAK
    # ----------------------------------------------------------------------------------------------

    if len(df) != 1:

        print()
        print("=" * 120)
        print("UNRESOLVED LOCKED-PROTOCOL TIE")
        print("=" * 120)

        print(label)

        print()

        print(
            df[
                [
                    "tau",
                    "alpha",
                    "validation_aa",
                    "validation_macro_f1",
                    "validation_nll",
                ]
            ].to_string(
                index=False
            )
        )

        print("=" * 120)


        raise RuntimeError(
            f"{label}: exact frozen selection rule "
            "did not produce one unique candidate."
        )


    return (
        df.iloc[
            0
        ]
        .to_dict()
    )


# ==================================================================================================
# 3. VERIFY ALL 9 STANDARD-LA + ALL 9 LTLC-SHIFT SELECTIONS
# ==================================================================================================

EXACT_STANDARD_LA_SELECTION = {}
EXACT_LTLC_SHIFT_SELECTION = {}


print("=" * 145)
print("NOTEBOOK 05 — EXACT FROZEN SHIFT-SELECTION VERIFICATION")
print("=" * 145)


for dataset, seed in RUN_ORDER:

    identity = (
        dataset,
        seed
    )


    run_df = (
        CORRECTED_SHIFT_GRID_DF[
            (
                CORRECTED_SHIFT_GRID_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                CORRECTED_SHIFT_GRID_DF[
                    "seed"
                ].astype(
                    int
                )
                ==
                int(
                    seed
                )
            )
        ]
        .copy()
    )


    # ----------------------------------------------------------------------------------------------
    # Standard LA
    # ----------------------------------------------------------------------------------------------

    standard_df = (
        run_df[
            run_df[
                "method"
            ]
            ==
            "Standard_LA"
        ]
        .copy()
    )


    assert len(
        standard_df
    ) == len(
        TAU_GRID
    )


    exact_standard = (
        select_by_exact_locked_shift_rule(
            standard_df,
            f"{dataset} seed {seed} Standard-LA"
        )
    )


    # ----------------------------------------------------------------------------------------------
    # LTLC Shift
    # ----------------------------------------------------------------------------------------------

    ltlc_df = (
        run_df[
            run_df[
                "method"
            ]
            ==
            "LTLC_Shift"
        ]
        .copy()
    )


    assert len(
        ltlc_df
    ) == (
        len(
            TAU_GRID
        )
        *
        len(
            ALPHA_GRID
        )
    )


    exact_ltlc = (
        select_by_exact_locked_shift_rule(
            ltlc_df,
            f"{dataset} seed {seed} LTLC-Shift"
        )
    )


    EXACT_STANDARD_LA_SELECTION[
        identity
    ] = exact_standard


    EXACT_LTLC_SHIFT_SELECTION[
        identity
    ] = exact_ltlc


    # ----------------------------------------------------------------------------------------------
    # Must match Cell 19A selected parameters
    # ----------------------------------------------------------------------------------------------

    old_standard = (
        CORRECTED_STANDARD_LA_SELECTION[
            identity
        ]
    )


    old_ltlc = (
        CORRECTED_LTLC_SHIFT_SELECTION[
            identity
        ]
    )


    assert np.isclose(
        float(
            exact_standard[
                "tau"
            ]
        ),
        float(
            old_standard[
                "tau"
            ]
        ),
        rtol=0.0,
        atol=1e-15
    )


    assert np.isclose(
        float(
            exact_ltlc[
                "tau"
            ]
        ),
        float(
            old_ltlc[
                "tau"
            ]
        ),
        rtol=0.0,
        atol=1e-15
    )


    assert np.isclose(
        float(
            exact_ltlc[
                "alpha"
            ]
        ),
        float(
            old_ltlc[
                "alpha"
            ]
        ),
        rtol=0.0,
        atol=1e-15
    )


    print(

        f"{dataset:11s} | "
        f"seed={seed:4d} | "

        f"LA tau="
        f"{float(exact_standard['tau']):.2f} | "

        f"LTLC tau="
        f"{float(exact_ltlc['tau']):.2f}, "
        f"alpha="
        f"{float(exact_ltlc['alpha']):.2f} | "

        "EXACT RULE UNIQUE | PASS"
    )


# ==================================================================================================
# 4. REPLACE IN-MEMORY SELECTION MAPS WITH EXACTLY VERIFIED MAPS
# ==================================================================================================

CORRECTED_STANDARD_LA_SELECTION = (
    EXACT_STANDARD_LA_SELECTION
)


CORRECTED_LTLC_SHIFT_SELECTION = (
    EXACT_LTLC_SHIFT_SELECTION
)


# ==================================================================================================
# 5. SCIENTIFIC CHECK
# ==================================================================================================

nonzero_alpha = [

    identity

    for identity, record
    in CORRECTED_LTLC_SHIFT_SELECTION.items()

    if float(
        record[
            "alpha"
        ]
    ) > 0.0
]


assert len(
    nonzero_alpha
) == 0


# ==================================================================================================
# 6. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False


NOTEBOOK05_EXACT_SHIFT_SELECTION_VERIFIED = True


print()
print("=" * 145)

print(
    "Standard-LA exact selections : 9/9 UNIQUE + PASS"
)

print(
    "LTLC-Shift exact selections  : 9/9 UNIQUE + PASS"
)

print(
    "Cell 19A selections unchanged: YES"
)

print(
    "Nonzero-alpha selections     :",
    len(
        nonzero_alpha
    ),
    "/ 9"
)

print()

print(
    "Training performed           :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used           :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 EXACT SHIFT-SELECTION RULE AUDIT: PASS"
)

print(
    "READY FOR EXACT TEMPERATURE SOURCE EXTRACTION"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 145)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 19B-E
# TARGETED EXTRACTION OF EXACT FROZEN CALIBRATION FUNCTIONS
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import ast
import json
from pathlib import Path


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_EXACT_SHIFT_SELECTION_VERIFIED is True
assert NOTEBOOK05_EXACT_CALIBRATION_SOURCE_AUDIT_PASS is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False


# ==================================================================================================
# 2. ORIGINAL NOTEBOOK 05 SOURCE
# ==================================================================================================

SOURCE_NOTEBOOK_PATH = (
    ROOT
    /
    "notebooks"
    /
    "05_Posthoc_Calibration_LTLC.ipynb"
)


assert SOURCE_NOTEBOOK_PATH.exists()


with open(
    SOURCE_NOTEBOOK_PATH,
    "r",
    encoding="utf-8"
) as f:

    source_notebook = json.load(f)


# ==================================================================================================
# 3. FUNCTIONS WE NEED EXACTLY
# ==================================================================================================

TARGET_FUNCTIONS = [

    "stable_softmax",

    "multiclass_nll_from_logits",

    "multiclass_brier_score",

    "expected_calibration_error",

    "apply_global_temperature",

    "apply_ltlc_temperature",

    "tail_ece",

    "fit_global_temperature",

    "fit_ltlc_temperature",

    "evaluate_logits",
]


# ==================================================================================================
# 4. EXTRACT FUNCTION SOURCE USING AST
# ==================================================================================================

EXACT_FUNCTION_SOURCE = {}

FUNCTION_SOURCE_CELL = {}


for cell_index, cell in enumerate(
    source_notebook.get(
        "cells",
        []
    )
):

    if cell.get(
        "cell_type"
    ) != "code":

        continue


    source = "".join(
        cell.get(
            "source",
            []
        )
    )


    try:

        tree = ast.parse(
            source
        )

    except SyntaxError:

        continue


    source_lines = source.splitlines()


    for node in tree.body:

        if not isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef
            )
        ):

            continue


        if node.name not in TARGET_FUNCTIONS:

            continue


        assert hasattr(
            node,
            "end_lineno"
        )


        function_text = "\n".join(

            source_lines[
                node.lineno - 1:
                node.end_lineno
            ]
        )


        # The same function may appear in later recovery/inspection cells.
        # Keep the FIRST definition from the original notebook execution order.
        if node.name not in EXACT_FUNCTION_SOURCE:

            EXACT_FUNCTION_SOURCE[
                node.name
            ] = function_text


            FUNCTION_SOURCE_CELL[
                node.name
            ] = int(
                cell_index
            )


# ==================================================================================================
# 5. REQUIRE ALL TARGET FUNCTIONS
# ==================================================================================================

missing = [

    name

    for name in TARGET_FUNCTIONS

    if name not in EXACT_FUNCTION_SOURCE
]


assert len(
    missing
) == 0, (
    f"Missing exact source functions: {missing}"
)


# ==================================================================================================
# 6. PRINT ONLY THE EXACT TARGET FUNCTIONS
# ==================================================================================================

print("=" * 135)
print("NOTEBOOK 05 — EXACT FROZEN CALIBRATION FUNCTION EXTRACTION")
print("=" * 135)


for name in TARGET_FUNCTIONS:

    print()
    print("#" * 135)

    print(
        f"FUNCTION: {name}"
    )

    print(
        f"SOURCE CELL INDEX: "
        f"{FUNCTION_SOURCE_CELL[name]}"
    )

    print("#" * 135)

    print(
        EXACT_FUNCTION_SOURCE[
            name
        ]
    )


# ==================================================================================================
# 7. SOURCE HASHES FOR EACH FUNCTION
# ==================================================================================================

import hashlib


EXACT_FUNCTION_SHA256 = {}


print()
print("=" * 135)
print("FUNCTION SOURCE SHA-256")
print("=" * 135)


for name in TARGET_FUNCTIONS:

    sha = hashlib.sha256(

        EXACT_FUNCTION_SOURCE[
            name
        ].encode(
            "utf-8"
        )

    ).hexdigest()


    EXACT_FUNCTION_SHA256[
        name
    ] = sha


    print(
        f"{name:32s} | "
        f"cell={FUNCTION_SOURCE_CELL[name]:3d} | "
        f"{sha}"
    )


# ==================================================================================================
# 8. SAVE TARGETED SOURCE AUDIT
# ==================================================================================================

TARGETED_FUNCTION_AUDIT_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_exact_calibration_functions.json"
)


audit_payload = {

    "project":
        "LTLC",

    "notebook":
        "05",

    "purpose":
        (
            "Exact frozen calibration-function extraction "
            "before corrected best-AA temperature fitting."
        ),

    "source_notebook":
        str(
            SOURCE_NOTEBOOK_PATH.relative_to(
                ROOT
            )
        ),

    "functions": {

        name: {

            "source_cell_index":
                FUNCTION_SOURCE_CELL[
                    name
                ],

            "source_sha256":
                EXACT_FUNCTION_SHA256[
                    name
                ],

        }

        for name in TARGET_FUNCTIONS
    },

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


audit_text = (
    json.dumps(
        audit_payload,
        indent=2,
        sort_keys=True
    )
    +
    "\n"
)


if TARGETED_FUNCTION_AUDIT_PATH.exists():

    existing_text = (
        TARGETED_FUNCTION_AUDIT_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing_text == audit_text, (
        "Existing targeted function audit differs. STOP."
    )


else:

    TARGETED_FUNCTION_AUDIT_PATH.write_text(
        audit_text,
        encoding="utf-8"
    )


TARGETED_FUNCTION_AUDIT_SHA = (
    sha256_file(
        TARGETED_FUNCTION_AUDIT_PATH
    )
)


# ==================================================================================================
# 9. FINAL GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_EXACT_CALIBRATION_FUNCTIONS_EXTRACTED = True


print()
print("=" * 135)

print(
    "Exact calibration functions found:",
    len(
        EXACT_FUNCTION_SOURCE
    ),
    "/",
    len(
        TARGET_FUNCTIONS
    ),
    "PASS"
)

print(
    "Source notebook                 :",
    SOURCE_NOTEBOOK_PATH.relative_to(
        ROOT
    )
)

print(
    "Targeted function audit SHA-256 :",
    TARGETED_FUNCTION_AUDIT_SHA
)

print(
    "Training performed              :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used              :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 EXACT CALIBRATION FUNCTIONS: PASS"
)

print(
    "READY FOR CORRECTED BEST-AA TEMPERATURE FITTING"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 135)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 19B
# CORRECTED BEST-VAL-AA VALIDATION-ONLY TEMPERATURE FITTING
#
# METHODS PER SEED:
#   1. CE
#   2. Standard_LA
#   3. Global_TS
#   4. LA_Global_TS
#   5. LTLC_Shift
#   6. LTLC_Shift_Global_TS
#   7. LTLC_Full
#
# TEMPERATURE FITS PER SEED:
#   - Global_TS
#   - LA_Global_TS
#   - LTLC_Shift_Global_TS
#   - LTLC_Full
#
# EXPECTED:
#   9 runs × 7 result rows     = 63
#   9 runs × 4 parameter rows  = 36
#
# CPU ONLY
# VALIDATION ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import ast
import json
import math
from pathlib import Path

import numpy as np
import pandas as pd

from scipy.optimize import minimize


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_EXACT_SHIFT_SELECTION_VERIFIED is True
assert NOTEBOOK05_EXACT_CALIBRATION_FUNCTIONS_EXTRACTED is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


# ==================================================================================================
# 2. EXACT ORIGINAL TEMPERATURE CONSTANTS
# ==================================================================================================

BOUND_LOW = -2.0
BOUND_HIGH = 2.0

BOUND_TOL = 1e-6
SIGNAL_TOL = 1e-12


assert TEMPERATURE_BOUNDS == {
    "a": (-2.0, 2.0),
    "b": (-2.0, 2.0),
}


assert TEMPERATURE_INITIAL == {
    "a": 0.0,
    "b": 0.0,
}


# ==================================================================================================
# 3. RESTORE REQUIRED EXACT ORIGINAL ENGINE DEPENDENCIES
#
# The previous cell extracted the main calibration functions.
# Here we also recover their original helper dependencies directly from
# the same Notebook 05 source, then execute those exact definitions.
# ==================================================================================================

SOURCE_FUNCTION_NAMES = [

    "validate_logits",
    "validate_class_vector",

    "stable_softmax",
    "multiclass_nll_from_logits",

    "confusion_matrix_numpy",
    "classification_metrics_from_predictions",

    "multiclass_brier_score",
    "expected_calibration_error",

    "frequency_group_metrics",

    "apply_global_temperature",
    "apply_ltlc_temperature",

    "tail_ece",

    "fit_global_temperature",
    "fit_ltlc_temperature",

    "evaluate_logits",
]


with open(
    SOURCE_NOTEBOOK_PATH,
    "r",
    encoding="utf-8"
) as f:

    exact_notebook = json.load(f)


EXACT_ENGINE_SOURCE = {}


for cell in exact_notebook.get(
    "cells",
    []
):

    if cell.get(
        "cell_type"
    ) != "code":

        continue


    source = "".join(
        cell.get(
            "source",
            []
        )
    )


    try:

        tree = ast.parse(
            source
        )

    except SyntaxError:

        continue


    source_lines = source.splitlines()


    for node in tree.body:

        if not isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef
            )
        ):

            continue


        if node.name not in SOURCE_FUNCTION_NAMES:

            continue


        # Keep first executed definition only.
        if node.name in EXACT_ENGINE_SOURCE:

            continue


        EXACT_ENGINE_SOURCE[
            node.name
        ] = "\n".join(

            source_lines[
                node.lineno - 1:
                node.end_lineno
            ]
        )


missing_functions = [

    name

    for name in SOURCE_FUNCTION_NAMES

    if name not in EXACT_ENGINE_SOURCE
]


assert missing_functions == [], (
    f"Missing original engine functions: "
    f"{missing_functions}"
)


# Execute exact original definitions into current runtime.
for name in SOURCE_FUNCTION_NAMES:

    exec(
        EXACT_ENGINE_SOURCE[
            name
        ],
        globals()
    )


print(
    "Exact original calibration engine restored:",
    len(
        SOURCE_FUNCTION_NAMES
    ),
    "/",
    len(
        SOURCE_FUNCTION_NAMES
    ),
    "PASS"
)


# ==================================================================================================
# 4. RESULT STORAGE
# ==================================================================================================

TEMPERATURE_PARAMETER_ROWS = []

TEMPERATURE_RESULT_ROWS = []

TEMPERATURE_FIT_OBJECTS = {}


# ==================================================================================================
# 5. PARAMETER-ROW HELPER
# ==================================================================================================

def append_temperature_parameter_row(
    dataset,
    seed,
    method,
    tau,
    alpha,
    a,
    b,
    temperatures,
    a_at_bound,
    b_at_bound,
    optimizer_success,
):

    temperatures = np.asarray(
        temperatures,
        dtype=np.float64
    )


    assert temperatures.ndim == 1

    assert np.isfinite(
        temperatures
    ).all()

    assert (
        temperatures
        >
        0.0
    ).all()


    TEMPERATURE_PARAMETER_ROWS.append({

        "dataset":
            dataset,

        "seed":
            int(seed),

        "method":
            method,

        "tau":
            float(tau),

        "alpha":
            float(alpha),

        "a":
            float(a),

        "b":
            float(b),

        "temperature_min":
            float(
                np.min(
                    temperatures
                )
            ),

        "temperature_median":
            float(
                np.median(
                    temperatures
                )
            ),

        "temperature_mean":
            float(
                np.mean(
                    temperatures
                )
            ),

        "temperature_max":
            float(
                np.max(
                    temperatures
                )
            ),

        "a_at_bound":
            bool(
                a_at_bound
            ),

        "b_at_bound":
            bool(
                b_at_bound
            ),

        "optimizer_success":
            bool(
                optimizer_success
            ),
    })


# ==================================================================================================
# 6. RESULT-ROW HELPER
# ==================================================================================================

def append_result_row(
    dataset,
    seed,
    method,
    tau,
    alpha,
    metrics,
):

    TEMPERATURE_RESULT_ROWS.append({

        "dataset":
            dataset,

        "seed":
            int(seed),

        "method":
            method,

        "tau":
            float(tau),

        "alpha":
            float(alpha),

        "oa":
            float(
                metrics[
                    "oa"
                ]
            ),

        "aa":
            float(
                metrics[
                    "aa"
                ]
            ),

        "macro_f1":
            float(
                metrics[
                    "macro_f1"
                ]
            ),

        "kappa":
            float(
                metrics[
                    "kappa"
                ]
            ),

        "tail_accuracy":
            float(
                metrics[
                    "tail_accuracy"
                ]
            ),

        "medium_accuracy":
            float(
                metrics[
                    "medium_accuracy"
                ]
            ),

        "head_accuracy":
            float(
                metrics[
                    "head_accuracy"
                ]
            ),

        "worst_class_accuracy":
            float(
                metrics[
                    "worst_class_accuracy"
                ]
            ),

        "ece":
            float(
                metrics[
                    "ece"
                ]
            ),

        "tail_ece":
            float(
                metrics[
                    "tail_ece"
                ]
            ),

        "nll":
            float(
                metrics[
                    "nll"
                ]
            ),

        "brier":
            float(
                metrics[
                    "brier"
                ]
            ),
    })


# ==================================================================================================
# 7. FIT ALL 9 BEST-AA VALIDATION RUNS
# ==================================================================================================

print()
print("=" * 160)
print("NOTEBOOK 05 — CORRECTED BEST-AA TEMPERATURE FITTING")
print("=" * 160)

print(
    "Validation runs:",
    len(
        RUN_ORDER
    )
)

print(
    "Result methods per run: 7"
)

print(
    "Temperature fits per run: 4"
)

print(
    "Optimizer: L-BFGS-B"
)

print(
    "Bounds: a,b ∈ [-2, 2]"
)

print(
    "Objective: validation multiclass NLL"
)

print(
    "Official test: SEALED"
)

print("=" * 160)


for dataset, seed in RUN_ORDER:

    identity = (
        dataset,
        seed
    )


    # ==============================================================================================
    # A. DATASET CONFIG
    # ==============================================================================================

    payload = (
        PRIOR_RARITY_BY_DATASET[
            dataset
        ]
    )


    prior = np.asarray(
        payload[
            "prior"
        ],
        dtype=np.float64
    )


    rarity = np.asarray(
        payload[
            "rarity"
        ],
        dtype=np.float64
    )


    frequency_groups = np.asarray(
        payload[
            "frequency_group"
        ],
        dtype=object
    )


    validation = load_bestAA_validation_output(

        FINAL_BEST_AA_VALIDATION_OUTPUTS[
            identity
        ]
    )


    ce_logits = np.asarray(
        validation[
            "logits"
        ],
        dtype=np.float64
    )


    labels = np.asarray(
        validation[
            "labels"
        ],
        dtype=np.int64
    )


    # ==============================================================================================
    # B. FROZEN SELECTED SHIFT PARAMETERS
    # ==============================================================================================

    standard_selection = (
        CORRECTED_STANDARD_LA_SELECTION[
            identity
        ]
    )


    ltlc_selection = (
        CORRECTED_LTLC_SHIFT_SELECTION[
            identity
        ]
    )


    la_tau = float(
        standard_selection[
            "tau"
        ]
    )


    ltlc_tau = float(
        ltlc_selection[
            "tau"
        ]
    )


    ltlc_alpha = float(
        ltlc_selection[
            "alpha"
        ]
    )


    # ==============================================================================================
    # C. SHIFTED LOGITS
    # ==============================================================================================

    standard_la = standard_la_logits(

        ce_logits,
        prior,
        tau=la_tau,
        eps=RARITY_EPSILON
    )


    ltlc_shift = ltlc_shift_logits(

        ce_logits,
        prior,
        rarity,
        tau=ltlc_tau,
        alpha=ltlc_alpha,
        eps=RARITY_EPSILON
    )


    # All selected alpha values are now known to be zero.
    assert ltlc_alpha == 0.0


    # Because selected tau is also identical between LA and LTLC
    # for every run, the arrays must match exactly.
    assert la_tau == ltlc_tau

    assert np.array_equal(
        standard_la,
        ltlc_shift
    )


    # ==============================================================================================
    # D. UNCALIBRATED METRICS
    # ==============================================================================================

    ce_eval = evaluate_logits(

        ce_logits,
        labels,
        frequency_groups
    )


    la_eval = evaluate_logits(

        standard_la,
        labels,
        frequency_groups
    )


    ltlc_shift_eval = evaluate_logits(

        ltlc_shift,
        labels,
        frequency_groups
    )


    append_result_row(

        dataset,
        seed,
        "CE",
        0.0,
        0.0,
        ce_eval
    )


    append_result_row(

        dataset,
        seed,
        "Standard_LA",
        la_tau,
        0.0,
        la_eval
    )


    # ==============================================================================================
    # E. GLOBAL TS ON CE
    # ==============================================================================================

    global_fit = fit_global_temperature(

        ce_logits,
        labels
    )


    global_logits = np.asarray(
        global_fit[
            "calibrated_logits"
        ],
        dtype=np.float64
    )


    global_eval = evaluate_logits(

        global_logits,
        labels,
        frequency_groups
    )


    global_temperatures = np.full(

        len(
            labels
        ),

        float(
            global_fit[
                "temperature"
            ]
        ),

        dtype=np.float64
    )


    append_result_row(

        dataset,
        seed,
        "Global_TS",
        0.0,
        0.0,
        global_eval
    )


    append_temperature_parameter_row(

        dataset=dataset,
        seed=seed,
        method="Global_TS",
        tau=0.0,
        alpha=0.0,

        a=
            global_fit[
                "a"
            ],

        b=0.0,

        temperatures=
            global_temperatures,

        a_at_bound=
            global_fit[
                "a_at_bound"
            ],

        b_at_bound=False,

        optimizer_success=
            global_fit[
                "success"
            ],
    )


    # ==============================================================================================
    # F. GLOBAL TS AFTER STANDARD LA
    # ==============================================================================================

    la_global_fit = fit_global_temperature(

        standard_la,
        labels
    )


    la_global_logits = np.asarray(
        la_global_fit[
            "calibrated_logits"
        ],
        dtype=np.float64
    )


    la_global_eval = evaluate_logits(

        la_global_logits,
        labels,
        frequency_groups
    )


    la_global_temperatures = np.full(

        len(
            labels
        ),

        float(
            la_global_fit[
                "temperature"
            ]
        ),

        dtype=np.float64
    )


    append_result_row(

        dataset,
        seed,
        "LA_Global_TS",
        la_tau,
        0.0,
        la_global_eval
    )


    append_temperature_parameter_row(

        dataset=dataset,
        seed=seed,
        method="LA_Global_TS",
        tau=la_tau,
        alpha=0.0,

        a=
            la_global_fit[
                "a"
            ],

        b=0.0,

        temperatures=
            la_global_temperatures,

        a_at_bound=
            la_global_fit[
                "a_at_bound"
            ],

        b_at_bound=False,

        optimizer_success=
            la_global_fit[
                "success"
            ],
    )


    # ==============================================================================================
    # G. LTLC SHIFT
    # ==============================================================================================

    append_result_row(

        dataset,
        seed,
        "LTLC_Shift",
        ltlc_tau,
        ltlc_alpha,
        ltlc_shift_eval
    )


    # ==============================================================================================
    # H. GLOBAL TS AFTER LTLC SHIFT
    # ==============================================================================================

    ltlc_global_fit = fit_global_temperature(

        ltlc_shift,
        labels
    )


    ltlc_global_logits = np.asarray(
        ltlc_global_fit[
            "calibrated_logits"
        ],
        dtype=np.float64
    )


    ltlc_global_eval = evaluate_logits(

        ltlc_global_logits,
        labels,
        frequency_groups
    )


    ltlc_global_temperatures = np.full(

        len(
            labels
        ),

        float(
            ltlc_global_fit[
                "temperature"
            ]
        ),

        dtype=np.float64
    )


    append_result_row(

        dataset,
        seed,
        "LTLC_Shift_Global_TS",
        ltlc_tau,
        ltlc_alpha,
        ltlc_global_eval
    )


    append_temperature_parameter_row(

        dataset=dataset,
        seed=seed,
        method="LTLC_Shift_Global_TS",
        tau=ltlc_tau,
        alpha=ltlc_alpha,

        a=
            ltlc_global_fit[
                "a"
            ],

        b=0.0,

        temperatures=
            ltlc_global_temperatures,

        a_at_bound=
            ltlc_global_fit[
                "a_at_bound"
            ],

        b_at_bound=False,

        optimizer_success=
            ltlc_global_fit[
                "success"
            ],
    )


    # ==============================================================================================
    # I. LTLC FULL — RARITY-CONDITIONED SAMPLEWISE TEMPERATURE
    # ==============================================================================================

    ltlc_full_fit = fit_ltlc_temperature(

        ltlc_shift,
        labels,
        rarity
    )


    ltlc_full_logits = np.asarray(
        ltlc_full_fit[
            "calibrated_logits"
        ],
        dtype=np.float64
    )


    ltlc_full_eval = evaluate_logits(

        ltlc_full_logits,
        labels,
        frequency_groups
    )


    # Positive samplewise scalar temperature MUST preserve hard predictions.
    assert np.array_equal(

        np.argmax(
            ltlc_shift,
            axis=1
        ),

        np.argmax(
            ltlc_full_logits,
            axis=1
        )
    )


    append_result_row(

        dataset,
        seed,
        "LTLC_Full",
        ltlc_tau,
        ltlc_alpha,
        ltlc_full_eval
    )


    predicted_before_temperature = np.argmax(

        ltlc_shift,
        axis=1
    )


    predicted_rarity = rarity[
        predicted_before_temperature
    ]


    full_temperatures = np.exp(

        float(
            ltlc_full_fit[
                "a"
            ]
        )

        +

        float(
            ltlc_full_fit[
                "b"
            ]
        )

        *

        predicted_rarity
    )


    append_temperature_parameter_row(

        dataset=dataset,
        seed=seed,
        method="LTLC_Full",
        tau=ltlc_tau,
        alpha=ltlc_alpha,

        a=
            ltlc_full_fit[
                "a"
            ],

        b=
            ltlc_full_fit[
                "b"
            ],

        temperatures=
            full_temperatures,

        a_at_bound=
            ltlc_full_fit[
                "a_at_bound"
            ],

        b_at_bound=
            ltlc_full_fit[
                "b_at_bound"
            ],

        optimizer_success=
            ltlc_full_fit[
                "success"
            ],
    )


    # ==============================================================================================
    # J. MANDATORY CALIBRATION SANITY CHECKS
    # ==============================================================================================

    # Global temperature cannot change hard predictions.
    assert np.array_equal(

        np.argmax(
            ce_logits,
            axis=1
        ),

        np.argmax(
            global_logits,
            axis=1
        )
    )


    assert np.array_equal(

        np.argmax(
            standard_la,
            axis=1
        ),

        np.argmax(
            la_global_logits,
            axis=1
        )
    )


    assert np.array_equal(

        np.argmax(
            ltlc_shift,
            axis=1
        ),

        np.argmax(
            ltlc_global_logits,
            axis=1
        )
    )


    # Selected LTLC shift = selected Standard LA.
    assert np.array_equal(

        np.argmax(
            standard_la,
            axis=1
        ),

        np.argmax(
            ltlc_shift,
            axis=1
        )
    )


    TEMPERATURE_FIT_OBJECTS[
        identity
    ] = {

        "Global_TS":
            global_fit,

        "LA_Global_TS":
            la_global_fit,

        "LTLC_Shift_Global_TS":
            ltlc_global_fit,

        "LTLC_Full":
            ltlc_full_fit,
    }


    # ==============================================================================================
    # K. COMPACT OUTPUT
    # ==============================================================================================

    print()

    print(
        f"{dataset:11s} | seed={seed:4d} | "
        f"tau={ltlc_tau:.2f} alpha={ltlc_alpha:.2f}"
    )


    print(

        f"  Standard LA : "
        f"ECE={la_eval['ece']:.9f} | "
        f"TailECE={la_eval['tail_ece']:.9f} | "
        f"NLL={la_eval['nll']:.9f} | "
        f"Brier={la_eval['brier']:.9f}"
    )


    print(

        f"  Global TS   : "
        f"T={global_fit['temperature']:.6f} | "
        f"ECE={global_eval['ece']:.9f} | "
        f"NLL={global_eval['nll']:.9f}"
    )


    print(

        f"  LA + TS     : "
        f"T={la_global_fit['temperature']:.6f} | "
        f"ECE={la_global_eval['ece']:.9f} | "
        f"NLL={la_global_eval['nll']:.9f}"
    )


    print(

        f"  LTLC Full   : "
        f"a={ltlc_full_fit['a']:+.6f} | "
        f"b={ltlc_full_fit['b']:+.6f} | "
        f"T=[{ltlc_full_fit['temperature_min']:.6f}, "
        f"{ltlc_full_fit['temperature_max']:.6f}] | "
        f"ECE={ltlc_full_eval['ece']:.9f} | "
        f"TailECE={ltlc_full_eval['tail_ece']:.9f} | "
        f"NLL={ltlc_full_eval['nll']:.9f} | "
        f"Brier={ltlc_full_eval['brier']:.9f}"
    )


    print(

        f"  LA - LTLC Full improvements: "
        f"ECE={la_eval['ece'] - ltlc_full_eval['ece']:+.9e} | "
        f"TailECE={la_eval['tail_ece'] - ltlc_full_eval['tail_ece']:+.9e} | "
        f"NLL={la_eval['nll'] - ltlc_full_eval['nll']:+.9e} | "
        f"Brier={la_eval['brier'] - ltlc_full_eval['brier']:+.9e}"
    )


# ==================================================================================================
# 8. DATAFRAMES + EXPECTED ROW COUNTS
# ==================================================================================================

CORRECTED_TEMPERATURE_PARAMETERS_DF = pd.DataFrame(
    TEMPERATURE_PARAMETER_ROWS
)


CORRECTED_TEMPERATURE_RESULTS_DF = pd.DataFrame(
    TEMPERATURE_RESULT_ROWS
)


assert len(
    CORRECTED_TEMPERATURE_PARAMETERS_DF
) == 36


assert len(
    CORRECTED_TEMPERATURE_RESULTS_DF
) == 63


assert (
    CORRECTED_TEMPERATURE_PARAMETERS_DF[
        "optimizer_success"
    ].all()
), (
    "At least one temperature optimizer reported failure."
)


# ==================================================================================================
# 9. METHOD COUNTS
# ==================================================================================================

EXPECTED_RESULT_METHODS = {

    "CE",
    "Standard_LA",
    "Global_TS",
    "LA_Global_TS",
    "LTLC_Shift",
    "LTLC_Shift_Global_TS",
    "LTLC_Full",
}


assert set(
    CORRECTED_TEMPERATURE_RESULTS_DF[
        "method"
    ].unique()
) == EXPECTED_RESULT_METHODS


for method in EXPECTED_RESULT_METHODS:

    assert (
        CORRECTED_TEMPERATURE_RESULTS_DF[
            "method"
        ]
        ==
        method
    ).sum() == 9


# ==================================================================================================
# 10. TANGDAOWAN REGRESSION AGAINST ORIGINAL FROZEN RESULTS
# ==================================================================================================

OLD_PARAMETER_PATH = (
    NB05_AUDIT_ROOT
    /
    "tangdaowan_temperature_validation_parameters.csv"
)


OLD_RESULT_PATH = (
    NB05_AUDIT_ROOT
    /
    "tangdaowan_temperature_validation_results.csv"
)


assert OLD_PARAMETER_PATH.exists()
assert OLD_RESULT_PATH.exists()


old_parameters = pd.read_csv(
    OLD_PARAMETER_PATH
)


old_results = pd.read_csv(
    OLD_RESULT_PATH
)


new_td_parameters = (

    CORRECTED_TEMPERATURE_PARAMETERS_DF[
        CORRECTED_TEMPERATURE_PARAMETERS_DF[
            "dataset"
        ]
        ==
        "Tangdaowan"
    ]
    .copy()
)


new_td_results = (

    CORRECTED_TEMPERATURE_RESULTS_DF[
        CORRECTED_TEMPERATURE_RESULTS_DF[
            "dataset"
        ]
        ==
        "Tangdaowan"
    ]
    .copy()
)


assert len(
    old_parameters
) == 12

assert len(
    new_td_parameters
) == 12

assert len(
    old_results
) == 21

assert len(
    new_td_results
) == 21


print()
print("=" * 160)
print("TANGDAOWAN ORIGINAL TEMPERATURE REGRESSION")
print("=" * 160)


# --------------------------------------------------------------------------------------------------
# Parameter regression
# --------------------------------------------------------------------------------------------------

parameter_numeric_columns = [

    "tau",
    "alpha",
    "a",
    "b",
    "temperature_min",
    "temperature_median",
    "temperature_mean",
    "temperature_max",
]


for _, old_row in old_parameters.iterrows():

    seed = int(
        old_row[
            "seed"
        ]
    )


    method = str(
        old_row[
            "method"
        ]
    )


    match = new_td_parameters[
        (
            new_td_parameters[
                "seed"
            ].astype(
                int
            )
            ==
            seed
        )
        &
        (
            new_td_parameters[
                "method"
            ]
            ==
            method
        )
    ]


    assert len(
        match
    ) == 1


    new_row = match.iloc[
        0
    ]


    for column in parameter_numeric_columns:

        difference = abs(

            float(
                new_row[
                    column
                ]
            )
            -
            float(
                old_row[
                    column
                ]
            )
        )


        assert difference <= 1e-8, (
            f"Tangdaowan temperature parameter regression mismatch: "
            f"seed={seed}, method={method}, "
            f"column={column}, difference={difference}"
        )


# --------------------------------------------------------------------------------------------------
# Result regression
# --------------------------------------------------------------------------------------------------

result_numeric_columns = [

    "tau",
    "alpha",

    "oa",
    "aa",
    "macro_f1",
    "kappa",

    "tail_accuracy",
    "medium_accuracy",
    "head_accuracy",
    "worst_class_accuracy",

    "ece",
    "tail_ece",
    "nll",
    "brier",
]


for _, old_row in old_results.iterrows():

    seed = int(
        old_row[
            "seed"
        ]
    )


    method = str(
        old_row[
            "method"
        ]
    )


    match = new_td_results[
        (
            new_td_results[
                "seed"
            ].astype(
                int
            )
            ==
            seed
        )
        &
        (
            new_td_results[
                "method"
            ]
            ==
            method
        )
    ]


    assert len(
        match
    ) == 1


    new_row = match.iloc[
        0
    ]


    for column in result_numeric_columns:

        difference = abs(

            float(
                new_row[
                    column
                ]
            )
            -
            float(
                old_row[
                    column
                ]
            )
        )


        tolerance = (

            1e-8

            if column in {
                "ece",
                "tail_ece",
                "nll",
                "brier",
            }

            else
            1e-12
        )


        assert difference <= tolerance, (
            f"Tangdaowan temperature-result regression mismatch: "
            f"seed={seed}, method={method}, "
            f"column={column}, difference={difference}"
        )


print(
    "Tangdaowan temperature parameters : 12/12 PASS"
)

print(
    "Tangdaowan calibration results    : 21/21 PASS"
)


# ==================================================================================================
# 11. SAVE CORRECTED TEMPERATURE ARTIFACTS
# ==================================================================================================

CORRECTED_TEMPERATURE_PARAMETER_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_temperature_validation_parameters.csv"
)


CORRECTED_TEMPERATURE_RESULT_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_temperature_validation_results.csv"
)


CORRECTED_TEMPERATURE_MANIFEST_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_temperature_validation_manifest.json"
)


# --------------------------------------------------------------------------------------------------
# CSV — parameters
# --------------------------------------------------------------------------------------------------

parameter_csv_text = (

    CORRECTED_TEMPERATURE_PARAMETERS_DF.to_csv(
        index=False,
        float_format="%.17g"
    )
)


if CORRECTED_TEMPERATURE_PARAMETER_PATH.exists():

    existing = (
        CORRECTED_TEMPERATURE_PARAMETER_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == parameter_csv_text, (
        "Existing corrected temperature parameter CSV differs."
    )


else:

    CORRECTED_TEMPERATURE_PARAMETER_PATH.write_text(

        parameter_csv_text,

        encoding="utf-8"
    )


# --------------------------------------------------------------------------------------------------
# CSV — results
# --------------------------------------------------------------------------------------------------

result_csv_text = (

    CORRECTED_TEMPERATURE_RESULTS_DF.to_csv(
        index=False,
        float_format="%.17g"
    )
)


if CORRECTED_TEMPERATURE_RESULT_PATH.exists():

    existing = (
        CORRECTED_TEMPERATURE_RESULT_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == result_csv_text, (
        "Existing corrected temperature result CSV differs."
    )


else:

    CORRECTED_TEMPERATURE_RESULT_PATH.write_text(

        result_csv_text,

        encoding="utf-8"
    )


CORRECTED_TEMPERATURE_PARAMETER_SHA = sha256_file(
    CORRECTED_TEMPERATURE_PARAMETER_PATH
)


CORRECTED_TEMPERATURE_RESULT_SHA = sha256_file(
    CORRECTED_TEMPERATURE_RESULT_PATH
)


# ==================================================================================================
# 12. DATASET-LEVEL CALIBRATION SUMMARY
# ==================================================================================================

DATASET_CALIBRATION_SUMMARY = {}


print()
print("=" * 160)
print("CORRECTED BEST-AA CALIBRATION SUMMARY")
print("=" * 160)


for dataset in [

    "Pingan",
    "Qingyun",
    "Tangdaowan",

]:

    dataset_df = (

        CORRECTED_TEMPERATURE_RESULTS_DF[
            CORRECTED_TEMPERATURE_RESULTS_DF[
                "dataset"
            ]
            ==
            dataset
        ]
    )


    standard_df = (

        dataset_df[
            dataset_df[
                "method"
            ]
            ==
            "Standard_LA"
        ]
        .sort_values(
            "seed"
        )
        .reset_index(
            drop=True
        )
    )


    full_df = (

        dataset_df[
            dataset_df[
                "method"
            ]
            ==
            "LTLC_Full"
        ]
        .sort_values(
            "seed"
        )
        .reset_index(
            drop=True
        )
    )


    assert len(
        standard_df
    ) == 3


    assert len(
        full_df
    ) == 3


    improvements = {

        "ECE":
            float(
                (
                    standard_df[
                        "ece"
                    ]
                    -
                    full_df[
                        "ece"
                    ]
                ).mean()
            ),

        "Tail_ECE":
            float(
                (
                    standard_df[
                        "tail_ece"
                    ]
                    -
                    full_df[
                        "tail_ece"
                    ]
                ).mean()
            ),

        "NLL":
            float(
                (
                    standard_df[
                        "nll"
                    ]
                    -
                    full_df[
                        "nll"
                    ]
                ).mean()
            ),

        "Brier":
            float(
                (
                    standard_df[
                        "brier"
                    ]
                    -
                    full_df[
                        "brier"
                    ]
                ).mean()
            ),
    }


    # Classification MUST remain identical between LTLC Shift and LTLC Full.
    shift_df = (

        dataset_df[
            dataset_df[
                "method"
            ]
            ==
            "LTLC_Shift"
        ]
        .sort_values(
            "seed"
        )
        .reset_index(
            drop=True
        )
    )


    for metric in [

        "oa",
        "aa",
        "macro_f1",
        "tail_accuracy",
        "worst_class_accuracy",

    ]:

        assert np.allclose(

            shift_df[
                metric
            ].to_numpy(
                dtype=np.float64
            ),

            full_df[
                metric
            ].to_numpy(
                dtype=np.float64
            ),

            rtol=0.0,

            atol=1e-12
        )


    calibration_signal = bool(

        improvements[
            "ECE"
        ] > SIGNAL_TOL

        or

        improvements[
            "Tail_ECE"
        ] > SIGNAL_TOL

        or

        improvements[
            "NLL"
        ] > SIGNAL_TOL
    )


    classification_summary = (

        CORRECTED_SHIFT_SELECTION_SUMMARY_DF[
            CORRECTED_SHIFT_SELECTION_SUMMARY_DF[
                "dataset"
            ]
            ==
            dataset
        ]
    )


    classification_signal = bool(

        (
            classification_summary[
                "delta_aa"
            ]
            >
            SIGNAL_TOL
        ).any()

        or

        (
            classification_summary[
                "delta_macro_f1"
            ]
            >
            SIGNAL_TOL
        ).any()

        or

        (
            classification_summary[
                "delta_tail_accuracy"
            ]
            >
            SIGNAL_TOL
        ).any()

        or

        (
            classification_summary[
                "delta_worst_class_accuracy"
            ]
            >
            SIGNAL_TOL
        ).any()
    )


    DATASET_CALIBRATION_SUMMARY[
        dataset
    ] = {

        "standard_la_minus_ltlc_full":
            improvements,

        "classification_signal":
            classification_signal,

        "calibration_signal":
            calibration_signal,
    }


    print()

    print(
        dataset
    )


    print(
        "  Mean Standard-LA − LTLC-Full ECE      :",
        f"{improvements['ECE']:+.9e}"
    )

    print(
        "  Mean Standard-LA − LTLC-Full Tail-ECE :",
        f"{improvements['Tail_ECE']:+.9e}"
    )

    print(
        "  Mean Standard-LA − LTLC-Full NLL      :",
        f"{improvements['NLL']:+.9e}"
    )

    print(
        "  Mean Standard-LA − LTLC-Full Brier    :",
        f"{improvements['Brier']:+.9e}"
    )

    print(
        "  Classification signal                 :",
        classification_signal
    )

    print(
        "  Calibration signal                    :",
        calibration_signal
    )


# ==================================================================================================
# 13. MANIFEST
# ==================================================================================================

temperature_manifest = {

    "project":
        "LTLC",

    "notebook":
        "05",

    "purpose":
        (
            "Corrected best-validation-AA validation-only "
            "temperature fitting using exact frozen Notebook 05 "
            "calibration implementation."
        ),

    "source_checkpoint_purpose":
        "best_validation_AA",

    "implementation_spec_sha256":
        ACTUAL_FROZEN_HASHES[
            "implementation_spec"
        ],

    "training_prior_rarity_manifest_sha256":
        ACTUAL_FROZEN_HASHES[
            "prior_rarity"
        ],

    "corrected_shift_selection_manifest_sha256":
        CORRECTED_SHIFT_SELECTION_SHA,

    "exact_calibration_function_audit_sha256":
        TARGETED_FUNCTION_AUDIT_SHA,

    "optimizer":
        "L-BFGS-B",

    "temperature_bounds":
        {
            "a":
                [-2.0, 2.0],

            "b":
                [-2.0, 2.0],
        },

    "temperature_initial":
        {
            "a":
                0.0,

            "b":
                0.0,
        },

    "objective":
        "validation_multiclass_NLL",

    "ece_bins":
        15,

    "parameter_rows":
        36,

    "validation_result_rows":
        63,

    "parameter_csv":
        str(
            CORRECTED_TEMPERATURE_PARAMETER_PATH.relative_to(
                ROOT
            )
        ),

    "parameter_csv_sha256":
        CORRECTED_TEMPERATURE_PARAMETER_SHA,

    "result_csv":
        str(
            CORRECTED_TEMPERATURE_RESULT_PATH.relative_to(
                ROOT
            )
        ),

    "result_csv_sha256":
        CORRECTED_TEMPERATURE_RESULT_SHA,

    "tangdaowan_original_regression_pass":
        True,

    "dataset_summary":
        DATASET_CALIBRATION_SUMMARY,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,

    "status":
        "CORRECTED_BEST_AA_TEMPERATURE_FITTING_COMPLETE",
}


manifest_text = (

    json.dumps(
        temperature_manifest,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )
    +
    "\n"
)


if CORRECTED_TEMPERATURE_MANIFEST_PATH.exists():

    existing = (
        CORRECTED_TEMPERATURE_MANIFEST_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == manifest_text, (
        "Existing corrected temperature manifest differs."
    )


else:

    CORRECTED_TEMPERATURE_MANIFEST_PATH.write_text(

        manifest_text,

        encoding="utf-8"
    )


CORRECTED_TEMPERATURE_MANIFEST_SHA = sha256_file(
    CORRECTED_TEMPERATURE_MANIFEST_PATH
)


# ==================================================================================================
# 14. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_CORRECTED_TEMPERATURE_GATE_PASS = True


print()
print("=" * 160)

print(
    "Temperature parameter fits          : 36/36 PASS"
)

print(
    "Validation method-result rows       : 63/63 PASS"
)

print(
    "Tangdaowan parameter regression     : 12/12 PASS"
)

print(
    "Tangdaowan result regression        : 21/21 PASS"
)

print()

print(
    "Temperature parameter CSV:"
)

print(
    CORRECTED_TEMPERATURE_PARAMETER_PATH
)

print(
    "SHA-256:"
)

print(
    CORRECTED_TEMPERATURE_PARAMETER_SHA
)

print()

print(
    "Temperature result CSV:"
)

print(
    CORRECTED_TEMPERATURE_RESULT_PATH
)

print(
    "SHA-256:"
)

print(
    CORRECTED_TEMPERATURE_RESULT_SHA
)

print()

print(
    "Temperature manifest:"
)

print(
    CORRECTED_TEMPERATURE_MANIFEST_PATH
)

print(
    "SHA-256:"
)

print(
    CORRECTED_TEMPERATURE_MANIFEST_SHA
)

print()

print(
    "Runtime                         :",
    DEVICE
)

print(
    "Training performed              :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official-test dataset created   :",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created:",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used              :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 CORRECTED BEST-AA TEMPERATURE FITTING: PASS"
)

print(
    "NEXT: FINAL VALIDATION-ONLY DEVELOPMENT GATE"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 160)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 19C
# FINAL VALIDATION-ONLY DEVELOPMENT GATE
#
# PURPOSE
#   Freeze the corrected best-val-AA development conclusion after:
#
#       - 9/9 exact Standard-LA selections
#       - 9/9 exact LTLC-Shift selections
#       - 36/36 validation-only temperature fits
#       - 63/63 validation method-result rows
#
# LOCKED DEVELOPMENT REQUIREMENT:
#
#   Recognition side:
#       LTLC-Shift must improve >= 1 of:
#           AA / Macro-F1 / Tail Accuracy / Worst-Class Accuracy
#       without a large OA collapse.
#
#   Calibration side:
#       LTLC-Full must improve >= 1 of:
#           ECE / Tail-ECE / NLL
#
# BOTH sides are required for GO.
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json

import numpy as np
import pandas as pd


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_CORRECTED_SHIFT_GRID_PASS is True
assert NOTEBOOK05_EXACT_SHIFT_SELECTION_VERIFIED is True
assert NOTEBOOK05_CORRECTED_TEMPERATURE_GATE_PASS is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


SIGNAL_TOL = 1e-12


# ==================================================================================================
# 2. LOCKED GATE METRICS
# ==================================================================================================

CLASSIFICATION_METRICS = [

    "delta_aa",

    "delta_macro_f1",

    "delta_tail_accuracy",

    "delta_worst_class_accuracy",
]


CALIBRATION_METRICS = [

    "ECE",

    "Tail_ECE",

    "NLL",
]


# ==================================================================================================
# 3. DATASET-LEVEL GATE EVALUATION
# ==================================================================================================

FINAL_DEVELOPMENT_GATE = {}


print("=" * 155)
print("NOTEBOOK 05 — FINAL CORRECTED VALIDATION-ONLY DEVELOPMENT GATE")
print("=" * 155)

print(
    "Required for GO:"
)

print(
    "  Recognition   : improve >=1 of "
    "AA / Macro-F1 / Tail Accuracy / Worst-Class Accuracy"
)

print(
    "  Calibration   : improve >=1 of "
    "ECE / Tail-ECE / NLL"
)

print(
    "  Both signals  : REQUIRED"
)

print(
    "Official test  : SEALED"
)

print("=" * 155)


for dataset in [

    "Pingan",

    "Qingyun",

    "Tangdaowan",

]:

    # ==============================================================================================
    # A. CLASSIFICATION SIDE
    # ==============================================================================================

    classification_df = (

        CORRECTED_SHIFT_SELECTION_SUMMARY_DF[
            CORRECTED_SHIFT_SELECTION_SUMMARY_DF[
                "dataset"
            ]
            ==
            dataset
        ]
        .copy()
    )


    assert len(
        classification_df
    ) == 3


    classification_max_improvement = {

        "AA":

            float(
                classification_df[
                    "delta_aa"
                ].max()
            ),


        "Macro_F1":

            float(
                classification_df[
                    "delta_macro_f1"
                ].max()
            ),


        "Tail_Accuracy":

            float(
                classification_df[
                    "delta_tail_accuracy"
                ].max()
            ),


        "Worst_Class_Accuracy":

            float(
                classification_df[
                    "delta_worst_class_accuracy"
                ].max()
            ),
    }


    classification_mean_delta = {

        "AA":

            float(
                classification_df[
                    "delta_aa"
                ].mean()
            ),


        "Macro_F1":

            float(
                classification_df[
                    "delta_macro_f1"
                ].mean()
            ),


        "Tail_Accuracy":

            float(
                classification_df[
                    "delta_tail_accuracy"
                ].mean()
            ),


        "Worst_Class_Accuracy":

            float(
                classification_df[
                    "delta_worst_class_accuracy"
                ].mean()
            ),
    }


    classification_signal = bool(

        any(

            value
            >
            SIGNAL_TOL

            for value in
            classification_max_improvement.values()
        )
    )


    # LTLC Shift == Standard LA in all selected runs,
    # therefore OA collapse is exactly absent.
    standard_rows = (

        CORRECTED_TEMPERATURE_RESULTS_DF[
            (
                CORRECTED_TEMPERATURE_RESULTS_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                CORRECTED_TEMPERATURE_RESULTS_DF[
                    "method"
                ]
                ==
                "Standard_LA"
            )
        ]
        .sort_values(
            "seed"
        )
        .reset_index(
            drop=True
        )
    )


    shift_rows = (

        CORRECTED_TEMPERATURE_RESULTS_DF[
            (
                CORRECTED_TEMPERATURE_RESULTS_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                CORRECTED_TEMPERATURE_RESULTS_DF[
                    "method"
                ]
                ==
                "LTLC_Shift"
            )
        ]
        .sort_values(
            "seed"
        )
        .reset_index(
            drop=True
        )
    )


    assert len(
        standard_rows
    ) == 3


    assert len(
        shift_rows
    ) == 3


    oa_deltas = (

        shift_rows[
            "oa"
        ].to_numpy(
            dtype=np.float64
        )

        -

        standard_rows[
            "oa"
        ].to_numpy(
            dtype=np.float64
        )
    )


    assert np.allclose(

        oa_deltas,

        0.0,

        rtol=0.0,

        atol=1e-12
    )


    large_oa_collapse = False


    # ==============================================================================================
    # B. CALIBRATION SIDE
    #
    # Positive means Standard-LA metric is worse than LTLC-Full,
    # therefore positive = LTLC-Full improvement.
    # ==============================================================================================

    calibration = (
        DATASET_CALIBRATION_SUMMARY[
            dataset
        ][
            "standard_la_minus_ltlc_full"
        ]
    )


    calibration_improvement = {

        "ECE":

            float(
                calibration[
                    "ECE"
                ]
            ),


        "Tail_ECE":

            float(
                calibration[
                    "Tail_ECE"
                ]
            ),


        "NLL":

            float(
                calibration[
                    "NLL"
                ]
            ),


        "Brier":

            float(
                calibration[
                    "Brier"
                ]
            ),
    }


    calibration_signal = bool(

        calibration_improvement[
            "ECE"
        ]
        >
        SIGNAL_TOL

        or

        calibration_improvement[
            "Tail_ECE"
        ]
        >
        SIGNAL_TOL

        or

        calibration_improvement[
            "NLL"
        ]
        >
        SIGNAL_TOL
    )


    # ==============================================================================================
    # C. ADAPTIVE-SHIFT SELECTION
    # ==============================================================================================

    ltlc_alpha_values = (

        classification_df[
            "ltlc_alpha"
        ].to_numpy(
            dtype=np.float64
        )
    )


    nonzero_alpha_seed_count = int(

        (
            ltlc_alpha_values
            >
            SIGNAL_TOL
        ).sum()
    )


    assert (
        nonzero_alpha_seed_count
        ==
        0
    )


    # ==============================================================================================
    # D. DATASET GATE
    # ==============================================================================================

    dataset_go = bool(

        classification_signal

        and

        calibration_signal

        and

        not large_oa_collapse
    )


    dataset_status = (

        "GO"

        if dataset_go

        else "REVISE"
    )


    FINAL_DEVELOPMENT_GATE[
        dataset
    ] = {

        "classification_signal":
            classification_signal,

        "calibration_signal":
            calibration_signal,

        "large_oa_collapse":
            large_oa_collapse,

        "nonzero_alpha_seed_count":
            nonzero_alpha_seed_count,

        "classification_max_improvement":
            classification_max_improvement,

        "classification_mean_delta":
            classification_mean_delta,

        "standard_la_minus_ltlc_full_calibration":
            calibration_improvement,

        "status":
            dataset_status,
    }


    print()

    print(
        dataset.upper()
    )

    print(
        "  Classification signal :",
        classification_signal
    )

    print(
        "  Calibration signal    :",
        calibration_signal
    )

    print(
        "  Large OA collapse     :",
        large_oa_collapse
    )

    print(
        "  Nonzero-alpha seeds   :",
        nonzero_alpha_seed_count,
        "/ 3"
    )

    print()

    print(
        "  Mean ΔAA              :",
        f"{classification_mean_delta['AA']:+.9e}"
    )

    print(
        "  Mean ΔMacro-F1        :",
        f"{classification_mean_delta['Macro_F1']:+.9e}"
    )

    print(
        "  Mean ΔTail Accuracy   :",
        f"{classification_mean_delta['Tail_Accuracy']:+.9e}"
    )

    print(
        "  Mean ΔWorst-Class Acc :",
        f"{classification_mean_delta['Worst_Class_Accuracy']:+.9e}"
    )

    print()

    print(
        "  LA − Full ECE         :",
        f"{calibration_improvement['ECE']:+.9e}"
    )

    print(
        "  LA − Full Tail-ECE    :",
        f"{calibration_improvement['Tail_ECE']:+.9e}"
    )

    print(
        "  LA − Full NLL         :",
        f"{calibration_improvement['NLL']:+.9e}"
    )

    print(
        "  LA − Full Brier       :",
        f"{calibration_improvement['Brier']:+.9e}"
    )

    print()

    print(
        "  DATASET GATE          :",
        dataset_status
    )


# ==================================================================================================
# 4. CROSS-DATASET SCIENTIFIC CONCLUSION
# ==================================================================================================

dataset_go_count = int(

    sum(

        gate[
            "status"
        ]
        ==
        "GO"

        for gate in
        FINAL_DEVELOPMENT_GATE.values()
    )
)


dataset_revise_count = int(

    sum(

        gate[
            "status"
        ]
        ==
        "REVISE"

        for gate in
        FINAL_DEVELOPMENT_GATE.values()
    )
)


classification_signal_dataset_count = int(

    sum(

        gate[
            "classification_signal"
        ]

        for gate in
        FINAL_DEVELOPMENT_GATE.values()
    )
)


calibration_signal_dataset_count = int(

    sum(

        gate[
            "calibration_signal"
        ]

        for gate in
        FINAL_DEVELOPMENT_GATE.values()
    )
)


assert dataset_go_count == 0

assert dataset_revise_count == 3

assert classification_signal_dataset_count == 0

assert calibration_signal_dataset_count == 3


OVERALL_DEVELOPMENT_STATUS = "REVISE"


# ==================================================================================================
# 5. FAILURE-MODE CLASSIFICATION
# ==================================================================================================

FAILURE_MODE = {

    "adaptive_shift_component_B":
        "NO_VALIDATION_CLASSIFICATION_SIGNAL",

    "temperature_calibration_component_C":
        "VALIDATION_CALIBRATION_SIGNAL_PRESENT",

    "long_tail_recognition_claim_supported":
        False,

    "calibration_claim_supported_on_validation":
        True,

    "tail_calibration_uniformly_improved":
        False,

    "interpretation":
        (
            "The rarity-adaptive shift was never selected with nonzero alpha "
            "and produced no AA, Macro-F1, Tail Accuracy, or Worst-Class "
            "Accuracy improvement on any corrected best-validation-AA run. "
            "The samplewise temperature component improved at least one "
            "calibration metric on all datasets, especially NLL, but this "
            "cannot establish a long-tail recognition improvement because "
            "positive scalar temperature preserves hard predictions."
        ),
}


# Mean Tail-ECE is worse for LTLC-Full on every dataset.
for dataset in [
    "Pingan",
    "Qingyun",
    "Tangdaowan",
]:

    assert (
        FINAL_DEVELOPMENT_GATE[
            dataset
        ][
            "standard_la_minus_ltlc_full_calibration"
        ][
            "Tail_ECE"
        ]
        <
        0.0
    )


# ==================================================================================================
# 6. OFFICIAL-TEST DECISION
# ==================================================================================================

OFFICIAL_TEST_DECISION = {

    "may_open_official_test_now":
        False,

    "reason":
        (
            "The locked validation development gate requires both a "
            "classification signal and a calibration signal. "
            "Classification signal is absent on all three datasets."
        ),

    "required_action":
        "REVISE_METHOD_BEFORE_OFFICIAL_TEST_INSPECTION",
}


assert (
    OFFICIAL_TEST_DECISION[
        "may_open_official_test_now"
    ]
    is False
)


# ==================================================================================================
# 7. SAVE FINAL DEVELOPMENT-GATE ARTIFACT
# ==================================================================================================

FINAL_DEVELOPMENT_GATE_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_corrected_final_validation_development_gate.json"
)


FINAL_DEVELOPMENT_GATE_OBJECT = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "status":
        OVERALL_DEVELOPMENT_STATUS,

    "source_checkpoint_purpose":
        "best_validation_AA",

    "gate_requirement": {

        "recognition":
            (
                "Improve at least one of AA, Macro-F1, "
                "Tail Accuracy, Worst-Class Accuracy "
                "without large OA collapse."
            ),

        "calibration":
            (
                "Improve at least one of ECE, Tail-ECE, NLL."
            ),

        "both_required":
            True,
    },

    "dataset_gates":
        FINAL_DEVELOPMENT_GATE,

    "cross_dataset_summary": {

        "dataset_go_count":
            dataset_go_count,

        "dataset_revise_count":
            dataset_revise_count,

        "classification_signal_dataset_count":
            classification_signal_dataset_count,

        "calibration_signal_dataset_count":
            calibration_signal_dataset_count,

        "nonzero_alpha_runs":
            0,

        "total_runs":
            9,
    },

    "failure_mode":
        FAILURE_MODE,

    "official_test_decision":
        OFFICIAL_TEST_DECISION,

    "corrected_shift_selection_manifest_sha256":
        CORRECTED_SHIFT_SELECTION_SHA,

    "corrected_temperature_manifest_sha256":
        CORRECTED_TEMPERATURE_MANIFEST_SHA,

    "exact_calibration_function_audit_sha256":
        TARGETED_FUNCTION_AUDIT_SHA,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


final_gate_text = (

    json.dumps(
        FINAL_DEVELOPMENT_GATE_OBJECT,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if FINAL_DEVELOPMENT_GATE_PATH.exists():

    existing = (
        FINAL_DEVELOPMENT_GATE_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == final_gate_text, (
        "Existing final development-gate artifact differs. STOP."
    )


else:

    FINAL_DEVELOPMENT_GATE_PATH.write_text(

        final_gate_text,

        encoding="utf-8"
    )


FINAL_DEVELOPMENT_GATE_SHA = sha256_file(
    FINAL_DEVELOPMENT_GATE_PATH
)


# ==================================================================================================
# 8. FINAL HARD SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False

assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False

assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_FINAL_VALIDATION_DEVELOPMENT_GATE_COMPLETE = True

NOTEBOOK05_DEVELOPMENT_GO = False

NOTEBOOK05_DEVELOPMENT_REVISE = True

NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED = False


# ==================================================================================================
# 9. FINAL REPORT
# ==================================================================================================

print()
print("=" * 155)
print("NOTEBOOK 05 — FINAL DEVELOPMENT DECISION")
print("=" * 155)

print(
    "Datasets with classification signal :",
    classification_signal_dataset_count,
    "/ 3"
)

print(
    "Datasets with calibration signal    :",
    calibration_signal_dataset_count,
    "/ 3"
)

print(
    "Datasets passing BOTH requirements  :",
    dataset_go_count,
    "/ 3"
)

print(
    "Nonzero-alpha LTLC selections       : 0 / 9"
)

print()

print(
    "Adaptive shift component            :",
    FAILURE_MODE[
        "adaptive_shift_component_B"
    ]
)

print(
    "Temperature calibration component   :",
    FAILURE_MODE[
        "temperature_calibration_component_C"
    ]
)

print(
    "Long-tail recognition claim         : NOT SUPPORTED"
)

print(
    "Validation calibration signal       : PRESENT"
)

print(
    "Mean Tail-ECE uniformly improved    : NO"
)

print()

print(
    "OVERALL DEVELOPMENT STATUS          :",
    OVERALL_DEVELOPMENT_STATUS
)

print(
    "OFFICIAL TEST MAY BE OPENED         :",
    NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED
)

print(
    "REQUIRED ACTION                     :",
    OFFICIAL_TEST_DECISION[
        "required_action"
    ]
)

print()

print(
    "Final development-gate artifact:"
)

print(
    FINAL_DEVELOPMENT_GATE_PATH
)

print(
    "SHA-256:"
)

print(
    FINAL_DEVELOPMENT_GATE_SHA
)

print()

print(
    "Training performed                  :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official-test dataset created       :",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created    :",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used                  :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 FINAL VALIDATION DEVELOPMENT GATE: REVISE"
)

print(
    "DO NOT OPEN OFFICIAL TEST"
)

print(
    "METHOD REVISION REQUIRED BEFORE NOTEBOOK 06"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 155)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 20A
# NONZERO-ALPHA RECOGNITION-HEADROOM DIAGNOSTIC
#
# PURPOSE
#   Diagnose WHY LTLC-Shift failed the final validation gate.
#
#   For every dataset / seed:
#       - use selected Standard-LA as the reference
#       - inspect ALL nonzero-alpha LTLC candidates
#       - find maximum possible gains in:
#           AA
#           Macro-F1
#           Tail Accuracy
#           Worst-Class Accuracy
#       - measure associated OA change
#       - count candidates satisfying the recognition side of Gate 5
#
# IMPORTANT:
#   This is DIAGNOSTIC ONLY.
#   It does NOT change the frozen selection rule.
#   It does NOT change tau/alpha grids.
#   It does NOT change the LTLC formula.
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
import numpy as np
import pandas as pd


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_FINAL_VALIDATION_DEVELOPMENT_GATE_COMPLETE is True
assert NOTEBOOK05_DEVELOPMENT_REVISE is True
assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_CORRECTED_SHIFT_GRID_PASS is True
assert NOTEBOOK05_EXACT_SHIFT_SELECTION_VERIFIED is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


DIAG_TOL = 1e-12


# ==================================================================================================
# 2. STORAGE
# ==================================================================================================

HEADROOM_ROWS = []

GATE5_CANDIDATE_ROWS = []


print("=" * 160)
print("NOTEBOOK 05 — NONZERO-ALPHA RECOGNITION-HEADROOM DIAGNOSTIC")
print("=" * 160)

print(
    "Reference:",
    "frozen selected Standard-LA per dataset/seed"
)

print(
    "Candidates:",
    "all LTLC-Shift grid points with alpha > 0"
)

print(
    "Formula changed:",
    False
)

print(
    "Selection rule changed:",
    False
)

print(
    "Official test:",
    "SEALED"
)

print("=" * 160)


# ==================================================================================================
# 3. PER-RUN DIAGNOSTIC
# ==================================================================================================

for dataset, seed in RUN_ORDER:

    identity = (
        dataset,
        seed
    )


    # ----------------------------------------------------------------------------------------------
    # Frozen selected Standard-LA baseline
    # ----------------------------------------------------------------------------------------------

    baseline = (
        CORRECTED_STANDARD_LA_SELECTION[
            identity
        ]
    )


    baseline_oa = float(
        baseline[
            "validation_oa"
        ]
    )

    baseline_aa = float(
        baseline[
            "validation_aa"
        ]
    )

    baseline_f1 = float(
        baseline[
            "validation_macro_f1"
        ]
    )

    baseline_tail = float(
        baseline[
            "validation_tail_accuracy"
        ]
    )

    baseline_worst = float(
        baseline[
            "validation_worst_class_accuracy"
        ]
    )


    # ----------------------------------------------------------------------------------------------
    # All LTLC candidates for this run
    # ----------------------------------------------------------------------------------------------

    candidate_df = (

        CORRECTED_SHIFT_GRID_DF[
            (
                CORRECTED_SHIFT_GRID_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                CORRECTED_SHIFT_GRID_DF[
                    "seed"
                ].astype(
                    int
                )
                ==
                int(
                    seed
                )
            )
            &
            (
                CORRECTED_SHIFT_GRID_DF[
                    "method"
                ]
                ==
                "LTLC_Shift"
            )
            &
            (
                CORRECTED_SHIFT_GRID_DF[
                    "alpha"
                ].astype(
                    float
                )
                >
                0.0
            )
        ]
        .copy()
        .reset_index(
            drop=True
        )
    )


    assert len(
        candidate_df
    ) == (
        len(
            TAU_GRID
        )
        *
        (
            len(
                ALPHA_GRID
            )
            -
            1
        )
    )


    # ==============================================================================================
    # 4. DELTAS VS SELECTED STANDARD LA
    # ==============================================================================================

    candidate_df[
        "delta_oa"
    ] = (
        candidate_df[
            "validation_oa"
        ]
        -
        baseline_oa
    )


    candidate_df[
        "delta_aa"
    ] = (
        candidate_df[
            "validation_aa"
        ]
        -
        baseline_aa
    )


    candidate_df[
        "delta_macro_f1"
    ] = (
        candidate_df[
            "validation_macro_f1"
        ]
        -
        baseline_f1
    )


    candidate_df[
        "delta_tail_accuracy"
    ] = (
        candidate_df[
            "validation_tail_accuracy"
        ]
        -
        baseline_tail
    )


    candidate_df[
        "delta_worst_class_accuracy"
    ] = (
        candidate_df[
            "validation_worst_class_accuracy"
        ]
        -
        baseline_worst
    )


    # ==============================================================================================
    # 5. MAXIMUM AVAILABLE GAIN IN EACH RECOGNITION METRIC
    # ==============================================================================================

    metric_map = {

        "AA":
            "delta_aa",

        "Macro_F1":
            "delta_macro_f1",

        "Tail_Accuracy":
            "delta_tail_accuracy",

        "Worst_Class_Accuracy":
            "delta_worst_class_accuracy",
    }


    best_metric_rows = {}


    for metric_name, delta_column in (
        metric_map.items()
    ):

        best_index = (
            candidate_df[
                delta_column
            ].idxmax()
        )


        best_row = (
            candidate_df.loc[
                best_index
            ]
        )


        best_metric_rows[
            metric_name
        ] = {

            "gain":
                float(
                    best_row[
                        delta_column
                    ]
                ),

            "oa_delta":
                float(
                    best_row[
                        "delta_oa"
                    ]
                ),

            "tau":
                float(
                    best_row[
                        "tau"
                    ]
                ),

            "alpha":
                float(
                    best_row[
                        "alpha"
                    ]
                ),
        }


    # ==============================================================================================
    # 6. GATE-5 RECOGNITION-SIDE CANDIDATES
    #
    # A candidate has recognition signal if it improves at least one of:
    #   AA / Macro-F1 / Tail Accuracy / Worst-Class Accuracy
    #
    # We record the OA delta rather than inventing a new OA-collapse threshold.
    # ==============================================================================================

    recognition_signal_mask = (

        (
            candidate_df[
                "delta_aa"
            ]
            >
            DIAG_TOL
        )

        |

        (
            candidate_df[
                "delta_macro_f1"
            ]
            >
            DIAG_TOL
        )

        |

        (
            candidate_df[
                "delta_tail_accuracy"
            ]
            >
            DIAG_TOL
        )

        |

        (
            candidate_df[
                "delta_worst_class_accuracy"
            ]
            >
            DIAG_TOL
        )
    )


    recognition_candidates = (
        candidate_df[
            recognition_signal_mask
        ]
        .copy()
    )


    recognition_candidate_count = int(
        len(
            recognition_candidates
        )
    )


    # ----------------------------------------------------------------------------------------------
    # Candidate with smallest OA penalty among those showing recognition signal
    # ----------------------------------------------------------------------------------------------

    if recognition_candidate_count > 0:

        best_oa_index = (
            recognition_candidates[
                "delta_oa"
            ].idxmax()
        )


        best_gate_candidate = (
            recognition_candidates.loc[
                best_oa_index
            ]
        )


        best_gate_candidate_record = {

            "dataset":
                dataset,

            "seed":
                int(
                    seed
                ),

            "tau":
                float(
                    best_gate_candidate[
                        "tau"
                    ]
                ),

            "alpha":
                float(
                    best_gate_candidate[
                        "alpha"
                    ]
                ),

            "delta_oa":
                float(
                    best_gate_candidate[
                        "delta_oa"
                    ]
                ),

            "delta_aa":
                float(
                    best_gate_candidate[
                        "delta_aa"
                    ]
                ),

            "delta_macro_f1":
                float(
                    best_gate_candidate[
                        "delta_macro_f1"
                    ]
                ),

            "delta_tail_accuracy":
                float(
                    best_gate_candidate[
                        "delta_tail_accuracy"
                    ]
                ),

            "delta_worst_class_accuracy":
                float(
                    best_gate_candidate[
                        "delta_worst_class_accuracy"
                    ]
                ),
        }


        GATE5_CANDIDATE_ROWS.append(
            best_gate_candidate_record
        )


        best_signal_oa_delta = float(
            best_gate_candidate[
                "delta_oa"
            ]
        )


    else:

        best_signal_oa_delta = float(
            "nan"
        )


    # ==============================================================================================
    # 7. HOW MANY NONZERO-ALPHA CANDIDATES ARE EXACTLY CLASSIFICATION-IDENTICAL?
    # ==============================================================================================

    classification_identical_mask = (

        np.isclose(
            candidate_df[
                "delta_oa"
            ],
            0.0,
            rtol=0.0,
            atol=DIAG_TOL
        )

        &

        np.isclose(
            candidate_df[
                "delta_aa"
            ],
            0.0,
            rtol=0.0,
            atol=DIAG_TOL
        )

        &

        np.isclose(
            candidate_df[
                "delta_macro_f1"
            ],
            0.0,
            rtol=0.0,
            atol=DIAG_TOL
        )

        &

        np.isclose(
            candidate_df[
                "delta_tail_accuracy"
            ],
            0.0,
            rtol=0.0,
            atol=DIAG_TOL
        )

        &

        np.isclose(
            candidate_df[
                "delta_worst_class_accuracy"
            ],
            0.0,
            rtol=0.0,
            atol=DIAG_TOL
        )
    )


    identical_candidate_count = int(
        classification_identical_mask.sum()
    )


    # ==============================================================================================
    # 8. STORE
    # ==============================================================================================

    HEADROOM_ROWS.append({

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "nonzero_alpha_candidates":
            int(
                len(
                    candidate_df
                )
            ),

        "recognition_signal_candidates":
            recognition_candidate_count,

        "classification_identical_candidates":
            identical_candidate_count,

        "max_delta_aa":
            best_metric_rows[
                "AA"
            ][
                "gain"
            ],

        "max_delta_aa_oa_delta":
            best_metric_rows[
                "AA"
            ][
                "oa_delta"
            ],

        "max_delta_aa_tau":
            best_metric_rows[
                "AA"
            ][
                "tau"
            ],

        "max_delta_aa_alpha":
            best_metric_rows[
                "AA"
            ][
                "alpha"
            ],

        "max_delta_macro_f1":
            best_metric_rows[
                "Macro_F1"
            ][
                "gain"
            ],

        "max_delta_macro_f1_oa_delta":
            best_metric_rows[
                "Macro_F1"
            ][
                "oa_delta"
            ],

        "max_delta_tail_accuracy":
            best_metric_rows[
                "Tail_Accuracy"
            ][
                "gain"
            ],

        "max_delta_tail_accuracy_oa_delta":
            best_metric_rows[
                "Tail_Accuracy"
            ][
                "oa_delta"
            ],

        "max_delta_worst_class_accuracy":
            best_metric_rows[
                "Worst_Class_Accuracy"
            ][
                "gain"
            ],

        "max_delta_worst_class_accuracy_oa_delta":
            best_metric_rows[
                "Worst_Class_Accuracy"
            ][
                "oa_delta"
            ],

        "best_recognition_signal_oa_delta":
            best_signal_oa_delta,
    })


    # ==============================================================================================
    # 9. REPORT
    # ==============================================================================================

    print()

    print("-" * 160)

    print(
        f"{dataset.upper()} — SEED {seed}"
    )

    print("-" * 160)


    print(
        "Nonzero-alpha candidates       :",
        len(
            candidate_df
        )
    )


    print(
        "Recognition-signal candidates :",
        recognition_candidate_count
    )


    print(
        "Classification-identical      :",
        identical_candidate_count
    )


    print()

    print(
        "Best possible ΔAA             :",
        f"{best_metric_rows['AA']['gain']:+.9e}",
        "| OA Δ:",
        f"{best_metric_rows['AA']['oa_delta']:+.9e}",
        "| tau:",
        f"{best_metric_rows['AA']['tau']:.2f}",
        "| alpha:",
        f"{best_metric_rows['AA']['alpha']:.2f}"
    )


    print(
        "Best possible ΔMacro-F1       :",
        f"{best_metric_rows['Macro_F1']['gain']:+.9e}",
        "| OA Δ:",
        f"{best_metric_rows['Macro_F1']['oa_delta']:+.9e}"
    )


    print(
        "Best possible ΔTail Accuracy  :",
        f"{best_metric_rows['Tail_Accuracy']['gain']:+.9e}",
        "| OA Δ:",
        f"{best_metric_rows['Tail_Accuracy']['oa_delta']:+.9e}"
    )


    print(
        "Best possible ΔWorst-Class Acc:",
        f"{best_metric_rows['Worst_Class_Accuracy']['gain']:+.9e}",
        "| OA Δ:",
        f"{best_metric_rows['Worst_Class_Accuracy']['oa_delta']:+.9e}"
    )


    if recognition_candidate_count > 0:

        print(
            "Best OA Δ among signal candidates:",
            f"{best_signal_oa_delta:+.9e}"
        )


    else:

        print(
            "Best OA Δ among signal candidates:",
            "N/A — NO RECOGNITION-SIGNAL CANDIDATE"
        )


# ==================================================================================================
# 10. DATAFRAMES
# ==================================================================================================

LTLC_HEADROOM_DIAGNOSTIC_DF = pd.DataFrame(
    HEADROOM_ROWS
)


LTLC_GATE5_CANDIDATE_DF = pd.DataFrame(
    GATE5_CANDIDATE_ROWS
)


assert len(
    LTLC_HEADROOM_DIAGNOSTIC_DF
) == 9


# ==================================================================================================
# 11. DATASET-LEVEL DIAGNOSIS
# ==================================================================================================

DATASET_HEADROOM_DIAGNOSIS = {}


print()
print("=" * 160)
print("DATASET-LEVEL ADAPTIVE-SHIFT DIAGNOSIS")
print("=" * 160)


for dataset in [

    "Pingan",
    "Qingyun",
    "Tangdaowan",

]:

    subset = (

        LTLC_HEADROOM_DIAGNOSTIC_DF[
            LTLC_HEADROOM_DIAGNOSTIC_DF[
                "dataset"
            ]
            ==
            dataset
        ]
    )


    assert len(
        subset
    ) == 3


    total_signal_candidates = int(
        subset[
            "recognition_signal_candidates"
        ].sum()
    )


    seeds_with_any_signal = int(

        (
            subset[
                "recognition_signal_candidates"
            ]
            >
            0
        ).sum()
    )


    max_aa_gain = float(
        subset[
            "max_delta_aa"
        ].max()
    )


    max_f1_gain = float(
        subset[
            "max_delta_macro_f1"
        ].max()
    )


    max_tail_gain = float(
        subset[
            "max_delta_tail_accuracy"
        ].max()
    )


    max_worst_gain = float(
        subset[
            "max_delta_worst_class_accuracy"
        ].max()
    )


    any_recognition_headroom = bool(

        max(
            max_aa_gain,
            max_f1_gain,
            max_tail_gain,
            max_worst_gain
        )
        >
        DIAG_TOL
    )


    DATASET_HEADROOM_DIAGNOSIS[
        dataset
    ] = {

        "seeds_with_any_nonzero_alpha_recognition_signal":
            seeds_with_any_signal,

        "total_nonzero_alpha_recognition_signal_candidates":
            total_signal_candidates,

        "maximum_available_gains": {

            "AA":
                max_aa_gain,

            "Macro_F1":
                max_f1_gain,

            "Tail_Accuracy":
                max_tail_gain,

            "Worst_Class_Accuracy":
                max_worst_gain,
        },

        "any_recognition_headroom":
            any_recognition_headroom,
    }


    print()

    print(
        dataset
    )


    print(
        "  Seeds with any signal candidate:",
        seeds_with_any_signal,
        "/ 3"
    )


    print(
        "  Total signal candidates        :",
        total_signal_candidates
    )


    print(
        "  Maximum available ΔAA          :",
        f"{max_aa_gain:+.9e}"
    )


    print(
        "  Maximum available ΔMacro-F1    :",
        f"{max_f1_gain:+.9e}"
    )


    print(
        "  Maximum available ΔTail Acc    :",
        f"{max_tail_gain:+.9e}"
    )


    print(
        "  Maximum available ΔWorst       :",
        f"{max_worst_gain:+.9e}"
    )


    print(
        "  Any recognition headroom       :",
        any_recognition_headroom
    )


# ==================================================================================================
# 12. CROSS-DATASET DIAGNOSIS
# ==================================================================================================

datasets_with_any_headroom = [

    dataset

    for dataset, result in
    DATASET_HEADROOM_DIAGNOSIS.items()

    if result[
        "any_recognition_headroom"
    ]
]


total_signal_candidates = int(

    LTLC_HEADROOM_DIAGNOSTIC_DF[
        "recognition_signal_candidates"
    ].sum()
)


seeds_with_any_signal = int(

    (
        LTLC_HEADROOM_DIAGNOSTIC_DF[
            "recognition_signal_candidates"
        ]
        >
        0
    ).sum()
)


# ==================================================================================================
# 13. INTERPRETATION
# ==================================================================================================

if len(
    datasets_with_any_headroom
) == 0:

    DIAGNOSTIC_CONCLUSION = (
        "FORMULATION_HAS_NO_RECOGNITION_HEADROOM_IN_FROZEN_GRID"
    )


    NEXT_REVISION_TARGET = (
        "LTLC_SHIFT_FORMULATION"
    )


else:

    DIAGNOSTIC_CONCLUSION = (
        "NONZERO_ALPHA_RECOGNITION_HEADROOM_EXISTS_BUT_LOCKED_SELECTION_REJECTS_IT"
    )


    NEXT_REVISION_TARGET = (
        "SELECTION_OBJECTIVE_OR_FORMULATION_AFTER_TRADEOFF_AUDIT"
    )


# ==================================================================================================
# 14. SAVE DIAGNOSTIC CSV
# ==================================================================================================

HEADROOM_DIAGNOSTIC_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_ltlc_nonzero_alpha_recognition_headroom.csv"
)


headroom_csv_text = (

    LTLC_HEADROOM_DIAGNOSTIC_DF.to_csv(
        index=False,
        float_format="%.17g"
    )
)


if HEADROOM_DIAGNOSTIC_PATH.exists():

    existing = (
        HEADROOM_DIAGNOSTIC_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == headroom_csv_text, (
        "Existing headroom diagnostic differs. STOP."
    )


else:

    HEADROOM_DIAGNOSTIC_PATH.write_text(

        headroom_csv_text,

        encoding="utf-8"
    )


HEADROOM_DIAGNOSTIC_SHA = sha256_file(
    HEADROOM_DIAGNOSTIC_PATH
)


# ==================================================================================================
# 15. SAVE DIAGNOSTIC MANIFEST
# ==================================================================================================

HEADROOM_MANIFEST_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_ltlc_nonzero_alpha_recognition_headroom_manifest.json"
)


HEADROOM_MANIFEST = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "purpose":
        (
            "Validation-only diagnostic of all nonzero-alpha "
            "LTLC-Shift candidates after final REVISE gate."
        ),

    "formula_changed":
        False,

    "selection_rule_changed":
        False,

    "tau_grid_changed":
        False,

    "alpha_grid_changed":
        False,

    "reference":
        "selected_Standard_LA_per_seed",

    "diagnosis_by_dataset":
        DATASET_HEADROOM_DIAGNOSIS,

    "cross_dataset": {

        "datasets_with_any_recognition_headroom":
            datasets_with_any_headroom,

        "dataset_count_with_headroom":
            len(
                datasets_with_any_headroom
            ),

        "seeds_with_any_signal":
            seeds_with_any_signal,

        "total_nonzero_alpha_signal_candidates":
            total_signal_candidates,
    },

    "diagnostic_conclusion":
        DIAGNOSTIC_CONCLUSION,

    "recommended_next_revision_target":
        NEXT_REVISION_TARGET,

    "source_corrected_shift_grid_sha256":
        CORRECTED_SHIFT_GRID_SHA,

    "source_final_development_gate_sha256":
        FINAL_DEVELOPMENT_GATE_SHA,

    "diagnostic_csv":
        str(
            HEADROOM_DIAGNOSTIC_PATH.relative_to(
                ROOT
            )
        ),

    "diagnostic_csv_sha256":
        HEADROOM_DIAGNOSTIC_SHA,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


manifest_text = (

    json.dumps(
        HEADROOM_MANIFEST,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if HEADROOM_MANIFEST_PATH.exists():

    existing = (
        HEADROOM_MANIFEST_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == manifest_text, (
        "Existing headroom manifest differs. STOP."
    )


else:

    HEADROOM_MANIFEST_PATH.write_text(

        manifest_text,

        encoding="utf-8"
    )


HEADROOM_MANIFEST_SHA = sha256_file(
    HEADROOM_MANIFEST_PATH
)


# ==================================================================================================
# 16. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_LTLC_HEADROOM_DIAGNOSTIC_PASS = True


print()
print("=" * 160)
print("NOTEBOOK 05 — ADAPTIVE-SHIFT FAILURE DIAGNOSIS")
print("=" * 160)

print(
    "Datasets with recognition headroom :",
    len(
        datasets_with_any_headroom
    ),
    "/ 3"
)

print(
    "Seeds with recognition headroom    :",
    seeds_with_any_signal,
    "/ 9"
)

print(
    "Nonzero-alpha signal candidates    :",
    total_signal_candidates
)

print()

print(
    "DIAGNOSTIC CONCLUSION              :",
    DIAGNOSTIC_CONCLUSION
)

print(
    "NEXT REVISION TARGET               :",
    NEXT_REVISION_TARGET
)

print()

print(
    "Diagnostic CSV:"
)

print(
    HEADROOM_DIAGNOSTIC_PATH
)

print(
    "SHA-256:"
)

print(
    HEADROOM_DIAGNOSTIC_SHA
)

print()

print(
    "Diagnostic manifest:"
)

print(
    HEADROOM_MANIFEST_PATH
)

print(
    "SHA-256:"
)

print(
    HEADROOM_MANIFEST_SHA
)

print()

print(
    "Formula changed                    : False"
)

print(
    "Selection rule changed             : False"
)

print(
    "Training performed                 :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used                 :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 LTLC HEADROOM DIAGNOSTIC: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 160)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 20B
# QINGYUN SEED-3407 NONZERO-ALPHA TRADE-OFF + PER-CLASS DIAGNOSTIC
#
# PURPOSE
#   Investigate the ONLY corrected best-AA run where nonzero-alpha LTLC
#   produced any recognition signal.
#
# QUESTIONS:
#   1. Which of the 14 candidates improve Macro-F1?
#   2. What AA/OA trade-off accompanies that improvement?
#   3. Does Tail-class F1 improve even though Tail Accuracy does not?
#   4. Which classes actually gain/lose?
#   5. Do candidates represent genuinely different prediction patterns?
#
# DIAGNOSTIC ONLY
# FORMULA UNCHANGED
# SELECTION RULE UNCHANGED
# OFFICIAL TEST SEALED
# ==================================================================================================

import hashlib
import json

import numpy as np
import pandas as pd

from sklearn.metrics import precision_recall_fscore_support


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_LTLC_HEADROOM_DIAGNOSTIC_PASS is True

assert NOTEBOOK05_FINAL_VALIDATION_DEVELOPMENT_GATE_COMPLETE is True
assert NOTEBOOK05_DEVELOPMENT_REVISE is True
assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


DIAG_TOL = 1e-12

DATASET = "Qingyun"
SEED = 3407

IDENTITY = (
    DATASET,
    SEED
)


# ==================================================================================================
# 2. LOAD FROZEN DATASET INFORMATION
# ==================================================================================================

payload = (
    PRIOR_RARITY_BY_DATASET[
        DATASET
    ]
)


prior = np.asarray(
    payload[
        "prior"
    ],
    dtype=np.float64
)


rarity = np.asarray(
    payload[
        "rarity"
    ],
    dtype=np.float64
)


frequency_groups = np.asarray(
    payload[
        "frequency_group"
    ],
    dtype=object
)


train_counts = np.asarray(
    payload[
        "train_counts"
    ],
    dtype=np.int64
)


K = int(
    payload[
        "class_count"
    ]
)


assert K == 6


validation = load_bestAA_validation_output(

    FINAL_BEST_AA_VALIDATION_OUTPUTS[
        IDENTITY
    ]
)


ce_logits = np.asarray(
    validation[
        "logits"
    ],
    dtype=np.float64
)


labels = np.asarray(
    validation[
        "labels"
    ],
    dtype=np.int64
)


# ==================================================================================================
# 3. SELECTED STANDARD-LA REFERENCE
# ==================================================================================================

baseline_selection = (
    CORRECTED_STANDARD_LA_SELECTION[
        IDENTITY
    ]
)


BASELINE_TAU = float(
    baseline_selection[
        "tau"
    ]
)


baseline_logits = standard_la_logits(

    ce_logits,
    prior,
    tau=BASELINE_TAU,
    eps=RARITY_EPSILON
)


baseline_predictions = np.argmax(
    baseline_logits,
    axis=1
)


baseline_eval = evaluate_logits(

    baseline_logits,
    labels,
    frequency_groups
)


assert BASELINE_TAU == 1.0


# ==================================================================================================
# 4. PER-CLASS PRECISION / RECALL / F1 HELPER
# ==================================================================================================

def per_class_prf(
    labels,
    predictions,
    k
):

    precision, recall, f1, support = (
        precision_recall_fscore_support(

            labels,
            predictions,

            labels=np.arange(
                k
            ),

            average=None,

            zero_division=0
        )
    )


    return {

        "precision":
            np.asarray(
                precision,
                dtype=np.float64
            ),

        "recall":
            np.asarray(
                recall,
                dtype=np.float64
            ),

        "f1":
            np.asarray(
                f1,
                dtype=np.float64
            ),

        "support":
            np.asarray(
                support,
                dtype=np.int64
            ),
    }


baseline_prf = per_class_prf(

    labels,
    baseline_predictions,
    K
)


# Recall must equal per-class accuracy in a single-label classification problem.
baseline_support = baseline_prf[
    "support"
]


assert (
    baseline_support
    >
    0
).all()


# ==================================================================================================
# 5. RECOVER THE 14 RECOGNITION-SIGNAL CANDIDATES
# ==================================================================================================

grid = (

    CORRECTED_SHIFT_GRID_DF[
        (
            CORRECTED_SHIFT_GRID_DF[
                "dataset"
            ]
            ==
            DATASET
        )
        &
        (
            CORRECTED_SHIFT_GRID_DF[
                "seed"
            ].astype(
                int
            )
            ==
            SEED
        )
        &
        (
            CORRECTED_SHIFT_GRID_DF[
                "method"
            ]
            ==
            "LTLC_Shift"
        )
        &
        (
            CORRECTED_SHIFT_GRID_DF[
                "alpha"
            ].astype(
                float
            )
            >
            0.0
        )
    ]
    .copy()
    .reset_index(
        drop=True
    )
)


assert len(
    grid
) == 54


grid[
    "delta_oa"
] = (
    grid[
        "validation_oa"
    ]
    -
    float(
        baseline_selection[
            "validation_oa"
        ]
    )
)


grid[
    "delta_aa"
] = (
    grid[
        "validation_aa"
    ]
    -
    float(
        baseline_selection[
            "validation_aa"
        ]
    )
)


grid[
    "delta_macro_f1"
] = (
    grid[
        "validation_macro_f1"
    ]
    -
    float(
        baseline_selection[
            "validation_macro_f1"
        ]
    )
)


grid[
    "delta_tail_accuracy"
] = (
    grid[
        "validation_tail_accuracy"
    ]
    -
    float(
        baseline_selection[
            "validation_tail_accuracy"
        ]
    )
)


grid[
    "delta_worst_class_accuracy"
] = (
    grid[
        "validation_worst_class_accuracy"
    ]
    -
    float(
        baseline_selection[
            "validation_worst_class_accuracy"
        ]
    )
)


signal_mask = (

    (
        grid[
            "delta_aa"
        ]
        >
        DIAG_TOL
    )

    |

    (
        grid[
            "delta_macro_f1"
        ]
        >
        DIAG_TOL
    )

    |

    (
        grid[
            "delta_tail_accuracy"
        ]
        >
        DIAG_TOL
    )

    |

    (
        grid[
            "delta_worst_class_accuracy"
        ]
        >
        DIAG_TOL
    )
)


signal_grid = (
    grid[
        signal_mask
    ]
    .copy()
    .reset_index(
        drop=True
    )
)


assert len(
    signal_grid
) == 14


# ==================================================================================================
# 6. RECOMPUTE EACH SIGNAL CANDIDATE AT PREDICTION / CLASS LEVEL
# ==================================================================================================

candidate_rows = []

candidate_class_rows = []

prediction_hashes = []


for _, row in signal_grid.iterrows():

    tau = float(
        row[
            "tau"
        ]
    )


    alpha = float(
        row[
            "alpha"
        ]
    )


    candidate_logits = ltlc_shift_logits(

        ce_logits,
        prior,
        rarity,

        tau=tau,
        alpha=alpha,

        eps=RARITY_EPSILON
    )


    candidate_predictions = np.argmax(

        candidate_logits,

        axis=1
    )


    candidate_eval = evaluate_logits(

        candidate_logits,
        labels,
        frequency_groups
    )


    candidate_prf = per_class_prf(

        labels,
        candidate_predictions,
        K
    )


    # ----------------------------------------------------------------------------------------------
    # Prediction changes
    # ----------------------------------------------------------------------------------------------

    prediction_changed = (
        candidate_predictions
        !=
        baseline_predictions
    )


    baseline_correct = (
        baseline_predictions
        ==
        labels
    )


    candidate_correct = (
        candidate_predictions
        ==
        labels
    )


    errors_fixed = int(

        (
            (~baseline_correct)
            &
            candidate_correct
        ).sum()
    )


    new_errors = int(

        (
            baseline_correct
            &
            (~candidate_correct)
        ).sum()
    )


    prediction_change_count = int(
        prediction_changed.sum()
    )


    prediction_hash = hashlib.sha256(

        np.asarray(
            candidate_predictions,
            dtype=np.int64
        ).tobytes()

    ).hexdigest()


    prediction_hashes.append(
        prediction_hash
    )


    # ----------------------------------------------------------------------------------------------
    # F1 group deltas
    # ----------------------------------------------------------------------------------------------

    class_f1_delta = (

        candidate_prf[
            "f1"
        ]

        -

        baseline_prf[
            "f1"
        ]
    )


    class_precision_delta = (

        candidate_prf[
            "precision"
        ]

        -

        baseline_prf[
            "precision"
        ]
    )


    class_recall_delta = (

        candidate_prf[
            "recall"
        ]

        -

        baseline_prf[
            "recall"
        ]
    )


    group_f1_delta = {}


    for group in [
        "Tail",
        "Medium",
        "Head",
    ]:

        group_mask = (
            frequency_groups
            ==
            group
        )


        group_f1_delta[
            group
        ] = float(

            class_f1_delta[
                group_mask
            ].mean()
        )


    # ----------------------------------------------------------------------------------------------
    # Candidate summary
    # ----------------------------------------------------------------------------------------------

    candidate_rows.append({

        "tau":
            tau,

        "alpha":
            alpha,

        "delta_oa":
            float(
                candidate_eval[
                    "oa"
                ]
                -
                baseline_eval[
                    "oa"
                ]
            ),

        "delta_aa":
            float(
                candidate_eval[
                    "aa"
                ]
                -
                baseline_eval[
                    "aa"
                ]
            ),

        "delta_macro_f1":
            float(
                candidate_eval[
                    "macro_f1"
                ]
                -
                baseline_eval[
                    "macro_f1"
                ]
            ),

        "delta_tail_accuracy":
            float(
                candidate_eval[
                    "tail_accuracy"
                ]
                -
                baseline_eval[
                    "tail_accuracy"
                ]
            ),

        "delta_medium_accuracy":
            float(
                candidate_eval[
                    "medium_accuracy"
                ]
                -
                baseline_eval[
                    "medium_accuracy"
                ]
            ),

        "delta_head_accuracy":
            float(
                candidate_eval[
                    "head_accuracy"
                ]
                -
                baseline_eval[
                    "head_accuracy"
                ]
            ),

        "delta_worst_class_accuracy":
            float(
                candidate_eval[
                    "worst_class_accuracy"
                ]
                -
                baseline_eval[
                    "worst_class_accuracy"
                ]
            ),

        "delta_nll":
            float(
                candidate_eval[
                    "nll"
                ]
                -
                baseline_eval[
                    "nll"
                ]
            ),

        "tail_mean_f1_delta":
            group_f1_delta[
                "Tail"
            ],

        "medium_mean_f1_delta":
            group_f1_delta[
                "Medium"
            ],

        "head_mean_f1_delta":
            group_f1_delta[
                "Head"
            ],

        "prediction_changes":
            prediction_change_count,

        "errors_fixed":
            errors_fixed,

        "new_errors":
            new_errors,

        "net_correct_gain":
            errors_fixed
            -
            new_errors,

        "prediction_sha256":
            prediction_hash,
    })


    # ----------------------------------------------------------------------------------------------
    # Per-class detail
    # ----------------------------------------------------------------------------------------------

    for class_index in range(
        K
    ):

        candidate_class_rows.append({

            "tau":
                tau,

            "alpha":
                alpha,

            "internal_class":
                int(
                    class_index
                ),

            "public_class":
                int(
                    class_index + 1
                ),

            "frequency_group":
                str(
                    frequency_groups[
                        class_index
                    ]
                ),

            "train_count":
                int(
                    train_counts[
                        class_index
                    ]
                ),

            "validation_support":
                int(
                    baseline_support[
                        class_index
                    ]
                ),

            "baseline_precision":
                float(
                    baseline_prf[
                        "precision"
                    ][
                        class_index
                    ]
                ),

            "candidate_precision":
                float(
                    candidate_prf[
                        "precision"
                    ][
                        class_index
                    ]
                ),

            "delta_precision":
                float(
                    class_precision_delta[
                        class_index
                    ]
                ),

            "baseline_recall":
                float(
                    baseline_prf[
                        "recall"
                    ][
                        class_index
                    ]
                ),

            "candidate_recall":
                float(
                    candidate_prf[
                        "recall"
                    ][
                        class_index
                    ]
                ),

            "delta_recall":
                float(
                    class_recall_delta[
                        class_index
                    ]
                ),

            "baseline_f1":
                float(
                    baseline_prf[
                        "f1"
                    ][
                        class_index
                    ]
                ),

            "candidate_f1":
                float(
                    candidate_prf[
                        "f1"
                    ][
                        class_index
                    ]
                ),

            "delta_f1":
                float(
                    class_f1_delta[
                        class_index
                    ]
                ),
        })


TRADEOFF_DF = pd.DataFrame(
    candidate_rows
)


PER_CLASS_TRADEOFF_DF = pd.DataFrame(
    candidate_class_rows
)


assert len(
    TRADEOFF_DF
) == 14


assert len(
    PER_CLASS_TRADEOFF_DF
) == (
    14
    *
    K
)


# ==================================================================================================
# 7. VERIFY GRID METRICS MATCH FRESH RECOMPUTATION
# ==================================================================================================

for _, original_row in signal_grid.iterrows():

    tau = float(
        original_row[
            "tau"
        ]
    )

    alpha = float(
        original_row[
            "alpha"
        ]
    )


    match = TRADEOFF_DF[
        np.isclose(
            TRADEOFF_DF[
                "tau"
            ],
            tau,
            rtol=0.0,
            atol=1e-15
        )
        &
        np.isclose(
            TRADEOFF_DF[
                "alpha"
            ],
            alpha,
            rtol=0.0,
            atol=1e-15
        )
    ]


    assert len(
        match
    ) == 1


    recomputed = match.iloc[
        0
    ]


    for column in [

        "delta_oa",
        "delta_aa",
        "delta_macro_f1",
        "delta_tail_accuracy",
        "delta_worst_class_accuracy",

    ]:

        assert np.isclose(

            float(
                original_row[
                    column
                ]
            ),

            float(
                recomputed[
                    column
                ]
            ),

            rtol=0.0,

            atol=1e-12
        )


# ==================================================================================================
# 8. PRINT ALL 14 SIGNAL CANDIDATES
# ==================================================================================================

DISPLAY_COLUMNS = [

    "tau",
    "alpha",

    "delta_oa",
    "delta_aa",
    "delta_macro_f1",

    "delta_tail_accuracy",
    "delta_worst_class_accuracy",

    "tail_mean_f1_delta",
    "medium_mean_f1_delta",
    "head_mean_f1_delta",

    "prediction_changes",
    "errors_fixed",
    "new_errors",
    "net_correct_gain",

    "delta_nll",
]


TRADEOFF_SORTED_DF = (

    TRADEOFF_DF
    .sort_values(

        by=[
            "delta_macro_f1",
            "delta_oa",
        ],

        ascending=[
            False,
            False,
        ]
    )
    .reset_index(
        drop=True
    )
)


print("=" * 175)
print("NOTEBOOK 05 — QINGYUN SEED-3407 NONZERO-ALPHA TRADE-OFF")
print("=" * 175)

print(
    "Selected Standard-LA tau:",
    BASELINE_TAU
)

print(
    "Signal candidates:",
    len(
        TRADEOFF_SORTED_DF
    )
)

print(
    "Official test:",
    "SEALED"
)

print()

print(
    TRADEOFF_SORTED_DF[
        DISPLAY_COLUMNS
    ].to_string(
        index=False
    )
)


# ==================================================================================================
# 9. UNIQUE PREDICTION PATTERNS
# ==================================================================================================

unique_prediction_patterns = int(
    TRADEOFF_DF[
        "prediction_sha256"
    ].nunique()
)


print()
print("=" * 175)
print("PREDICTION-PATTERN DIAGNOSTIC")
print("=" * 175)

print(
    "Signal candidates          :",
    len(
        TRADEOFF_DF
    )
)

print(
    "Unique prediction patterns :",
    unique_prediction_patterns
)


pattern_counts = (

    TRADEOFF_DF
    .groupby(
        "prediction_sha256"
    )
    .size()
    .sort_values(
        ascending=False
    )
)


print()

print(
    "Candidates per prediction pattern:"
)

print(
    pattern_counts.to_string()
)


# ==================================================================================================
# 10. MAXIMUM MACRO-F1 CANDIDATE SET
#
# Do not invent a tie-break.
# Inspect every candidate tied at maximum Macro-F1 improvement.
# ==================================================================================================

MAX_MACRO_F1_GAIN = float(

    TRADEOFF_DF[
        "delta_macro_f1"
    ].max()
)


MAX_F1_CANDIDATES = (

    TRADEOFF_DF[
        np.isclose(

            TRADEOFF_DF[
                "delta_macro_f1"
            ],

            MAX_MACRO_F1_GAIN,

            rtol=0.0,

            atol=DIAG_TOL
        )
    ]
    .copy()
    .sort_values(
        [
            "tau",
            "alpha",
        ]
    )
    .reset_index(
        drop=True
    )
)


print()
print("=" * 175)
print("MAXIMUM MACRO-F1 CANDIDATE SET")
print("=" * 175)

print(
    "Maximum ΔMacro-F1:",
    f"{MAX_MACRO_F1_GAIN:+.12e}"
)

print(
    "Candidates tied at maximum:",
    len(
        MAX_F1_CANDIDATES
    )
)

print()

print(
    MAX_F1_CANDIDATES[
        DISPLAY_COLUMNS
    ].to_string(
        index=False
    )
)


# ==================================================================================================
# 11. PER-CLASS DETAILS FOR ALL MAX-F1 CANDIDATES
# ==================================================================================================

print()
print("=" * 175)
print("PER-CLASS EFFECTS FOR MAXIMUM-MACRO-F1 CANDIDATE(S)")
print("=" * 175)


for _, candidate in MAX_F1_CANDIDATES.iterrows():

    tau = float(
        candidate[
            "tau"
        ]
    )

    alpha = float(
        candidate[
            "alpha"
        ]
    )


    class_subset = (

        PER_CLASS_TRADEOFF_DF[
            np.isclose(
                PER_CLASS_TRADEOFF_DF[
                    "tau"
                ],
                tau,
                rtol=0.0,
                atol=1e-15
            )
            &
            np.isclose(
                PER_CLASS_TRADEOFF_DF[
                    "alpha"
                ],
                alpha,
                rtol=0.0,
                atol=1e-15
            )
        ]
        .copy()
        .sort_values(
            "public_class"
        )
    )


    assert len(
        class_subset
    ) == K


    print()

    print("-" * 175)

    print(
        f"tau={tau:.2f} | alpha={alpha:.2f}"
    )

    print("-" * 175)


    print(
        class_subset[
            [
                "public_class",
                "frequency_group",
                "train_count",
                "validation_support",

                "baseline_precision",
                "candidate_precision",
                "delta_precision",

                "baseline_recall",
                "candidate_recall",
                "delta_recall",

                "baseline_f1",
                "candidate_f1",
                "delta_f1",
            ]
        ].to_string(
            index=False
        )
    )


# ==================================================================================================
# 12. GROUP-LEVEL F1 DIAGNOSIS ACROSS ALL 14 SIGNAL CANDIDATES
# ==================================================================================================

GROUP_F1_DIAGNOSIS = {}


for group, column in [

    (
        "Tail",
        "tail_mean_f1_delta"
    ),

    (
        "Medium",
        "medium_mean_f1_delta"
    ),

    (
        "Head",
        "head_mean_f1_delta"
    ),

]:

    values = TRADEOFF_DF[
        column
    ].to_numpy(
        dtype=np.float64
    )


    GROUP_F1_DIAGNOSIS[
        group
    ] = {

        "maximum_mean_f1_gain":
            float(
                np.max(
                    values
                )
            ),

        "minimum_mean_f1_change":
            float(
                np.min(
                    values
                )
            ),

        "candidates_with_positive_mean_f1_gain":
            int(
                (
                    values
                    >
                    DIAG_TOL
                ).sum()
            ),
    }


print()
print("=" * 175)
print("FREQUENCY-GROUP F1 DIAGNOSIS")
print("=" * 175)


for group in [
    "Tail",
    "Medium",
    "Head",
]:

    info = (
        GROUP_F1_DIAGNOSIS[
            group
        ]
    )


    print()

    print(
        group
    )

    print(
        "  Max mean F1 gain       :",
        f"{info['maximum_mean_f1_gain']:+.12e}"
    )

    print(
        "  Min mean F1 change     :",
        f"{info['minimum_mean_f1_change']:+.12e}"
    )

    print(
        "  Positive-gain candidates:",
        info[
            "candidates_with_positive_mean_f1_gain"
        ],
        "/ 14"
    )


# ==================================================================================================
# 13. DETERMINE WHETHER THE SIGNAL IS ACTUALLY TAIL-RELATED
# ==================================================================================================

TAIL_F1_SIGNAL = bool(

    GROUP_F1_DIAGNOSIS[
        "Tail"
    ][
        "maximum_mean_f1_gain"
    ]
    >
    DIAG_TOL
)


TAIL_RECALL_SIGNAL = bool(

    (
        TRADEOFF_DF[
            "delta_tail_accuracy"
        ]
        >
        DIAG_TOL
    ).any()
)


AA_SIGNAL = bool(

    (
        TRADEOFF_DF[
            "delta_aa"
        ]
        >
        DIAG_TOL
    ).any()
)


WORST_CLASS_SIGNAL = bool(

    (
        TRADEOFF_DF[
            "delta_worst_class_accuracy"
        ]
        >
        DIAG_TOL
    ).any()
)


MACRO_F1_SIGNAL = bool(

    (
        TRADEOFF_DF[
            "delta_macro_f1"
        ]
        >
        DIAG_TOL
    ).any()
)


# ==================================================================================================
# 14. DIAGNOSTIC INTERPRETATION
#
# This does NOT revise anything yet.
# ==================================================================================================

if (
    TAIL_F1_SIGNAL
    or
    TAIL_RECALL_SIGNAL
    or
    WORST_CLASS_SIGNAL
):

    TRADEOFF_INTERPRETATION = (
        "LOCAL_LONG_TAIL_RECOGNITION_SIGNAL_PRESENT_IN_QINGYUN_SEED3407"
    )


    NEXT_AUDIT_TARGET = (
        "ROBUSTNESS_AND_SELECTION_TRADEOFF"
    )


elif MACRO_F1_SIGNAL:

    TRADEOFF_INTERPRETATION = (
        "MACRO_F1_ONLY_SIGNAL_WITHOUT_TAIL_RECALL_OR_WORST_CLASS_GAIN"
    )


    NEXT_AUDIT_TARGET = (
        "CLASS_LEVEL_ORIGIN_AND_FORMULATION_ROBUSTNESS"
    )


else:

    TRADEOFF_INTERPRETATION = (
        "NO_RECOGNITION_SIGNAL_AFTER_EXACT_RECOMPUTATION"
    )


    NEXT_AUDIT_TARGET = (
        "LTLC_SHIFT_FORMULATION"
    )


# ==================================================================================================
# 15. SAVE DIAGNOSTIC CSVs
# ==================================================================================================

TRADEOFF_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_qingyun_seed3407_nonzero_alpha_tradeoff.csv"
)


PER_CLASS_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_qingyun_seed3407_nonzero_alpha_per_class_tradeoff.csv"
)


tradeoff_text = TRADEOFF_DF.to_csv(
    index=False,
    float_format="%.17g"
)


per_class_text = PER_CLASS_TRADEOFF_DF.to_csv(
    index=False,
    float_format="%.17g"
)


if TRADEOFF_PATH.exists():

    assert (
        TRADEOFF_PATH.read_text(
            encoding="utf-8"
        )
        ==
        tradeoff_text
    )

else:

    TRADEOFF_PATH.write_text(
        tradeoff_text,
        encoding="utf-8"
    )


if PER_CLASS_PATH.exists():

    assert (
        PER_CLASS_PATH.read_text(
            encoding="utf-8"
        )
        ==
        per_class_text
    )

else:

    PER_CLASS_PATH.write_text(
        per_class_text,
        encoding="utf-8"
    )


TRADEOFF_SHA = sha256_file(
    TRADEOFF_PATH
)


PER_CLASS_SHA = sha256_file(
    PER_CLASS_PATH
)


# ==================================================================================================
# 16. SAVE MANIFEST
# ==================================================================================================

TRADEOFF_MANIFEST_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_qingyun_seed3407_nonzero_alpha_tradeoff_manifest.json"
)


TRADEOFF_MANIFEST = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "purpose":
        (
            "Validation-only trade-off and per-class diagnosis "
            "of the only run with nonzero-alpha recognition headroom."
        ),

    "dataset":
        DATASET,

    "seed":
        SEED,

    "baseline":
        {
            "method":
                "Standard_LA",

            "tau":
                BASELINE_TAU,
        },

    "signal_candidate_count":
        int(
            len(
                TRADEOFF_DF
            )
        ),

    "unique_prediction_patterns":
        unique_prediction_patterns,

    "maximum_macro_f1_gain":
        MAX_MACRO_F1_GAIN,

    "max_macro_f1_candidate_count":
        int(
            len(
                MAX_F1_CANDIDATES
            )
        ),

    "frequency_group_f1_diagnosis":
        GROUP_F1_DIAGNOSIS,

    "signals":
        {
            "aa":
                AA_SIGNAL,

            "macro_f1":
                MACRO_F1_SIGNAL,

            "tail_accuracy":
                TAIL_RECALL_SIGNAL,

            "tail_mean_f1":
                TAIL_F1_SIGNAL,

            "worst_class_accuracy":
                WORST_CLASS_SIGNAL,
        },

    "interpretation":
        TRADEOFF_INTERPRETATION,

    "recommended_next_audit_target":
        NEXT_AUDIT_TARGET,

    "formula_changed":
        False,

    "selection_rule_changed":
        False,

    "official_test_used":
        False,

    "training_performed":
        False,

    "tradeoff_csv_sha256":
        TRADEOFF_SHA,

    "per_class_csv_sha256":
        PER_CLASS_SHA,

    "source_headroom_manifest_sha256":
        HEADROOM_MANIFEST_SHA,
}


manifest_text = (

    json.dumps(
        TRADEOFF_MANIFEST,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if TRADEOFF_MANIFEST_PATH.exists():

    assert (
        TRADEOFF_MANIFEST_PATH.read_text(
            encoding="utf-8"
        )
        ==
        manifest_text
    )

else:

    TRADEOFF_MANIFEST_PATH.write_text(
        manifest_text,
        encoding="utf-8"
    )


TRADEOFF_MANIFEST_SHA = sha256_file(
    TRADEOFF_MANIFEST_PATH
)


# ==================================================================================================
# 17. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_QINGYUN_3407_TRADEOFF_DIAGNOSTIC_PASS = True


print()
print("=" * 175)
print("NOTEBOOK 05 — QINGYUN SEED-3407 TRADE-OFF DIAGNOSIS")
print("=" * 175)

print(
    "Macro-F1 signal                 :",
    MACRO_F1_SIGNAL
)

print(
    "AA signal                       :",
    AA_SIGNAL
)

print(
    "Tail Accuracy signal            :",
    TAIL_RECALL_SIGNAL
)

print(
    "Tail mean-F1 signal             :",
    TAIL_F1_SIGNAL
)

print(
    "Worst-Class Accuracy signal     :",
    WORST_CLASS_SIGNAL
)

print()

print(
    "Maximum Macro-F1 gain           :",
    f"{MAX_MACRO_F1_GAIN:+.12e}"
)

print(
    "Unique prediction patterns      :",
    unique_prediction_patterns
)

print()

print(
    "TRADE-OFF INTERPRETATION        :",
    TRADEOFF_INTERPRETATION
)

print(
    "NEXT AUDIT TARGET               :",
    NEXT_AUDIT_TARGET
)

print()

print(
    "Formula changed                 : False"
)

print(
    "Selection rule changed          : False"
)

print(
    "Training performed              :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used              :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 QINGYUN-3407 TRADE-OFF DIAGNOSTIC: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 175)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 20C
# PURE ADAPTIVE-ALPHA EFFECT DIAGNOSTIC
#
# PURPOSE
#   Isolate the actual effect of LTLC rarity adaptation.
#
#   For each dataset / seed / tau:
#
#       reference:
#           Standard LA at SAME tau
#           == LTLC with alpha = 0
#
#       candidate:
#           LTLC with SAME tau and alpha > 0
#
#   Therefore every reported delta is attributable ONLY to alpha.
#
# IMPORTANT:
#   - tau=0 candidates MUST have exactly zero alpha effect.
#   - no comparison against a differently selected tau is allowed here.
#
# DIAGNOSTIC ONLY
# FORMULA UNCHANGED
# SELECTION RULE UNCHANGED
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json

import numpy as np
import pandas as pd


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_QINGYUN_3407_TRADEOFF_DIAGNOSTIC_PASS is True
assert NOTEBOOK05_LTLC_HEADROOM_DIAGNOSTIC_PASS is True

assert NOTEBOOK05_DEVELOPMENT_REVISE is True
assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


DIAG_TOL = 1e-12


# ==================================================================================================
# 2. STORAGE
# ==================================================================================================

PURE_ALPHA_ROWS = []


print("=" * 165)
print("NOTEBOOK 05 — PURE ADAPTIVE-ALPHA EFFECT DIAGNOSTIC")
print("=" * 165)

print(
    "Reference:",
    "Standard LA at SAME tau"
)

print(
    "Candidate:",
    "LTLC-Shift at SAME tau with alpha > 0"
)

print(
    "Interpretation:",
    "all deltas isolate alpha only"
)

print(
    "Official test:",
    "SEALED"
)

print("=" * 165)


# ==================================================================================================
# 3. RUN ALL DATASET / SEED / TAU / ALPHA COMPARISONS
# ==================================================================================================

for dataset, seed in RUN_ORDER:

    run_df = (

        CORRECTED_SHIFT_GRID_DF[
            (
                CORRECTED_SHIFT_GRID_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                CORRECTED_SHIFT_GRID_DF[
                    "seed"
                ].astype(
                    int
                )
                ==
                int(
                    seed
                )
            )
        ]
        .copy()
    )


    standard_df = (

        run_df[
            run_df[
                "method"
            ]
            ==
            "Standard_LA"
        ]
        .copy()
    )


    ltlc_df = (

        run_df[
            run_df[
                "method"
            ]
            ==
            "LTLC_Shift"
        ]
        .copy()
    )


    assert len(
        standard_df
    ) == len(
        TAU_GRID
    )


    assert len(
        ltlc_df
    ) == (
        len(
            TAU_GRID
        )
        *
        len(
            ALPHA_GRID
        )
    )


    for tau in TAU_GRID:

        # ------------------------------------------------------------------------------------------
        # SAME-TAU alpha=0 reference
        # ------------------------------------------------------------------------------------------

        reference = standard_df[
            np.isclose(
                standard_df[
                    "tau"
                ],
                float(
                    tau
                ),
                rtol=0.0,
                atol=1e-15
            )
        ]


        assert len(
            reference
        ) == 1


        reference = reference.iloc[
            0
        ]


        # ------------------------------------------------------------------------------------------
        # Verify LTLC alpha=0 is exactly Standard LA at same tau
        # ------------------------------------------------------------------------------------------

        alpha_zero = ltlc_df[
            np.isclose(
                ltlc_df[
                    "tau"
                ],
                float(
                    tau
                ),
                rtol=0.0,
                atol=1e-15
            )
            &
            np.isclose(
                ltlc_df[
                    "alpha"
                ],
                0.0,
                rtol=0.0,
                atol=1e-15
            )
        ]


        assert len(
            alpha_zero
        ) == 1


        alpha_zero = alpha_zero.iloc[
            0
        ]


        for metric in [

            "validation_oa",
            "validation_aa",
            "validation_macro_f1",
            "validation_tail_accuracy",
            "validation_medium_accuracy",
            "validation_head_accuracy",
            "validation_worst_class_accuracy",
            "validation_nll",

        ]:

            assert np.isclose(

                float(
                    reference[
                        metric
                    ]
                ),

                float(
                    alpha_zero[
                        metric
                    ]
                ),

                rtol=0.0,

                atol=1e-12
            )


        # ------------------------------------------------------------------------------------------
        # All alpha > 0 candidates at SAME tau
        # ------------------------------------------------------------------------------------------

        candidates = ltlc_df[
            np.isclose(
                ltlc_df[
                    "tau"
                ],
                float(
                    tau
                ),
                rtol=0.0,
                atol=1e-15
            )
            &
            (
                ltlc_df[
                    "alpha"
                ].astype(
                    float
                )
                >
                0.0
            )
        ]


        assert len(
            candidates
        ) == (
            len(
                ALPHA_GRID
            )
            -
            1
        )


        for _, candidate in candidates.iterrows():

            PURE_ALPHA_ROWS.append({

                "dataset":
                    dataset,

                "seed":
                    int(
                        seed
                    ),

                "tau":
                    float(
                        tau
                    ),

                "alpha":
                    float(
                        candidate[
                            "alpha"
                        ]
                    ),

                # ----------------------------------------------------------------------------------
                # PURE alpha contribution:
                # LTLC(tau, alpha>0) - StandardLA(tau)
                # ----------------------------------------------------------------------------------

                "alpha_delta_oa":
                    float(
                        candidate[
                            "validation_oa"
                        ]
                        -
                        reference[
                            "validation_oa"
                        ]
                    ),

                "alpha_delta_aa":
                    float(
                        candidate[
                            "validation_aa"
                        ]
                        -
                        reference[
                            "validation_aa"
                        ]
                    ),

                "alpha_delta_macro_f1":
                    float(
                        candidate[
                            "validation_macro_f1"
                        ]
                        -
                        reference[
                            "validation_macro_f1"
                        ]
                    ),

                "alpha_delta_tail_accuracy":
                    float(
                        candidate[
                            "validation_tail_accuracy"
                        ]
                        -
                        reference[
                            "validation_tail_accuracy"
                        ]
                    ),

                "alpha_delta_medium_accuracy":
                    float(
                        candidate[
                            "validation_medium_accuracy"
                        ]
                        -
                        reference[
                            "validation_medium_accuracy"
                        ]
                    ),

                "alpha_delta_head_accuracy":
                    float(
                        candidate[
                            "validation_head_accuracy"
                        ]
                        -
                        reference[
                            "validation_head_accuracy"
                        ]
                    ),

                "alpha_delta_worst_class_accuracy":
                    float(
                        candidate[
                            "validation_worst_class_accuracy"
                        ]
                        -
                        reference[
                            "validation_worst_class_accuracy"
                        ]
                    ),

                # Negative = alpha improves NLL.
                "alpha_delta_nll":
                    float(
                        candidate[
                            "validation_nll"
                        ]
                        -
                        reference[
                            "validation_nll"
                        ]
                    ),
            })


PURE_ALPHA_DF = pd.DataFrame(
    PURE_ALPHA_ROWS
)


EXPECTED_ROWS = (

    len(
        RUN_ORDER
    )
    *
    len(
        TAU_GRID
    )
    *
    (
        len(
            ALPHA_GRID
        )
        -
        1
    )
)


assert len(
    PURE_ALPHA_DF
) == EXPECTED_ROWS


assert EXPECTED_ROWS == 486


# ==================================================================================================
# 4. MANDATORY TAU=0 IDENTITY CHECK
#
# alpha cannot possibly matter when tau=0.
# ==================================================================================================

tau_zero_df = (

    PURE_ALPHA_DF[
        np.isclose(
            PURE_ALPHA_DF[
                "tau"
            ],
            0.0,
            rtol=0.0,
            atol=1e-15
        )
    ]
)


assert len(
    tau_zero_df
) == (
    9
    *
    (
        len(
            ALPHA_GRID
        )
        -
        1
    )
)


PURE_EFFECT_COLUMNS = [

    "alpha_delta_oa",
    "alpha_delta_aa",
    "alpha_delta_macro_f1",
    "alpha_delta_tail_accuracy",
    "alpha_delta_medium_accuracy",
    "alpha_delta_head_accuracy",
    "alpha_delta_worst_class_accuracy",
    "alpha_delta_nll",
]


for column in PURE_EFFECT_COLUMNS:

    assert np.allclose(

        tau_zero_df[
            column
        ].to_numpy(
            dtype=np.float64
        ),

        0.0,

        rtol=0.0,

        atol=1e-12
    ), (
        f"tau=0 alpha identity failed for {column}"
    )


print()
print(
    "tau=0 alpha-identity comparisons:",
    len(
        tau_zero_df
    ),
    "/",
    len(
        tau_zero_df
    ),
    "PASS"
)


# ==================================================================================================
# 5. PURE RECOGNITION-SIGNAL MASK
# ==================================================================================================

PURE_ALPHA_DF[
    "pure_recognition_signal"
] = (

    (
        PURE_ALPHA_DF[
            "alpha_delta_aa"
        ]
        >
        DIAG_TOL
    )

    |

    (
        PURE_ALPHA_DF[
            "alpha_delta_macro_f1"
        ]
        >
        DIAG_TOL
    )

    |

    (
        PURE_ALPHA_DF[
            "alpha_delta_tail_accuracy"
        ]
        >
        DIAG_TOL
    )

    |

    (
        PURE_ALPHA_DF[
            "alpha_delta_worst_class_accuracy"
        ]
        >
        DIAG_TOL
    )
)


# ==================================================================================================
# 6. PER-RUN SUMMARY
# ==================================================================================================

PURE_ALPHA_RUN_SUMMARY = []


print()
print("=" * 165)
print("PER-RUN PURE ALPHA EFFECT")
print("=" * 165)


for dataset, seed in RUN_ORDER:

    subset = (

        PURE_ALPHA_DF[
            (
                PURE_ALPHA_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                PURE_ALPHA_DF[
                    "seed"
                ].astype(
                    int
                )
                ==
                int(
                    seed
                )
            )
        ]
    )


    assert len(
        subset
    ) == (
        len(
            TAU_GRID
        )
        *
        (
            len(
                ALPHA_GRID
            )
            -
            1
        )
    )


    signal_subset = subset[
        subset[
            "pure_recognition_signal"
        ]
    ]


    signal_count = int(
        len(
            signal_subset
        )
    )


    max_aa = float(
        subset[
            "alpha_delta_aa"
        ].max()
    )


    max_f1 = float(
        subset[
            "alpha_delta_macro_f1"
        ].max()
    )


    max_tail = float(
        subset[
            "alpha_delta_tail_accuracy"
        ].max()
    )


    max_worst = float(
        subset[
            "alpha_delta_worst_class_accuracy"
        ].max()
    )


    best_f1_row = subset.loc[
        subset[
            "alpha_delta_macro_f1"
        ].idxmax()
    ]


    PURE_ALPHA_RUN_SUMMARY.append({

        "dataset":
            dataset,

        "seed":
            int(
                seed
            ),

        "pure_signal_candidates":
            signal_count,

        "max_alpha_delta_aa":
            max_aa,

        "max_alpha_delta_macro_f1":
            max_f1,

        "max_alpha_delta_tail_accuracy":
            max_tail,

        "max_alpha_delta_worst_class_accuracy":
            max_worst,

        "best_macro_f1_tau":
            float(
                best_f1_row[
                    "tau"
                ]
            ),

        "best_macro_f1_alpha":
            float(
                best_f1_row[
                    "alpha"
                ]
            ),

        "best_macro_f1_oa_delta":
            float(
                best_f1_row[
                    "alpha_delta_oa"
                ]
            ),

        "any_pure_alpha_recognition_signal":
            bool(
                signal_count
                >
                0
            ),
    })


    print()

    print(
        f"{dataset:11s} | seed={seed:4d}"
    )


    print(
        "  Pure-alpha signal candidates:",
        signal_count
    )

    print(
        "  Max pure ΔAA              :",
        f"{max_aa:+.12e}"
    )

    print(
        "  Max pure ΔMacro-F1        :",
        f"{max_f1:+.12e}"
    )

    print(
        "  Max pure ΔTail Accuracy   :",
        f"{max_tail:+.12e}"
    )

    print(
        "  Max pure ΔWorst-Class Acc :",
        f"{max_worst:+.12e}"
    )

    print(
        "  Best-F1 tau / alpha       :",
        f"{float(best_f1_row['tau']):.2f}",
        "/",
        f"{float(best_f1_row['alpha']):.2f}"
    )


PURE_ALPHA_RUN_SUMMARY_DF = pd.DataFrame(
    PURE_ALPHA_RUN_SUMMARY
)


assert len(
    PURE_ALPHA_RUN_SUMMARY_DF
) == 9


# ==================================================================================================
# 7. DATASET-LEVEL SUMMARY
# ==================================================================================================

PURE_ALPHA_DATASET_SUMMARY = {}


print()
print("=" * 165)
print("DATASET-LEVEL PURE ADAPTIVE-ALPHA DIAGNOSIS")
print("=" * 165)


for dataset in [

    "Pingan",
    "Qingyun",
    "Tangdaowan",

]:

    subset = (

        PURE_ALPHA_RUN_SUMMARY_DF[
            PURE_ALPHA_RUN_SUMMARY_DF[
                "dataset"
            ]
            ==
            dataset
        ]
    )


    assert len(
        subset
    ) == 3


    seeds_with_signal = int(

        subset[
            "any_pure_alpha_recognition_signal"
        ].sum()
    )


    grid_subset = (

        PURE_ALPHA_DF[
            PURE_ALPHA_DF[
                "dataset"
            ]
            ==
            dataset
        ]
    )


    signal_candidate_count = int(

        grid_subset[
            "pure_recognition_signal"
        ].sum()
    )


    max_gains = {

        "AA":
            float(
                grid_subset[
                    "alpha_delta_aa"
                ].max()
            ),

        "Macro_F1":
            float(
                grid_subset[
                    "alpha_delta_macro_f1"
                ].max()
            ),

        "Tail_Accuracy":
            float(
                grid_subset[
                    "alpha_delta_tail_accuracy"
                ].max()
            ),

        "Worst_Class_Accuracy":
            float(
                grid_subset[
                    "alpha_delta_worst_class_accuracy"
                ].max()
            ),
    }


    any_signal = bool(

        seeds_with_signal
        >
        0
    )


    PURE_ALPHA_DATASET_SUMMARY[
        dataset
    ] = {

        "seeds_with_signal":
            seeds_with_signal,

        "pure_alpha_signal_candidate_count":
            signal_candidate_count,

        "maximum_pure_alpha_gains":
            max_gains,

        "any_pure_alpha_recognition_signal":
            any_signal,
    }


    print()

    print(
        dataset
    )

    print(
        "  Seeds with pure alpha signal:",
        seeds_with_signal,
        "/ 3"
    )

    print(
        "  Pure-alpha signal candidates:",
        signal_candidate_count
    )

    print(
        "  Maximum pure ΔAA          :",
        f"{max_gains['AA']:+.12e}"
    )

    print(
        "  Maximum pure ΔMacro-F1    :",
        f"{max_gains['Macro_F1']:+.12e}"
    )

    print(
        "  Maximum pure ΔTail Acc    :",
        f"{max_gains['Tail_Accuracy']:+.12e}"
    )

    print(
        "  Maximum pure ΔWorst       :",
        f"{max_gains['Worst_Class_Accuracy']:+.12e}"
    )

    print(
        "  Any pure adaptive signal  :",
        any_signal
    )


# ==================================================================================================
# 8. CROSS-DATASET CONCLUSION
# ==================================================================================================

datasets_with_pure_signal = [

    dataset

    for dataset, info
    in PURE_ALPHA_DATASET_SUMMARY.items()

    if info[
        "any_pure_alpha_recognition_signal"
    ]
]


seeds_with_pure_signal = int(

    PURE_ALPHA_RUN_SUMMARY_DF[
        "any_pure_alpha_recognition_signal"
    ].sum()
)


total_pure_signal_candidates = int(

    PURE_ALPHA_DF[
        "pure_recognition_signal"
    ].sum()
)


if len(
    datasets_with_pure_signal
) == 0:

    PURE_ALPHA_CONCLUSION = (
        "NO_ADAPTIVE_ALPHA_RECOGNITION_SIGNAL"
    )

    NEXT_REVISION_TARGET = (
        "LTLC_SHIFT_FORMULATION"
    )


elif (
    len(
        datasets_with_pure_signal
    )
    ==
    1
    and
    seeds_with_pure_signal
    ==
    1
):

    PURE_ALPHA_CONCLUSION = (
        "ISOLATED_NONROBUST_ADAPTIVE_ALPHA_SIGNAL"
    )

    NEXT_REVISION_TARGET = (
        "FORMULATION_ROBUSTNESS_BEFORE_ANY_SELECTION_RULE_CHANGE"
    )


else:

    PURE_ALPHA_CONCLUSION = (
        "REPEATABLE_ADAPTIVE_ALPHA_SIGNAL_PRESENT"
    )

    NEXT_REVISION_TARGET = (
        "SELECTION_OBJECTIVE_TRADEOFF_AUDIT"
    )


# ==================================================================================================
# 9. SAVE FULL DIAGNOSTIC CSV
# ==================================================================================================

PURE_ALPHA_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_ltlc_pure_alpha_effect_grid.csv"
)


pure_alpha_text = PURE_ALPHA_DF.to_csv(
    index=False,
    float_format="%.17g"
)


if PURE_ALPHA_PATH.exists():

    assert (
        PURE_ALPHA_PATH.read_text(
            encoding="utf-8"
        )
        ==
        pure_alpha_text
    )

else:

    PURE_ALPHA_PATH.write_text(
        pure_alpha_text,
        encoding="utf-8"
    )


PURE_ALPHA_SHA = sha256_file(
    PURE_ALPHA_PATH
)


# ==================================================================================================
# 10. SAVE MANIFEST
# ==================================================================================================

PURE_ALPHA_MANIFEST_PATH = (

    NB05_AUDIT_ROOT
    /
    "bestAA_ltlc_pure_alpha_effect_manifest.json"
)


PURE_ALPHA_MANIFEST = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "purpose":
        (
            "Validation-only isolation of the rarity-adaptive alpha "
            "effect by comparing LTLC(tau,alpha>0) against "
            "Standard-LA at the same tau."
        ),

    "comparison":
        "LTLC_same_tau_alpha_gt_0_minus_StandardLA_same_tau",

    "tau_zero_identity_verified":
        True,

    "rows":
        int(
            len(
                PURE_ALPHA_DF
            )
        ),

    "dataset_summary":
        PURE_ALPHA_DATASET_SUMMARY,

    "datasets_with_pure_alpha_signal":
        datasets_with_pure_signal,

    "seeds_with_pure_alpha_signal":
        seeds_with_pure_signal,

    "total_pure_alpha_signal_candidates":
        total_pure_signal_candidates,

    "conclusion":
        PURE_ALPHA_CONCLUSION,

    "recommended_next_revision_target":
        NEXT_REVISION_TARGET,

    "formula_changed":
        False,

    "selection_rule_changed":
        False,

    "tau_grid_changed":
        False,

    "alpha_grid_changed":
        False,

    "source_shift_grid_sha256":
        CORRECTED_SHIFT_GRID_SHA,

    "source_qingyun_tradeoff_manifest_sha256":
        TRADEOFF_MANIFEST_SHA,

    "diagnostic_csv_sha256":
        PURE_ALPHA_SHA,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


manifest_text = (

    json.dumps(
        PURE_ALPHA_MANIFEST,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if PURE_ALPHA_MANIFEST_PATH.exists():

    assert (
        PURE_ALPHA_MANIFEST_PATH.read_text(
            encoding="utf-8"
        )
        ==
        manifest_text
    )

else:

    PURE_ALPHA_MANIFEST_PATH.write_text(
        manifest_text,
        encoding="utf-8"
    )


PURE_ALPHA_MANIFEST_SHA = sha256_file(
    PURE_ALPHA_MANIFEST_PATH
)


# ==================================================================================================
# 11. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_PURE_ALPHA_DIAGNOSTIC_PASS = True


print()
print("=" * 165)
print("NOTEBOOK 05 — PURE ADAPTIVE-ALPHA CONCLUSION")
print("=" * 165)

print(
    "Datasets with pure alpha signal :",
    len(
        datasets_with_pure_signal
    ),
    "/ 3"
)

print(
    "Seeds with pure alpha signal    :",
    seeds_with_pure_signal,
    "/ 9"
)

print(
    "Pure-alpha signal candidates    :",
    total_pure_signal_candidates
)

print()

print(
    "PURE ALPHA CONCLUSION           :",
    PURE_ALPHA_CONCLUSION
)

print(
    "NEXT REVISION TARGET            :",
    NEXT_REVISION_TARGET
)

print()

print(
    "Diagnostic CSV:"
)

print(
    PURE_ALPHA_PATH
)

print(
    "SHA-256:"
)

print(
    PURE_ALPHA_SHA
)

print()

print(
    "Diagnostic manifest:"
)

print(
    PURE_ALPHA_MANIFEST_PATH
)

print(
    "SHA-256:"
)

print(
    PURE_ALPHA_MANIFEST_SHA
)

print()

print(
    "Formula changed                 : False"
)

print(
    "Selection rule changed          : False"
)

print(
    "Training performed              :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used              :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 PURE ALPHA DIAGNOSTIC: PASS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 165)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 20D
# FREEZE LTLC REVISION R1 — DECOUPLED ADAPTIVE SHIFT
#
# ORIGINAL:
#
#   z_B[k] =
#       z[k]
#       - tau * (1 + alpha * rarity[k]) * log(pi[k] + eps)
#
# EXPANDED:
#
#   z_B[k] =
#       z[k]
#       - tau * log(pi[k] + eps)
#       - tau * alpha * rarity[k] * log(pi[k] + eps)
#
# PROBLEM:
#
#   adaptive strength = tau * alpha
#
#   Therefore tau = 0 forces the adaptive component to zero.
#
# REVISION R1:
#
#   z_R1[k] =
#       z[k]
#       - tau  * log(pi[k] + eps)
#       - beta * rarity[k] * log(pi[k] + eps)
#
#   Equivalently:
#
#   z_R1[k] =
#       z[k]
#       - (tau + beta * rarity[k]) * log(pi[k] + eps)
#
# beta is now independent of tau.
#
# IMPORTANT:
#   THIS CELL ONLY FREEZES THE REVISION.
#   IT DOES NOT EVALUATE R1.
#   IT DOES NOT LOOK AT OFFICIAL TEST.
# ==================================================================================================

from pathlib import Path
import json

import numpy as np


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_PURE_ALPHA_DIAGNOSTIC_PASS is True

assert NOTEBOOK05_FINAL_VALIDATION_DEVELOPMENT_GATE_COMPLETE is True
assert NOTEBOOK05_DEVELOPMENT_REVISE is True
assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


# ==================================================================================================
# 2. FREEZE OBSERVED FAILURE EVIDENCE
# ==================================================================================================

selected_standard_tau = {

    f"{dataset}_{seed}":
        float(
            CORRECTED_STANDARD_LA_SELECTION[
                (
                    dataset,
                    seed
                )
            ][
                "tau"
            ]
        )

    for dataset, seed in RUN_ORDER
}


selected_tau_zero_count = int(

    sum(

        np.isclose(
            tau,
            0.0,
            rtol=0.0,
            atol=1e-15
        )

        for tau in selected_standard_tau.values()
    )
)


selected_tau_nonzero_count = (

    len(
        selected_standard_tau
    )
    -
    selected_tau_zero_count
)


assert selected_tau_zero_count == 8

assert selected_tau_nonzero_count == 1


assert (
    PURE_ALPHA_MANIFEST[
        "seeds_with_pure_alpha_signal"
    ]
    ==
    1
)


assert (
    PURE_ALPHA_MANIFEST[
        "datasets_with_pure_alpha_signal"
    ]
    ==
    [
        "Qingyun"
    ]
)


# ==================================================================================================
# 3. ALGEBRAIC FAILURE MODE
# ==================================================================================================

ORIGINAL_FORMULA = (
    "z_B[k] = z[k] - tau * "
    "(1 + alpha * rarity[k]) * "
    "log(pi[k] + eps)"
)


ORIGINAL_EXPANDED_FORMULA = (
    "z_B[k] = z[k] "
    "- tau * log(pi[k] + eps) "
    "- tau * alpha * rarity[k] * log(pi[k] + eps)"
)


ORIGINAL_EFFECTIVE_ADAPTIVE_COEFFICIENT = (
    "beta_effective = tau * alpha"
)


STRUCTURAL_LIMITATION = (
    "The original adaptive strength is multiplicatively coupled "
    "to tau. Therefore tau=0 forces the rarity-adaptive term to "
    "zero for every alpha."
)


# ==================================================================================================
# 4. REVISION R1 — DECOUPLE ADAPTIVE COEFFICIENT
# ==================================================================================================

REVISION_R1_FORMULA = (
    "z_R1[k] = z[k] "
    "- tau * log(pi[k] + eps) "
    "- beta * rarity[k] * log(pi[k] + eps)"
)


REVISION_R1_EQUIVALENT_FORM = (
    "z_R1[k] = z[k] "
    "- (tau + beta * rarity[k]) * log(pi[k] + eps)"
)


# Standard LA must be recovered exactly when beta = 0.
REVISION_R1_STANDARD_LA_IDENTITY = (
    "beta = 0 reproduces Standard LA exactly"
)


# Unlike the original LTLC form, tau=0 does NOT force beta=0.
REVISION_R1_TAU_ZERO_PROPERTY = (
    "tau = 0 preserves an independent beta-controlled "
    "rarity-adaptive shift"
)


# ==================================================================================================
# 5. FREEZE SEARCH GRID BEFORE R1 EVALUATION
#
# Keep EXACTLY 7 adaptive values — same tuning-budget size as the
# original alpha grid.
#
# We do NOT broaden the search after seeing R1 results.
# ==================================================================================================

REVISION_R1_TAU_GRID = np.asarray(
    TAU_GRID,
    dtype=np.float64
)


REVISION_R1_BETA_GRID = np.asarray(
    [
        0.00,
        0.25,
        0.50,
        0.75,
        1.00,
        1.50,
        2.00,
    ],
    dtype=np.float64
)


assert np.array_equal(
    REVISION_R1_TAU_GRID,
    TAU_GRID
)


assert np.array_equal(
    REVISION_R1_BETA_GRID,
    ALPHA_GRID
)


assert len(
    REVISION_R1_TAU_GRID
) == 9


assert len(
    REVISION_R1_BETA_GRID
) == 7


assert (
    len(
        REVISION_R1_TAU_GRID
    )
    *
    len(
        REVISION_R1_BETA_GRID
    )
) == 63


# ==================================================================================================
# 6. FREEZE R1 SELECTION PROCEDURE
#
# Stronger robustness rule:
#
#   - Tune (tau, beta) using SEED 42 VALIDATION ONLY for each dataset.
#   - Freeze that dataset's selected (tau, beta).
#   - Apply unchanged to seeds 123 and 3407 validation.
#
# This prevents per-seed hyperparameter cherry-picking during R1.
#
# Exact selection hierarchy stays aligned with the frozen Notebook05 rule:
#
#   1. highest validation AA
#   2. highest validation Macro-F1
#   3. lowest validation NLL
#   4. smaller beta
#   5. unresolved tie => STOP
# ==================================================================================================

REVISION_R1_TUNING_SEED = 42


REVISION_R1_EVALUATION_SEEDS = [
    42,
    123,
    3407,
]


REVISION_R1_SELECTION_SCOPE = (
    "select_on_seed42_validation_per_dataset_then_freeze_to_seeds123_3407"
)


REVISION_R1_SELECTION_RULE = [

    "highest_validation_AA",

    "tie_highest_validation_Macro_F1",

    "tie_lowest_validation_NLL",

    "tie_smaller_beta",

    "unresolved_tie_STOP",
]


REVISION_R1_SELECTION_TOLERANCE = 1e-12


# ==================================================================================================
# 7. TEMPERATURE COMPONENT REMAINS UNCHANGED
#
# We change ONLY the shift formulation.
# Temperature is NOT changed in R1.
# ==================================================================================================

REVISION_R1_TEMPERATURE_CHANGED = False


REVISION_R1_TEMPERATURE_FORMULA = (
    LTLC_IMPLEMENTATION_SPEC[
        "LTLC_temperature"
    ]
)


REVISION_R1_CALIBRATED_LOGITS_FORMULA = (
    LTLC_IMPLEMENTATION_SPEC[
        "LTLC_calibrated_logits"
    ]
)


assert (
    REVISION_R1_TEMPERATURE_FORMULA
    ==
    "T(x) = exp(a + b * rarity[argmax(z_B(x))])"
)


# ==================================================================================================
# 8. MANDATORY R1 SANITY CHECK DEFINITIONS
# ==================================================================================================

REVISION_R1_MANDATORY_SANITY_CHECKS = {

    "beta_zero_reproduces_standard_LA":
        True,

    "tau_zero_beta_zero_reproduces_CE":
        True,

    "tau_zero_nonzero_beta_remains_adaptive":
        True,

    "positive_temperature_preserves_argmax":
        True,

    "softmax_rows_sum_to_one":
        True,

    "official_test_remains_sealed_during_development":
        True,
}


# ==================================================================================================
# 9. DEVELOPMENT GATE FOR R1
#
# We retain the same scientific criterion.
#
# Recognition:
#   improve >=1 of
#       AA / Macro-F1 / Tail Accuracy / Worst-Class Accuracy
#   without large OA collapse
#
# Calibration:
#   improve >=1 of
#       ECE / Tail-ECE / NLL
#
# BOTH required.
# ==================================================================================================

REVISION_R1_GATE = {

    "recognition_requirement":

        (
            "Improve at least one of AA, Macro-F1, "
            "Tail Accuracy, or Worst-Class Accuracy "
            "relative to Standard LA."
        ),


    "calibration_requirement":

        (
            "Improve at least one of ECE, Tail-ECE, or NLL "
            "relative to Standard LA."
        ),


    "large_oa_collapse_not_allowed":
        True,


    "both_recognition_and_calibration_required":
        True,
}


# ==================================================================================================
# 10. REVISION STATUS / SCIENTIFIC DISCLOSURE
# ==================================================================================================

REVISION_R1_DISCLOSURE = {

    "revision_type":
        "validation_driven_method_revision_after_predeclared_development_gate_failure",


    "trigger":

        (
            "Original LTLC-Shift selected alpha=0 in all 9 corrected "
            "best-validation-AA runs and produced no selected recognition "
            "gain. Pure-alpha analysis found an isolated signal in only "
            "Qingyun seed 3407."
        ),


    "mechanistic_reason":

        (
            "The original adaptive coefficient equals tau*alpha. "
            "Eight of nine selected Standard-LA runs use tau=0, "
            "which disables rarity adaptation under the original "
            "parameterization."
        ),


    "revision":

        (
            "Replace the coupled coefficient tau*alpha with an "
            "independent adaptive coefficient beta."
        ),


    "what_is_not_changed": [

        "training data",

        "validation data",

        "official test sealing",

        "training priors",

        "rarity definition",

        "frequency groups",

        "tau grid",

        "number of adaptive grid values",

        "temperature parameterization",

        "temperature optimizer",

        "temperature bounds",

        "calibration metrics",
    ],


    "publication_requirement":

        (
            "R1 must be disclosed as a validation-driven protocol "
            "revision made before any official-test inspection."
        ),
}


# ==================================================================================================
# 11. BUILD FROZEN R1 PROTOCOL
# ==================================================================================================

REVISION_R1_PROTOCOL = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "revision":
        "R1",

    "status":
        "FROZEN_BEFORE_R1_EVALUATION",

    "original_formula":
        ORIGINAL_FORMULA,

    "original_expanded_formula":
        ORIGINAL_EXPANDED_FORMULA,

    "original_effective_adaptive_coefficient":
        ORIGINAL_EFFECTIVE_ADAPTIVE_COEFFICIENT,

    "structural_limitation":
        STRUCTURAL_LIMITATION,

    "revised_formula":
        REVISION_R1_FORMULA,

    "revised_equivalent_form":
        REVISION_R1_EQUIVALENT_FORM,

    "standard_la_identity":
        REVISION_R1_STANDARD_LA_IDENTITY,

    "tau_zero_property":
        REVISION_R1_TAU_ZERO_PROPERTY,

    "tau_grid":
        REVISION_R1_TAU_GRID.tolist(),

    "beta_grid":
        REVISION_R1_BETA_GRID.tolist(),

    "grid_points_per_tuning_run":
        63,

    "tuning_seed":
        REVISION_R1_TUNING_SEED,

    "evaluation_seeds":
        REVISION_R1_EVALUATION_SEEDS,

    "selection_scope":
        REVISION_R1_SELECTION_SCOPE,

    "selection_rule":
        REVISION_R1_SELECTION_RULE,

    "selection_tolerance":
        REVISION_R1_SELECTION_TOLERANCE,

    "temperature_changed":
        REVISION_R1_TEMPERATURE_CHANGED,

    "temperature_formula":
        REVISION_R1_TEMPERATURE_FORMULA,

    "calibrated_logits_formula":
        REVISION_R1_CALIBRATED_LOGITS_FORMULA,

    "mandatory_sanity_checks":
        REVISION_R1_MANDATORY_SANITY_CHECKS,

    "development_gate":
        REVISION_R1_GATE,

    "failure_evidence": {

        "selected_standard_tau_zero_runs":
            selected_tau_zero_count,

        "selected_standard_tau_nonzero_runs":
            selected_tau_nonzero_count,

        "total_runs":
            len(
                RUN_ORDER
            ),

        "pure_alpha_signal_datasets":
            PURE_ALPHA_MANIFEST[
                "datasets_with_pure_alpha_signal"
            ],

        "pure_alpha_signal_seeds":
            PURE_ALPHA_MANIFEST[
                "seeds_with_pure_alpha_signal"
            ],

        "pure_alpha_signal_candidates":
            PURE_ALPHA_MANIFEST[
                "total_pure_alpha_signal_candidates"
            ],
    },

    "disclosure":
        REVISION_R1_DISCLOSURE,

    "source_final_development_gate_sha256":
        FINAL_DEVELOPMENT_GATE_SHA,

    "source_pure_alpha_manifest_sha256":
        PURE_ALPHA_MANIFEST_SHA,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


# ==================================================================================================
# 12. SAVE FROZEN R1 PROTOCOL
# ==================================================================================================

REVISION_R1_PROTOCOL_PATH = (

    NB05_AUDIT_ROOT
    /
    "notebook05_ltlc_revision_R1_protocol.json"
)


revision_text = (

    json.dumps(
        REVISION_R1_PROTOCOL,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if REVISION_R1_PROTOCOL_PATH.exists():

    existing = (
        REVISION_R1_PROTOCOL_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == revision_text, (
        "Existing R1 protocol differs. STOP."
    )


else:

    REVISION_R1_PROTOCOL_PATH.write_text(
        revision_text,
        encoding="utf-8"
    )


REVISION_R1_PROTOCOL_SHA = sha256_file(
    REVISION_R1_PROTOCOL_PATH
)


# ==================================================================================================
# 13. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_REVISION_R1_PROTOCOL_FROZEN = True


# ==================================================================================================
# 14. REPORT
# ==================================================================================================

print("=" * 155)
print("NOTEBOOK 05 — LTLC REVISION R1 PROTOCOL FREEZE")
print("=" * 155)

print()

print(
    "Original adaptive coefficient : tau * alpha"
)

print(
    "Revised adaptive coefficient  : beta"
)

print()

print(
    "Original:"
)

print(
    "  z - tau*(1 + alpha*r)*log(pi)"
)

print()

print(
    "Revision R1:"
)

print(
    "  z - tau*log(pi) - beta*r*log(pi)"
)

print()

print(
    "Selected tau=0 runs           :",
    selected_tau_zero_count,
    "/ 9"
)

print(
    "Pure-alpha signal datasets    :",
    PURE_ALPHA_MANIFEST[
        "datasets_with_pure_alpha_signal"
    ]
)

print(
    "Pure-alpha signal seeds       :",
    PURE_ALPHA_MANIFEST[
        "seeds_with_pure_alpha_signal"
    ],
    "/ 9"
)

print()

print(
    "Tau grid                      :",
    REVISION_R1_TAU_GRID.tolist()
)

print(
    "Beta grid                     :",
    REVISION_R1_BETA_GRID.tolist()
)

print(
    "Grid points                   :",
    63
)

print()

print(
    "Tuning seed                   :",
    REVISION_R1_TUNING_SEED
)

print(
    "Selection scope               :",
    REVISION_R1_SELECTION_SCOPE
)

print(
    "Temperature changed           :",
    REVISION_R1_TEMPERATURE_CHANGED
)

print()

print(
    "R1 protocol artifact:"
)

print(
    REVISION_R1_PROTOCOL_PATH
)

print(
    "SHA-256:"
)

print(
    REVISION_R1_PROTOCOL_SHA
)

print()

print(
    "Training performed            :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used            :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 LTLC REVISION R1 PROTOCOL: FROZEN"
)

print(
    "NO R1 RESULT HAS BEEN INSPECTED YET"
)

print(
    "NEXT: R1 VALIDATION-ONLY EVALUATION"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 155)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 20D-A
# FREEZE OPERATIONAL R1 ROBUSTNESS / OA-COLLAPSE GATE
#
# IMPORTANT:
#   RUN BEFORE ANY R1 GRID OR RESULT IS COMPUTED.
#
# THIS CELL DOES NOT:
#   - evaluate R1
#   - inspect R1 results
#   - change formula
#   - change tau/beta grids
#   - use official test
# ==================================================================================================

import json


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_REVISION_R1_PROTOCOL_FROZEN is True

assert NOTEBOOK05_DEVELOPMENT_REVISE is True
assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


# ==================================================================================================
# 2. OPERATIONALIZE "LARGE OA COLLAPSE"
#
# Absolute OA drop > 0.01 = more than 1 percentage point.
#
# This threshold is frozen BEFORE R1 result inspection.
# ==================================================================================================

REVISION_R1_LARGE_OA_COLLAPSE_THRESHOLD = 0.01


assert (
    REVISION_R1_LARGE_OA_COLLAPSE_THRESHOLD
    ==
    0.01
)


# ==================================================================================================
# 3. ROBUSTNESS SEEDS
#
# Seed 42:
#   tuning / model-selection seed
#
# Seeds 123 and 3407:
#   frozen-parameter robustness seeds
#
# Seed 42 may NOT by itself establish R1 robustness.
# ==================================================================================================

REVISION_R1_TUNING_SEED = 42


REVISION_R1_ROBUSTNESS_SEEDS = [
    123,
    3407,
]


assert REVISION_R1_TUNING_SEED not in (
    REVISION_R1_ROBUSTNESS_SEEDS
)


# ==================================================================================================
# 4. ROBUST RECOGNITION RULE
#
# For each dataset:
#
#   1. Select (tau,beta) on seed 42 only.
#
#   2. Freeze those exact parameters.
#
#   3. Apply to seeds 123 and 3407.
#
#   4. Compute:
#
#        R1 - selected Standard LA
#
#      separately for:
#        AA
#        Macro-F1
#        Tail Accuracy
#        Worst-Class Accuracy
#
#   5. Recognition robustness exists if the MEAN across seeds
#      123 and 3407 is > 0 for at least one recognition metric.
#
#   6. No robustness seed may suffer an OA decrease > 0.01.
# ==================================================================================================

REVISION_R1_ROBUSTNESS_RULE = {

    "tuning_seed":
        42,

    "robustness_seeds":
        [
            123,
            3407,
        ],

    "comparison":
        "R1_minus_selected_Standard_LA_same_dataset_seed",

    "recognition_metrics":
        [
            "AA",
            "Macro_F1",
            "Tail_Accuracy",
            "Worst_Class_Accuracy",
        ],

    "recognition_signal_rule":
        (
            "mean delta across robustness seeds 123 and 3407 "
            "must be > 0 for at least one recognition metric"
        ),

    "large_oa_collapse_definition":
        (
            "R1 OA decrease greater than 0.01 absolute "
            "on either robustness seed"
        ),

    "large_oa_collapse_threshold":
        0.01,

    "tuning_seed_alone_can_establish_robustness":
        False,
}


# ==================================================================================================
# 5. CALIBRATION RULE REMAINS AS ALREADY FROZEN
#
# Temperature evaluation occurs ONLY after shift recognition evaluation.
# ==================================================================================================

REVISION_R1_EVALUATION_SEQUENCE = [

    "seed42_shift_grid_and_parameter_selection",

    "freeze_selected_tau_beta",

    "evaluate_shift_on_seeds123_3407",

    "evaluate_recognition_robustness_gate",

    "only_if_recognition_gate_supports_continuation_fit_temperature",

    "evaluate_calibration_gate",

    "both_recognition_and_calibration_required_for_final_GO",
]


# ==================================================================================================
# 6. IMPORTANT STOP RULE
#
# If R1 shows no robust recognition signal, we do NOT keep redesigning
# additional variants on these same validation splits inside Notebook 05.
# ==================================================================================================

REVISION_R1_STOP_RULE = {

    "if_no_robust_recognition_signal":
        (
            "STOP_METHOD_DEVELOPMENT_ON_CURRENT_VALIDATION_SPLITS"
        ),

    "reason":
        (
            "Repeated redesign after observing the same validation "
            "results would increase validation overfitting risk."
        ),

    "official_test_may_then_be_opened":
        False,
}


# ==================================================================================================
# 7. BUILD AMENDMENT
# ==================================================================================================

REVISION_R1_GATE_AMENDMENT = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "revision":
        "R1",

    "status":
        "ROBUSTNESS_GATE_FROZEN_BEFORE_R1_EVALUATION",

    "r1_protocol_sha256":
        REVISION_R1_PROTOCOL_SHA,

    "large_oa_collapse_threshold":
        REVISION_R1_LARGE_OA_COLLAPSE_THRESHOLD,

    "robustness_rule":
        REVISION_R1_ROBUSTNESS_RULE,

    "evaluation_sequence":
        REVISION_R1_EVALUATION_SEQUENCE,

    "stop_rule":
        REVISION_R1_STOP_RULE,

    "formula_changed":
        False,

    "tau_grid_changed":
        False,

    "beta_grid_changed":
        False,

    "temperature_changed":
        False,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


# ==================================================================================================
# 8. SAVE
# ==================================================================================================

REVISION_R1_GATE_AMENDMENT_PATH = (

    NB05_AUDIT_ROOT
    /
    "notebook05_ltlc_revision_R1_robustness_gate.json"
)


gate_text = (

    json.dumps(
        REVISION_R1_GATE_AMENDMENT,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if REVISION_R1_GATE_AMENDMENT_PATH.exists():

    existing = (
        REVISION_R1_GATE_AMENDMENT_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == gate_text, (
        "Existing R1 robustness gate differs. STOP."
    )


else:

    REVISION_R1_GATE_AMENDMENT_PATH.write_text(
        gate_text,
        encoding="utf-8"
    )


REVISION_R1_GATE_AMENDMENT_SHA = sha256_file(
    REVISION_R1_GATE_AMENDMENT_PATH
)


# ==================================================================================================
# 9. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_REVISION_R1_ROBUSTNESS_GATE_FROZEN = True


# ==================================================================================================
# 10. REPORT
# ==================================================================================================

print("=" * 155)
print("NOTEBOOK 05 — R1 ROBUSTNESS GATE FREEZE")
print("=" * 155)

print()

print(
    "R1 protocol SHA-256          :",
    REVISION_R1_PROTOCOL_SHA
)

print()

print(
    "Tuning seed                  :",
    REVISION_R1_TUNING_SEED
)

print(
    "Robustness seeds             :",
    REVISION_R1_ROBUSTNESS_SEEDS
)

print()

print(
    "Recognition robustness rule:"
)

print(
    "  Mean delta over seeds 123 + 3407 must be > 0"
)

print(
    "  for at least one of:"
)

print(
    "  AA / Macro-F1 / Tail Accuracy / Worst-Class Accuracy"
)

print()

print(
    "Large OA collapse threshold  :",
    REVISION_R1_LARGE_OA_COLLAPSE_THRESHOLD
)

print(
    "Interpretation               : >1 percentage-point OA decrease"
)

print()

print(
    "Tuning seed alone sufficient : False"
)

print(
    "Formula changed              : False"
)

print(
    "Tau grid changed             : False"
)

print(
    "Beta grid changed            : False"
)

print()

print(
    "If R1 recognition still fails:"
)

print(
    "  STOP further method redesign on these validation splits"
)

print(
    "  Official test remains sealed"
)

print()

print(
    "R1 robustness-gate artifact:"
)

print(
    REVISION_R1_GATE_AMENDMENT_PATH
)

print(
    "SHA-256:"
)

print(
    REVISION_R1_GATE_AMENDMENT_SHA
)

print()

print(
    "Training performed           :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used           :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 R1 ROBUSTNESS GATE: FROZEN BEFORE EVALUATION"
)

print(
    "NEXT: R1 SEED-42 GRID + FROZEN-SEED RECOGNITION EVALUATION"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 155)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 21A
# LTLC REVISION R1 — SEED-42 GRID + FROZEN-SEED RECOGNITION EVALUATION
#
# R1:
#
#   z_R1[k]
#       =
#       z[k]
#       - tau  * log(pi[k] + eps)
#       - beta * rarity[k] * log(pi[k] + eps)
#
# DEVELOPMENT PROCEDURE:
#
#   Dataset independently:
#
#       Seed 42:
#           full 9 x 7 = 63 point (tau,beta) validation grid
#           select exactly once using frozen rule
#
#       Freeze selected (tau,beta)
#
#       Seeds 123 / 3407:
#           evaluate frozen pair only
#           NO retuning
#
# RECOGNITION ROBUSTNESS:
#
#   Mean R1 - selected Standard-LA delta across seeds 123 + 3407
#   must be > 0 for >=1 of:
#
#       AA
#       Macro-F1
#       Tail Accuracy
#       Worst-Class Accuracy
#
#   AND neither robustness seed may lose >0.01 OA.
#
# CPU ONLY
# VALIDATION ONLY
# NO TRAINING
# NO TEMPERATURE FITTING
# NO OFFICIAL TEST
# ==================================================================================================

import json

import numpy as np
import pandas as pd


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_REVISION_R1_PROTOCOL_FROZEN is True
assert NOTEBOOK05_REVISION_R1_ROBUSTNESS_GATE_FROZEN is True

assert NOTEBOOK05_DEVELOPMENT_REVISE is True
assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


R1_TOL = 1e-12

R1_DATASETS = [
    "Pingan",
    "Qingyun",
    "Tangdaowan",
]

R1_TUNING_SEED = 42

R1_ROBUSTNESS_SEEDS = [
    123,
    3407,
]


assert R1_TUNING_SEED == REVISION_R1_TUNING_SEED

assert (
    R1_ROBUSTNESS_SEEDS
    ==
    REVISION_R1_ROBUSTNESS_SEEDS
)


# ==================================================================================================
# 2. EXACT R1 SHIFT
# ==================================================================================================

def revision_r1_logits(
    logits,
    prior,
    rarity,
    tau,
    beta,
    eps
):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    prior = np.asarray(
        prior,
        dtype=np.float64
    )

    rarity = np.asarray(
        rarity,
        dtype=np.float64
    )


    tau = float(
        tau
    )

    beta = float(
        beta
    )

    eps = float(
        eps
    )


    assert logits.ndim == 2

    assert prior.ndim == 1

    assert rarity.ndim == 1

    assert logits.shape[1] == len(prior)

    assert len(prior) == len(rarity)

    assert (
        prior
        >
        0.0
    ).all()


    log_prior = np.log(
        prior
        +
        eps
    )


    shift = (

        tau
        *
        log_prior

        +

        beta
        *
        rarity
        *
        log_prior
    )


    revised_logits = (

        logits
        -
        shift[
            None,
            :
        ]
    )


    assert revised_logits.shape == logits.shape

    assert np.isfinite(
        revised_logits
    ).all()


    return revised_logits


# ==================================================================================================
# 3. MANDATORY R1 ALGEBRAIC SANITY TESTS
# ==================================================================================================

print("=" * 160)
print("NOTEBOOK 05 — LTLC REVISION R1 VALIDATION EVALUATION")
print("=" * 160)

print(
    "R1 formula:"
)

print(
    "  z - tau*log(pi) - beta*rarity*log(pi)"
)

print()

print(
    "Tuning seed      :",
    R1_TUNING_SEED
)

print(
    "Robustness seeds :",
    R1_ROBUSTNESS_SEEDS
)

print(
    "Grid points/tuning dataset:",
    len(
        REVISION_R1_TAU_GRID
    )
    *
    len(
        REVISION_R1_BETA_GRID
    )
)

print(
    "Official test    : SEALED"
)

print("=" * 160)


for dataset in R1_DATASETS:

    identity = (
        dataset,
        42
    )


    validation = load_bestAA_validation_output(

        FINAL_BEST_AA_VALIDATION_OUTPUTS[
            identity
        ]
    )


    logits = np.asarray(
        validation[
            "logits"
        ],
        dtype=np.float64
    )


    payload = PRIOR_RARITY_BY_DATASET[
        dataset
    ]


    prior = np.asarray(
        payload[
            "prior"
        ],
        dtype=np.float64
    )


    rarity = np.asarray(
        payload[
            "rarity"
        ],
        dtype=np.float64
    )


    # ----------------------------------------------------------------------------------------------
    # beta = 0 must reproduce Standard LA exactly
    # ----------------------------------------------------------------------------------------------

    for tau in REVISION_R1_TAU_GRID:

        r1_beta_zero = revision_r1_logits(

            logits,
            prior,
            rarity,

            tau=float(
                tau
            ),

            beta=0.0,

            eps=RARITY_EPSILON
        )


        standard_reference = standard_la_logits(

            logits,
            prior,

            tau=float(
                tau
            ),

            eps=RARITY_EPSILON
        )


        assert np.array_equal(

            r1_beta_zero,

            standard_reference
        )


    # ----------------------------------------------------------------------------------------------
    # tau=0, beta=0 must reproduce CE
    # ----------------------------------------------------------------------------------------------

    r1_identity = revision_r1_logits(

        logits,
        prior,
        rarity,

        tau=0.0,

        beta=0.0,

        eps=RARITY_EPSILON
    )


    assert np.array_equal(

        r1_identity,

        logits
    )


    # ----------------------------------------------------------------------------------------------
    # tau=0, beta>0 must NOT algebraically collapse to CE
    # ----------------------------------------------------------------------------------------------

    r1_adaptive_tau_zero = revision_r1_logits(

        logits,
        prior,
        rarity,

        tau=0.0,

        beta=0.25,

        eps=RARITY_EPSILON
    )


    assert not np.array_equal(

        r1_adaptive_tau_zero,

        logits
    )


    print(
        f"{dataset:11s} | "
        "beta=0 LA identity PASS | "
        "tau=0,beta=0 CE identity PASS | "
        "tau=0,beta>0 adaptive PASS"
    )


# ==================================================================================================
# 4. EXACT FROZEN R1 SELECTOR
# ==================================================================================================

def select_r1_seed42_candidate(
    candidate_df,
    label
):

    df = candidate_df.copy()


    assert len(
        df
    ) > 0


    # ----------------------------------------------------------------------------------------------
    # 1. highest validation AA
    # ----------------------------------------------------------------------------------------------

    best_aa = float(
        df[
            "validation_aa"
        ].max()
    )


    df = df[
        np.isclose(
            df[
                "validation_aa"
            ].to_numpy(
                dtype=np.float64
            ),
            best_aa,
            rtol=0.0,
            atol=R1_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 2. highest validation Macro-F1
    # ----------------------------------------------------------------------------------------------

    best_f1 = float(
        df[
            "validation_macro_f1"
        ].max()
    )


    df = df[
        np.isclose(
            df[
                "validation_macro_f1"
            ].to_numpy(
                dtype=np.float64
            ),
            best_f1,
            rtol=0.0,
            atol=R1_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 3. lowest validation NLL
    # ----------------------------------------------------------------------------------------------

    best_nll = float(
        df[
            "validation_nll"
        ].min()
    )


    df = df[
        np.isclose(
            df[
                "validation_nll"
            ].to_numpy(
                dtype=np.float64
            ),
            best_nll,
            rtol=0.0,
            atol=R1_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # 4. smaller beta
    # ----------------------------------------------------------------------------------------------

    best_beta = float(
        df[
            "beta"
        ].min()
    )


    df = df[
        np.isclose(
            df[
                "beta"
            ].to_numpy(
                dtype=np.float64
            ),
            best_beta,
            rtol=0.0,
            atol=R1_TOL
        )
    ].copy()


    # ----------------------------------------------------------------------------------------------
    # NO UNDECLARED TAU TIE-BREAK
    # ----------------------------------------------------------------------------------------------

    if len(
        df
    ) != 1:

        print()
        print("=" * 120)
        print("R1 UNRESOLVED LOCKED-PROTOCOL TIE")
        print("=" * 120)

        print(
            label
        )

        print()

        print(
            df[
                [
                    "tau",
                    "beta",
                    "validation_oa",
                    "validation_aa",
                    "validation_macro_f1",
                    "validation_nll",
                ]
            ].to_string(
                index=False
            )
        )

        print("=" * 120)


        raise RuntimeError(
            f"{label}: frozen R1 selection rule "
            "did not produce exactly one candidate."
        )


    return (
        df.iloc[
            0
        ]
        .to_dict()
    )


# ==================================================================================================
# 5. SEED-42 R1 GRID
# ==================================================================================================

R1_TUNING_GRID_ROWS = []

R1_SELECTED_PARAMETERS = {}


print()
print("=" * 160)
print("R1 SEED-42 TUNING GRID")
print("=" * 160)


for dataset in R1_DATASETS:

    identity = (
        dataset,
        R1_TUNING_SEED
    )


    payload = PRIOR_RARITY_BY_DATASET[
        dataset
    ]


    prior = np.asarray(
        payload[
            "prior"
        ],
        dtype=np.float64
    )


    rarity = np.asarray(
        payload[
            "rarity"
        ],
        dtype=np.float64
    )


    frequency_groups = np.asarray(
        payload[
            "frequency_group"
        ],
        dtype=object
    )


    validation = load_bestAA_validation_output(

        FINAL_BEST_AA_VALIDATION_OUTPUTS[
            identity
        ]
    )


    logits = np.asarray(
        validation[
            "logits"
        ],
        dtype=np.float64
    )


    labels = np.asarray(
        validation[
            "labels"
        ],
        dtype=np.int64
    )


    dataset_rows = []


    for tau in REVISION_R1_TAU_GRID:

        for beta in REVISION_R1_BETA_GRID:

            r1_logits = revision_r1_logits(

                logits,
                prior,
                rarity,

                tau=float(
                    tau
                ),

                beta=float(
                    beta
                ),

                eps=RARITY_EPSILON
            )


            metrics = evaluate_logits(

                r1_logits,
                labels,
                frequency_groups
            )


            row = {

                "dataset":
                    dataset,

                "seed":
                    int(
                        R1_TUNING_SEED
                    ),

                "tau":
                    float(
                        tau
                    ),

                "beta":
                    float(
                        beta
                    ),

                "validation_oa":
                    float(
                        metrics[
                            "oa"
                        ]
                    ),

                "validation_aa":
                    float(
                        metrics[
                            "aa"
                        ]
                    ),

                "validation_macro_f1":
                    float(
                        metrics[
                            "macro_f1"
                        ]
                    ),

                "validation_tail_accuracy":
                    float(
                        metrics[
                            "tail_accuracy"
                        ]
                    ),

                "validation_medium_accuracy":
                    float(
                        metrics[
                            "medium_accuracy"
                        ]
                    ),

                "validation_head_accuracy":
                    float(
                        metrics[
                            "head_accuracy"
                        ]
                    ),

                "validation_worst_class_accuracy":
                    float(
                        metrics[
                            "worst_class_accuracy"
                        ]
                    ),

                "validation_nll":
                    float(
                        metrics[
                            "nll"
                        ]
                    ),
            }


            R1_TUNING_GRID_ROWS.append(
                row
            )


            dataset_rows.append(
                row
            )


    dataset_grid_df = pd.DataFrame(
        dataset_rows
    )


    assert len(
        dataset_grid_df
    ) == 63


    selected = select_r1_seed42_candidate(

        dataset_grid_df,

        f"{dataset} R1 seed-42"
    )


    R1_SELECTED_PARAMETERS[
        dataset
    ] = {

        "tau":
            float(
                selected[
                    "tau"
                ]
            ),

        "beta":
            float(
                selected[
                    "beta"
                ]
            ),

        "seed42_validation_oa":
            float(
                selected[
                    "validation_oa"
                ]
            ),

        "seed42_validation_aa":
            float(
                selected[
                    "validation_aa"
                ]
            ),

        "seed42_validation_macro_f1":
            float(
                selected[
                    "validation_macro_f1"
                ]
            ),

        "seed42_validation_tail_accuracy":
            float(
                selected[
                    "validation_tail_accuracy"
                ]
            ),

        "seed42_validation_worst_class_accuracy":
            float(
                selected[
                    "validation_worst_class_accuracy"
                ]
            ),

        "seed42_validation_nll":
            float(
                selected[
                    "validation_nll"
                ]
            ),
    }


    print()

    print(
        dataset
    )

    print(
        "  Selected tau  :",
        f"{selected['tau']:.2f}"
    )

    print(
        "  Selected beta :",
        f"{selected['beta']:.2f}"
    )

    print(
        "  Seed42 OA     :",
        f"{selected['validation_oa']:.9f}"
    )

    print(
        "  Seed42 AA     :",
        f"{selected['validation_aa']:.9f}"
    )

    print(
        "  Seed42 F1     :",
        f"{selected['validation_macro_f1']:.9f}"
    )

    print(
        "  Seed42 Tail   :",
        f"{selected['validation_tail_accuracy']:.9f}"
    )

    print(
        "  Seed42 Worst  :",
        f"{selected['validation_worst_class_accuracy']:.9f}"
    )


R1_TUNING_GRID_DF = pd.DataFrame(
    R1_TUNING_GRID_ROWS
)


assert len(
    R1_TUNING_GRID_DF
) == (
    3
    *
    63
)


# ==================================================================================================
# 6. FREEZE SELECTED PARAMETERS IN MEMORY
# ==================================================================================================

assert set(
    R1_SELECTED_PARAMETERS.keys()
) == {
    "Pingan",
    "Qingyun",
    "Tangdaowan",
}


print()
print("=" * 160)
print("R1 PARAMETERS FROZEN AFTER SEED-42 SELECTION")
print("=" * 160)


for dataset in R1_DATASETS:

    selection = R1_SELECTED_PARAMETERS[
        dataset
    ]


    print(

        f"{dataset:11s} | "
        f"tau={selection['tau']:.2f} | "
        f"beta={selection['beta']:.2f}"
    )


# ==================================================================================================
# 7. APPLY FROZEN PARAMETERS TO ALL THREE SEEDS
#
# Seed42 = tuning result
# Seeds123/3407 = robustness evaluation only
# ==================================================================================================

R1_FROZEN_EVALUATION_ROWS = []


print()
print("=" * 160)
print("R1 FROZEN-PARAMETER VALIDATION EVALUATION")
print("=" * 160)


for dataset in R1_DATASETS:

    frozen_tau = float(
        R1_SELECTED_PARAMETERS[
            dataset
        ][
            "tau"
        ]
    )


    frozen_beta = float(
        R1_SELECTED_PARAMETERS[
            dataset
        ][
            "beta"
        ]
    )


    payload = PRIOR_RARITY_BY_DATASET[
        dataset
    ]


    prior = np.asarray(
        payload[
            "prior"
        ],
        dtype=np.float64
    )


    rarity = np.asarray(
        payload[
            "rarity"
        ],
        dtype=np.float64
    )


    frequency_groups = np.asarray(
        payload[
            "frequency_group"
        ],
        dtype=object
    )


    for seed in [
        42,
        123,
        3407,
    ]:

        identity = (
            dataset,
            seed
        )


        validation = load_bestAA_validation_output(

            FINAL_BEST_AA_VALIDATION_OUTPUTS[
                identity
            ]
        )


        logits = np.asarray(
            validation[
                "logits"
            ],
            dtype=np.float64
        )


        labels = np.asarray(
            validation[
                "labels"
            ],
            dtype=np.int64
        )


        r1_logits = revision_r1_logits(

            logits,
            prior,
            rarity,

            tau=frozen_tau,

            beta=frozen_beta,

            eps=RARITY_EPSILON
        )


        r1_metrics = evaluate_logits(

            r1_logits,
            labels,
            frequency_groups
        )


        # ------------------------------------------------------------------------------------------
        # Reference = already frozen selected Standard LA for THIS seed
        # ------------------------------------------------------------------------------------------

        baseline_selection = (

            CORRECTED_STANDARD_LA_SELECTION[
                identity
            ]
        )


        baseline_tau = float(
            baseline_selection[
                "tau"
            ]
        )


        baseline_logits = standard_la_logits(

            logits,
            prior,

            tau=baseline_tau,

            eps=RARITY_EPSILON
        )


        baseline_metrics = evaluate_logits(

            baseline_logits,
            labels,
            frequency_groups
        )


        row = {

            "dataset":
                dataset,

            "seed":
                int(
                    seed
                ),

            "role":
                (
                    "tuning"
                    if seed == 42
                    else "robustness"
                ),

            "r1_tau":
                frozen_tau,

            "r1_beta":
                frozen_beta,

            "baseline_standard_la_tau":
                baseline_tau,

            "r1_oa":
                float(
                    r1_metrics[
                        "oa"
                    ]
                ),

            "r1_aa":
                float(
                    r1_metrics[
                        "aa"
                    ]
                ),

            "r1_macro_f1":
                float(
                    r1_metrics[
                        "macro_f1"
                    ]
                ),

            "r1_tail_accuracy":
                float(
                    r1_metrics[
                        "tail_accuracy"
                    ]
                ),

            "r1_worst_class_accuracy":
                float(
                    r1_metrics[
                        "worst_class_accuracy"
                    ]
                ),

            "r1_nll":
                float(
                    r1_metrics[
                        "nll"
                    ]
                ),

            "baseline_oa":
                float(
                    baseline_metrics[
                        "oa"
                    ]
                ),

            "baseline_aa":
                float(
                    baseline_metrics[
                        "aa"
                    ]
                ),

            "baseline_macro_f1":
                float(
                    baseline_metrics[
                        "macro_f1"
                    ]
                ),

            "baseline_tail_accuracy":
                float(
                    baseline_metrics[
                        "tail_accuracy"
                    ]
                ),

            "baseline_worst_class_accuracy":
                float(
                    baseline_metrics[
                        "worst_class_accuracy"
                    ]
                ),

            "baseline_nll":
                float(
                    baseline_metrics[
                        "nll"
                    ]
                ),

            "delta_oa":
                float(
                    r1_metrics[
                        "oa"
                    ]
                    -
                    baseline_metrics[
                        "oa"
                    ]
                ),

            "delta_aa":
                float(
                    r1_metrics[
                        "aa"
                    ]
                    -
                    baseline_metrics[
                        "aa"
                    ]
                ),

            "delta_macro_f1":
                float(
                    r1_metrics[
                        "macro_f1"
                    ]
                    -
                    baseline_metrics[
                        "macro_f1"
                    ]
                ),

            "delta_tail_accuracy":
                float(
                    r1_metrics[
                        "tail_accuracy"
                    ]
                    -
                    baseline_metrics[
                        "tail_accuracy"
                    ]
                ),

            "delta_worst_class_accuracy":
                float(
                    r1_metrics[
                        "worst_class_accuracy"
                    ]
                    -
                    baseline_metrics[
                        "worst_class_accuracy"
                    ]
                ),

            "delta_nll":
                float(
                    r1_metrics[
                        "nll"
                    ]
                    -
                    baseline_metrics[
                        "nll"
                    ]
                ),
        }


        R1_FROZEN_EVALUATION_ROWS.append(
            row
        )


        print()

        print(

            f"{dataset:11s} | "
            f"seed={seed:4d} | "
            f"{row['role']:10s} | "
            f"R1(tau={frozen_tau:.2f}, beta={frozen_beta:.2f})"
        )


        print(

            f"  ΔOA    = {row['delta_oa']:+.9e} | "
            f"ΔAA    = {row['delta_aa']:+.9e} | "
            f"ΔF1    = {row['delta_macro_f1']:+.9e}"
        )


        print(

            f"  ΔTail  = {row['delta_tail_accuracy']:+.9e} | "
            f"ΔWorst = {row['delta_worst_class_accuracy']:+.9e} | "
            f"ΔNLL   = {row['delta_nll']:+.9e}"
        )


R1_FROZEN_EVALUATION_DF = pd.DataFrame(
    R1_FROZEN_EVALUATION_ROWS
)


assert len(
    R1_FROZEN_EVALUATION_DF
) == 9


# ==================================================================================================
# 8. ROBUSTNESS GATE — SEEDS 123 + 3407 ONLY
# ==================================================================================================

R1_RECOGNITION_GATE = {}


print()
print("=" * 160)
print("R1 ROBUSTNESS-SEED RECOGNITION GATE")
print("=" * 160)


for dataset in R1_DATASETS:

    subset = (

        R1_FROZEN_EVALUATION_DF[
            (
                R1_FROZEN_EVALUATION_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                R1_FROZEN_EVALUATION_DF[
                    "role"
                ]
                ==
                "robustness"
            )
        ]
        .copy()
        .sort_values(
            "seed"
        )
        .reset_index(
            drop=True
        )
    )


    assert list(
        subset[
            "seed"
        ].astype(
            int
        )
    ) == [
        123,
        3407,
    ]


    mean_deltas = {

        "AA":
            float(
                subset[
                    "delta_aa"
                ].mean()
            ),

        "Macro_F1":
            float(
                subset[
                    "delta_macro_f1"
                ].mean()
            ),

        "Tail_Accuracy":
            float(
                subset[
                    "delta_tail_accuracy"
                ].mean()
            ),

        "Worst_Class_Accuracy":
            float(
                subset[
                    "delta_worst_class_accuracy"
                ].mean()
            ),

        "OA":
            float(
                subset[
                    "delta_oa"
                ].mean()
            ),
    }


    recognition_metric_signal = {

        metric:
            bool(
                value
                >
                R1_TOL
            )

        for metric, value
        in mean_deltas.items()

        if metric != "OA"
    }


    recognition_signal = bool(

        any(
            recognition_metric_signal.values()
        )
    )


    per_seed_large_oa_collapse = {

        str(
            int(
                row[
                    "seed"
                ]
            )
        ):
            bool(
                float(
                    row[
                        "delta_oa"
                    ]
                )
                <
                -
                REVISION_R1_LARGE_OA_COLLAPSE_THRESHOLD
            )

        for _, row
        in subset.iterrows()
    }


    large_oa_collapse = bool(

        any(
            per_seed_large_oa_collapse.values()
        )
    )


    recognition_gate_pass = bool(

        recognition_signal

        and

        not large_oa_collapse
    )


    R1_RECOGNITION_GATE[
        dataset
    ] = {

        "frozen_tau":
            float(
                R1_SELECTED_PARAMETERS[
                    dataset
                ][
                    "tau"
                ]
            ),

        "frozen_beta":
            float(
                R1_SELECTED_PARAMETERS[
                    dataset
                ][
                    "beta"
                ]
            ),

        "robustness_seeds":
            [
                123,
                3407,
            ],

        "mean_deltas":
            mean_deltas,

        "metric_signal":
            recognition_metric_signal,

        "recognition_signal":
            recognition_signal,

        "per_seed_large_oa_collapse":
            per_seed_large_oa_collapse,

        "large_oa_collapse":
            large_oa_collapse,

        "recognition_gate_pass":
            recognition_gate_pass,
    }


    print()

    print(
        dataset.upper()
    )

    print(
        "  Frozen tau / beta       :",
        f"{R1_SELECTED_PARAMETERS[dataset]['tau']:.2f}",
        "/",
        f"{R1_SELECTED_PARAMETERS[dataset]['beta']:.2f}"
    )

    print(
        "  Mean robustness ΔAA     :",
        f"{mean_deltas['AA']:+.12e}"
    )

    print(
        "  Mean robustness ΔF1     :",
        f"{mean_deltas['Macro_F1']:+.12e}"
    )

    print(
        "  Mean robustness ΔTail   :",
        f"{mean_deltas['Tail_Accuracy']:+.12e}"
    )

    print(
        "  Mean robustness ΔWorst  :",
        f"{mean_deltas['Worst_Class_Accuracy']:+.12e}"
    )

    print(
        "  Mean robustness ΔOA     :",
        f"{mean_deltas['OA']:+.12e}"
    )

    print(
        "  Recognition signal      :",
        recognition_signal
    )

    print(
        "  Large OA collapse       :",
        large_oa_collapse
    )

    print(
        "  RECOGNITION GATE        :",
        (
            "PASS"
            if recognition_gate_pass
            else "FAIL"
        )
    )


# ==================================================================================================
# 9. CROSS-DATASET R1 RECOGNITION RESULT
# ==================================================================================================

datasets_passing_recognition = [

    dataset

    for dataset, gate
    in R1_RECOGNITION_GATE.items()

    if gate[
        "recognition_gate_pass"
    ]
]


datasets_with_nonzero_beta = [

    dataset

    for dataset, selection
    in R1_SELECTED_PARAMETERS.items()

    if float(
        selection[
            "beta"
        ]
    ) > 0.0
]


R1_RECOGNITION_PASS_DATASET_COUNT = len(
    datasets_passing_recognition
)


# Conservative continuation rule:
#
# R1 must demonstrate robustness on at least one dataset before
# temperature fitting is scientifically warranted.
#
# This does NOT imply final GO.
# Calibration would still have to pass afterwards.

R1_RECOGNITION_SUPPORTS_CALIBRATION_STAGE = bool(

    R1_RECOGNITION_PASS_DATASET_COUNT
    >
    0
)


# ==================================================================================================
# 10. SAVE SEED-42 GRID
# ==================================================================================================

R1_TUNING_GRID_PATH = (

    NB05_AUDIT_ROOT
    /
    "revision_R1_seed42_validation_grid.csv"
)


r1_grid_text = R1_TUNING_GRID_DF.to_csv(
    index=False,
    float_format="%.17g"
)


if R1_TUNING_GRID_PATH.exists():

    assert (
        R1_TUNING_GRID_PATH.read_text(
            encoding="utf-8"
        )
        ==
        r1_grid_text
    ), (
        "Existing R1 seed42 grid differs. STOP."
    )

else:

    R1_TUNING_GRID_PATH.write_text(
        r1_grid_text,
        encoding="utf-8"
    )


R1_TUNING_GRID_SHA = sha256_file(
    R1_TUNING_GRID_PATH
)


# ==================================================================================================
# 11. SAVE FROZEN-PARAMETER EVALUATION CSV
# ==================================================================================================

R1_EVALUATION_PATH = (

    NB05_AUDIT_ROOT
    /
    "revision_R1_frozen_parameter_validation_evaluation.csv"
)


r1_eval_text = R1_FROZEN_EVALUATION_DF.to_csv(
    index=False,
    float_format="%.17g"
)


if R1_EVALUATION_PATH.exists():

    assert (
        R1_EVALUATION_PATH.read_text(
            encoding="utf-8"
        )
        ==
        r1_eval_text
    ), (
        "Existing R1 evaluation CSV differs. STOP."
    )

else:

    R1_EVALUATION_PATH.write_text(
        r1_eval_text,
        encoding="utf-8"
    )


R1_EVALUATION_SHA = sha256_file(
    R1_EVALUATION_PATH
)


# ==================================================================================================
# 12. SAVE R1 RECOGNITION MANIFEST
# ==================================================================================================

R1_RECOGNITION_MANIFEST_PATH = (

    NB05_AUDIT_ROOT
    /
    "revision_R1_recognition_validation_manifest.json"
)


R1_RECOGNITION_MANIFEST = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "revision":
        "R1",

    "status":
        (
            "RECOGNITION_STAGE_COMPLETE"
        ),

    "r1_protocol_sha256":
        REVISION_R1_PROTOCOL_SHA,

    "r1_robustness_gate_sha256":
        REVISION_R1_GATE_AMENDMENT_SHA,

    "tuning_seed":
        42,

    "robustness_seeds":
        [
            123,
            3407,
        ],

    "selected_parameters":
        R1_SELECTED_PARAMETERS,

    "datasets_with_nonzero_beta":
        datasets_with_nonzero_beta,

    "dataset_recognition_gates":
        R1_RECOGNITION_GATE,

    "datasets_passing_recognition":
        datasets_passing_recognition,

    "recognition_pass_dataset_count":
        R1_RECOGNITION_PASS_DATASET_COUNT,

    "recognition_supports_calibration_stage":
        R1_RECOGNITION_SUPPORTS_CALIBRATION_STAGE,

    "seed42_grid_rows":
        int(
            len(
                R1_TUNING_GRID_DF
            )
        ),

    "frozen_parameter_evaluation_rows":
        int(
            len(
                R1_FROZEN_EVALUATION_DF
            )
        ),

    "seed42_grid_sha256":
        R1_TUNING_GRID_SHA,

    "frozen_parameter_evaluation_sha256":
        R1_EVALUATION_SHA,

    "formula_changed_after_freeze":
        False,

    "grid_changed_after_freeze":
        False,

    "selection_rule_changed_after_freeze":
        False,

    "temperature_fitted":
        False,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


manifest_text = (

    json.dumps(
        R1_RECOGNITION_MANIFEST,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if R1_RECOGNITION_MANIFEST_PATH.exists():

    assert (
        R1_RECOGNITION_MANIFEST_PATH.read_text(
            encoding="utf-8"
        )
        ==
        manifest_text
    ), (
        "Existing R1 recognition manifest differs. STOP."
    )

else:

    R1_RECOGNITION_MANIFEST_PATH.write_text(
        manifest_text,
        encoding="utf-8"
    )


R1_RECOGNITION_MANIFEST_SHA = sha256_file(
    R1_RECOGNITION_MANIFEST_PATH
)


# ==================================================================================================
# 13. APPLY PRE-FROZEN STOP RULE
# ==================================================================================================

if R1_RECOGNITION_SUPPORTS_CALIBRATION_STAGE:

    R1_NEXT_ACTION = (
        "PROCEED_TO_R1_VALIDATION_TEMPERATURE_CALIBRATION"
    )

else:

    R1_NEXT_ACTION = (
        "STOP_METHOD_DEVELOPMENT_ON_CURRENT_VALIDATION_SPLITS"
    )


# ==================================================================================================
# 14. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_REVISION_R1_RECOGNITION_STAGE_COMPLETE = True


print()
print("=" * 160)
print("NOTEBOOK 05 — R1 RECOGNITION-STAGE DECISION")
print("=" * 160)

print(
    "Datasets selecting beta > 0       :",
    len(
        datasets_with_nonzero_beta
    ),
    "/ 3"
)

print(
    "Datasets with beta > 0            :",
    datasets_with_nonzero_beta
)

print()

print(
    "Datasets passing robustness gate  :",
    R1_RECOGNITION_PASS_DATASET_COUNT,
    "/ 3"
)

print(
    "Passing datasets                  :",
    datasets_passing_recognition
)

print()

print(
    "Recognition supports calibration  :",
    R1_RECOGNITION_SUPPORTS_CALIBRATION_STAGE
)

print(
    "NEXT ACTION                       :",
    R1_NEXT_ACTION
)

print()

print(
    "Seed-42 grid artifact:"
)

print(
    R1_TUNING_GRID_PATH
)

print(
    "SHA-256:"
)

print(
    R1_TUNING_GRID_SHA
)

print()

print(
    "Frozen evaluation artifact:"
)

print(
    R1_EVALUATION_PATH
)

print(
    "SHA-256:"
)

print(
    R1_EVALUATION_SHA
)

print()

print(
    "Recognition manifest:"
)

print(
    R1_RECOGNITION_MANIFEST_PATH
)

print(
    "SHA-256:"
)

print(
    R1_RECOGNITION_MANIFEST_SHA
)

print()

print(
    "Temperature fitted               : False"
)

print(
    "Training performed                :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used                :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 R1 RECOGNITION STAGE: COMPLETE"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 160)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 21A-R
# R1 RECOGNITION-STAGE COMPARATOR CORRECTION
#
# ISSUE FOUND:
#
#   Cell 21A tuned R1 on seed 42 and froze its parameters,
#   BUT compared robustness seeds against Standard-LA parameters
#   re-selected independently for each seed.
#
#   This creates an unfair comparator mismatch.
#
# CORRECTION:
#
#   Standard LA must follow the SAME development structure:
#
#       select tau on seed 42
#       freeze tau
#       apply unchanged to seeds 123 and 3407
#
#   R1 is then compared against this frozen seed-42 Standard-LA baseline.
#
# NO NEW TUNING
# NO FORMULA CHANGE
# NO GRID CHANGE
# NO TEMPERATURE FITTING
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json

import numpy as np
import pandas as pd


# ==================================================================================================
# 1. PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_REVISION_R1_RECOGNITION_STAGE_COMPLETE is True

assert NOTEBOOK05_REVISION_R1_PROTOCOL_FROZEN is True
assert NOTEBOOK05_REVISION_R1_ROBUSTNESS_GATE_FROZEN is True

assert DEVICE.type == "cpu"

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


CORRECTION_TOL = 1e-12


# ==================================================================================================
# 2. FREEZE STANDARD-LA PARAMETERS FROM SEED 42 ONLY
# ==================================================================================================

R1_FROZEN_STANDARD_LA_PARAMETERS = {}


print("=" * 165)
print("NOTEBOOK 05 — R1 COMPARATOR CORRECTION")
print("=" * 165)

print()
print(
    "Problem in Cell 21A:"
)

print(
    "  R1 parameters were frozen from seed 42,"
)

print(
    "  but Standard-LA reference tau was selected independently per seed."
)

print()

print(
    "Correct comparison:"
)

print(
    "  frozen R1(seed42 tau,beta)"
)

print(
    "  versus"
)

print(
    "  frozen Standard-LA(seed42 tau)"
)

print()

print(
    "Official test: SEALED"
)

print("=" * 165)


for dataset in R1_DATASETS:

    seed42_standard = (

        CORRECTED_STANDARD_LA_SELECTION[
            (
                dataset,
                42
            )
        ]
    )


    frozen_tau = float(
        seed42_standard[
            "tau"
        ]
    )


    R1_FROZEN_STANDARD_LA_PARAMETERS[
        dataset
    ] = {

        "tau":
            frozen_tau,

        "selection_seed":
            42,
    }


    print(

        f"{dataset:11s} | "
        f"Frozen Standard-LA tau={frozen_tau:.2f} | "
        f"Frozen R1 tau={R1_SELECTED_PARAMETERS[dataset]['tau']:.2f}, "
        f"beta={R1_SELECTED_PARAMETERS[dataset]['beta']:.2f}"
    )


# ==================================================================================================
# 3. EXPECTED PARAMETER RELATION
#
# Seed42 selected:
#
#   Standard LA tau = 0
#   R1 tau = 0, beta = 0
#
# for all three datasets.
# ==================================================================================================

for dataset in R1_DATASETS:

    assert np.isclose(

        R1_FROZEN_STANDARD_LA_PARAMETERS[
            dataset
        ][
            "tau"
        ],

        0.0,

        rtol=0.0,
        atol=1e-15
    )


    assert np.isclose(

        R1_SELECTED_PARAMETERS[
            dataset
        ][
            "tau"
        ],

        0.0,

        rtol=0.0,
        atol=1e-15
    )


    assert np.isclose(

        R1_SELECTED_PARAMETERS[
            dataset
        ][
            "beta"
        ],

        0.0,

        rtol=0.0,
        atol=1e-15
    )


# ==================================================================================================
# 4. CORRECTED FROZEN-PARAMETER EVALUATION
# ==================================================================================================

R1_CORRECTED_EVALUATION_ROWS = []


print()
print("=" * 165)
print("CORRECTED R1 VS FROZEN STANDARD-LA EVALUATION")
print("=" * 165)


for dataset in R1_DATASETS:

    payload = PRIOR_RARITY_BY_DATASET[
        dataset
    ]


    prior = np.asarray(
        payload[
            "prior"
        ],
        dtype=np.float64
    )


    rarity = np.asarray(
        payload[
            "rarity"
        ],
        dtype=np.float64
    )


    frequency_groups = np.asarray(
        payload[
            "frequency_group"
        ],
        dtype=object
    )


    r1_tau = float(
        R1_SELECTED_PARAMETERS[
            dataset
        ][
            "tau"
        ]
    )


    r1_beta = float(
        R1_SELECTED_PARAMETERS[
            dataset
        ][
            "beta"
        ]
    )


    frozen_la_tau = float(

        R1_FROZEN_STANDARD_LA_PARAMETERS[
            dataset
        ][
            "tau"
        ]
    )


    for seed in [
        42,
        123,
        3407,
    ]:

        identity = (
            dataset,
            seed
        )


        validation = load_bestAA_validation_output(

            FINAL_BEST_AA_VALIDATION_OUTPUTS[
                identity
            ]
        )


        logits = np.asarray(
            validation[
                "logits"
            ],
            dtype=np.float64
        )


        labels = np.asarray(
            validation[
                "labels"
            ],
            dtype=np.int64
        )


        # ------------------------------------------------------------------------------------------
        # R1
        # ------------------------------------------------------------------------------------------

        r1_logits = revision_r1_logits(

            logits,
            prior,
            rarity,

            tau=r1_tau,
            beta=r1_beta,

            eps=RARITY_EPSILON
        )


        r1_metrics = evaluate_logits(

            r1_logits,
            labels,
            frequency_groups
        )


        # ------------------------------------------------------------------------------------------
        # FAIR FROZEN STANDARD-LA BASELINE
        # ------------------------------------------------------------------------------------------

        frozen_la_logits = standard_la_logits(

            logits,
            prior,

            tau=frozen_la_tau,

            eps=RARITY_EPSILON
        )


        frozen_la_metrics = evaluate_logits(

            frozen_la_logits,
            labels,
            frequency_groups
        )


        # ------------------------------------------------------------------------------------------
        # Because R1 beta=0 and tau matches frozen Standard LA,
        # logits MUST be exactly identical.
        # ------------------------------------------------------------------------------------------

        assert np.array_equal(

            r1_logits,

            frozen_la_logits
        ), (
            f"{dataset} seed {seed}: "
            "R1 should exactly equal frozen Standard LA."
        )


        row = {

            "dataset":
                dataset,

            "seed":
                int(
                    seed
                ),

            "role":
                (
                    "tuning"
                    if seed == 42
                    else "robustness"
                ),

            "r1_tau":
                r1_tau,

            "r1_beta":
                r1_beta,

            "frozen_standard_la_tau":
                frozen_la_tau,

            "delta_oa":
                float(
                    r1_metrics[
                        "oa"
                    ]
                    -
                    frozen_la_metrics[
                        "oa"
                    ]
                ),

            "delta_aa":
                float(
                    r1_metrics[
                        "aa"
                    ]
                    -
                    frozen_la_metrics[
                        "aa"
                    ]
                ),

            "delta_macro_f1":
                float(
                    r1_metrics[
                        "macro_f1"
                    ]
                    -
                    frozen_la_metrics[
                        "macro_f1"
                    ]
                ),

            "delta_tail_accuracy":
                float(
                    r1_metrics[
                        "tail_accuracy"
                    ]
                    -
                    frozen_la_metrics[
                        "tail_accuracy"
                    ]
                ),

            "delta_worst_class_accuracy":
                float(
                    r1_metrics[
                        "worst_class_accuracy"
                    ]
                    -
                    frozen_la_metrics[
                        "worst_class_accuracy"
                    ]
                ),

            "delta_nll":
                float(
                    r1_metrics[
                        "nll"
                    ]
                    -
                    frozen_la_metrics[
                        "nll"
                    ]
                ),
        }


        R1_CORRECTED_EVALUATION_ROWS.append(
            row
        )


        print()

        print(

            f"{dataset:11s} | "
            f"seed={seed:4d} | "
            f"{row['role']:10s}"
        )


        print(

            f"  ΔOA    = {row['delta_oa']:+.12e} | "
            f"ΔAA    = {row['delta_aa']:+.12e} | "
            f"ΔF1    = {row['delta_macro_f1']:+.12e}"
        )


        print(

            f"  ΔTail  = {row['delta_tail_accuracy']:+.12e} | "
            f"ΔWorst = {row['delta_worst_class_accuracy']:+.12e} | "
            f"ΔNLL   = {row['delta_nll']:+.12e}"
        )


R1_CORRECTED_EVALUATION_DF = pd.DataFrame(
    R1_CORRECTED_EVALUATION_ROWS
)


assert len(
    R1_CORRECTED_EVALUATION_DF
) == 9


# ==================================================================================================
# 5. ALL DELTAS MUST BE ZERO
# ==================================================================================================

DELTA_COLUMNS = [

    "delta_oa",
    "delta_aa",
    "delta_macro_f1",
    "delta_tail_accuracy",
    "delta_worst_class_accuracy",
    "delta_nll",
]


for column in DELTA_COLUMNS:

    assert np.allclose(

        R1_CORRECTED_EVALUATION_DF[
            column
        ].to_numpy(
            dtype=np.float64
        ),

        0.0,

        rtol=0.0,

        atol=CORRECTION_TOL
    ), (
        f"Expected zero corrected R1 delta: {column}"
    )


# ==================================================================================================
# 6. CORRECTED ROBUSTNESS GATE
# ==================================================================================================

R1_CORRECTED_RECOGNITION_GATE = {}


print()
print("=" * 165)
print("CORRECTED R1 ROBUSTNESS-SEED RECOGNITION GATE")
print("=" * 165)


for dataset in R1_DATASETS:

    subset = (

        R1_CORRECTED_EVALUATION_DF[
            (
                R1_CORRECTED_EVALUATION_DF[
                    "dataset"
                ]
                ==
                dataset
            )
            &
            (
                R1_CORRECTED_EVALUATION_DF[
                    "role"
                ]
                ==
                "robustness"
            )
        ]
    )


    assert len(
        subset
    ) == 2


    mean_deltas = {

        "AA":
            float(
                subset[
                    "delta_aa"
                ].mean()
            ),

        "Macro_F1":
            float(
                subset[
                    "delta_macro_f1"
                ].mean()
            ),

        "Tail_Accuracy":
            float(
                subset[
                    "delta_tail_accuracy"
                ].mean()
            ),

        "Worst_Class_Accuracy":
            float(
                subset[
                    "delta_worst_class_accuracy"
                ].mean()
            ),

        "OA":
            float(
                subset[
                    "delta_oa"
                ].mean()
            ),
    }


    recognition_signal = bool(

        any(

            value
            >
            CORRECTION_TOL

            for metric, value
            in mean_deltas.items()

            if metric != "OA"
        )
    )


    large_oa_collapse = bool(

        (
            subset[
                "delta_oa"
            ]
            <
            -
            REVISION_R1_LARGE_OA_COLLAPSE_THRESHOLD
        ).any()
    )


    gate_pass = bool(

        recognition_signal

        and

        not large_oa_collapse
    )


    R1_CORRECTED_RECOGNITION_GATE[
        dataset
    ] = {

        "mean_deltas":
            mean_deltas,

        "recognition_signal":
            recognition_signal,

        "large_oa_collapse":
            large_oa_collapse,

        "recognition_gate_pass":
            gate_pass,
    }


    print()

    print(
        dataset.upper()
    )

    print(
        "  Mean robustness ΔAA     :",
        f"{mean_deltas['AA']:+.12e}"
    )

    print(
        "  Mean robustness ΔF1     :",
        f"{mean_deltas['Macro_F1']:+.12e}"
    )

    print(
        "  Mean robustness ΔTail   :",
        f"{mean_deltas['Tail_Accuracy']:+.12e}"
    )

    print(
        "  Mean robustness ΔWorst  :",
        f"{mean_deltas['Worst_Class_Accuracy']:+.12e}"
    )

    print(
        "  Mean robustness ΔOA     :",
        f"{mean_deltas['OA']:+.12e}"
    )

    print(
        "  Recognition signal      :",
        recognition_signal
    )

    print(
        "  Large OA collapse       :",
        large_oa_collapse
    )

    print(
        "  RECOGNITION GATE        :",
        (
            "PASS"
            if gate_pass
            else "FAIL"
        )
    )


# ==================================================================================================
# 7. CORRECTED CROSS-DATASET DECISION
# ==================================================================================================

R1_CORRECTED_PASSING_DATASETS = [

    dataset

    for dataset, gate
    in R1_CORRECTED_RECOGNITION_GATE.items()

    if gate[
        "recognition_gate_pass"
    ]
]


assert len(
    R1_CORRECTED_PASSING_DATASETS
) == 0


R1_CORRECTED_RECOGNITION_SUPPORTS_CALIBRATION = False


R1_CORRECTED_NEXT_ACTION = (
    "STOP_METHOD_DEVELOPMENT_ON_CURRENT_VALIDATION_SPLITS"
)


# ==================================================================================================
# 8. SAVE CORRECTED EVALUATION
# ==================================================================================================

R1_CORRECTED_EVALUATION_PATH = (

    NB05_AUDIT_ROOT
    /
    "revision_R1_corrected_frozen_baseline_validation_evaluation.csv"
)


corrected_csv_text = (

    R1_CORRECTED_EVALUATION_DF.to_csv(
        index=False,
        float_format="%.17g"
    )
)


if R1_CORRECTED_EVALUATION_PATH.exists():

    assert (
        R1_CORRECTED_EVALUATION_PATH.read_text(
            encoding="utf-8"
        )
        ==
        corrected_csv_text
    ), (
        "Existing corrected R1 evaluation differs. STOP."
    )

else:

    R1_CORRECTED_EVALUATION_PATH.write_text(

        corrected_csv_text,

        encoding="utf-8"
    )


R1_CORRECTED_EVALUATION_SHA = sha256_file(
    R1_CORRECTED_EVALUATION_PATH
)


# ==================================================================================================
# 9. SAVE CORRECTION MANIFEST
# ==================================================================================================

R1_CORRECTION_MANIFEST_PATH = (

    NB05_AUDIT_ROOT
    /
    "revision_R1_recognition_comparator_correction_manifest.json"
)


R1_CORRECTION_MANIFEST = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "revision":
        "R1",

    "status":
        "RECOGNITION_STAGE_CORRECTED",

    "correction_reason":
        (
            "Cell 21A compared seed42-frozen R1 parameters against "
            "Standard-LA parameters selected independently per seed. "
            "This confounded the Qingyun seed3407 comparison because "
            "R1 used tau=0,beta=0 while the per-seed Standard-LA "
            "reference used tau=1."
        ),

    "correct_comparator":
        (
            "Seed42-selected Standard-LA tau frozen across "
            "seeds 42,123,3407."
        ),

    "frozen_standard_la_parameters":
        R1_FROZEN_STANDARD_LA_PARAMETERS,

    "r1_selected_parameters":
        R1_SELECTED_PARAMETERS,

    "corrected_dataset_recognition_gates":
        R1_CORRECTED_RECOGNITION_GATE,

    "corrected_passing_datasets":
        R1_CORRECTED_PASSING_DATASETS,

    "corrected_recognition_supports_calibration_stage":
        False,

    "corrected_next_action":
        R1_CORRECTED_NEXT_ACTION,

    "supersedes_recognition_manifest_sha256":
        R1_RECOGNITION_MANIFEST_SHA,

    "corrected_evaluation_csv_sha256":
        R1_CORRECTED_EVALUATION_SHA,

    "r1_protocol_sha256":
        REVISION_R1_PROTOCOL_SHA,

    "r1_robustness_gate_sha256":
        REVISION_R1_GATE_AMENDMENT_SHA,

    "formula_changed":
        False,

    "grid_changed":
        False,

    "parameters_retuned":
        False,

    "temperature_fitted":
        False,

    "training_performed":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


manifest_text = (

    json.dumps(
        R1_CORRECTION_MANIFEST,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if R1_CORRECTION_MANIFEST_PATH.exists():

    assert (
        R1_CORRECTION_MANIFEST_PATH.read_text(
            encoding="utf-8"
        )
        ==
        manifest_text
    ), (
        "Existing R1 correction manifest differs. STOP."
    )

else:

    R1_CORRECTION_MANIFEST_PATH.write_text(

        manifest_text,

        encoding="utf-8"
    )


R1_CORRECTION_MANIFEST_SHA = sha256_file(
    R1_CORRECTION_MANIFEST_PATH
)


# ==================================================================================================
# 10. OVERRIDE IN-MEMORY INVALID CELL-21A DECISION
# ==================================================================================================

R1_RECOGNITION_SUPPORTS_CALIBRATION_STAGE = False

R1_RECOGNITION_PASS_DATASET_COUNT = 0

datasets_passing_recognition = []

R1_NEXT_ACTION = (
    "STOP_METHOD_DEVELOPMENT_ON_CURRENT_VALIDATION_SPLITS"
)


NOTEBOOK05_REVISION_R1_RECOGNITION_COMPARATOR_CORRECTED = True


# ==================================================================================================
# 11. FINAL SAFETY GATE
# ==================================================================================================

assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False
assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False

assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


# ==================================================================================================
# 12. REPORT
# ==================================================================================================

print()
print("=" * 165)
print("NOTEBOOK 05 — CORRECTED R1 RECOGNITION DECISION")
print("=" * 165)

print()

print(
    "R1 selected beta > 0 datasets       : 0 / 3"
)

print(
    "R1 differs from frozen Standard LA  : NO"
)

print()

print(
    "Datasets passing corrected gate     :",
    len(
        R1_CORRECTED_PASSING_DATASETS
    ),
    "/ 3"
)

print(
    "Passing datasets                    :",
    R1_CORRECTED_PASSING_DATASETS
)

print()

print(
    "Recognition supports calibration    :",
    R1_RECOGNITION_SUPPORTS_CALIBRATION_STAGE
)

print(
    "Temperature fitting warranted       : False"
)

print(
    "NEXT ACTION                         :",
    R1_NEXT_ACTION
)

print()

print(
    "Correction manifest:"
)

print(
    R1_CORRECTION_MANIFEST_PATH
)

print(
    "SHA-256:"
)

print(
    R1_CORRECTION_MANIFEST_SHA
)

print()

print(
    "Formula changed                     : False"
)

print(
    "Parameters retuned                  : False"
)

print(
    "Training performed                  :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official test used                  :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 R1 RECOGNITION COMPARATOR CORRECTION: PASS"
)

print(
    "R1 RECOGNITION STAGE: FAIL"
)

print(
    "DO NOT FIT R1 TEMPERATURE"
)

print(
    "STOP FURTHER METHOD REDESIGN ON CURRENT VALIDATION SPLITS"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 165)
```

```python
# ==================================================================================================
# NOTEBOOK 05 — CELL 21B
# TERMINAL VALIDATION-ONLY SCIENTIFIC DECISION
#
# PURPOSE
#   Freeze the final Notebook 05 method-development conclusion after:
#
#   ORIGINAL LTLC:
#       - selected alpha = 0 in all 9 corrected best-val-AA runs
#       - no selected recognition signal
#       - calibration signal present
#       - Tail-ECE not consistently improved
#       - final original gate = REVISE
#
#   REVISION R1:
#       - adaptive coefficient decoupled from tau
#       - protocol frozen before evaluation
#       - parameters selected on seed 42 only
#       - frozen to seeds 123 and 3407
#       - selected beta = 0 for all 3 datasets
#       - fair frozen Standard-LA comparator used
#       - R1 == frozen Standard LA exactly
#       - corrected recognition gate = FAIL for 3/3 datasets
#
# PRE-FROZEN STOP RULE:
#       - do not redesign another method on these same validation splits
#       - do not fit R1 temperature
#       - do not open official test
#
# CPU ONLY
# NO TRAINING
# NO OFFICIAL TEST
# ==================================================================================================

import json
import numpy as np


# ==================================================================================================
# 1. HARD PRECONDITIONS
# ==================================================================================================

assert NOTEBOOK05_FINAL_VALIDATION_DEVELOPMENT_GATE_COMPLETE is True

assert NOTEBOOK05_DEVELOPMENT_REVISE is True

assert NOTEBOOK05_REVISION_R1_PROTOCOL_FROZEN is True

assert NOTEBOOK05_REVISION_R1_ROBUSTNESS_GATE_FROZEN is True

assert NOTEBOOK05_REVISION_R1_RECOGNITION_STAGE_COMPLETE is True

assert NOTEBOOK05_REVISION_R1_RECOGNITION_COMPARATOR_CORRECTED is True


assert DEVICE.type == "cpu"


assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False

assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False

assert NOTEBOOK05_OFFICIAL_TEST_USED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False


assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


# ==================================================================================================
# 2. ORIGINAL LTLC TERMINAL FINDING
# ==================================================================================================

ORIGINAL_LTLC_FINAL_FINDING = {

    "status":
        "REVISE",

    "selected_runs":
        9,

    "selected_nonzero_alpha_runs":
        0,

    "classification_signal_dataset_count":
        0,

    "calibration_signal_dataset_count":
        3,

    "long_tail_recognition_claim_supported":
        False,

    "validation_calibration_signal_present":
        True,

    "tail_ece_uniformly_improved":
        False,

    "interpretation":
        (
            "The original rarity-adaptive shift did not produce a "
            "selected recognition improvement. The samplewise "
            "temperature component produced validation calibration "
            "improvements, particularly NLL, but calibration alone "
            "cannot establish long-tail recognition because positive "
            "scalar temperature preserves hard predictions."
        ),
}


assert (
    FINAL_DEVELOPMENT_GATE_OBJECT[
        "status"
    ]
    ==
    "REVISE"
)


assert (
    FINAL_DEVELOPMENT_GATE_OBJECT[
        "cross_dataset_summary"
    ][
        "nonzero_alpha_runs"
    ]
    ==
    0
)


assert (
    FINAL_DEVELOPMENT_GATE_OBJECT[
        "cross_dataset_summary"
    ][
        "classification_signal_dataset_count"
    ]
    ==
    0
)


assert (
    FINAL_DEVELOPMENT_GATE_OBJECT[
        "cross_dataset_summary"
    ][
        "calibration_signal_dataset_count"
    ]
    ==
    3
)


# ==================================================================================================
# 3. R1 TERMINAL FINDING
# ==================================================================================================

R1_SELECTED_BETA_VALUES = {

    dataset:
        float(
            R1_SELECTED_PARAMETERS[
                dataset
            ][
                "beta"
            ]
        )

    for dataset in [
        "Pingan",
        "Qingyun",
        "Tangdaowan",
    ]
}


assert all(

    np.isclose(
        beta,
        0.0,
        rtol=0.0,
        atol=1e-15
    )

    for beta in R1_SELECTED_BETA_VALUES.values()
)


assert (
    R1_CORRECTED_PASSING_DATASETS
    ==
    []
)


assert (
    R1_CORRECTED_RECOGNITION_SUPPORTS_CALIBRATION
    is False
)


for dataset in [
    "Pingan",
    "Qingyun",
    "Tangdaowan",
]:

    gate = (
        R1_CORRECTED_RECOGNITION_GATE[
            dataset
        ]
    )


    assert gate[
        "recognition_signal"
    ] is False


    assert gate[
        "recognition_gate_pass"
    ] is False


    for value in gate[
        "mean_deltas"
    ].values():

        assert np.isclose(
            float(value),
            0.0,
            rtol=0.0,
            atol=1e-12
        )


R1_FINAL_FINDING = {

    "status":
        "FAIL_RECOGNITION_STAGE",

    "selected_parameters": {

        dataset: {

            "tau":
                float(
                    R1_SELECTED_PARAMETERS[
                        dataset
                    ][
                        "tau"
                    ]
                ),

            "beta":
                float(
                    R1_SELECTED_PARAMETERS[
                        dataset
                    ][
                        "beta"
                    ]
                ),
        }

        for dataset in [
            "Pingan",
            "Qingyun",
            "Tangdaowan",
        ]
    },

    "datasets_selecting_nonzero_beta":
        [],

    "datasets_passing_corrected_recognition_gate":
        [],

    "recognition_pass_dataset_count":
        0,

    "r1_differs_from_frozen_standard_la":
        False,

    "temperature_fitting_warranted":
        False,

    "interpretation":
        (
            "After correcting the robustness comparator, R1 is exactly "
            "identical to the seed-42-frozen Standard-LA reference on "
            "all datasets and seeds because seed 42 selected tau=0 and "
            "beta=0 for every dataset. R1 therefore provides no "
            "recognition evidence."
        ),
}


# ==================================================================================================
# 4. INVALID INTERMEDIATE CELL-21A DECISION
#
# Preserve history rather than deleting it.
# ==================================================================================================

SUPERSEDED_INTERMEDIATE_RESULT = {

    "artifact_sha256":
        R1_RECOGNITION_MANIFEST_SHA,

    "status":
        "SUPERSEDED_BY_COMPARATOR_CORRECTION",

    "reason":
        (
            "The original Cell 21A robustness comparison used "
            "per-seed-selected Standard-LA tau values while R1 used "
            "seed-42-frozen parameters. This caused the apparent "
            "Qingyun robustness signal. The result is superseded by "
            "the fair frozen-baseline comparator correction."
        ),

    "superseding_artifact_sha256":
        R1_CORRECTION_MANIFEST_SHA,
}


# ==================================================================================================
# 5. APPLY THE PRE-FROZEN STOP RULE
# ==================================================================================================

assert (
    REVISION_R1_STOP_RULE[
        "if_no_robust_recognition_signal"
    ]
    ==
    "STOP_METHOD_DEVELOPMENT_ON_CURRENT_VALIDATION_SPLITS"
)


NOTEBOOK05_FURTHER_METHOD_REDESIGN_ALLOWED = False

NOTEBOOK05_R1_TEMPERATURE_FITTING_ALLOWED = False

NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED = False


TERMINAL_REQUIRED_ACTION = (
    "STOP_METHOD_DEVELOPMENT_ON_CURRENT_VALIDATION_SPLITS"
)


# ==================================================================================================
# 6. CLAIM BOUNDARIES
# ==================================================================================================

NOTEBOOK05_CLAIM_BOUNDARIES = {

    "may_claim_original_ltlc_long_tail_recognition_improvement":
        False,

    "may_claim_r1_long_tail_recognition_improvement":
        False,

    "may_claim_validation_calibration_signal":
        True,

    "may_claim_uniform_tail_calibration_improvement":
        False,

    "may_claim_successful_final_ltlc_method":
        False,

    "allowed_interpretation":
        (
            "The experiments provide evidence that the tested LTLC "
            "shift formulations did not yield robust long-tail "
            "recognition gains under the locked validation protocol. "
            "The temperature component showed calibration improvements "
            "on validation, but this constitutes calibration evidence "
            "rather than recognition evidence."
        ),
}


# ==================================================================================================
# 7. PUBLICATION / AUDIT DISCLOSURES
# ==================================================================================================

NOTEBOOK05_REQUIRED_DISCLOSURES = {

    "best_val_aa_checkpoint_recovery":
        (
            "Five missing best-validation-AA CE checkpoints were "
            "recovered by exact historical retraining before official "
            "test inspection."
        ),

    "tangdaowan_saturation_amendment":
        (
            "The original Tangdaowan-only development gate was found "
            "to be classification-saturated, motivating an additional "
            "validation-only diagnostic on Pingan and Qingyun."
        ),

    "original_ltlc_result":
        (
            "Original LTLC selected alpha=0 in all corrected runs and "
            "failed the recognition side of the development gate."
        ),

    "r1_revision":
        (
            "R1 was a validation-driven protocol revision that "
            "decoupled the adaptive coefficient from tau and was frozen "
            "before R1 evaluation."
        ),

    "r1_result":
        (
            "R1 selected beta=0 for all datasets and failed the "
            "corrected frozen-baseline recognition robustness gate."
        ),

    "cell21a_correction":
        (
            "An intermediate R1 recognition result was superseded after "
            "detecting an unfair comparator mismatch; the corrected "
            "comparison used seed-42-frozen Standard-LA parameters."
        ),

    "official_test":
        (
            "No Notebook 05 development decision used the official "
            "test set."
        ),
}


# ==================================================================================================
# 8. NOTEBOOK 05 TERMINAL STATUS
# ==================================================================================================

NOTEBOOK05_TERMINAL_STATUS = (
    "VALIDATION_METHOD_DEVELOPMENT_TERMINATED_NO_ROBUST_RECOGNITION_SIGNAL"
)


NOTEBOOK05_TERMINAL_DECISION = {

    "project":
        "LTLC",

    "notebook":
        "05_Posthoc_Calibration_LTLC",

    "status":
        NOTEBOOK05_TERMINAL_STATUS,

    "original_ltlc":
        ORIGINAL_LTLC_FINAL_FINDING,

    "revision_R1":
        R1_FINAL_FINDING,

    "superseded_intermediate_result":
        SUPERSEDED_INTERMEDIATE_RESULT,

    "claim_boundaries":
        NOTEBOOK05_CLAIM_BOUNDARIES,

    "required_disclosures":
        NOTEBOOK05_REQUIRED_DISCLOSURES,

    "stop_rule_applied":
        True,

    "required_action":
        TERMINAL_REQUIRED_ACTION,

    "further_method_redesign_allowed":
        False,

    "r1_temperature_fitting_allowed":
        False,

    "official_test_may_be_opened":
        False,

    "source_original_final_gate_sha256":
        FINAL_DEVELOPMENT_GATE_SHA,

    "source_r1_protocol_sha256":
        REVISION_R1_PROTOCOL_SHA,

    "source_r1_robustness_gate_sha256":
        REVISION_R1_GATE_AMENDMENT_SHA,

    "source_r1_intermediate_recognition_sha256":
        R1_RECOGNITION_MANIFEST_SHA,

    "source_r1_comparator_correction_sha256":
        R1_CORRECTION_MANIFEST_SHA,

    "training_performed_in_terminal_phase":
        False,

    "official_test_dataset_created":
        False,

    "official_test_dataloader_created":
        False,

    "official_test_used":
        False,
}


# ==================================================================================================
# 9. SAVE TERMINAL DECISION
# ==================================================================================================

NOTEBOOK05_TERMINAL_DECISION_PATH = (

    NB05_AUDIT_ROOT
    /
    "notebook05_terminal_validation_scientific_decision.json"
)


terminal_text = (

    json.dumps(
        NOTEBOOK05_TERMINAL_DECISION,
        indent=2,
        sort_keys=True,
        allow_nan=False
    )

    +

    "\n"
)


if NOTEBOOK05_TERMINAL_DECISION_PATH.exists():

    existing = (
        NOTEBOOK05_TERMINAL_DECISION_PATH.read_text(
            encoding="utf-8"
        )
    )


    assert existing == terminal_text, (
        "Existing Notebook05 terminal decision differs. STOP."
    )


else:

    NOTEBOOK05_TERMINAL_DECISION_PATH.write_text(

        terminal_text,

        encoding="utf-8"
    )


NOTEBOOK05_TERMINAL_DECISION_SHA = sha256_file(
    NOTEBOOK05_TERMINAL_DECISION_PATH
)


# ==================================================================================================
# 10. FINAL SAFETY ASSERTIONS
# ==================================================================================================

assert NOTEBOOK05_FURTHER_METHOD_REDESIGN_ALLOWED is False

assert NOTEBOOK05_R1_TEMPERATURE_FITTING_ALLOWED is False

assert NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED is False


assert NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED is False

assert NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED is False

assert NOTEBOOK05_OFFICIAL_TEST_USED is False


assert NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE is False


NOTEBOOK05_TERMINAL_SCIENTIFIC_DECISION_FROZEN = True


# ==================================================================================================
# 11. REPORT
# ==================================================================================================

print("=" * 165)
print("NOTEBOOK 05 — TERMINAL VALIDATION SCIENTIFIC DECISION")
print("=" * 165)

print()

print("ORIGINAL LTLC")
print("-" * 165)

print(
    "Selected nonzero-alpha runs      : 0 / 9"
)

print(
    "Classification-signal datasets   : 0 / 3"
)

print(
    "Calibration-signal datasets      : 3 / 3"
)

print(
    "Long-tail recognition supported  : False"
)

print()

print("REVISION R1")
print("-" * 165)

print(
    "Datasets selecting beta > 0      : 0 / 3"
)

print(
    "Corrected recognition passes     : 0 / 3"
)

print(
    "R1 differs from frozen LA        : False"
)

print(
    "R1 temperature fitting warranted : False"
)

print()

print("FINAL CLAIM BOUNDARY")
print("-" * 165)

print(
    "Successful LTLC recognition claim: NOT SUPPORTED"
)

print(
    "Validation calibration signal    : SUPPORTED"
)

print(
    "Uniform Tail-ECE improvement     : NOT SUPPORTED"
)

print()

print(
    "Further method redesign allowed  :",
    NOTEBOOK05_FURTHER_METHOD_REDESIGN_ALLOWED
)

print(
    "Official test may be opened      :",
    NOTEBOOK05_OFFICIAL_TEST_MAY_BE_OPENED
)

print()

print(
    "TERMINAL STATUS:"
)

print(
    NOTEBOOK05_TERMINAL_STATUS
)

print()

print(
    "REQUIRED ACTION:"
)

print(
    TERMINAL_REQUIRED_ACTION
)

print()

print(
    "Terminal decision artifact:"
)

print(
    NOTEBOOK05_TERMINAL_DECISION_PATH
)

print(
    "SHA-256:"
)

print(
    NOTEBOOK05_TERMINAL_DECISION_SHA
)

print()

print(
    "Training performed               :",
    NOTEBOOK05_TRAINING_PERFORMED_IN_CPU_PHASE
)

print(
    "Official-test dataset created    :",
    NOTEBOOK05_OFFICIAL_TEST_DATASET_CREATED
)

print(
    "Official-test DataLoader created :",
    NOTEBOOK05_OFFICIAL_TEST_DATALOADER_CREATED
)

print(
    "Official test used               :",
    NOTEBOOK05_OFFICIAL_TEST_USED
)

print()

print(
    "NOTEBOOK 05 TERMINAL SCIENTIFIC DECISION: FROZEN"
)

print(
    "NEXT: NOTEBOOK 05 MASTER AUDIT + COMPLETION GATE"
)

print(
    "OFFICIAL TEST REMAINS SEALED"
)

print("=" * 165)
```
